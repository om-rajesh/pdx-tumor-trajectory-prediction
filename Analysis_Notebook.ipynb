# Depth and durability of response in patient-derived xenografts: analysis notebook

This notebook carries out every analysis of the study, from the source data to the result tables, the result records and the data behind every figure. It draws no figures: they are drawn by the separate figures notebook, which reads only the files this notebook saves, so a change to a figure never requires rerunning the analysis. Each section opens with a note that states what the section does, the inputs it reads and the objects it produces. The code cells then carry out the analysis. Every quantity that the study reports is stored in a results register as it is computed, and Section 15 writes that register, the tables and a run manifest to `outputs/`.

### How to run

1. Create the environment from `environment.yml` and register its kernel under the name `pdxe`.
2. Place the input files under `data/` as listed below, and the plan files under `prespecification/`.
3. Run the notebook from the project folder (the folder that holds `pdxe_utils.py`) with *Restart and run all*.
4. Run the figures notebook, which reads the files written to `outputs/`.

Every random seed is fixed in Section 1.1, and each analysis cell creates its own random generator from its own seed. Section 15.2 checks the internal consistency of the stored results, and Section 15.3 records the software versions, the checksums of every input and output file, and the run time, and lists any file under `outputs/` that the run did not rewrite, so that no figure is drawn from a stale record (figures under `outputs/figures/`, which the figures notebook writes after this run, are left out of that check).

## 1 · Configuration and shared functions

Section 1.1 makes every import, fixes every path, seed and analysis constant, and prints the software versions. It also checks that every required file is present, stopping with the full list if one is missing, records the checksum of `pdxe_utils.py` and runs that module's self-test. Sections 1.2 and 1.3 define every function the analyses use, and Section 1.4 tests the notebook's own functions on short synthetic series with known answers. No later cell imports a package or defines a function.

Every analysis cell creates its own random generator from its seed in `SEEDS`, so the results of one cell do not depend on how many random numbers an earlier cell drew. Seeds added for the robustness analyses are new entries, so they leave every earlier stream unchanged. Each reported quantity is stored with `record()` as it is computed, and Section 15 writes the full register to `outputs/`.

**Outputs:** paths under `outputs/`, `SEEDS`, `SOFTWARE_VERSIONS`, `UTILS_SHA256`, `RUN_STARTED_UTC`, the shared functions, `NOTEBOOK_SELF_TESTS`.

```python
# ==========================================================================
# 1.1 Imports, software versions, paths, random seeds and analysis constants
#
# Every import in the notebook is made here, and every path, seed and
# constant shared by more than one section is fixed here before any data is
# read. Later cells import nothing; a constant used by a single analysis is
# set, with a comment, at the top of the cell that runs that analysis. The
# required input files are checked here too, so a missing one stops the run
# before any analysis starts.
# ==========================================================================

import hashlib
import json
import platform
import re
import sys
import time
import urllib.request
import warnings
from datetime import datetime, timezone
from itertools import combinations, product
from pathlib import Path

import numpy as np
import pandas as pd
import scipy
from scipy import stats
from scipy.optimize import least_squares
from scipy.sparse import coo_matrix
from scipy.sparse.csgraph import connected_components
import sklearn
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import GroupKFold, StratifiedKFold
from sklearn.preprocessing import StandardScaler
import statsmodels
import statsmodels.api as sm
import statsmodels.formula.api as smf
import jax
import numpyro
import numpyro.distributions as dist
from numpyro.diagnostics import effective_sample_size, split_gelman_rubin
from numpyro.infer import MCMC, NUTS, Predictive
import openpyxl
import pyarrow

# The run clock starts here; Section 15.3 records the elapsed time and the
# start time (UTC) in the manifest.
RUN_STARTED = time.perf_counter()
RUN_STARTED_UTC = datetime.now(timezone.utc).isoformat(timespec="seconds")
# Wall-clock start, used in Section 15.3 to flag any file under outputs/ that
# this run did not rewrite (a stale file the figures notebook could still read).
RUN_STARTED_EPOCH = time.time()


# --- project folder --------------------------------------------------------
# The project folder is the nearest folder, at or above the notebook's own
# folder, that contains pdxe_utils.py. Every other path is built from it, so
# the notebook runs unchanged from the project root or from notebooks/.
# JupyterLab starts the kernel in the notebook's folder; VS Code may start it
# in the workspace folder instead, so the notebook's own path (which VS Code
# provides as __vsc_ipynb_file__) is tried first.
def locate_project_root(marker="pdxe_utils.py"):
    """Return the nearest folder at or above the notebook (or working directory) holding `marker`."""
    starts = [Path.cwd()]
    if "__vsc_ipynb_file__" in globals():
        starts.insert(0, Path(globals()["__vsc_ipynb_file__"]).resolve().parent)
    for start in starts:
        for folder in (start, *start.parents):
            if (folder / marker).exists():
                return folder
    # Last resort: exactly one folder directly below the starting folder holds it.
    below = [folder for folder in Path.cwd().iterdir() if folder.is_dir() and (folder / marker).exists()]
    if len(below) == 1:
        print(f"Project folder found below the working directory: {below[0].name}")
        return below[0]
    raise FileNotFoundError(
        f"{marker} was not found at or above {', '.join(str(s) for s in starts)}. "
        "Open the notebook from the project folder that holds pdxe_utils.py.")


PROJECT_DIR = locate_project_root()
if str(PROJECT_DIR) not in sys.path:
    sys.path.insert(0, str(PROJECT_DIR))
import pdxe_utils as utils                     # shared endpoint functions

DATA_DIR = PROJECT_DIR / "data"                # inputs, never modified (9.3 may add the hallmark library once)
OUTPUT_DIR = PROJECT_DIR / "outputs"           # everything this run writes
PRESPEC_DIR = PROJECT_DIR / "prespecification" # plan files fixed before testing
FROZEN_DIR = OUTPUT_DIR / "frozen"             # per-analysis result records
TABLE_DIR = OUTPUT_DIR / "tables"
INTERIM_DIR = OUTPUT_DIR / "interim"           # Parquet copies of input sheets
for folder in (FROZEN_DIR, TABLE_DIR, INTERIM_DIR):
    folder.mkdir(parents=True, exist_ok=True)
# Written by the figures notebook, never by this one. Figure files are always
# older than this run (the figures notebook runs after it), so Section 15.3
# leaves this folder out of the stale-file check and of this manifest; the
# figures notebook writes its own manifest there.
FIGURE_DIR = OUTPUT_DIR / "figures"
# Run stamp: the figures notebook compares it with manifest.json, so figures
# can never be drawn from the records of an earlier run.
(FROZEN_DIR / "run_stamp.txt").write_text(RUN_STARTED_UTC, encoding="utf-8")

# Input files. The workbook is required. The others are optional or
# restricted (see the project layout at the top of the notebook), and the
# sections that use them print a skip line when they are absent.
WORKBOOK_PATH = DATA_DIR / "gao2015" / "41591_2015_BFnm3954_MOESM10_ESM.xlsx"
PDXNET_DIR = DATA_DIR / "pdxnet"               # PDXNet-PDTC-Export-*.csv
OBERG_PATH = DATA_DIR / "oberg2021" / "supp2_data.csv"
MSIGDB_DIR = DATA_DIR / "msigdb"               # hallmark gene sets, .gmt or Enrichr .txt
PDS_DIR = DATA_DIR / "project_data_sphere"     # A6181087/, CA031/, EFC10261/
# Hallmark gene sets for Section 9.3. When no library file is in data/msigdb/,
# the notebook downloads the Enrichr text version of the MSigDB hallmark
# collection once, keeps it only if it holds every set 9.3 needs, and saves
# it there, so later runs read the saved file (its checksum is in the
# manifest) and download nothing. False skips the download, and 9.3 then
# prints a skip line unless a library file is already present.
FETCH_HALLMARK_LIBRARY = True
HALLMARK_LIBRARY_URL = ("https://maayanlab.cloud/Enrichr/geneSetLibrary?mode=text"
                        "&libraryName=MSigDB_Hallmark_2020")
HALLMARK_LIBRARY_FILE = "MSigDB_Hallmark_2020.txt"

# --- checksums recorded before analysis ------------------------------------
# The workbook checksum is the published supplementary file. Each plan entry
# gives the file name in PRESPEC_DIR and the first 16 hexadecimal digits of
# the SHA-256 recorded when that plan was fixed.
EXPECTED_WORKBOOK_SHA256 = (
    "c4b9a6903a4d1f76e3ddca4199039776d56bb99970aa5b7abe4f3abd732a0c6d")
PRESPECIFIED_PLANS = {
    "mechanism_class": ("mechanism_family_plan.json", "7a013fd2991e9848"),
    "clinical_single_trial": ("clinical_single_trial_plan.json", "971249c51d66f110"),
    "clinical_pooled": ("clinical_pooled_plan.json", "2a34fec1d3aef86d"),
}

# --- required files, checked before anything else runs ----------------------
# The workbook, the utility module and the three plan files are required. A
# missing one stops the run here with the complete list, instead of part-way
# through Section 11 (mechanism plan) or Section 14 (clinical plans). The plan
# files must be committed under prespecification/ in the repository.
REQUIRED_FILES = [WORKBOOK_PATH, PROJECT_DIR / "pdxe_utils.py",
                  *[PRESPEC_DIR / file_name for file_name, _ in PRESPECIFIED_PLANS.values()]]
missing_required = [path.relative_to(PROJECT_DIR).as_posix() for path in REQUIRED_FILES
                    if not path.exists()]
if missing_required:
    raise FileNotFoundError(
        "Required files are missing: " + ", ".join(missing_required)
        + ". Place the published workbook under data/gao2015/ and commit the three plan "
          "files fixed before testing to prespecification/ (see the project layout above).")

# --- random seeds ----------------------------------------------------------
# One named seed per analysis. Each analysis cell creates its own generator
# from its seed, so no random stream carries over from one cell to the next.
SEED = utils.SEED                               # 20260905, the project seed
SEEDS = {
    "parameter_recovery": SEED,
    "mechanism_discrimination": SEED + 1,
    "design_map": SEED + 2,
    "prior_predictive": SEED,
    "calibration": SEED + 1000,                 # base for per-replicate keys
    "hazard_bootstrap": SEED + 4,
    "forecast_bootstrap": SEED + 5,
    "regret_bootstrap": SEED + 6,
    "predictor_controls": SEED + 7,
    "molecular_null": SEED + 8,
    "combination_bootstrap": SEED + 9,
    "molecular_bootstrap": SEED + 10,
    "pair_bootstrap": SEED + 11,
    "mechanism_demonstration": SEED + 12,
    "positive_control": SEED + 13,
    "separability_bootstrap": SEED + 15,
    "untreated_forecast_bootstrap": SEED + 16,
    "exposure_separation": 20260909,
    "exposure_window": 20260912,
    "exposure_confirmation": 20260910,
    "exposure_thinning": 20260913,
    "mechanism_class_model": 20260921,          # +1 depth, +2 durability
    "mechanism_class_compound": 20260922,       # +101 sign-flip, +200 + k LOCO
    "mechanism_label_permutation": 20260923,
    "external_noise_bootstrap": 20260930,
    "external_cohort": 20260931,
    "clinical_single_trial": 20260921,
    "clinical_pooled": 20260922,
    # Robustness analyses. Each has its own new seed, so adding them changes
    # no random stream used by the analyses above. They sit at SEED + 501
    # onward, clear of the seeds above and of the offsets derived from them
    # (+1, +2, +101, +200 + k, and SEED + 1000 + replicate).
    "paired_discrimination": SEED + 501,
    "rolling_bootstrap": SEED + 502,
    "holdout_bootstrap": SEED + 503,
    "baseline_slope_bootstrap": SEED + 504,
    "hallmark_null": SEED + 505,
    "hallmark_bootstrap": SEED + 506,
    "separation_growth_bootstrap": SEED + 507,
    # Seeds added with the review fixes, again new entries, so no stream above
    # changes. Each new analysis has its own seed.
    "noise_estimates": SEED + 508,               # 3.6 interpolation noise intervals
    "noise_baseline": SEED + 509,                # 4.5 noise-only discordance baseline
    "rho_bootstrap": SEED + 510,                 # 4.2 clustered intervals for rho
    "sensitivity_bootstrap": SEED + 511,         # 4.3 clustered intervals, every row
    "endpoint_sensitivity_bootstrap": SEED + 512,  # 7.4 window and zero rules
    "all_arm_selection_bootstrap": SEED + 513,   # 8.6 selection among all treated arms
    "combination_corrected_bootstrap": SEED + 514,  # 10 corrected combination gains
    "mechanism_dependence_bootstrap": SEED + 515,   # 11.4 three-way bootstrap
    "molecular_treatment_null": SEED + 516,      # 9.4 shuffle null, treatment adjusted
    "molecular_treatment_bootstrap": SEED + 517,  # 9.4 dissociation interval
    "external_ratio_bootstrap": SEED + 518,      # 13.3 corrected variance ratio
    "exposure_cluster_bootstrap": SEED + 519,    # 12.5 replicate-clustered intervals
    "exposure_weak_kill": SEED + 520,            # 12.5 weaker-kill simulations
    "design_map_bootstrap": SEED + 521,          # 5.4 intervals on median errors
    "separation_strata_bootstrap": SEED + 522,   # 8.4 strata (one generator for all)
    "scenario_bootstrap": SEED + 523,            # 8.1 censoring-scenario intervals
    "comparator_bootstrap": SEED + 524,          # 8.7 trajectory against the strongest comparators
    "stability_refits": SEED + 525,              # 8.8 refits on resampled training PDX models
    "termination_selection_bootstrap": SEED + 526,  # 8.8 selection with censored arms reinserted
    "replicate_noise": SEED + 527,               # 13.4 selection judged against a second animal
    "leakage_audit": SEED + 528,                 # 7.1 perturbation of readings after the origin
    "fold_partitions": SEED + 529,               # 8.8 repeated grouped partitions into folds
    # Seeds added with the second round of review fixes, again new entries,
    # so no stream above changes.
    "responder_discordance_bootstrap": SEED + 530,   # 4.6 discordance among responders
    "floor_plateau_bootstrap": SEED + 531,           # 4.6 complete responses held at the floor
    "learner_contrast_bootstrap": SEED + 532,        # 8.9 boosting against least squares
    "learner_free_bootstrap": SEED + 533,            # 8.9 ranking by raw day-10 quantities
    "change_outcome_bootstrap": SEED + 534,          # 8.9 regret on change from baseline
    "long_horizon_selection_bootstrap": SEED + 535,  # 8.9 selection at later start days and horizons
    "v600_combination_bootstrap": SEED + 536,        # 10.3 approved combinations by BRAF V600 status
    "encorafenib_v600_bootstrap": SEED + 537,        # 11.5 the encorafenib exception by BRAF V600 status
    "between_animal_discordance": SEED + 538,        # 13.5 discordance with between-animal variation
}
# Some analyses share a seed VALUE: the date-like seeds of Sections 11, 12
# and 14 coincide with SEED + k of other analyses, the clinical seeds are
# fixed in the plan files (checked in 14.1), the offsets of 11.2 (+1, +2)
# reach the seeds of 11.3 and of the label permutation, and the prior
# predictive simulation uses the project seed as a JAX key. Every analysis
# builds its own generator from its seed and draws for different data, so a
# shared value makes no reported result depend on another. The values are
# kept so that the plans' recorded seeds and every reported number stay as
# they are; the shared values are printed with the configuration below.
# The robustness seeds must not coincide with any seed used before them.
ROBUSTNESS_SEED_KEYS = ("paired_discrimination", "rolling_bootstrap", "holdout_bootstrap",
                        "baseline_slope_bootstrap", "hallmark_null", "hallmark_bootstrap",
                        "separation_growth_bootstrap", "noise_estimates", "noise_baseline",
                        "rho_bootstrap", "sensitivity_bootstrap", "endpoint_sensitivity_bootstrap",
                        "all_arm_selection_bootstrap", "combination_corrected_bootstrap",
                        "mechanism_dependence_bootstrap", "molecular_treatment_null",
                        "molecular_treatment_bootstrap", "external_ratio_bootstrap",
                        "exposure_cluster_bootstrap", "exposure_weak_kill", "design_map_bootstrap",
                        "separation_strata_bootstrap", "scenario_bootstrap", "comparator_bootstrap",
                        "stability_refits", "termination_selection_bootstrap", "replicate_noise",
                        "leakage_audit", "fold_partitions", "responder_discordance_bootstrap",
                        "floor_plateau_bootstrap", "learner_contrast_bootstrap", "learner_free_bootstrap",
                        "change_outcome_bootstrap", "long_horizon_selection_bootstrap",
                        "v600_combination_bootstrap", "encorafenib_v600_bootstrap",
                        "between_animal_discordance")
_earlier_seeds = {value for key, value in SEEDS.items() if key not in ROBUSTNESS_SEED_KEYS}
_robustness_seeds = [SEEDS[key] for key in ROBUSTNESS_SEED_KEYS]
assert len(set(_robustness_seeds)) == len(_robustness_seeds), "two robustness analyses share a seed"
assert not _earlier_seeds & set(_robustness_seeds), "a robustness seed repeats an earlier seed"

# --- measurement and eligibility constants ---------------------------------
VOLUME_FLOOR = utils.VOLUME_FLOOR                # 1 mm3
MIN_DAY_FOR_BEST_AVERAGE_RESPONSE = utils.MIN_DAY_FOR_BEST_AVERAGE_RESPONSE  # 10
CONTROL_LABEL = "untreated"                      # the only untreated label
MIN_MEASUREMENTS_PER_ARM = 5    # line fits: untreated growth, kill rates
MIN_POINTS_AFTER_NADIR = 2      # post-nadir arms need >= 2 later points
# Durability is the least-squares slope of log volume over the measurements
# AFTER the nadir. The nadir itself is left out: it is selected as the minimum
# of noisy readings, so its error is biased downward and a line through it
# overstates regrowth. Every durability value in the notebook comes from
# post_nadir_fit (Section 1.2) with this setting; True gives the fit that
# includes the nadir, which Section 4.3 reports as a sensitivity analysis.
DURABILITY_INCLUDES_NADIR = False
REGRESSION_FRACTION = 0.01      # durability phenotype: shrank >= 1%
REGROWTH_MULTIPLE = 1.20        # durability phenotype: regrew >= 1.2x nadir
TERMINATION_VOLUME = 2000.0     # mm3; simulated arms end above this volume
SIMULATION_NOISE_SD = 0.15      # log units; above the measured noise on purpose
RECOVERY_THRESHOLD = 0.25       # median absolute log error counted as recovered
KILL_RATE_RANGE = (1e-4, 1.0)   # per day; fits outside are failed fits

# --- prespecified forecasting endpoints ------------------------------------
FORECAST_ORIGIN_DAY = 10
FORECAST_HORIZON_DAY = 24
# Readings below 1 mm³ (recorded as zero or below 1 mm³, and set to the
# floor) in the day-24 outcome: "floor" (kept at 1 mm³, the rule used
# everywhere else) or "drop" (left out, as in the forecasting analysis first
# run). "drop" removes arms still at the floor on day 24, the best responders;
# Section 7.4 reports both endpoints under "floor".
FORECAST_OUTCOME_ZEROS = "drop"
# Best average response in the day-10 window is defined from day 10 onward,
# so with readings on or before day 10 it exists only for arms measured ON
# day 10. "exact_day" keeps that rule (the analysis first run); "last_reading"
# falls back to the running average at the last reading on or before the
# origin, if that reading is no more than WINDOW_RESPONSE_TOLERANCE_DAYS
# before it. No reading after the origin is used under either rule.
FORECAST_WINDOW_RESPONSE = "exact_day"
WINDOW_RESPONSE_TOLERANCE_DAYS = 3.5      # one median measurement interval
FORECAST_MARGIN = 0.10          # relative RMSE reduction
REGRET_MARGIN = 0.15            # relative reduction in mean regret
N_FOLDS = 5
N_BOOTSTRAP = 2000              # default bootstrap replicates
N_BOOTSTRAP_CONTROLS = 1500     # control predictor sets

# --- display ---------------------------------------------------------------
pd.set_option("display.max_columns", 60)
pd.set_option("display.width", 200)
pd.set_option("display.max_colwidth", 60)

# --- software versions -----------------------------------------------------
SOFTWARE_VERSIONS = {
    "python": platform.python_version(), "numpy": np.__version__,
    "pandas": pd.__version__, "scipy": scipy.__version__,
    "scikit-learn": sklearn.__version__, "statsmodels": statsmodels.__version__,
    "numpyro": numpyro.__version__, "jax": jax.__version__,
    "openpyxl": openpyxl.__version__,
    "pyarrow": pyarrow.__version__,
}
PLATFORM = f"{platform.system()} {platform.release()} {platform.machine()}"
print("SOFTWARE")
for name, version in SOFTWARE_VERSIONS.items():
    print(f"  {name:<13} {version}")
print(f"  platform      {PLATFORM}")
print(f"  run started   {RUN_STARTED_UTC} (UTC)")

# --- the shared utility module ---------------------------------------------
# pdxe_utils holds best average response, the per-arm feature table, the
# volume floor and same-day averaging. Its checksum is recorded with the
# results, and its self-test (synthetic trajectories with known answers)
# must pass before anything else runs.
UTILS_SHA256 = hashlib.sha256((PROJECT_DIR / "pdxe_utils.py").read_bytes()).hexdigest()
print("\nUTILITY MODULE")
print(f"  pdxe_utils.py SHA-256  {UTILS_SHA256}")
print(f"  version                {getattr(utils, '__version__', 'not set')}")
assert utils.run_self_test(), "pdxe_utils self-test failed"

print("\nCONFIGURATION")
print(f"  seed                          {SEED}")
print(f"  volume floor                  {VOLUME_FLOOR:g} mm³")
print(f"  best average response from    day {MIN_DAY_FOR_BEST_AVERAGE_RESPONSE:g}")
print(f"  durability fit                nadir {'included' if DURABILITY_INCLUDES_NADIR else 'left out'}, "
      f">= {MIN_POINTS_AFTER_NADIR} points after it")
print(f"  forecast                      day {FORECAST_ORIGIN_DAY} to day {FORECAST_HORIZON_DAY}")
print(f"  window response rule          {FORECAST_WINDOW_RESPONSE} "
      f"(tolerance {WINDOW_RESPONSE_TOLERANCE_DAYS:g} days for last_reading)")
print(f"  day-24 readings at the floor  {FORECAST_OUTCOME_ZEROS}")
print(f"  margins                       RMSE {FORECAST_MARGIN:.0%}, regret {REGRET_MARGIN:.0%}")
print(f"  hallmark library download     {'on' if FETCH_HALLMARK_LIBRARY else 'off'} "
      "(Section 9.3, only when no library file is present)")
# Seed values used by more than one analysis (see the note on SEEDS above).
seed_users = {}
for key, value in SEEDS.items():
    seed_users.setdefault(value, []).append(key)
print("  seed values shared by analyses (separate generators, different data):")
for value, keys in seed_users.items():
    if len(keys) > 1:
        print(f"    {value}: {', '.join(keys)}")
```

```python
# ==========================================================================
# 1.2 Functions: records, checksums, statistics, the canonical
#     observation table, the durability endpoint, mechanistic models and
#     model fitting
#
# Every function used by more than one analysis is defined here, once.
# Functions used by a single analysis are here too, because no later cell
# defines a function (Section 1.4 tests the ones the analyses depend on).
# ==========================================================================

# --------------------------------------------------------------------------
# Results register. Each analysis stores the quantities it reports under a
# descriptive key as it computes them; Section 15 writes the register to
# outputs/ and checks it for internal consistency. RESULT_SECTION remembers
# the section that stored each key, for the register's section column.
# --------------------------------------------------------------------------
RESULTS = {}
RESULT_SECTION = {}
CURRENT_SECTION = {"label": "1"}


def set_section(label):
    """Name the section whose values the following record() calls store."""
    CURRENT_SECTION["label"] = str(label)


def record(key, value):
    """Store one reported value under `key` and return it unchanged.

    NumPy scalars are converted to Python numbers so that the register can be
    written to JSON. A key stored twice keeps the later value, and the
    section column follows the later call.
    """
    if isinstance(value, np.generic):
        value = value.item()
    RESULTS[key] = value
    RESULT_SECTION[key] = CURRENT_SECTION["label"]
    return value


def relative(path):
    """A path relative to the project folder, so no output shows local folders."""
    try:
        return Path(path).resolve().relative_to(PROJECT_DIR.resolve()).as_posix()
    except ValueError:
        return Path(path).name


def sha256_of_file(path, chunk_bytes=1 << 20):
    """Streaming SHA-256 of a file's bytes."""
    digest = hashlib.sha256()
    with open(path, "rb") as handle:
        for block in iter(lambda: handle.read(chunk_bytes), b""):
            digest.update(block)
    return digest.hexdigest()


def plan_sha256(path):
    """SHA-256 of a plan file with line endings normalized to LF.

    Each plan's checksum was computed on the plan text before it was written,
    and the written file carries the line endings of the system that wrote
    it, so carriage returns are removed before hashing.
    """
    return hashlib.sha256(Path(path).read_bytes().replace(b"\r\n", b"\n")).hexdigest()


def load_prespecified_plan(key):
    """Load a plan from PRESPEC_DIR and assert its recorded checksum."""
    file_name, recorded_prefix = PRESPECIFIED_PLANS[key]
    path = PRESPEC_DIR / file_name
    if not path.exists():
        raise FileNotFoundError(
            f"{relative(path)} is missing. Copy the plan file fixed before "
            "testing into the prespecification folder under this name.")
    digest = plan_sha256(path)
    assert digest.startswith(recorded_prefix), (
        f"{file_name}: SHA-256 {digest[:16]} does not match the recorded "
        f"{recorded_prefix}. The file is not the plan that was fixed.")
    return json.loads(path.read_text(encoding="utf-8")), digest


def save_record(frame, name, index=False):
    """Write one analysis record to FROZEN_DIR as CSV."""
    frame.to_csv(FROZEN_DIR / f"{name}.csv", index=index)


def file_record(path):
    """Relative path, size in bytes and SHA-256 of one file, for the manifest."""
    path = Path(path)
    return {"path": relative(path), "bytes": path.stat().st_size, "sha256": sha256_of_file(path)}


def usable_value(value):
    """True for a stored value that a check can use: not None and, for a
    float, finite."""
    return value is not None and not (isinstance(value, float) and not np.isfinite(value))


def json_ready(value):
    """A value converted for JSON: NumPy scalars to Python numbers, NaN and
    infinities to None, paths to project-relative text, containers
    element by element."""
    if isinstance(value, np.generic):
        value = value.item()
    if isinstance(value, float) and not np.isfinite(value):
        return None
    if isinstance(value, Path):
        return relative(value)
    if isinstance(value, dict):
        return {str(key): json_ready(item) for key, item in value.items()}
    if isinstance(value, (list, tuple, set)):
        return [json_ready(item) for item in value]
    return value


# --------------------------------------------------------------------------
# Number formatting for printed output
# --------------------------------------------------------------------------
_SUPERSCRIPT = str.maketrans("-0123456789", "⁻⁰¹²³⁴⁵⁶⁷⁸⁹")


def format_p(p, resolution=None):
    """A P value with two significant figures, and m × 10⁻ᵏ below 0.001.

    `resolution` is the smallest non-zero value a resampling P value can
    take; below it the result is a bound ("< 0.0005"). A P value that
    underflows to zero is printed as a bound as well. The returned text
    starts with "<" exactly when it is a bound, which p_label relies on.
    """
    if p is None or not np.isfinite(p):
        return "not computed"
    if resolution is not None and p < resolution:
        return f"< {resolution:.2g}"
    if p == 0:
        return "< 1 × 10⁻³⁰⁰"
    if p < 0.001:
        mantissa, exponent = f"{p:.1e}".split("e")
        return f"{mantissa} × 10{str(int(exponent)).translate(_SUPERSCRIPT)}"
    return f"{p:.2g}"


def format_number(value, style):
    """A stored value for the summary: style "int", "pct" (as a percentage
    with one decimal), "p" (a P value), "pb" (a bootstrap P value, shown as
    a bound below the resampling resolution 1 / N_BOOTSTRAP) or "fN" (N
    decimals). A value that was not computed is shown as such."""
    if not usable_value(value):
        return "not computed"
    if style == "int":
        return f"{int(value):,}"
    if style == "pct":
        return f"{100 * value:.1f}%"
    if style == "p":
        return format_p(value)
    if style == "pb":
        return format_p(value, 1 / N_BOOTSTRAP)
    return f"{value:.{int(style[1:])}f}"


def p_label(p, resolution=None):
    """'P = 0.026' or 'P < 0.0005': the P value with the right relation sign,
    so that a bound is never printed as 'P = < ...'."""
    text = format_p(p, resolution)
    if text == "not computed":
        return "P not computed"
    return f"P {text}" if text.startswith("<") else f"P = {text}"


def format_ci(estimate, low, high, scale=100.0, decimals=1, unit="%"):
    """'19.7% (95% CI 13.1 to 26.0)' style, with a true minus sign."""
    def number(value):
        return f"{scale * value:.{decimals}f}".replace("-", "−")
    return f"{number(estimate)}{unit} (95% CI {number(low)} to {number(high)})"


# --------------------------------------------------------------------------
# Multiple testing and resampling
# --------------------------------------------------------------------------
def benjamini_hochberg(p_values):
    """Benjamini–Hochberg q values. Missing P values pass through and do not
    count toward the number of tests."""
    p = np.asarray(p_values, float)
    finite = np.isfinite(p)
    adjusted = np.full_like(p, np.nan)
    if not finite.any():
        return adjusted
    order = np.argsort(p[finite])
    ranked = p[finite][order]
    m = len(ranked)
    stepped = np.minimum.accumulate((ranked * m / np.arange(1, m + 1))[::-1])[::-1]
    restored = np.empty(m)
    restored[order] = np.clip(stepped, 0, 1)
    adjusted[finite] = restored
    return adjusted


def holm(p_values, alpha=0.05):
    """Holm step-down. Returns the per-test threshold and whether the test is
    rejected; once one test fails, every larger P value also fails."""
    p = np.asarray(p_values, float)
    order = np.argsort(p)
    thresholds = np.empty(len(p))
    rejected = np.zeros(len(p), dtype=bool)
    still_rejecting = True
    for rank, index in enumerate(order):
        thresholds[index] = alpha / (len(p) - rank)
        if still_rejecting and p[index] <= thresholds[index]:
            rejected[index] = True
        else:
            still_rejecting = False
    return thresholds, rejected


def rows_by_cluster(labels):
    """Row positions for each cluster label, clusters in sorted order."""
    labels = np.asarray(labels)
    return {cluster: np.flatnonzero(labels == cluster) for cluster in np.unique(labels)}


def cluster_bootstrap(cluster_rows, statistic, n_bootstrap, rng):
    """Resample whole clusters with replacement and evaluate `statistic` on the
    concatenated rows. Non-finite replicates are dropped."""
    clusters = np.array(list(cluster_rows))
    draws = []
    for _ in range(n_bootstrap):
        drawn = rng.choice(clusters, size=len(clusters), replace=True)
        value = statistic(np.concatenate([cluster_rows[c] for c in drawn]))
        if np.isfinite(value):
            draws.append(value)
    return np.asarray(draws)


def pigeonhole_bootstrap(first_labels, second_labels, statistic, n_bootstrap, rng):
    """Two-way (crossed) cluster bootstrap: the pigeonhole bootstrap of Owen
    (2007, Ann. Appl. Stat. 1:386-411).

    The levels of two crossed factors (for example PDX models and treatments)
    are resampled independently with replacement, and each row enters the
    resample as many times as the product of the counts of its two levels.
    `statistic` receives the (repeated) row positions. For crossed designs
    the resulting variance is slightly conservative. Resamples with fewer
    than three rows and non-finite replicates are dropped.
    """
    first_codes, first_levels = pd.factorize(pd.Series(np.asarray(first_labels)).astype(str))
    second_codes, second_levels = pd.factorize(pd.Series(np.asarray(second_labels)).astype(str))
    positions = np.arange(len(first_codes))
    draws = []
    for _ in range(n_bootstrap):
        # How many times each level was drawn, for each factor separately.
        first_count = np.bincount(rng.integers(0, len(first_levels), len(first_levels)),
                                  minlength=len(first_levels))
        second_count = np.bincount(rng.integers(0, len(second_levels), len(second_levels)),
                                   minlength=len(second_levels))
        rows = np.repeat(positions, first_count[first_codes] * second_count[second_codes])
        if len(rows) < 3:
            continue
        value = statistic(rows)
        if np.isfinite(value):
            draws.append(value)
    return np.asarray(draws)


def bootstrap_median_interval(values, n_bootstrap, rng):
    """95% percentile interval of the median, resampling values (n >= 4)."""
    values = np.asarray(values, float)
    values = values[np.isfinite(values)]
    if len(values) < 4:
        return np.nan, np.nan
    medians = np.median(values[rng.integers(0, len(values), size=(n_bootstrap, len(values)))], axis=1)
    return float(np.percentile(medians, 2.5)), float(np.percentile(medians, 97.5))


def bootstrap_p_two_sided(draws):
    """Twice the smaller share of bootstrap estimates on either side of zero."""
    draws = np.asarray(draws, float)
    return float(min(1.0, 2.0 * min(np.mean(draws <= 0), np.mean(draws >= 0))))


def rmse(observed, predicted):
    return float(np.sqrt(np.mean((np.asarray(observed) - np.asarray(predicted)) ** 2)))


def relative_rmse_reduction(observed, baseline, challenger):
    """(RMSE of baseline − RMSE of challenger) / RMSE of baseline."""
    base = rmse(observed, baseline)
    return (base - rmse(observed, challenger)) / base if base > 0 else np.nan


def reduction_with_interval(observed, baseline, challenger, cluster_rows, n_bootstrap, rng):
    """Relative RMSE reduction with a 95% percentile interval from resampling
    whole clusters (PDX models), predictions held fixed. Returns the
    estimate, the interval bounds and the bootstrap draws."""
    observed, baseline, challenger = (np.asarray(observed, float), np.asarray(baseline, float),
                                      np.asarray(challenger, float))
    draws = cluster_bootstrap(
        cluster_rows,
        lambda rows: relative_rmse_reduction(observed[rows], baseline[rows], challenger[rows]),
        n_bootstrap, rng)
    low, high = np.percentile(draws, [2.5, 97.5]) if len(draws) else (np.nan, np.nan)
    return relative_rmse_reduction(observed, baseline, challenger), float(low), float(high), draws


def wilson_interval(successes, trials, z=1.96):
    """Wilson score interval for a binomial proportion."""
    if trials == 0:
        return np.nan, np.nan
    share = successes / trials
    centre = (share + z * z / (2 * trials)) / (1 + z * z / trials)
    half = z * np.sqrt(share * (1 - share) / trials + z * z / (4 * trials ** 2)) / (1 + z * z / trials)
    return centre - half, centre + half


def spearman(x, y):
    """Spearman correlation and P value, or NaN when either input is constant."""
    x, y = np.asarray(x, float), np.asarray(y, float)
    if len(x) < 3 or np.ptp(x) == 0 or np.ptp(y) == 0:
        return np.nan, np.nan
    result = stats.spearmanr(x, y)
    return float(result.statistic), float(result.pvalue)


def correlation_row(analysis, subset, x, y):
    """One row of a correlation table: Spearman ρ, its P value and n."""
    value, p_value = spearman(x, y)
    return {"analysis": analysis, "subset": subset, "n_arms": len(x),
            "spearman_rho": value, "p_value": p_value}


def clustered_correlation_row(analysis, subset, x, y, clusters, n_bootstrap, rng):
    """correlation_row plus a 95% interval and a two-sided P value from
    resampling whole clusters (PDX models), so that arms of one PDX model are
    not counted as independent.

    The P value in `p_value` treats arms as independent and is kept for
    comparison; `p_cluster` is the bootstrap P value, whose resolution is
    1 / n_bootstrap (zero means below it). Rows with fewer than ten arms or
    fewer than five clusters get no interval.
    """
    row = correlation_row(analysis, subset, x, y)
    x, y = np.asarray(x, float), np.asarray(y, float)
    clusters = np.asarray(clusters).astype(str)
    row.update(n_clusters=len(np.unique(clusters)), rho_low=np.nan, rho_high=np.nan, p_cluster=np.nan)
    if len(x) >= 10 and row["n_clusters"] >= 5:
        draws = cluster_bootstrap(rows_by_cluster(clusters),
                                  lambda rows: spearman(x[rows], y[rows])[0], n_bootstrap, rng)
        if len(draws):
            row.update(rho_low=float(np.percentile(draws, 2.5)), rho_high=float(np.percentile(draws, 97.5)),
                       p_cluster=bootstrap_p_two_sided(draws))
    return row


def partial_rank_correlation(x, y, z):
    """Spearman partial correlation of x and y given z: the Pearson
    correlation of the rank residuals left after regressing the ranks of x
    and of y on the rank of z (with an intercept). Ties take average ranks,
    as pandas .rank() does, so the value equals the 4.3 computation of the
    analysis first run."""
    ranks = [stats.rankdata(np.asarray(values, float)) for values in (x, y, z)]
    design = np.column_stack([np.ones(len(ranks[2])), ranks[2]])
    residuals = [rank - design @ np.linalg.lstsq(design, rank, rcond=None)[0] for rank in ranks[:2]]
    return float(np.corrcoef(*residuals)[0, 1])


def slope_difference_exceedance(sxx_a, sxx_b, sigma, gap, between_sd=0.0):
    """Probability that two least-squares slopes with the SAME true value
    differ by at least `gap` in absolute value through measurement noise
    alone. Each slope's error is Normal(0, σ²/Sxx) under independent
    Normal(0, σ²) measurement errors, so the difference is
    Normal(0, σ²(1/Sxx_a + 1/Sxx_b)).

    With `between_sd` = τ > 0, each arm's slope also carries between-animal
    variation of SD τ (the two arms are different animals), independent of
    the measurement error, so the difference is
    Normal(0, σ²(1/Sxx_a + 1/Sxx_b) + 2τ²) (Section 13.5). With τ = 0 the
    result is exactly the noise-only probability of Section 4.5."""
    sd = np.sqrt(sigma ** 2 * (1.0 / np.asarray(sxx_a, float) + 1.0 / np.asarray(sxx_b, float))
                 + 2.0 * float(between_sd) ** 2)
    return 2.0 * stats.norm.sf(gap / sd)


# --------------------------------------------------------------------------
# Regression fits shared by several sections
# --------------------------------------------------------------------------
MIXED_MODEL_OPTIMIZERS = ["lbfgs", "bfgs", "powell", "nm"]


def fit_mixed_model(formula, data, **model_arguments):
    """REML fit of a statsmodels linear mixed model that tries several
    optimizers in turn and stops at the first that converges.

    statsmodels moves to the next optimizer in the list only when the
    previous one fails, so a model that converges with L-BFGS gives the same
    estimates as a fit with L-BFGS alone. Returns the fit, whether it
    converged, and the distinct warning messages raised while fitting.
    """
    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter("always")
        fitted = smf.mixedlm(formula, data, **model_arguments).fit(
            reml=True, method=MIXED_MODEL_OPTIMIZERS)
    messages = sorted({str(w.message) for w in caught})
    return fitted, bool(getattr(fitted, "converged", False)), messages


def unpenalized_logistic():
    """Logistic regression without a penalty. C = ∞ is the form scikit-learn
    1.8 and later accept for an unpenalized fit (penalty=None is deprecated);
    the estimates are the maximum-likelihood estimates either way."""
    return LogisticRegression(C=np.inf, max_iter=5000)



# --------------------------------------------------------------------------
# The canonical observation table and per-arm summaries
# --------------------------------------------------------------------------
def canonical_observations(raw):
    """Apply the two preprocessing rules once, for every later analysis.

    Volumes below VOLUME_FLOOR (zero or recorded below 1 mm³) are set to the
    floor, and replicate readings on the same day of the same arm are averaged
    on the log scale (geometric mean), giving one row per arm per day.
    """
    frame = raw.copy()
    frame["volume"] = np.maximum(frame.volume.to_numpy(float), VOLUME_FLOOR)
    frame["log_volume"] = np.log(frame.volume)
    collapsed = (frame.groupby(["arm_id", "day"], as_index=False, sort=True)
                 .agg(model_id=("model_id", "first"),
                      tumor_type=("tumor_type", "first"),
                      treatment=("treatment", "first"),
                      is_control=("is_control", "first"),
                      log_volume=("log_volume", "mean"),
                      n_readings=("log_volume", "size")))
    collapsed["volume"] = np.exp(collapsed.log_volume)
    return collapsed


def arm_series(canonical, arm_ids=None):
    """Yield (arm_id, block) for each arm of the canonical table, sorted by day."""
    frame = canonical if arm_ids is None else canonical.loc[canonical.arm_id.isin(arm_ids)]
    for arm_id, block in frame.groupby("arm_id", sort=True):
        yield arm_id, block


def fit_line(day, log_volume):
    """Least-squares line on log volume: slope, intercept and residual SD with
    the two fitted parameters removed from the degrees of freedom."""
    day = np.asarray(day, float)
    log_volume = np.asarray(log_volume, float)
    slope, intercept = np.polyfit(day, log_volume, 1)
    residual = log_volume - (slope * day + intercept)
    residual_sd = (float(np.sqrt(np.sum(residual ** 2) / (len(day) - 2)))
                   if len(day) > 2 else np.nan)
    return float(slope), float(intercept), residual_sd


# Columns that post_nadir_fit returns, in the order they are stored per arm.
DURABILITY_COLUMNS = ("nadir_index", "nadir_day", "n_after_nadir", "post_nadir_growth_rate",
                      "post_nadir_intercept", "post_nadir_first_day", "post_nadir_last_day",
                      "post_nadir_points", "post_nadir_sxx")


def post_nadir_fit(day, log_volume, include_nadir=None):
    """The durability endpoint of one arm: the least-squares slope of log
    volume after the nadir. Every durability value in the notebook (the
    endpoint table of Section 2.3, the regrowth summaries, the windowed and
    negative-control analyses) comes from this one function.

    The nadir is the first occurrence of the smallest log volume. With
    include_nadir False (DURABILITY_INCLUDES_NADIR, the default) the line is
    fitted to the measurements after the nadir; with True it is fitted from
    the nadir on. The slope is defined when at least MIN_POINTS_AFTER_NADIR
    measurements follow the nadir, so with the nadir left out and exactly two
    later measurements it is a two-point slope.

    Returns the nadir index and day, the number of measurements after the
    nadir, the slope (per day), the fitted log volume at the nadir day
    (intercept: a figure draws exactly the fitted line,
    log V = intercept + slope × (t − nadir day), between the first and last
    fitted days), the first and last fitted days, the number of fitted
    points, and Sxx = Σ(t − mean t)² over the fitted days, which sets the
    slope's standard error σ / √Sxx.
    """
    day = np.asarray(day, float)
    log_volume = np.asarray(log_volume, float)
    include_nadir = DURABILITY_INCLUDES_NADIR if include_nadir is None else bool(include_nadir)
    nadir_index = int(np.argmin(log_volume))
    n_after_nadir = len(day) - 1 - nadir_index
    fit = {"nadir_index": nadir_index, "nadir_day": float(day[nadir_index]),
           "n_after_nadir": n_after_nadir, "post_nadir_growth_rate": np.nan,
           "post_nadir_intercept": np.nan, "post_nadir_first_day": np.nan,
           "post_nadir_last_day": np.nan, "post_nadir_points": 0, "post_nadir_sxx": np.nan}
    if n_after_nadir >= MIN_POINTS_AFTER_NADIR:
        start = nadir_index if include_nadir else nadir_index + 1
        # Time is measured from the nadir day, so the intercept is the fitted
        # log volume at the nadir whether or not the nadir is in the fit.
        elapsed = day[start:] - day[nadir_index]
        slope, intercept = np.polyfit(elapsed, log_volume[start:], 1)
        fit.update(post_nadir_growth_rate=float(slope), post_nadir_intercept=float(intercept),
                   post_nadir_first_day=float(day[start]), post_nadir_last_day=float(day[-1]),
                   post_nadir_points=int(len(elapsed)),
                   post_nadir_sxx=float(np.sum((elapsed - elapsed.mean()) ** 2)))
    return fit


def floor_plateau_fit(day, log_volume, floor_log=None):
    """Regrowth of an arm whose nadir sits at the volume floor, measured from
    the end of its stay at the floor (Section 4.6).

    An arm held at the floor (1 mm³, a complete response by caliper) for
    several readings and then relapsing gets a small post-nadir slope from
    post_nadir_fit, because the line averages the flat readings with the
    relapse. Here the plateau is the run of consecutive readings at the floor
    that starts at the nadir (the first occurrence of the minimum, as in
    post_nadir_fit). The slope is fitted to the readings AFTER the plateau's
    last reading (at least MIN_POINTS_AFTER_NADIR of them), with that last
    floor reading left out, as post_nadir_fit leaves out the nadir, and time
    measured from it. Returns whether the nadir is at the floor, the plateau
    length (readings and days), the number of readings after it, and that
    slope (NaN when the nadir is above the floor, the plateau has a single
    reading, or too few readings follow it).
    """
    day = np.asarray(day, float)
    log_volume = np.asarray(log_volume, float)
    floor_log = np.log(VOLUME_FLOOR) if floor_log is None else floor_log
    nadir_index = int(np.argmin(log_volume))
    at_floor = np.isclose(log_volume, floor_log, atol=1e-9)
    result = {"nadir_at_floor": bool(at_floor[nadir_index]), "plateau_readings": 0, "plateau_days": 0.0,
              "readings_after_plateau": 0, "post_plateau_growth_rate": np.nan}
    if not at_floor[nadir_index]:
        return result
    # Walk forward through consecutive readings at the floor.
    end = nadir_index
    while end + 1 < len(day) and at_floor[end + 1]:
        end += 1
    after = len(day) - 1 - end
    result.update(plateau_readings=end - nadir_index + 1, plateau_days=float(day[end] - day[nadir_index]),
                  readings_after_plateau=after)
    if end > nadir_index and after >= MIN_POINTS_AFTER_NADIR:
        result["post_plateau_growth_rate"] = float(np.polyfit(day[end + 1:] - day[end], log_volume[end + 1:], 1)[0])
    return result


def summarize_regrowth(day, log_volume):
    """Model-free regrowth summary of one arm's series (sorted by day).

    The nadir, the number of measurements after it and the post-nadir growth
    rate come from post_nadir_fit, so they are the durability endpoint
    itself. (In the analysis first run this function fitted the slope from
    the nadir on while the endpoint table left the nadir out, so Sections
    3.3, 4.3 and 8.3 used a second definition.)
    """
    day = np.asarray(day, float)
    log_volume = np.asarray(log_volume, float)
    if len(day) < 3:
        return None
    fit = post_nadir_fit(day, log_volume)
    nadir_index = fit["nadir_index"]
    return {
        "n_measurements": len(day),
        "nadir_index": nadir_index,
        "n_after_nadir": fit["n_after_nadir"],
        "nadir_ratio": float(np.exp(log_volume[nadir_index] - log_volume[0])),
        "regrowth_multiple": float(np.exp(log_volume[-1] - log_volume[nadir_index])),
        "post_nadir_growth_rate": fit["post_nadir_growth_rate"],
    }


def durability_phenotype(summary):
    """The durability phenotype applied to treated arms: shrank by at least
    REGRESSION_FRACTION, an interior nadir followed by at least two points,
    and regrowth to at least REGROWTH_MULTIPLE times the nadir."""
    return (summary["nadir_ratio"] <= 1 - REGRESSION_FRACTION
            and summary["nadir_index"] > 0
            and summary["n_after_nadir"] >= MIN_POINTS_AFTER_NADIR
            and summary["regrowth_multiple"] >= REGROWTH_MULTIPLE)


def series_frames(frame, id_column, day_column, value_column):
    """{series id: (days, values)} for each series of a long table, by day."""
    return {series: (block[day_column].to_numpy(float), block[value_column].to_numpy(float))
            for series, block in frame.sort_values(day_column).groupby(id_column, sort=True)}


def interpolation_noise(frames, span_range=None):
    """Measurement-error SD from interpolation residuals, no growth model assumed.

    `frames` maps a series id to (days, values) sorted by day. For consecutive
    readings (t0, y0), (t1, y1), (t2, y2) the middle one is predicted by
    linear interpolation between its neighbors. With locally linear growth
    the residual r has variance σ²(1 + w0² + w2²), so each triple contributes
    r²/(1 + w0² + w2²) as an estimate of σ². Equal spacing reduces to the
    second-difference estimator. With `span_range` = (low, high), only
    triples with low ≤ t2 − t0 ≤ high are used. Returns per-series arrays so
    that resampling can be clustered on series.
    """
    per_series = {}
    for series_id, (t, y) in frames.items():
        estimates = []
        for i in range(1, len(y) - 1):
            span = t[i + 1] - t[i - 1]
            if span <= 0:
                continue
            if span_range is not None and not (span_range[0] <= span <= span_range[1]):
                continue
            w0 = (t[i + 1] - t[i]) / span
            w2 = (t[i] - t[i - 1]) / span
            residual = y[i] - (w0 * y[i - 1] + w2 * y[i + 1])
            estimates.append(residual ** 2 / (1.0 + w0 ** 2 + w2 ** 2))
        if estimates:
            per_series[series_id] = np.asarray(estimates)
    return per_series


def summarize_interpolation_noise(per_series, rng, n_bootstrap=N_BOOTSTRAP):
    """Pooled σ with a bootstrap interval that resamples whole series."""
    if not per_series:
        return {"sigma": np.nan, "low": np.nan, "high": np.nan,
                "n_triples": 0, "n_series": 0}
    # The pooled estimate is sqrt(sum of contributions / number of triples),
    # so a resample only needs each series' sum and count.
    sums = np.array([values.sum() for values in per_series.values()])
    counts = np.array([len(values) for values in per_series.values()])
    sigma = float(np.sqrt(sums.sum() / counts.sum()))
    draws = np.empty(n_bootstrap)
    for b in range(n_bootstrap):
        picked = rng.integers(0, len(sums), size=len(sums))
        draws[b] = np.sqrt(sums[picked].sum() / counts[picked].sum())
    low, high = np.percentile(draws, [2.5, 97.5])
    return {"sigma": sigma, "low": float(low), "high": float(high),
            "n_triples": int(counts.sum()), "n_series": len(sums)}


# --------------------------------------------------------------------------
# Mechanistic models, on the log-volume scale
# --------------------------------------------------------------------------
def log_volume_exponential(t, log_baseline, growth_rate):
    """Untreated growth: log V(t) = log V0 + g t."""
    return log_baseline + growth_rate * np.asarray(t, float)


def log_volume_constant_kill(t, log_baseline, growth_rate, kill_rate):
    """Treatment lowers the growth rate by a fixed kill rate: dV/dt = (g − k)V."""
    return log_baseline + (growth_rate - kill_rate) * np.asarray(t, float)


def log_volume_tolerant(t, log_baseline, growth_rate, kill_rate, tolerant_fraction):
    """Tolerant subpopulation: V(t) = V0[(1 − f)e^((g − k)t) + f e^(gt)].

    Evaluated with logaddexp, so neither exponential is formed, and with
    log1p(−f), which keeps precision when f is small.
    """
    t = np.asarray(t, float)
    fraction = np.clip(tolerant_fraction, 1e-12, 1.0 - 1e-12)
    sensitive = np.log1p(-fraction) + (growth_rate - kill_rate) * t
    tolerant = np.log(fraction) + growth_rate * t
    return log_baseline + np.logaddexp(sensitive, tolerant)


def log_volume_declining(t, log_baseline, growth_rate, initial_kill_rate, decay_rate):
    """Declining drug effect: dV/dt = [g − k e^(−δt)]V, integrated in closed form.
    As δ → 0 the model is the constant-kill model, handled explicitly."""
    t = np.asarray(t, float)
    if decay_rate < 1e-10:
        return log_baseline + (growth_rate - initial_kill_rate) * t
    integrated_kill = (initial_kill_rate / decay_rate) * (-np.expm1(-decay_rate * t))
    return log_baseline + growth_rate * t - integrated_kill


def sensitivity_matrix(model, days, parameters, relative_step=1e-5):
    """Central-difference derivatives of predicted log volume (rows: days)
    with respect to each parameter (columns)."""
    parameters = np.asarray(parameters, float)
    columns = []
    for index in range(len(parameters)):
        step = relative_step * max(1.0, abs(parameters[index]))
        forward, backward = parameters.copy(), parameters.copy()
        forward[index] += step
        backward[index] -= step
        columns.append((model(days, *forward) - model(days, *backward)) / (2.0 * step))
    return np.column_stack(columns)


MECHANISMS = ("tolerant subpopulation", "declining drug effect")

# Fitting bounds on the unconstrained scale (log for rates and volume, logit
# for the fraction). A fit that ends on a bound is a degenerate limit and is
# flagged, never used silently.
MECHANISM_BOUNDS = {
    "tolerant subpopulation": [("log_baseline", 0.0, 11.5), ("log_growth", -6.9, 0.0),
                               ("log_kill", -6.9, 0.69), ("logit_fraction", -12.0, 4.0)],
    "declining drug effect": [("log_baseline", 0.0, 11.5), ("log_growth", -6.9, 0.0),
                              ("log_kill", -6.9, 0.69), ("log_decay", -11.5, 0.69)],
}


def constant_exposure_forward(mechanism):
    """Forward model on the unconstrained scale, under constant exposure."""
    def forward(theta, days):
        log_v0, log_growth, log_kill, fourth = theta
        if mechanism == "tolerant subpopulation":
            return log_volume_tolerant(days, log_v0, np.exp(log_growth), np.exp(log_kill),
                                       1.0 / (1.0 + np.exp(-fourth)))
        return log_volume_declining(days, log_v0, np.exp(log_growth), np.exp(log_kill),
                                    float(np.exp(fourth)))
    return forward


def mechanism_starts(observed):
    """Three starting points shared by every mechanistic fit."""
    return [np.array([observed[0], np.log(0.05), np.log(0.08), -2.0]),
            np.array([observed[0], np.log(0.02), np.log(0.15), -1.0]),
            np.array([observed[0], np.log(0.09), np.log(0.04), -3.0])]


def fit_bounded(forward, bounds, days, observed, starts, max_nfev=2000,
                pin_tolerance=1e-6):
    """Bounded multi-start least squares. Returns the best fit as a dict with
    the residual sum of squares, RMSE, Akaike criterion and pinned parameters."""
    names = [b[0] for b in bounds]
    low = np.array([b[1] for b in bounds])
    high = np.array([b[2] for b in bounds])
    margin = 1e-3 * (high - low)
    best = None
    for start in starts:
        start = np.clip(start, low + margin, high - margin)
        try:
            solution = least_squares(lambda theta: forward(theta, days) - observed,
                                     start, bounds=(low, high), max_nfev=max_nfev)
        except (ValueError, np.linalg.LinAlgError):
            continue
        rss = float(np.sum(solution.fun ** 2))
        if np.isfinite(rss) and (best is None or rss < best["rss"]):
            best = {"rss": rss, "x": solution.x}
    if best is None:
        return {"rss": np.inf, "rmse": np.inf, "criterion": np.inf,
                "n_pinned": len(names), "pinned": "all starts failed",
                "x": np.full(len(names), np.nan), "names": names}
    span = np.maximum(high - low, 1e-12)
    at_low = (best["x"] - low) / span < pin_tolerance
    at_high = (high - best["x"]) / span < pin_tolerance
    pinned = [f"{name}@{'low' if lo else 'high'}"
              for name, lo, hi in zip(names, at_low, at_high) if lo or hi]
    n = len(observed)
    return {"rss": best["rss"], "rmse": float(np.sqrt(best["rss"] / n)),
            "criterion": float(n * np.log(max(best["rss"], 1e-12) / n) + 2 * len(names)),
            "n_pinned": len(pinned), "pinned": ";".join(pinned),
            "x": best["x"], "names": names}


def fit_mechanism(mechanism, days, observed):
    """Fit one regrowth mechanism under constant exposure."""
    return fit_bounded(constant_exposure_forward(mechanism), MECHANISM_BOUNDS[mechanism],
                       days, observed, mechanism_starts(observed))


def tolerant_estimates(theta):
    """Growth rate, kill rate and tolerant fraction from a fitted parameter vector."""
    return {"growth_rate": float(np.exp(theta[1])), "kill_rate": float(np.exp(theta[2])),
            "tolerant_fraction": float(1.0 / (1.0 + np.exp(-theta[3])))}


def recovery_summary(log_errors, log_truths):
    """Median absolute log error, bias, error-to-prior ratio (RMSE of the log
    error over the SD of the log true values) and whether the prespecified
    recovery threshold is met."""
    log_errors, log_truths = np.asarray(log_errors), np.asarray(log_truths)
    median_error = float(np.median(np.abs(log_errors)))
    return {"replicates": len(log_errors), "median_abs_log_error": median_error,
            "bias_log": float(np.mean(log_errors)),
            "error_to_prior_ratio": float(np.sqrt(np.mean(log_errors ** 2)) / np.std(log_truths)),
            "recovered": median_error < RECOVERY_THRESHOLD}


def aic_verdict(fits, tie_threshold=None):
    """Which mechanism the Akaike criterion prefers. With `tie_threshold`, a
    difference smaller than it is recorded as no decision."""
    delta = fits["declining drug effect"]["criterion"] - fits["tolerant subpopulation"]["criterion"]
    if not np.isfinite(delta):
        return "undecidable", delta
    if tie_threshold is not None and abs(delta) < tie_threshold:
        return "tie", delta
    return ("tolerant subpopulation" if delta > 0 else "declining drug effect"), delta


def apply_termination(days, log_volume, threshold=TERMINATION_VOLUME):
    """End a simulated arm at its first measurement above the threshold volume,
    keeping that measurement, as the source study removed large tumors."""
    above = np.flatnonzero(log_volume > np.log(threshold))
    if len(above):
        cut = above[0] + 1
        return days[:cut], log_volume[:cut], True
    return days, log_volume, False


def sample_truncated_normal(rng, centre, spread, upper):
    """One draw from Normal(centre, spread) truncated above at `upper`."""
    while True:
        value = rng.normal(centre, spread)
        if value <= upper:
            return value


def draw_prior_parameters(rng, prior):
    """One parameter set of the tolerant-subpopulation model from the priors of
    the Bayesian model in Section 5.6 (prior dictionary built there)."""
    return {
        "log_baseline": rng.normal(np.log(200.0), 0.5),
        "growth_rate": float(np.exp(sample_truncated_normal(
            rng, prior["growth_centre"], prior["growth_spread"],
            np.log(prior["growth_upper"])))),
        "kill_rate": float(abs(rng.normal(0.0, 0.10))),
        "tolerant_fraction": float(1.0 / (1.0 + np.exp(-rng.normal(-2.2, 1.5)))),
    }


def tolerant_subpopulation_model(days, prior, log_volume=None):
    """The tolerant-subpopulation model as a probabilistic program.

    Priors: log baseline volume Normal(log 200, 0.5); log growth rate
    TruncatedNormal(centre, spread, upper) from the untreated arms (Section
    5.6); kill rate HalfNormal(0.10); logit tolerant fraction Normal(−2.2,
    1.5); measurement SD HalfNormal(0.30). The likelihood is on log volume.
    """
    log_baseline = numpyro.sample("log_baseline", dist.Normal(np.log(200.0), 0.5))
    log_growth = numpyro.sample("log_growth_rate", dist.TruncatedNormal(
        loc=prior["growth_centre"], scale=prior["growth_spread"],
        high=np.log(prior["growth_upper"])))
    growth_rate = numpyro.deterministic("growth_rate", jax.numpy.exp(log_growth))
    kill_rate = numpyro.sample("kill_rate", dist.HalfNormal(0.10))
    fraction_logit = numpyro.sample("fraction_logit", dist.Normal(-2.2, 1.5))
    fraction = numpyro.deterministic("tolerant_fraction", jax.nn.sigmoid(fraction_logit))
    noise_scale = numpyro.sample("noise_scale", dist.HalfNormal(0.30))
    sensitive = jax.numpy.log1p(-fraction) + (growth_rate - kill_rate) * days
    tolerant = jax.numpy.log(fraction) + growth_rate * days
    predicted = log_baseline + jax.numpy.logaddexp(sensitive, tolerant)
    numpyro.sample("observed", dist.Normal(predicted, noise_scale), obs=log_volume)


# --------------------------------------------------------------------------
# Drug exposure (Section 12): one-compartment kinetics, analytic integrals
# --------------------------------------------------------------------------
def dosing_days(interval_days, duration_days):
    """Dosing days at a fixed interval (1.0 = once daily) up to duration_days."""
    return np.arange(0.0, duration_days + 1e-9, interval_days)


def cumulative_exposure(t, dose_days, half_life_days, normalize_over=None):
    """∫₀ᵗ q(s) ds for superposed first-order doses with instantaneous
    absorption. With `normalize_over` = T, the result is scaled so that mean
    exposure over [0, T] is 1, which keeps kill rates comparable with the
    constant-exposure models. T is the last day of the grid being modeled, so
    the scaling (a constant factor absorbed by the kill rate) depends on it."""
    t = np.atleast_1d(np.asarray(t, float))
    decay = np.log(2.0) / half_life_days
    elapsed = np.maximum(t[:, None] - dose_days[None, :], 0.0)
    active = t[:, None] >= dose_days[None, :]
    total = np.where(active, (1.0 - np.exp(-decay * elapsed)) / decay, 0.0).sum(axis=1)
    if normalize_over is not None and normalize_over > 0:
        reference = cumulative_exposure(np.array([normalize_over]), dose_days,
                                        half_life_days)[0]
        if reference > 0:
            total = total * (normalize_over / reference)
    return total


def log_volume_tolerant_exposure(t, log_baseline, growth_rate, kill_rate,
                                 tolerant_fraction, dose_days, half_life_days):
    """Tolerant subpopulation with the kill term driven by drug exposure:
    sensitive cells dV/dt = (g − k q(t))V; tolerant cells dV/dt = gV."""
    t = np.asarray(t, float)
    fraction = np.clip(tolerant_fraction, 1e-12, 1.0 - 1e-12)
    integrated = cumulative_exposure(t, dose_days, half_life_days,
                                     normalize_over=float(np.max(t)) or 1.0)
    sensitive = np.log1p(-fraction) + growth_rate * t - kill_rate * integrated
    tolerant = np.log(fraction) + growth_rate * t
    return log_baseline + np.logaddexp(sensitive, tolerant)


def log_volume_declining_exposure(t, log_baseline, growth_rate, kill_rate,
                                  biological_decay, dose_days, half_life_days):
    """Declining effect with clearance separated from biology:
    dV/dt = (g − k q(t) e^(−δt))V. The integral stays analytic, and it is
    scaled by the same exposure-only factor as the tolerant model."""
    t = np.asarray(t, float)
    pk_decay = np.log(2.0) / half_life_days
    elapsed = np.maximum(t[:, None] - dose_days[None, :], 0.0)
    active = t[:, None] >= dose_days[None, :]
    combined = pk_decay + biological_decay
    if combined < 1e-12:
        per_dose = np.where(active, elapsed, 0.0)
    else:
        per_dose = np.where(active, np.exp(-biological_decay * dose_days[None, :])
                            * (1.0 - np.exp(-combined * elapsed)) / combined, 0.0)
    integrated = per_dose.sum(axis=1)
    reference_days = float(np.max(t)) or 1.0
    reference = cumulative_exposure(np.array([reference_days]), dose_days, half_life_days)[0]
    if reference > 0:
        integrated = integrated * (reference_days / reference)
    return log_baseline + growth_rate * t - kill_rate * integrated


def exposure_forward(mechanism, dose_days, half_life_days):
    """Forward model on the unconstrained scale, with explicit exposure."""
    def forward(theta, days):
        log_v0, log_growth, log_kill, fourth = theta
        if mechanism == "tolerant subpopulation":
            return log_volume_tolerant_exposure(
                days, log_v0, np.exp(log_growth), np.exp(log_kill),
                1.0 / (1.0 + np.exp(-fourth)), dose_days, half_life_days)
        return log_volume_declining_exposure(
            days, log_v0, np.exp(log_growth), np.exp(log_kill), float(np.exp(fourth)),
            dose_days, half_life_days)
    return forward


def fit_mechanism_exposure(mechanism, days, observed, dose_days, half_life_days):
    """Fit one regrowth mechanism with explicit exposure."""
    return fit_bounded(exposure_forward(mechanism, dose_days, half_life_days),
                       MECHANISM_BOUNDS[mechanism], days, observed,
                       mechanism_starts(observed))


def draw_schedule(rng, pool, window_days, keep_every=1):
    """One measurement schedule drawn from `pool`, cut at `window_days` and,
    with keep_every = k, thinned to every k-th measurement. Returns None when
    fewer than five measurements remain (no further random draws are made)."""
    grid = pool[rng.integers(len(pool))]
    days = grid[grid <= window_days][::keep_every]
    return days if len(days) >= 5 else None


def typical_schedule(pool, window_days):
    """The schedule, cut at `window_days`, whose number of measurements is
    closest to the median of the pool (the first such schedule in pool order).
    Figures use it so that they show a typical design, not the densest one."""
    cut = [grid[grid <= window_days] for grid in pool]
    counts = np.array([len(days) for days in cut], float)
    return cut[int(np.argmin(np.abs(counts - np.median(counts))))]


def paired_exposure_trajectories(rng, days, dose_days, half_life_days,
                                 kill_centre=0.13, kill_spread=0.04):
    """Both mechanisms from one shared draw of baseline volume, growth rate,
    kill rate and noise realization, so that only the mechanism differs.

    log V0 ~ Normal(log 200, 0.3); growth rate ~ LogNormal(log 0.05, 0.5);
    kill rate ~ |Normal(kill_centre, kill_spread)|, by default
    |Normal(0.13, 0.04)| (strong kill; Section 12.5 also uses weaker kill);
    noise ~ Normal(0, 1) per measurement (scaled by the noise SD when
    scored). Mechanism-specific: logit tolerant fraction ~ Normal(−2.5, 1.0);
    biological decay ~ LogNormal(log 0.04, 0.5). Every result of Section 12
    is conditional on these distributions. The draw order is fixed, so a
    seed reproduces the same arms."""
    log_v0 = np.log(200.0) + rng.normal(0, 0.3)
    growth = float(np.exp(rng.normal(np.log(0.05), 0.5)))
    kill = float(abs(rng.normal(kill_centre, kill_spread)))
    noise = rng.normal(0, 1.0, size=len(days))
    fraction = float(1.0 / (1.0 + np.exp(-rng.normal(-2.5, 1.0))))
    decay = float(np.exp(rng.normal(np.log(0.04), 0.5)))
    clean = {
        "tolerant subpopulation": log_volume_tolerant_exposure(
            days, log_v0, growth, kill, fraction, dose_days, half_life_days),
        "declining drug effect": log_volume_declining_exposure(
            days, log_v0, growth, kill, decay, dose_days, half_life_days),
    }
    return clean, noise


def score_exposure_pair(clean, noise, noise_sd, days, dose_days, half_life_days, tie_threshold):
    """Fit both mechanisms to each simulated arm and record the AIC verdict,
    once with the full schedule and once with the arm ended at its first
    measurement above TERMINATION_VOLUME (`termination` = "not applied" or
    "applied"). An arm left with fewer than five measurements is undecidable.
    When the rule does not shorten the arm, the full-schedule fits are reused.
    """
    rows = []
    for truth in MECHANISMS:
        observed = clean[truth] + noise_sd * noise
        cut_days, cut_observed, stopped = apply_termination(days, observed)
        results_by_length = {}
        for rule, (arm_days, arm_observed) in (("not applied", (days, observed)),
                                               ("applied", (cut_days, cut_observed))):
            if len(arm_days) not in results_by_length:
                if len(arm_days) < 5:
                    result = {"verdict": "undecidable", "criterion_delta": np.nan,
                              "rmse_right": np.nan, "rmse_wrong": np.nan, "any_pinned": False}
                else:
                    fits = {m: fit_mechanism_exposure(m, arm_days, arm_observed, dose_days,
                                                      half_life_days) for m in MECHANISMS}
                    verdict, delta = aic_verdict(fits, tie_threshold)
                    wrong = [m for m in MECHANISMS if m != truth][0]
                    result = {"verdict": verdict, "criterion_delta": delta,
                              "rmse_right": fits[truth]["rmse"], "rmse_wrong": fits[wrong]["rmse"],
                              "any_pinned": (fits[truth]["n_pinned"] + fits[wrong]["n_pinned"]) > 0}
                results_by_length[len(arm_days)] = result
            result = results_by_length[len(arm_days)]
            rows.append({"termination": rule, "true_mechanism": truth,
                         "correct": result["verdict"] == truth,
                         "terminated": stopped and rule == "applied",
                         "n_measurements": len(arm_days), **result})
    return rows


def summarize_verdicts(block):
    """Decisive rate and accuracy among decisive calls, each with counts and a
    Wilson 95% interval, and the share choosing the tolerant-subpopulation
    model. The decisive-rate interval treats the arms of a cell as
    independent; the two arms of one paired replicate share their parameters,
    so it is somewhat narrow."""
    decisive = block.verdict.isin(MECHANISMS)
    n_decisive = int(decisive.sum())
    n_correct = int(block.loc[decisive, "correct"].sum())
    low, high = wilson_interval(n_correct, n_decisive)
    decisive_low, decisive_high = wilson_interval(n_decisive, len(block))
    return pd.Series({
        "n": len(block),
        "median_measurements": float(block.n_measurements.median()),
        "decisive_rate": float(decisive.mean()),
        "decisive_low": decisive_low, "decisive_high": decisive_high,
        "decisive_calls": n_decisive,
        "correct_calls": n_correct,
        "accuracy_when_decisive": n_correct / n_decisive if n_decisive else np.nan,
        "accuracy_low": low, "accuracy_high": high,
        "chose_tolerant": float((block.verdict == "tolerant subpopulation").mean()),
        "terminated_share": float(block.terminated.mean()),
    })
```

```python
# ==========================================================================
# 1.3 Functions: forecasting features and learners, within-model ranking,
#     molecular scores and grouped ridge regression (with an optional
#     treatment adjustment), comparisons of combinations with their
#     components, mechanism-class resampling, and the clinical tumor-burden
#     endpoints
# ==========================================================================

# --------------------------------------------------------------------------
# Forecasting (Sections 7 and 8)
# --------------------------------------------------------------------------
def window_response(window_day, window_volume, origin_day, rule=None):
    """Best average response from the readings of one window (on or before
    `origin_day`), under the window rule.

    Best average response is defined from day MIN_DAY_FOR_BEST_AVERAGE_RESPONSE
    (10) onward, so for a day-10 origin the only eligible reading is one taken
    on day 10, and the value is the running average of percentage change at
    that reading (a single value, not a best over several).

    rule "exact_day" (FORECAST_WINDOW_RESPONSE default, the analysis first
         run): that value, or NaN when no reading falls in
         [day 10, origin].
    rule "last_reading": the same value when it exists; otherwise the
         running average at the last reading on or before the origin,
         provided that reading is no more than
         WINDOW_RESPONSE_TOLERANCE_DAYS before the origin. It is computed by
         pdxe_utils.best_average_response with min_day set to that reading's
         day, so the running average is the module's own.
    Neither rule uses a reading after the origin.
    """
    rule = FORECAST_WINDOW_RESPONSE if rule is None else rule
    assert rule in ("exact_day", "last_reading"), f"unknown window rule {rule}"
    value = utils.best_average_response(window_day, window_volume,
                                        min_day=MIN_DAY_FOR_BEST_AVERAGE_RESPONSE)
    if rule == "exact_day" or np.isfinite(value):
        return value
    if len(window_day) and window_day[-1] >= origin_day - WINDOW_RESPONSE_TOLERANCE_DAYS:
        return utils.best_average_response(window_day, window_volume, min_day=float(window_day[-1]))
    return np.nan


def early_window_features(day, log_volume, origin_day, response_rule=None):
    """Predictors from one arm's measurements on or before `origin_day`.

    Depth-derived quantities: baseline log volume, best average response in
    the window (window_response, under `response_rule`), nadir depth (log
    baseline − log minimum), nadir-to-baseline ratio, minimum log volume and
    the current (last in-window) log volume.
    Shape descriptors: overall slope, initial slope (first three points),
    steepest decline, time to nadir, post-nadir slope inside the window
    (0.0 when the in-window minimum is the last point), curvature (quadratic
    coefficient) and change in slope (post-nadir slope − initial slope).
    The in-window post-nadir slope is a predictor fixed before the endpoints
    were evaluated; it is fitted from the in-window nadir on, unlike the
    durability endpoint, which leaves the nadir out (post_nadir_fit).
    """
    in_window = day <= origin_day
    window_day, window_log = day[in_window], log_volume[in_window]
    if len(window_day) < 3 or np.ptp(window_day) == 0:
        return None
    nadir_index = int(np.argmin(window_log))
    initial_slope = float(np.polyfit(window_day[:3], window_log[:3], 1)[0])
    post_nadir_slope = (float(np.polyfit(window_day[nadir_index:], window_log[nadir_index:], 1)[0])
                        if len(window_day) - nadir_index >= 2 else 0.0)
    return {
        "baseline_log_volume": float(window_log[0]),
        "response_in_window": window_response(window_day, np.exp(window_log), origin_day,
                                              response_rule),
        "nadir_depth": float(window_log[0] - window_log.min()),
        "nadir_ratio": float(np.exp(window_log.min() - window_log[0])),
        "minimum_log_volume": float(window_log.min()),
        "current_log_volume": float(window_log[-1]),
        "overall_slope": float(np.polyfit(window_day, window_log, 1)[0]),
        "initial_slope": initial_slope,
        "steepest_decline": float(np.min(np.diff(window_log) / np.diff(window_day))),
        "time_to_nadir": float(window_day[nadir_index] - window_day[0]),
        "post_nadir_slope_in_window": post_nadir_slope,
        "curvature": float(np.polyfit(window_day, window_log, 2)[0]),
        "slope_change": post_nadir_slope - initial_slope,
    }


def build_forecast_table(canonical, origin_day, horizon_day, arm_ids=None,
                         outcome_zeros=None, response_rule=None):
    """One row per arm: early-window predictors and the log volume at the
    horizon, linearly interpolated between the measurements around it. Arms
    whose observation ended before the horizon are not included.

    outcome_zeros (default FORECAST_OUTCOME_ZEROS): "floor" interpolates the
    outcome from every reading, with readings below 1 mm³ (recorded as zero
    or below 1 mm³) at 1 mm³; "drop" interpolates it from readings above the
    floor only, as in the forecasting analysis first run, so an arm still at
    the floor on the horizon day, with no later reading above it, is left
    out. Predictors always use the floored readings.
    response_rule (default FORECAST_WINDOW_RESPONSE): see window_response."""
    outcome_zeros = outcome_zeros or FORECAST_OUTCOME_ZEROS
    rows = []
    for arm_id, block in arm_series(canonical, arm_ids):
        day = block.day.to_numpy(float)
        log_volume = block.log_volume.to_numpy(float)
        kept = block.volume.to_numpy(float) > VOLUME_FLOOR if outcome_zeros == "drop" else np.ones(len(day), bool)
        outcome_day, outcome_log_volume = day[kept], log_volume[kept]
        if (len(day) < 3 or len(outcome_day) == 0 or outcome_day.max() < horizon_day
                or outcome_day.min() > horizon_day):
            continue
        features = early_window_features(day, log_volume, origin_day, response_rule)
        if features is None:
            continue
        rows.append({"arm_id": arm_id, "model_id": block.model_id.iloc[0],
                     "treatment": block.treatment.iloc[0],
                     "tumor_type": block.tumor_type.iloc[0],
                     "outcome_log_volume": float(np.interp(horizon_day, outcome_day, outcome_log_volume)),
                     **features})
    return pd.DataFrame(rows)


def make_forecaster():
    """The one gradient-boosted regressor used for every predictor set."""
    return HistGradientBoostingRegressor(max_iter=300, learning_rate=0.05,
                                         max_leaf_nodes=15, l2_regularization=1.0,
                                         random_state=SEED)


def make_linear_forecaster():
    """Ordinary least squares, the simplest well-specified learner. With the
    baseline-plus-slope set it is log-linear extrapolation of early growth;
    Section 8.7 uses it to show whether the comparisons depend on the
    gradient-boosted learner."""
    return LinearRegression()


def grouped_folds(frame, n_folds=N_FOLDS):
    """Cross-validation folds that keep every arm of a PDX model in one fold."""
    groups = frame.model_id.astype(str).to_numpy()
    n_splits = min(n_folds, len(np.unique(groups)))
    folds = list(GroupKFold(n_splits=n_splits).split(frame, groups=groups))
    for train, test in folds:
        assert set(groups[train]).isdisjoint(groups[test]), "a PDX model is in two folds"
    return folds


def out_of_fold_predictions(frame, predictors, outcome, folds, learner_factory=None):
    """Predictions for every row, each from a model that did not see its PDX
    model. `learner_factory` (default make_forecaster) builds a fresh,
    unfitted learner for each fold."""
    learner_factory = make_forecaster if learner_factory is None else learner_factory
    design = frame[predictors].to_numpy(float)
    predicted = np.full(len(outcome), np.nan)
    for train, test in folds:
        learner = learner_factory()
        learner.fit(design[train], outcome[train])
        predicted[test] = learner.predict(design[test])
    return predicted


def compare_predictor_sets(frame, predictor_sets, folds=None):
    """Out-of-fold predictions and RMSE for several predictor sets on one frame."""
    outcome = frame.outcome_log_volume.to_numpy(float)
    folds = grouped_folds(frame) if folds is None else folds
    predictions = {name: out_of_fold_predictions(frame, predictors, outcome, folds)
                   for name, predictors in predictor_sets.items()}
    errors = {name: rmse(outcome, predicted) for name, predicted in predictions.items()}
    return predictions, errors


def selection_by_model(selection, prediction_columns):
    """For each PDX model with two or more arms: the regret of the arm each
    prediction would select (its observed log volume minus the smallest
    observed log volume among the model's arms) and whether the selections
    differ."""
    rows = []
    for model_id, block in selection.groupby("model_id", sort=True):
        if len(block) < 2:
            continue
        best = float(block.outcome_log_volume.min())
        row = {"model_id": model_id, "n_treatments": len(block)}
        chosen = {}
        for column in prediction_columns:
            choice = block.loc[block[column].idxmin()]
            chosen[column] = choice.arm_id
            row[f"{column}_regret"] = float(choice.outcome_log_volume - best)
        row["different_choice"] = len(set(chosen.values())) > 1
        rows.append(row)
    return pd.DataFrame(rows)


def within_model_pairs(selection, prediction_columns, extra_gaps=()):
    """Every pair of arms within a PDX model with distinct observed outcomes.

    For each prediction column the pair scores 1 when the predicted order
    matches the observed order, 0.5 when the predictions are tied and 0
    otherwise. Pairs with tied observed outcomes have no order and are left
    out. For each column named in `extra_gaps` the absolute difference
    between the two arms is stored as `<column>_gap` (for example the day-10
    log volume, a separation known when a treatment is selected).
    """
    rows = []
    for model_id, block in selection.groupby("model_id", sort=True):
        outcome = block.outcome_log_volume.to_numpy(float)
        predictions = {c: block[c].to_numpy(float) for c in prediction_columns}
        extras = {c: block[c].to_numpy(float) for c in extra_gaps}
        for i, j in combinations(range(len(block)), 2):
            observed_gap = outcome[i] - outcome[j]
            if observed_gap == 0:
                continue
            row = {"model_id": model_id, "observed_gap": abs(observed_gap)}
            for column, values in extras.items():
                row[f"{column}_gap"] = abs(values[i] - values[j])
            for column, predicted in predictions.items():
                predicted_gap = predicted[i] - predicted[j]
                row[column] = 0.5 if predicted_gap == 0 else float(
                    np.sign(predicted_gap) == np.sign(observed_gap))
            rows.append(row)
    return pd.DataFrame(rows)


def ranking_difference(pairs, column_a, column_b, n_bootstrap, rng):
    """Pairwise ranking accuracy of `column_a` minus that of `column_b` over
    the rows of a within_model_pairs table, with a 95% percentile interval
    from resampling PDX models (each model's pairs move together)."""
    scores_a, scores_b = pairs[column_a].to_numpy(float), pairs[column_b].to_numpy(float)
    draws = cluster_bootstrap(rows_by_cluster(pairs.model_id),
                              lambda rows: scores_a[rows].mean() - scores_b[rows].mean(),
                              n_bootstrap, rng)
    low, high = np.percentile(draws, [2.5, 97.5])
    return float(scores_a.mean() - scores_b.mean()), float(low), float(high)


def evaluate_endpoints(frame, depth_set, trajectory_set, n_bootstrap, rng, folds=None):
    """Both prespecified endpoints on one forecasting table, computed exactly
    as in Section 7: the same learner and grouped folds, out-of-fold
    predictions, the relative RMSE reduction with an interval resampling PDX
    models, the regret of the arm each set selects in every PDX model with
    two or more arms (interval resampling those PDX models), and pairwise
    ranking accuracy. Used for the sensitivity analyses of 7.4, 8.6 and 8.8.

    `folds` (default grouped_folds(frame), as in Section 7) lets a caller
    hold the assignment of PDX models to folds fixed while arms are added,
    so that a change in the result comes from the added arms and not from a
    reshuffle of the folds. Row indices refer to `frame` after reset_index.

    Returns a dict of results and the frame with both predictions added.
    """
    frame = frame.reset_index(drop=True)
    outcome = frame.outcome_log_volume.to_numpy(float)
    folds = grouped_folds(frame) if folds is None else folds
    depth = out_of_fold_predictions(frame, depth_set, outcome, folds)
    trajectory = out_of_fold_predictions(frame, trajectory_set, outcome, folds)
    reduction, low, high, rmse_draws = reduction_with_interval(
        outcome, depth, trajectory, rows_by_cluster(frame.model_id.astype(str)), n_bootstrap, rng)
    selected = frame.assign(depth_prediction=depth, trajectory_prediction=trajectory)
    regret_table = selection_by_model(selected, ["depth_prediction", "trajectory_prediction"])
    result = {"arms": len(frame), "pdx_models": frame.model_id.nunique(),
              "depth_rmse": rmse(outcome, depth), "trajectory_rmse": rmse(outcome, trajectory),
              "rmse_reduction": reduction, "rmse_low": low, "rmse_high": high,
              "rmse_p_vs_margin": float(np.mean(rmse_draws <= FORECAST_MARGIN)) if len(rmse_draws) else np.nan,
              "regret_models": len(regret_table), "depth_regret": np.nan, "trajectory_regret": np.nan,
              "regret_reduction": np.nan, "regret_low": np.nan, "regret_high": np.nan,
              "regret_p_vs_margin": np.nan,
              "share_different_choice": np.nan, "ranking_pairs": 0,
              "ranking_depth": np.nan, "ranking_trajectory": np.nan}
    if len(regret_table):
        depth_regret = regret_table.depth_prediction_regret.to_numpy(float)
        trajectory_regret = regret_table.trajectory_prediction_regret.to_numpy(float)
        regret_draws = []
        for _ in range(n_bootstrap):
            # Each row of the regret table is one PDX model, so resampling
            # rows resamples PDX models.
            picked = rng.integers(0, len(regret_table), len(regret_table))
            depth_mean = depth_regret[picked].mean()
            if depth_mean > 0:
                regret_draws.append((depth_mean - trajectory_regret[picked].mean()) / depth_mean)
        pairs = within_model_pairs(selected, ["depth_prediction", "trajectory_prediction"])
        result.update(
            depth_regret=float(depth_regret.mean()), trajectory_regret=float(trajectory_regret.mean()),
            regret_reduction=float((depth_regret.mean() - trajectory_regret.mean()) / depth_regret.mean())
            if depth_regret.mean() > 0 else np.nan,
            regret_low=float(np.percentile(regret_draws, 2.5)) if regret_draws else np.nan,
            regret_high=float(np.percentile(regret_draws, 97.5)) if regret_draws else np.nan,
            regret_p_vs_margin=float(np.mean(np.asarray(regret_draws) <= REGRET_MARGIN))
            if regret_draws else np.nan,
            share_different_choice=float(regret_table.different_choice.mean()),
            ranking_pairs=len(pairs),
            ranking_depth=float(pairs.depth_prediction.mean()) if len(pairs) else np.nan,
            ranking_trajectory=float(pairs.trajectory_prediction.mean()) if len(pairs) else np.nan)
    return result, selected


def paired_selection_contrast(selected, baseline_column, challenger_column, n_bootstrap, rng):
    """The selection endpoints of two predictions made for the same arms (the
    same held-out folds): mean regret of each, the relative reduction in
    regret of the challenger against the baseline with a 95% interval
    resampling PDX models, and the difference in pairwise ranking accuracy
    (challenger minus baseline) with a 95% interval resampling PDX models.
    Because both predictions score the same models and pairs, the intervals
    are for the paired difference."""
    regret_table = selection_by_model(selected, [baseline_column, challenger_column])
    baseline_regret = regret_table[f"{baseline_column}_regret"].to_numpy(float)
    challenger_regret = regret_table[f"{challenger_column}_regret"].to_numpy(float)
    draws = []
    for _ in range(n_bootstrap):
        picked = rng.integers(0, len(regret_table), len(regret_table))
        baseline_mean = baseline_regret[picked].mean()
        if baseline_mean > 0:
            draws.append((baseline_mean - challenger_regret[picked].mean()) / baseline_mean)
    pairs = within_model_pairs(selected, [baseline_column, challenger_column])
    ranking_gain, ranking_low, ranking_high = ranking_difference(pairs, challenger_column, baseline_column,
                                                                 n_bootstrap, rng)
    reduction = ((baseline_regret.mean() - challenger_regret.mean()) / baseline_regret.mean()
                 if baseline_regret.mean() > 0 else np.nan)
    return {"regret_models": len(regret_table), "regret_baseline": float(baseline_regret.mean()),
            "regret_challenger": float(challenger_regret.mean()), "regret_reduction": float(reduction),
            "regret_low": float(np.percentile(draws, 2.5)) if draws else np.nan,
            "regret_high": float(np.percentile(draws, 97.5)) if draws else np.nan,
            "different_choice": float(regret_table.different_choice.mean()),
            "ranking_pairs": len(pairs), "ranking_baseline": float(pairs[baseline_column].mean()),
            "ranking_challenger": float(pairs[challenger_column].mean()),
            "ranking_difference": ranking_gain, "ranking_low": ranking_low, "ranking_high": ranking_high}


def selection_stability(frame, depth_set, trajectory_set, n_refits, rng, learner_factory=None):
    """How stable each predictor set's choice of arm is when the learner is
    refitted on resampled training data.

    The folds of Section 7 stay fixed. In every refit, each fold's training
    set is replaced by a bootstrap resample of its PDX models (whole models
    drawn with replacement), the learner is refitted for both predictor
    sets, and the held-out arms are predicted again. For every PDX model with
    two or more arms the arm each set selects is compared with the arm it
    selects in the unperturbed fit.

    Returns a per-model table (share of refits in which each set keeps its
    original choice) and a per-refit table (relative regret reduction,
    share of models where the two sets choose differently, ranking
    accuracies), whose spread shows how much the headline numbers depend on
    which PDX models happened to be in training.
    """
    learner_factory = make_forecaster if learner_factory is None else learner_factory
    frame = frame.reset_index(drop=True)
    outcome = frame.outcome_log_volume.to_numpy(float)
    groups = frame.model_id.astype(str).to_numpy()
    folds = grouped_folds(frame)
    sets = {"depth": depth_set, "trajectory": trajectory_set}
    designs = {name: frame[predictors].to_numpy(float) for name, predictors in sets.items()}
    rows_of_model = rows_by_cluster(groups)
    eligible = [model for model, rows in rows_of_model.items() if len(rows) >= 2]

    def chosen_arms(predictions):
        """The row each prediction selects in every eligible PDX model (first minimum)."""
        return np.array([rows_of_model[model][np.argmin(predictions[rows_of_model[model]])] for model in eligible])

    original = {name: out_of_fold_predictions(frame, sets[name], outcome, folds, learner_factory) for name in sets}
    original_choice = {name: chosen_arms(original[name]) for name in sets}
    kept = {name: np.zeros(len(eligible)) for name in sets}
    refit_rows = []
    for _ in range(n_refits):
        predictions = {name: np.full(len(frame), np.nan) for name in sets}
        for train, test in folds:
            training_models = np.unique(groups[train])
            drawn = rng.choice(training_models, size=len(training_models), replace=True)
            rows = np.concatenate([rows_of_model[model] for model in drawn])
            for name in sets:
                learner = learner_factory().fit(designs[name][rows], outcome[rows])
                predictions[name][test] = learner.predict(designs[name][test])
        choice = {name: chosen_arms(predictions[name]) for name in sets}
        for name in sets:
            kept[name] += choice[name] == original_choice[name]
        selected = frame.assign(depth_prediction=predictions["depth"], trajectory_prediction=predictions["trajectory"])
        regret_table = selection_by_model(selected, ["depth_prediction", "trajectory_prediction"])
        pairs = within_model_pairs(selected, ["depth_prediction", "trajectory_prediction"])
        depth_mean = regret_table.depth_prediction_regret.mean()
        refit_rows.append({"regret_reduction": (depth_mean - regret_table.trajectory_prediction_regret.mean()) / depth_mean
                           if depth_mean > 0 else np.nan,
                           "different_choice": float(np.mean(choice["depth"] != choice["trajectory"])),
                           "ranking_depth": float(pairs.depth_prediction.mean()),
                           "ranking_trajectory": float(pairs.trajectory_prediction.mean())})
    per_model = pd.DataFrame({"model_id": eligible,
                              "arms": [len(rows_of_model[model]) for model in eligible],
                              "stability_depth": kept["depth"] / n_refits,
                              "stability_trajectory": kept["trajectory"] / n_refits})
    return per_model, pd.DataFrame(refit_rows)


def replicate_outcome_regret(selected, prediction_columns, replicate_sd, n_draws, rng):
    """Selection judged against a hypothetical second animal, because the
    best observed arm is not necessarily the truly best treatment.

    With one animal per arm, the observed day-24 log volume is y = μ + e. A
    second animal would give y' = μ + e', so y' − y has SD √2 × the SD of one
    animal's deviation; `replicate_sd` is that SD of y' − y. For each draw,
    y' = y + Normal(0, replicate_sd) is generated for every arm, and each
    prediction's chosen arm (fixed, from the predictions) is scored by its
    regret against y'. Also scored: the arm with the best OBSERVED outcome
    (its regret against y' is the regret that outcome noise alone leaves,
    the floor no predictor can beat on average), and a random choice (the
    mean regret over the model's arms). Returns per-draw mean regrets and the
    share of models whose best arm stays best.
    """
    results = {column: [] for column in [*prediction_columns, "observed best", "random choice"]}
    best_kept = []
    for _, block in selected.groupby("model_id", sort=True):
        if len(block) < 2:
            continue
        observed = block.outcome_log_volume.to_numpy(float)
        replicate = observed[None, :] + rng.normal(0.0, replicate_sd, size=(n_draws, len(observed)))
        best = replicate.min(axis=1)
        for column in prediction_columns:
            results[column].append(replicate[:, int(np.argmin(block[column].to_numpy(float)))] - best)
        results["observed best"].append(replicate[:, int(np.argmin(observed))] - best)
        results["random choice"].append(replicate.mean(axis=1) - best)
        best_kept.append(np.argmin(replicate, axis=1) == int(np.argmin(observed)))
    per_draw = {column: np.mean(np.vstack(values), axis=0) for column, values in results.items()}
    return per_draw, float(np.mean(np.vstack(best_kept)))


def censoring_scenario_table(canonical, origin_day, horizon_day, arm_ids=None, outcome_zeros=None,
                             response_rule=None):
    """One row per arm, as build_forecast_table, but with arms whose
    observation ended before the horizon reinserted under two scenarios.

    Arms observed through the horizon get exactly the outcome of
    build_forecast_table under `outcome_zeros` (both scenario columns equal
    it): under "drop", an arm still at the floor on the horizon day is left
    out, as in the primary analysis. An arm whose observation ended before
    the horizon (ended_early) gets two outcomes: growth stopped at its last
    reading, or growth continued at its own whole-arm slope (non-negative).
    Under "drop", a censored arm whose last reading is at the floor has no
    reading above the floor to carry forward and is left out.
    Returns the table and the number of arms left out at the floor.
    """
    outcome_zeros = outcome_zeros or FORECAST_OUTCOME_ZEROS
    rows, left_out_at_floor = [], 0
    for arm_id, block in arm_series(canonical, arm_ids):
        day, log_volume = block.day.to_numpy(float), block.log_volume.to_numpy(float)
        features = early_window_features(day, log_volume, origin_day, response_rule) if len(day) >= 3 else None
        if features is None:
            continue
        kept = (block.volume.to_numpy(float) > VOLUME_FLOOR if outcome_zeros == "drop"
                else np.ones(len(day), bool))
        if day.max() >= horizon_day:
            kept_day, kept_log = day[kept], log_volume[kept]
            if len(kept_day) == 0 or kept_day.max() < horizon_day or kept_day.min() > horizon_day:
                left_out_at_floor += 1
                continue
            stopped = continued = float(np.interp(horizon_day, kept_day, kept_log))
        else:
            if not kept[-1]:
                left_out_at_floor += 1
                continue
            slope = float(np.polyfit(day, log_volume, 1)[0])
            stopped = float(log_volume[-1])
            continued = stopped + (horizon_day - day.max()) * max(0.0, slope)
        rows.append({"arm_id": arm_id, "model_id": block.model_id.iloc[0],
                     "treatment": block.treatment.iloc[0], "tumor_type": block.tumor_type.iloc[0],
                     "ended_early": day.max() < horizon_day, "growth_stopped": stopped,
                     "growth_continued": continued, **features})
    return pd.DataFrame(rows), left_out_at_floor


def model_fold_assignment(frame, folds):
    """The fold index of every PDX model under a set of grouped folds."""
    models = frame.model_id.astype(str).to_numpy()
    assignment = {}
    for index, (_, test) in enumerate(folds):
        for model in np.unique(models[test]):
            assignment[model] = index
    return assignment


def folds_from_assignment(frame, assignment, n_folds):
    """Grouped folds for `frame` with every PDX model in its fold from
    `assignment`. A PDX model missing from `assignment` goes to the fold that
    holds the fewest arms so far (models taken in sorted order, so the result
    is deterministic). Used to add arms to a pool without reshuffling the
    folds: GroupKFold balances folds by arm counts, so adding even one arm
    can move many PDX models to other folds and change every prediction."""
    models = frame.model_id.astype(str).to_numpy()
    assignment = dict(assignment)
    fold_sizes = np.zeros(n_folds, int)
    for model, index in assignment.items():
        fold_sizes[index] += int((models == model).sum())
    for model in sorted(set(models) - set(assignment)):
        index = int(np.argmin(fold_sizes))
        assignment[model] = index
        fold_sizes[index] += int((models == model).sum())
    fold_of_row = np.array([assignment[model] for model in models])
    return [(np.flatnonzero(fold_of_row != index), np.flatnonzero(fold_of_row == index))
            for index in range(n_folds) if (fold_of_row == index).any()]


def random_grouped_folds(frame, n_folds, rng):
    """A random partition of the PDX models into `n_folds` folds of similar
    size: models are shuffled, then each goes to the fold holding the fewest
    arms so far. GroupKFold (Section 7) gives one deterministic partition;
    repeating the analysis over random partitions shows how much a result
    depends on which PDX models happen to share a fold."""
    models = frame.model_id.astype(str).to_numpy()
    unique_models = np.unique(models)
    n_folds = min(n_folds, len(unique_models))
    fold_sizes = np.zeros(n_folds, int)
    assignment = {}
    for model in rng.permutation(unique_models):
        index = int(np.argmin(fold_sizes))
        assignment[model] = index
        fold_sizes[index] += int((models == model).sum())
    return folds_from_assignment(frame, assignment, n_folds)


# --------------------------------------------------------------------------
# Molecular correlates (Section 9)
# --------------------------------------------------------------------------
RIDGE_ALPHAS = np.logspace(-3, 4, 30)


def _ridge_path(train_x, train_y, test_x, alphas):
    """Predictions on test_x for every penalty in `alphas`, by SVD of the
    standardized training design (the ridge solution path in closed form)."""
    centre, scale = train_x.mean(axis=0), train_x.std(axis=0)
    keep = scale > 0
    if not keep.any():
        return np.repeat(np.full((len(test_x), 1), train_y.mean()), len(alphas), axis=1)
    x = (train_x[:, keep] - centre[keep]) / scale[keep]
    x_test = (test_x[:, keep] - centre[keep]) / scale[keep]
    y_mean = train_y.mean()
    u, s, vt = np.linalg.svd(x, full_matrices=False)
    uty = u.T @ (train_y - y_mean)
    factors = s[None, :] / (s[None, :] ** 2 + alphas[:, None])
    coefficients = (factors * uty[None, :]) @ vt
    return x_test @ coefficients.T + y_mean


def grouped_ridge_r2(design, target, groups, alphas=RIDGE_ALPHAS, n_outer=5, n_inner=5,
                     treatments=None):
    """Out-of-fold R² of ridge regression with PDX-model-grouped folds.

    The penalty is chosen inside each training fold by an inner grouped
    cross-validation, and missing feature values are imputed with
    training-fold medians, so nothing from a test fold reaches its own
    prediction.

    With `treatments` (one label per row), each outcome is first expressed
    relative to its treatment's mean, computed on the training fold only (a
    treatment absent from the training fold takes the training mean). The
    ridge model is fitted to those within-treatment residuals, and the R²
    returned is 1 − SSE(treatment mean + ridge) / SSE(treatment mean): the
    share of the within-treatment variance that the features explain beyond
    a treatment-mean model. Inner folds reuse the outer training fold's
    treatment means, which affects only the choice of penalty, never a test
    prediction.
    """
    outer_splits = min(n_outer, len(np.unique(groups)))
    if outer_splits < 2:
        return np.nan
    predicted = np.full(len(target), np.nan)
    reference = np.full(len(target), np.nan)       # treatment-mean prediction (adjusted R² only)
    for train, test in GroupKFold(n_splits=outer_splits).split(design, groups=groups):
        medians = np.nanmedian(design[train], axis=0)
        medians = np.where(np.isfinite(medians), medians, 0.0)
        train_x = np.where(np.isfinite(design[train]), design[train], medians)
        test_x = np.where(np.isfinite(design[test]), design[test], medians)
        train_y, train_groups = target[train], groups[train]
        train_offset, test_offset = 0.0, 0.0
        if treatments is not None:
            means = pd.Series(target[train]).groupby(np.asarray(treatments)[train]).mean()
            train_offset = means.reindex(np.asarray(treatments)[train]).to_numpy(float)
            test_offset = (means.reindex(np.asarray(treatments)[test])
                           .fillna(float(target[train].mean())).to_numpy(float))
            train_y = train_y - train_offset
            reference[test] = test_offset
        inner_splits = min(n_inner, len(np.unique(train_groups)))
        inner_error = np.zeros(len(alphas))
        for inner_train, inner_test in GroupKFold(n_splits=inner_splits).split(
                train_x, groups=train_groups):
            path = _ridge_path(train_x[inner_train], train_y[inner_train],
                               train_x[inner_test], alphas)
            inner_error += ((path - train_y[inner_test][:, None]) ** 2).sum(axis=0)
        best_alpha = alphas[[int(np.argmin(inner_error))]]
        predicted[test] = test_offset + _ridge_path(train_x, train_y, test_x, best_alpha)[:, 0]
    if treatments is None:
        return float(r2_score(target, predicted))
    return float(1.0 - np.sum((target - predicted) ** 2) / np.sum((target - reference) ** 2))


def block_r2(frame, features, outcome_column, adjust_treatment=False):
    """Out-of-fold R² of one feature block for one outcome. With
    `adjust_treatment`, the R² beyond a treatment-mean model (see
    grouped_ridge_r2)."""
    return grouped_ridge_r2(frame[features].to_numpy(float),
                            frame[outcome_column].to_numpy(float),
                            frame.model_id.astype(str).to_numpy(),
                            treatments=frame.treatment.astype(str).to_numpy() if adjust_treatment else None)


def treatment_mean_r2(frame, outcome_column, n_outer=5):
    """Out-of-fold R² of the treatment-mean model alone (training-fold mean of
    each treatment; a treatment absent from the training fold takes the
    training mean), with the PDX-model-grouped folds of grouped_ridge_r2. It
    shows how much of an outcome's variance is drug-to-drug."""
    target = frame[outcome_column].to_numpy(float)
    groups = frame.model_id.astype(str).to_numpy()
    treatments = frame.treatment.astype(str).to_numpy()
    predicted = np.full(len(target), np.nan)
    for train, test in GroupKFold(n_splits=min(n_outer, len(np.unique(groups)))).split(target, groups=groups):
        means = pd.Series(target[train]).groupby(treatments[train]).mean()
        predicted[test] = means.reindex(treatments[test]).fillna(float(target[train].mean())).to_numpy(float)
    return float(r2_score(target, predicted))


def block_features(block):
    """All feature columns of one block specification."""
    return block["model"] + block["arm"] + block["fixed"]


def dissociation_statistic(frame, feature_blocks, outcomes, adjust_treatment=False):
    """D = [R²(signatures, durability) − R²(context, durability)]
         − [R²(signatures, depth) − R²(context, depth)].
    With `adjust_treatment`, every R² is the R² beyond a treatment-mean model."""
    r2 = {(block, label): block_r2(frame, block_features(feature_blocks[block]), column, adjust_treatment)
          for block in ("signatures", "context") for label, column in outcomes.items()}
    return ((r2[("signatures", "durability")] - r2[("context", "durability")])
            - (r2[("signatures", "depth")] - r2[("context", "depth")]))


def normalized_set_name(name):
    """A gene-set name reduced to upper-case letters and digits, without the
    HALLMARK prefix, so that the MSigDB GMT name
    (HALLMARK_EPITHELIAL_MESENCHYMAL_TRANSITION) and the Enrichr name
    (Epithelial Mesenchymal Transition) of one set compare equal."""
    text = "".join(character for character in str(name).upper() if character.isalnum())
    return text[len("HALLMARK"):] if text.startswith("HALLMARK") else text


def read_gene_set_library(path):
    """{normalized name: (name as written, [genes])} from a GMT file or an
    Enrichr text library.

    Both formats hold one set per line, tab-separated: the name, a second
    field (a description or URL in GMT, usually empty in Enrichr) and then
    the genes. Some Enrichr libraries append ",weight" to each gene; the
    weight is removed. Gene symbols are upper-cased and duplicates dropped.
    """
    library = {}
    for line in Path(path).read_text(encoding="utf-8", errors="replace").splitlines():
        fields = [field.strip() for field in line.rstrip("\n").split("\t")]
        if len(fields) < 3 or not fields[0]:
            continue
        genes = [field.split(",")[0].strip().upper() for field in fields[2:] if field.strip()]
        library[normalized_set_name(fields[0])] = (fields[0], list(dict.fromkeys(genes)))
    return library


def gene_set_score(expression_ranks, genes):
    """Mean within-model percentile rank of the genes present, minus 0.5 (the
    scoring rule of Section 9.1), and the number of genes present."""
    present = [gene for gene in genes if gene in expression_ranks.index]
    if not present:
        return pd.Series(np.nan, index=expression_ranks.columns), 0
    return expression_ranks.loc[present].mean(axis=0) - 0.5, len(present)


def permute_within_tumor_type(frame, model_features, arm_features, rng):
    """Break the feature–outcome link while keeping tumor-type structure.

    PDX-model-level features are reassigned from another PDX model of the
    same tumor type (every arm of a model receives the donor's values);
    arm-level features are shuffled across arms of the same tumor type.
    Tumor-type indicators cannot change under either shuffle.
    """
    permuted = frame.copy()
    for _, positions in frame.groupby("tumor_type", sort=True).indices.items():
        block = frame.iloc[positions]
        if model_features:
            models = block.model_id.unique()
            donor = dict(zip(models, rng.permutation(models)))
            values = block.groupby("model_id", sort=False)[model_features].first()
            permuted.iloc[positions, [frame.columns.get_loc(c) for c in model_features]] = (
                values.reindex(block.model_id.map(donor)).to_numpy())
        if arm_features:
            permuted.iloc[positions, [frame.columns.get_loc(c) for c in arm_features]] = (
                block[arm_features].to_numpy()[rng.permutation(len(block))])
    return permuted


# --------------------------------------------------------------------------
# Combinations against their components (Section 10)
# --------------------------------------------------------------------------
def combination_durability_values(frame):
    """Durability value of each arm for comparisons with combinations, defined
    for every arm whose regrowth status is known.

    - A post-nadir growth rate where one is defined, including arms whose
      nadir is their first measurement (they never shrank; the rate is their
      growth after the first measurement).
    - −∞ for an arm whose smallest volume is its last measurement (with at
      least three measurements): no regrowth was seen during follow-up, so
      its durability is right-censored and it counts as more durable than
      any arm with a measured regrowth rate.
    - NaN for an arm with one measurement after an interior nadir: regrowth
      had begun but its rate cannot be measured.
    """
    rate = frame.post_nadir_growth_rate.to_numpy(float)
    no_regrowth = ((frame.n_after_nadir.to_numpy(float) == 0)
                   & (frame.n_measurements.to_numpy(float) >= 3))
    return np.where(no_regrowth, -np.inf, np.where(np.isfinite(rate), rate, np.nan))


def combination_evaluations_table(depth_values, durability_values, require_all_components):
    """Every combination–PDX model evaluation against the best single component.

    `depth_values` and `durability_values` are Series indexed by (model_id,
    treatment). An evaluation needs the combination and every component in
    `depth_values`. Depth gain is the lowest component best average response
    minus the combination's. Durability gain is the lowest component
    durability value minus the combination's (positive: the combination
    regrows more slowly than its best component), with the infinite values of
    combination_durability_values: a combination with no observed regrowth
    against a component that regrew gains +∞, the reverse −∞, and two arms
    without regrowth tie at 0. With `require_all_components`, the durability
    gain is computed only when every component has a durability value;
    otherwise the best is taken among the components that have one (the rule
    of the analysis first run).
    """
    labels = sorted(t for t in depth_values.index.get_level_values("treatment").unique() if "+" in t)
    rows = []
    for combination in labels:
        components = [part.strip() for part in combination.split("+")]
        for model_id in depth_values.xs(combination, level="treatment").index:
            if not all((model_id, part) in depth_values.index for part in components):
                continue
            combination_value = durability_values.get((model_id, combination), np.nan)
            component_values = np.array([durability_values.get((model_id, part), np.nan)
                                         for part in components], float)
            missing = bool(np.isnan(component_values).any())
            if (np.isnan(combination_value) or np.isnan(component_values).all()
                    or (require_all_components and missing)):
                durability_gain = np.nan
            else:
                best = np.nanmin(component_values)
                durability_gain = (0.0 if np.isinf(best) and np.isinf(combination_value)
                                   else float(best - combination_value))
            rows.append({
                "model_id": model_id, "combination": combination, "components": len(components),
                "depth_gain": min(depth_values[(model_id, part)] for part in components)
                              - depth_values[(model_id, combination)],
                "durability_gain": durability_gain,
                "combination_durability": combination_value,
                "best_component_durability": (np.nanmin(component_values)
                                              if not np.isnan(component_values).all() else np.nan),
                "component_without_durability": missing,
                "combination_without_regrowth": bool(np.isneginf(combination_value))})
    return pd.DataFrame(rows)


def finite_rank_values(values):
    """Values with +∞ and −∞ moved just beyond the finite range, so a rank
    statistic (Spearman) treats them as the largest and smallest values. A
    monotone change, so ranks and Spearman ρ are those of the extended
    values."""
    values = np.asarray(values, float).copy()
    finite = values[np.isfinite(values)]
    if finite.size:
        span = float(np.ptp(finite)) or 1.0
        values[np.isposinf(values)] = finite.max() + span
        values[np.isneginf(values)] = finite.min() - span
    return values


def extended_median(values, axis=None):
    """Median that stays defined when the values include ±∞: the ordinary
    median when every value is finite, otherwise the lower median (an
    observed value), so +∞ and −∞ are never averaged into NaN."""
    values = np.asarray(values, float)
    if np.isfinite(values).all():
        return np.median(values, axis=axis)
    return np.quantile(values, 0.5, axis=axis, method="inverted_cdf")


def extended_median_interval(values, n_bootstrap, rng):
    """95% bootstrap interval of extended_median (resampling values; order
    statistics of the bootstrap medians, without interpolation, so infinite
    bounds stay defined). NaN with fewer than 4 values."""
    values = np.asarray(values, float)
    values = values[~np.isnan(values)]
    if len(values) < 4:
        return np.nan, np.nan
    medians = np.sort(extended_median(values[rng.integers(0, len(values), size=(n_bootstrap, len(values)))], axis=1))
    return float(medians[int(np.floor(0.025 * (n_bootstrap - 1)))]), float(medians[int(np.ceil(0.975 * (n_bootstrap - 1)))])


def sign_test_p(gains):
    """Two-sided exact binomial (sign) test of positive against negative
    gains; zero gains (ties) are left out. NaN with no untied gains."""
    gains = np.asarray(gains, float)
    gains = gains[~np.isnan(gains)]
    positive, negative = int((gains > 0).sum()), int((gains < 0).sum())
    if positive + negative == 0:
        return np.nan
    return float(stats.binomtest(positive, positive + negative, 0.5).pvalue)


def gain_summary(depth_gains, durability_gains, n_bootstrap, rng, min_evaluations):
    """The per-combination summary of 10.2 for any set of evaluations (used
    by 10.3 within BRAF V600 groups): median depth gain with a 95% bootstrap
    interval and a two-sided Wilcoxon signed-rank P; median durability gain
    (an extended median, since gains can be infinite) with its interval, a
    two-sided sign-test P, and the counts of evaluations in which the
    combination regrew more slowly, faster or the same as its best
    component. Tests need at least `min_evaluations` evaluations; missing
    durability gains are left out."""
    depth = np.asarray(depth_gains, float)
    depth = depth[np.isfinite(depth)]
    durability = np.asarray(durability_gains, float)
    durability = durability[~np.isnan(durability)]
    depth_low, depth_high = bootstrap_median_interval(depth, n_bootstrap, rng)
    durability_low, durability_high = extended_median_interval(durability, n_bootstrap, rng)
    return {
        "evaluations_depth": len(depth), "evaluations": len(durability),
        "median_depth_gain": float(np.median(depth)) if len(depth) else np.nan,
        "depth_low": depth_low, "depth_high": depth_high,
        "depth_p": (float(stats.wilcoxon(depth).pvalue)
                    if len(depth) >= min_evaluations and np.ptp(depth) > 0 else np.nan),
        "median_durability_gain": float(extended_median(durability)) if len(durability) else np.nan,
        "durability_low": durability_low, "durability_high": durability_high,
        "better_durability": int((durability > 0).sum()), "worse_durability": int((durability < 0).sum()),
        "tied_durability": int((durability == 0).sum()),
        "durability_p": sign_test_p(durability) if len(durability) >= min_evaluations else np.nan}


# --------------------------------------------------------------------------
# Mechanism class (Section 11)
# --------------------------------------------------------------------------
def weighted_compound_effect(values, cytostatic_codes, comparator_codes, model_codes,
                             cytostatic_weights, comparator_weights, model_weights):
    """The compound-level mechanism-class effect of 11.3 with integer
    resampling weights on cytostatic compounds, comparator compounds and PDX
    models (all weights 1 gives the 11.3 estimate).

    `values` holds one durability difference per (cytostatic compound,
    comparator compound, PDX model) cell. Each cytostatic compound's effect is
    the weighted mean of its cells (cell weight = comparator weight × model
    weight); the result is the mean of those effects weighted by the
    cytostatic weights, over compounds with positive weight and at least one
    weighted cell.
    """
    n_compounds = len(cytostatic_weights)
    cell_weights = comparator_weights[comparator_codes] * model_weights[model_codes]
    numerator = np.bincount(cytostatic_codes, weights=cell_weights * values, minlength=n_compounds)
    denominator = np.bincount(cytostatic_codes, weights=cell_weights, minlength=n_compounds)
    present = (denominator > 0) & (cytostatic_weights > 0)
    if not present.any():
        return np.nan
    effects = numerator[present] / denominator[present]
    return float(np.sum(cytostatic_weights[present] * effects) / np.sum(cytostatic_weights[present]))


def bootstrap_mean(values, n_bootstrap, seed):
    """Mean with a 95% percentile interval and two-sided bootstrap P value."""
    rng = np.random.default_rng(seed)
    values = np.asarray(values, float)
    values = values[np.isfinite(values)]
    draws = np.array([rng.choice(values, size=len(values), replace=True).mean()
                      for _ in range(n_bootstrap)])
    low, high = np.percentile(draws, [2.5, 97.5])
    return {"estimate": float(values.mean()), "low": float(low), "high": float(high),
            "p_two_sided": bootstrap_p_two_sided(draws), "n": len(values)}


def family_contrast_statistic(model_codes, code_a, code_b, differences, group_one, group_two):
    """Mean over PDX models of the mean oriented difference (group one minus
    group two) across within-model pairs with one member in each group.
    `differences` are member a minus member b. Returns the statistic, the
    number of PDX models and the number of pairs."""
    a_first = np.isin(code_a, group_one) & np.isin(code_b, group_two)
    b_first = np.isin(code_b, group_one) & np.isin(code_a, group_two)
    usable = (a_first | b_first) & np.isfinite(differences)
    if not usable.any():
        return np.nan, 0, 0
    oriented = np.where(a_first, differences, -differences)[usable]
    models = model_codes[usable]
    sums = np.bincount(models, weights=oriented)
    counts = np.bincount(models)
    present = counts > 0
    return float((sums[present] / counts[present]).mean()), int(present.sum()), int(usable.sum())


# --------------------------------------------------------------------------
# Clinical tumor-burden endpoints (Section 14)
# --------------------------------------------------------------------------
PROGRESSION_RATIO = 1.20          # RECIST 1.1: >= 20% above the reference sum
PROGRESSION_ABSOLUTE_MM = 5.0     # and >= 5 mm absolute increase


def read_sas_table(path):
    """A Project Data Sphere SAS table, with the trailing spaces of
    fixed-width character storage stripped."""
    frame = pd.read_sas(path, encoding="latin-1")
    for column in frame.select_dtypes(include=["object", "string"]).columns:
        frame[column] = frame[column].str.strip()
    return frame


def crosses_progression(sld, reference):
    """RECIST 1.1 sum criterion: at least 20% and at least 5 mm above the reference."""
    return bool(np.isfinite(reference) and sld >= max(PROGRESSION_RATIO * reference,
                                                       reference + PROGRESSION_ABSOLUTE_MM))


def progressed_before(block, day_limit, new_lesion_day):
    """True when RECIST progression occurred strictly before `day_limit`: a new
    lesion, or a sum that crossed the threshold relative to the smallest
    complete-assessment sum recorded before it. `block` holds one patient's
    assessments (day, sld, complete) sorted by day."""
    if pd.notna(new_lesion_day) and new_lesion_day < day_limit:
        return True
    reference = np.inf
    for day, sld, complete in block.loc[block.day < day_limit, ["day", "sld", "complete"]].itertuples(index=False):
        if crosses_progression(sld, reference):
            return True
        if complete:
            reference = min(reference, sld)
    return False


def time_to_progression(block, origin_day, new_lesion_day, new_lesions="event",
                        drop_lowest=False):
    """Time from `origin_day` to RECIST 1.1 progression, or to censoring.

    Each assessment is compared with the smallest complete-assessment sum
    recorded before it, from baseline on, as RECIST 1.1 specifies; no later
    assessment enters the reference. An incomplete assessment (fewer target
    lesions measured than at baseline) cannot set the reference but can show
    progression. Only assessments on or after the origin can be events.

    new_lesions  "event": a new lesion on or after the origin is progression;
                 "censor": follow-up ends, censored, on the day a new lesion
                 appears (the sum-only definition).
    drop_lowest  remove the patient's lowest complete sum before scoring, so
                 that no single low reading sets the reference.
    Returns {"duration", "event", "source"}, or None without an assessment
    on or after the origin.
    """
    block = block.sort_values("day")
    if drop_lowest and block.complete.any():
        block = block.drop(index=block.loc[block.complete, "sld"].idxmin())
    forward = block.loc[block.day >= origin_day]
    if forward.empty:
        return None
    reference, sld_event_day = np.inf, np.nan
    for day, sld, complete in block[["day", "sld", "complete"]].itertuples(index=False):
        if day >= origin_day and crosses_progression(sld, reference):
            sld_event_day = float(day)
            break
        if complete:
            reference = min(reference, sld)
    last_seen = float(forward.day.max())
    lesion_day = (float(new_lesion_day) if pd.notna(new_lesion_day) and new_lesion_day >= origin_day
                  else np.nan)
    if new_lesions == "censor":
        if np.isfinite(sld_event_day) and not (np.isfinite(lesion_day) and lesion_day < sld_event_day):
            return {"duration": sld_event_day - origin_day, "event": True, "source": "SLD"}
        end = min(lesion_day, last_seen) if np.isfinite(lesion_day) else last_seen
        return {"duration": end - origin_day, "event": False, "source": None}
    candidates = {source: day for source, day in (("new lesion", lesion_day), ("SLD", sld_event_day))
                  if np.isfinite(day)}
    if candidates:
        source = min(candidates, key=candidates.get)
        return {"duration": candidates[source] - origin_day, "event": True, "source": source}
    return {"duration": last_seen - origin_day, "event": False, "source": None}


def burden_features(block, cutoff_day, inclusive=False):
    """Depth, final-interval slope of log burden, and an indicator that burden
    rose over the final interval, from complete assessments before the cutoff
    (on or before it when `inclusive`). None with fewer than two."""
    window = block.loc[(block.day <= cutoff_day) if inclusive else (block.day < cutoff_day)]
    window = window.loc[window.complete].sort_values("day")
    if len(window) < 2 or window.sld.iloc[0] <= 0:
        return None
    baseline, last, previous = (float(window.sld.iloc[0]), float(window.sld.iloc[-1]),
                                float(window.sld.iloc[-2]))
    span = max(float(window.day.iloc[-1] - window.day.iloc[-2]), 1.0)
    return {
        "depth_at_landmark": 1.0 - float(window.sld.min()) / baseline,
        "slope_to_landmark": (np.log(max(last, 1.0)) - np.log(max(previous, 1.0))) / span,
        "already_rising": float(last > previous),
        "log_baseline_sld": float(np.log(max(baseline, 1.0))),
        "n_assessments": len(window),
        "last_feature_day": float(window.day.iloc[-1]),
        "slope_window_days": span,
    }


def _outcome_columns(outcome, prefix):
    if outcome is None:
        return {f"{prefix}duration": np.nan, f"{prefix}event": False, f"{prefix}source": None}
    return {f"{prefix}{key}": value for key, value in outcome.items()}


def score_patients(panel, new_lesion_days, landmark_day, study, alternative_new_lesion_days=None):
    """Landmark and nadir-clock outcomes and features for every patient of one study.

    `panel` has one row per assessment (patient, day, sld, complete);
    `new_lesion_days` gives each patient's first new-lesion day (absent =
    none). Outcomes from the landmark: the primary (new lesions are events),
    sum only (censored at a new lesion) and with the lowest complete sum
    removed. The nadir clock starts at the lowest complete sum, with features
    from assessments on or before it. With `alternative_new_lesion_days`, the
    primary outcome and the pre-landmark progression check are repeated with
    that definition of a new lesion (prefix `harmonized_`).
    """
    rows = []
    for patient, block in panel.sort_values("day").groupby("patient", sort=True):
        new_lesion_day = new_lesion_days.get(patient, np.nan)
        features = burden_features(block, landmark_day)
        row = {"study": study, "patient": patient, "assessments": len(block),
               "last_day": float(block.day.max()),
               "reached_landmark": bool(block.day.max() >= landmark_day),
               "progressed_before_landmark": progressed_before(block, landmark_day, new_lesion_day),
               "has_features": features is not None, **(features or {})}
        row.update(_outcome_columns(time_to_progression(block, landmark_day, new_lesion_day), ""))
        row.update(_outcome_columns(time_to_progression(block, landmark_day, new_lesion_day,
                                                        new_lesions="censor"), "sld_only_"))
        row.update(_outcome_columns(time_to_progression(block, landmark_day, new_lesion_day,
                                                        drop_lowest=True), "nadir_excluded_"))
        complete = block.loc[block.complete]
        if len(complete):
            nadir_day = float(complete.loc[complete.sld.idxmin(), "day"])
            row["nadir_day"] = nadir_day
            row.update(_outcome_columns(time_to_progression(block, nadir_day, new_lesion_day), "nadir_"))
            nadir_features = burden_features(block, nadir_day, inclusive=True)
            row.update({f"nadir_{key}": value for key, value in (nadir_features or {}).items()})
        if alternative_new_lesion_days is not None:
            alternative = alternative_new_lesion_days.get(patient, np.nan)
            row["harmonized_progressed_before_landmark"] = progressed_before(block, landmark_day, alternative)
            row.update(_outcome_columns(time_to_progression(block, landmark_day, alternative), "harmonized_"))
        rows.append(row)
    return pd.DataFrame(rows)


def harrell_c(durations, events, risk, strata=None):
    """Harrell's concordance index. A pair is comparable when the member with
    the shorter time had an event; tied risks count one half. With `strata`,
    only pairs within the same stratum are compared."""
    durations = np.asarray(durations, float)
    events = np.asarray(events, bool)
    risk = np.asarray(risk, float)
    same = (np.ones((len(durations), len(durations)), bool) if strata is None
            else np.asarray(strata)[:, None] == np.asarray(strata)[None, :])
    comparable = (durations[None, :] > durations[:, None]) & events[:, None] & same
    n_comparable = comparable.sum()
    if n_comparable == 0:
        return np.nan
    concordant = ((risk[:, None] > risk[None, :]) & comparable).sum()
    tied = ((risk[:, None] == risk[None, :]) & comparable).sum()
    return float((concordant + 0.5 * tied) / n_comparable)


def variance_inflation(frame, predictors):
    """Variance inflation factor of each predictor: 1 / (1 − R²) from the
    least-squares regression of that predictor on the others (with an
    intercept). Values above about 10 mean the predictor is close to a
    linear function of the others, so its Cox coefficient is unstable. A
    predictor that is constant in `frame` has no defined factor (NaN)."""
    design = frame[predictors].to_numpy(float)
    factors = {}
    for index, name in enumerate(predictors):
        target = design[:, index]
        total = np.sum((target - target.mean()) ** 2)
        if total <= 0:
            factors[name] = np.nan
            continue
        others = np.column_stack([np.ones(len(design)), np.delete(design, index, axis=1)])
        fitted = others @ np.linalg.lstsq(others, target, rcond=None)[0]
        r2 = 1.0 - np.sum((target - fitted) ** 2) / total
        factors[name] = float(1.0 / max(1.0 - r2, 1e-12))
    return factors


def _standardize(train, test):
    """Training-set centring and scaling; constant columns are left out."""
    centre, scale = train.mean(axis=0), train.std(axis=0)
    keep = scale > 1e-12
    return (train[:, keep] - centre[keep]) / scale[keep], (test[:, keep] - centre[keep]) / scale[keep], keep


def cox_out_of_fold_risk(design, durations, events, fold_labels, seed, cox_strata=None):
    """Cross-validated Cox linear predictors (five folds stratified on
    `fold_labels`). Predictors are standardized with training-fold
    statistics; a predictor constant within a training fold is left out of
    that fold's fit. Returns risk scores, each patient's fold and the
    per-fold coefficients (NaN where a predictor was left out)."""
    risk = np.full(len(durations), np.nan)
    fold_of = np.full(len(durations), -1)
    coefficients = []
    splitter = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=seed)
    for fold, (train, test) in enumerate(splitter.split(design, fold_labels)):
        train_design, test_design, keep = _standardize(design[train], design[test])
        fitted = sm.PHReg(durations[train], train_design, status=events[train].astype(int), ties="efron",
                          strata=None if cox_strata is None else cox_strata[train]).fit()
        params = np.asarray(fitted.params, float)
        risk[test] = test_design @ params
        fold_of[test] = fold
        full = np.full(design.shape[1], np.nan)
        full[keep] = params
        coefficients.append(full)
    return risk, fold_of, np.vstack(coefficients)


def paired_bootstrap_c(durations, events, risk_a, risk_b, resample_strata, c_strata,
                       seed, min_events):
    """Paired bootstrap of C(risk_b) − C(risk_a), resampling patients within
    each stratum of `resample_strata`; resamples with fewer than `min_events`
    events are skipped. Returns the 95% interval, two-sided P and the number
    of resamples used."""
    rng = np.random.default_rng(seed + 7)
    index_by_stratum = [np.flatnonzero(resample_strata == s) for s in np.unique(resample_strata)]
    differences = []
    for _ in range(N_BOOTSTRAP):
        drawn = np.concatenate([rng.choice(idx, size=len(idx), replace=True)
                                for idx in index_by_stratum])
        if events[drawn].sum() < min_events:
            continue
        strata = None if c_strata is None else c_strata[drawn]
        c_a = harrell_c(durations[drawn], events[drawn], risk_a[drawn], strata)
        c_b = harrell_c(durations[drawn], events[drawn], risk_b[drawn], strata)
        if np.isfinite(c_a) and np.isfinite(c_b):
            differences.append(c_b - c_a)
    differences = np.asarray(differences)
    low, high = np.percentile(differences, [2.5, 97.5])
    return float(low), float(high), bootstrap_p_two_sided(differences), len(differences)


def _survival_arrays(frame, duration, event):
    """Durations (a time of 0 or less becomes 0.5 day, so that a patient who
    progressed at the clock start stays in the cohort), events and study codes."""
    durations = frame[duration].to_numpy(float)
    durations = np.where(durations <= 0, 0.5, durations)
    return durations, frame[event].to_numpy(bool), pd.Categorical(frame.study.astype(str)).codes


def compare_cox_models(frame, comparator, trajectory, seed, duration="duration",
                       event="event", stratified=False, label="", min_events=5):
    """Depth-only against depth-plus-shape Cox models on the same patients and
    folds, scored by out-of-fold concordance.

    Folds are stratified on event status and, with more than one study, on
    study; bootstrap resamples are drawn within study. With `stratified`,
    the Cox model has a baseline hazard per study and the concordance index
    compares patients only within the same study. Also returns the
    concordance by fold and by study as checks.
    """
    durations, events, study_codes = _survival_arrays(frame, duration, event)
    several_studies = len(np.unique(study_codes)) > 1
    fold_labels = (np.char.add(events.astype(int).astype(str), study_codes.astype(str))
                   if several_studies else events.astype(int))
    cox_strata = study_codes if stratified else None
    c_strata = study_codes if stratified else None
    risk_a, fold_of, coef_a = cox_out_of_fold_risk(frame[comparator].to_numpy(float), durations,
                                                   events, fold_labels, seed, cox_strata)
    risk_b, _, coef_b = cox_out_of_fold_risk(frame[trajectory].to_numpy(float), durations,
                                             events, fold_labels, seed, cox_strata)
    c_a = harrell_c(durations, events, risk_a, c_strata)
    c_b = harrell_c(durations, events, risk_b, c_strata)
    low, high, p_value, n_draws = paired_bootstrap_c(durations, events, risk_a, risk_b,
                                                     study_codes, c_strata, seed, min_events)
    checks = []
    for kind, labels in (("fold", fold_of), ("study", frame.study.astype(str).to_numpy())):
        for value in np.unique(labels):
            rows = labels == value
            strata = None if c_strata is None else c_strata[rows]
            checks.append({"by": kind, "value": value, "patients": int(rows.sum()),
                           "events": int(events[rows].sum()),
                           "c_depth": harrell_c(durations[rows], events[rows], risk_a[rows], strata),
                           "c_trajectory": harrell_c(durations[rows], events[rows], risk_b[rows], strata)})
    return {"analysis": label, "n": len(frame), "events": int(events.sum()),
            "c_depth": c_a, "c_trajectory": c_b, "difference": c_b - c_a,
            "low": low, "high": high, "p_value": p_value, "bootstrap_draws": n_draws,
            "coefficients_depth": pd.DataFrame(coef_a, columns=comparator),
            "coefficients": pd.DataFrame(coef_b, columns=trajectory), "checks": pd.DataFrame(checks)}


def leave_one_study_out(frame, comparator, trajectory, seed, duration="duration", event="event",
                        label="", min_events=5):
    """Each study's patients scored by Cox models fitted on the other studies.
    Concordance is computed within study and the paired bootstrap resamples
    within study, as for the stratified primary."""
    durations, events, study_codes = _survival_arrays(frame, duration, event)
    risk = {"depth": np.full(len(frame), np.nan), "trajectory": np.full(len(frame), np.nan)}
    by_study = []
    for held_out in np.unique(study_codes):
        train, test = study_codes != held_out, study_codes == held_out
        for name, predictors in (("depth", comparator), ("trajectory", trajectory)):
            design = frame[predictors].to_numpy(float)
            train_design, test_design, _ = _standardize(design[train], design[test])
            fitted = sm.PHReg(durations[train], train_design, status=events[train].astype(int),
                              ties="efron").fit()
            risk[name][test] = test_design @ np.asarray(fitted.params, float)
        by_study.append({"held_out": frame.study.astype(str).to_numpy()[test][0],
                         "patients": int(test.sum()), "events": int(events[test].sum()),
                         "c_depth": harrell_c(durations[test], events[test], risk["depth"][test]),
                         "c_trajectory": harrell_c(durations[test], events[test], risk["trajectory"][test])})
    c_a = harrell_c(durations, events, risk["depth"], study_codes)
    c_b = harrell_c(durations, events, risk["trajectory"], study_codes)
    low, high, p_value, n_draws = paired_bootstrap_c(durations, events, risk["depth"], risk["trajectory"],
                                                     study_codes, study_codes, seed, min_events)
    return {"analysis": label, "n": len(frame), "events": int(events.sum()),
            "c_depth": c_a, "c_trajectory": c_b, "difference": c_b - c_a,
            "low": low, "high": high, "p_value": p_value, "bootstrap_draws": n_draws,
            "checks": pd.DataFrame(by_study)}


def cox_full_fit(frame, predictors, duration="duration", event="event", stratified=False):
    """Cox model on all patients, predictors standardized: coefficient (log
    hazard per SD), standard error, 95% Wald interval and P value."""
    durations, events, study_codes = _survival_arrays(frame, duration, event)
    design = frame[predictors].to_numpy(float)
    standardized, _, keep = _standardize(design, design)
    fitted = sm.PHReg(durations, standardized, status=events.astype(int), ties="efron",
                      strata=study_codes if stratified else None).fit()
    params, errors = np.asarray(fitted.params, float), np.asarray(fitted.bse, float)
    return pd.DataFrame({"predictor": np.asarray(predictors)[keep], "coefficient": params,
                         "se": errors, "low": params - 1.96 * errors, "high": params + 1.96 * errors,
                         "p_value": np.asarray(fitted.pvalues, float)})


def report_comparison(result):
    """Print one depth-only against depth-plus-shape comparison: patients,
    events, both concordance indices and their difference with its 95%
    interval and paired-bootstrap P value."""
    print(f"  {result['analysis']}")
    print(f"    patients {result['n']:,}   events {result['events']:,}")
    print(f"    C, depth {result['c_depth']:.3f}   C, depth plus shape "
          f"{result['c_trajectory']:.3f}   difference "
          f"{format_ci(result['difference'], result['low'], result['high'], 1, 3, '')}"
          f"   {p_label(result['p_value'], 1 / result['bootstrap_draws'])}")
```

```python
# ==========================================================================
# 1.4 Self-tests of the notebook's own functions
#     pdxe_utils tests its own module (Section 1.1). These tests cover the
#     functions this notebook defines and that the reported results depend
#     on, each on a short synthetic series whose answer is known. A failure
#     stops the run here.
# ==========================================================================

NOTEBOOK_SELF_TESTS = []

# --- the durability endpoint (post_nadir_fit, summarize_regrowth) -----------
test_day = np.array([0.0, 3.0, 7.0, 10.0, 14.0])
test_log = np.array([5.0, 4.0, 3.5, 3.8, 4.1])            # nadir on day 7
nadir_left_out = post_nadir_fit(test_day, test_log, include_nadir=False)
nadir_included = post_nadir_fit(test_day, test_log, include_nadir=True)
still_shrinking = post_nadir_fit(test_day, np.array([5.0, 4.5, 4.0, 3.5, 3.0]))
NOTEBOOK_SELF_TESTS += [
    ("nadir found on day 7", nadir_left_out["nadir_index"] == 2 and nadir_left_out["nadir_day"] == 7.0),
    ("two measurements after the nadir", nadir_left_out["n_after_nadir"] == 2),
    ("nadir left out: two-point slope (4.1 - 3.8) / 4",
     np.isclose(nadir_left_out["post_nadir_growth_rate"], 0.075)),
    ("nadir left out: the fitted line passes through the day-10 reading",
     np.isclose(nadir_left_out["post_nadir_intercept"] + 0.075 * 3.0, 3.8)),
    ("nadir left out: Sxx of a two-point fit is (dt)^2 / 2", np.isclose(nadir_left_out["post_nadir_sxx"], 8.0)),
    ("nadir included: least-squares slope over days 7, 10 and 14",
     np.isclose(nadir_included["post_nadir_growth_rate"], np.polyfit([0.0, 3.0, 7.0], [3.5, 3.8, 4.1], 1)[0])),
    ("no slope when the nadir is the last reading",
     np.isnan(still_shrinking["post_nadir_growth_rate"]) and still_shrinking["n_after_nadir"] == 0),
    ("the regrowth summary uses the endpoint itself",
     np.isclose(summarize_regrowth(test_day, test_log)["post_nadir_growth_rate"],
                post_nadir_fit(test_day, test_log)["post_nadir_growth_rate"])),
]

# --- the window response rule (window_response) -----------------------------
window_volume = np.exp(np.array([5.0, 4.9, 4.8, 4.7]))
measured_day_10 = np.array([0.0, 3.0, 7.0, 10.0])
not_measured_day_10 = np.array([0.0, 3.0, 7.0])
NOTEBOOK_SELF_TESTS += [
    ("exact-day rule: no value without a reading on day 10",
     np.isnan(window_response(not_measured_day_10, window_volume[:3], 10, "exact_day"))),
    ("both rules agree when day 10 was measured",
     window_response(measured_day_10, window_volume, 10, "exact_day")
     == window_response(measured_day_10, window_volume, 10, "last_reading")),
    ("last-reading rule: running average at day 7, within the tolerance",
     np.isclose(window_response(not_measured_day_10, window_volume[:3], 10, "last_reading"),
                utils.best_average_response(not_measured_day_10, window_volume[:3], min_day=7.0))),
    ("last-reading rule: no value from a reading 7 days before the origin",
     np.isnan(window_response(np.array([0.0, 3.0]), window_volume[:2], 10, "last_reading"))),
]

# --- noise-only exceedance of the durability gap ---------------------------
# Two 3-point slopes 3.5 days apart (Sxx = 24.5) at sigma 0.112 differ by at
# least 0.02 per day about 53% of the time (the review's figure).
NOTEBOOK_SELF_TESTS += [
    ("noise-only exceedance for two 3-point slopes is about 0.53",
     abs(slope_difference_exceedance(24.5, 24.5, 0.112, 0.02) - 0.532) < 0.005),
]

# --- partial rank correlation equals the pandas-rank computation ------------
test_rng = np.random.default_rng(SEED)
x_test, z_test = test_rng.normal(size=60), test_rng.normal(size=60)
y_test = x_test + z_test + test_rng.normal(size=60)
ranked = pd.DataFrame({"x": x_test, "y": y_test, "z": z_test}).rank()
design_test = np.column_stack([np.ones(60), ranked.z])
residual_test = [ranked[c] - design_test @ np.linalg.lstsq(design_test, ranked[c], rcond=None)[0] for c in ("x", "y")]
NOTEBOOK_SELF_TESTS += [
    ("partial rank correlation matches the pandas-rank computation",
     np.isclose(partial_rank_correlation(x_test, y_test, z_test), np.corrcoef(*residual_test)[0, 1])),
]

# --- two-way bootstrap ------------------------------------------------------
two_way_draws = pigeonhole_bootstrap(np.repeat(list("abcde"), 4), np.tile(list("wxyz"), 5),
                                     lambda rows: float(np.mean(x_test[:20][rows])), 50, test_rng)
NOTEBOOK_SELF_TESTS += [
    ("two-way bootstrap returns finite replicates", len(two_way_draws) > 40 and np.isfinite(two_way_draws).all()),
]

# --- combination comparisons -------------------------------------------------
durability_toy = combination_durability_values(pd.DataFrame({
    "post_nadir_growth_rate": [0.03, np.nan, np.nan, 0.01], "n_after_nadir": [3, 0, 1, 2],
    "n_measurements": [8, 8, 8, 8]}))
toy_index = pd.MultiIndex.from_tuples([("M", "A"), ("M", "B"), ("M", "A + B")], names=["model_id", "treatment"])
toy_depth = pd.Series([10.0, -5.0, -30.0], index=toy_index)
toy_evaluation = combination_evaluations_table(toy_depth, pd.Series([0.03, 0.02, -np.inf], index=toy_index),
                                               True).iloc[0]
toy_strict = combination_evaluations_table(toy_depth, pd.Series([0.03, np.nan, 0.01], index=toy_index),
                                           True).iloc[0]
toy_loose = combination_evaluations_table(toy_depth, pd.Series([0.03, np.nan, 0.01], index=toy_index),
                                          False).iloc[0]
NOTEBOOK_SELF_TESTS += [
    ("durability values: rate, no regrowth (-inf), not estimable (NaN)",
     np.isclose(durability_toy[0], 0.03) and np.isneginf(durability_toy[1]) and np.isnan(durability_toy[2])
     and np.isclose(durability_toy[3], 0.01)),
    ("depth gain against the best component is 25 points", np.isclose(toy_evaluation.depth_gain, 25.0)),
    ("a combination that never regrew gains +inf against components that did",
     np.isposinf(toy_evaluation.durability_gain)),
    ("strict rule needs every component; the first-run rule takes the best available",
     np.isnan(toy_strict.durability_gain) and np.isclose(toy_loose.durability_gain, 0.02)),
    ("infinite values rank beyond the finite ones",
     list(np.argsort(finite_rank_values([1.0, np.inf, -np.inf, 2.0]))) == [2, 0, 3, 1]),
    ("median interval stays defined with infinite values",
     not np.isnan(extended_median_interval([0.01, np.inf, 0.02, -0.01, np.inf], 200, test_rng)).any()),
]

# --- compound-level effect with resampling weights --------------------------
cell_values = np.array([1.0, 3.0, 5.0])
cytostatic_test, comparator_test, model_test = np.array([0, 0, 1]), np.array([0, 1, 0]), np.array([0, 1, 1])
NOTEBOOK_SELF_TESTS += [
    ("compound effect with unit weights is the mean of the compound means (3.5)",
     np.isclose(weighted_compound_effect(cell_values, cytostatic_test, comparator_test, model_test,
                                         np.ones(2), np.ones(2), np.ones(2)), 3.5)),
    ("compound effect drops a PDX model with weight 0 (4.0)",
     np.isclose(weighted_compound_effect(cell_values, cytostatic_test, comparator_test, model_test,
                                         np.ones(2), np.ones(2), np.array([0.0, 2.0])), 4.0)),
]

# --- treatment-adjusted ridge R² ---------------------------------------------
# Outcome = treatment effect + 0.5 x feature + noise (SD 0.5): treatment means
# explain most of the variance, and the feature about half of what is left.
ridge_test = pd.DataFrame({"model_id": np.repeat(np.arange(30), 4).astype(str),
                           "treatment": np.tile(list("ABCD"), 30)})
ridge_test["feature"] = test_rng.normal(size=len(ridge_test))
ridge_test["outcome"] = (ridge_test.treatment.map({"A": -2.0, "B": 0.0, "C": 1.0, "D": 3.0})
                         + 0.5 * ridge_test.feature + test_rng.normal(0, 0.5, len(ridge_test)))
NOTEBOOK_SELF_TESTS += [
    ("treatment means alone explain most of the variance", treatment_mean_r2(ridge_test, "outcome") > 0.6),
    ("the feature explains within-treatment variance beyond the treatment means",
     0.3 < block_r2(ridge_test, ["feature"], "outcome", adjust_treatment=True) < 0.7),
]

# --- folds, paired selection contrasts, replicate scoring, censored arms ---
fold_test = pd.DataFrame({"model_id": np.repeat([f"M{i}" for i in range(12)], 3)})
folds_first, folds_second = grouped_folds(fold_test), grouped_folds(fold_test)
toy_selected = pd.DataFrame({"model_id": np.repeat(["M1", "M2", "M3"], 3),
                             "arm_id": [f"arm{i}" for i in range(9)],
                             "outcome_log_volume": [5.0, 5.5, 6.0, 4.0, 4.4, 3.9, 6.2, 6.0, 6.5],
                             "prediction_a": [5.2, 5.1, 6.1, 4.1, 4.5, 4.0, 6.0, 6.3, 6.4]})
toy_selected["prediction_b"] = toy_selected.prediction_a
toy_contrast = paired_selection_contrast(toy_selected, "prediction_a", "prediction_b", 50, test_rng)
toy_draws, toy_best_kept = replicate_outcome_regret(toy_selected, ["prediction_a"], 0.0, 5, test_rng)
toy_regret = selection_by_model(toy_selected, ["prediction_a"]).prediction_a_regret.mean()
toy_canonical = pd.DataFrame({
    "arm_id": ["A"] * 9 + ["B"] * 7, "model_id": "M", "tumor_type": "T", "treatment": ["x"] * 9 + ["y"] * 7,
    "is_control": False,
    "day": [0, 3, 7, 10, 14, 17, 21, 24, 28] + [0, 3, 7, 10, 14, 17, 20]})
toy_canonical["log_volume"] = 5.0 + np.where(toy_canonical.arm_id == "A", 0.02, 0.05) * toy_canonical.day
toy_canonical["volume"] = np.exp(toy_canonical.log_volume)
toy_scenarios, _ = censoring_scenario_table(toy_canonical, 10, 24, outcome_zeros="floor")
toy_scenarios = toy_scenarios.set_index("arm_id")
# fixed and random fold assignments
fold_assignment = model_fold_assignment(fold_test, folds_first)
folds_rebuilt = folds_from_assignment(fold_test, fold_assignment, len(folds_first))
fold_test_added = pd.concat([fold_test, pd.DataFrame({"model_id": ["M0", "M_new"]})], ignore_index=True)
folds_added = folds_from_assignment(fold_test_added, fold_assignment, len(folds_first))
added_assignment = model_fold_assignment(fold_test_added, folds_added)
folds_random = random_grouped_folds(fold_test, N_FOLDS, test_rng)
random_tests = np.concatenate([test for _, test in folds_random])
NOTEBOOK_SELF_TESTS += [
    ("grouped folds are reproducible and keep each PDX model in one fold",
     all(np.array_equal(a[1], b[1]) for a, b in zip(folds_first, folds_second))
     and all(set(fold_test.model_id.iloc[train]).isdisjoint(fold_test.model_id.iloc[test])
             for train, test in folds_first)),
    ("identical predictions: no regret reduction and no ranking difference",
     np.isclose(toy_contrast["regret_reduction"], 0.0) and np.isclose(toy_contrast["ranking_difference"], 0.0)),
    ("without replicate noise, regret against the replicate is the observed regret",
     np.allclose(toy_draws["prediction_a"], toy_regret) and np.allclose(toy_draws["observed best"], 0.0)
     and toy_best_kept == 1.0),
    ("an arm observed past day 24 is interpolated, in both scenarios",
     np.isclose(toy_scenarios.loc["A", "growth_stopped"], 5.0 + 0.02 * 24)
     and toy_scenarios.loc["A", "growth_stopped"] == toy_scenarios.loc["A", "growth_continued"]),
    ("an arm ended on day 20 stops at its last reading or continues at its slope",
     bool(toy_scenarios.loc["B", "ended_early"]) and np.isclose(toy_scenarios.loc["B", "growth_stopped"], 6.0)
     and np.isclose(toy_scenarios.loc["B", "growth_continued"], 6.0 + 4 * 0.05)),
    ("a fold assignment rebuilds the grouped folds it came from",
     all(np.array_equal(a[0], b[0]) and np.array_equal(a[1], b[1]) for a, b in zip(folds_first, folds_rebuilt))),
    ("adding arms keeps every PDX model in its fold; a new model joins the smallest fold",
     all(added_assignment[model] == fold for model, fold in fold_assignment.items())
     and "M_new" in added_assignment),
    ("random grouped folds cover every row once and never split a PDX model",
     np.array_equal(np.sort(random_tests), np.arange(len(fold_test)))
     and all(set(fold_test.model_id.iloc[train]).isdisjoint(fold_test.model_id.iloc[test])
             for train, test in folds_random)),
]

# --- tests added with the second round of review fixes ------------------------
# Noise-only exceedance with between-animal variation (Section 13.5).
NOTEBOOK_SELF_TESTS += [
    ("a between-animal SD of zero leaves the noise-only exceedance unchanged",
     np.isclose(slope_difference_exceedance(24.5, 24.5, 0.112, 0.02, between_sd=0.0),
                slope_difference_exceedance(24.5, 24.5, 0.112, 0.02))),
    ("between-animal variation alone gives 2Φ(−gap / (√2 τ))",
     np.isclose(slope_difference_exceedance(24.5, 24.5, 0.0, 0.02, between_sd=0.015),
                2 * stats.norm.sf(0.02 / (np.sqrt(2) * 0.015)))),
]
# Regrowth after a plateau at the floor (Section 4.6): floor on days 7-14,
# then relapse on days 17 and 21.
plateau_test = floor_plateau_fit(np.array([0.0, 3.0, 7.0, 10.0, 14.0, 17.0, 21.0]),
                                 np.array([5.0, 2.0, 0.0, 0.0, 0.0, 0.6, 1.2]))
above_floor_test = floor_plateau_fit(test_day, test_log)
NOTEBOOK_SELF_TESTS += [
    ("floor plateau: three readings over seven days",
     plateau_test["nadir_at_floor"] and plateau_test["plateau_readings"] == 3
     and np.isclose(plateau_test["plateau_days"], 7.0)),
    ("floor plateau: relapse slope from the readings after the plateau (0.6 / 4 per day)",
     np.isclose(plateau_test["post_plateau_growth_rate"], 0.15)),
    ("a nadir above the floor has no plateau",
     not above_floor_test["nadir_at_floor"] and np.isnan(above_floor_test["post_plateau_growth_rate"])),
]
# The per-group combination summary (Section 10.3).
gain_test = gain_summary([5.0, 7.0, -1.0, 9.0, 3.0], [0.01, np.inf, -0.002, 0.004, np.nan], 200, test_rng, 5)
NOTEBOOK_SELF_TESTS += [
    ("gain summary: median depth gain 5; four durability gains (NaN left out), three better and one worse",
     np.isclose(gain_test["median_depth_gain"], 5.0) and gain_test["evaluations"] == 4
     and gain_test["better_durability"] == 3 and gain_test["worse_durability"] == 1
     and np.isfinite(gain_test["depth_p"])),
]

print("1.4 NOTEBOOK SELF-TESTS")
for label, passed in NOTEBOOK_SELF_TESTS:
    print(f"  [{'PASS' if passed else 'FAIL'}] {label}")
failed_self_tests = [label for label, passed in NOTEBOOK_SELF_TESTS if not passed]
assert not failed_self_tests, f"notebook self-tests failed: {failed_self_tests}"
print(f"\n  all {len(NOTEBOOK_SELF_TESTS)} notebook self-tests pass")
```

## 2 · Data and preprocessing

Serial tumor volumes are read from the supplementary workbook of the Novartis PDX encyclopedia (Gao et al., 2015). Each PDX model received each treatment in a single animal, apart from a small number of arms with replicate readings on the same day, which are counted below. The workbook's SHA-256 checksum is compared with that of the published file before anything else is read. Treatment labels are normalized before arm identifiers are built, because one label in the workbook carries a stray quotation mark and would otherwise fail to join the curve-metric sheet.

Two preprocessing rules are then applied once, to one canonical table that every later section uses. Volumes below 1 mm³ (recorded as zero or below 1 mm³) are set to 1 mm³, and replicate readings of an arm on the same day are averaged on the log scale. An arm is one PDX model–treatment pair. The per-arm table is built by `pdxe_utils.arm_features` from the readings as recorded. That function floors volumes but keeps the first of replicate same-day readings instead of averaging them, so the nadir, the durability endpoint and the baseline are recomputed from the canonical table with the notebook's own function (`post_nadir_fit`), and every disagreement with the module's values is counted. The source study's own response metrics (best average response, treatment target) are then merged in by arm.

**Inputs:** `data/gao2015/41591_2015_BFnm3954_MOESM10_ESM.xlsx`.
**Outputs:** `raw_observations` (readings as recorded), `observations` (canonical: one row per arm per day), `arms` (one row per arm), `model_sets` (PDX models with each molecular data type).

```python
# ==========================================================================
# 2.1 Source workbook: checksum and conversion of the sheets to Parquet
#     The workbook must be the published file. Sheets are converted to
#     Parquet once and reused while the workbook checksum is unchanged.
# ==========================================================================

set_section("2")
assert WORKBOOK_PATH.exists(), (f"{relative(WORKBOOK_PATH)} not found; place the published workbook "
                                "there (see the project layout at the top of the notebook)")
workbook_sha256 = sha256_of_file(WORKBOOK_PATH)
print("SOURCE WORKBOOK")
print(f"  file      {relative(WORKBOOK_PATH)}")
print(f"  size      {WORKBOOK_PATH.stat().st_size / 2 ** 20:.1f} MiB")
print(f"  SHA-256   {workbook_sha256}")
assert workbook_sha256 == EXPECTED_WORKBOOK_SHA256, "the workbook is not the published file"
print("  matches the published file")

WORKBOOK_SHEETS = {"raw": "PCT raw data", "metrics": "PCT curve metrics",
                   "expression": "RNAseq_fpkm", "mutation": "pdxe_mut_and_cn2"}
COPY_NUMBER_SHEET = "copy number"     # only its header row (PDX model ids) is read
sheet_paths = {key: INTERIM_DIR / f"{key}.parquet" for key in WORKBOOK_SHEETS}
stamp_path = INTERIM_DIR / "workbook_sha256.txt"
cache_valid = (stamp_path.exists() and stamp_path.read_text().strip() == workbook_sha256
               and all(path.exists() for path in sheet_paths.values()))

if not cache_valid:
    for key, sheet in WORKBOOK_SHEETS.items():
        frame = pd.read_excel(WORKBOOK_PATH, sheet_name=sheet, engine="openpyxl")
        # Text columns hold mixed cell types in places; they are stored as text.
        for column in frame.columns[frame.dtypes == object]:
            frame[column] = frame[column].map(lambda v: None if pd.isna(v) else str(v))
        frame.to_parquet(sheet_paths[key], index=False)
    stamp_path.write_text(workbook_sha256)
sheets = {key: pd.read_parquet(path) for key, path in sheet_paths.items()}
print(f"\n  sheets {'read from the Parquet copies' if cache_valid else 'converted to Parquet'}")
for key, sheet in WORKBOOK_SHEETS.items():
    print(f"    {sheet:<20} {sheets[key].shape[0]:>9,} rows x {sheets[key].shape[1]:,} columns")

workbook = openpyxl.load_workbook(WORKBOOK_PATH, read_only=True, data_only=True)
copy_number_header = next(workbook[COPY_NUMBER_SHEET].iter_rows(max_row=1, values_only=True))
workbook.close()

# ==========================================================================
# 2.2 The canonical observation table
#     Columns are mapped by name, counts are asserted, and the floor and
#     same-day averaging are applied once (canonical_observations).
# ==========================================================================

RAW_COLUMNS = {"Model": "model_id", "Tumor Type": "tumor_type", "Treatment": "treatment",
               "Days Post T0": "day", "Volume (mm3)": "volume"}
utils.require_columns(sheets["raw"], list(RAW_COLUMNS), "PCT raw data")
raw_observations = sheets["raw"].rename(columns=RAW_COLUMNS)[list(RAW_COLUMNS.values())].copy()
for column in ("day", "volume"):
    raw_observations[column] = pd.to_numeric(raw_observations[column], errors="coerce")
non_numeric = int(raw_observations[["day", "volume"]].isna().any(axis=1).sum())
assert non_numeric == 0, f"{non_numeric} readings have a non-numeric day or volume"

# Treatment labels are normalized before the arm identifier is built: one
# label in the sheet carries a stray quotation mark.
raw_observations["treatment"] = raw_observations.treatment.map(utils.normalise_treatment)
raw_observations["arm_id"] = (raw_observations.model_id.astype(str) + " || "
                              + raw_observations.treatment.astype(str))
raw_observations["is_control"] = raw_observations.treatment.eq(CONTROL_LABEL)

# Tumor type is blank on a few rows; each PDX model has one tumor type, so a
# blank is filled from the same model's other rows.
missing_type = raw_observations.tumor_type.isna()
type_by_model = (raw_observations.loc[~missing_type].groupby("model_id").tumor_type
                 .agg(lambda values: values.mode().iloc[0]))
assert (raw_observations.loc[~missing_type].groupby("model_id").tumor_type.nunique() == 1).all()
filled_arms = sorted(raw_observations.loc[missing_type, "arm_id"].unique())
raw_observations["tumor_type"] = raw_observations.tumor_type.fillna(
    raw_observations.model_id.map(type_by_model))
assert raw_observations.tumor_type.notna().all(), "a PDX model has no recorded tumor type"
raw_observations = raw_observations.sort_values(["arm_id", "day"], kind="stable").reset_index(drop=True)

n_nonpositive = int((raw_observations.volume <= 0).sum())
n_below_floor = int(((raw_observations.volume > 0) & (raw_observations.volume < VOLUME_FLOOR)).sum())
floored_arms = raw_observations.loc[raw_observations.volume < VOLUME_FLOOR, "arm_id"].nunique()

observations = canonical_observations(raw_observations)
replicate_days = observations.loc[observations.n_readings > 1]
replicate_arms = replicate_days.groupby("arm_id").agg(
    days_with_replicates=("day", "size"), max_readings=("n_readings", "max"))

treatments = sorted(raw_observations.treatment.unique())
regimens = [t for t in treatments if t != CONTROL_LABEL]
combination_regimens = [t for t in regimens if "+" in t]

print("\nCANONICAL OBSERVATION TABLE")
print(f"  readings as recorded                 {record('n_readings', len(raw_observations)):,}")
print(f"  PDX models                           {record('n_models', raw_observations.model_id.nunique()):,}")
print(f"  tumor types                          {record('n_tumor_types', raw_observations.tumor_type.nunique())}"
      f"   {sorted(raw_observations.tumor_type.unique())}")
print(f"  arms                                 {record('n_arms', raw_observations.arm_id.nunique()):,}")
print(f"  untreated arms                       {record('n_untreated_arms', raw_observations.loc[raw_observations.is_control, 'arm_id'].nunique()):,}")
print(f"  treatment regimens                   {record('n_regimens', len(regimens))}"
      f"  ({record('n_single_agent_regimens', len(regimens) - len(combination_regimens))} single-agent, "
      f"{record('n_combination_regimens', len(combination_regimens))} combination)")
print(f"  rows with a blank tumor type         {int(missing_type.sum())}, in arm(s) {filled_arms}; filled from the PDX model")
print(f"  readings at or below 0 mm³           {n_nonpositive:,}")
print(f"  readings between 0 and 1 mm³         {n_below_floor:,}")
print(f"  arms with a reading set to the floor {floored_arms:,}")
print(f"  rows after same-day averaging        {len(observations):,}")
print(f"  arms with replicate same-day readings {record('n_arms_with_replicate_readings', len(replicate_arms)):,} "
      f"({int(replicate_arms.days_with_replicates.sum())} arm-days, up to "
      f"{int(replicate_arms.max_readings.max()) if len(replicate_arms) else 0} readings in a day)")
if len(replicate_arms):
    print(replicate_arms.sort_values("days_with_replicates", ascending=False).head(10).to_string())

assert len(raw_observations) == 67_445
assert raw_observations.arm_id.nunique() == 4_706
assert raw_observations.model_id.nunique() == 277
assert raw_observations.loc[raw_observations.is_control, "arm_id"].nunique() == 223
assert observations.groupby(["arm_id", "day"]).size().max() == 1
# The canonical table, for the figures notebook (example trajectories).
save_record(observations, "observations_canonical")

# ==========================================================================
# 2.3 The per-arm endpoint table and the source study's response metrics
#     pdxe_utils.arm_features receives each arm's readings as recorded and
#     returns nothing for arms that cannot support its features; those arms
#     are listed. The module floors volumes but keeps the FIRST of replicate
#     same-day readings instead of averaging them, so the nadir, the
#     durability endpoint and the baseline are recomputed below from the
#     canonical table with post_nadir_fit, the notebook's one durability
#     function, and compared with the module's values.
# ==========================================================================

utils_input = raw_observations.rename(columns={"tumor_type": "indication"})[
    ["model_id", "indication", "treatment", "day", "volume", "arm_id", "is_control"]]
feature_rows, arms_without_features = [], []
for arm_id, block in utils_input.groupby("arm_id", sort=False):
    row = utils.arm_features(block)
    if row is None:
        arms_without_features.append(arm_id)
    else:
        feature_rows.append(row)
# pdxe_utils uses compact column names; the notebook uses spelled-out ones,
# renamed once here so every later section reads one convention.
UTILITY_COLUMN_NAMES = {
    "indication": "tumor_type",
    "BAR_self": "response_reimplemented",           # best average response, recomputed
    "g_late": "post_nadir_growth_rate",              # durability: slope of log volume after the nadir
    "post_n": "n_after_nadir",
    "n_obs": "n_measurements",
    "duration": "duration_days",
    "baseline": "baseline_volume",
    "logV_baseline": "baseline_log_volume",
    "nadir_depth_log": "nadir_depth",
    "durable_estimable": "durability_estimable",
}
arms = pd.DataFrame(feature_rows).rename(columns=UTILITY_COLUMN_NAMES)
arms["treatment"] = arms.treatment.map(utils.normalise_treatment)
arms["is_control"] = arms.treatment.eq(CONTROL_LABEL)
utils.require_columns(arms, ["arm_id", "model_id", "tumor_type", "treatment",
                             "post_nadir_growth_rate", "interior_nadir", "n_after_nadir",
                             "nadir_day", "baseline_volume", "response_reimplemented",
                             "n_measurements", "duration_days"], "arm feature table")

print("\nARM FEATURE TABLE")
print(f"  arms with features                   {len(arms):,}")
print(f"  arms without features                {len(arms_without_features)}")
if arms_without_features:
    print(observations.loc[observations.arm_id.isin(arms_without_features)]
          .groupby("arm_id").agg(readings=("day", "size"), first_day=("day", "min"),
                                 last_day=("day", "max")).to_string())

# --------------------------------------------------------------------------
# The nadir, the durability endpoint and the baseline, recomputed from the
# canonical table (floored, same-day readings averaged on the log scale).
# Every durability quantity from here on comes from post_nadir_fit; the
# module's post-nadir growth rate is kept as post_nadir_growth_rate_utils,
# and every disagreement between the two is counted.
# --------------------------------------------------------------------------
recomputed_rows = []
for arm_id, block in arm_series(observations, set(arms.arm_id)):
    day = block.day.to_numpy(float)
    log_volume = block.log_volume.to_numpy(float)
    fit = post_nadir_fit(day, log_volume)
    recomputed_rows.append({"arm_id": arm_id, "baseline_volume": float(np.exp(log_volume[0])),
                            "baseline_log_volume": float(log_volume[0]),
                            "nadir_depth": float(log_volume[0] - log_volume.min()),
                            # Interior: the nadir is neither the first nor the last
                            # measurement, the definition pdxe_utils uses. (The
                            # analysis first run stored "nadir after the first
                            # measurement" under this name, which also counted the
                            # arms still shrinking at their last reading, so the
                            # comparison with the module below reported those 150
                            # arms as disagreements.) The post-nadir set, which needs
                            # at least two measurements after the nadir, is unchanged.
                            "interior_nadir": 0 < fit["nadir_index"] < len(day) - 1,
                            **{column: fit[column] for column in DURABILITY_COLUMNS}})
recomputed = pd.DataFrame(recomputed_rows).set_index("arm_id")
module_values = arms.set_index("arm_id")
rate_module = module_values.post_nadir_growth_rate.astype(float)
rate_notebook = recomputed.post_nadir_growth_rate.reindex(module_values.index)
both_rates = rate_module.notna() & rate_notebook.notna()
disagreements = pd.DataFrame({
    "rate defined in one only": rate_module.notna() != rate_notebook.notna(),
    "rate differs (> 1e-9 per day)": both_rates & ((rate_module - rate_notebook).abs() > 1e-9),
    "interior nadir differs": (module_values.interior_nadir.astype(bool)
                               != recomputed.interior_nadir.reindex(module_values.index).astype(bool)),
    # A value missing on both sides is agreement, not a difference.
    "measurements after the nadir differ": ~np.isclose(
        module_values.n_after_nadir.astype(float).to_numpy(),
        recomputed.n_after_nadir.reindex(module_values.index).astype(float).to_numpy(), equal_nan=True),
    "nadir day differs": ~np.isclose(
        module_values.nadir_day.astype(float).to_numpy(),
        recomputed.nadir_day.reindex(module_values.index).astype(float).to_numpy(), equal_nan=True),
    "baseline volume differs (> 1e-6 relative)": ((module_values.baseline_volume.astype(float)
                                                   / recomputed.baseline_volume.reindex(module_values.index) - 1)
                                                  .abs() > 1e-6),
})
durability_check = pd.DataFrame({
    "check": disagreements.columns,
    "all_arms": disagreements.sum().to_numpy(),
    "treated_arms": disagreements.loc[~module_values.is_control.astype(bool)].sum().to_numpy()})
any_disagreement = disagreements.any(axis=1)
record("n_arms_durability_differs_from_utils", int(any_disagreement.sum()))
record("n_treated_arms_durability_differs_from_utils",
       int(any_disagreement.loc[~module_values.is_control.astype(bool)].sum()))
print("\n  durability endpoint recomputed from the canonical table "
      f"(nadir {'included in' if DURABILITY_INCLUDES_NADIR else 'left out of'} the fit)")
print(durability_check.to_string(index=False))
if any_disagreement.any():
    print("  arms that disagree with pdxe_utils (first 10):")
    print(pd.DataFrame({"is_control": module_values.is_control, "rate_utils": rate_module,
                        "rate_notebook": rate_notebook}).loc[any_disagreement].head(10).to_string())

# Replace the module's durability columns with the recomputed ones. The
# module's durability_estimable flag uses a 10% regression threshold that
# the notebook does not use (the notebook's threshold is 1%), so it is
# dropped rather than left to be picked up by mistake.
replaced_columns = ["interior_nadir", "baseline_volume", "baseline_log_volume", "nadir_depth",
                    "durability_estimable", *DURABILITY_COLUMNS]
arms = (arms.assign(post_nadir_growth_rate_utils=arms.post_nadir_growth_rate)
        .drop(columns=[column for column in replaced_columns if column in arms.columns])
        .join(recomputed, on="arm_id"))
save_record(durability_check, "durability_definition_check")

METRIC_COLUMNS = {"BestAvgResponse": "best_average_response", "BestResponse": "best_response",
                  "Day_BestAvgResponse": "day_best_average_response",
                  "TimeToDouble": "time_to_double_published",
                  "ResponseCategory": "response_category", "Day_Last": "day_last",
                  "Treatment target": "treatment_target", "Treatment type": "treatment_type"}
metrics = sheets["metrics"].rename(columns={"Model": "model_id", "Treatment": "treatment"})
utils.require_columns(metrics, list(METRIC_COLUMNS), "PCT curve metrics")
metrics["treatment"] = metrics.treatment.map(utils.normalise_treatment)
metrics["arm_id"] = metrics.model_id.astype(str) + " || " + metrics.treatment.astype(str)
metrics = metrics.rename(columns=METRIC_COLUMNS)[["arm_id", *METRIC_COLUMNS.values()]]
duplicated_rows = metrics.duplicated("arm_id", keep=False)
disagreeing = metrics.loc[duplicated_rows].groupby("arm_id").nunique(dropna=False).gt(1).any(axis=1)
print(f"\n  curve-metric rows                    {len(metrics):,} for {metrics.arm_id.nunique():,} arms")
print(f"  arms listed more than once           {int(duplicated_rows.groupby(metrics.arm_id).any().sum()):,}"
      f" ({int(disagreeing.sum())} with differing values; the first row is used)")
arms = arms.merge(metrics.drop_duplicates("arm_id"), on="arm_id", how="left")
print(f"  arms matched to a published metric   {int(arms.best_average_response.notna().sum()):,} of {len(arms):,}")

both = arms.dropna(subset=["response_reimplemented", "best_average_response"])
agreement, _ = spearman(both.response_reimplemented, both.best_average_response)
print(f"  best average response, reimplemented against published: "
      f"Spearman {agreement:.3f} (n = {len(both):,}); the published value is used throughout")

# PDX models carrying each molecular data type (expression and copy number
# hold one column per model; the mutation sheet holds one row per finding).
known_models = set(raw_observations.model_id.astype(str))
model_sets = {
    "expression": known_models & set(map(str, sheets["expression"].columns[1:])),
    "copy number": known_models & {str(value) for value in copy_number_header[1:] if value is not None},
    "mutation": known_models & set(sheets["mutation"].iloc[:, 0].astype(str)),
}
```

## 3 · Design characteristics and attrition

This section describes what the design of the source study can support before any model is fitted. It covers the measurement schedules (which the simulations in Sections 5 and 12 reuse), arms with replicate same-day readings, and how many PDX models have an untreated arm. It also covers how the PDX model × treatment matrix is connected, how often treated arms show the durability phenotype, three signatures of termination at large volumes, and which PDX models have molecular data. It ends by fitting a line to log volume in every untreated arm with at least five measurements. The slope is that PDX model's untreated growth rate and the median residual SD is the measurement noise. Both are used from Section 4 onward and reported in Section 6. The median of per-arm residual SDs is below the typical (root-mean-square) noise, and a straight line overstates the noise in arms that curve, so the pooled residual SD of the linear and of the quadratic fits and the interpolation estimate of Section 13.2 (for untreated and for treated arms) are printed beside it. The median remains the value used downstream; Sections 4.5 and 8.4 repeat their analyses with the others.

**Inputs:** `observations`, `arms`, `model_sets`.
**Outputs:** `arm_design` (per-arm schedule summary), `empirical_grids` (measurement days of arms with ≥5 measurements), `regrowth` (model-free regrowth summary of treated arms), `untreated_fits`, `untreated_growth` (per PDX model), `MEASUREMENT_NOISE`, `noise_estimates_table`, `TREATED_NOISE`, `UNTREATED_RESIDUALS`; figure data `arm_design.csv` and `duration_quartiles.csv`.

```python
# ==========================================================================
# 3.1 Measurement schedules
#     The measurement days of every arm with at least five measurements are
#     kept as `empirical_grids`; the simulations draw schedules from them.
# ==========================================================================

set_section("3")
by_arm = observations.groupby("arm_id", sort=True)
arm_design = pd.DataFrame({
    "model_id": by_arm.model_id.first(), "tumor_type": by_arm.tumor_type.first(),
    "treatment": by_arm.treatment.first(), "is_control": by_arm.is_control.first(),
    "n_measurements": by_arm.size(), "first_day": by_arm.day.min(),
    "last_day": by_arm.day.max(), "last_volume": by_arm.volume.last(),
    "max_volume": by_arm.volume.max(),
    "median_gap_days": by_arm.day.agg(lambda d: float(np.median(np.diff(d))) if len(d) > 1 else np.nan),
})
arm_design["duration_days"] = arm_design.last_day - arm_design.first_day

print("3.1 MEASUREMENT SCHEDULES (canonical table, one measurement per arm per day)")
print(arm_design[["n_measurements", "duration_days", "median_gap_days"]]
      .describe(percentiles=[.05, .25, .5, .75, .95]).round(1).to_string())
print("\n  arms with at least k measurements")
for k in (3, 4, 5, 6, 8, 10):
    count = int((arm_design.n_measurements >= k).sum())
    print(f"    k = {k:>2}   {count:>5,}  ({100 * count / len(arm_design):.1f}%)")
print("\n  duration (days) by tumor type")
print(arm_design.groupby("tumor_type").duration_days
      .describe()[["count", "25%", "50%", "75%", "max"]].round(0).to_string())

empirical_grids = [block.day.to_numpy(float) for _, block in arm_series(observations)
                   if len(block) >= MIN_MEASUREMENTS_PER_ARM]
share_reaching_84 = float(np.mean([grid.max() >= 84 for grid in empirical_grids]))
print(f"\n  measurement schedules kept for simulation      {len(empirical_grids):,}")
print(f"  of those extending to day 84                   "
      f"{100 * record('share_schedules_to_day_84', share_reaching_84):.1f}%")

# ==========================================================================
# 3.2 Untreated-arm coverage and connectivity of the design
#     A treated arm can be paired with an untreated growth rate only when its
#     PDX model has an untreated arm. Connectivity is the number of connected
#     components of the bipartite PDX model–treatment graph.
# ==========================================================================

models_with_untreated = set(arm_design.loc[arm_design.is_control, "model_id"])
treated_design = arm_design.loc[~arm_design.is_control]
treated_without_untreated = int((~treated_design.model_id.isin(models_with_untreated)).sum())
print("\n3.2 UNTREATED-ARM COVERAGE AND CONNECTIVITY")
print(f"  PDX models with an untreated arm                {len(models_with_untreated)} of "
      f"{arm_design.model_id.nunique()}")
print(f"  treated arms in PDX models without one          {treated_without_untreated:,} of "
      f"{len(treated_design):,} ({100 * treated_without_untreated / len(treated_design):.1f}%)")

treatments_per_model = treated_design.groupby("model_id").treatment.nunique()
print(f"  treatments per PDX model, median [range]        {treatments_per_model.median():.0f} "
      f"[{treatments_per_model.min()}, {treatments_per_model.max()}]")
print(f"  PDX models with at least two treatments         {int((treatments_per_model >= 2).sum())}")
# Categorical codes are stored in the smallest integer type that holds them
# (int8 below 128 categories); they are widened before the node indices of
# the two node types are added, so the sum cannot overflow.
model_codes = pd.Categorical(treated_design.model_id).codes.astype(np.int64)
treatment_codes = pd.Categorical(treated_design.treatment).codes.astype(np.int64)
n_model_nodes = model_codes.max() + 1
graph = coo_matrix((np.ones(len(treated_design)), (model_codes, n_model_nodes + treatment_codes)),
                   shape=(n_model_nodes + treatment_codes.max() + 1,) * 2)
n_components, _ = connected_components(graph, directed=False)
print(f"  connected components of the design              {n_components}")
save_record(pd.crosstab(treated_design.tumor_type, treated_design.treatment),
            "tumor_type_by_treatment", index=True)

# ==========================================================================
# 3.3 Regrowth base rate (treated arms, canonical table)
#     The durability phenotype: shrank by at least 1%, an interior nadir with
#     at least two later measurements, and regrowth to at least 1.2 times the
#     nadir. The thresholds are varied in the grid below.
# ==========================================================================

regrowth_rows = []
for arm_id, block in arm_series(observations.loc[~observations.is_control]):
    summary = summarize_regrowth(block.day.to_numpy(float), block.log_volume.to_numpy(float))
    if summary is not None:
        regrowth_rows.append({"arm_id": arm_id, "tumor_type": block.tumor_type.iloc[0], **summary})
regrowth = pd.DataFrame(regrowth_rows).set_index("arm_id")
shrank = regrowth.nadir_ratio <= 1 - REGRESSION_FRACTION
interior = (regrowth.nadir_index > 0) & (regrowth.n_after_nadir >= MIN_POINTS_AFTER_NADIR)
regrew = regrowth.regrowth_multiple >= REGROWTH_MULTIPLE
regrowth["durability_phenotype"] = shrank & interior & regrew

print("\n3.3 REGROWTH BASE RATE (treated arms with at least three measurements)")
print(f"  arms                                            {len(regrowth):,}")
for label, mask in [("shrank by at least 1%", shrank),
                    ("interior nadir, >= 2 later measurements", interior),
                    ("regrew to >= 1.2 times the nadir", regrew),
                    ("all three (durability phenotype)", regrowth.durability_phenotype)]:
    print(f"  {label:<47} {int(mask.sum()):>5,}  ({100 * mask.mean():.1f}%)")
record("treated_phenotype_share", float(regrowth.durability_phenotype.mean()))
record("treated_interior_nadir_share", float(interior.mean()))

grid_rows = []
for regression in (0.0, 0.05, 0.10, 0.20, 0.30):
    for multiple in (1.10, 1.20, 1.50, 2.00):
        share = ((regrowth.nadir_ratio <= 1 - regression) & interior
                 & (regrowth.regrowth_multiple >= multiple)).mean()
        grid_rows.append({"minimum regression": regression, "minimum regrowth multiple": multiple,
                          "percent of arms": round(100 * share, 1)})
print("\n  percent of arms with the phenotype, by threshold")
print(pd.DataFrame(grid_rows).pivot(index="minimum regression",
                                    columns="minimum regrowth multiple",
                                    values="percent of arms").to_string())
print("\n  arms with the phenotype, by tumor type")
print(regrowth.groupby("tumor_type").durability_phenotype
      .agg(arms="size", with_phenotype="sum", share="mean").round(3).to_string())

# ==========================================================================
# 3.4 Termination at large volumes
#     Three signatures: arms ending at their largest volume, shorter arms in
#     faster-growing tumors, and larger final volumes in shorter arms.
# ==========================================================================

overall_rate = {}
for arm_id, block in arm_series(observations):
    if len(block) >= 3:
        overall_rate[arm_id] = fit_line(block.day.to_numpy(float) - block.day.iloc[0],
                                        block.log_volume.to_numpy(float))[0]
arm_design["overall_growth_rate"] = pd.Series(overall_rate)
complete_design = arm_design.dropna(subset=["overall_growth_rate"]).copy()
duration_growth_r = float(np.corrcoef(complete_design.duration_days,
                                      complete_design.overall_growth_rate)[0, 1])
complete_design["duration_quartile"] = pd.qcut(complete_design.duration_days, 4,
                                               labels=["Q1", "Q2", "Q3", "Q4"])
quartile_summary = complete_design.groupby("duration_quartile", observed=True).agg(
    days=("duration_days", lambda d: f"{d.min():.0f}-{d.max():.0f}"),
    mean_last_volume=("last_volume", "mean"), mean_growth_rate=("overall_growth_rate", "mean"),
    mean_measurements=("n_measurements", "mean"))

print("\n3.4 TERMINATION")
print(f"  arms whose last volume is their largest         "
      f"{100 * (arm_design.last_volume >= arm_design.max_volume).mean():.1f}%")
print("  final volume (mm³), arms per band")
for low, high in [(800, 1200), (1200, 1600), (1600, 2000), (2000, 2400), (2400, 3200)]:
    print(f"    {low:>5}-{high:<5} {int(arm_design.last_volume.between(low, high, inclusive='left').sum()):>6,}")
print(f"  Pearson r, duration against whole-arm growth rate  {duration_growth_r:.3f} "
      f"(n = {len(complete_design):,})")
print("  by duration quartile")
print(quartile_summary.round(4).to_string())
print("  untreated against treated arms")
print(arm_design.groupby("is_control")[["duration_days", "n_measurements", "last_volume"]]
      .mean().rename(index={False: "treated", True: "untreated"}).round(1).to_string())

# Data for the design and termination figure (drawn by the figures
# notebook): the per-arm schedule summary, with each arm's whole-arm growth
# rate, and the duration-quartile summary.
save_record(arm_design.reset_index(), "arm_design")
save_record(quartile_summary.reset_index(), "duration_quartiles")

# ==========================================================================
# 3.5 Molecular data coverage
# ==========================================================================

treated_models = set(treated_design.model_id.astype(str))
print("\n3.5 MOLECULAR DATA (PDX models in the response data)")
combinations_of_types = {
    "expression": model_sets["expression"], "copy number": model_sets["copy number"],
    "mutation": model_sets["mutation"],
    "all three": model_sets["expression"] & model_sets["copy number"] & model_sets["mutation"],
}
for label, model_ids in combinations_of_types.items():
    arms_covered = int(treated_design.model_id.astype(str).isin(model_ids).sum())
    print(f"  {label:<12} {len(model_ids):>4} PDX models   {arms_covered:>5,} treated arms")
expression_types = (arm_design.loc[arm_design.model_id.astype(str).isin(model_sets["expression"])]
                    .groupby("tumor_type").model_id.nunique())
print("  PDX models with expression data, by tumor type")
print(expression_types.reindex(sorted(arm_design.tumor_type.unique()), fill_value=0).to_string())

# ==========================================================================
# 3.6 Untreated growth and measurement noise
#     A least-squares line on log volume in each untreated arm with at least
#     five measurements. Slope: the PDX model's untreated growth rate.
#     Residual SD (two parameters removed): that arm's measurement noise.
#     A quadratic fit is recorded for the growth-law comparison in Section 6.
# ==========================================================================

untreated_rows, untreated_residuals = [], []
for arm_id, block in arm_series(observations.loc[observations.is_control]):
    if len(block) < MIN_MEASUREMENTS_PER_ARM:
        continue
    day = block.day.to_numpy(float) - block.day.iloc[0]
    log_volume = block.log_volume.to_numpy(float)
    slope, intercept, residual_sd = fit_line(day, log_volume)
    linear_residual = log_volume - (slope * day + intercept)
    # Residuals rescaled by √(n / (n − 2)), so that their variance estimates
    # σ² rather than σ²(n − 2)/n; Section 4.5 resamples them as measurement
    # errors.
    untreated_residuals.append(linear_residual * np.sqrt(len(day) / (len(day) - 2)))
    quadratic = np.polyfit(day, log_volume, 2)
    quadratic_residual = log_volume - np.polyval(quadratic, day)
    sxx = float(np.sum((day - day.mean()) ** 2))
    untreated_rows.append({
        "arm_id": arm_id, "model_id": block.model_id.iloc[0],
        "tumor_type": block.tumor_type.iloc[0], "n_measurements": len(block),
        "growth_rate": slope, "residual_sd": residual_sd,
        "quadratic_residual_sd": float(np.sqrt(np.sum(quadratic_residual ** 2) / (len(day) - 3))),
        "curvature": float(quadratic[0]),
        # Slope standard error and residual sums of squares, for the pooled
        # noise estimates below and the variance decomposition of Section 13.3.
        "sxx": sxx, "slope_se": residual_sd / np.sqrt(sxx),
        "rss_linear": float(np.sum(linear_residual ** 2)),
        "rss_quadratic": float(np.sum(quadratic_residual ** 2)),
    })
untreated_fits = pd.DataFrame(untreated_rows)
UNTREATED_RESIDUALS = np.concatenate(untreated_residuals)
assert untreated_fits.model_id.is_unique, "a PDX model has more than one untreated arm"
untreated_growth = untreated_fits.set_index("model_id").growth_rate
MEASUREMENT_NOISE = record("measurement_noise", float(untreated_fits.residual_sd.median()))
print("\n3.6 UNTREATED GROWTH AND MEASUREMENT NOISE")
print(f"  untreated arms with >= {MIN_MEASUREMENTS_PER_ARM} measurements          {len(untreated_fits)}")
print(f"  measurement noise (median residual SD)          {MEASUREMENT_NOISE:.4f} log units "
      f"({np.exp(MEASUREMENT_NOISE):.3f}-fold)")
print(f"  untreated growth rate, median [5th, 95th]       {untreated_growth.median():.4f} "
      f"[{untreated_growth.quantile(.05):.4f}, {untreated_growth.quantile(.95):.4f}] per day")

# Other estimates of the same measurement noise. The median of per-arm
# residual SDs is below the typical (root-mean-square) noise, and a straight
# line overstates the noise in the arms that curve (Section 6.1). Pooled RMS
# residual SDs from the linear and the quadratic fits, and the
# model-free interpolation estimate (Section 13.2's method) for untreated
# and treated arms, are printed beside it. MEASUREMENT_NOISE stays the value
# used downstream; Sections 4.5 and 8.4 repeat their analyses with the others.
rms_linear = float(np.sqrt(untreated_fits.rss_linear.sum() / (untreated_fits.n_measurements - 2).sum()))
rms_quadratic = float(np.sqrt(untreated_fits.rss_quadratic.sum() / (untreated_fits.n_measurements - 3).sum()))
noise_rng = np.random.default_rng(SEEDS["noise_estimates"])
interpolation_untreated = summarize_interpolation_noise(interpolation_noise(series_frames(
    observations.loc[observations.is_control], "arm_id", "day", "log_volume")), noise_rng)
interpolation_treated = summarize_interpolation_noise(interpolation_noise(series_frames(
    observations.loc[~observations.is_control], "arm_id", "day", "log_volume")), noise_rng)
# Treated-arm interpolation noise includes the curvature of regression and
# regrowth, so it is an upper value for the treated arms' measurement noise.
TREATED_NOISE = float(interpolation_treated["sigma"])
record("measurement_noise_rms_linear", rms_linear)
record("measurement_noise_rms_quadratic", rms_quadratic)
for tag, estimate in (("untreated", interpolation_untreated), ("treated", interpolation_treated)):
    record(f"measurement_noise_interp_{tag}", estimate["sigma"])
    record(f"measurement_noise_interp_{tag}_low", estimate["low"])
    record(f"measurement_noise_interp_{tag}_high", estimate["high"])
noise_estimates_table = pd.DataFrame([
    {"estimate": "median per-arm residual SD, linear fits (used downstream)", "sigma": MEASUREMENT_NOISE,
     "low": np.nan, "high": np.nan, "arms": len(untreated_fits)},
    {"estimate": "pooled RMS residual SD, linear fits", "sigma": rms_linear, "low": np.nan, "high": np.nan,
     "arms": len(untreated_fits)},
    {"estimate": "pooled RMS residual SD, quadratic fits", "sigma": rms_quadratic, "low": np.nan,
     "high": np.nan, "arms": len(untreated_fits)},
    {"estimate": "interpolation residuals, untreated arms", "sigma": interpolation_untreated["sigma"],
     "low": interpolation_untreated["low"], "high": interpolation_untreated["high"],
     "arms": interpolation_untreated["n_series"]},
    {"estimate": "interpolation residuals, treated arms (upper value)", "sigma": TREATED_NOISE,
     "low": interpolation_treated["low"], "high": interpolation_treated["high"],
     "arms": interpolation_treated["n_series"]},
])
print("\n  measurement noise, alternative estimates (log units)")
print(noise_estimates_table.round(4).to_string(index=False))
```

## 4 · Endpoints

Depth is the best average response published by the source study: from day 10 onward, the lowest value of the running average of percentage change in volume from baseline. Durability is the post-nadir growth rate, the least-squares slope of log volume over the measurements after the nadir (the smallest observed volume). The nadir itself is left out of the fit: it is selected as the minimum of noisy readings, so its error is biased downward and a line through it would overstate regrowth. With exactly two measurements after the nadir the slope is a two-point slope. Every durability value in the notebook comes from one function (`post_nadir_fit`); `DURABILITY_INCLUDES_NADIR` in Section 1.1 switches to the fit that includes the nadir, which Section 4.3 also reports as a sensitivity analysis. When the nadir is the first measurement this slope is the growth rate of the trajectory after its first measurement, so durability analyses use **post-nadir arms**: treated arms whose nadir comes after the first measurement and is followed by at least two measurements. Depth and durability are compared by Spearman correlation. The sensitivity analyses vary the minimum arm duration and the observation window, split by tumor type and by treatment, and adjust for the untreated growth rate of each PDX model. Treatments are then compared in pairs within each PDX model. A pair is depth-matched when best average response differs by less than 5 percentage points, and durability-discordant when post-nadir growth rates differ by at least 0.02 per day. Both thresholds are varied. Post-nadir arms whose post-nadir slope is not positive are left out of the pair analysis, because their nadir is not followed by regrowth.

The arms share PDX models and treatments, so P values that treat them as independent are far too small. Every correlation therefore also carries a 95% interval and a P value from resampling whole PDX models, and the headline correlation is resampled by treatment and by PDX model and treatment together (a two-way bootstrap). Section 4.3 adds a sweep over the number of points in the durability fit. Section 4.5 asks how much of the observed discordance, and of the low correlation, measurement noise alone would produce: it gives the share of depth-matched pairs that would be discordant if both arms of every pair had the same true regrowth rate, on each pair's real post-nadir schedule, and the reliability of the durability endpoint given each arm's slope error.

Section 4.6 (added after review) repeats the discordance among pairs whose two arms both responded (best average response at most 0%, −30% and −50%), with the noise-only expectation of 4.5 for each subset, because many depth-matched pairs hold arms that barely responded. It also counts arms held at the 1 mm³ floor for two or more readings before relapsing, whose post-nadir slope averages the flat readings with the relapse, and repeats the depth–durability correlation and the discordance with those arms left out and with their durability measured from the end of the plateau. The durability endpoint itself is unchanged.

All quantities here use the complete trajectory, so this section is descriptive. Forecasting is in Section 7.

**Inputs:** `arms`, `observations`, `untreated_growth`.
**Outputs:** `treated_with_endpoints`, `post_nadir_arms`, `depth_durability_sensitivity` (table `depth_durability_sensitivity.csv`), `model_pairs`, `rho_resampling`, `noise_baseline`, `attenuation`, `discordance_responders`, `floor_plateaus`, `floor_plateau_sensitivity`; figure data `arm_endpoints.csv` and `treatment_pairs.csv`.

```python
# ==========================================================================
# 4.1 Analysis sets and attrition
# ==========================================================================

set_section("4")
treated_arms = arms.loc[~arms.is_control].copy()
treated_with_endpoints = treated_arms.loc[treated_arms.best_average_response.notna()
                                          & treated_arms.post_nadir_growth_rate.notna()].copy()
post_nadir_arms = treated_with_endpoints.loc[
    treated_with_endpoints.interior_nadir.astype(bool)
    & (treated_with_endpoints.n_after_nadir >= MIN_POINTS_AFTER_NADIR)].copy()
post_nadir_ids = set(post_nadir_arms.arm_id)

attrition = pd.DataFrame([
    ("arms", observations.arm_id.nunique()),
    ("arms with an endpoint table entry", len(arms)),
    ("treated arms", len(treated_arms)),
    ("treated arms with both endpoints", len(treated_with_endpoints)),
    ("post-nadir arms", record("n_post_nadir_arms", len(post_nadir_arms))),
], columns=["stage", "arms"])
print("4.1 ATTRITION TO THE DURABILITY ANALYSIS SET")
print(attrition.to_string(index=False))
print(f"  PDX models with post-nadir arms                 {post_nadir_arms.model_id.nunique()}")

# ==========================================================================
# 4.2 Depth against durability
# ==========================================================================

rho, rho_p = spearman(post_nadir_arms.best_average_response, post_nadir_arms.post_nadir_growth_rate)
pearson_r, pearson_p = stats.pearsonr(post_nadir_arms.best_average_response,
                                      post_nadir_arms.post_nadir_growth_rate)
rho_all, rho_all_p = spearman(treated_with_endpoints.best_average_response,
                              treated_with_endpoints.post_nadir_growth_rate)
record("depth_durability_rho", rho)
record("depth_durability_p", rho_p)
record("depth_durability_shared_variance", rho ** 2)
print("\n4.2 DEPTH AGAINST DURABILITY")
print(f"  post-nadir arms (n = {len(post_nadir_arms):,})")
print(f"    Spearman ρ   {rho:.3f}   {p_label(rho_p)}   shared rank variance {100 * rho ** 2:.0f}%")
print(f"    Pearson r    {pearson_r:.3f}   {p_label(pearson_p)}")
print(f"  all treated arms with both endpoints (n = {len(treated_with_endpoints):,}), including arms "
      f"whose nadir is the first measurement")
print(f"    Spearman ρ   {rho_all:.3f}   {p_label(rho_all_p)}")

# The P values above treat the 1,832 arms as independent, but the arms sit in
# PDX models and in treatments that each hold many of them. The intervals
# and P values below resample whole PDX models, whole treatments, and both
# at once (pigeonhole bootstrap, Section 1.2). These are the ones to report;
# the P value of a bootstrap has resolution 1 / replicates.
rho_rng = np.random.default_rng(SEEDS["rho_bootstrap"])
depth_values = post_nadir_arms.best_average_response.to_numpy(float)
durability_values = post_nadir_arms.post_nadir_growth_rate.to_numpy(float)
rho_of_rows = lambda rows: spearman(depth_values[rows], durability_values[rows])[0]
rho_draws_by_scheme = {
    "PDX models": cluster_bootstrap(rows_by_cluster(post_nadir_arms.model_id.astype(str)),
                                    rho_of_rows, N_BOOTSTRAP, rho_rng),
    "treatments": cluster_bootstrap(rows_by_cluster(post_nadir_arms.treatment.astype(str)),
                                    rho_of_rows, N_BOOTSTRAP, rho_rng),
    "PDX models and treatments (two-way)": pigeonhole_bootstrap(
        post_nadir_arms.model_id, post_nadir_arms.treatment, rho_of_rows, N_BOOTSTRAP, rho_rng),
}
rho_resampling = pd.DataFrame([
    {"resampled": scheme, "replicates": len(draws), "spearman_rho": rho,
     "low": float(np.percentile(draws, 2.5)), "high": float(np.percentile(draws, 97.5)),
     "p_two_sided": bootstrap_p_two_sided(draws)}
    for scheme, draws in rho_draws_by_scheme.items()])
for tag, scheme in (("", "PDX models"), ("_treatments", "treatments"),
                    ("_twoway", "PDX models and treatments (two-way)")):
    row = rho_resampling.set_index("resampled").loc[scheme]
    record(f"depth_durability_rho{tag}_low", float(row.low))
    record(f"depth_durability_rho{tag}_high", float(row.high))
    record(f"depth_durability_p_cluster{tag}", float(row.p_two_sided))
print("  resampling instead of treating arms as independent")
for row in rho_resampling.itertuples(index=False):
    print(f"    {row.resampled:<38} ρ {row.spearman_rho:.3f} (95% CI {row.low:.3f} to {row.high:.3f})   "
          f"{p_label(row.p_two_sided, 1 / row.replicates)}")
save_record(rho_resampling, "depth_durability_resampling")

# ==========================================================================
# 4.3 Sensitivity analyses
# ==========================================================================

DEPTH, DURABILITY = "best_average_response", "post_nadir_growth_rate"
# Every row carries a 95% interval and a two-sided P value from resampling
# PDX models (clustered_correlation_row), beside the P value that treats the
# arms as independent. One generator serves every row, in the order below.
sensitivity_rng = np.random.default_rng(SEEDS["sensitivity_bootstrap"])
sensitivity_rows = []
for minimum_days in (0, 21, 28, 35, 42, 56):
    subset = post_nadir_arms.loc[post_nadir_arms.duration_days >= minimum_days]
    sensitivity_rows.append(clustered_correlation_row(
        "minimum arm duration", f">= {minimum_days} days", subset[DEPTH], subset[DURABILITY],
        subset.model_id, N_BOOTSTRAP, sensitivity_rng))
record("rho_duration_56", sensitivity_rows[-1]["spearman_rho"])
record("n_duration_56", sensitivity_rows[-1]["n_arms"])
record("p_duration_56", sensitivity_rows[-1]["p_value"])
record("rho_duration_56_low", sensitivity_rows[-1]["rho_low"])
record("rho_duration_56_high", sensitivity_rows[-1]["rho_high"])

# Minimum number of measurements in the durability fit. A slope fitted to
# more points is measured more precisely (Section 4.5), so if slope noise
# were what keeps ρ low, ρ would rise along this sweep; the duration sweep
# above shows it falling in the longer-followed arms.
for minimum_points in (2, 3, 4, 6):
    subset = post_nadir_arms.loc[post_nadir_arms.post_nadir_points >= minimum_points]
    sensitivity_rows.append(clustered_correlation_row(
        "minimum points in the durability fit", f">= {minimum_points} points", subset[DEPTH],
        subset[DURABILITY], subset.model_id, N_BOOTSTRAP, sensitivity_rng))
    record(f"rho_fit_points_{minimum_points}", sensitivity_rows[-1]["spearman_rho"])
    record(f"n_fit_points_{minimum_points}", sensitivity_rows[-1]["n_arms"])

# The durability fit under the other nadir rule (the nadir included in the
# slope when the primary leaves it out, and the reverse), on the same arms.
alternative_nadir_rule = not DURABILITY_INCLUDES_NADIR
alternative_rate = {arm_id: post_nadir_fit(block.day.to_numpy(float), block.log_volume.to_numpy(float),
                                           include_nadir=alternative_nadir_rule)["post_nadir_growth_rate"]
                    for arm_id, block in arm_series(observations, post_nadir_ids)}
post_nadir_arms["post_nadir_growth_rate_other_rule"] = post_nadir_arms.arm_id.map(alternative_rate)
sensitivity_rows.append(clustered_correlation_row(
    "durability fit", "nadir included" if alternative_nadir_rule else "nadir left out",
    post_nadir_arms[DEPTH], post_nadir_arms.post_nadir_growth_rate_other_rule, post_nadir_arms.model_id,
    N_BOOTSTRAP, sensitivity_rng))
record("rho_other_nadir_rule", sensitivity_rows[-1]["spearman_rho"])
record("rho_other_nadir_rule_low", sensitivity_rows[-1]["rho_low"])
record("rho_other_nadir_rule_high", sensitivity_rows[-1]["rho_high"])

# The same two endpoints recomputed with every arm truncated at a common day.
# Durability comes from summarize_regrowth, which now uses the endpoint's own
# fit (the analysis first run fitted these rows from the nadir on).
for window_days in (14, 21, 28, 35):
    window_rows = []
    for arm_id, block in arm_series(observations.loc[~observations.is_control]):
        block = block.loc[block.day <= window_days]
        summary = summarize_regrowth(block.day.to_numpy(float), block.log_volume.to_numpy(float))
        if (summary is None or summary["nadir_index"] == 0
                or summary["n_after_nadir"] < MIN_POINTS_AFTER_NADIR):
            continue
        response = utils.best_average_response(block.day.to_numpy(float), block.volume.to_numpy(float),
                                               min_day=MIN_DAY_FOR_BEST_AVERAGE_RESPONSE)
        if np.isfinite(response):
            window_rows.append({"model_id": block.model_id.iloc[0], "depth": response,
                                "durability": summary["post_nadir_growth_rate"]})
    window_values = pd.DataFrame(window_rows, columns=["model_id", "depth", "durability"])
    sensitivity_rows.append(clustered_correlation_row(
        "common observation window", f"first {window_days} days", window_values.depth,
        window_values.durability, window_values.model_id, N_BOOTSTRAP, sensitivity_rng))

for tumor_type, subset in post_nadir_arms.groupby("tumor_type"):
    sensitivity_rows.append(clustered_correlation_row(
        "tumor type", tumor_type, subset[DEPTH], subset[DURABILITY], subset.model_id,
        N_BOOTSTRAP, sensitivity_rng))
for treatment, subset in post_nadir_arms.groupby("treatment"):
    if len(subset) >= 25:
        # Within one treatment every arm is a different PDX model, so the
        # clustered interval resamples arms.
        sensitivity_rows.append(clustered_correlation_row(
            "treatment (>= 25 arms)", treatment, subset[DEPTH], subset[DURABILITY], subset.model_id,
            N_BOOTSTRAP, sensitivity_rng))

# Adjustment for the PDX model's untreated growth rate: rank-partial
# correlation, with the unadjusted correlation on the same arms beside it.
adjusted = post_nadir_arms.assign(untreated_growth=post_nadir_arms.model_id.map(untreated_growth)
                                  ).dropna(subset=["untreated_growth"])
sensitivity_rows.append(clustered_correlation_row(
    "untreated growth rate", "arms in PDX models with an untreated fit, unadjusted",
    adjusted[DEPTH], adjusted[DURABILITY], adjusted.model_id, N_BOOTSTRAP, sensitivity_rng))
unadjusted_rho = sensitivity_rows[-1]["spearman_rho"]
partial_rho = partial_rank_correlation(adjusted[DEPTH], adjusted[DURABILITY], adjusted.untreated_growth)
partial_df = len(adjusted) - 3
partial_p = float(2 * stats.t.sf(abs(partial_rho) * np.sqrt(partial_df / (1 - partial_rho ** 2)), partial_df))
adjusted_values = adjusted[[DEPTH, DURABILITY, "untreated_growth"]].to_numpy(float)
partial_draws = cluster_bootstrap(rows_by_cluster(adjusted.model_id.astype(str)),
                                  lambda rows: partial_rank_correlation(*adjusted_values[rows].T),
                                  N_BOOTSTRAP, sensitivity_rng)
sensitivity_rows.append({"analysis": "untreated growth rate",
                         "subset": "same arms, partial correlation adjusted for untreated growth",
                         "n_arms": len(adjusted), "spearman_rho": partial_rho, "p_value": partial_p,
                         "n_clusters": adjusted.model_id.nunique(),
                         "rho_low": float(np.percentile(partial_draws, 2.5)),
                         "rho_high": float(np.percentile(partial_draws, 97.5)),
                         "p_cluster": bootstrap_p_two_sided(partial_draws)})
record("partial_rho_untreated_growth", partial_rho)
record("partial_rho_untreated_growth_low", sensitivity_rows[-1]["rho_low"])
record("partial_rho_untreated_growth_high", sensitivity_rows[-1]["rho_high"])
record("unadjusted_rho_untreated_growth_subset", unadjusted_rho)

depth_durability_sensitivity = pd.DataFrame(sensitivity_rows)
treatment_rows = depth_durability_sensitivity.analysis.str.startswith("treatment")
depth_durability_sensitivity.loc[treatment_rows, "q_value"] = benjamini_hochberg(
    depth_durability_sensitivity.loc[treatment_rows, "p_value"])
# q values from the clustered P values. A bootstrap P value of zero is set to
# the resampling resolution (1 / replicates) before the adjustment, so no q
# value claims more precision than the resampling has.
depth_durability_sensitivity.loc[treatment_rows, "q_cluster"] = benjamini_hochberg(
    np.maximum(depth_durability_sensitivity.loc[treatment_rows, "p_cluster"].to_numpy(float), 1 / N_BOOTSTRAP))
by_type = depth_durability_sensitivity.loc[depth_durability_sensitivity.analysis == "tumor type"]
record("rho_tumor_type_min", by_type.spearman_rho.min())
record("rho_tumor_type_max", by_type.spearman_rho.max())

print("\n4.3 SENSITIVITY ANALYSES (interval and P_cluster resample PDX models)")
print(depth_durability_sensitivity.assign(
    spearman_rho=lambda d: d.spearman_rho.round(3),
    interval=lambda d: [f"{low:.3f} to {high:.3f}" if np.isfinite(low) else ""
                        for low, high in zip(d.rho_low, d.rho_high)],
    p_value=lambda d: d.p_value.map(format_p),
    p_cluster=lambda d: d.p_cluster.map(lambda p: format_p(p, 1 / N_BOOTSTRAP) if pd.notna(p) else ""),
    q_value=lambda d: d.q_value.map(lambda q: format_p(q) if pd.notna(q) else ""),
    q_cluster=lambda d: d.q_cluster.map(lambda q: format_p(q) if pd.notna(q) else "")
)[["analysis", "subset", "n_arms", "n_clusters", "spearman_rho", "interval", "p_value", "p_cluster",
   "q_value", "q_cluster"]].to_string(index=False))
print(f"\n  adjustment for untreated growth: {len(adjusted):,} arms in "
      f"{adjusted.model_id.nunique()} PDX models; ρ {unadjusted_rho:.3f} unadjusted, "
      f"{partial_rho:.3f} adjusted ({p_label(partial_p)}, {partial_df} df)")

# ==========================================================================
# 4.4 Treatment pairs within PDX models
# ==========================================================================

RESPONSE_TOLERANCE = 5.0     # percentage points of best average response
GROWTH_RATE_GAP = 0.02       # per day
PAIR_HORIZON_DAYS = 28       # days over which a rate gap is expressed as a fold change

positive_slope = post_nadir_arms.post_nadir_growth_rate > 0
pair_arms = post_nadir_arms.loc[positive_slope]
pair_rows = []
for model_id, block in pair_arms.groupby("model_id", sort=True):
    response = block.best_average_response.to_numpy(float)
    growth = block.post_nadir_growth_rate.to_numpy(float)
    pair_arm_ids = block.arm_id.to_numpy()
    for i, j in combinations(range(len(block)), 2):
        # The two arm identifiers are kept for the noise baseline of 4.5.
        pair_rows.append({"model_id": model_id, "arm_a": pair_arm_ids[i], "arm_b": pair_arm_ids[j],
                          "response_gap": abs(response[i] - response[j]),
                          "growth_rate_gap": abs(growth[i] - growth[j])})
model_pairs = pd.DataFrame(pair_rows)
depth_matched = model_pairs.response_gap < RESPONSE_TOLERANCE
discordant = depth_matched & (model_pairs.growth_rate_gap >= GROWTH_RATE_GAP)
discordant_share = discordant.sum() / depth_matched.sum()
pair_rows_by_model = rows_by_cluster(model_pairs.model_id)
matched_values, discordant_values = depth_matched.to_numpy(), discordant.to_numpy()
pair_draws = cluster_bootstrap(
    pair_rows_by_model,
    lambda rows: discordant_values[rows].sum() / max(matched_values[rows].sum(), 1),
    N_BOOTSTRAP, np.random.default_rng(SEEDS["pair_bootstrap"]))
pair_low, pair_high = np.percentile(pair_draws, [2.5, 97.5])

record("n_depth_matched_pairs", int(depth_matched.sum()))
record("n_discordant_pairs", int(discordant.sum()))
record("discordant_share", float(discordant_share))
record("fold_change_28_days", float(np.exp(GROWTH_RATE_GAP * PAIR_HORIZON_DAYS)))
print("\n4.4 TREATMENT PAIRS WITHIN PDX MODELS")
print(f"  post-nadir arms with a non-positive post-nadir slope, left out   "
      f"{int((~positive_slope).sum()):,} ({100 * (~positive_slope).mean():.1f}%)")
print(f"  arms in the pair analysis  {len(pair_arms):,} in {pair_arms.model_id.nunique()} PDX models")
print(f"  pairs                      {len(model_pairs):,}")
print(f"  depth-matched pairs        {int(depth_matched.sum()):,}")
print(f"  of which discordant        {int(discordant.sum()):,}  "
      f"{format_ci(discordant_share, pair_low, pair_high)}, interval resampling PDX models")
print(f"  a rate gap of {GROWTH_RATE_GAP} per day is a {np.exp(GROWTH_RATE_GAP * PAIR_HORIZON_DAYS):.2f}-fold "
      f"volume difference over {PAIR_HORIZON_DAYS} days after the nadir")

sweep_rows = []
for tolerance in (2.5, 5.0, 10.0):
    for gap in (0.01, 0.02, 0.03, 0.05):
        matched = model_pairs.response_gap < tolerance
        hits = matched & (model_pairs.growth_rate_gap >= gap)
        sweep_rows.append({"response tolerance (points)": tolerance, "rate gap (per day)": gap,
                           "fold change over 28 days": round(float(np.exp(gap * PAIR_HORIZON_DAYS)), 2),
                           "depth-matched pairs": int(matched.sum()),
                           "discordant pairs": int(hits.sum()),
                           "share of depth-matched": round(hits.sum() / max(matched.sum(), 1), 3)})
pair_sweep = pd.DataFrame(sweep_rows)
print("\n  threshold sweep")
print(pair_sweep.to_string(index=False))
save_record(depth_durability_sensitivity, "depth_durability_sensitivity")
save_record(pair_sweep, "pair_threshold_sweep")

# Data for the depth and durability figure (drawn by the figures notebook):
# every arm's endpoints with its analysis-set membership, and every
# within-model treatment pair of 4.4.
save_record(arms.assign(treated_with_endpoints=arms.arm_id.isin(treated_with_endpoints.arm_id),
                        post_nadir=arms.arm_id.isin(post_nadir_ids)), "arm_endpoints")
save_record(model_pairs, "treatment_pairs")
```

```python
# ==========================================================================
# 4.5 Discordance and attenuation expected from measurement noise alone
#     Two arms with the same true regrowth still differ in their fitted
#     post-nadir slopes, because each slope carries measurement error with SD
#     σ/√Sxx (Sxx from that arm's own fitted days). This cell asks how many
#     of the depth-matched pairs of 4.4 would be "discordant" (slopes 0.02
#     per day or more apart) if every pair had identical true rates, on each
#     pair's real post-nadir schedule, and how much slope error lowers the
#     depth–durability correlation of 4.2.
#
#     Noise scenarios: the measurement noise of Section 3.6 (median per-arm
#     residual SD), the pooled RMS residual SD, the treated-arm interpolation
#     estimate (an upper value: it includes regrowth curvature), and the
#     untreated linear-fit residuals resampled as errors (keeps their
#     heavier-than-normal tails). Gaussian scenarios use the exact
#     exceedance probability per pair; every scenario is also simulated at
#     the arm level, so that pairs sharing an arm share its error.
# ==========================================================================

N_NULL_REPLICATES = N_BOOTSTRAP
noise_baseline_rng = np.random.default_rng(SEEDS["noise_baseline"])
NOISE_SCENARIOS = {
    "median residual SD (Section 3.6)": ("median", MEASUREMENT_NOISE),
    "pooled RMS residual SD": ("rms", RESULTS["measurement_noise_rms_linear"]),
    "treated-arm interpolation (upper value)": ("treated", TREATED_NOISE),
    "untreated residuals, resampled": ("residuals", float(np.sqrt(np.mean(UNTREATED_RESIDUALS ** 2)))),
}

matched = model_pairs.loc[depth_matched].reset_index(drop=True)
observed_discordant = (matched.growth_rate_gap >= GROWTH_RATE_GAP).to_numpy()
assert np.isclose(observed_discordant.mean(), RESULTS["discordant_share"]), "4.4 pairs changed"
by_arm = pair_arms.set_index("arm_id")
pair_sxx = (matched.arm_a.map(by_arm.post_nadir_sxx).to_numpy(float),
            matched.arm_b.map(by_arm.post_nadir_sxx).to_numpy(float))
pair_min_points = np.minimum(matched.arm_a.map(by_arm.post_nadir_points).to_numpy(float),
                             matched.arm_b.map(by_arm.post_nadir_points).to_numpy(float))

# Arm-level layout for simulation: every arm in a matched pair, with the
# least-squares weights (t − mean t)/Sxx of its fitted days, so that a slope
# error is Σ weight × error over that arm's fitted measurements.
null_arms = sorted(set(matched.arm_a) | set(matched.arm_b))
arm_position = {arm_id: position for position, arm_id in enumerate(null_arms)}
weight_arm, weight_value = [], []
for arm_id, block in arm_series(observations, set(null_arms)):
    fit = post_nadir_fit(block.day.to_numpy(float), block.log_volume.to_numpy(float))
    start = fit["nadir_index"] + (0 if DURABILITY_INCLUDES_NADIR else 1)
    fitted_days = block.day.to_numpy(float)[start:]
    weight_arm += [arm_position[arm_id]] * len(fitted_days)
    weight_value += list((fitted_days - fitted_days.mean()) / np.sum((fitted_days - fitted_days.mean()) ** 2))
weight_arm, weight_value = np.asarray(weight_arm), np.asarray(weight_value)
null_sxx = np.array([by_arm.post_nadir_sxx[arm_id] for arm_id in null_arms])
index_a = matched.arm_a.map(arm_position).to_numpy()
index_b = matched.arm_b.map(arm_position).to_numpy()

baseline_rows, pair_null_probability = [], {}
matched_rows_by_model = rows_by_cluster(matched.model_id)
for scenario, (tag, sigma) in NOISE_SCENARIOS.items():
    # Simulated null shares: one draw of every arm's slope error per replicate.
    null_shares, pair_hits = np.empty(N_NULL_REPLICATES), np.zeros(len(matched))
    for replicate in range(N_NULL_REPLICATES):
        if tag == "residuals":
            errors = noise_baseline_rng.choice(UNTREATED_RESIDUALS, size=len(weight_value), replace=True)
            slope_errors = np.bincount(weight_arm, weights=weight_value * errors, minlength=len(null_arms))
        else:
            slope_errors = noise_baseline_rng.normal(0.0, 1.0, len(null_arms)) * sigma / np.sqrt(null_sxx)
        hits = np.abs(slope_errors[index_a] - slope_errors[index_b]) >= GROWTH_RATE_GAP
        null_shares[replicate] = hits.mean()
        pair_hits += hits
    # Per-pair null probability: exact for Gaussian errors, simulated for
    # resampled residuals. The interval on the excess resamples PDX models.
    probability = (pair_hits / N_NULL_REPLICATES if tag == "residuals"
                   else slope_difference_exceedance(*pair_sxx, sigma, GROWTH_RATE_GAP))
    pair_null_probability[tag] = probability
    excess_draws = cluster_bootstrap(
        matched_rows_by_model, lambda rows: observed_discordant[rows].mean() - probability[rows].mean(),
        N_BOOTSTRAP, noise_baseline_rng)
    expected = float(probability.mean())
    row = {"noise": scenario, "sigma": sigma, "depth_matched_pairs": len(matched),
           "observed_share": float(observed_discordant.mean()), "null_expected_share": expected,
           "null_simulated_mean": float(null_shares.mean()),
           "null_2_5": float(np.percentile(null_shares, 2.5)), "null_97_5": float(np.percentile(null_shares, 97.5)),
           "p_null_at_least_observed": (1 + np.sum(null_shares >= observed_discordant.mean())) / (1 + N_NULL_REPLICATES),
           "excess_over_noise": float(observed_discordant.mean() - expected),
           "excess_low": float(np.percentile(excess_draws, 2.5)),
           "excess_high": float(np.percentile(excess_draws, 97.5))}
    baseline_rows.append(row)
    record(f"discordance_null_expected_{tag}", expected)
    record(f"discordance_null_p_{tag}", float(row["p_null_at_least_observed"]))
    record(f"discordance_excess_{tag}", row["excess_over_noise"])
    record(f"discordance_excess_{tag}_low", row["excess_low"])
    record(f"discordance_excess_{tag}_high", row["excess_high"])
noise_baseline = pd.DataFrame(baseline_rows)

# The same comparison among pairs whose two slopes both rest on several
# points, where slope noise is smaller.
points_rows = []
for minimum_points in (2, 3, 4, 6):
    keep = pair_min_points >= minimum_points
    if keep.sum() < 20:
        continue
    for scenario, (tag, sigma) in NOISE_SCENARIOS.items():
        points_rows.append({"minimum_points_both_arms": minimum_points, "noise": scenario,
                            "pairs": int(keep.sum()), "observed_share": float(observed_discordant[keep].mean()),
                            "null_expected_share": float(pair_null_probability[tag][keep].mean())})
noise_baseline_by_points = pd.DataFrame(points_rows)
noise_baseline_by_points["excess_over_noise"] = (noise_baseline_by_points.observed_share
                                                 - noise_baseline_by_points.null_expected_share)

# The threshold sweep of 4.4 with the noise-only expectation beside each cell
# (Gaussian scenarios, exact per-pair probabilities).
pair_sxx_all = (model_pairs.arm_a.map(by_arm.post_nadir_sxx).to_numpy(float),
                model_pairs.arm_b.map(by_arm.post_nadir_sxx).to_numpy(float))
null_sweep_rows = []
for tolerance in (2.5, 5.0, 10.0):
    in_tolerance = (model_pairs.response_gap < tolerance).to_numpy()
    for gap in (0.01, 0.02, 0.03, 0.05):
        row = {"response tolerance (points)": tolerance, "rate gap (per day)": gap,
               "depth-matched pairs": int(in_tolerance.sum()),
               "observed share": float((model_pairs.growth_rate_gap.to_numpy()[in_tolerance] >= gap).mean())}
        for scenario, (tag, sigma) in NOISE_SCENARIOS.items():
            if tag != "residuals":
                row[f"null share, {tag}"] = float(slope_difference_exceedance(
                    pair_sxx_all[0][in_tolerance], pair_sxx_all[1][in_tolerance], sigma, gap).mean())
        null_sweep_rows.append(row)
null_sweep = pd.DataFrame(null_sweep_rows)

# Attenuation. With slope error variance σ²/Sxx per arm, the reliability of
# durability across post-nadir arms is 1 − mean(σ²/Sxx) / var(slopes), and the
# Pearson correlation corrected for that error is r / √reliability (depth's
# own error is ignored, so the correction is a lower bound on the true
# attenuation). A reliability near or below zero means slope noise alone
# accounts for the spread of durability in that set.
attenuation_rows = []
for minimum_points in (2, 3, 4, 6):
    subset = post_nadir_arms.loc[post_nadir_arms.post_nadir_points >= minimum_points]
    if len(subset) < 30:
        continue
    slope_variance = float(np.var(subset.post_nadir_growth_rate, ddof=1))
    subset_r = float(stats.pearsonr(subset.best_average_response, subset.post_nadir_growth_rate)[0])
    for scenario, (tag, sigma) in NOISE_SCENARIOS.items():
        error_variance = float(np.mean(sigma ** 2 / subset.post_nadir_sxx))
        reliability = 1.0 - error_variance / slope_variance
        corrected = subset_r / np.sqrt(reliability) if reliability > 0 else np.nan
        attenuation_rows.append({"minimum_points": minimum_points, "noise": scenario, "arms": len(subset),
                                 "slope_variance": slope_variance, "error_variance": error_variance,
                                 "reliability": reliability, "pearson_r": subset_r,
                                 "pearson_r_corrected": corrected if np.isfinite(corrected) and abs(corrected) <= 1
                                 else np.nan})
        if minimum_points == 2:
            record(f"durability_reliability_{tag}", reliability)
            record(f"depth_durability_pearson_corrected_{tag}", attenuation_rows[-1]["pearson_r_corrected"])
attenuation = pd.DataFrame(attenuation_rows)

print("4.5 DISCORDANCE EXPECTED FROM MEASUREMENT NOISE ALONE (identical true rates in every pair)")
print(f"  depth-matched pairs {len(matched):,}; discordant when slopes differ by >= {GROWTH_RATE_GAP} per day")
print(noise_baseline.round(4).to_string(index=False))
print("\n  by the number of points in the less-sampled arm's fit")
print(noise_baseline_by_points.round(4).to_string(index=False))
print("\n  threshold sweep of 4.4 with the noise-only expectation")
print(null_sweep.round(3).to_string(index=False))
print("\n  attenuation of the depth-durability correlation by slope error (Pearson, post-nadir arms)")
print(attenuation.round(4).to_string(index=False))
print("  pearson_r_corrected is left blank when the reliability is not positive or the corrected value "
      "exceeds 1:\n  slope noise then accounts for the spread of durability in that set.")
save_record(noise_baseline, "discordance_noise_baseline")
save_record(noise_baseline_by_points, "discordance_noise_baseline_by_points")
save_record(null_sweep, "discordance_null_sweep")
save_record(attenuation, "durability_attenuation")
```

```python
# ==========================================================================
# 4.6 Discordance among responders, and complete responses held at the floor
#     Two sensitivity analyses added after review; the durability endpoint
#     itself is unchanged.
#     (a) Responders. Many depth-matched pairs of 4.4 hold arms that barely
#         responded, in which the post-nadir slope is mostly growth. The
#         discordance of 4.4 and its noise-only expectation (4.5; σ from
#         Section 3.6, and the treated-arm upper value) are repeated among
#         pairs whose two arms BOTH reached a best average response of at
#         most 0%, −30% (a RECIST partial response) and −50%. Intervals
#         resample PDX models; the row of all pairs keeps the intervals of
#         4.4 and 4.5, so one estimate carries one interval.
#     (b) Plateaus at the floor. An arm held at 1 mm³ (a complete response by
#         caliper) for two or more readings and then relapsing gets a small
#         post-nadir slope, because the line averages the flat readings with
#         the relapse. Such arms are counted, and the depth–durability
#         correlation of 4.2 and the discordance of 4.4 are repeated with
#         them left out, and with their durability measured from the end of
#         the plateau instead (floor_plateau_fit, Section 1.2).
# ==========================================================================

# --- (a) discordance among responders -----------------------------------------
RESPONDER_THRESHOLDS = (0.0, -30.0, -50.0)       # best average response (%) that BOTH arms must reach
responder_rng = np.random.default_rng(SEEDS["responder_discordance_bootstrap"])
pair_lookup = post_nadir_arms.set_index("arm_id")
matched_pairs = model_pairs.loc[model_pairs.response_gap < RESPONSE_TOLERANCE].reset_index(drop=True)
# The shallower of the two responses decides whether a pair counts as a
# pair of responders (both arms at or below the threshold).
pair_worse_response = np.maximum(matched_pairs.arm_a.map(pair_lookup.best_average_response).to_numpy(float),
                                 matched_pairs.arm_b.map(pair_lookup.best_average_response).to_numpy(float))
pair_discordant = (matched_pairs.growth_rate_gap >= GROWTH_RATE_GAP).to_numpy()
pair_sxx_a = matched_pairs.arm_a.map(pair_lookup.post_nadir_sxx).to_numpy(float)
pair_sxx_b = matched_pairs.arm_b.map(pair_lookup.post_nadir_sxx).to_numpy(float)
# Noise-only probability that each pair is discordant (exact, Gaussian), as
# in 4.5: σ from Section 3.6 and the treated-arm interpolation upper value.
pair_null = {"median": slope_difference_exceedance(pair_sxx_a, pair_sxx_b, MEASUREMENT_NOISE, GROWTH_RATE_GAP),
             "treated": slope_difference_exceedance(pair_sxx_a, pair_sxx_b, TREATED_NOISE, GROWTH_RATE_GAP)}
assert np.isclose(pair_discordant.mean(), RESULTS["discordant_share"]), "the depth-matched pairs of 4.4 changed"

responder_rows = []
for threshold in (None, *RESPONDER_THRESHOLDS):
    keep = np.ones(len(matched_pairs), bool) if threshold is None else pair_worse_response <= threshold
    row = {"pairs": ("all depth-matched pairs (4.4)" if threshold is None
                     else f"both arms at or below {threshold:g}%"),
           "depth_matched_pairs": int(keep.sum()), "pdx_models": int(matched_pairs.model_id[keep].nunique()),
           "observed_share": float(pair_discordant[keep].mean()) if keep.any() else np.nan,
           "null_expected_share": float(pair_null["median"][keep].mean()) if keep.any() else np.nan,
           "null_expected_share_treated": float(pair_null["treated"][keep].mean()) if keep.any() else np.nan}
    row["excess_over_noise"] = row["observed_share"] - row["null_expected_share"]
    if threshold is None:
        # The full set is the analysis of 4.4 and 4.5: their intervals are
        # reused instead of drawing new ones.
        row.update(share_low=float(pair_low), share_high=float(pair_high),
                   excess_low=RESULTS["discordance_excess_median_low"],
                   excess_high=RESULTS["discordance_excess_median_high"])
    elif keep.sum() >= 10 and matched_pairs.model_id[keep].nunique() >= 5:
        # Positions are taken within the subset, so the cluster bootstrap
        # resamples the PDX models of the responder pairs only.
        subset_discordant, subset_null = pair_discordant[keep], pair_null["median"][keep]
        subset_rows = rows_by_cluster(matched_pairs.model_id[keep].to_numpy())
        share_draws = cluster_bootstrap(subset_rows, lambda rows: subset_discordant[rows].mean(),
                                        N_BOOTSTRAP, responder_rng)
        excess_draws = cluster_bootstrap(subset_rows,
                                         lambda rows: subset_discordant[rows].mean() - subset_null[rows].mean(),
                                         N_BOOTSTRAP, responder_rng)
        row.update(share_low=float(np.percentile(share_draws, 2.5)), share_high=float(np.percentile(share_draws, 97.5)),
                   excess_low=float(np.percentile(excess_draws, 2.5)),
                   excess_high=float(np.percentile(excess_draws, 97.5)))
    else:
        # Too few pairs or PDX models for an interval.
        row.update(share_low=np.nan, share_high=np.nan, excess_low=np.nan, excess_high=np.nan)
    responder_rows.append(row)
    if threshold is not None:
        tag = f"responders_{abs(int(threshold))}"
        record(f"n_depth_matched_pairs_{tag}", row["depth_matched_pairs"])
        record(f"discordant_share_{tag}", row["observed_share"])
        record(f"discordant_share_{tag}_low", row["share_low"])
        record(f"discordant_share_{tag}_high", row["share_high"])
        record(f"discordance_null_expected_{tag}", row["null_expected_share"])
        record(f"discordance_excess_{tag}", row["excess_over_noise"])
        record(f"discordance_excess_{tag}_low", row["excess_low"])
        record(f"discordance_excess_{tag}_high", row["excess_high"])
discordance_responders = pd.DataFrame(responder_rows)

print("4.6 SENSITIVITY OF THE DISCORDANCE AND OF DURABILITY")
print(f"  (a) depth-matched pairs (best average response within {RESPONSE_TOLERANCE:g} points) whose two arms both "
      "reached the threshold; intervals resample PDX models")
print(discordance_responders.round(4).to_string(index=False))

# --- (b) complete responses held at the floor -----------------------------------
# Every treated arm with at least three measurements, so that arms still at
# the floor at their last reading (no relapse seen) are counted as well.
plateau_rows = []
for arm_id, block in arm_series(observations.loc[~observations.is_control]):
    if len(block) < 3:
        continue
    plateau_rows.append({"arm_id": arm_id, "model_id": block.model_id.iloc[0],
                         "treatment": block.treatment.iloc[0],
                         **floor_plateau_fit(block.day.to_numpy(float), block.log_volume.to_numpy(float))})
floor_plateaus = pd.DataFrame(plateau_rows).set_index("arm_id")
floor_plateaus["post_nadir"] = floor_plateaus.index.isin(post_nadir_ids)
floor_plateaus["post_nadir_growth_rate"] = pair_lookup.post_nadir_growth_rate.reindex(floor_plateaus.index)
held_at_floor = floor_plateaus.plateau_readings >= 2
held_post_nadir = held_at_floor & floor_plateaus.post_nadir
held_ids = set(floor_plateaus.index[held_post_nadir])
record("n_floor_nadir_treated", int(floor_plateaus.nadir_at_floor.sum()))
record("n_floor_plateau_treated", int(held_at_floor.sum()))
record("n_floor_plateau_post_nadir", int(held_post_nadir.sum()))
record("n_floor_plateau_post_nadir_with_relapse_rate",
       int(floor_plateaus.loc[held_post_nadir, "post_plateau_growth_rate"].notna().sum()))

# Three versions of the post-nadir set: as analyzed (4.2), with the arms held
# at the floor left out, and with their durability measured from the end of
# the plateau (arms without at least two readings after it are left out).
plateau_rng = np.random.default_rng(SEEDS["floor_plateau_bootstrap"])
relapse_rate = floor_plateaus.post_plateau_growth_rate
held_mask = post_nadir_arms.arm_id.isin(held_ids)
plateau_variants = {
    "post-nadir arms (4.2, 4.4)": post_nadir_arms,
    "arms held at the floor left out": post_nadir_arms.loc[~held_mask],
    "arms held at the floor: durability from the end of the plateau": (
        post_nadir_arms.assign(post_nadir_growth_rate=np.where(held_mask, post_nadir_arms.arm_id.map(relapse_rate),
                                                               post_nadir_arms.post_nadir_growth_rate))
        .dropna(subset=["post_nadir_growth_rate"])),
}
plateau_sensitivity_rows = []
for label, frame in plateau_variants.items():
    variant_rho, _ = spearman(frame.best_average_response, frame.post_nadir_growth_rate)
    if label.startswith("post-nadir arms"):
        # The analysis of 4.2: its interval is reused.
        rho_low, rho_high = RESULTS["depth_durability_rho_low"], RESULTS["depth_durability_rho_high"]
    else:
        depth_array = frame.best_average_response.to_numpy(float)
        durability_array = frame.post_nadir_growth_rate.to_numpy(float)
        draws = cluster_bootstrap(rows_by_cluster(frame.model_id.astype(str)),
                                  lambda rows: spearman(depth_array[rows], durability_array[rows])[0],
                                  N_BOOTSTRAP, plateau_rng)
        rho_low, rho_high = float(np.percentile(draws, 2.5)), float(np.percentile(draws, 97.5))
    # Discordance by the rule of 4.4: arms with a positive post-nadir slope,
    # every pair within a PDX model, depth-matched within RESPONSE_TOLERANCE,
    # discordant at GROWTH_RATE_GAP.
    matched_count, discordant_count = 0, 0
    for _, block in frame.loc[frame.post_nadir_growth_rate > 0].groupby("model_id", sort=True):
        response = block.best_average_response.to_numpy(float)
        growth = block.post_nadir_growth_rate.to_numpy(float)
        for i, j in combinations(range(len(block)), 2):
            if abs(response[i] - response[j]) < RESPONSE_TOLERANCE:
                matched_count += 1
                discordant_count += int(abs(growth[i] - growth[j]) >= GROWTH_RATE_GAP)
    plateau_sensitivity_rows.append({
        "arms": label, "n_arms": len(frame), "pdx_models": frame.model_id.nunique(),
        "spearman_rho": variant_rho, "rho_low": rho_low, "rho_high": rho_high,
        "depth_matched_pairs": matched_count,
        "discordant_share": discordant_count / matched_count if matched_count else np.nan})
floor_plateau_sensitivity = pd.DataFrame(plateau_sensitivity_rows)
assert np.isclose(floor_plateau_sensitivity.spearman_rho.iloc[0], RESULTS["depth_durability_rho"])
assert np.isclose(floor_plateau_sensitivity.discordant_share.iloc[0], RESULTS["discordant_share"])
for tag, index in (("without_floor_plateaus", 1), ("floor_plateau_from_relapse", 2)):
    row = floor_plateau_sensitivity.iloc[index]
    record(f"rho_{tag}", float(row.spearman_rho))
    record(f"rho_{tag}_low", float(row.rho_low))
    record(f"rho_{tag}_high", float(row.rho_high))
    record(f"discordant_share_{tag}", float(row.discordant_share))

print(f"\n  (b) treated arms whose nadir is at the floor (1 mm³)            "
      f"{int(floor_plateaus.nadir_at_floor.sum()):,}")
print(f"      held at the floor for >= 2 readings                       {int(held_at_floor.sum()):,} "
      f"(post-nadir arms {int(held_post_nadir.sum()):,}; with >= 2 readings after the plateau "
      f"{RESULTS['n_floor_plateau_post_nadir_with_relapse_rate']:,})")
if held_post_nadir.any():
    print("      post-nadir arms held at the floor: plateau, endpoint slope and slope from the end of the plateau")
    print(floor_plateaus.loc[held_post_nadir, ["model_id", "treatment", "plateau_readings", "plateau_days",
                                               "readings_after_plateau", "post_nadir_growth_rate",
                                               "post_plateau_growth_rate"]]
          .sort_values("plateau_days", ascending=False).round(4).head(20).to_string())
print("\n      depth against durability and the discordance of 4.4, with and without these arms "
      "(interval resamples PDX models)")
print(floor_plateau_sensitivity.round(4).to_string(index=False))
save_record(discordance_responders, "discordance_responders")
save_record(floor_plateaus.reset_index(), "floor_plateau_arms")
save_record(floor_plateau_sensitivity, "floor_plateau_sensitivity")
```

## 5 · Mechanistic models and identifiability

Tumor volume is described by four models: exponential growth, constant kill (treatment lowers the growth rate g by a fixed kill rate k), and two models in which shrinkage is followed by regrowth. In the tolerant-subpopulation model a fraction f of the tumor does not respond: V(t) = V0[(1 − f)e^((g − k)t) + f·e^(gt)]. In the declining-effect model the whole tumor responds and the kill rate decays at rate δ: dV/dt = [g − k·e^(−δt)]V. The declining-effect model is not nested within the tolerant-subpopulation model, so the two are compared as competing explanations.

Section 5.1 measures how closely the declining-effect model can reproduce a tolerant-subpopulation trajectory on the study's own measurement schedules. Section 5.2 checks local identifiability from sensitivity matrices; condition numbers are given for the parameters as fitted and for relative sensitivities (∂ log V / ∂ log θ), which do not depend on the parameters' units. Section 5.3 tests whether the tolerant-subpopulation parameters can be recovered. Parameter sets are drawn from the priors of Section 5.6 and simulated on the study's measurement schedules, with measurement error of SD 0.15 log units (deliberately above the measured noise). Each simulated arm ends at its first measurement above 2,000 mm³, and the model is refitted by bounded least squares from three starting points. A parameter counts as recovered when the median absolute log error is below 0.25. Recovery is also reported among draws whose kill is strong enough for the tolerant fraction to be identifiable in principle ((k − g) × observed span ≥ 1, so that the sensitive population falls at least e-fold), because many draws from the HalfNormal(0.10) kill-rate prior give negligible kill; and the tolerant fraction is also scored by its absolute error on the logit and on the natural scale, because a log error exaggerates misses on a small fraction. Section 5.4 repeats this on regular designs that vary the measurement interval, duration and number of animals per arm. Additional animals are modeled as repeated measurements of one shared trajectory, so the design map is an upper bound on what replication can achieve. Section 5.5 simulates from each regrowth model, fits both, and keeps the model with the lower Akaike information criterion (AIC). In 5.5 the two generating models draw their kill rates from different distributions (centred on 0.12 and 0.14 per day), so part of what the criterion separates can be kill-rate magnitude rather than mechanism. Section 5.5b therefore repeats the comparison with paired replicates: baseline volume, growth rate, kill rate, schedule and noise vector are drawn once and shared by both mechanisms, so the mechanism is the only systematic difference, as in Section 12. Section 5.6 checks a Bayesian version of the tolerant-subpopulation model by prior predictive simulation and simulation-based calibration (Talts et al., 2018); that model is not fitted to observed data. Each calibration replicate runs one chain, so convergence is judged by split R̂ within the chain; the uniformity tests are repeated on the replicates whose split R̂ is at most 1.05 for every calibrated parameter and that had no divergent transitions, to show whether poorly mixing replicates drive the result.

The design map uses 150 replicates per design with common random numbers: within one interval and duration, every number of animals uses the same true parameters and the same standard-normal noise, scaled by σ/√animals, so that differences between animal counts are not Monte Carlo noise. The median absolute log errors and the upper bounds of their 95% intervals are printed beside the recovery calls, and a design is also marked as recovered with confidence when that upper bound is below the threshold. Replicates whose drawn schedule has too few measurements to simulate are counted and printed in 5.3, 5.5 and 5.5b.

**Inputs:** `empirical_grids`, `untreated_growth`, `MEASUREMENT_NOISE`.
**Outputs:** `GROWTH_PRIOR`, `approximation`, `local_identifiability`, `parameter_recovery`, `conditional_recovery`, `design_map`, `mechanism_discrimination`, `calibration`, `paired_discrimination`; figure data `approximation_example.csv` and `prior_predictive.csv`.

```python
# ==========================================================================
# 5.0 Priors shared by the simulations and the Bayesian model
#     The growth-rate prior is centred on the untreated arms: log-normal with
#     the median and 1.15 times the SD of the positive untreated log growth
#     rates, truncated at 0.20 per day (a doubling time of 3.5 days).
# ==========================================================================

set_section("5")
positive_untreated = untreated_growth.loc[untreated_growth > 0]
GROWTH_PRIOR = {
    "growth_centre": float(np.median(np.log(positive_untreated))),
    "growth_spread": float(1.15 * np.std(np.log(positive_untreated))),
    "growth_upper": 0.20,
}
upper_z = (np.log(GROWTH_PRIOR["growth_upper"]) - GROWTH_PRIOR["growth_centre"]) / GROWTH_PRIOR["growth_spread"]
prior_rate_quantiles = np.exp(GROWTH_PRIOR["growth_centre"] + GROWTH_PRIOR["growth_spread"]
                              * stats.truncnorm.ppf([0.025, 0.5, 0.975], -np.inf, upper_z))
print("5.0 PRIORS")
print(f"  untreated arms with a positive growth rate      {len(positive_untreated)} "
      f"(fastest {positive_untreated.max():.3f} per day)")
print(f"  log growth rate ~ TruncatedNormal({GROWTH_PRIOR['growth_centre']:.3f}, "
      f"{GROWTH_PRIOR['growth_spread']:.3f}), upper limit log {GROWTH_PRIOR['growth_upper']}")
print(f"  implied doubling time, median [95% range]       {np.log(2) / prior_rate_quantiles[1]:.1f} "
      f"[{np.log(2) / prior_rate_quantiles[2]:.1f}, {np.log(2) / prior_rate_quantiles[0]:.1f}] days")
print("  kill rate ~ HalfNormal(0.10); logit tolerant fraction ~ Normal(−2.2, 1.5) "
      f"(median fraction {1 / (1 + np.exp(2.2)):.3f}); log baseline ~ Normal(log 200, 0.5)")

# ==========================================================================
# 5.1 How closely the declining-effect model reproduces a tolerant-
#     subpopulation trajectory on the study's measurement schedules
#     The declining-effect model is fitted to noiseless tolerant-subpopulation
#     curves; the largest absolute residual is the closest the two models
#     come on that schedule. Two settings: one fixed parameter set on 200
#     schedules, and 200 parameter sets drawn from the priors.
# ==========================================================================

ILLUSTRATIVE_TOLERANT = {"log_baseline": np.log(200.0), "growth_rate": 0.05,
                         "kill_rate": 0.12, "tolerant_fraction": 0.05}
demonstration_rng = np.random.default_rng(SEEDS["mechanism_demonstration"])
declining_forward = constant_exposure_forward("declining drug effect")
schedules_56 = [grid[grid <= 56] for grid in empirical_grids if grid.max() >= 56]
approximation_rows = []
for setting in ("fixed parameters", "prior draws"):
    picks = demonstration_rng.choice(len(schedules_56), size=min(200, len(schedules_56)), replace=False)
    for index in picks:
        days = schedules_56[index]
        truth = (ILLUSTRATIVE_TOLERANT if setting == "fixed parameters"
                 else draw_prior_parameters(demonstration_rng, GROWTH_PRIOR))
        curve = log_volume_tolerant(days, truth["log_baseline"], truth["growth_rate"],
                                    truth["kill_rate"], truth["tolerant_fraction"])
        fit = fit_mechanism("declining drug effect", days, curve)
        residual = declining_forward(fit["x"], days) - curve
        approximation_rows.append({"setting": setting, "n_measurements": len(days),
                                   "max_abs_residual": float(np.max(np.abs(residual))),
                                   "rmse": float(np.sqrt(np.mean(residual ** 2)))})
approximation = pd.DataFrame(approximation_rows)
approximation_summary = approximation.groupby("setting").agg(
    schedules=("max_abs_residual", "size"),
    median_measurements=("n_measurements", "median"),
    median_max_abs_residual=("max_abs_residual", "median"),
    p90_max_abs_residual=("max_abs_residual", lambda v: v.quantile(0.9)),
    median_rmse=("rmse", "median"))
fixed_separation = record("approximation_max_residual",
                          float(approximation_summary.loc["fixed parameters", "median_max_abs_residual"]))
record("approximation_to_noise_ratio", fixed_separation / MEASUREMENT_NOISE)
print("\n5.1 DECLINING-EFFECT FIT TO TOLERANT-SUBPOPULATION CURVES (schedules up to day 56)")
print(approximation_summary.round(4).to_string())
print(f"  fixed parameters: median largest residual {fixed_separation:.4f} log units, "
      f"{fixed_separation / MEASUREMENT_NOISE:.2f} times the measurement noise")

# ==========================================================================
# 5.2 Local identifiability
#     Sensitivity matrices on a dense grid (a property of the models, not of
#     the schedules). Numerical rank uses a relative tolerance of 1e-8 times
#     the largest singular value.
# ==========================================================================

REPRESENTATIVE_PARAMETERS = {
    "constant kill": (log_volume_constant_kill, [np.log(100.0), 0.05, 0.10]),
    "tolerant subpopulation": (log_volume_tolerant, [np.log(100.0), 0.05, 0.10, 0.20]),
    "declining drug effect": (log_volume_declining, [np.log(100.0), 0.05, 0.15, 0.05]),
}
identifiability_rows = []
for window_days in (7, 14, 21, 28, 42, 56):
    days = np.linspace(0.0, float(window_days), max(5, window_days + 1))
    for model_name, (model, parameters) in REPRESENTATIVE_PARAMETERS.items():
        matrix = sensitivity_matrix(model, days, parameters)
        singular = np.linalg.svd(matrix, compute_uv=False)
        # Relative sensitivities: each rate or fraction column multiplied by
        # its parameter value (∂ log V / ∂ log θ), so that parameters on
        # different scales (a rate of 0.05 per day against a log volume of
        # 4.6) do not inflate the condition number. The log baseline is
        # already on a log scale and is left as it is.
        scale = np.where(np.arange(len(parameters)) == 0, 1.0, np.asarray(parameters, float))
        singular_scaled = np.linalg.svd(matrix * scale, compute_uv=False)
        identifiability_rows.append({
            "model": model_name, "window_days": window_days, "parameters": len(parameters),
            "numerical_rank": int(np.sum(singular > 1e-8 * singular.max())),
            "condition_number": float(singular.max() / singular.min()) if singular.min() > 0 else np.inf,
            "condition_number_scaled": (float(singular_scaled.max() / singular_scaled.min())
                                        if singular_scaled.min() > 0 else np.inf)})
local_identifiability = pd.DataFrame(identifiability_rows)
print("\n5.2 LOCAL IDENTIFIABILITY")
print(local_identifiability.pivot(index="window_days", columns="model",
                                  values="numerical_rank").to_string())
print("  condition number, parameters as fitted")
print(local_identifiability.pivot(index="window_days", columns="model", values="condition_number")
      .map(lambda v: f"{v:.3g}").to_string())
print("  condition number, relative sensitivities (d log V / d log parameter); report these")
print(local_identifiability.pivot(index="window_days", columns="model", values="condition_number_scaled")
      .map(lambda v: f"{v:.3g}").to_string())
print("  In the constant-kill model growth and kill rates enter only as g − k, so its rank is 2 of 3.")

# ==========================================================================
# 5.3 Recovery of the tolerant-subpopulation parameters on the study's schedules
#     Error-to-prior ratio: RMSE of the log error over the SD of the log true
#     values; below 1 the estimates carry information beyond the prior.
# ==========================================================================

RECOVERY_WINDOWS = (14, 21, 28, 42, 56)
N_RECOVERY_REPLICATES = 200
RECOVERED_PARAMETERS = ("growth_rate", "kill_rate", "tolerant_fraction")

recovery_rng = np.random.default_rng(SEEDS["parameter_recovery"])
recovery_rows, recovery_replicate_rows = [], []
for window_days in RECOVERY_WINDOWS:
    usable = [grid[grid <= window_days] for grid in empirical_grids if grid.max() >= window_days]
    log_errors = {name: [] for name in RECOVERED_PARAMETERS}
    log_truths = {name: [] for name in RECOVERED_PARAMETERS}
    terminated = schedules_too_short = ended_too_short = failed_fits = 0
    for _ in range(N_RECOVERY_REPLICATES):
        days = usable[recovery_rng.integers(len(usable))]
        if len(days) < 4:
            # A replicate whose drawn schedule has fewer than four measurements
            # is not simulated (no further random draws, so the stream of the
            # analysis first run is unchanged); it is counted here.
            schedules_too_short += 1
            continue
        truth = draw_prior_parameters(recovery_rng, GROWTH_PRIOR)
        noisy = (log_volume_tolerant(days, truth["log_baseline"], truth["growth_rate"],
                                     truth["kill_rate"], truth["tolerant_fraction"])
                 + recovery_rng.normal(0, SIMULATION_NOISE_SD, size=len(days)))
        fit_days, noisy, stopped = apply_termination(days, noisy)
        terminated += stopped
        if len(fit_days) < 4:
            ended_too_short += 1
            continue
        fit = fit_mechanism("tolerant subpopulation", fit_days, noisy)
        if not np.isfinite(fit["rss"]):
            failed_fits += 1
            continue
        estimate = tolerant_estimates(fit["x"])
        for name in RECOVERED_PARAMETERS:
            log_errors[name].append(np.log(max(estimate[name], 1e-12)) - np.log(max(truth[name], 1e-12)))
            log_truths[name].append(np.log(max(truth[name], 1e-12)))
        recovery_replicate_rows.append({
            "window_days": window_days, "fit_span_days": float(fit_days[-1] - fit_days[0]),
            "n_fit": len(fit_days), "pinned": fit["pinned"],
            **{f"true_{name}": truth[name] for name in RECOVERED_PARAMETERS},
            **{f"estimated_{name}": estimate[name] for name in RECOVERED_PARAMETERS}})
    for name in RECOVERED_PARAMETERS:
        recovery_rows.append({"parameter": name, "window_days": window_days,
                              "terminated_share": terminated / N_RECOVERY_REPLICATES,
                              "schedules_too_short": schedules_too_short,
                              "too_short_after_termination": ended_too_short, "failed_fits": failed_fits,
                              **recovery_summary(log_errors[name], log_truths[name])})
parameter_recovery = pd.DataFrame(recovery_rows)
record("n_recovered_parameter_windows", int(parameter_recovery.recovered.sum()))
print("\n5.3 PARAMETER RECOVERY (tolerant-subpopulation model, study schedules)")
print(f"  simulation noise SD {SIMULATION_NOISE_SD} log units; arms end above "
      f"{TERMINATION_VOLUME:,.0f} mm³; recovered when the median |log error| < {RECOVERY_THRESHOLD}")
print(parameter_recovery.round(3).to_string(index=False))
print(f"  parameter-window combinations recovered         {int(parameter_recovery.recovered.sum())} "
      f"of {len(parameter_recovery)}")

# Recovery conditional on draws where the tolerant fraction is identifiable
# in principle, and with error metrics suited to a small fraction.
# The HalfNormal(0.10) kill-rate prior puts much of its mass on kill rates
# below the growth rate, where the sensitive cells never shrink and the
# fraction cannot be told apart from the rest of the tumor in any design. A
# draw counts as "kill strong enough" when (k − g) × observed span >= 1, i.e.
# the sensitive population falls at least e-fold within the arm's fitted
# span. A log error also exaggerates misses on a small fraction (0.01 against
# 0.02 is a log error of 0.69), so the tolerant fraction is also scored by
# its absolute error on the logit scale and on the natural scale.
recovery_replicates = pd.DataFrame(recovery_replicate_rows)
recovery_replicates["kill_identifiable"] = (
    (recovery_replicates.true_kill_rate - recovery_replicates.true_growth_rate)
    * recovery_replicates.fit_span_days >= 1.0)
logit = lambda fraction: np.log(np.clip(fraction, 1e-12, 1 - 1e-12) / (1 - np.clip(fraction, 1e-12, 1 - 1e-12)))
conditional_rows = []
for window_days, window_block in recovery_replicates.groupby("window_days"):
    for subset_label, subset in (("all draws", window_block),
                                 ("kill strong enough", window_block.loc[window_block.kill_identifiable])):
        for name in RECOVERED_PARAMETERS:
            if len(subset) < 10:
                continue
            log_error = (np.log(np.maximum(subset[f"estimated_{name}"], 1e-12))
                         - np.log(np.maximum(subset[f"true_{name}"], 1e-12))).to_numpy()
            row = {"window_days": window_days, "draws": subset_label, "parameter": name,
                   "share_of_draws": len(subset) / len(window_block),
                   **recovery_summary(log_error, np.log(np.maximum(subset[f"true_{name}"], 1e-12)))}
            if name == "tolerant_fraction":
                row["median_abs_logit_error"] = float(np.median(np.abs(
                    logit(subset.estimated_tolerant_fraction) - logit(subset.true_tolerant_fraction))))
                row["median_abs_error"] = float(np.median(np.abs(
                    subset.estimated_tolerant_fraction - subset.true_tolerant_fraction)))
            conditional_rows.append(row)
conditional_recovery = pd.DataFrame(conditional_rows)
strong_recovery = conditional_recovery.loc[conditional_recovery.draws == "kill strong enough"]
record("share_draws_kill_strong_enough",
       float(recovery_replicates.kill_identifiable.mean()))
record("n_recovered_parameter_windows_kill_strong", int(strong_recovery.recovered.sum()))
print("\n  recovery among all draws and among draws with kill strong enough to identify the fraction "
      "((k − g) × span >= 1)")
print(conditional_recovery.round(3).to_string(index=False))
print(f"  draws with kill strong enough: {100 * recovery_replicates.kill_identifiable.mean():.1f}% overall; "
      f"parameter-window combinations recovered among them "
      f"{int(strong_recovery.recovered.sum())} of {len(strong_recovery)}")

# ==========================================================================
# 5.4 Design map: regular schedules with varied interval, duration and
#     number of animals per arm
#     Extra animals enter as repeated measurements of one shared trajectory
#     (noise SD divided by the square root of the number of animals), with no
#     between-animal variation; the map is therefore an upper bound.
# ==========================================================================

DESIGN_INTERVALS = (2, 3, 4, 7)
DESIGN_DURATIONS = (14, 21, 28, 42, 56)
DESIGN_ANIMALS = (1, 2, 3, 5)
# 150 replicates per design (60 in the analysis first run, where Monte Carlo
# noise alone moved designs across the recovery threshold and made the
# counts non-monotone in the number of animals). Common random numbers:
# within one interval and duration, every number of animals uses the same
# true parameters and the same standard-normal noise vectors, scaled by
# σ/√animals, so differences between animal counts are not sampling noise.
N_DESIGN_REPLICATES = 150

design_rng = np.random.default_rng(SEEDS["design_map"])
design_bootstrap_rng = np.random.default_rng(SEEDS["design_map_bootstrap"])
design_replicate_rows = []
for interval_days in DESIGN_INTERVALS:
    for duration_days in DESIGN_DURATIONS:
        days = np.arange(0.0, duration_days + 1, interval_days)
        if len(days) < 4:
            continue
        truths = [draw_prior_parameters(design_rng, GROWTH_PRIOR) for _ in range(N_DESIGN_REPLICATES)]
        unit_noise = design_rng.normal(0.0, 1.0, size=(N_DESIGN_REPLICATES, len(days)))
        for animals in DESIGN_ANIMALS:
            noise_sd = SIMULATION_NOISE_SD / np.sqrt(animals)
            for replicate, truth in enumerate(truths):
                noisy = (log_volume_tolerant(days, truth["log_baseline"], truth["growth_rate"],
                                             truth["kill_rate"], truth["tolerant_fraction"])
                         + noise_sd * unit_noise[replicate])
                fit_days, noisy, _ = apply_termination(days, noisy)
                if len(fit_days) < 4:
                    continue
                fit = fit_mechanism("tolerant subpopulation", fit_days, noisy)
                if not np.isfinite(fit["rss"]):
                    continue
                estimate = tolerant_estimates(fit["x"])
                design_replicate_rows.append({
                    "interval_days": interval_days, "duration_days": duration_days,
                    "animals_per_arm": animals, "replicate": replicate,
                    "fit_span_days": float(fit_days[-1] - fit_days[0]),
                    **{f"true_{name}": truth[name] for name in RECOVERED_PARAMETERS},
                    **{f"estimated_{name}": estimate[name] for name in RECOVERED_PARAMETERS}})
design_replicates = pd.DataFrame(design_replicate_rows)
design_replicates["kill_identifiable"] = (
    (design_replicates.true_kill_rate - design_replicates.true_growth_rate)
    * design_replicates.fit_span_days >= 1.0)

# Per design and parameter: the summary of 5.3, a 95% interval on the median
# absolute log error (resampling replicates), "recovered_confidently" when
# that interval's upper bound is below the threshold, and the median error
# among draws whose kill is strong enough to identify the fraction.
design_rows = []
for (interval_days, duration_days, animals), block in design_replicates.groupby(
        ["interval_days", "duration_days", "animals_per_arm"], sort=True):
    for name in RECOVERED_PARAMETERS:
        if len(block) < 10:
            continue
        log_error = (np.log(np.maximum(block[f"estimated_{name}"], 1e-12))
                     - np.log(np.maximum(block[f"true_{name}"], 1e-12))).to_numpy()
        low, high = bootstrap_median_interval(np.abs(log_error), N_BOOTSTRAP, design_bootstrap_rng)
        strong = block.kill_identifiable.to_numpy()
        row = {"parameter": name, "interval_days": interval_days, "duration_days": duration_days,
               "animals_per_arm": animals,
               **recovery_summary(log_error, np.log(np.maximum(block[f"true_{name}"], 1e-12))),
               "median_error_low": low, "median_error_high": high,
               "recovered_confidently": bool(high < RECOVERY_THRESHOLD),
               "replicates_kill_strong": int(strong.sum()),
               "median_abs_log_error_kill_strong": (float(np.median(np.abs(log_error[strong])))
                                                    if strong.sum() >= 10 else np.nan)}
        if name == "tolerant_fraction":
            row["median_abs_logit_error"] = float(np.median(np.abs(
                logit(block.estimated_tolerant_fraction) - logit(block.true_tolerant_fraction))))
            row["median_abs_error"] = float(np.median(np.abs(
                block.estimated_tolerant_fraction - block.true_tolerant_fraction)))
        design_rows.append(row)
design_map = pd.DataFrame(design_rows)
n_designs = design_map.groupby(["interval_days", "duration_days", "animals_per_arm"]).ngroups
fraction_designs = design_map.loc[design_map.parameter == "tolerant_fraction"]
# Monotonicity in the number of animals: count the steps where adding animals
# raised the median error. With common random numbers these should be few.
animal_increases = (design_map.sort_values("animals_per_arm")
                    .groupby(["parameter", "interval_days", "duration_days"]).median_abs_log_error
                    .apply(lambda errors: int((np.diff(errors.to_numpy()) > 0).sum())))
record("n_alternative_designs", n_designs)
record("n_designs_fraction_recovered", int(fraction_designs.recovered.sum()))
record("n_designs_fraction_recovered_confidently", int(fraction_designs.recovered_confidently.sum()))
for name in ("growth_rate", "kill_rate"):
    subset = design_map.loc[design_map.parameter == name]
    record(f"n_designs_{name}_recovered", int(subset.recovered.sum()))
    record(f"n_designs_{name}_recovered_confidently", int(subset.recovered_confidently.sum()))
record("n_design_steps_error_rose_with_animals", int(animal_increases.sum()))
print("\n5.4 DESIGN MAP")
print(f"  designs evaluated                               {n_designs}; {N_DESIGN_REPLICATES} replicates "
      "each, common random numbers across animals per arm")
for name in RECOVERED_PARAMETERS:
    subset = design_map.loc[design_map.parameter == name]
    print(f"  {name:<18} recovered in {int(subset.recovered.sum())} of {len(subset)} designs; "
          f"interval wholly below the threshold in {int(subset.recovered_confidently.sum())}")
print(f"  steps where an extra animal raised the median error: {int(animal_increases.sum())} of "
      f"{int(sum(len(errors) - 1 for _, errors in design_map.groupby(['parameter', 'interval_days', 'duration_days'])))}")
for name in RECOVERED_PARAMETERS:
    print(f"\n  {name}: recovered (rows: interval, animals per arm; columns: duration in days)")
    print(design_map.loc[design_map.parameter == name]
          .pivot_table(index=["interval_days", "animals_per_arm"], columns="duration_days",
                       values="recovered", aggfunc="first").to_string())
    # The continuous error behind each call: a call can flip between seeds
    # when the median error sits near the threshold, the error itself less so.
    print(f"  {name}: median absolute log error (threshold {RECOVERY_THRESHOLD})")
    print(design_map.loc[design_map.parameter == name]
          .pivot_table(index=["interval_days", "animals_per_arm"], columns="duration_days",
                       values="median_abs_log_error", aggfunc="first").round(2).to_string())
    print(f"  {name}: upper 95% bound of the median absolute log error")
    print(design_map.loc[design_map.parameter == name]
          .pivot_table(index=["interval_days", "animals_per_arm"], columns="duration_days",
                       values="median_error_high", aggfunc="first").round(2).to_string())
print("\n  tolerant_fraction: median absolute error on the logit scale")
print(fraction_designs.pivot_table(index=["interval_days", "animals_per_arm"], columns="duration_days",
                                   values="median_abs_logit_error", aggfunc="first").round(2).to_string())
save_record(parameter_recovery, "parameter_recovery")
save_record(conditional_recovery, "parameter_recovery_conditional")
save_record(design_map, "design_map")
save_record(approximation, "declining_effect_approximation")

# The worked example of 5.1, for the identifiability figure: the fixed
# tolerant-subpopulation curve on the typical 56-day schedule and its best
# declining-effect fit, on that schedule and on a dense grid of days.
schedule_56 = typical_schedule(schedules_56, 56)
example_fit = fit_mechanism("declining drug effect", schedule_56,
                            log_volume_tolerant(schedule_56, **ILLUSTRATIVE_TOLERANT))
dense_days = np.linspace(0.0, float(schedule_56.max()), 300)
approximation_example = pd.concat([
    pd.DataFrame({"grid": grid, "day": days,
                  "tolerant": log_volume_tolerant(days, **ILLUSTRATIVE_TOLERANT),
                  "declining_fit": declining_forward(example_fit["x"], days)})
    for grid, days in (("schedule", schedule_56), ("dense", dense_days))], ignore_index=True)
save_record(approximation_example, "approximation_example")
```

```python
# ==========================================================================
# 5.5 Discrimination between the two regrowth models
#     Trajectories are simulated from each model on the study's measurement
#     schedules, with SD 0.15 log units of noise, and end at their first
#     measurement above 2,000 mm³. Both models are fitted to every
#     trajectory by bounded multi-start least squares, and the model with the
#     lower AIC is selected. Both models have four parameters, so the AIC
#     comparison is a comparison of residual fit.
# ==========================================================================

DISCRIMINATION_WINDOWS = (14, 21, 28, 42, 56)
N_DISCRIMINATION_REPLICATES = 200

discrimination_rng = np.random.default_rng(SEEDS["mechanism_discrimination"])
discrimination_rows = []
# Replicates whose drawn schedule has fewer than five measurements are not
# simulated at all (and draw nothing further, so the random stream of the
# analysis first run is unchanged); they are counted here and reported.
schedules_skipped = {}
for window_days in DISCRIMINATION_WINDOWS:
    usable = [grid[grid <= window_days] for grid in empirical_grids if grid.max() >= window_days]
    for truth in MECHANISMS:
        for _ in range(N_DISCRIMINATION_REPLICATES):
            days = usable[discrimination_rng.integers(len(usable))]
            if len(days) < 5:
                schedules_skipped[(window_days, truth)] = schedules_skipped.get((window_days, truth), 0) + 1
                continue
            log_v0 = np.log(200.0) + discrimination_rng.normal(0, 0.3)
            growth = float(np.exp(discrimination_rng.normal(np.log(0.05), 0.5)))
            if truth == "tolerant subpopulation":
                curve = log_volume_tolerant(
                    days, log_v0, growth, float(abs(discrimination_rng.normal(0.12, 0.04))),
                    float(1.0 / (1.0 + np.exp(-discrimination_rng.normal(-2.5, 1.0)))))
            else:
                curve = log_volume_declining(
                    days, log_v0, growth, float(abs(discrimination_rng.normal(0.14, 0.04))),
                    float(np.exp(discrimination_rng.normal(np.log(0.06), 0.5))))
            noisy = curve + discrimination_rng.normal(0, SIMULATION_NOISE_SD, size=len(days))
            fit_days, noisy, stopped = apply_termination(days, noisy)
            if len(fit_days) < 5:
                discrimination_rows.append({"true_mechanism": truth, "window_days": window_days,
                                            "verdict": "too few measurements", "terminated": stopped})
                continue
            fits = {mechanism: fit_mechanism(mechanism, fit_days, noisy) for mechanism in MECHANISMS}
            verdict, delta = aic_verdict(fits)
            discrimination_rows.append({"true_mechanism": truth, "window_days": window_days,
                                        "verdict": verdict, "terminated": stopped,
                                        "criterion_gap": abs(delta)})
discrimination_fits = pd.DataFrame(discrimination_rows)
scored_fits = discrimination_fits.loc[discrimination_fits.verdict.isin(MECHANISMS)]
mechanism_discrimination = scored_fits.groupby(["true_mechanism", "window_days"]).agg(
    replicates=("verdict", "size"),
    chose_tolerant=("verdict", lambda v: float((v == "tolerant subpopulation").mean())),
    median_criterion_gap=("criterion_gap", "median")).reset_index()
# Accuracy is the share choosing the model that generated the data.
mechanism_discrimination["accuracy"] = np.where(
    mechanism_discrimination.true_mechanism == "tolerant subpopulation",
    mechanism_discrimination.chose_tolerant, 1 - mechanism_discrimination.chose_tolerant)
mechanism_discrimination["replicates_attempted"] = N_DISCRIMINATION_REPLICATES
mechanism_discrimination["schedules_too_short"] = [
    schedules_skipped.get((window, truth), 0)
    for window, truth in zip(mechanism_discrimination.window_days, mechanism_discrimination.true_mechanism)]
excluded = discrimination_fits.groupby("window_days").verdict.apply(
    lambda v: int((v == "too few measurements").sum()))

record("chose_tolerant_min", float(mechanism_discrimination.chose_tolerant.min()))
record("chose_tolerant_max", float(mechanism_discrimination.chose_tolerant.max()))
record("discrimination_mean_accuracy", float(mechanism_discrimination.accuracy.mean()))
print("5.5 DISCRIMINATION BETWEEN THE TWO REGROWTH MODELS")
print(mechanism_discrimination.round(3).to_string(index=False))
print(f"\n  share selecting the tolerant-subpopulation model, range over conditions   "
      f"{mechanism_discrimination.chose_tolerant.min():.3f} to {mechanism_discrimination.chose_tolerant.max():.3f}")
print(f"  mean accuracy over conditions                    {mechanism_discrimination.accuracy.mean():.3f}")
# summed over the two true mechanisms, so the denominator is twice the replicates per mechanism
print(f"  drawn schedules with fewer than five measurements, not simulated (both mechanisms, of "
      f"{2 * N_DISCRIMINATION_REPLICATES} per window): "
      f"{ {window: sum(count for (w, _), count in schedules_skipped.items() if w == window) for window in DISCRIMINATION_WINDOWS} }")
print(f"  simulated arms ending with fewer than five measurements after termination, by window: "
      f"{excluded.to_dict()}")
save_record(mechanism_discrimination, "mechanism_discrimination")

# ==========================================================================
# 5.5b Paired discrimination: the mechanism is the only systematic difference
#     In 5.5 the two generating models draw kill rates from different
#     distributions (centred on 0.12 and 0.14 per day). Here each replicate
#     draws one schedule, baseline volume, growth rate, kill rate (centred on
#     0.13, between the two) and standard-normal noise vector, and generates
#     both mechanisms from them. Only the tolerant fraction (tolerant model)
#     and the decay rate (declining model) are mechanism-specific, with the
#     distributions of 5.5. Each arm ends at its first measurement above
#     2,000 mm³, as in 5.5. Verdicts are recorded twice: always choosing the
#     lower AIC, as in 5.5, and with |ΔAIC| < 2 recorded as no decision, as
#     in Section 12.
# ==========================================================================

N_PAIRED_DISCRIMINATION_REPLICATES = 200        # pairs per window
PAIRED_DISCRIMINATION_TIE = 2.0                 # |ΔAIC| below this is no decision

paired_discrimination_rng = np.random.default_rng(SEEDS["paired_discrimination"])
paired_discrimination_rows = []
paired_schedules_skipped = {window: 0 for window in DISCRIMINATION_WINDOWS}
for window_days in DISCRIMINATION_WINDOWS:
    usable = [grid[grid <= window_days] for grid in empirical_grids if grid.max() >= window_days]
    for replicate in range(N_PAIRED_DISCRIMINATION_REPLICATES):
        # Fixed draw order, so the seed reproduces the same pairs.
        days = usable[paired_discrimination_rng.integers(len(usable))]
        log_v0 = np.log(200.0) + paired_discrimination_rng.normal(0, 0.3)
        growth = float(np.exp(paired_discrimination_rng.normal(np.log(0.05), 0.5)))
        kill = float(abs(paired_discrimination_rng.normal(0.13, 0.04)))
        noise = paired_discrimination_rng.normal(0, 1.0, size=len(days))
        fraction = float(1.0 / (1.0 + np.exp(-paired_discrimination_rng.normal(-2.5, 1.0))))
        decay = float(np.exp(paired_discrimination_rng.normal(np.log(0.06), 0.5)))
        if len(days) < 5:
            paired_schedules_skipped[window_days] += 1     # counted, as in 5.5
            continue
        curves = {"tolerant subpopulation": log_volume_tolerant(days, log_v0, growth, kill, fraction),
                  "declining drug effect": log_volume_declining(days, log_v0, growth, kill, decay)}
        for truth, curve in curves.items():
            noisy = curve + SIMULATION_NOISE_SD * noise
            fit_days, fit_values, stopped = apply_termination(days, noisy)
            row = {"window_days": window_days, "replicate": replicate, "true_mechanism": truth,
                   "terminated": stopped, "n_measurements": len(fit_days)}
            if len(fit_days) < 5:
                paired_discrimination_rows.append({**row, "verdict": "too few measurements",
                                                   "verdict_with_tie": "too few measurements",
                                                   "criterion_gap": np.nan})
                continue
            fits = {mechanism: fit_mechanism(mechanism, fit_days, fit_values) for mechanism in MECHANISMS}
            verdict, delta = aic_verdict(fits)
            verdict_with_tie, _ = aic_verdict(fits, PAIRED_DISCRIMINATION_TIE)
            paired_discrimination_rows.append({**row, "verdict": verdict, "verdict_with_tie": verdict_with_tie,
                                               "criterion_gap": abs(delta)})
paired_discrimination_fits = pd.DataFrame(paired_discrimination_rows)
paired_scored = paired_discrimination_fits.loc[paired_discrimination_fits.verdict.isin(MECHANISMS)]
paired_discrimination = paired_scored.groupby(["true_mechanism", "window_days"]).agg(
    replicates=("verdict", "size"),
    chose_tolerant=("verdict", lambda v: float((v == "tolerant subpopulation").mean())),
    decisive_rate=("verdict_with_tie", lambda v: float(v.isin(MECHANISMS).mean())),
    median_criterion_gap=("criterion_gap", "median")).reset_index()
paired_discrimination["accuracy"] = np.where(
    paired_discrimination.true_mechanism == "tolerant subpopulation",
    paired_discrimination.chose_tolerant, 1 - paired_discrimination.chose_tolerant)
decisive_calls = paired_scored.loc[paired_scored.verdict_with_tie.isin(MECHANISMS)]
paired_decisive_accuracy = (decisive_calls.assign(correct=decisive_calls.verdict_with_tie
                                                   == decisive_calls.true_mechanism)
                            .groupby("window_days").correct.agg(["mean", "sum", "size"]))
# Balanced accuracy per window: the mean of the two per-mechanism accuracies,
# so that a fitter preferring one model scores 0.5 however strong the bias.
balanced_by_window = paired_discrimination.groupby("window_days").accuracy.mean()

record("paired_chose_tolerant_min", float(paired_discrimination.chose_tolerant.min()))
record("paired_chose_tolerant_max", float(paired_discrimination.chose_tolerant.max()))
record("paired_discrimination_mean_accuracy", float(paired_discrimination.accuracy.mean()))
record("paired_balanced_accuracy_min", float(balanced_by_window.min()))
record("paired_balanced_accuracy_max", float(balanced_by_window.max()))
paired_discrimination["schedules_too_short"] = paired_discrimination.window_days.map(paired_schedules_skipped)
print("\n5.5b PAIRED DISCRIMINATION (shared baseline, growth, kill rate, schedule and noise)")
print(f"  pairs drawn per window {N_PAIRED_DISCRIMINATION_REPLICATES}; schedules with fewer than five "
      f"measurements, not simulated: {paired_schedules_skipped}")
print(paired_discrimination.round(3).to_string(index=False))
print("\n  balanced accuracy by window (mean of the two per-mechanism accuracies)")
print(balanced_by_window.round(3).to_string())
print(f"\n  with |ΔAIC| < {PAIRED_DISCRIMINATION_TIE:g} recorded as no decision: accuracy among decisive calls by window")
print(paired_decisive_accuracy.rename(columns={"mean": "accuracy", "sum": "correct", "size": "decisive_calls"})
      .round(3).to_string())
print(f"\n  unpaired (5.5) against paired: mean accuracy over conditions "
      f"{RESULTS['discrimination_mean_accuracy']:.3f} against {RESULTS['paired_discrimination_mean_accuracy']:.3f}; "
      f"share selecting the tolerant model {RESULTS['chose_tolerant_min']:.3f}-{RESULTS['chose_tolerant_max']:.3f} "
      f"against {RESULTS['paired_chose_tolerant_min']:.3f}-{RESULTS['paired_chose_tolerant_max']:.3f}")
save_record(paired_discrimination, "mechanism_discrimination_paired")
```

```python
# ==========================================================================
# 5.6 Bayesian tolerant-subpopulation model: prior predictive simulation and
#     simulation-based calibration
#     Calibration: a parameter set is drawn from the priors, data are
#     simulated on a 28-day schedule, the posterior is sampled with NUTS,
#     and the rank of the true value among posterior draws is recorded.
#     Posterior draws are thinned to roughly their effective sample size
#     before ranking, so that ranks are computed among nearly independent
#     draws. Under correct calibration the normalized ranks are uniform.
# ==========================================================================

prior_days = np.linspace(0.0, 42.0, 15)
prior_draws = Predictive(tolerant_subpopulation_model, num_samples=400)(
    jax.random.PRNGKey(SEEDS["prior_predictive"]), days=prior_days, prior=GROWTH_PRIOR)
prior_volumes = np.exp(np.asarray(prior_draws["observed"]))
print("5.6 PRIOR PREDICTIVE SIMULATION (400 trajectories, days 0 to 42)")
for label, column in (("day 0", 0), ("day 42", -1)):
    values = prior_volumes[:, column]
    print(f"  volume at {label:<7} median {np.median(values):,.0f} mm³ "
          f"[5th {np.percentile(values, 5):,.0f}, 95th {np.percentile(values, 95):,.0f}]")
print(f"  trajectories above 20,000 mm³ at day 42          "
      f"{100 * np.mean(prior_volumes[:, -1] > 20_000):.1f}% (the study removed animals near 2,000 mm³)")

CALIBRATION_DAYS = np.array([0.0, 3.0, 7.0, 10.0, 14.0, 17.0, 21.0, 24.0, 28.0])
N_CALIBRATION_REPLICATES = 100
CALIBRATION_WARMUP, CALIBRATION_DRAWS = 300, 500
CALIBRATED_PARAMETERS = ("growth_rate", "kill_rate", "tolerant_fraction", "noise_scale")

uniform_rng = np.random.default_rng(SEEDS["calibration"])
calibration_rows = []
for replicate in range(N_CALIBRATION_REPLICATES):
    truth_key, sampler_key = jax.random.split(jax.random.PRNGKey(SEEDS["calibration"] + replicate))
    truth = Predictive(tolerant_subpopulation_model, num_samples=1)(
        truth_key, days=CALIBRATION_DAYS, prior=GROWTH_PRIOR)
    sampler = MCMC(NUTS(tolerant_subpopulation_model, target_accept_prob=0.90),
                   num_warmup=CALIBRATION_WARMUP, num_samples=CALIBRATION_DRAWS,
                   num_chains=1, progress_bar=False)
    sampler.run(sampler_key, days=CALIBRATION_DAYS, prior=GROWTH_PRIOR,
                log_volume=np.asarray(truth["observed"])[0], extra_fields=("diverging",))
    posterior = sampler.get_samples()
    divergences = int(np.asarray(sampler.get_extra_fields()["diverging"]).sum())
    for name in CALIBRATED_PARAMETERS:
        draws = np.asarray(posterior[name], dtype=float)
        ess = float(effective_sample_size(draws[None, :]))
        thin = max(1, int(np.ceil(len(draws) / max(ess, 1.0))))
        kept = draws[::thin]
        rank = int(np.sum(kept < float(np.asarray(truth[name])[0])))
        calibration_rows.append({
            "replicate": replicate, "parameter": name, "divergences": divergences,
            "ess": ess, "split_rhat": float(split_gelman_rubin(draws[None, :])),
            "kept_draws": len(kept), "rank": rank,
            "normalized_rank": (rank + uniform_rng.uniform()) / (len(kept) + 1)})
calibration_ranks = pd.DataFrame(calibration_rows)
calibration = calibration_ranks.groupby("parameter").agg(
    replicates=("rank", "size"), median_ess=("ess", "median"),
    median_kept_draws=("kept_draws", "median"), max_split_rhat=("split_rhat", "max"),
    ks_statistic=("normalized_rank", lambda r: stats.kstest(r, "uniform").statistic),
    ks_p_value=("normalized_rank", lambda r: stats.kstest(r, "uniform").pvalue))
replicate_divergences = calibration_ranks.drop_duplicates("replicate").divergences
print("\n5.6 SIMULATION-BASED CALIBRATION")
print(f"  replicates {N_CALIBRATION_REPLICATES}; NUTS, {CALIBRATION_WARMUP} warm-up and "
      f"{CALIBRATION_DRAWS} draws, one chain; schedule {CALIBRATION_DAYS.astype(int).tolist()}")
print(f"  divergent transitions: {int(replicate_divergences.sum())} in total, in "
      f"{int((replicate_divergences > 0).sum())} replicates")
print(calibration.round(3).to_string())
save_record(calibration_ranks, "calibration_ranks")

# Convergence. With one chain per replicate, split R̂ compares the two halves
# of that chain. A replicate is counted as converged when every calibrated
# parameter has split R̂ <= 1.05 and the chain had no divergent transition.
# The uniformity tests are repeated on the converged replicates only; if the
# poorly mixing replicates were driving the result, the two sets would differ.
MAX_CALIBRATION_RHAT = 1.05
by_replicate = calibration_ranks.groupby("replicate").agg(max_rhat=("split_rhat", "max"),
                                                          divergences=("divergences", "first"))
converged_replicates = by_replicate.index[(by_replicate.max_rhat <= MAX_CALIBRATION_RHAT)
                                          & (by_replicate.divergences == 0)]
converged_ranks = calibration_ranks.loc[calibration_ranks.replicate.isin(converged_replicates)]
calibration_converged = converged_ranks.groupby("parameter").agg(
    replicates=("rank", "size"),
    ks_statistic=("normalized_rank", lambda r: stats.kstest(r, "uniform").statistic),
    ks_p_value=("normalized_rank", lambda r: stats.kstest(r, "uniform").pvalue))
record("calibration_replicates", N_CALIBRATION_REPLICATES)
record("calibration_converged_replicates", len(converged_replicates))
record("calibration_replicates_with_divergences", int((replicate_divergences > 0).sum()))
record("calibration_min_ks_p", float(calibration.ks_p_value.min()))
record("calibration_converged_min_ks_p", float(calibration_converged.ks_p_value.min())
       if len(calibration_converged) else np.nan)
print(f"\n  replicates with split R̂ > {MAX_CALIBRATION_RHAT} for some parameter: "
      f"{int((by_replicate.max_rhat > MAX_CALIBRATION_RHAT).sum())}; with a divergent transition: "
      f"{int((by_replicate.divergences > 0).sum())}; converged on both counts: {len(converged_replicates)} "
      f"of {N_CALIBRATION_REPLICATES}")
print("  uniformity of normalized ranks, converged replicates only")
print(calibration_converged.round(3).to_string())

# Data for the Bayesian model-checks figure: the prior predictive
# trajectories, one row per draw and day (the SBC ranks are saved above).
save_record(pd.DataFrame({"draw": np.repeat(np.arange(prior_volumes.shape[0]), len(prior_days)),
                          "day": np.tile(prior_days, prior_volumes.shape[0]),
                          "volume": prior_volumes.ravel()}), "prior_predictive")
```

## 6 · Measurement noise, kinetic estimates and termination hazard

Measurement noise is the median residual SD across untreated arms after a straight line is fitted to log volume (Section 3.6). The intrinsic growth rate of each PDX model is the slope of that line. Exponential growth is the working growth law; the curvature check below reports how often a quadratic fits the untreated arms better.

Kill rates are estimated for each treated arm by least squares under the tolerant-subpopulation model, with the growth rate held at the PDX model's untreated value. The fit starts from three points, as every other mechanistic fit does, and keeps the best; the share of failed fits from the first start alone, the single start of the analysis first run, is printed beside it. The fit is bounded (log kill rate between −18.4 and 6.9), so a fit that finds no signal stops at a bound instead of overflowing. Fits with kill rates below 0.0001 or above 1 per day are failed fits. The variance of the log kill rate over the remaining fits is divided into between-treatment, between-model and remainder parts, by two methods. The first is descriptive marginal shares (the weighted variance of treatment means and of model means, each over the total), which overlap and do not sum to one. The second is a crossed random-effects model fitted by restricted maximum likelihood (REML). Each PDX model–treatment pair is one animal, so the model × treatment interaction cannot be separated from residual error, and both sit in the remainder. The REML fit tries L-BFGS first and moves on to BFGS, Powell and Nelder–Mead only if the previous optimizer fails to converge; whether the final fit converged is printed and stored with the shares, because variance components from a fit that has not converged are not estimates.

Arms were ended when tumors became large. The per-visit probability that a measurement is the arm's last is modeled by logistic regression on standardized log volume and elapsed day, without a penalty, with a bootstrap that resamples arms. The fitted probabilities give inverse-probability-of-censoring weights for the mean post-nadir growth rate. Each post-nadir arm is weighted by the inverse of its estimated probability of remaining under observation through its second post-nadir measurement, the point at which it entered the analysis set (probabilities below 0.05 are set to 0.05).

**Inputs:** `observations`, `untreated_fits`, `untreated_growth`, `post_nadir_arms`.
**Outputs:** `kinetic_estimates`, `kill_rate_partition`, `visits`, `hazard_model`; figure data `termination_hazard_curve.csv`, `termination_hazard_points.csv`, `termination_by_volume_decile.csv`, `last_visits.csv` and `censoring_weights.csv`.

```python
# ==========================================================================
# 6.1 Measurement noise, untreated growth and the growth law
#     The quadratic is counted as fitting better when its residual SD (three
#     parameters removed) is more than 10% below the linear fit's (two removed).
# ==========================================================================

set_section("6")
doubling_days = np.log(2) / untreated_growth.loc[untreated_growth > 0]
curved = untreated_fits.quadratic_residual_sd < 0.9 * untreated_fits.residual_sd
print("6.1 MEASUREMENT NOISE AND UNTREATED GROWTH")
print(f"  untreated arms fitted                           {len(untreated_fits)} "
      f"(>= {MIN_MEASUREMENTS_PER_ARM} measurements)")
print(f"  measurement noise, median residual SD           {MEASUREMENT_NOISE:.4f} log units")
print(f"  residual SD, interquartile range                {untreated_fits.residual_sd.quantile(.25):.4f}"
      f" to {untreated_fits.residual_sd.quantile(.75):.4f}")
print(f"  untreated growth rate, median [5th, 95th]       {untreated_growth.median():.4f} "
      f"[{untreated_growth.quantile(.05):.4f}, {untreated_growth.quantile(.95):.4f}] per day")
print(f"  doubling time, median [5th, 95th]               {doubling_days.median():.1f} "
      f"[{doubling_days.quantile(.05):.1f}, {doubling_days.quantile(.95):.1f}] days")
print(f"  arms where a quadratic fits better              {int(curved.sum())} of {len(untreated_fits)} "
      f"({100 * curved.mean():.1f}%), decelerating in "
      f"{int((curved & (untreated_fits.curvature < 0)).sum())}")
print("  untreated growth rate by tumor type")
print(untreated_fits.groupby("tumor_type").growth_rate.describe()[["count", "25%", "50%", "75%"]]
      .round(4).to_string())

# ==========================================================================
# 6.2 Kill rates and the variance partition
# ==========================================================================

KILL_FIT_BOUNDS = (np.array([-5.0, np.log(1e-8), -30.0]), np.array([15.0, np.log(1e3), 30.0]))
# Three starting points for (log kill rate, logit tolerant fraction), as for
# every other mechanistic fit in the notebook; the best residual sum of
# squares is kept. The first start is the single start of the analysis first
# run, and its kill rate is kept beside the result for comparison.
KILL_FIT_STARTS = ((np.log(0.05), -2.0), (np.log(0.15), -1.0), (np.log(0.02), -3.0))
kill_rows = []
for arm_id, block in arm_series(observations.loc[~observations.is_control]):
    model_id = block.model_id.iloc[0]
    if model_id not in untreated_growth.index or len(block) < MIN_MEASUREMENTS_PER_ARM:
        continue
    day = block.day.to_numpy(float) - block.day.iloc[0]
    log_volume = block.log_volume.to_numpy(float)
    growth = float(untreated_growth[model_id])
    kill_residuals = lambda theta: (log_volume_tolerant(day, theta[0], growth, np.exp(theta[1]),
                                                        1.0 / (1.0 + np.exp(-theta[2]))) - log_volume)
    solutions = [least_squares(kill_residuals, np.clip([log_volume[0], log_kill, logit_fraction], *KILL_FIT_BOUNDS),
                               bounds=KILL_FIT_BOUNDS, max_nfev=2000)
                 for log_kill, logit_fraction in KILL_FIT_STARTS]
    solution = min(solutions, key=lambda candidate: candidate.cost)
    kill_rows.append({"arm_id": arm_id, "model_id": model_id, "treatment": block.treatment.iloc[0],
                      "tumor_type": block.tumor_type.iloc[0], "kill_rate": float(np.exp(solution.x[1])),
                      "tolerant_fraction": float(1.0 / (1.0 + np.exp(-solution.x[2]))),
                      "residual_sd": float(np.sqrt(np.sum(solution.fun ** 2) / max(len(day) - 3, 1))),
                      "kill_rate_first_start": float(np.exp(solutions[0].x[1])),
                      "rss_first_start_minus_best": float(2.0 * (solutions[0].cost - solution.cost))})
kinetic_estimates = pd.DataFrame(kill_rows)
kinetic_estimates["log_kill_rate"] = np.log(kinetic_estimates.kill_rate)
kinetic_estimates["failed_fit"] = ~kinetic_estimates.kill_rate.between(*KILL_RATE_RANGE)
retained = kinetic_estimates.loc[~kinetic_estimates.failed_fit].copy()
below = float((kinetic_estimates.kill_rate < KILL_RATE_RANGE[0]).mean())
above = float((kinetic_estimates.kill_rate > KILL_RATE_RANGE[1]).mean())
record("failed_kill_fit_share", below + above)

print("\n6.2 KILL RATES (tolerant-subpopulation model, growth held at the untreated rate)")
print(f"  treated arms fitted                             {len(kinetic_estimates):,} in "
      f"{kinetic_estimates.model_id.nunique()} PDX models, {kinetic_estimates.treatment.nunique()} treatments")
print(f"  kill rate below {KILL_RATE_RANGE[0]:g} per day (failed)          {100 * below:.1f}%")
print(f"  kill rate above {KILL_RATE_RANGE[1]:g} per day (failed)              {100 * above:.1f}%")
# The same shares from the first start alone (the single start of the
# analysis first run), so the part of the failures that was an optimizer
# artefact is visible.
first_start_failed = ~kinetic_estimates.kill_rate_first_start.between(*KILL_RATE_RANGE)
improved_fits = kinetic_estimates.rss_first_start_minus_best > 1e-8
record("failed_kill_fit_share_first_start", float(first_start_failed.mean()))
print(f"  failed fits, first start only (as first run)    {100 * first_start_failed.mean():.1f}%; "
      f"best of three starts {100 * (below + above):.1f}%")
print(f"  arms whose fit improved with more starts        {int(improved_fits.sum()):,} "
      f"({100 * improved_fits.mean():.1f}%); failed under one start but not three: "
      f"{int((first_start_failed & ~kinetic_estimates.failed_fit).sum()):,}")
print(f"  kill rate on the lower fitting bound (1e-8)     "
      f"{100 * (kinetic_estimates.kill_rate <= 1.0001e-8).mean():.1f}%; on the upper bound (1e3) "
      f"{100 * (kinetic_estimates.kill_rate >= 0.9999e3).mean():.1f}%")
print(f"  retained fits                                   {len(retained):,}; median kill rate "
      f"{retained.kill_rate.median():.4f} per day; log kill rate mean {retained.log_kill_rate.mean():.3f}, "
      f"SD {retained.log_kill_rate.std():.3f}")
kill_quantiles = kinetic_estimates.kill_rate.quantile([0, .01, .05, .25, .5, .75, .95, .99, 1])
print("  kill rate quantiles, all fits: " + ", ".join(f"p{100 * q:g} {v:.3g}" for q, v in kill_quantiles.items()))

total_variance = float(retained.log_kill_rate.var(ddof=1))
grand_mean = float(retained.log_kill_rate.mean())
marginal_shares = {}
for factor, column in (("treatment", "treatment"), ("PDX model", "model_id")):
    means = retained.groupby(column).log_kill_rate.agg(["mean", "size"])
    marginal_shares[factor] = float(np.average((means["mean"] - grand_mean) ** 2,
                                               weights=means["size"]) / total_variance)

# Crossed random intercepts for treatment and PDX model: one group holding
# every arm, with both factors as variance components (the statsmodels form
# for crossed effects). fit_mixed_model moves to another optimizer only if
# L-BFGS fails to converge.
reml_fit, reml_converged, reml_warnings = fit_mixed_model(
    "log_kill_rate ~ 1", retained, groups=np.ones(len(retained)), re_formula="0",
    vc_formula={"treatment": "0 + C(treatment)", "model": "0 + C(model_id)"})
reml_components = dict(zip(reml_fit.model.exog_vc.names, np.asarray(reml_fit.vcomp, float)))
reml_total = sum(reml_components.values()) + float(reml_fit.scale)
kill_rate_partition = pd.DataFrame({
    "component": ["treatment", "PDX model", "remainder (interaction and residual)"],
    "marginal_share": [marginal_shares["treatment"], marginal_shares["PDX model"], np.nan],
    "reml_variance": [reml_components["treatment"], reml_components["model"], float(reml_fit.scale)]})
kill_rate_partition["reml_share"] = kill_rate_partition.reml_variance / reml_total
print(f"\n  variance partition of the log kill rate, retained fits (total variance {total_variance:.3f})")
print(kill_rate_partition.round(3).to_string(index=False))
print(f"  REML fit converged: {reml_converged}")
for message in reml_warnings:
    print(f"  REML fit reported: {message}")
record("reml_converged", int(reml_converged))
record("reml_share_treatment", float(kill_rate_partition.reml_share.iloc[0]))
record("reml_share_model", float(kill_rate_partition.reml_share.iloc[1]))
record("marginal_share_treatment", marginal_shares["treatment"])
record("marginal_share_model", marginal_shares["PDX model"])

# The kill-rate figure is drawn by the figures notebook from
# kinetic_estimates.csv and kill_rate_partition.csv (saved at the end of
# this cell).

# ==========================================================================
# 6.3 Termination hazard and inverse-probability-of-censoring weights
# ==========================================================================

visit_rows = []
for arm_id, block in arm_series(observations):
    if len(block) < 3:
        continue
    for index, (day, log_volume) in enumerate(zip(block.day, block.log_volume)):
        visit_rows.append((arm_id, index, day, log_volume, int(index == len(block) - 1),
                           bool(block.is_control.iloc[0])))
visits = pd.DataFrame(visit_rows, columns=["arm_id", "visit", "day", "log_volume",
                                           "arm_ends_here", "is_control"])
hazard_scaler = StandardScaler().fit(visits[["log_volume", "day"]])
hazard_design = hazard_scaler.transform(visits[["log_volume", "day"]])
hazard_outcome = visits.arm_ends_here.to_numpy()
hazard_model = unpenalized_logistic().fit(hazard_design, hazard_outcome)
volume_coefficient, time_coefficient = hazard_model.coef_[0]
hazard_rng = np.random.default_rng(SEEDS["hazard_bootstrap"])
coefficient_draws = cluster_bootstrap(
    rows_by_cluster(visits.arm_id),
    lambda rows: unpenalized_logistic().fit(hazard_design[rows], hazard_outcome[rows]).coef_[0][0],
    N_BOOTSTRAP, hazard_rng)
coefficient_low, coefficient_high = np.percentile(coefficient_draws, [2.5, 97.5])

median_day = float(visits.day.median())
volume_points = np.percentile(np.exp(visits.log_volume), [5, 25, 50, 75, 95])
hazard_points = hazard_model.predict_proba(hazard_scaler.transform(pd.DataFrame(
    {"log_volume": np.log(volume_points), "day": median_day})))[:, 1]
record("hazard_low_volume", float(volume_points[0]))
record("hazard_low", float(hazard_points[0]))
record("hazard_high_volume", float(volume_points[-1]))
record("hazard_high", float(hazard_points[-1]))
print("\n6.3 TERMINATION HAZARD")
print(f"  visits {len(visits):,} in {visits.arm_id.nunique():,} arms; last visits "
      f"{int(hazard_outcome.sum()):,}")
print(f"  standardized log-volume coefficient            {volume_coefficient:.3f} "
      f"(95% CI {coefficient_low:.3f} to {coefficient_high:.3f}, {len(coefficient_draws):,} replicates)")
print(f"  standardized elapsed-day coefficient           {time_coefficient:.3f}")
print(f"  probability a visit is the arm's last, at the median elapsed day ({median_day:.0f})")
for volume, hazard in zip(volume_points, hazard_points):
    print(f"    {volume:>7,.0f} mm³   {100 * hazard:.1f}%")

continue_probability = 1.0 - hazard_model.predict_proba(hazard_design)[:, 1]
visits["continue_probability"] = continue_probability
continuation = visits.groupby("arm_id").continue_probability.apply(np.asarray)
visit_log_volume = visits.groupby("arm_id").log_volume.apply(np.asarray)
weights = {}
for arm_id in post_nadir_arms.arm_id:
    # The arm entered the analysis set at its second measurement after the
    # nadir, so it had to remain under observation at every earlier visit.
    entry_index = int(np.argmin(visit_log_volume[arm_id])) + MIN_POINTS_AFTER_NADIR
    weights[arm_id] = 1.0 / max(np.prod(continuation[arm_id][:entry_index]), 0.05)
weighted_set = post_nadir_arms.assign(ipc_weight=post_nadir_arms.arm_id.map(weights))
unweighted_mean = float(weighted_set.post_nadir_growth_rate.mean())
weighted_mean = float(np.average(weighted_set.post_nadir_growth_rate, weights=weighted_set.ipc_weight))
record("ipcw_relative_change", (weighted_mean - unweighted_mean) / unweighted_mean)
print(f"\n  mean post-nadir growth rate, unweighted          {unweighted_mean:.5f} per day")
print(f"  weighted for censoring                           {weighted_mean:.5f} per day "
      f"({100 * (weighted_mean - unweighted_mean) / unweighted_mean:+.1f}%)")
print(f"  weights: median {weighted_set.ipc_weight.median():.2f}, 99th percentile "
      f"{weighted_set.ipc_weight.quantile(.99):.2f}, at the cap {int((weighted_set.ipc_weight >= 20).sum())}")

# Data for the termination-hazard figure: the fitted probability of a last
# visit across volumes at the median elapsed day, the same at the five
# volume percentiles printed above, the observed share by volume decile,
# each arm's last measurement, and the censoring weights.
dense_volume = np.exp(np.linspace(*np.log(np.exp(visits.log_volume).quantile([0.02, 0.98])), 120))
dense_hazard = hazard_model.predict_proba(hazard_scaler.transform(pd.DataFrame(
    {"log_volume": np.log(dense_volume), "day": median_day})))[:, 1]
deciles = visits.assign(decile=pd.qcut(visits.log_volume, 10, labels=False))
empirical = deciles.groupby("decile").agg(volume=("log_volume", lambda v: np.exp(v.median())),
                                          end_rate=("arm_ends_here", "mean"),
                                          visits=("arm_ends_here", "size"))
save_record(pd.DataFrame({"volume": dense_volume, "probability_last_visit": dense_hazard,
                          "day": median_day}), "termination_hazard_curve")
save_record(pd.DataFrame({"percentile": [5, 25, 50, 75, 95], "volume": volume_points,
                          "probability_last_visit": hazard_points, "day": median_day}),
            "termination_hazard_points")
save_record(empirical.reset_index(), "termination_by_volume_decile")
save_record(visits.loc[visits.arm_ends_here == 1, ["arm_id", "day", "is_control"]], "last_visits")
save_record(weighted_set[["arm_id", "post_nadir_growth_rate", "ipc_weight"]], "censoring_weights")
save_record(kinetic_estimates, "kinetic_estimates")
save_record(kill_rate_partition, "kill_rate_partition")
```

## 7 · Forecasting and selection of treatment

Tumor volume at day 24 is forecast from measurements taken on or before day 10, the first day on which best average response is defined. The outcome is log volume at day 24, linearly interpolated between the positive readings around it. Readings below 1 mm³ (recorded as zero or below 1 mm³) are left out of the outcome, as in the analysis on which the endpoints were first evaluated; the predictors use readings floored at 1 mm³ (`FORECAST_OUTCOME_ZEROS` in Section 1.1 switches the outcome to the floor for the sensitivity analysis). Arms observed for less than 24 days are not included here (Section 8.1 reinserts them). Two predictor sets are compared here; they and the control sets of Section 8 are listed in `predictor_sets.csv`. The depth set holds baseline log volume and best average response within the window. The trajectory set adds eight shape descriptors; one of them, nadir depth, is itself depth-derived and is also in the expanded depth set. Volume at the end of the window is in neither set. Both sets are fitted with the same gradient-boosted regressor (300 iterations, learning rate 0.05, L2 penalty 1.0, 15 leaves per tree) on the same five folds, each PDX model kept within one fold. The primary population is post-nadir arms. That criterion uses each arm's complete trajectory, but no measurement after day 10 enters any predictor; the cell checks this by replacing every reading after day 10 with a random value, rebuilding the table, and requiring every predictor to come out identical.

Best average response is defined from day 10 onward, so within a window that ends on day 10 the only eligible reading is one taken on day 10 itself: best average response in the window is the running average of percentage change at that reading, and an arm without a day-10 reading has no value and is not in the primary cohort. Leaving readings at the floor out of the outcome also removes arms still at the floor on day 24, the best responders, from the forecasting and selection pools. Section 7.4 repeats both endpoints with the running average at the last reading on or before day 10 (no more than 3.5 days earlier; `FORECAST_WINDOW_RESPONSE`), with readings at the floor kept in the outcome, and with both changes, and compares the arms measured on day 10 with those the last-reading rule recovers.

**Prespecified endpoints.** The first is the relative reduction in out-of-fold root-mean-square error (RMSE), with a margin of 10%. The second is the relative reduction in mean regret, with a margin of 15%. For each PDX model with at least two eligible arms, each predictor set selects the arm with the smallest predicted day-24 volume. Regret is that arm's observed log volume minus the smallest observed log volume among the model's eligible arms. An endpoint is met when its margin is reached and its 95% bootstrap interval, resampling PDX models 2,000 times with out-of-fold predictions held fixed, excludes zero. Holm correction is applied across the two endpoints. Because a lower bound that clears the margin is the stronger statement, the cell also reports whether each lower bound clears its margin and gives a one-sided bootstrap P value against the margin (the share of resamples at or below it), with Holm correction applied to those P values too. Pairwise ranking accuracy counts tied predictions as one half and leaves out pairs whose observed outcomes are tied.

**Inputs:** `observations`, `post_nadir_ids`.
**Outputs:** `PREDICTOR_SETS`, `forecast_primary`, `selection`, `regret`, `forecast_pairs`, `endpoint_sensitivity`, `day10_census`, table `predictor_sets.csv`; figure data `forecast_predictions.csv`, `forecast_bootstrap_draws.csv` and `ranking_pairs.csv`.

```python
# ==========================================================================
# 7.1 Predictor sets and the forecasting table
#     The depth and trajectory sets are the two compared by the prespecified
#     endpoints; the others are controls, fitted in Section 8.
# ==========================================================================

set_section("7")
DEPTH_SET = ["baseline_log_volume", "response_in_window"]
SHAPE_DESCRIPTORS = ["overall_slope", "initial_slope", "steepest_decline", "nadir_depth",
                     "time_to_nadir", "post_nadir_slope_in_window", "curvature", "slope_change"]
TRAJECTORY_SET = DEPTH_SET + SHAPE_DESCRIPTORS
PREDICTOR_SETS = {
    "depth": DEPTH_SET,
    "trajectory": TRAJECTORY_SET,
    "baseline only": ["baseline_log_volume"],
    "expanded depth": DEPTH_SET + ["nadir_depth", "nadir_ratio", "minimum_log_volume"],
    "noise-padded depth": DEPTH_SET + [f"noise_{i}" for i in range(len(SHAPE_DESCRIPTORS))],
    "permuted shape": ["baseline_log_volume"] + [f"permuted_{c}" for c in SHAPE_DESCRIPTORS],
    "depth plus current volume": DEPTH_SET + ["current_log_volume"],
    # Robustness sets (Section 8): the permuted descriptors added to the full
    # depth set, so that the comparison with the trajectory set differs only
    # in whether the shape descriptors carry their real values; and the
    # simplest trajectory summary, baseline volume and the net slope of log
    # volume over the window (in treated arms the slope includes the drug's
    # effect, so this is not a growth-only set).
    "depth plus permuted shape": DEPTH_SET + [f"permuted_{c}" for c in SHAPE_DESCRIPTORS],
    "baseline plus slope": ["baseline_log_volume", "overall_slope"],
}
PREDICTOR_SET_ROLES = {
    "depth": "Comparator for both primary endpoints",
    "trajectory": "Tested set",
    "baseline only": "Floor",
    "expanded depth": "Tests whether shape adds to all depth information",
    "noise-padded depth": "Controls for the number of predictors",
    "permuted shape": "Keeps each descriptor's distribution, removes its association with outcome",
    "depth plus current volume": "Tests whether the advantage is the tumor's size at the end of the window",
    "depth plus permuted shape": "Same predictors as the trajectory set, shape descriptors permuted across arms",
    "baseline plus slope": "Simplest trajectory summary; shows how much of the advantage one early slope carries",
}
predictor_set_table = pd.DataFrame({"set": list(PREDICTOR_SETS),
                        "predictors": ["; ".join(v) for v in PREDICTOR_SETS.values()],
                        "role": [PREDICTOR_SET_ROLES[k] for k in PREDICTOR_SETS]})
predictor_set_table.to_csv(TABLE_DIR / "predictor_sets.csv", index=False)

forecast_table = build_forecast_table(observations.loc[~observations.is_control],
                                      FORECAST_ORIGIN_DAY, FORECAST_HORIZON_DAY)
forecast_primary = (forecast_table.loc[forecast_table.arm_id.isin(post_nadir_ids)]
                    .dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"]).reset_index(drop=True))
outcome = forecast_primary.outcome_log_volume.to_numpy(float)
groups = forecast_primary.model_id.astype(str).to_numpy()
print("7.1 FORECASTING TABLE")
print(f"  treated arms with a day-{FORECAST_ORIGIN_DAY} window and an observed day-{FORECAST_HORIZON_DAY} "
      f"volume   {len(forecast_table):,}")
print(f"  post-nadir arms with every predictor            "
      f"{record('forecast_n_arms', len(forecast_primary)):,} in "
      f"{record('forecast_n_models', len(np.unique(groups)))} PDX models")
# Best average response in the window exists only for arms measured on day
# 10 under the exact-day rule (the window rule in force is printed), so most
# arms without it are lost here; Section 7.4 repeats both endpoints with the
# last-reading rule.
window_response_defined = forecast_table.response_in_window.notna()
record("forecast_table_arms", len(forecast_table))
record("forecast_table_arms_with_window_response", int(window_response_defined.sum()))
print(f"  of the {len(forecast_table):,} treated arms, best average response defined in the window "
      f"(rule {FORECAST_WINDOW_RESPONSE}): {int(window_response_defined.sum()):,} "
      f"({100 * window_response_defined.mean():.1f}%)")

# Leakage audit. Every predictor must be computable from readings taken on or
# before the origin day. Every reading after the origin is replaced by a
# random value, the forecasting table is rebuilt, and every predictor of every
# primary arm must come out identical (only outcomes may change; arms whose
# perturbed outcome falls to the floor drop out of the comparison).
leakage_rng = np.random.default_rng(SEEDS["leakage_audit"])
perturbed = observations.loc[~observations.is_control].copy()
after_origin = (perturbed.day > FORECAST_ORIGIN_DAY).to_numpy()
perturbed.loc[after_origin, "log_volume"] = (perturbed.loc[after_origin, "log_volume"]
                                             + leakage_rng.normal(0.0, 1.0, int(after_origin.sum())))
perturbed.loc[after_origin, "volume"] = np.exp(perturbed.loc[after_origin, "log_volume"])
perturbed_table = build_forecast_table(perturbed, FORECAST_ORIGIN_DAY, FORECAST_HORIZON_DAY,
                                       arm_ids=set(forecast_primary.arm_id)).set_index("arm_id")
predictor_columns = [column for column in forecast_table.columns
                     if column not in ("arm_id", "model_id", "treatment", "tumor_type", "outcome_log_volume")]
audited = forecast_primary.arm_id[forecast_primary.arm_id.isin(perturbed_table.index)].to_numpy()
original_predictors = forecast_primary.set_index("arm_id").loc[audited, predictor_columns].to_numpy(float)
perturbed_predictors = perturbed_table.loc[audited, predictor_columns].to_numpy(float)
assert np.array_equal(np.isnan(original_predictors), np.isnan(perturbed_predictors)), \
    "a predictor's availability depends on readings after the origin"
leakage_difference = float(np.nanmax(np.abs(original_predictors - perturbed_predictors)))
assert leakage_difference == 0.0, "a predictor changes when readings after the origin change"
record("leakage_audit_arms", len(audited))
record("leakage_audit_max_difference", leakage_difference)
print(f"  leakage audit: readings after day {FORECAST_ORIGIN_DAY} replaced by random values; "
      f"{len(predictor_columns)} predictors of {len(audited):,} arms unchanged (largest difference "
      f"{leakage_difference:g})")

# ==========================================================================
# 7.2 First endpoint: out-of-fold forecasting error
# ==========================================================================

forecast_folds = grouped_folds(forecast_primary)
# The fold assignment is deterministic: rebuilding it gives the same folds.
assert all(np.array_equal(train_a, train_b) and np.array_equal(test_a, test_b)
           for (train_a, test_a), (train_b, test_b) in zip(forecast_folds, grouped_folds(forecast_primary))), \
    "grouped folds are not reproducible"
depth_prediction = out_of_fold_predictions(forecast_primary, DEPTH_SET, outcome, forecast_folds)
trajectory_prediction = out_of_fold_predictions(forecast_primary, TRAJECTORY_SET, outcome, forecast_folds)
forecast_reduction = relative_rmse_reduction(outcome, depth_prediction, trajectory_prediction)
forecast_rows_by_model = rows_by_cluster(groups)
forecast_draws = cluster_bootstrap(
    forecast_rows_by_model,
    lambda rows: relative_rmse_reduction(outcome[rows], depth_prediction[rows], trajectory_prediction[rows]),
    N_BOOTSTRAP, np.random.default_rng(SEEDS["forecast_bootstrap"]))
forecast_low, forecast_high = np.percentile(forecast_draws, [2.5, 97.5])
forecast_p = bootstrap_p_two_sided(forecast_draws)
# One-sided bootstrap P value against the margin: the share of resamples in
# which the reduction is at or below the margin. Stronger than the P value
# against zero, and the one that matches the margin rule.
forecast_p_margin = float(np.mean(forecast_draws <= FORECAST_MARGIN))

record("depth_rmse", rmse(outcome, depth_prediction))
record("trajectory_rmse", rmse(outcome, trajectory_prediction))
record("depth_r2", r2_score(outcome, depth_prediction))
record("trajectory_r2", r2_score(outcome, trajectory_prediction))
record("forecast_reduction", forecast_reduction)
record("forecast_reduction_low", forecast_low)
record("forecast_reduction_high", forecast_high)
print(f"\n7.2 FORECASTING, day {FORECAST_ORIGIN_DAY} to day {FORECAST_HORIZON_DAY} "
      f"({len(forecast_primary)} arms, {len(np.unique(groups))} PDX models)")
print(f"  depth set        RMSE {RESULTS['depth_rmse']:.3f}   R² {RESULTS['depth_r2']:.3f}")
print(f"  trajectory set   RMSE {RESULTS['trajectory_rmse']:.3f}   R² {RESULTS['trajectory_r2']:.3f}")
print(f"  relative reduction in RMSE   {format_ci(forecast_reduction, forecast_low, forecast_high)}")
print(f"  margin {FORECAST_MARGIN:.0%} met: {forecast_reduction >= FORECAST_MARGIN}   "
      f"interval excludes zero: {forecast_low > 0}")

# ==========================================================================
# 7.3 Second endpoint: treatment-selection regret, and ranking accuracy
# ==========================================================================

selection = forecast_primary.assign(depth_prediction=depth_prediction,
                                    trajectory_prediction=trajectory_prediction)
regret = selection_by_model(selection, ["depth_prediction", "trajectory_prediction"])
depth_regret = float(regret.depth_prediction_regret.mean())
trajectory_regret = float(regret.trajectory_prediction_regret.mean())
regret_reduction = (depth_regret - trajectory_regret) / depth_regret
regret_rng = np.random.default_rng(SEEDS["regret_bootstrap"])
regret_draws = []
for _ in range(N_BOOTSTRAP):
    drawn = regret.sample(len(regret), replace=True, random_state=int(regret_rng.integers(1 << 31)))
    depth_mean = drawn.depth_prediction_regret.mean()
    if depth_mean > 0:
        regret_draws.append((depth_mean - drawn.trajectory_prediction_regret.mean()) / depth_mean)
regret_low, regret_high = np.percentile(regret_draws, [2.5, 97.5])
regret_p = bootstrap_p_two_sided(regret_draws)
regret_p_margin = float(np.mean(np.asarray(regret_draws) <= REGRET_MARGIN))
holm_thresholds, holm_rejected = holm([forecast_p, regret_p])
holm_margin_thresholds, holm_margin_rejected = holm([forecast_p_margin, regret_p_margin])

# The day-10 log volume of each arm is carried into the pair table, so that
# Section 8.4 can stratify pairs on a separation known at selection time.
forecast_pairs = within_model_pairs(selection, ["depth_prediction", "trajectory_prediction"],
                                    extra_gaps=("current_log_volume",))
record("regret_n_models", len(regret))
record("depth_regret", depth_regret)
record("trajectory_regret", trajectory_regret)
record("depth_regret_fold", float(np.exp(depth_regret)))
record("trajectory_regret_fold", float(np.exp(trajectory_regret)))
record("regret_reduction", regret_reduction)
record("regret_reduction_low", regret_low)
record("regret_reduction_high", regret_high)
record("forecast_p", forecast_p)
record("regret_p", regret_p)
record("n_different_choice", int(regret.different_choice.sum()))
record("share_different_choice", float(regret.different_choice.mean()))
record("ranking_accuracy_depth", float(forecast_pairs.depth_prediction.mean()))
record("ranking_accuracy_trajectory", float(forecast_pairs.trajectory_prediction.mean()))
record("n_ranking_pairs", len(forecast_pairs))
# How the regret reduction is spread over PDX models (no resampling): models
# where each set's choice has the lower regret and ties (the same arm, or
# arms with equal outcomes), the share of models in which each set picks the
# best arm, and the fraction of the net reduction in summed regret carried
# by the 10 largest per-model differences (the quantities printed with
# Figure 2d and 2e, now in the register).
regret_difference = (regret.depth_prediction_regret - regret.trajectory_prediction_regret).to_numpy(float)
net_regret_gain = float(regret_difference.sum())
record("regret_models_trajectory_lower", int((regret_difference > 0).sum()))
record("regret_models_depth_lower", int((regret_difference < 0).sum()))
record("regret_models_tied", int((regret_difference == 0).sum()))
record("best_arm_share_depth", float(np.mean(regret.depth_prediction_regret <= 1e-12)))
record("best_arm_share_trajectory", float(np.mean(regret.trajectory_prediction_regret <= 1e-12)))
# Named "fraction", not "share": it can exceed 1 when other models favor depth.
record("regret_top10_fraction_of_net_reduction",
       float(np.sort(regret_difference)[::-1][:10].sum() / net_regret_gain) if net_regret_gain > 0 else np.nan)
# Each endpoint is met when its margin is reached and its interval excludes
# zero; the Holm decisions are stored beside the endpoint decisions.
record("forecast_endpoint_met", int(forecast_reduction >= FORECAST_MARGIN and forecast_low > 0))
record("regret_endpoint_met", int(regret_reduction >= REGRET_MARGIN and regret_low > 0))
record("forecast_holm_significant", int(holm_rejected[0]))
record("regret_holm_significant", int(holm_rejected[1]))
# The stronger statement: the lower bound of each interval clears its margin,
# and the one-sided P value against the margin survives Holm correction.
record("forecast_lower_bound_clears_margin", int(forecast_low >= FORECAST_MARGIN))
record("regret_lower_bound_clears_margin", int(regret_low >= REGRET_MARGIN))
record("forecast_p_vs_margin", forecast_p_margin)
record("regret_p_vs_margin", regret_p_margin)
record("forecast_holm_margin_significant", int(holm_margin_rejected[0]))
record("regret_holm_margin_significant", int(holm_margin_rejected[1]))

print(f"\n7.3 TREATMENT SELECTION ({len(regret)} PDX models with at least two eligible arms)")
print(f"  mean regret, depth set        {depth_regret:.3f} log units ({np.exp(depth_regret):.2f}-fold)")
print(f"  mean regret, trajectory set   {trajectory_regret:.3f} log units ({np.exp(trajectory_regret):.2f}-fold)")
print(f"  relative reduction in regret  {format_ci(regret_reduction, regret_low, regret_high)}")
print(f"  margin {REGRET_MARGIN:.0%} met: {regret_reduction >= REGRET_MARGIN}   "
      f"interval excludes zero: {regret_low > 0}")
print(f"  PDX models where the two sets select different arms   "
      f"{int(regret.different_choice.sum())} ({100 * regret.different_choice.mean():.1f}%)")
print(f"  pairwise ranking accuracy ({len(forecast_pairs):,} pairs)   depth "
      f"{forecast_pairs.depth_prediction.mean():.3f}   trajectory {forecast_pairs.trajectory_prediction.mean():.3f}")
print(f"  lower regret with the trajectory set in {RESULTS['regret_models_trajectory_lower']} PDX models, with the "
      f"depth set in {RESULTS['regret_models_depth_lower']}, equal in {RESULTS['regret_models_tied']}; best arm "
      f"picked in {100 * RESULTS['best_arm_share_trajectory']:.0f}% (trajectory) and "
      f"{100 * RESULTS['best_arm_share_depth']:.0f}% (depth) of models; the 10 largest differences carry "
      f"{100 * RESULTS['regret_top10_fraction_of_net_reduction']:.0f}% of the net reduction")
print("\n  Holm correction across the two endpoints")
for name, p_value, threshold, rejected in zip(("forecasting", "regret"), (forecast_p, regret_p),
                                              holm_thresholds, holm_rejected):
    print(f"    {name:<12} {p_label(p_value, 1 / N_BOOTSTRAP):<12} threshold {threshold:.3f}   "
          f"significant: {rejected}")
print("\n  against the margins (one-sided: reduction at or below the margin), Holm correction")
for name, p_value, threshold, rejected, low, margin in zip(
        ("forecasting", "regret"), (forecast_p_margin, regret_p_margin), holm_margin_thresholds,
        holm_margin_rejected, (forecast_low, regret_low), (FORECAST_MARGIN, REGRET_MARGIN)):
    print(f"    {name:<12} lower bound {100 * low:.1f}% against margin {100 * margin:.0f}%: "
          f"{'clears it' if low >= margin else 'does not clear it'};   "
          f"{p_label(p_value, 1 / N_BOOTSTRAP):<12} threshold {threshold:.3f}   significant: {rejected}")
save_record(pd.DataFrame([{
    "n_arms": len(forecast_primary), "n_models": len(np.unique(groups)),
    "depth_rmse": RESULTS["depth_rmse"], "trajectory_rmse": RESULTS["trajectory_rmse"],
    "rmse_reduction": forecast_reduction, "rmse_low": forecast_low, "rmse_high": forecast_high,
    "rmse_p": forecast_p, "regret_models": len(regret), "depth_regret": depth_regret,
    "trajectory_regret": trajectory_regret, "regret_reduction": regret_reduction,
    "regret_low": regret_low, "regret_high": regret_high, "regret_p": regret_p}]),
    "forecasting_endpoints")
save_record(regret, "treatment_selection_regret")

# Data for the forecasting and selection figure: each arm's observed day-24
# log volume with both out-of-fold predictions, the bootstrap distribution
# of the relative RMSE reduction, and every within-model pair of 7.3 (the
# per-model regret table is saved above).
save_record(selection[["arm_id", "model_id", "treatment", "tumor_type", "outcome_log_volume",
                       "depth_prediction", "trajectory_prediction"]], "forecast_predictions")
save_record(pd.DataFrame({"relative_reduction": forecast_draws}), "forecast_bootstrap_draws")
save_record(forecast_pairs, "ranking_pairs")
```

```python
# ==========================================================================
# 7.4 Sensitivity of the prespecified endpoints to the window rule and to
#     readings at the floor
#     Window rule: "exact_day" needs a reading on day 10 for best average
#     response; "last_reading" uses the running average at the last reading
#     on or before day 10 (no more than WINDOW_RESPONSE_TOLERANCE_DAYS
#     earlier). Neither uses a reading after day 10.
#     Zero rule: "drop" leaves readings at the floor out of the day-24
#     outcome, which removes arms still at the floor on day 24 (the best
#     responders); "floor" keeps them at 1 mm³.
#     The four combinations are run on the post-nadir arms with the learner,
#     folds and endpoints of Sections 7.2 and 7.3 (evaluate_endpoints). The
#     combination set in Section 1.1 must reproduce 7.2 and 7.3 exactly.
# ==========================================================================

WINDOW_RULES = ("exact_day", "last_reading")
ZERO_RULES = ("drop", "floor")
WINDOW_TAG = {"exact_day": "exact", "last_reading": "last"}
endpoint_rng = np.random.default_rng(SEEDS["endpoint_sensitivity_bootstrap"])

# --- which arms the exact-day rule loses, and whether they differ -----------
census_tables = {rule: build_forecast_table(observations.loc[~observations.is_control], FORECAST_ORIGIN_DAY,
                                            FORECAST_HORIZON_DAY, arm_ids=post_nadir_ids, response_rule=rule)
                 for rule in WINDOW_RULES}
census = census_tables["last_reading"][["arm_id", "tumor_type", "outcome_log_volume", "baseline_log_volume"]].copy()
census["exact_rule"] = census.arm_id.map(census_tables["exact_day"].set_index("arm_id").response_in_window).notna()
census["last_rule"] = census_tables["last_reading"].response_in_window.notna().to_numpy()
census["group"] = np.select([census.exact_rule, census.last_rule],
                            ["measured on day 10", "recovered by the last-reading rule"], "no response in the window")
census["published_best_average_response"] = census.arm_id.map(arms.set_index("arm_id").best_average_response)
last_window_day = observations.loc[observations.day <= FORECAST_ORIGIN_DAY].groupby("arm_id").day.max()
census["last_day_in_window"] = census.arm_id.map(last_window_day)
day10_census = census.groupby("group").agg(
    arms=("arm_id", "size"), median_outcome_log_volume=("outcome_log_volume", "median"),
    median_baseline_log_volume=("baseline_log_volume", "median"),
    median_published_response=("published_best_average_response", "median"),
    median_last_day_in_window=("last_day_in_window", "median"))
measured = census.loc[census.group == "measured on day 10"]
recovered = census.loc[census.group == "recovered by the last-reading rule"]
outcome_test = (stats.mannwhitneyu(measured.outcome_log_volume, recovered.outcome_log_volume)
                if len(measured) and len(recovered) else None)
type_table = pd.crosstab(census.group, census.tumor_type)
type_test = stats.chi2_contingency(type_table) if type_table.shape[0] > 1 else None
record("day10_measured_arms", len(measured))
record("day10_recovered_arms", len(recovered))
if outcome_test is not None:
    record("day10_outcome_difference_p", float(outcome_test.pvalue))
if type_test is not None:
    record("day10_tumor_type_difference_p", float(type_test.pvalue))

print("7.4 SENSITIVITY TO THE WINDOW RULE AND TO READINGS AT THE FLOOR")
print(f"  post-nadir arms with an observed day-{FORECAST_HORIZON_DAY} volume (readings at the floor dropped)")
print(day10_census.round(3).to_string())
if outcome_test is not None:
    print(f"  day-24 log volume, measured on day 10 against recovered: Mann–Whitney {p_label(outcome_test.pvalue)}")
if type_test is not None:
    print(f"  tumor-type mix across the groups: chi-square {p_label(type_test.pvalue)}")
    print(type_table.to_string())

# --- both endpoints under the four combinations --------------------------------
grid_rows = []
for window_rule in WINDOW_RULES:
    for zero_rule in ZERO_RULES:
        table = (build_forecast_table(observations.loc[~observations.is_control], FORECAST_ORIGIN_DAY,
                                      FORECAST_HORIZON_DAY, arm_ids=post_nadir_ids, outcome_zeros=zero_rule,
                                      response_rule=window_rule)
                 .dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"]).reset_index(drop=True))
        result, _ = evaluate_endpoints(table, DEPTH_SET, TRAJECTORY_SET, N_BOOTSTRAP, endpoint_rng)
        if (window_rule, zero_rule) == (FORECAST_WINDOW_RESPONSE, FORECAST_OUTCOME_ZEROS):
            # The primary combination is the analysis of Sections 7.2 and 7.3.
            # Its point estimates are recomputed here (and asserted equal
            # below), but its intervals are taken from 7.2 and 7.3 so that one
            # estimate carries one interval everywhere it is reported; a fresh
            # bootstrap would give slightly different bounds from other draws.
            for mine, theirs in (("rmse_low", "forecast_reduction_low"), ("rmse_high", "forecast_reduction_high"),
                                 ("regret_low", "regret_reduction_low"), ("regret_high", "regret_reduction_high"),
                                 ("rmse_p_vs_margin", "forecast_p_vs_margin"),
                                 ("regret_p_vs_margin", "regret_p_vs_margin")):
                result[mine] = RESULTS[theirs]
        at_floor = int((table.outcome_log_volume <= np.log(VOLUME_FLOOR) + 1e-9).sum())
        grid_rows.append({"window_response": window_rule, "outcome_zeros": zero_rule,
                          "arms_at_floor_on_day_24": at_floor, **result})
        tag = f"{WINDOW_TAG[window_rule]}_{zero_rule}"
        record(f"forecast_reduction_{tag}", result["rmse_reduction"])
        record(f"forecast_reduction_{tag}_low", result["rmse_low"])
        record(f"forecast_reduction_{tag}_high", result["rmse_high"])
        record(f"regret_reduction_{tag}", result["regret_reduction"])
        record(f"regret_reduction_{tag}_low", result["regret_low"])
        record(f"regret_reduction_{tag}_high", result["regret_high"])
        record(f"forecast_n_arms_{tag}", result["arms"])
endpoint_sensitivity = pd.DataFrame(grid_rows)

# The combination fixed in Section 1.1 must reproduce Sections 7.2 and 7.3.
primary_cell = endpoint_sensitivity.loc[(endpoint_sensitivity.window_response == FORECAST_WINDOW_RESPONSE)
                                        & (endpoint_sensitivity.outcome_zeros == FORECAST_OUTCOME_ZEROS)].iloc[0]
assert primary_cell.arms == len(forecast_primary), "7.4 does not rebuild the primary cohort"
assert np.isclose(primary_cell.rmse_reduction, RESULTS["forecast_reduction"]), "7.4 does not reproduce 7.2"
assert np.isclose(primary_cell.regret_reduction, RESULTS["regret_reduction"]), "7.4 does not reproduce 7.3"

columns = ["window_response", "outcome_zeros", "arms", "pdx_models", "arms_at_floor_on_day_24",
           "depth_rmse", "trajectory_rmse", "rmse_reduction", "rmse_low", "rmse_high", "regret_models",
           "regret_reduction", "regret_low", "regret_high", "ranking_depth", "ranking_trajectory",
           "share_different_choice"]
print("\n  both endpoints, post-nadir arms (the row matching Section 1.1 reproduces 7.2 and 7.3)")
print(endpoint_sensitivity[columns].round(3).to_string(index=False))
save_record(endpoint_sensitivity, "forecasting_endpoint_sensitivity")
save_record(day10_census.reset_index(), "forecasting_day10_census")
```

## 8 · Controls and robustness analyses

**Control predictor sets (8.1).** Seven control sets are fitted with the same learner and folds as Section 7 (all sets are listed in `predictor_sets.csv`). Baseline only is the floor. Expanded depth tests whether shape adds to every depth-derived quantity in the window. Noise-padded depth adds eight Gaussian noise predictors to the depth set, so its predictor count matches the trajectory set's. Permuted shape is baseline volume plus the eight shape descriptors permuted across arms: it keeps each descriptor's distribution but removes its association with the outcome. Because it also lacks best average response, its comparison with the trajectory set mixes the loss of shape with the loss of depth. Depth plus permuted shape removes that mixture: it holds exactly the predictors of the trajectory set, with the shape descriptors permuted, so its comparison with the trajectory set isolates the information carried by the real shape values. Depth plus current volume tests whether the advantage reduces to the tumor's size at the end of the window. Baseline plus slope holds the two simplest trajectory quantities, baseline volume and the slope of log volume over the window; it shows how much of the advantage one early slope already carries. In a treated arm that slope is the net rate of change under treatment, so it includes the drug's effect and is not a growth-only control. Each comparison is a relative reduction in RMSE with a 95% interval from 1,500 bootstrap resamples of PDX models.

**Robustness (8.1).** Forecasts are repeated at three start days (10, 14 and 21) and three prediction horizons (10, 14 and 21 days later), and with each tumor type held out of training in turn. Each of these comparisons carries a 95% interval from 2,000 resamples of PDX models with predictions held fixed. Arms that ended before day 24 are reinserted under two scenarios: growth stopped at the last observed volume, or growth continued at the arm's own whole-arm slope (non-negative). These are scenarios, not bounds. Arms observed through day 24 keep the outcome of the primary analysis, with readings at the floor handled by its rule, so the scenarios change only the treatment of censored arms; the same scenarios with readings at the floor kept are printed beside them, and each scenario carries a 95% interval from resampling PDX models.

**Without the post-nadir restriction (8.2).** The primary population is restricted to post-nadir arms, a criterion that uses the complete trajectory. The forecast is repeated on every treated arm with a complete day-10 window and an observed day-24 volume.

**Negative and positive controls (8.3).** Untreated arms cannot truly regress, so they serve as negative controls in three ways. First, the durability-phenotype rule applied to treated arms (Section 3.3) is applied unchanged to untreated arms. Second, the tolerant-subpopulation model is fitted to untreated arms, whose true kill rate is zero; treated arms are fitted the same way (growth rate free) for comparison. Third, the forecast is repeated in untreated arms and in all treated arms over the same windows. Forecasting starts at day 10 or later, because best average response is not defined before day 10. The positive control compares depth between BRAF V600-mutant and wild-type PDX models under MAPK-pathway inhibitors. The approved combinations are the second positive control and appear in Section 10.

**Ranking accuracy by separation (8.4).** Within-model treatment pairs from Section 7 are grouped by the observed gap between their day-24 log volumes, in units of √2σ, the measurement-noise SD of a difference between two readings (σ from Section 3.6); the grouping is repeated with σ the treated-arm interpolation estimate, which is an upper value. The observed day-24 gap is not known when a treatment is chosen, so these groups describe how ranking accuracy depends on the size of the true difference, not a rule that could be applied prospectively. A further grouping uses the two arms' gap in log volume at day 10, which is known at selection time, and the grouping of the analysis first run (gap / σ) is kept for comparison. Ranking accuracy is reported within each group, with the trajectory-minus-depth difference and a 95% interval from resampling PDX models.

**Baseline plus slope and treatment selection (8.5).** The baseline-plus-slope set is taken through the selection endpoints of Section 7 (regret and pairwise ranking accuracy, overall and by separation) and compared with the trajectory set, with intervals from resampling PDX models. The comparison shows what the remaining shape descriptors add beyond a single early slope. It does not separate drug effect from intrinsic growth; that separation comes from the design of the selection endpoint itself: treatments are ranked within a PDX model, and a PDX model's intrinsic growth rate is shared by all of its arms, so it cannot order them.

**Treatment selection among all treated arms (8.6).** The primary selection pool holds post-nadir arms, a criterion that uses each arm's complete trajectory, so the pool itself is defined by outcome. Both endpoints are repeated with every treated arm that has a day-10 window and an observed day-24 volume eligible for selection, with readings at the floor dropped from the outcome, kept, and kept under the last-reading window rule.

**The strongest comparators (8.7).** Depth plus the volume at the end of the window, and baseline volume plus the window slope, recover much of the advantage over the depth set, so every predictor set is carried through both endpoints (RMSE, regret and pairwise ranking) on the same held-out folds, in the post-nadir population and in the practical pool of every treated arm with readings at the floor kept, with the gradient-boosted learner and with ordinary least squares. Every contrast is a paired difference with an interval from resampling PDX models. Least squares on baseline volume and window slope is log-linear extrapolation of early growth, the simplest growth-kinetic comparator. The primary result is also recomputed with each treatment, and each PDX model, left out in turn.

**Stability and censored arms (8.8).** Each predictor set's choice of arm is recomputed after refitting the learner 50 times on bootstrap resamples of the training PDX models, and the share of refits that keep the original choice is reported per PDX model. Arms whose observation ended before day 24, mostly poor responders ended for size, are not candidates in the pools above; in the practical pool they are reinserted under the two scenarios of 8.1 and both endpoints are recomputed, with every PDX model kept in the fold it has without them (otherwise the folds are rebalanced and the comparison mixes the censored arms with a reshuffle of the folds). Finally, both endpoints are recomputed over 20 random partitions of the PDX models into folds; Sections 7 and 8.7 use one deterministic partition, and the spread over partitions shows how much the reported values owe to which PDX models share a fold. That spread describes the cross-validation and is not a confidence interval.

**One interval per estimate (8.7b).** Several estimates are computed twice from identical predictions (contrasts of 8.1 repeated in 8.7, the selection contrasts of 8.5, and two pools of 8.6), and each computation drew its own bootstrap. Each such estimate now carries one interval, from 7.2 for the primary forecasting comparison, from 8.2 for forecasting without the post-nadir restriction, and from 8.7 otherwise; the point estimates are asserted equal, the earlier tables are rewritten with the source of each interval, and no random number is drawn.

**Learners, rankings without a model, change from baseline and later horizons (8.9).** Added after review. The trajectory set fitted by gradient boosting is compared with the depth set fitted by least squares, and each set's two learners are compared, in both populations of 8.7. Arms are also ranked within each PDX model by one raw day-10 quantity (best average response in the window, window slope, day-10 log volume) with no model at all. Selection is repeated with the outcome expressed as change in log volume from baseline, so that a small tumor at baseline does not make an arm look best. Finally both endpoints, regret and ranking included, are computed at every start day and horizon of 8.1, so selection is also judged on volumes up to day 42. Intervals resample PDX models with predictions held fixed; where a window repeats an earlier analysis its interval is taken from there.

**Inputs:** `forecast_primary`, `observations`, `post_nadir_ids`, `PREDICTOR_SETS`, `forecast_pairs`, `regrowth`, `MEASUREMENT_NOISE`, `sheets["mutation"]`.
**Outputs:** `predictor_set_performance`, `predictor_contrasts`, `rolling_origins`, `tumor_type_holdout`, `censoring_scenarios`, `untreated_controls`, `untreated_forecasting`, `positive_control`, `separability`, `separability_all`, `baseline_slope_selection`, `all_arm_selection`, `comparator_contrasts`, `selection_influence`, `selection_stability_summary`, `selection_with_censored`, `fold_partition_summary`, `interval_reconciliation`, `learner_contrasts`, `model_free_levels`, `model_free_selection`, `selection_change_from_baseline`, `selection_by_horizon`; figure data `forecast_predictions_all_sets.csv`, `untreated_regrowth_summary.csv`, `treated_regrowth_summary.csv`, `comparator_predictions_primary.csv` and `comparator_predictions_practical.csv`.

```python
# ==========================================================================
# 8.1 Control predictor sets, start days and horizons, held-out tumor types,
#     and reinsertion of arms that ended before day 24
# ==========================================================================

set_section("8")
control_rng = np.random.default_rng(SEEDS["predictor_controls"])
validation = forecast_primary.copy()
for index in range(len(SHAPE_DESCRIPTORS)):
    validation[f"noise_{index}"] = control_rng.normal(size=len(validation))
for column in SHAPE_DESCRIPTORS:
    validation[f"permuted_{column}"] = validation[column].to_numpy()[control_rng.permutation(len(validation))]

set_predictions, set_errors = compare_predictor_sets(validation, PREDICTOR_SETS, forecast_folds)
predictor_set_performance = pd.DataFrame([
    {"predictor_set": name, "predictors": len(PREDICTOR_SETS[name]), "rmse": set_errors[name],
     "r2": r2_score(outcome, set_predictions[name])} for name in PREDICTOR_SETS]).sort_values("rmse")
print("8.1 PREDICTOR SETS, same learner and folds as Section 7")
print(predictor_set_performance.round(3).to_string(index=False))

CONTRASTS = [("depth", "trajectory"), ("expanded depth", "trajectory"),
             ("noise-padded depth", "trajectory"), ("permuted shape", "trajectory"),
             ("depth", "expanded depth"), ("depth", "noise-padded depth"),
             ("depth plus current volume", "trajectory"), ("depth", "depth plus current volume"),
             # Robustness contrasts, appended so that the resamples drawn for
             # the contrasts above are unchanged.
             ("depth plus permuted shape", "trajectory"), ("baseline plus slope", "trajectory"),
             ("depth", "baseline plus slope")]
contrast_rows = []
for baseline, challenger in CONTRASTS:
    draws = cluster_bootstrap(
        forecast_rows_by_model,
        lambda rows: relative_rmse_reduction(outcome[rows], set_predictions[baseline][rows],
                                             set_predictions[challenger][rows]),
        N_BOOTSTRAP_CONTROLS, control_rng)
    low, high = np.percentile(draws, [2.5, 97.5])
    contrast_rows.append({"challenger": challenger, "baseline": baseline,
                          "relative_reduction": relative_rmse_reduction(
                              outcome, set_predictions[baseline], set_predictions[challenger]),
                          "low": low, "high": high})
predictor_contrasts = pd.DataFrame(contrast_rows)
print(f"\n  relative reduction in RMSE, 95% interval from {N_BOOTSTRAP_CONTROLS:,} resamples of PDX models")
for row in predictor_contrasts.itertuples(index=False):
    print(f"    {row.challenger:<26} against {row.baseline:<26} "
          f"{format_ci(row.relative_reduction, row.low, row.high)}")
contrast_lookup = predictor_contrasts.set_index(["baseline", "challenger"])
for key, (baseline, challenger) in {"expanded_over_depth": ("depth", "expanded depth"),
                                    "trajectory_over_expanded": ("expanded depth", "trajectory"),
                                    "noise_over_depth": ("depth", "noise-padded depth"),
                                    "trajectory_over_current_volume": ("depth plus current volume", "trajectory"),
                                    "trajectory_over_permuted_shape": ("depth plus permuted shape", "trajectory"),
                                    "trajectory_over_baseline_slope": ("baseline plus slope", "trajectory"),
                                    "baseline_slope_over_depth": ("depth", "baseline plus slope"),
                                    }.items():
    row = contrast_lookup.loc[(baseline, challenger)]
    record(key, row.relative_reduction)
    record(f"{key}_low", row.low)
    record(f"{key}_high", row.high)
record("permuted_shape_rmse", set_errors["permuted shape"])
record("baseline_only_rmse", set_errors["baseline only"])
record("baseline_slope_rmse", set_errors["baseline plus slope"])
record("depth_plus_permuted_shape_rmse", set_errors["depth plus permuted shape"])

# --- start days and horizons -----------------------------------------------
# Each window carries a 95% interval from resampling PDX models with its
# out-of-fold predictions held fixed (one generator for all windows).
rolling_rng = np.random.default_rng(SEEDS["rolling_bootstrap"])
rolling_rows = []
for origin_day in (10, 14, 21):
    for lead_days in (10, 14, 21):
        table = build_forecast_table(observations.loc[~observations.is_control], origin_day,
                                     origin_day + lead_days, arm_ids=post_nadir_ids)
        table = table.dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"]).reset_index(drop=True)
        if len(table) < 60 or table.model_id.nunique() < 2:
            continue
        predictions, errors = compare_predictor_sets(table, {"depth": DEPTH_SET, "trajectory": TRAJECTORY_SET})
        _, low, high, _ = reduction_with_interval(
            table.outcome_log_volume.to_numpy(float), predictions["depth"], predictions["trajectory"],
            rows_by_cluster(table.model_id.astype(str)), N_BOOTSTRAP, rolling_rng)
        rolling_rows.append({"start_day": origin_day, "horizon_day": origin_day + lead_days,
                             "arms": len(table), "pdx_models": table.model_id.nunique(),
                             "depth_rmse": errors["depth"], "trajectory_rmse": errors["trajectory"],
                             "relative_reduction": (errors["depth"] - errors["trajectory"]) / errors["depth"],
                             "low": low, "high": high})
rolling_origins = pd.DataFrame(rolling_rows)
record("n_rolling_windows", len(rolling_origins))
record("rolling_reduction_min", rolling_origins.relative_reduction.min())
record("rolling_reduction_max", rolling_origins.relative_reduction.max())
record("rolling_windows_interval_above_zero", int((rolling_origins.low > 0).sum()))
record("rolling_windows_above_margin", int((rolling_origins.relative_reduction >= FORECAST_MARGIN).sum()))
print("\n  start days and horizons (post-nadir arms)")
print(rolling_origins.round(3).to_string(index=False))

# --- each tumor type held out of training ----------------------------------
# Intervals resample the held-out PDX models with predictions held fixed.
holdout_rng = np.random.default_rng(SEEDS["holdout_bootstrap"])
holdout_rows = []
for held_out in sorted(forecast_primary.tumor_type.unique()):
    train = forecast_primary.loc[forecast_primary.tumor_type != held_out]
    test = forecast_primary.loc[forecast_primary.tumor_type == held_out]
    if len(test) < 30 or len(train) < 100:
        continue
    row = {"held_out_tumor_type": held_out, "train_arms": len(train), "test_arms": len(test),
           "test_pdx_models": test.model_id.nunique()}
    held_out_predictions = {}
    for name, predictors in (("depth", DEPTH_SET), ("trajectory", TRAJECTORY_SET)):
        learner = make_forecaster().fit(train[predictors].to_numpy(float),
                                        train.outcome_log_volume.to_numpy(float))
        predicted = learner.predict(test[predictors].to_numpy(float))
        held_out_predictions[name] = predicted
        row[f"{name}_rmse"] = rmse(test.outcome_log_volume, predicted)
        row[f"{name}_r2"] = r2_score(test.outcome_log_volume, predicted)
    row["relative_reduction"] = (row["depth_rmse"] - row["trajectory_rmse"]) / row["depth_rmse"]
    _, row["low"], row["high"], _ = reduction_with_interval(
        test.outcome_log_volume.to_numpy(float), held_out_predictions["depth"],
        held_out_predictions["trajectory"], rows_by_cluster(test.model_id.astype(str)),
        N_BOOTSTRAP, holdout_rng)
    holdout_rows.append(row)
    record(f"holdout_{held_out}", row["relative_reduction"])
tumor_type_holdout = pd.DataFrame(holdout_rows)
lowest_holdout = tumor_type_holdout.loc[tumor_type_holdout.relative_reduction.idxmin()]
others = tumor_type_holdout.drop(index=lowest_holdout.name)
record("holdout_types_above_margin", int((tumor_type_holdout.relative_reduction >= FORECAST_MARGIN).sum()))
record("holdout_others_min", others.relative_reduction.min())
record("holdout_others_max", others.relative_reduction.max())
record("holdout_lowest_type", str(lowest_holdout.held_out_tumor_type))
record("holdout_types_interval_above_zero", int((tumor_type_holdout.low > 0).sum()))
print("\n  each tumor type held out of training")
print(tumor_type_holdout.round(3).to_string(index=False))

# --- arms that ended before day 24, reinserted ------------------------------
# Arms observed through day 24 keep exactly the outcome of the primary
# analysis under each zero rule, so the scenarios change only the treatment
# of censored arms. (In the analysis first run every arm used floored
# readings, which changed the zero rule and the censoring treatment at once.)
# Under "drop", an arm still at the floor on day 24 is left out as in Section
# 7, and so is a censored arm whose last reading is at the floor (it has no
# reading above the floor to carry forward).
scenario_rng = np.random.default_rng(SEEDS["scenario_bootstrap"])
scenario_results, scenario_frames = [], {}
for zero_rule in (FORECAST_OUTCOME_ZEROS, *[rule for rule in ("drop", "floor") if rule != FORECAST_OUTCOME_ZEROS]):
    # censoring_scenario_table (Section 1.3) builds the outcome exactly as
    # build_forecast_table does for arms observed through day 24, and the
    # two scenario outcomes for arms that ended before it.
    scenario_frame, left_out_at_floor = censoring_scenario_table(
        observations, FORECAST_ORIGIN_DAY, FORECAST_HORIZON_DAY, arm_ids=post_nadir_ids, outcome_zeros=zero_rule)
    scenario_frame = scenario_frame.dropna(subset=TRAJECTORY_SET).reset_index(drop=True)
    scenario_frames[zero_rule] = scenario_frame
    for scenario in ("growth_stopped", "growth_continued"):
        frame = scenario_frame.assign(outcome_log_volume=scenario_frame[scenario])
        predictions, errors = compare_predictor_sets(frame, {"depth": DEPTH_SET, "trajectory": TRAJECTORY_SET})
        reduction, low, high, _ = reduction_with_interval(
            frame.outcome_log_volume.to_numpy(float), predictions["depth"], predictions["trajectory"],
            rows_by_cluster(frame.model_id.astype(str)), N_BOOTSTRAP, scenario_rng)
        scenario_results.append({"outcome_zeros": zero_rule, "scenario": scenario.replace("_", " "),
                                 "arms": len(frame), "observed_to_day_24": int((~frame.ended_early).sum()),
                                 "ended_early": int(frame.ended_early.sum()),
                                 "left_out_at_floor": left_out_at_floor,
                                 "depth_rmse": errors["depth"], "trajectory_rmse": errors["trajectory"],
                                 "relative_reduction": reduction, "low": low, "high": high})
        suffix = "" if zero_rule == FORECAST_OUTCOME_ZEROS else f"_{zero_rule}"
        name = "scenario_stopped" if scenario == "growth_stopped" else "scenario_continued"
        record(f"{name}{suffix}", reduction)
        record(f"{name}{suffix}_low", low)
        record(f"{name}{suffix}_high", high)
censoring_scenarios = pd.DataFrame(scenario_results)
primary_scenario_frame = scenario_frames[FORECAST_OUTCOME_ZEROS]
record("n_ended_early", int(primary_scenario_frame.ended_early.sum()))
record("share_ended_early", float(primary_scenario_frame.ended_early.mean()))
print(f"\n  arms that ended before day {FORECAST_HORIZON_DAY}, reinserted (readings at the floor: "
      f"{FORECAST_OUTCOME_ZEROS}, as in Section 7): {int(primary_scenario_frame.ended_early.sum())} of "
      f"{len(primary_scenario_frame)} ({100 * primary_scenario_frame.ended_early.mean():.1f}%); arms observed to "
      f"day {FORECAST_HORIZON_DAY} {int((~primary_scenario_frame.ended_early).sum())} (Section 7: {len(forecast_primary)})")
print(censoring_scenarios.round(3).to_string(index=False))
for name, frame in (("predictor_set_performance", predictor_set_performance),
                    ("predictor_contrasts", predictor_contrasts), ("rolling_origins", rolling_origins),
                    ("tumor_type_holdout", tumor_type_holdout), ("censoring_scenarios", censoring_scenarios)):
    save_record(frame, name)

# Data for the controls figure: the out-of-fold predictions of every
# predictor set on the Section 7 arms (the performance, contrast, window and
# held-out tables are saved above).
save_record(validation[["arm_id", "model_id", "outcome_log_volume"]].assign(
    **{f"prediction_{name.replace(' ', '_')}": values for name, values in set_predictions.items()}),
    "forecast_predictions_all_sets")
```

```python
# ==========================================================================
# 8.2 Forecasting without the post-nadir restriction, day 10 to day 24
#     Same learner, predictor sets and grouped folds as Section 7; every
#     treated arm with a complete day-10 window and an observed day-24 volume.
# ==========================================================================

forecast_all = (forecast_table.dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"])
                .reset_index(drop=True))
outcome_all = forecast_all.outcome_log_volume.to_numpy(float)
groups_all = forecast_all.model_id.astype(str).to_numpy()
folds_all = grouped_folds(forecast_all)
depth_all = out_of_fold_predictions(forecast_all, DEPTH_SET, outcome_all, folds_all)
trajectory_all = out_of_fold_predictions(forecast_all, TRAJECTORY_SET, outcome_all, folds_all)
reduction_all = relative_rmse_reduction(outcome_all, depth_all, trajectory_all)
# Resample PDX models as for the primary endpoint. The seed is the forecasting
# seed plus one, which leaves the Section 7 stream unchanged.
draws_all = cluster_bootstrap(
    rows_by_cluster(groups_all),
    lambda rows: relative_rmse_reduction(outcome_all[rows], depth_all[rows], trajectory_all[rows]),
    N_BOOTSTRAP, np.random.default_rng(SEEDS["forecast_bootstrap"] + 1))
low_all, high_all = np.percentile(draws_all, [2.5, 97.5])
record("forecast_all_n_arms", len(forecast_all))
record("forecast_all_reduction", reduction_all)
record("forecast_all_reduction_low", low_all)
record("forecast_all_reduction_high", high_all)
record("forecast_all_n_models", len(np.unique(groups_all)))
print(f"8.2 FORECASTING WITHOUT THE POST-NADIR RESTRICTION, "
      f"day {FORECAST_ORIGIN_DAY} to day {FORECAST_HORIZON_DAY}")
print(f"  arms {len(forecast_all):,} in {len(np.unique(groups_all))} PDX models")
print(f"  relative reduction in RMSE   {format_ci(reduction_all, low_all, high_all)}")
```

```python
# ==========================================================================
# 8.3 Negative controls (untreated arms) and the mutation-matched positive
#     control
# ==========================================================================

# --- (a) the durability-phenotype rule applied to untreated arms -----------
untreated_regrowth_rows = []
for arm_id, block in arm_series(observations.loc[observations.is_control]):
    summary = summarize_regrowth(block.day.to_numpy(float), block.log_volume.to_numpy(float))
    if summary is not None:
        untreated_regrowth_rows.append({"arm_id": arm_id, **summary,
                               "durability_phenotype": durability_phenotype(summary)})
untreated_controls = pd.DataFrame(untreated_regrowth_rows)
untreated_interior = ((untreated_controls.nadir_index > 0)
                      & (untreated_controls.n_after_nadir >= MIN_POINTS_AFTER_NADIR))
record("untreated_phenotype_share", float(untreated_controls.durability_phenotype.mean()))
print("8.3 NEGATIVE CONTROLS")
print("  (a) durability-phenotype rule          untreated            treated")
print(f"      arms assessed                      {len(untreated_controls):>9,}         {len(regrowth):>9,}")
print(f"      interior nadir with >= 2 later      {100 * untreated_interior.mean():>8.1f}%"
      f"         {100 * RESULTS['treated_interior_nadir_share']:>8.1f}%")
print(f"      all three conditions (phenotype)    {100 * untreated_controls.durability_phenotype.mean():>8.1f}%"
      f"         {100 * RESULTS['treated_phenotype_share']:>8.1f}%")
spurious = untreated_controls.loc[untreated_controls.durability_phenotype, "post_nadir_growth_rate"]
print(f"      post-nadir growth rate, median      {spurious.median():>9.4f}         "
      f"{regrowth.loc[regrowth.durability_phenotype, 'post_nadir_growth_rate'].median():>9.4f}  per day")

# --- (b) apparent kill rates, growth rate free in both groups ---------------
tolerant_forward = constant_exposure_forward("tolerant subpopulation")
free_fit_rows = []
for arm_id, block in arm_series(observations):
    if len(block) < MIN_MEASUREMENTS_PER_ARM:
        continue
    day = block.day.to_numpy(float) - block.day.iloc[0]
    log_volume = block.log_volume.to_numpy(float)
    fit = fit_bounded(tolerant_forward, MECHANISM_BOUNDS["tolerant subpopulation"], day,
                      log_volume, mechanism_starts(log_volume))
    free_fit_rows.append({"arm_id": arm_id, "is_control": bool(block.is_control.iloc[0]),
                          "kill_rate": tolerant_estimates(fit["x"])["kill_rate"],
                          "pinned": fit["pinned"]})
free_fits = pd.DataFrame(free_fit_rows)
treated_free_median = float(free_fits.loc[~free_fits.is_control, "kill_rate"].median())
untreated_free = free_fits.loc[free_fits.is_control, "kill_rate"]
record("untreated_kill_above_treated_median", float((untreated_free > treated_free_median).mean()))
print(f"\n  (b) tolerant-subpopulation fits with the growth rate free "
      f"({int(free_fits.is_control.sum())} untreated, {int((~free_fits.is_control).sum()):,} treated arms)")
print(f"      median apparent kill rate, untreated   {untreated_free.median():.4f} per day")
print(f"      median kill rate, treated              {treated_free_median:.4f} per day")
print(f"      untreated arms above the treated median   "
      f"{100 * (untreated_free > treated_free_median).mean():.1f}%")
print(f"      fits ending on the lower kill-rate bound (0.001 per day): untreated "
      f"{100 * free_fits.loc[free_fits.is_control, 'pinned'].str.contains('log_kill@low').mean():.1f}%, "
      f"treated {100 * free_fits.loc[~free_fits.is_control, 'pinned'].str.contains('log_kill@low').mean():.1f}%")

# --- (c) forecasting in untreated arms and in all treated arms --------------
CONTROL_WINDOWS = [(10, 17), (10, 20), (10, 24), (14, 21), (14, 28)]
MIN_UNTREATED_ARMS, MIN_UNTREATED_MODELS = 80, 20
# One generator for every window and population (the analysis first run
# re-created it from the same seed inside the loop, so all windows and both
# populations shared one random stream).
untreated_forecast_rng = np.random.default_rng(SEEDS["untreated_forecast_bootstrap"])
forecast_rows = []
window_census = []
for origin_day, horizon_day in CONTROL_WINDOWS:
    populations = {}
    for label, is_control in (("untreated", True), ("treated", False)):
        table = build_forecast_table(observations.loc[observations.is_control == is_control],
                                     origin_day, horizon_day)
        populations[label] = (table.dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"])
                              .reset_index(drop=True) if len(table) else table)
    # In-window nadir position of every arm (first occurrence of the minimum,
    # as in early_window_features): interior means neither the first nor the
    # last reading in the window. The analysis first run printed the share
    # whose nadir merely came after the first reading under the interior
    # name; both shares are printed now.
    in_window = observations.loc[observations.day <= origin_day].sort_values(["arm_id", "day"])
    window_interior = in_window.groupby("arm_id").log_volume.agg(
        lambda values: 0 < int(np.argmin(values.to_numpy())) < len(values) - 1)
    window_census.append({
        "start_day": origin_day, "horizon_day": horizon_day,
        **{f"{label}_arms": len(frame) for label, frame in populations.items()},
        **{f"{label}_nadir_after_first_share": float((frame.time_to_nadir > 0).mean()) if len(frame) else np.nan
           for label, frame in populations.items()},
        **{f"{label}_interior_nadir_share": (float(frame.arm_id.map(window_interior).eq(True).mean())
                                             if len(frame) else np.nan)
           for label, frame in populations.items()}})
    untreated_frame = populations["untreated"]
    if len(untreated_frame) < MIN_UNTREATED_ARMS or untreated_frame.model_id.nunique() < MIN_UNTREATED_MODELS:
        continue
    for label, frame in populations.items():
        frame_outcome = frame.outcome_log_volume.to_numpy(float)
        predictions, errors = compare_predictor_sets(frame, {"depth": DEPTH_SET, "trajectory": TRAJECTORY_SET})
        draws = cluster_bootstrap(
            rows_by_cluster(frame.model_id.astype(str)),
            lambda rows: relative_rmse_reduction(frame_outcome[rows], predictions["depth"][rows],
                                                 predictions["trajectory"][rows]),
            N_BOOTSTRAP, untreated_forecast_rng)
        low, high = np.percentile(draws, [2.5, 97.5])
        forecast_rows.append({"population": label, "start_day": origin_day, "horizon_day": horizon_day,
                              "arms": len(frame), "pdx_models": frame.model_id.nunique(),
                              "relative_reduction": (errors["depth"] - errors["trajectory"]) / errors["depth"],
                              "low": low, "high": high})
untreated_forecasting = pd.DataFrame(forecast_rows)
print("\n  (c) forecasting in untreated and treated arms over the same windows")
print(pd.DataFrame(window_census).round(3).to_string(index=False))
print(f"  windows with >= {MIN_UNTREATED_ARMS} untreated arms in >= {MIN_UNTREATED_MODELS} PDX models are fitted")
for row in untreated_forecasting.itertuples(index=False):
    print(f"    {row.population:<9} day {row.start_day} to {row.horizon_day}   {row.arms:>5,} arms   "
          f"{format_ci(row.relative_reduction, row.low, row.high)}")
for row in untreated_forecasting.loc[(untreated_forecasting.start_day == 14)
                                     & (untreated_forecasting.horizon_day == 21)].itertuples(index=False):
    record(f"{row.population}_forecast_14_21", row.relative_reduction)
    record(f"{row.population}_forecast_14_21_low", row.low)
    record(f"{row.population}_forecast_14_21_high", row.high)

# --- positive control: depth by BRAF V600 status under MAPK-pathway agents ---
# The mutation sheet holds one row per finding; the first column is the PDX
# model. The gene column is the one in which "BRAF" occurs most often. A
# finding is BRAF V600 when any of its text fields contains "V600".
mutation = sheets["mutation"]
mutation_model_column = mutation.columns[0]
braf_counts = {column: int((mutation[column].astype(str).str.upper() == "BRAF").sum())
               for column in mutation.columns[1:]}
gene_column = max(braf_counts, key=braf_counts.get)
braf_rows = mutation.loc[mutation[gene_column].astype(str).str.upper() == "BRAF"]
other_text = braf_rows.drop(columns=[mutation_model_column, gene_column]).astype(str)
v600_rows = braf_rows.loc[other_text.apply(lambda c: c.str.contains("V600", case=False)).any(axis=1)]
braf_v600_models = set(v600_rows[mutation_model_column].astype(str))
MAPK_AGENTS = ["encorafenib", "binimetinib", "binimetinib-3.5mpk", "trametinib", "encorafenib + binimetinib"]
positive_rows = []
for agent in MAPK_AGENTS:
    agent_arms = arms.loc[(arms.treatment == agent) & arms.model_id.astype(str).isin(model_sets["mutation"])
                          & arms.best_average_response.notna()]
    mutant = agent_arms.model_id.astype(str).isin(braf_v600_models)
    if mutant.sum() < 3 or (~mutant).sum() < 3:
        continue
    test = stats.mannwhitneyu(agent_arms.loc[mutant, "best_average_response"],
                              agent_arms.loc[~mutant, "best_average_response"], alternative="less")
    positive_rows.append({"agent": agent, "braf_v600_models": int(mutant.sum()),
                          "wild_type_models": int((~mutant).sum()),
                          "median_response_v600": agent_arms.loc[mutant, "best_average_response"].median(),
                          "median_response_wild_type": agent_arms.loc[~mutant, "best_average_response"].median(),
                          "one_sided_p": test.pvalue})
positive_control = pd.DataFrame(positive_rows)
record("braf_v600_models", len(braf_v600_models))
for row in positive_control.itertuples(index=False):
    tag = "".join(character if character.isalnum() else "_" for character in row.agent)
    record(f"positive_control_{tag}_median_v600", float(row.median_response_v600))
    record(f"positive_control_{tag}_median_wild_type", float(row.median_response_wild_type))
    record(f"positive_control_{tag}_p", float(row.one_sided_p))
print("\n  POSITIVE CONTROL: best average response by BRAF V600 status")
print(f"    gene column '{gene_column}'; BRAF V600 findings in {len(braf_v600_models)} PDX models "
      f"of {len(model_sets['mutation'])} with mutation data")
if len(positive_control):
    print(positive_control.assign(one_sided_p=positive_control.one_sided_p.map(format_p))
          .round(2).to_string(index=False))
save_record(untreated_forecasting, "untreated_forecasting")
# The phenotype rule's inputs for untreated and treated arms, for the
# negative-control panels of the figures notebook.
save_record(untreated_controls, "untreated_regrowth_summary")
save_record(regrowth.reset_index(), "treated_regrowth_summary")
save_record(free_fits, "growth_free_kill_rates")
save_record(positive_control, "positive_control_braf")
```

```python
# ==========================================================================
# 8.4 Ranking accuracy by the separation of the two treatments
#     The observed gap between two arms' day-24 log volumes is a difference
#     of two readings, so its measurement-noise SD is √2σ, not σ. The
#     primary strata are in units of √2σ, with σ the measurement noise of
#     Section 3.6; they are repeated with σ the treated-arm interpolation
#     noise (an upper value). The observed day-24 gap is not known when a
#     treatment is chosen, so these strata describe how accuracy depends on
#     the true size of the difference, not a rule that can be applied
#     prospectively. The day-10 strata use the two arms' gap in log volume
#     at day 10, which is known at selection time. The strata of the analysis
#     first run (gap / σ) are kept for comparison.
# ==========================================================================

PAIR_GAP_SD = np.sqrt(2.0) * MEASUREMENT_NOISE
SEPARATION_EDGES = [0, 1, 2, 4, 8, np.inf]
SEPARATION_LABELS = ["< 1", "1-2", "2-4", "4-8", "> 8"]
PRIMARY_SEPARATION = "day-24 gap / √2σ (σ from Section 3.6)"
SEPARATION_SCHEMES = {
    PRIMARY_SEPARATION: forecast_pairs.observed_gap / PAIR_GAP_SD,
    "day-24 gap / √2σ (σ from treated interpolation)": forecast_pairs.observed_gap / (np.sqrt(2.0) * TREATED_NOISE),
    "day-10 gap / √2σ (known at selection)": forecast_pairs.current_log_volume_gap / PAIR_GAP_SD,
    "day-24 gap / σ (as first run)": forecast_pairs.observed_gap / MEASUREMENT_NOISE,
}
# One generator for every stratum of every scheme (the analysis first run
# re-created the generator from the same seed for each stratum).
separation_rng = np.random.default_rng(SEEDS["separation_strata_bootstrap"])
separability_rows = []
for scheme, units in SEPARATION_SCHEMES.items():
    strata = pd.cut(units, SEPARATION_EDGES, labels=SEPARATION_LABELS, include_lowest=True)
    for label in SEPARATION_LABELS:
        stratum = forecast_pairs.loc[strata == label].reset_index(drop=True)
        if stratum.empty:
            continue
        depth_scores = stratum.depth_prediction.to_numpy()
        trajectory_scores = stratum.trajectory_prediction.to_numpy()
        draws = cluster_bootstrap(rows_by_cluster(stratum.model_id),
                                  lambda rows: trajectory_scores[rows].mean() - depth_scores[rows].mean(),
                                  N_BOOTSTRAP, separation_rng)
        low, high = np.percentile(draws, [2.5, 97.5]) if len(draws) else (np.nan, np.nan)
        separability_rows.append({
            "scheme": scheme, "noise_sd_units": label, "pairs": len(stratum),
            "share_of_pairs": len(stratum) / len(forecast_pairs), "pdx_models": stratum.model_id.nunique(),
            # always the day-24 gap in log volume, whichever gap defines the stratum
            "median_day24_gap": stratum.observed_gap.median(), "depth_accuracy": depth_scores.mean(),
            "trajectory_accuracy": trajectory_scores.mean(),
            "difference": trajectory_scores.mean() - depth_scores.mean(), "low": low, "high": high})
separability_all = pd.DataFrame(separability_rows)
separability = (separability_all.loc[separability_all.scheme == PRIMARY_SEPARATION]
                .drop(columns="scheme").reset_index(drop=True))
# The top stratum by its label, not by position, so that an empty stratum
# cannot shift the row reported as "> 8".
top = separability.loc[separability.noise_sd_units == SEPARATION_LABELS[-1]].iloc[0]
record("share_pairs_below_2_sd", float(separability.share_of_pairs.iloc[:2].sum()))
record("top_stratum_share", float(top.share_of_pairs))
record("top_stratum_depth_accuracy", float(top.depth_accuracy))
record("top_stratum_trajectory_accuracy", float(top.trajectory_accuracy))
record("top_stratum_difference", float(top.difference))
record("top_stratum_difference_low", float(top.low))
record("top_stratum_difference_high", float(top.high))
# The top stratum under the other schemes, for comparison with the
# deposited "91%" (first-run units) and with the selection-time strata.
for tag, scheme in (("_treated_noise", "day-24 gap / √2σ (σ from treated interpolation)"),
                    ("_day10", "day-10 gap / √2σ (known at selection)"),
                    ("_first_run", "day-24 gap / σ (as first run)")):
    rows = separability_all.loc[(separability_all.scheme == scheme) & (separability_all.noise_sd_units == "> 8")]
    if len(rows):
        row = rows.iloc[0]
        record(f"top_stratum_share{tag}", float(row.share_of_pairs))
        record(f"top_stratum_depth_accuracy{tag}", float(row.depth_accuracy))
        record(f"top_stratum_trajectory_accuracy{tag}", float(row.trajectory_accuracy))
        record(f"top_stratum_difference{tag}", float(row.difference))
        record(f"top_stratum_difference{tag}_low", float(row.low))
        record(f"top_stratum_difference{tag}_high", float(row.high))
print("8.4 RANKING ACCURACY BY SEPARATION")
print(f"  {len(forecast_pairs):,} pairs in {forecast_pairs.model_id.nunique()} PDX models; measurement noise "
      f"σ {MEASUREMENT_NOISE:.4f}, pair-difference SD √2σ {PAIR_GAP_SD:.4f}; treated interpolation σ "
      f"{TREATED_NOISE:.4f}")
for scheme, block in separability_all.groupby("scheme", sort=False):
    print(f"\n  {scheme}")
    print(block.drop(columns="scheme").round(3).to_string(index=False))
save_record(separability, "ranking_by_separation")
save_record(separability_all, "ranking_by_separation_all_schemes")
```

```python
# ==========================================================================
# 8.5 Baseline volume and the window slope: forecasting, treatment selection
#     and ranking
#     The baseline-plus-slope set (baseline log volume and the least-squares
#     slope of log volume over the window) was fitted in 8.1 with the learner
#     and folds of Section 7. In a treated arm the slope is the net rate of
#     change under treatment, so it carries the drug's effect as well as the
#     tumor's growth: this set is the simplest trajectory summary, not a
#     growth-only control. Its out-of-fold predictions select an arm in each
#     PDX model and rank every within-model pair, exactly as the depth and
#     trajectory sets do in Section 7, so the comparison with the trajectory
#     set shows what the remaining shape descriptors add beyond one early
#     slope. Intervals resample PDX models with predictions held fixed.
# ==========================================================================

slope_rng = np.random.default_rng(SEEDS["baseline_slope_bootstrap"])
slope_prediction = set_predictions["baseline plus slope"]
# validation (8.1) is a copy of forecast_primary, so its rows are in the order
# of the Section 7 selection table.
assert len(slope_prediction) == len(selection)
assert (validation.arm_id.to_numpy() == selection.arm_id.to_numpy()).all()
slope_selection = selection.assign(slope_prediction=slope_prediction)

# --- regret of the arm each set selects --------------------------------------
slope_regret = selection_by_model(slope_selection, ["slope_prediction", "trajectory_prediction"])
slope_mean_regret = float(slope_regret.slope_prediction_regret.mean())
trajectory_mean_regret = float(slope_regret.trajectory_prediction_regret.mean())
slope_regret_reduction = (slope_mean_regret - trajectory_mean_regret) / slope_mean_regret
slope_regret_draws = []
for _ in range(N_BOOTSTRAP):
    drawn = slope_regret.iloc[slope_rng.integers(0, len(slope_regret), size=len(slope_regret))]
    slope_mean = drawn.slope_prediction_regret.mean()
    if slope_mean > 0:
        slope_regret_draws.append((slope_mean - drawn.trajectory_prediction_regret.mean()) / slope_mean)
slope_regret_low, slope_regret_high = np.percentile(slope_regret_draws, [2.5, 97.5])

# --- pairwise ranking accuracy, overall and by separation ---------------------
ranking_columns = ["depth_prediction", "trajectory_prediction", "slope_prediction"]
slope_pairs = within_model_pairs(slope_selection, ranking_columns)
assert len(slope_pairs) == len(forecast_pairs), "the pairs differ from those of Section 7"
# Strata in units of √2σ, as the primary strata of 8.4.
slope_pairs["separation"] = pd.cut(slope_pairs.observed_gap / PAIR_GAP_SD,
                                   SEPARATION_EDGES, labels=SEPARATION_LABELS, include_lowest=True)

slope_rows = []
for label, stratum in [("all pairs", slope_pairs)] + [
        (label, slope_pairs.loc[slope_pairs.separation == label]) for label in SEPARATION_LABELS]:
    if stratum.empty:
        continue
    stratum = stratum.reset_index(drop=True)
    difference, low, high = ranking_difference(stratum, "trajectory_prediction", "slope_prediction",
                                               N_BOOTSTRAP, slope_rng)
    slope_rows.append({"pairs": label if label == "all pairs" else f"{label} pair-noise SDs (√2σ)",
                       "n_pairs": len(stratum), "depth_accuracy": stratum.depth_prediction.mean(),
                       "baseline_slope_accuracy": stratum.slope_prediction.mean(),
                       "trajectory_accuracy": stratum.trajectory_prediction.mean(),
                       "trajectory_minus_baseline_slope": difference, "low": low, "high": high})
baseline_slope_selection = pd.DataFrame(slope_rows)
overall_row = baseline_slope_selection.iloc[0]
# The top stratum by its label, not by position (as in 8.4).
top_row = baseline_slope_selection.loc[baseline_slope_selection.pairs
                                       == f"{SEPARATION_LABELS[-1]} pair-noise SDs (√2σ)"].iloc[0]

record("baseline_slope_regret", slope_mean_regret)
record("trajectory_vs_baseline_slope_regret_reduction", slope_regret_reduction)
record("trajectory_vs_baseline_slope_regret_reduction_low", float(slope_regret_low))
record("trajectory_vs_baseline_slope_regret_reduction_high", float(slope_regret_high))
record("share_different_choice_trajectory_baseline_slope", float(slope_regret.different_choice.mean()))
record("ranking_accuracy_baseline_slope", float(overall_row.baseline_slope_accuracy))
record("ranking_trajectory_minus_baseline_slope", float(overall_row.trajectory_minus_baseline_slope))
record("ranking_trajectory_minus_baseline_slope_low", float(overall_row.low))
record("ranking_trajectory_minus_baseline_slope_high", float(overall_row.high))
record("top_stratum_baseline_slope_accuracy", float(top_row.baseline_slope_accuracy))
record("top_stratum_trajectory_minus_baseline_slope", float(top_row.trajectory_minus_baseline_slope))
record("top_stratum_trajectory_minus_baseline_slope_low", float(top_row.low))
record("top_stratum_trajectory_minus_baseline_slope_high", float(top_row.high))

print("8.5 BASELINE VOLUME AND WINDOW SLOPE (the simplest trajectory summary)")
print(f"  forecasting RMSE: baseline plus slope {RESULTS['baseline_slope_rmse']:.3f}, "
      f"depth {RESULTS['depth_rmse']:.3f}, trajectory {RESULTS['trajectory_rmse']:.3f}")
print(f"  baseline plus slope against depth, relative reduction in RMSE      "
      f"{format_ci(RESULTS['baseline_slope_over_depth'], RESULTS['baseline_slope_over_depth_low'], RESULTS['baseline_slope_over_depth_high'])}")
print(f"  trajectory against baseline plus slope, relative reduction in RMSE "
      f"{format_ci(RESULTS['trajectory_over_baseline_slope'], RESULTS['trajectory_over_baseline_slope_low'], RESULTS['trajectory_over_baseline_slope_high'])}")
print(f"\n  treatment selection ({len(slope_regret)} PDX models)")
print(f"    mean regret, baseline plus slope   {slope_mean_regret:.3f} log units ({np.exp(slope_mean_regret):.2f}-fold)")
print(f"    mean regret, trajectory set        {trajectory_mean_regret:.3f} log units "
      f"({np.exp(trajectory_mean_regret):.2f}-fold)")
print(f"    relative reduction in regret, trajectory against baseline plus slope   "
      f"{format_ci(slope_regret_reduction, slope_regret_low, slope_regret_high)}")
print(f"    PDX models where the two select different arms   {int(slope_regret.different_choice.sum())} "
      f"({100 * slope_regret.different_choice.mean():.1f}%)")
print("\n  pairwise ranking accuracy; trajectory minus baseline plus slope, with a 95% interval "
      "(PDX models resampled)")
print(baseline_slope_selection.round(3).to_string(index=False))
save_record(baseline_slope_selection, "baseline_slope_ranking")
save_record(slope_regret, "baseline_slope_regret")
```

```python
# ==========================================================================
# 8.6 Treatment selection among all treated arms
#     The primary selection pool holds post-nadir arms only, a criterion
#     that uses each arm's complete trajectory, so the pool is defined by
#     outcome. Both endpoints are repeated with every treated arm that has a
#     day-10 window and an observed day-24 volume eligible for selection:
#     readings at the floor dropped from the outcome (the population of
#     8.2); readings at the floor kept, so complete responders return to the
#     pool; and readings at the floor kept under the last-reading window
#     rule (the largest pool). Same learner, folds and endpoints as Section 7.
# ==========================================================================

all_arm_rng = np.random.default_rng(SEEDS["all_arm_selection_bootstrap"])
ALL_ARM_POOLS = [("exact_day", "drop"), ("exact_day", "floor"), ("last_reading", "floor")]
all_arm_rows = []
for window_rule, zero_rule in ALL_ARM_POOLS:
    table = (build_forecast_table(observations.loc[~observations.is_control], FORECAST_ORIGIN_DAY,
                                  FORECAST_HORIZON_DAY, outcome_zeros=zero_rule, response_rule=window_rule)
             .dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"]).reset_index(drop=True))
    result, _ = evaluate_endpoints(table, DEPTH_SET, TRAJECTORY_SET, N_BOOTSTRAP, all_arm_rng)
    all_arm_rows.append({"window_response": window_rule, "outcome_zeros": zero_rule,
                         "post_nadir_share_of_pool": float(table.arm_id.isin(post_nadir_ids).mean()),
                         "arms_at_floor_on_day_24": int((table.outcome_log_volume <= np.log(VOLUME_FLOOR) + 1e-9).sum()),
                         **result})
    tag = f"{WINDOW_TAG[window_rule]}_{zero_rule}"
    record(f"all_arms_forecast_reduction_{tag}", result["rmse_reduction"])
    record(f"all_arms_forecast_reduction_{tag}_low", result["rmse_low"])
    record(f"all_arms_forecast_reduction_{tag}_high", result["rmse_high"])
    record(f"all_arms_regret_reduction_{tag}", result["regret_reduction"])
    record(f"all_arms_regret_reduction_{tag}_low", result["regret_low"])
    record(f"all_arms_regret_reduction_{tag}_high", result["regret_high"])
    record(f"all_arms_ranking_accuracy_depth_{tag}", result["ranking_depth"])
    record(f"all_arms_ranking_accuracy_trajectory_{tag}", result["ranking_trajectory"])
    record(f"all_arms_regret_models_{tag}", result["regret_models"])
all_arm_selection = pd.DataFrame(all_arm_rows)
# The first pool is the population of 8.2 and must reproduce its forecasting result.
if FORECAST_WINDOW_RESPONSE == "exact_day" and FORECAST_OUTCOME_ZEROS == "drop":
    assert np.isclose(all_arm_selection.rmse_reduction.iloc[0], RESULTS["forecast_all_reduction"]), \
        "8.6 does not reproduce the forecasting result of 8.2"
print("8.6 TREATMENT SELECTION AMONG ALL TREATED ARMS")
print(all_arm_selection[["window_response", "outcome_zeros", "arms", "pdx_models", "post_nadir_share_of_pool",
                         "arms_at_floor_on_day_24", "rmse_reduction", "rmse_low", "rmse_high", "regret_models",
                         "regret_reduction", "regret_low", "regret_high", "ranking_depth", "ranking_trajectory",
                         "share_different_choice"]].round(3).to_string(index=False))
save_record(all_arm_selection, "selection_all_treated_arms")
```

```python
# ==========================================================================
# 8.7 The trajectory set against the strongest comparators
#     The headline comparison is against the depth set (baseline volume and
#     best average response). Two simple sets recover much of the
#     advantage: depth plus the volume at the end of the window, and
#     baseline volume plus the window slope (Section 8.1). Here every set is
#     carried through BOTH endpoints (RMSE, regret and pairwise ranking) on
#     the same held-out folds, so every contrast is a paired difference with
#     an interval from resampling PDX models, in two populations: the
#     post-nadir arms of Section 7 and every treated arm with a day-10 window
#     and an observed day-24 volume (readings at the floor kept, the
#     practical screening pool of 8.6). Each set is fitted with the
#     gradient-boosted learner and with ordinary least squares; least
#     squares on baseline volume and window slope is log-linear
#     extrapolation of early growth, the simplest well-specified
#     growth-kinetic comparator.
#     The cell ends by leaving out one treatment, and one PDX model, at a
#     time, to show whether a few of them carry the primary result
#     (predictions held fixed).
# ==========================================================================

COMPARATOR_SETS = {
    "depth": DEPTH_SET,
    "depth plus current volume": PREDICTOR_SETS["depth plus current volume"],
    "baseline plus slope": PREDICTOR_SETS["baseline plus slope"],
    "expanded depth": PREDICTOR_SETS["expanded depth"],
    "trajectory": TRAJECTORY_SET,
}
LEARNERS = {"gradient boosting": make_forecaster, "least squares": make_linear_forecaster}
# (baseline, challenger) pairs; each row reports the challenger against the baseline.
STRONG_CONTRASTS = {
    "trajectory_vs_depth": (("gradient boosting", "depth"), ("gradient boosting", "trajectory")),
    "trajectory_vs_current_volume": (("gradient boosting", "depth plus current volume"),
                                     ("gradient boosting", "trajectory")),
    "trajectory_vs_baseline_slope": (("gradient boosting", "baseline plus slope"), ("gradient boosting", "trajectory")),
    "trajectory_vs_expanded_depth": (("gradient boosting", "expanded depth"), ("gradient boosting", "trajectory")),
    "baseline_slope_vs_depth": (("gradient boosting", "depth"), ("gradient boosting", "baseline plus slope")),
    "current_volume_vs_depth": (("gradient boosting", "depth"), ("gradient boosting", "depth plus current volume")),
    "trajectory_vs_ls_baseline_slope": (("least squares", "baseline plus slope"), ("gradient boosting", "trajectory")),
    "ls_baseline_slope_vs_ls_depth": (("least squares", "depth"), ("least squares", "baseline plus slope")),
    "ls_trajectory_vs_ls_baseline_slope": (("least squares", "baseline plus slope"), ("least squares", "trajectory")),
}
comparator_rng = np.random.default_rng(SEEDS["comparator_bootstrap"])
practical_pool = (build_forecast_table(observations.loc[~observations.is_control], FORECAST_ORIGIN_DAY,
                                       FORECAST_HORIZON_DAY, outcome_zeros="floor",
                                       response_rule=FORECAST_WINDOW_RESPONSE)
                  .dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"]).reset_index(drop=True))
SELECTION_POPULATIONS = {"primary": ("post-nadir arms (Section 7)", forecast_primary),
                         "practical": ("all treated arms, readings at the floor kept", practical_pool)}
comparator_rows, comparator_selected = [], {}
for tag, (population, frame) in SELECTION_POPULATIONS.items():
    frame = frame.reset_index(drop=True)
    population_outcome = frame.outcome_log_volume.to_numpy(float)
    population_folds = grouped_folds(frame)
    predictions = {(learner, name): out_of_fold_predictions(frame, predictors, population_outcome,
                                                            population_folds, factory)
                   for learner, factory in LEARNERS.items() for name, predictors in COMPARATOR_SETS.items()}
    if tag == "primary":
        # The same learner, folds and arms as Section 7, so its predictions recur exactly.
        assert np.allclose(predictions[("gradient boosting", "trajectory")], trajectory_prediction)
        assert np.allclose(predictions[("gradient boosting", "depth")], depth_prediction)
    column_of = {key: f"{key[0]}: {key[1]}" for key in predictions}
    selected = frame.assign(**{column_of[key]: values for key, values in predictions.items()})
    comparator_selected[tag] = selected
    model_clusters = rows_by_cluster(frame.model_id.astype(str))
    for contrast, (baseline, challenger) in STRONG_CONTRASTS.items():
        reduction, low, high, _ = reduction_with_interval(population_outcome, predictions[baseline],
                                                          predictions[challenger], model_clusters,
                                                          N_BOOTSTRAP, comparator_rng)
        selection_contrast = paired_selection_contrast(selected, column_of[baseline], column_of[challenger],
                                                       N_BOOTSTRAP, comparator_rng)
        if tag == "primary" and contrast == "trajectory_vs_depth":
            # This row is the analysis of Sections 7.2 and 7.3 (same arms,
            # learner, folds and predictions, asserted above). Its point
            # estimates must agree, and its intervals are taken from 7.2 and
            # 7.3 so that one estimate carries one interval in the paper.
            assert np.isclose(reduction, RESULTS["forecast_reduction"]), "8.7 does not reproduce 7.2"
            assert np.isclose(selection_contrast["regret_reduction"], RESULTS["regret_reduction"]), \
                "8.7 does not reproduce 7.3"
            low, high = RESULTS["forecast_reduction_low"], RESULTS["forecast_reduction_high"]
            selection_contrast["regret_low"] = RESULTS["regret_reduction_low"]
            selection_contrast["regret_high"] = RESULTS["regret_reduction_high"]
        comparator_rows.append({"population": population, "contrast": contrast, "arms": len(frame),
                                "pdx_models": frame.model_id.nunique(), "baseline": column_of[baseline],
                                "challenger": column_of[challenger],
                                "rmse_baseline": rmse(population_outcome, predictions[baseline]),
                                "rmse_challenger": rmse(population_outcome, predictions[challenger]),
                                "rmse_reduction": reduction, "rmse_low": low, "rmse_high": high,
                                **selection_contrast})
        stem = f"comparator_{tag}_{contrast}"
        for quantity, (value, low_value, high_value) in {
                "rmse_reduction": (reduction, low, high),
                "regret_reduction": (selection_contrast["regret_reduction"], selection_contrast["regret_low"],
                                     selection_contrast["regret_high"]),
                "ranking_difference": (selection_contrast["ranking_difference"], selection_contrast["ranking_low"],
                                       selection_contrast["ranking_high"])}.items():
            record(f"{stem}_{quantity}", value)
            record(f"{stem}_{quantity}_low", low_value)
            record(f"{stem}_{quantity}_high", high_value)
comparator_contrasts = pd.DataFrame(comparator_rows)

print("8.7 THE TRAJECTORY SET AGAINST THE STRONGEST COMPARATORS (paired, PDX models resampled)")
for population, block in comparator_contrasts.groupby("population", sort=False):
    print(f"\n  {population}: {block.arms.iloc[0]:,} arms in {block.pdx_models.iloc[0]} PDX models; "
          f"selection in {block.regret_models.iloc[0]} PDX models with >= 2 arms")
    print(block[["challenger", "baseline", "rmse_reduction", "rmse_low", "rmse_high", "regret_baseline",
                 "regret_challenger", "regret_reduction", "regret_low", "regret_high", "ranking_baseline",
                 "ranking_challenger", "ranking_difference", "ranking_low", "ranking_high", "different_choice"]]
          .round(3).to_string(index=False))

# --- does one treatment or one PDX model carry the primary result? ----------
influence_rows = []
for unit_label, unit_column in (("treatment", "treatment"), ("PDX model", "model_id")):
    for unit in sorted(selection[unit_column].astype(str).unique()):
        remaining = selection.loc[selection[unit_column].astype(str) != unit]
        remaining_regret = selection_by_model(remaining, ["depth_prediction", "trajectory_prediction"])
        depth_mean = remaining_regret.depth_prediction_regret.mean()
        influence_rows.append({
            "left_out": unit_label, "unit": unit,
            "arms_left_out": int((selection[unit_column].astype(str) == unit).sum()),
            "rmse_reduction": relative_rmse_reduction(remaining.outcome_log_volume, remaining.depth_prediction,
                                                      remaining.trajectory_prediction),
            "regret_reduction": (depth_mean - remaining_regret.trajectory_prediction_regret.mean()) / depth_mean
            if depth_mean > 0 else np.nan})
selection_influence = pd.DataFrame(influence_rows)
print("\n  primary result with one treatment, or one PDX model, left out (predictions held fixed)")
for unit_label, block in selection_influence.groupby("left_out", sort=False):
    for quantity in ("rmse_reduction", "regret_reduction"):
        lowest = block.loc[block[quantity].idxmin()]
        record(f"influence_{'treatment' if unit_label == 'treatment' else 'model'}_{quantity}_min",
               float(block[quantity].min()))
        record(f"influence_{'treatment' if unit_label == 'treatment' else 'model'}_{quantity}_max",
               float(block[quantity].max()))
        print(f"    {unit_label:<10} {quantity:<17} range {100 * block[quantity].min():.1f}% to "
              f"{100 * block[quantity].max():.1f}% (all units {100 * RESULTS['forecast_reduction' if quantity == 'rmse_reduction' else 'regret_reduction']:.1f}%); "
              f"lowest without {lowest.unit} ({lowest.arms_left_out} arms)")
save_record(comparator_contrasts, "comparator_contrasts")
# Figure data: the out-of-fold predictions of every set under both learners,
# one file per population (Figure 3a and Supplementary Figure S7 of the
# figures notebook draw them). Columns are named "<learner>: <set>".
for tag, selected in comparator_selected.items():
    prediction_columns = [column for column in selected.columns
                          if column.startswith(("gradient boosting: ", "least squares: "))]
    save_record(selected[["arm_id", "model_id", "outcome_log_volume", "baseline_log_volume",
                          *prediction_columns]], f"comparator_predictions_{tag}")
save_record(selection_influence, "selection_influence")
```

```python
# ==========================================================================
# 8.7b One interval per estimate
#     Several estimates are computed twice from identical predictions: the
#     contrasts of 8.1 that 8.7 repeats in the post-nadir arms with the same
#     learner and folds, the baseline-plus-slope selection contrasts of 8.5,
#     and two pools of 8.6 (the population of 8.2, and the practical pool of
#     8.7). Each computation drew its own bootstrap, so one point estimate
#     carried two slightly different intervals (for example, trajectory
#     against expanded depth, 3.1 to 17.1 in 8.1 and 3.5 to 17.0 in 8.7).
#     Here every point estimate is asserted equal to its twin, and each takes
#     one interval: 7.2 for the primary forecasting comparison, 8.2 for
#     forecasting without the post-nadir restriction, and 8.7 (paired,
#     2,000 resamples of PDX models) for the rest. The tables of 8.1, 8.5 and
#     8.6 are rewritten with an interval_source column, and the register keys
#     of these estimates are rewritten with the same intervals. No random
#     number is drawn, so no other result changes.
# ==========================================================================

primary_label = SELECTION_POPULATIONS["primary"][0]
practical_label = SELECTION_POPULATIONS["practical"][0]
comparator_primary = comparator_contrasts.loc[comparator_contrasts.population == primary_label].set_index("contrast")
comparator_practical = comparator_contrasts.loc[comparator_contrasts.population == practical_label].set_index("contrast")
SOURCE_8_7 = f"8.7 (paired, {N_BOOTSTRAP:,} resamples of PDX models)"
reconciliation_rows = []

# --- (1) contrasts of 8.1 repeated in 8.7 (gradient boosting, post-nadir arms) ---
# (baseline, challenger) of 8.1 -> the 8.7 contrast with the same sets, and
# the register key 8.1 stored the contrast under (None: no key in 8.1).
CONTRASTS_REPEATED_IN_8_7 = {
    ("depth", "trajectory"): ("trajectory_vs_depth", None),
    ("expanded depth", "trajectory"): ("trajectory_vs_expanded_depth", "trajectory_over_expanded"),
    ("depth plus current volume", "trajectory"): ("trajectory_vs_current_volume", "trajectory_over_current_volume"),
    ("baseline plus slope", "trajectory"): ("trajectory_vs_baseline_slope", "trajectory_over_baseline_slope"),
    ("depth", "baseline plus slope"): ("baseline_slope_vs_depth", "baseline_slope_over_depth"),
    ("depth", "depth plus current volume"): ("current_volume_vs_depth", None),
}
predictor_contrasts["interval_source"] = f"8.1 ({N_BOOTSTRAP_CONTROLS:,} resamples of PDX models)"
for (baseline, challenger), (contrast, register_key) in CONTRASTS_REPEATED_IN_8_7.items():
    position = predictor_contrasts.index[(predictor_contrasts.baseline == baseline)
                                         & (predictor_contrasts.challenger == challenger)]
    if not len(position) or contrast not in comparator_primary.index:
        continue
    source = comparator_primary.loc[contrast]
    assert np.isclose(predictor_contrasts.loc[position[0], "relative_reduction"], source.rmse_reduction), \
        f"8.1 and 8.7 give different estimates for {challenger} against {baseline}"
    before = predictor_contrasts.loc[position[0], ["low", "high"]].to_numpy(float)
    predictor_contrasts.loc[position[0], ["low", "high"]] = [float(source.rmse_low), float(source.rmse_high)]
    # The 8.7 row of the primary comparison already carries the interval of 7.2.
    predictor_contrasts.loc[position[0], "interval_source"] = (
        "7.2 (primary endpoint)" if contrast == "trajectory_vs_depth" else SOURCE_8_7)
    if register_key:
        record(f"{register_key}_low", float(source.rmse_low))
        record(f"{register_key}_high", float(source.rmse_high))
    reconciliation_rows.append({"section": "8.1", "estimate": f"RMSE reduction, {challenger} against {baseline}",
                                "value": float(source.rmse_reduction),
                                "interval_before": tuple(round(float(value), 4) for value in before),
                                "interval_now": (round(float(source.rmse_low), 4), round(float(source.rmse_high), 4)),
                                "source": predictor_contrasts.loc[position[0], "interval_source"]})

# --- (2) the baseline-plus-slope selection contrasts of 8.5 -------------------------
slope_source = comparator_primary.loc["trajectory_vs_baseline_slope"]
assert np.isclose(slope_source.regret_reduction, RESULTS["trajectory_vs_baseline_slope_regret_reduction"]), \
    "8.5 and 8.7 give different regret reductions for trajectory against baseline plus slope"
assert np.isclose(slope_source.ranking_difference, RESULTS["ranking_trajectory_minus_baseline_slope"]), \
    "8.5 and 8.7 give different ranking differences for trajectory against baseline plus slope"
for quantity, key, low, high in (
        ("regret reduction", "trajectory_vs_baseline_slope_regret_reduction", slope_source.regret_low,
         slope_source.regret_high),
        ("ranking accuracy difference", "ranking_trajectory_minus_baseline_slope", slope_source.ranking_low,
         slope_source.ranking_high)):
    reconciliation_rows.append({"section": "8.5", "estimate": f"{quantity}, trajectory against baseline plus slope",
                                "value": float(RESULTS[key]),
                                "interval_before": (round(float(RESULTS[f"{key}_low"]), 4),
                                                    round(float(RESULTS[f"{key}_high"]), 4)),
                                "interval_now": (round(float(low), 4), round(float(high), 4)), "source": SOURCE_8_7})
    record(f"{key}_low", float(low))
    record(f"{key}_high", float(high))
# The "all pairs" row of the 8.5 ranking table takes the same interval; the
# separation strata exist only in 8.5 and keep theirs.
baseline_slope_selection["interval_source"] = "8.5"
all_pairs_row = baseline_slope_selection.index[baseline_slope_selection.pairs == "all pairs"]
baseline_slope_selection.loc[all_pairs_row, ["low", "high"]] = [float(slope_source.ranking_low),
                                                                float(slope_source.ranking_high)]
baseline_slope_selection.loc[all_pairs_row, "interval_source"] = SOURCE_8_7

# --- (3) pools of 8.6 that repeat 8.2 and the practical pool of 8.7 ------------------
all_arm_selection["rmse_interval_source"] = "8.6"
all_arm_selection["regret_interval_source"] = "8.6"
primary_rule_pool = all_arm_selection.index[(all_arm_selection.window_response == FORECAST_WINDOW_RESPONSE)
                                            & (all_arm_selection.outcome_zeros == FORECAST_OUTCOME_ZEROS)]
practical_rule_pool = all_arm_selection.index[(all_arm_selection.window_response == FORECAST_WINDOW_RESPONSE)
                                              & (all_arm_selection.outcome_zeros == "floor")]
if len(primary_rule_pool):
    # The population of 8.2 (every treated arm with a day-10 window and an
    # observed day-24 volume, rules of Section 1.1): forecasting interval of 8.2.
    index = primary_rule_pool[0]
    assert np.isclose(all_arm_selection.loc[index, "rmse_reduction"], RESULTS["forecast_all_reduction"])
    tag = f"{WINDOW_TAG[FORECAST_WINDOW_RESPONSE]}_{FORECAST_OUTCOME_ZEROS}"
    reconciliation_rows.append({"section": "8.6", "estimate": f"RMSE reduction, all treated arms ({tag})",
                                "value": float(RESULTS["forecast_all_reduction"]),
                                "interval_before": (round(float(all_arm_selection.loc[index, "rmse_low"]), 4),
                                                    round(float(all_arm_selection.loc[index, "rmse_high"]), 4)),
                                "interval_now": (round(float(RESULTS["forecast_all_reduction_low"]), 4),
                                                 round(float(RESULTS["forecast_all_reduction_high"]), 4)),
                                "source": "8.2"})
    all_arm_selection.loc[index, ["rmse_low", "rmse_high"]] = [RESULTS["forecast_all_reduction_low"],
                                                               RESULTS["forecast_all_reduction_high"]]
    all_arm_selection.loc[index, "rmse_interval_source"] = "8.2"
    record(f"all_arms_forecast_reduction_{tag}_low", float(RESULTS["forecast_all_reduction_low"]))
    record(f"all_arms_forecast_reduction_{tag}_high", float(RESULTS["forecast_all_reduction_high"]))
if len(practical_rule_pool):
    # The practical pool of 8.7 (readings at the floor kept): both intervals of 8.7.
    index = practical_rule_pool[0]
    practical = comparator_practical.loc["trajectory_vs_depth"]
    assert all_arm_selection.loc[index, "arms"] == len(practical_pool), "8.6 and 8.7 hold different practical pools"
    assert np.isclose(all_arm_selection.loc[index, "rmse_reduction"], practical.rmse_reduction)
    assert np.isclose(all_arm_selection.loc[index, "regret_reduction"], practical.regret_reduction)
    tag = f"{WINDOW_TAG[FORECAST_WINDOW_RESPONSE]}_floor"
    for quantity, low, high in (("rmse", practical.rmse_low, practical.rmse_high),
                                ("regret", practical.regret_low, practical.regret_high)):
        reconciliation_rows.append({
            "section": "8.6", "estimate": f"{quantity} reduction, all treated arms ({tag})",
            "value": float(all_arm_selection.loc[index, f"{quantity}_reduction"]),
            "interval_before": (round(float(all_arm_selection.loc[index, f"{quantity}_low"]), 4),
                                round(float(all_arm_selection.loc[index, f"{quantity}_high"]), 4)),
            "interval_now": (round(float(low), 4), round(float(high), 4)), "source": SOURCE_8_7})
        all_arm_selection.loc[index, [f"{quantity}_low", f"{quantity}_high"]] = [float(low), float(high)]
        all_arm_selection.loc[index, f"{quantity}_interval_source"] = SOURCE_8_7
        register_stem = "forecast" if quantity == "rmse" else "regret"
        record(f"all_arms_{register_stem}_reduction_{tag}_low", float(low))
        record(f"all_arms_{register_stem}_reduction_{tag}_high", float(high))

interval_reconciliation = pd.DataFrame(reconciliation_rows)
print("8.7b ONE INTERVAL PER ESTIMATE (point estimates asserted equal; intervals taken from the source shown)")
print(interval_reconciliation.to_string(index=False))
# The tables saved earlier in 8.1, 8.5 and 8.6 are rewritten with these intervals.
save_record(predictor_contrasts, "predictor_contrasts")
save_record(baseline_slope_selection, "baseline_slope_ranking")
save_record(all_arm_selection, "selection_all_treated_arms")
save_record(interval_reconciliation, "interval_reconciliation")
```

```python
# ==========================================================================
# 8.8 Stability of the selected arm, and censored arms in the selection pool
#     (a) Stability. Each set's choice of arm in a PDX model comes from one
#         fit. The learner is refitted N_STABILITY_REFITS times with each
#         training fold replaced by a bootstrap resample of its PDX models
#         (selection_stability), and the share of refits in which each set
#         keeps its original choice is reported per PDX model, with the
#         spread over refits of the regret reduction and of the share of PDX
#         models where the two sets choose differently.
#     (b) Censored arms. Arms whose observation ended before day 24 have no
#         day-24 volume and are not candidates in the pools above; arms are
#         ended when tumors grow large, so they are mostly poor responders.
#         In the practical pool they are reinserted under the two scenarios
#         of Section 8.1 (growth stopped at the last reading, or continued at
#         the arm's own slope), and both endpoints are recomputed.
# ==========================================================================

N_STABILITY_REFITS = 50       # 50 refits x 5 folds x 2 sets per population; several minutes
stability_rng = np.random.default_rng(SEEDS["stability_refits"])
stability_model_tables, stability_refit_tables, stability_rows = [], [], []
for tag, (population, frame) in SELECTION_POPULATIONS.items():
    per_model, per_refit = selection_stability(frame, DEPTH_SET, TRAJECTORY_SET, N_STABILITY_REFITS, stability_rng)
    stability_model_tables.append(per_model.assign(population=population))
    stability_refit_tables.append(per_refit.assign(population=population))
    row = {"population": population, "pdx_models": len(per_model), "refits": len(per_refit)}
    for name in ("depth", "trajectory"):
        row[f"mean_stability_{name}"] = float(per_model[f"stability_{name}"].mean())
        row[f"share_stable_{name}"] = float((per_model[f"stability_{name}"] >= 0.8).mean())
        record(f"stability_{tag}_{name}_mean", row[f"mean_stability_{name}"])
        record(f"stability_{tag}_{name}_share_stable", row[f"share_stable_{name}"])
    for quantity in ("regret_reduction", "different_choice", "ranking_depth", "ranking_trajectory"):
        values = per_refit[quantity].dropna()
        row[f"{quantity}_median"] = float(values.median())
        row[f"{quantity}_2_5"] = float(values.quantile(0.025))
        row[f"{quantity}_97_5"] = float(values.quantile(0.975))
    record(f"refit_{tag}_regret_reduction", row["regret_reduction_median"])
    # The 2.5th and 97.5th percentiles over refits show how much the result
    # moves with the training data; they are not a confidence interval, so
    # they are stored under _refits_ names, which Section 15 does not read
    # as an interval.
    record(f"refit_{tag}_regret_reduction_refits_low", row["regret_reduction_2_5"])
    record(f"refit_{tag}_regret_reduction_refits_high", row["regret_reduction_97_5"])
    stability_rows.append(row)
selection_stability_summary = pd.DataFrame(stability_rows)
selection_stability_models = pd.concat(stability_model_tables, ignore_index=True)
selection_stability_refits = pd.concat(stability_refit_tables, ignore_index=True)
print("8.8 STABILITY OF THE SELECTED ARM")
print(f"  {N_STABILITY_REFITS} refits, each training fold resampled by PDX model; a choice is stable in a PDX "
      "model when it is kept in >= 80% of refits")
print(selection_stability_summary.T.to_string(header=False))

# --- (b) censored arms reinserted in the practical pool -------------------------
termination_rng = np.random.default_rng(SEEDS["termination_selection_bootstrap"])
scenario_pool, _ = censoring_scenario_table(observations.loc[~observations.is_control], FORECAST_ORIGIN_DAY,
                                            FORECAST_HORIZON_DAY, outcome_zeros="floor",
                                            response_rule=FORECAST_WINDOW_RESPONSE)
scenario_pool = scenario_pool.dropna(subset=TRAJECTORY_SET).reset_index(drop=True)
observed_pool = scenario_pool.loc[~scenario_pool.ended_early].reset_index(drop=True)
# The PDX models keep the folds they have in the observed-only pool (8.7);
# censored arms join their model's fold, and a PDX model present only through
# censored arms joins the smallest fold. Without this, GroupKFold rebalances
# the folds when arms are added and the comparison mixes the effect of the
# censored arms with a reshuffle of the folds.
observed_folds = grouped_folds(observed_pool)
observed_assignment = model_fold_assignment(observed_pool, observed_folds)
scenario_folds = folds_from_assignment(scenario_pool, observed_assignment, len(observed_folds))
termination_rows = []
for label, frame, folds in (
        ("observed to day 24 only (8.7 practical pool)",
         observed_pool.assign(outcome_log_volume=observed_pool.growth_stopped), observed_folds),
        ("censored arms added, growth stopped",
         scenario_pool.assign(outcome_log_volume=scenario_pool.growth_stopped), scenario_folds),
        ("censored arms added, growth continued",
         scenario_pool.assign(outcome_log_volume=scenario_pool.growth_continued), scenario_folds)):
    result, _ = evaluate_endpoints(frame, DEPTH_SET, TRAJECTORY_SET, N_BOOTSTRAP, termination_rng, folds=folds)
    termination_rows.append({"pool": label, "censored_arms": int(frame.ended_early.sum()), **result})
selection_with_censored = pd.DataFrame(termination_rows)
assert selection_with_censored.arms.iloc[0] == len(practical_pool), "the observed-only pool differs from 8.7"
# The observed-only row is the 8.7 practical-pool contrast of the trajectory
# and depth sets (same arms, folds and learner): same point estimates, and
# the 8.7 intervals are reported so that one estimate has one interval.
practical_stem = "comparator_practical_trajectory_vs_depth"
assert np.isclose(selection_with_censored.rmse_reduction.iloc[0], RESULTS[f"{practical_stem}_rmse_reduction"]), \
    "8.8 does not reproduce the 8.7 practical pool"
assert np.isclose(selection_with_censored.regret_reduction.iloc[0], RESULTS[f"{practical_stem}_regret_reduction"]), \
    "8.8 does not reproduce the 8.7 practical pool"
for mine, theirs in (("rmse_low", "rmse_reduction_low"), ("rmse_high", "rmse_reduction_high"),
                     ("regret_low", "regret_reduction_low"), ("regret_high", "regret_reduction_high")):
    selection_with_censored.loc[0, mine] = RESULTS[f"{practical_stem}_{theirs}"]
for row, tag in zip(selection_with_censored.iloc[1:].itertuples(index=False), ("stopped", "continued")):
    record(f"censored_selection_{tag}_regret_reduction", row.regret_reduction)
    record(f"censored_selection_{tag}_regret_reduction_low", row.regret_low)
    record(f"censored_selection_{tag}_regret_reduction_high", row.regret_high)
    record(f"censored_selection_{tag}_forecast_reduction", row.rmse_reduction)
print("\n  censored arms in the selection pool (all treated arms, readings at the floor kept)")
print(selection_with_censored[["pool", "arms", "censored_arms", "pdx_models", "rmse_reduction", "rmse_low",
                               "rmse_high", "regret_models", "regret_reduction", "regret_low", "regret_high",
                               "ranking_depth", "ranking_trajectory", "share_different_choice"]]
      .round(3).to_string(index=False))

# --- (c) repeated random partitions of the PDX models into folds ----------------
# Sections 7 and 8.7 use one deterministic partition (GroupKFold). Both
# endpoints are recomputed over N_FOLD_PARTITIONS random grouped partitions,
# with the same learner and arms, to show how much the reported values depend
# on which PDX models happen to share a fold. The spread is a property of the
# cross-validation, not a confidence interval.
N_FOLD_PARTITIONS = 20
partition_rng = np.random.default_rng(SEEDS["fold_partitions"])
partition_rows = []
for tag, (population, frame) in SELECTION_POPULATIONS.items():
    frame = frame.reset_index(drop=True)
    for partition in range(N_FOLD_PARTITIONS):
        folds = random_grouped_folds(frame, N_FOLDS, partition_rng)
        # n_bootstrap 0: only the point estimates are needed for each partition
        result, _ = evaluate_endpoints(frame, DEPTH_SET, TRAJECTORY_SET, 0, partition_rng, folds=folds)
        partition_rows.append({"population": population, "tag": tag, "partition": partition,
                               "rmse_reduction": result["rmse_reduction"],
                               "regret_reduction": result["regret_reduction"],
                               "ranking_depth": result["ranking_depth"],
                               "ranking_trajectory": result["ranking_trajectory"],
                               "share_different_choice": result["share_different_choice"]})
fold_partitions = pd.DataFrame(partition_rows)
partition_summary_rows = []
for tag, block in fold_partitions.groupby("tag", sort=False):
    groupkfold_rmse = (RESULTS["forecast_reduction"] if tag == "primary"
                       else RESULTS["comparator_practical_trajectory_vs_depth_rmse_reduction"])
    groupkfold_regret = (RESULTS["regret_reduction"] if tag == "primary"
                         else RESULTS["comparator_practical_trajectory_vs_depth_regret_reduction"])
    summary = {"population": block.population.iloc[0], "partitions": len(block)}
    for quantity, margin, groupkfold in (("rmse_reduction", FORECAST_MARGIN, groupkfold_rmse),
                                         ("regret_reduction", REGRET_MARGIN, groupkfold_regret)):
        values = block[quantity].to_numpy(float)
        summary.update({
            f"{quantity}_groupkfold": groupkfold,
            f"{quantity}_median": float(np.nanmedian(values)),
            f"{quantity}_min": float(np.nanmin(values)), f"{quantity}_max": float(np.nanmax(values)),
            # share of partitions whose point estimate reaches the margin
            f"{quantity}_share_at_margin": float(np.mean(values >= margin)),
            # where the reported (GroupKFold) value falls among the partitions
            f"{quantity}_groupkfold_percentile": float(100 * np.mean(values <= groupkfold))})
        record(f"partition_{tag}_{quantity}_median", summary[f"{quantity}_median"])
        record(f"partition_{tag}_{quantity}_min", summary[f"{quantity}_min"])
        record(f"partition_{tag}_{quantity}_max", summary[f"{quantity}_max"])
        record(f"partition_{tag}_{quantity}_share_at_margin", summary[f"{quantity}_share_at_margin"])
        record(f"partition_{tag}_{quantity}_groupkfold_percentile", summary[f"{quantity}_groupkfold_percentile"])
    partition_summary_rows.append(summary)
fold_partition_summary = pd.DataFrame(partition_summary_rows)
print(f"\n  {N_FOLD_PARTITIONS} random partitions of the PDX models into {N_FOLDS} folds (spread from the "
      "cross-validation alone, not a confidence interval)")
print(fold_partition_summary.set_index("population").T.round(3).to_string())
save_record(fold_partitions, "fold_partitions")
save_record(fold_partition_summary, "fold_partition_summary")
save_record(selection_stability_summary, "selection_stability_summary")
save_record(selection_stability_models, "selection_stability_models")
save_record(selection_stability_refits, "selection_stability_refits")
save_record(selection_with_censored, "selection_with_censored_arms")
```

```python
# ==========================================================================
# 8.9 Treatment selection: learners, rankings without a model, change from
#     baseline and later horizons (all added after review)
#     (a) Learners. 8.7 compares sets fitted with the same learner. Here the
#         trajectory set fitted by gradient boosting is set against the depth
#         set fitted by least squares (a straight line on the same two depth
#         predictors), and each set's two learners are compared, on both
#         endpoints and in both populations of 8.7. The predictions are
#         those of 8.7 (same arms, folds and fits); only the bootstrap is new,
#         with its own seed, so no 8.7 interval moves.
#     (b) No model. Within each PDX model the arms are ranked directly by one
#         raw day-10 quantity (lower is better): best average response in
#         the window, the window slope of log volume, or the day-10 log
#         volume. Nothing is fitted, so these rankings show how much of the
#         selection result needs a learner at all.
#     (c) Change from baseline. Regret above is in absolute day-24 log
#         volume, so part of a "best arm" can be a mouse whose tumor was
#         small at baseline. Here the outcome is the change in log volume
#         from baseline to day 24: each set's prediction minus the arm's
#         baseline log volume (a predictor in every set) is its predicted
#         change, each set picks the arm with the smallest predicted change,
#         and regret is that arm's observed change minus the smallest
#         observed change among the model's arms.
#     (d) Later horizons. 8.1 repeated the forecasting comparison at three
#         start days and three horizons; here both endpoints, regret and
#         ranking included, are computed at each window, so selection is
#         also judged on volumes up to day 42, after most regrowth has begun.
#         Where a window is the analysis of Section 7, of 8.1 or of 8.7, its
#         interval is taken from there (one interval per estimate).
#     Intervals resample PDX models with the predictions held fixed.
# ==========================================================================

# --- (a) gradient boosting against least squares ------------------------------------
learner_rng = np.random.default_rng(SEEDS["learner_contrast_bootstrap"])
# contrast name -> ((learner, set) of the baseline, (learner, set) of the challenger)
LEARNER_CONTRASTS = {
    "trajectory_gb_vs_depth_ls": (("least squares", "depth"), ("gradient boosting", "trajectory")),
    "depth_ls_vs_depth_gb": (("gradient boosting", "depth"), ("least squares", "depth")),
    "trajectory_ls_vs_depth_ls": (("least squares", "depth"), ("least squares", "trajectory")),
    "trajectory_gb_vs_trajectory_ls": (("least squares", "trajectory"), ("gradient boosting", "trajectory")),
    "baseline_slope_ls_vs_depth_gb": (("gradient boosting", "depth"), ("least squares", "baseline plus slope")),
}
learner_rows = []
for tag, selected in comparator_selected.items():
    selected = selected.reset_index(drop=True)
    population_outcome = selected.outcome_log_volume.to_numpy(float)
    model_clusters = rows_by_cluster(selected.model_id.astype(str))
    for contrast, (baseline, challenger) in LEARNER_CONTRASTS.items():
        baseline_column, challenger_column = f"{baseline[0]}: {baseline[1]}", f"{challenger[0]}: {challenger[1]}"
        reduction, low, high, _ = reduction_with_interval(
            population_outcome, selected[baseline_column].to_numpy(float), selected[challenger_column].to_numpy(float),
            model_clusters, N_BOOTSTRAP, learner_rng)
        contrast_result = paired_selection_contrast(selected, baseline_column, challenger_column, N_BOOTSTRAP,
                                                    learner_rng)
        learner_rows.append({"population": SELECTION_POPULATIONS[tag][0], "contrast": contrast,
                             "challenger": challenger_column, "baseline": baseline_column,
                             "arms": len(selected), "pdx_models": selected.model_id.nunique(),
                             "rmse_baseline": rmse(population_outcome, selected[baseline_column]),
                             "rmse_challenger": rmse(population_outcome, selected[challenger_column]),
                             "rmse_reduction": reduction, "rmse_low": low, "rmse_high": high, **contrast_result})
        stem = f"learner_{tag}_{contrast}"
        for quantity, (value, value_low, value_high) in {
                "rmse_reduction": (reduction, low, high),
                "regret_reduction": (contrast_result["regret_reduction"], contrast_result["regret_low"],
                                     contrast_result["regret_high"]),
                "ranking_difference": (contrast_result["ranking_difference"], contrast_result["ranking_low"],
                                       contrast_result["ranking_high"])}.items():
            record(f"{stem}_{quantity}", value)
            record(f"{stem}_{quantity}_low", value_low)
            record(f"{stem}_{quantity}_high", value_high)
learner_contrasts = pd.DataFrame(learner_rows)
print("8.9 TREATMENT SELECTION: LEARNERS, NO MODEL, CHANGE FROM BASELINE, LATER HORIZONS")
print("  (a) gradient boosting against least squares (predictions of 8.7; paired, PDX models resampled)")
for population, block in learner_contrasts.groupby("population", sort=False):
    print(f"\n  {population}: {block.arms.iloc[0]:,} arms in {block.pdx_models.iloc[0]} PDX models")
    print(block[["challenger", "baseline", "rmse_reduction", "rmse_low", "rmse_high", "regret_baseline",
                 "regret_challenger", "regret_reduction", "regret_low", "regret_high", "ranking_baseline",
                 "ranking_challenger", "ranking_difference", "ranking_low", "ranking_high", "different_choice"]]
          .round(3).to_string(index=False))

# --- (b) rankings by raw day-10 quantities, no model ---------------------------------
free_rng = np.random.default_rng(SEEDS["learner_free_bootstrap"])
# column added to the selection frame -> the window quantity it ranks by (lower is better)
RAW_SCORES = {"raw: best average response in the window": "response_in_window",
              "raw: window slope": "overall_slope",
              "raw: day-10 log volume": "current_log_volume"}
# (baseline, challenger) pairs; raw scores against each other and against fitted sets
FREE_CONTRASTS = {
    "slope_vs_response": ("raw: best average response in the window", "raw: window slope"),
    "day10_volume_vs_response": ("raw: best average response in the window", "raw: day-10 log volume"),
    "slope_vs_gb_depth": ("gradient boosting: depth", "raw: window slope"),
    "gb_trajectory_vs_slope": ("raw: window slope", "gradient boosting: trajectory"),
    "ls_baseline_slope_vs_slope": ("raw: window slope", "least squares: baseline plus slope"),
}
free_rows, free_level_rows = [], []
for tag, selected in comparator_selected.items():
    frame = selected.reset_index(drop=True)
    frame = frame.assign(**{name: frame[column].to_numpy(float) for name, column in RAW_SCORES.items()})
    # Mean regret and ranking accuracy of every raw score and fitted set alone.
    level_columns = [*RAW_SCORES, "gradient boosting: depth", "gradient boosting: trajectory",
                     "least squares: depth", "least squares: baseline plus slope"]
    level_regret = selection_by_model(frame, level_columns)
    level_pairs = within_model_pairs(frame, level_columns)
    for column in level_columns:
        free_level_rows.append({"population": SELECTION_POPULATIONS[tag][0], "ranked_by": column,
                                "mean_regret": float(level_regret[f"{column}_regret"].mean()),
                                "best_arm_share": float(np.mean(level_regret[f"{column}_regret"] <= 1e-12)),
                                "ranking_accuracy": float(level_pairs[column].mean())})
    for contrast, (baseline_column, challenger_column) in FREE_CONTRASTS.items():
        contrast_result = paired_selection_contrast(frame, baseline_column, challenger_column, N_BOOTSTRAP, free_rng)
        free_rows.append({"population": SELECTION_POPULATIONS[tag][0], "contrast": contrast,
                          "challenger": challenger_column, "baseline": baseline_column, **contrast_result})
        stem = f"model_free_{tag}_{contrast}"
        record(f"{stem}_regret_reduction", contrast_result["regret_reduction"])
        record(f"{stem}_regret_reduction_low", contrast_result["regret_low"])
        record(f"{stem}_regret_reduction_high", contrast_result["regret_high"])
        record(f"{stem}_ranking_difference", contrast_result["ranking_difference"])
        record(f"{stem}_ranking_difference_low", contrast_result["ranking_low"])
        record(f"{stem}_ranking_difference_high", contrast_result["ranking_high"])
model_free_levels = pd.DataFrame(free_level_rows)
model_free_selection = pd.DataFrame(free_rows)
for row in model_free_levels.itertuples(index=False):
    population_tag = "primary" if row.population == SELECTION_POPULATIONS["primary"][0] else "practical"
    name_tag = "".join(character if character.isalnum() else "_" for character in row.ranked_by)
    record(f"selection_level_{population_tag}_{name_tag}_ranking_accuracy", row.ranking_accuracy)
    record(f"selection_level_{population_tag}_{name_tag}_mean_regret", row.mean_regret)
print("\n  (b) ranking arms within each PDX model by one raw day-10 quantity, no model (lower is better)")
print(model_free_levels.round(3).to_string(index=False))
print("\n  paired contrasts (challenger against baseline; PDX models resampled)")
print(model_free_selection[["population", "challenger", "baseline", "regret_baseline", "regret_challenger",
                            "regret_reduction", "regret_low", "regret_high", "ranking_baseline",
                            "ranking_challenger", "ranking_difference", "ranking_low", "ranking_high"]]
      .round(3).to_string(index=False))

# --- (c) regret on the change from baseline ---------------------------------------
change_rng = np.random.default_rng(SEEDS["change_outcome_bootstrap"])
CHANGE_COLUMNS = ["gradient boosting: depth", "gradient boosting: trajectory", "least squares: depth",
                  "least squares: baseline plus slope"]
CHANGE_CONTRASTS = {
    "trajectory_vs_depth": ("gradient boosting: depth", "gradient boosting: trajectory"),
    "ls_baseline_slope_vs_ls_depth": ("least squares: depth", "least squares: baseline plus slope"),
    "trajectory_gb_vs_depth_ls": ("least squares: depth", "gradient boosting: trajectory"),
}
change_rows = []
for tag, selected in comparator_selected.items():
    frame = selected.reset_index(drop=True)
    baseline_log = frame.baseline_log_volume.to_numpy(float)
    # Outcome and every prediction expressed as change from the arm's own baseline.
    change_frame = frame.assign(outcome_log_volume=frame.outcome_log_volume.to_numpy(float) - baseline_log,
                                **{column: frame[column].to_numpy(float) - baseline_log for column in CHANGE_COLUMNS})
    for contrast, (baseline_column, challenger_column) in CHANGE_CONTRASTS.items():
        contrast_result = paired_selection_contrast(change_frame, baseline_column, challenger_column, N_BOOTSTRAP,
                                                    change_rng)
        change_rows.append({"population": SELECTION_POPULATIONS[tag][0], "outcome": "change from baseline to day 24",
                            "contrast": contrast, "challenger": challenger_column, "baseline": baseline_column,
                            **contrast_result})
        stem = f"change_{tag}_{contrast}"
        record(f"{stem}_regret_reduction", contrast_result["regret_reduction"])
        record(f"{stem}_regret_reduction_low", contrast_result["regret_low"])
        record(f"{stem}_regret_reduction_high", contrast_result["regret_high"])
        record(f"{stem}_ranking_difference", contrast_result["ranking_difference"])
        record(f"{stem}_ranking_difference_low", contrast_result["ranking_low"])
        record(f"{stem}_ranking_difference_high", contrast_result["ranking_high"])
selection_change_from_baseline = pd.DataFrame(change_rows)
print("\n  (c) selection judged on the change in log volume from baseline to day 24 (PDX models resampled)")
print(selection_change_from_baseline[["population", "challenger", "baseline", "regret_models", "regret_baseline",
                                      "regret_challenger", "regret_reduction", "regret_low", "regret_high",
                                      "ranking_baseline", "ranking_challenger", "ranking_difference",
                                      "ranking_low", "ranking_high", "different_choice"]]
      .round(3).to_string(index=False))

# --- (d) both endpoints at later start days and horizons ------------------------------
horizon_rng = np.random.default_rng(SEEDS["long_horizon_selection_bootstrap"])
# population tag -> (label, readings at the floor in the outcome, arms allowed)
HORIZON_POPULATIONS = {"primary": (SELECTION_POPULATIONS["primary"][0], FORECAST_OUTCOME_ZEROS, post_nadir_ids),
                       "practical": (SELECTION_POPULATIONS["practical"][0], "floor", None)}
rolling_lookup = rolling_origins.set_index(["start_day", "horizon_day"])
horizon_rows = []
for origin_day in (10, 14, 21):
    for lead_days in (10, 14, 21):
        horizon_day = origin_day + lead_days
        for tag, (population, zero_rule, arm_ids) in HORIZON_POPULATIONS.items():
            table = (build_forecast_table(observations.loc[~observations.is_control], origin_day, horizon_day,
                                          arm_ids=arm_ids, outcome_zeros=zero_rule)
                     .dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"]).reset_index(drop=True))
            if len(table) < 60 or table.model_id.nunique() < 2:
                continue
            result, _ = evaluate_endpoints(table, DEPTH_SET, TRAJECTORY_SET, N_BOOTSTRAP, horizon_rng)
            source = "8.9"
            is_primary_window = (origin_day, horizon_day) == (FORECAST_ORIGIN_DAY, FORECAST_HORIZON_DAY)
            if tag == "primary" and is_primary_window:
                # The analysis of Section 7: its point estimates and intervals.
                assert np.isclose(result["rmse_reduction"], RESULTS["forecast_reduction"])
                assert np.isclose(result["regret_reduction"], RESULTS["regret_reduction"])
                result.update(rmse_low=RESULTS["forecast_reduction_low"], rmse_high=RESULTS["forecast_reduction_high"],
                              regret_low=RESULTS["regret_reduction_low"], regret_high=RESULTS["regret_reduction_high"])
                source = "7.2 and 7.3"
            elif tag == "practical" and is_primary_window:
                # The practical pool of 8.7: its point estimates and intervals.
                stem = "comparator_practical_trajectory_vs_depth"
                assert np.isclose(result["rmse_reduction"], RESULTS[f"{stem}_rmse_reduction"])
                assert np.isclose(result["regret_reduction"], RESULTS[f"{stem}_regret_reduction"])
                result.update(rmse_low=RESULTS[f"{stem}_rmse_reduction_low"],
                              rmse_high=RESULTS[f"{stem}_rmse_reduction_high"],
                              regret_low=RESULTS[f"{stem}_regret_reduction_low"],
                              regret_high=RESULTS[f"{stem}_regret_reduction_high"])
                source = "8.7"
            elif tag == "primary" and (origin_day, horizon_day) in rolling_lookup.index:
                # The window of 8.1 (same arms, learner and folds): its
                # forecasting interval; the selection endpoints are new here.
                window = rolling_lookup.loc[(origin_day, horizon_day)]
                assert np.isclose(result["rmse_reduction"], window.relative_reduction)
                result.update(rmse_low=float(window.low), rmse_high=float(window.high))
                source = "8.1 (forecasting) and 8.9 (selection)"
            horizon_rows.append({"population": population, "start_day": origin_day, "horizon_day": horizon_day,
                                 "interval_source": source, **result})
            stem = f"horizon_{tag}_{origin_day}_{horizon_day}"
            record(f"{stem}_rmse_reduction", result["rmse_reduction"])
            record(f"{stem}_rmse_reduction_low", result["rmse_low"])
            record(f"{stem}_rmse_reduction_high", result["rmse_high"])
            record(f"{stem}_regret_reduction", result["regret_reduction"])
            record(f"{stem}_regret_reduction_low", result["regret_low"])
            record(f"{stem}_regret_reduction_high", result["regret_high"])
            record(f"{stem}_ranking_accuracy_depth", result["ranking_depth"])
            record(f"{stem}_ranking_accuracy_trajectory", result["ranking_trajectory"])
selection_by_horizon = pd.DataFrame(horizon_rows)
print("\n  (d) both endpoints at each start day and horizon (trajectory against depth, gradient boosting)")
print(selection_by_horizon[["population", "start_day", "horizon_day", "arms", "pdx_models", "rmse_reduction",
                            "rmse_low", "rmse_high", "regret_models", "regret_reduction", "regret_low",
                            "regret_high", "ranking_depth", "ranking_trajectory", "share_different_choice",
                            "interval_source"]].round(3).to_string(index=False))
save_record(learner_contrasts, "learner_contrasts")
save_record(model_free_levels, "model_free_ranking_levels")
save_record(model_free_selection, "model_free_selection")
save_record(selection_change_from_baseline, "selection_change_from_baseline")
save_record(selection_by_horizon, "selection_by_horizon")
```

## 9 · Molecular correlates of depth and durability

Expression signatures are scored from the RNA-seq data of the source study. Genes are ranked within each PDX model, and a signature score is the mean percentile rank of its genes minus 0.5. Three tumor-cell programs are scored: epithelial–mesenchymal transition (EMT), stemness, and cell-cycle arrest. The arrest score is signed: the mean rank of four arrest genes minus the mean rank of six proliferation genes. A stromal score serves as a proxy for stromal content. The genes of every set are listed in `gene_sets.csv`. The signature set holds the three program scores. The context set holds the stromal score, baseline log volume, the PDX model's untreated growth rate, and tumor-type indicators.

The analysis set is post-nadir arms with at least five measurements, in PDX models that have both an untreated growth rate and expression data. Gastric cancer models have no expression data, so gastric arms are not included. Depth and durability are each regressed on each set by ridge regression with five folds grouped by PDX model. The penalty is chosen within each training fold by an inner grouped cross-validation. Performance is the out-of-fold R².

The null distribution comes from 500 shuffles that break the link between features and outcome but keep tumor-type structure. PDX-model-level features are reassigned between PDX models of the same tumor type, and baseline volume is shuffled across arms of the same tumor type. Tumor-type indicators cannot change under either shuffle, so a context set that beats its null carries information beyond tumor type. Permutation P values are (1 + number of null values ≥ observed) / (1 + 500).

The dissociation statistic is D = [R²(signatures, durability) − R²(context, durability)] − [R²(signatures, depth) − R²(context, depth)]. D is positive when the signatures explain durability relatively better than they explain depth. Its interval comes from 2,000 bootstrap resamples of PDX models.

**Robustness to the choice of gene sets (9.3).** The three program scores of 9.1 come from curated marker genes. Section 9.3 rebuilds them from the MSigDB hallmark collection, scored by the same rank rule, and repeats the block R², shuffle null and dissociation statistic with the hallmark signature block in place of the curated one; the context block is unchanged. EMT is the hallmark epithelial–mesenchymal transition set. Stemness, which has no hallmark set, is represented by the mean score of the three developmental self-renewal pathways (Wnt/β-catenin, Notch and Hedgehog signalling). Cell-cycle arrest is the hallmark p53 pathway score minus the mean of the E2F-target and G2/M-checkpoint scores, the same arrest-minus-proliferation construction as in 9.1. The section runs when a hallmark library is present in `data/msigdb/` (a GMT file, or the Enrichr `MSigDB_Hallmark_2020` text library). When none is present it downloads the Enrichr library once (`FETCH_HALLMARK_LIBRARY` in Section 1.1), keeps it only if it holds every set needed, and saves it there for later runs; if that fails it prints a skip line.

**Treatment-adjusted models (9.4).** Each PDX model received up to 21 treatments, and an arm's depth and durability depend far more on the drug than on the model's signature scores. The block R² of 9.2 pools arms across treatments with no treatment term, so drug-to-drug variance sits in the residual of both blocks. Section 9.4 repeats the block R², the shuffle null and the dissociation statistic with each outcome expressed relative to its treatment's mean, computed within each training fold; R² is then the share of within-treatment variance a block explains beyond a treatment-mean model. The R² of the treatment means alone is printed beside it.

**Inputs:** `sheets["expression"]`, `post_nadir_arms`, `arm_design`, `untreated_growth`, `data/msigdb/` (optional).
**Outputs:** `signature_scores`, `gene_set_table` (table `gene_sets.csv`), `molecular_frame`, `molecular_blocks`, `dissociation`, `hallmark_blocks`, `molecular_blocks_within_treatment`.

```python
# ==========================================================================
# 9.1 Signature scores
# ==========================================================================

set_section("9")
expression = sheets["expression"].rename(columns={sheets["expression"].columns[0]: "gene"})
expression["gene"] = expression.gene.astype(str).str.strip().str.upper()
duplicated_genes = int(expression.gene.duplicated().sum())
expression = expression.groupby("gene").mean(numeric_only=True)       # duplicate symbols averaged
expression = expression.loc[expression.notna().mean(axis=1) > 0.5]
expression_ranks = expression.rank(axis=0, pct=True)

GENE_SETS = {
    "EMT": ["VIM", "CDH2", "FN1", "SNAI1", "SNAI2", "TWIST1", "ZEB1", "ZEB2", "MMP2", "MMP9",
            "SPARC", "TGFB1", "COL1A1", "COL1A2", "SERPINE1"],
    "stemness": ["SOX2", "POU5F1", "NANOG", "KLF4", "MYC", "ALDH1A1", "PROM1", "CD44", "LGR5",
                 "BMI1", "NOTCH1"],
    "cell-cycle arrest": ["CDKN1A", "CDKN1B", "CDKN2A", "RB1"],
    "proliferation": ["MKI67", "PCNA", "CCNB1", "TOP2A", "AURKA", "PLK1"],
    "stromal": ["PECAM1", "VWF", "PTPRC", "CD68", "COL3A1", "ACTA2", "PDGFRB", "THY1"],
}
# Source for each gene set, written to gene_sets.csv. These are marker lists
# curated for this study, not published signatures; a set taken from a
# publication should name that publication here. An empty entry is printed
# below as a set without a recorded source.
GENE_SET_SOURCES = {
    "EMT": "Curated for this study: EMT transcription factors (SNAI1, SNAI2, TWIST1, ZEB1, ZEB2) "
           "and mesenchymal and extracellular-matrix markers",
    "stemness": "Curated for this study: pluripotency factors (SOX2, POU5F1, NANOG, KLF4, MYC) "
                "and cancer stem-cell markers",
    "cell-cycle arrest": "Curated for this study: CDK inhibitors (CDKN1A, CDKN1B, CDKN2A) and RB1",
    "proliferation": "Curated for this study: S-phase and G2/M proliferation markers",
    "stromal": "Curated for this study: endothelial (PECAM1, VWF), immune (PTPRC, CD68) and "
               "fibroblast (COL3A1, ACTA2, PDGFRB, THY1) markers",
}

set_scores, gene_set_rows = {}, []
for name, genes in GENE_SETS.items():
    present = [gene for gene in genes if gene in expression_ranks.index]
    set_scores[name] = expression_ranks.loc[present].mean(axis=0) - 0.5
    gene_set_rows.append({"gene_set": name, "genes": ", ".join(genes), "genes_in_data": len(present),
                          "genes_in_set": len(genes), "source": GENE_SET_SOURCES[name]})
gene_set_table = pd.DataFrame(gene_set_rows)
signature_scores = pd.DataFrame({
    "signature_emt": set_scores["EMT"],
    "signature_stemness": set_scores["stemness"],
    "signature_cell_cycle_arrest": set_scores["cell-cycle arrest"] - set_scores["proliferation"],
    "context_stromal": set_scores["stromal"],
})
signature_scores.index = signature_scores.index.astype(str)

print("9.1 SIGNATURE SCORES")
print(f"  expression matrix {expression.shape[0]:,} genes x {expression.shape[1]:,} PDX models "
      f"({duplicated_genes} duplicate gene symbols averaged)")
print(gene_set_table[["gene_set", "genes_in_data", "genes_in_set"]].to_string(index=False))
missing_sources = [name for name, source in GENE_SET_SOURCES.items() if not source]
if missing_sources:
    print(f"  gene sets without a recorded source (GENE_SET_SOURCES): {missing_sources}")
print("  Spearman correlation of each program score with the stromal score")
for column in ("signature_emt", "signature_stemness", "signature_cell_cycle_arrest"):
    value, _ = spearman(signature_scores[column], signature_scores.context_stromal)
    print(f"    {column:<30} ρ = {value:.3f}")

# ==========================================================================
# 9.2 Analysis set, block R² against the shuffle null, and the dissociation
# ==========================================================================

molecular_frame = post_nadir_arms.loc[
    post_nadir_arms.arm_id.map(arm_design.n_measurements).ge(MIN_MEASUREMENTS_PER_ARM)
    & post_nadir_arms.model_id.isin(untreated_growth.index)
    & post_nadir_arms.model_id.astype(str).isin(signature_scores.index),
    ["arm_id", "model_id", "tumor_type", "treatment", "best_average_response", "post_nadir_growth_rate",
     "baseline_volume"]].reset_index(drop=True)
molecular_frame = molecular_frame.join(signature_scores, on=molecular_frame.model_id.astype(str))
molecular_frame["context_untreated_growth"] = molecular_frame.model_id.map(untreated_growth)
molecular_frame["context_baseline_log_volume"] = np.log(molecular_frame.baseline_volume.clip(lower=VOLUME_FLOOR))
tumor_type_columns = []
for tumor_type in sorted(molecular_frame.tumor_type.unique()):
    molecular_frame[f"context_type_{tumor_type}"] = (molecular_frame.tumor_type == tumor_type).astype(float)
    tumor_type_columns.append(f"context_type_{tumor_type}")
FEATURE_BLOCKS = {
    "signatures": {"model": ["signature_emt", "signature_stemness", "signature_cell_cycle_arrest"],
                   "arm": [], "fixed": []},
    "context": {"model": ["context_stromal", "context_untreated_growth"],
                "arm": ["context_baseline_log_volume"], "fixed": tumor_type_columns},
}
MOLECULAR_OUTCOMES = {"depth": "best_average_response", "durability": "post_nadir_growth_rate"}
N_PERMUTATIONS = 500
print("\n9.2 ANALYSIS SET")
print(f"  arms {record('molecular_n_arms', len(molecular_frame)):,} in "
      f"{record('molecular_n_models', molecular_frame.model_id.nunique())} PDX models; tumor types "
      f"{sorted(molecular_frame.tumor_type.unique())}")

null_rng = np.random.default_rng(SEEDS["molecular_null"])
block_rows = []
for outcome_label, outcome_column in MOLECULAR_OUTCOMES.items():
    for block_name, block in FEATURE_BLOCKS.items():
        features = block_features(block)
        observed_r2 = block_r2(molecular_frame, features, outcome_column)
        null_r2 = np.array([block_r2(permute_within_tumor_type(molecular_frame, block["model"],
                                                               block["arm"], null_rng),
                                     features, outcome_column)
                            for _ in range(N_PERMUTATIONS)])
        block_rows.append({"outcome": outcome_label, "block": block_name, "features": len(features),
                           "r2": observed_r2, "null_mean": null_r2.mean(),
                           "null_95th": np.percentile(null_r2, 95),
                           "permutation_p": (1 + np.sum(null_r2 >= observed_r2)) / (1 + N_PERMUTATIONS)})
        record(f"r2_{block_name}_{outcome_label}", observed_r2)
        record(f"null95_{block_name}_{outcome_label}", float(np.percentile(null_r2, 95)))
        record(f"permutation_p_{block_name}_{outcome_label}", float(block_rows[-1]["permutation_p"]))
molecular_blocks = pd.DataFrame(block_rows)
print(f"\n  out-of-fold R² against {N_PERMUTATIONS} within-tumor-type shuffles")
print(molecular_blocks.round(4).to_string(index=False))

dissociation = dissociation_statistic(molecular_frame, FEATURE_BLOCKS, MOLECULAR_OUTCOMES)
dissociation_draws = cluster_bootstrap(
    rows_by_cluster(molecular_frame.model_id.astype(str)),
    lambda rows: dissociation_statistic(molecular_frame.iloc[rows].reset_index(drop=True),
                                        FEATURE_BLOCKS, MOLECULAR_OUTCOMES),
    N_BOOTSTRAP, np.random.default_rng(SEEDS["molecular_bootstrap"]))
dissociation_low, dissociation_high = np.percentile(dissociation_draws, [2.5, 97.5])
record("dissociation", dissociation)
record("dissociation_low", dissociation_low)
record("dissociation_high", dissociation_high)
print(f"\n  dissociation statistic D   {format_ci(dissociation, dissociation_low, dissociation_high, 1, 3, '')}"
      f", {len(dissociation_draws):,} resamples of PDX models")
save_record(molecular_blocks, "molecular_blocks")
save_record(molecular_frame, "molecular_frame")
```

```python
# ==========================================================================
# 9.3 Robustness to the choice of gene sets: MSigDB hallmark programs
#     The curated signature block of 9.1 is rebuilt from hallmark sets and
#     scored by the same rank rule. The analysis set, the context block, the
#     ridge regression, the shuffle null and the dissociation statistic are
#     those of 9.2; only the values in the three signature columns change.
#     The context block's R² is therefore the one printed in 9.2.
# ==========================================================================

# Hallmark sets (normalized names, see normalized_set_name) behind each
# program. A program score is the mean score of its "plus" sets minus the mean
# score of its "minus" sets.
HALLMARK_PROGRAMS = {
    "signature_emt": {"plus": ["EPITHELIALMESENCHYMALTRANSITION"], "minus": []},
    "signature_stemness": {"plus": ["WNTBETACATENINSIGNALING", "NOTCHSIGNALING", "HEDGEHOGSIGNALING"],
                           "minus": []},
    "signature_cell_cycle_arrest": {"plus": ["P53PATHWAY"], "minus": ["E2FTARGETS", "G2MCHECKPOINT"]},
}
assert set(HALLMARK_PROGRAMS) == set(FEATURE_BLOCKS["signatures"]["model"]), \
    "the hallmark programs must replace the curated signature columns one for one"

hallmark_files = (sorted([*MSIGDB_DIR.glob("*.gmt"), *MSIGDB_DIR.glob("*.txt")])
                  if MSIGDB_DIR.exists() else [])
HALLMARK_AVAILABLE = False
hallmark_blocks, hallmark_sets_table = pd.DataFrame(), pd.DataFrame()
print("9.3 ROBUSTNESS TO THE CHOICE OF GENE SETS (MSigDB hallmark collection)")
# No library on disk: download the Enrichr text version of the hallmark
# collection once (FETCH_HALLMARK_LIBRARY, Section 1.1). It is kept only if it
# parses and holds every set the programs above need, and it is saved under
# data/msigdb/, so later runs read the saved file and its checksum enters the
# manifest. A failed download (no network, a changed address) only skips 9.3.
hallmark_downloaded = False
if not hallmark_files and FETCH_HALLMARK_LIBRARY:
    needed_sets = {name for program in HALLMARK_PROGRAMS.values() for name in program["plus"] + program["minus"]}
    downloaded_path = MSIGDB_DIR / HALLMARK_LIBRARY_FILE
    try:
        with urllib.request.urlopen(HALLMARK_LIBRARY_URL, timeout=60) as response:
            library_text = response.read().decode("utf-8", errors="replace")
        MSIGDB_DIR.mkdir(parents=True, exist_ok=True)
        downloaded_path.write_text(library_text, encoding="utf-8")
        if needed_sets <= set(read_gene_set_library(downloaded_path)):
            hallmark_files, hallmark_downloaded = [downloaded_path], True
            print(f"  downloaded {HALLMARK_LIBRARY_URL}\n  saved as {relative(downloaded_path)}")
        else:
            downloaded_path.unlink()
            print("  the downloaded library lacks a needed hallmark set, so it was not kept")
    except Exception as error:          # any download or parsing failure only skips this section
        if downloaded_path.exists() and not hallmark_downloaded:
            downloaded_path.unlink()
        print(f"  download of the hallmark library failed ({type(error).__name__}: {error})")
record("hallmark_library_downloaded", int(hallmark_downloaded))
if not hallmark_files:
    print(f"  skipped: no .gmt or .txt hallmark library in {relative(MSIGDB_DIR)}. Put the GMT file or the Enrichr "
          f"text library there ({HALLMARK_LIBRARY_URL}), or set FETCH_HALLMARK_LIBRARY = True in Section 1.1")
else:
    library_path = hallmark_files[0]
    library = read_gene_set_library(library_path)
    needed = sorted({name for program in HALLMARK_PROGRAMS.values()
                     for name in program["plus"] + program["minus"]})
    missing_sets = [name for name in needed if name not in library]
    print(f"  library {relative(library_path)}: {len(library)} sets "
          f"(SHA-256 {sha256_of_file(library_path)[:16]})")
    if missing_sets:
        print(f"  skipped: the library lacks {missing_sets}")
    else:
        HALLMARK_AVAILABLE = True
        set_rows, set_score = [], {}
        for name in needed:
            written, genes = library[name]
            set_score[name], present = gene_set_score(expression_ranks, genes)
            set_rows.append({"hallmark_set": written, "genes_in_set": len(genes), "genes_in_data": present,
                             "coverage": present / len(genes) if genes else np.nan})
        hallmark_sets_table = pd.DataFrame(set_rows)
        print(hallmark_sets_table.round(3).to_string(index=False))

        hallmark_scores = pd.DataFrame({
            program: (pd.concat([set_score[name] for name in spec["plus"]], axis=1).mean(axis=1)
                      - (pd.concat([set_score[name] for name in spec["minus"]], axis=1).mean(axis=1)
                         if spec["minus"] else 0.0))
            for program, spec in HALLMARK_PROGRAMS.items()})
        hallmark_scores.index = hallmark_scores.index.astype(str)

        # How closely each hallmark program follows its curated counterpart and
        # the stromal proxy, across PDX models with expression data.
        print("\n  Spearman correlation across PDX models: hallmark program with the curated program, "
              "and with the stromal score")
        for program in HALLMARK_PROGRAMS:
            shared = hallmark_scores.index.intersection(signature_scores.index)
            with_curated, _ = spearman(hallmark_scores.loc[shared, program], signature_scores.loc[shared, program])
            with_stroma, _ = spearman(hallmark_scores.loc[shared, program],
                                      signature_scores.loc[shared, "context_stromal"])
            print(f"    {program:<30} curated ρ = {with_curated:+.3f}   stromal ρ = {with_stroma:+.3f}")

        # The 9.2 analysis frame with the hallmark values in the signature columns.
        hallmark_frame = molecular_frame.copy()
        for program in HALLMARK_PROGRAMS:
            hallmark_frame[program] = hallmark_frame.model_id.astype(str).map(hallmark_scores[program])
        print(f"\n  arms {len(hallmark_frame):,} in {hallmark_frame.model_id.nunique()} PDX models "
              f"(the 9.2 analysis set); arms without a hallmark score "
              f"{int(hallmark_frame[list(HALLMARK_PROGRAMS)].isna().any(axis=1).sum())}")

        hallmark_null_rng = np.random.default_rng(SEEDS["hallmark_null"])
        signature_features = block_features(FEATURE_BLOCKS["signatures"])
        hallmark_rows = []
        for outcome_label, outcome_column in MOLECULAR_OUTCOMES.items():
            observed_r2 = block_r2(hallmark_frame, signature_features, outcome_column)
            null_r2 = np.array([block_r2(permute_within_tumor_type(
                hallmark_frame, FEATURE_BLOCKS["signatures"]["model"], [], hallmark_null_rng),
                signature_features, outcome_column) for _ in range(N_PERMUTATIONS)])
            curated = molecular_blocks.set_index(["outcome", "block"]).loc[(outcome_label, "signatures")]
            hallmark_rows.append({"outcome": outcome_label, "r2_hallmark": observed_r2,
                                  "null_mean": null_r2.mean(), "null_95th": np.percentile(null_r2, 95),
                                  "permutation_p": (1 + np.sum(null_r2 >= observed_r2)) / (1 + N_PERMUTATIONS),
                                  "r2_curated": curated.r2, "permutation_p_curated": curated.permutation_p})
            record(f"r2_hallmark_signatures_{outcome_label}", observed_r2)
            record(f"null95_hallmark_signatures_{outcome_label}", float(np.percentile(null_r2, 95)))
            record(f"permutation_p_hallmark_signatures_{outcome_label}", float(hallmark_rows[-1]["permutation_p"]))
        hallmark_blocks = pd.DataFrame(hallmark_rows)
        print(f"\n  signature block, out-of-fold R² against {N_PERMUTATIONS} within-tumor-type shuffles "
              "(hallmark beside curated)")
        print(hallmark_blocks.round(4).to_string(index=False))

        hallmark_dissociation = dissociation_statistic(hallmark_frame, FEATURE_BLOCKS, MOLECULAR_OUTCOMES)
        hallmark_draws = cluster_bootstrap(
            rows_by_cluster(hallmark_frame.model_id.astype(str)),
            lambda rows: dissociation_statistic(hallmark_frame.iloc[rows].reset_index(drop=True),
                                                FEATURE_BLOCKS, MOLECULAR_OUTCOMES),
            N_BOOTSTRAP, np.random.default_rng(SEEDS["hallmark_bootstrap"]))
        hallmark_low, hallmark_high = np.percentile(hallmark_draws, [2.5, 97.5])
        record("dissociation_hallmark", hallmark_dissociation)
        record("dissociation_hallmark_low", float(hallmark_low))
        record("dissociation_hallmark_high", float(hallmark_high))
        print(f"\n  dissociation statistic D, hallmark signatures   "
              f"{format_ci(hallmark_dissociation, hallmark_low, hallmark_high, 1, 3, '')}, "
              f"{len(hallmark_draws):,} resamples of PDX models")
        print(f"  dissociation statistic D, curated signatures (9.2)   "
              f"{format_ci(RESULTS['dissociation'], RESULTS['dissociation_low'], RESULTS['dissociation_high'], 1, 3, '')}")
        save_record(hallmark_blocks, "molecular_blocks_hallmark")
        save_record(hallmark_sets_table, "hallmark_sets")
```

```python
# ==========================================================================
# 9.4 Molecular blocks with each outcome expressed within treatment
#     Each PDX model received up to 21 treatments, and an arm's depth and
#     durability depend far more on the drug than on the model's signature
#     scores. 9.2 pools arms across 61 treatments with no treatment term, so
#     drug-to-drug variance sits in the residual of both blocks and caps
#     their R². Here each outcome is expressed relative to its treatment's
#     mean, computed within each training fold (grouped_ridge_r2 with
#     treatments), and R² is the share of within-treatment variance a block
#     explains beyond a treatment-mean model. The analysis set, blocks,
#     ridge regression, shuffle null and dissociation statistic are those of
#     9.2; the treatment column is never shuffled.
# ==========================================================================

treatment_null_rng = np.random.default_rng(SEEDS["molecular_treatment_null"])
treatment_variance = {label: treatment_mean_r2(molecular_frame, column)
                      for label, column in MOLECULAR_OUTCOMES.items()}
for label, value in treatment_variance.items():
    record(f"r2_treatment_means_{label}", value)

within_rows = []
pooled_blocks = molecular_blocks.set_index(["outcome", "block"])
for outcome_label, outcome_column in MOLECULAR_OUTCOMES.items():
    for block_name, block in FEATURE_BLOCKS.items():
        features = block_features(block)
        observed_r2 = block_r2(molecular_frame, features, outcome_column, adjust_treatment=True)
        null_r2 = np.array([block_r2(permute_within_tumor_type(molecular_frame, block["model"], block["arm"],
                                                               treatment_null_rng),
                                     features, outcome_column, adjust_treatment=True)
                            for _ in range(N_PERMUTATIONS)])
        within_rows.append({"outcome": outcome_label, "block": block_name, "features": len(features),
                            "r2_within_treatment": observed_r2, "null_mean": null_r2.mean(),
                            "null_95th": np.percentile(null_r2, 95),
                            "permutation_p": (1 + np.sum(null_r2 >= observed_r2)) / (1 + N_PERMUTATIONS),
                            "r2_pooled_9_2": pooled_blocks.loc[(outcome_label, block_name)].r2,
                            "permutation_p_pooled_9_2": pooled_blocks.loc[(outcome_label, block_name)].permutation_p})
        record(f"r2_within_treatment_{block_name}_{outcome_label}", observed_r2)
        record(f"null95_within_treatment_{block_name}_{outcome_label}", float(np.percentile(null_r2, 95)))
        record(f"permutation_p_within_treatment_{block_name}_{outcome_label}",
               float(within_rows[-1]["permutation_p"]))
molecular_blocks_within_treatment = pd.DataFrame(within_rows)

dissociation_within = dissociation_statistic(molecular_frame, FEATURE_BLOCKS, MOLECULAR_OUTCOMES,
                                             adjust_treatment=True)
dissociation_within_draws = cluster_bootstrap(
    rows_by_cluster(molecular_frame.model_id.astype(str)),
    lambda rows: dissociation_statistic(molecular_frame.iloc[rows].reset_index(drop=True),
                                        FEATURE_BLOCKS, MOLECULAR_OUTCOMES, adjust_treatment=True),
    N_BOOTSTRAP, np.random.default_rng(SEEDS["molecular_treatment_bootstrap"]))
dissociation_within_low, dissociation_within_high = np.percentile(dissociation_within_draws, [2.5, 97.5])
record("dissociation_within_treatment", dissociation_within)
record("dissociation_within_treatment_low", float(dissociation_within_low))
record("dissociation_within_treatment_high", float(dissociation_within_high))

print("9.4 MOLECULAR BLOCKS WITHIN TREATMENT")
print(f"  out-of-fold R² of the treatment means alone: depth {treatment_variance['depth']:.3f}, "
      f"durability {treatment_variance['durability']:.3f}")
print(f"  R² beyond a treatment-mean model, against {N_PERMUTATIONS} within-tumor-type shuffles "
      "(the pooled values of 9.2 beside them)")
print(molecular_blocks_within_treatment.round(4).to_string(index=False))
print(f"\n  dissociation statistic D within treatment   "
      f"{format_ci(dissociation_within, dissociation_within_low, dissociation_within_high, 1, 3, '')}, "
      f"{len(dissociation_within_draws):,} resamples of PDX models")
print(f"  dissociation statistic D pooled (9.2)        "
      f"{format_ci(RESULTS['dissociation'], RESULTS['dissociation_low'], RESULTS['dissociation_high'], 1, 3, '')}")

# The hallmark signature block of 9.3, within treatment, when the library is present.
molecular_blocks_hallmark_within = pd.DataFrame()
if HALLMARK_AVAILABLE:
    hallmark_within_rows = []
    signature_features = block_features(FEATURE_BLOCKS["signatures"])
    for outcome_label, outcome_column in MOLECULAR_OUTCOMES.items():
        observed_r2 = block_r2(hallmark_frame, signature_features, outcome_column, adjust_treatment=True)
        null_r2 = np.array([block_r2(permute_within_tumor_type(
            hallmark_frame, FEATURE_BLOCKS["signatures"]["model"], [], treatment_null_rng),
            signature_features, outcome_column, adjust_treatment=True) for _ in range(N_PERMUTATIONS)])
        hallmark_within_rows.append({"outcome": outcome_label, "r2_within_treatment": observed_r2,
                                     "null_95th": np.percentile(null_r2, 95),
                                     "permutation_p": (1 + np.sum(null_r2 >= observed_r2)) / (1 + N_PERMUTATIONS)})
        record(f"r2_within_treatment_hallmark_{outcome_label}", observed_r2)
        record(f"permutation_p_within_treatment_hallmark_{outcome_label}",
               float(hallmark_within_rows[-1]["permutation_p"]))
    molecular_blocks_hallmark_within = pd.DataFrame(hallmark_within_rows)
    print("\n  hallmark signature block within treatment")
    print(molecular_blocks_hallmark_within.round(4).to_string(index=False))
save_record(molecular_blocks_within_treatment, "molecular_blocks_within_treatment")
```

## 10 · Analysis of combinations

A combination is evaluated in a PDX model only if every one of its components was also tested alone in that model. The combination is then compared with its own single agents on each endpoint. Depth gain is the lowest best average response among the single agents minus the combination's best average response, using the published value of every treated arm. Durability gain is the lowest durability value among the single agents minus the combination's. An arm's durability value is its post-nadir growth rate where one is defined, including arms whose nadir is their first measurement (they never shrank; the rate is their growth after the first measurement). An arm whose smallest volume is its last measurement showed no regrowth during follow-up: its durability is right-censored, and it counts as more durable than any arm with a measured regrowth rate (a value of −∞, so the gains it enters are +∞ or −∞). An arm with a single measurement after an interior nadir had begun to regrow but its rate cannot be measured, so it has no value. A durability gain is computed only when the combination and every one of its components have a value, so that the best component is the best of all of them. A positive gain means the combination is better than its best component on that endpoint. Evaluations are classified by depth gain > 5 percentage points and durability gain > 0.005 per day. Evaluations under the rule of the analysis first run (depth only from arms with a defined post-nadir growth rate, which left out arms still shrinking at their last measurement; durability only from post-nadir arms; the best component taken among those that had a value) are computed beside the corrected ones and compared combination by combination.

For combinations with at least eight evaluations, the median depth gain is compared with zero by a Wilcoxon signed-rank test. Because some durability gains are infinite, durability is tested by a sign test on the evaluations in which the combination and its best component differ, with the Wilcoxon test on the evaluations with finite gains printed beside it. The Benjamini–Hochberg procedure is applied across those combinations. Both tests are two-sided, so a combination counts as giving a deeper response only when its depth q value is below 0.05 and its median depth gain is positive, and as also regrowing more slowly only when its durability q value is below 0.10 and more of its evaluations favour the combination than its best component; combinations that regrow significantly faster are listed separately, whatever their depth result, and so are combinations whose response was significantly shallower than their best component. The analysis first run filtered on the q values alone, and its counts are printed beside the corrected ones. Medians carry 95% bootstrap intervals (a median over values that include infinite gains is the lower median). The association between the two gains is the Spearman correlation across evaluations, with infinite gains ranked beyond every finite one and an interval from resampling PDX models, and across combinations. Two combinations in the panel were later approved for patients carrying BRAF V600 mutations, and they serve as a positive control.

**The approved combinations by BRAF V600 status (10.3).** Added after review. Both approvals are for BRAF V600-mutant tumors, but most evaluations come from BRAF wild-type models. The evaluations of every encorafenib-containing combination are therefore split by the BRAF V600 status of the PDX model (Section 8.3), and the summary of 10.2 is repeated within each group, with the depth gain compared between V600 and wild-type models; single-agent encorafenib and binimetinib durability is compared between the groups as well. The groups are small, so this is descriptive.

**Inputs:** `arms`, `treated_with_endpoints`, `post_nadir_arms`, `braf_v600_models` (Section 8.3).
**Outputs:** `combination_evaluations`, `all_evaluations`, `first_run_evaluations`, `by_combination`, `combination_comparison`, `combination_v600`, `single_agent_v600`, `V600_STATUS`; figure data `braf_v600_status.csv`.

```python
# ==========================================================================
# 10.1 Combination–PDX model evaluations against the best single component
#     Corrected comparison (the one reported):
#       depth       the published best average response of every treated
#                   arm (the analysis first run used only arms with a
#                   defined post-nadir growth rate, which left out arms still
#                   shrinking at their last measurement, the deepest ones);
#       durability  combination_durability_values (Section 1.3): the
#                   post-nadir growth rate where it is defined (including
#                   arms that never shrank), −∞ for an arm whose smallest
#                   volume is its last measurement (no regrowth during
#                   follow-up, right-censored), no value for an arm with a
#                   single measurement after its nadir;
#       an evaluation has a durability gain only when the combination AND
#       every component have a durability value, so the best component is
#       truly the best of all components.
#     The rule of the analysis first run (durability only from post-nadir
#     arms, best component taken among those that had a value) is computed
#     beside it for comparison.
# ==========================================================================

set_section("10")
DEPTH_GAIN_THRESHOLD = 5.0          # percentage points of best average response
DURABILITY_GAIN_THRESHOLD = 0.005   # per day
MIN_EVALUATIONS_TO_TEST = 8
APPROVED_COMBINATIONS = {
    "encorafenib + binimetinib": "FDA 2018, BRAF V600E/K melanoma",
    "cetuximab + encorafenib": "FDA 2020, BRAF V600E colorectal cancer",
}
combination_rng = np.random.default_rng(SEEDS["combination_corrected_bootstrap"])

# --- durability status of every treated arm ----------------------------------
treated_all = arms.loc[~arms.is_control].copy()
treated_all["durability_value"] = combination_durability_values(treated_all)
treated_all["durability_status"] = np.select(
    [np.isneginf(treated_all.durability_value), treated_all.durability_value.notna()],
    ["no regrowth during follow-up (-inf)", "growth rate measured"], "regrowth not measurable")

# --- corrected and first-run evaluations ---------------------------------------
corrected_depth = (treated_all.dropna(subset=["best_average_response"])
                   .groupby(["model_id", "treatment"]).best_average_response.mean())
corrected_durability = treated_all.groupby(["model_id", "treatment"]).durability_value.mean()
all_evaluations = combination_evaluations_table(corrected_depth, corrected_durability,
                                                require_all_components=True)
combination_evaluations = all_evaluations.dropna(subset=["durability_gain"]).reset_index(drop=True)

first_run_all = combination_evaluations_table(
    treated_with_endpoints.groupby(["model_id", "treatment"]).best_average_response.mean(),
    post_nadir_arms.groupby(["model_id", "treatment"]).post_nadir_growth_rate.mean(),
    require_all_components=False)
first_run_evaluations = first_run_all.dropna(subset=["durability_gain"]).reset_index(drop=True)

deeper = combination_evaluations.depth_gain > DEPTH_GAIN_THRESHOLD
longer = combination_evaluations.durability_gain > DURABILITY_GAIN_THRESHOLD
combination_evaluations["category"] = np.select(
    [deeper & longer, deeper & ~longer, ~deeper & longer],
    ["superior on both", "superior in depth only", "superior in durability only"], "superior on neither")

print("10.1 COMBINATION EVALUATIONS AGAINST THE BEST SINGLE COMPONENT")
print("  durability status of treated arms")
print(treated_all.durability_status.value_counts().to_string())
no_combination_value = all_evaluations.combination_durability.isna()
print(f"\n  combination labels                                {all_evaluations.combination.nunique()}")
print(f"  evaluations with a depth gain                     {len(all_evaluations):,}")
print(f"    without a durability gain: combination has no value {int(no_combination_value.sum()):,}; "
      f"a component has none {int((~no_combination_value & all_evaluations.component_without_durability).sum()):,}")
print(f"  evaluations with both gains                       "
      f"{record('combination_evaluations', len(combination_evaluations)):,} across "
      f"{record('combinations_evaluated', combination_evaluations.combination.nunique())} combinations")
print(f"    combination with no regrowth during follow-up    "
      f"{int(combination_evaluations.combination_without_regrowth.sum()):,}")
print(f"  first-run rule: {len(first_run_evaluations):,} evaluations with both gains, of which "
      f"{int(first_run_evaluations.component_without_durability.sum()):,} chose the best among "
      f"incomplete components")
record("combination_evaluations_first_run", len(first_run_evaluations))
record("combination_evaluations_depth", len(all_evaluations))
category_shares = combination_evaluations.category.value_counts(normalize=True)
for category in ("superior on both", "superior in depth only", "superior in durability only",
                 "superior on neither"):
    share = float(category_shares.get(category, 0.0))
    record(f"share_{category.replace(' ', '_')}", share)
    print(f"  {category:<30} {int(round(share * len(combination_evaluations))):>4}  ({100 * share:.1f}%)")

# Depth gain against durability gain. Infinite durability gains rank beyond
# every finite one (finite_rank_values). The P value treats evaluations as
# independent; the interval and P_cluster resample PDX models, which share
# their single-agent arms across evaluations.
depth_ranked = finite_rank_values(combination_evaluations.depth_gain)
durability_ranked = finite_rank_values(combination_evaluations.durability_gain)
gain_rho, gain_p = spearman(depth_ranked, durability_ranked)
gain_draws = cluster_bootstrap(rows_by_cluster(combination_evaluations.model_id.astype(str)),
                               lambda rows: spearman(depth_ranked[rows], durability_ranked[rows])[0],
                               N_BOOTSTRAP, combination_rng)
gain_low, gain_high = np.percentile(gain_draws, [2.5, 97.5])
record("gain_rho", gain_rho)
record("gain_p", gain_p)
record("gain_rho_low", float(gain_low))
record("gain_rho_high", float(gain_high))
record("gain_p_cluster", bootstrap_p_two_sided(gain_draws))
record("gain_shared_variance", gain_rho ** 2)
first_run_rho, _ = spearman(first_run_evaluations.depth_gain, first_run_evaluations.durability_gain)
record("gain_rho_first_run", first_run_rho)
print(f"\n  depth gain against durability gain, evaluations   ρ = {gain_rho:.3f} "
      f"(95% CI {gain_low:.3f} to {gain_high:.3f}, PDX models resampled), "
      f"{p_label(RESULTS['gain_p_cluster'], 1 / len(gain_draws))}; arms as independent {p_label(gain_p)}; "
      f"first-run rule ρ = {first_run_rho:.3f}")

# ==========================================================================
# 10.2 Per-combination tests and the approved-combination positive control
#     Depth: Wilcoxon signed-rank test on every evaluation with a depth gain.
#     Durability: some gains are infinite, so the test is a sign test on the
#     evaluations where the combination and its best component differ; the
#     Wilcoxon test on the evaluations with finite gains is printed beside
#     it. Benjamini–Hochberg across combinations with >= 8 evaluations.
# ==========================================================================

combination_rows = []
for combination in sorted(all_evaluations.combination.unique()):
    depth_values = all_evaluations.loc[all_evaluations.combination == combination, "depth_gain"].to_numpy(float)
    group = combination_evaluations.loc[combination_evaluations.combination == combination]
    durability_values = group.durability_gain.to_numpy(float)
    finite_values = durability_values[np.isfinite(durability_values)]
    depth_low, depth_high = bootstrap_median_interval(depth_values, N_BOOTSTRAP, combination_rng)
    durability_low, durability_high = extended_median_interval(durability_values, N_BOOTSTRAP, combination_rng)
    combination_rows.append({
        "combination": combination, "evaluations_depth": len(depth_values), "evaluations": len(durability_values),
        "median_depth_gain": float(np.median(depth_values)), "depth_low": depth_low, "depth_high": depth_high,
        "depth_p": (stats.wilcoxon(depth_values).pvalue
                    if len(depth_values) >= MIN_EVALUATIONS_TO_TEST and np.ptp(depth_values) > 0 else np.nan),
        "median_durability_gain": float(extended_median(durability_values)) if len(durability_values) else np.nan,
        "durability_low": durability_low, "durability_high": durability_high,
        "better_durability": int((durability_values > 0).sum()), "worse_durability": int((durability_values < 0).sum()),
        "tied_durability": int((durability_values == 0).sum()),
        "combination_without_regrowth": int(group.combination_without_regrowth.sum()),
        "durability_p": (sign_test_p(durability_values)
                         if len(durability_values) >= MIN_EVALUATIONS_TO_TEST else np.nan),
        "durability_p_finite_wilcoxon": (stats.wilcoxon(finite_values).pvalue
                                         if len(finite_values) >= MIN_EVALUATIONS_TO_TEST
                                         and np.ptp(finite_values) > 0 else np.nan),
        "share_superior_on_both": float((group.category == "superior on both").mean()) if len(group) else np.nan,
        "approval": APPROVED_COMBINATIONS.get(combination, "")})
by_combination = pd.DataFrame(combination_rows).sort_values("evaluations", ascending=False)
# Benjamini–Hochberg across the tested combinations only (at least
# MIN_EVALUATIONS_TO_TEST evaluations with both gains). A combination with
# fewer evaluations gets no q value even when its depth gains alone would
# allow a test, so both endpoints share one family. (Every combination is
# tested in this run, so no q value changes.)
is_tested = by_combination.evaluations >= MIN_EVALUATIONS_TO_TEST
for p_column, q_column in (("depth_p", "depth_q"), ("durability_p", "durability_q"),
                           ("durability_p_finite_wilcoxon", "durability_q_finite_wilcoxon")):
    by_combination[q_column] = np.nan
    by_combination.loc[is_tested, q_column] = benjamini_hochberg(by_combination.loc[is_tested, p_column])
tested = by_combination.loc[is_tested]
# Both tests are two-sided, so a small q says only that the gains differ from
# zero. "Deeper response" and "slower regrowth" also need the direction: a
# positive median depth gain, and more evaluations with slower than with
# faster regrowth (the direction of the sign test). The analysis first run
# filtered on q alone, so a combination whose response was significantly
# shallower, or whose regrowth was significantly faster, would have been
# counted as deeper or more durable. The direction-blind counts are recorded
# beside the corrected ones.
depth_significant = tested.loc[(tested.depth_q < 0.05) & (tested.median_depth_gain > 0)]
also_durable = depth_significant.loc[(depth_significant.durability_q < 0.10)
                                     & (depth_significant.better_durability > depth_significant.worse_durability)]
faster_regrowth = depth_significant.loc[(depth_significant.durability_q < 0.10)
                                        & (depth_significant.better_durability < depth_significant.worse_durability)]
# Both directions, whatever the other endpoint: combinations whose response
# was significantly SHALLOWER than their best component (q < 0.05, negative
# median depth gain), and every tested combination that regrew significantly
# more slowly or FASTER (q < 0.10, by the direction of the sign test).
# (faster_regrowth above is restricted to the deeper combinations, as in the
# analysis first run, and is kept for comparison.)
shallower = tested.loc[(tested.depth_q < 0.05) & (tested.median_depth_gain < 0)]
slower_regrowth_any = tested.loc[(tested.durability_q < 0.10) & (tested.better_durability > tested.worse_durability)]
faster_regrowth_any = tested.loc[(tested.durability_q < 0.10) & (tested.better_durability < tested.worse_durability)]
record("combinations_shallower", len(shallower))
if len(shallower):
    record("shallower_q_min", float(shallower.depth_q.min()))
    record("shallower_q_max", float(shallower.depth_q.max()))
record("combinations_slower_regrowth_any_depth", len(slower_regrowth_any))
record("combinations_faster_regrowth_any_depth", len(faster_regrowth_any))
for row in faster_regrowth_any.itertuples(index=False):
    tag = "".join(character if character.isalnum() else "_" for character in row.combination)
    record(f"faster_regrowth_{tag}_median_durability_gain",
           float(row.median_durability_gain) if np.isfinite(row.median_durability_gain) else np.nan)
    record(f"faster_regrowth_{tag}_durability_q", float(row.durability_q))
    record(f"faster_regrowth_{tag}_median_depth_gain", float(row.median_depth_gain))
    record(f"faster_regrowth_{tag}_depth_q", float(row.depth_q))
depth_significant_two_sided = tested.loc[tested.depth_q < 0.05]
also_durable_two_sided = depth_significant_two_sided.loc[depth_significant_two_sided.durability_q < 0.10]
combination_rho, combination_rho_p = spearman(finite_rank_values(tested.median_depth_gain),
                                              finite_rank_values(tested.median_durability_gain))
record("combinations_tested", len(tested))
record("combinations_depth_significant", len(depth_significant))
record("combinations_also_durable", len(also_durable))
record("combinations_faster_regrowth", len(faster_regrowth))
record("combinations_depth_significant_two_sided", len(depth_significant_two_sided))
record("combinations_also_durable_two_sided", len(also_durable_two_sided))
finite_gains = also_durable.median_durability_gain[np.isfinite(also_durable.median_durability_gain)]
if len(finite_gains):
    record("also_durable_gain_min", finite_gains.min())
    record("also_durable_gain_max", finite_gains.max())
if len(also_durable):
    record("also_durable_q_min", also_durable.durability_q.min())
    record("also_durable_q_max", also_durable.durability_q.max())
# deeper combinations without significantly slower regrowth (including any
# with significantly faster regrowth)
not_durable = depth_significant.loc[~depth_significant.combination.isin(also_durable.combination)]
if len(not_durable):
    record("not_durable_q_min", not_durable.durability_q.min())

# The first-run per-combination summary, for comparison (Wilcoxon on both gains).
first_run_rows = []
for combination, group in first_run_evaluations.groupby("combination"):
    depth_values, durability_values = group.depth_gain.to_numpy(float), group.durability_gain.to_numpy(float)
    testable = len(group) >= MIN_EVALUATIONS_TO_TEST
    first_run_rows.append({
        "combination": combination, "evaluations_first_run": len(group),
        "median_depth_gain_first_run": float(np.median(depth_values)),
        "median_durability_gain_first_run": float(np.median(durability_values)),
        "depth_p_first_run": stats.wilcoxon(depth_values).pvalue if testable and np.ptp(depth_values) > 0 else np.nan,
        "durability_p_first_run": (stats.wilcoxon(durability_values).pvalue
                                   if testable and np.ptp(durability_values) > 0 else np.nan)})
by_combination_first_run = pd.DataFrame(first_run_rows)
by_combination_first_run["depth_q_first_run"] = benjamini_hochberg(by_combination_first_run.depth_p_first_run)
by_combination_first_run["durability_q_first_run"] = benjamini_hochberg(by_combination_first_run.durability_p_first_run)
combination_comparison = by_combination[["combination", "evaluations", "median_depth_gain", "depth_q",
                                         "median_durability_gain", "durability_q"]].merge(
    by_combination_first_run[["combination", "evaluations_first_run", "median_depth_gain_first_run",
                              "depth_q_first_run", "median_durability_gain_first_run", "durability_q_first_run"]],
    on="combination", how="outer")

print(f"\n10.2 COMBINATIONS WITH >= {MIN_EVALUATIONS_TO_TEST} EVALUATIONS "
      f"({len(tested)}; depth Wilcoxon, durability sign test, Benjamini–Hochberg across these)")
print(tested.drop(columns=["depth_p", "durability_p", "durability_p_finite_wilcoxon"]).round(4).to_string(index=False))
print(f"\n  deeper response, q < 0.05                         {len(depth_significant)}: "
      f"{', '.join(depth_significant.combination)}")
print(f"  of these, slower regrowth with q < 0.10           {len(also_durable)}: "
      f"{', '.join(also_durable.combination)}")
print(f"  of these, FASTER regrowth with q < 0.10           {len(faster_regrowth)}: "
      f"{', '.join(faster_regrowth.combination)}")
print(f"  shallower response, q < 0.05 (negative median)    {len(shallower)}: "
      + ", ".join(f"{row.combination} (q = {format_p(row.depth_q)})" for row in shallower.itertuples(index=False)))
print(f"  slower regrowth, q < 0.10, any depth result       {len(slower_regrowth_any)}: "
      f"{', '.join(slower_regrowth_any.combination)}")
print(f"  FASTER regrowth, q < 0.10, any depth result       {len(faster_regrowth_any)}: "
      + ", ".join(f"{row.combination} (median {row.median_durability_gain:+.4f} per day, q = "
                  f"{format_p(row.durability_q)})" for row in faster_regrowth_any.itertuples(index=False)))
print(f"  direction-blind, as first run: q < 0.05 for depth {len(depth_significant_two_sided)}, "
      f"and q < 0.10 for durability {len(also_durable_two_sided)}")
print(f"  depth gain against durability gain, combinations  ρ = {combination_rho:.3f}, "
      f"{p_label(combination_rho_p)} (n = {len(tested)})")
print("\n  corrected against first-run rule, by combination")
print(combination_comparison.sort_values("evaluations", ascending=False).round(4).to_string(index=False))
record("combination_level_gain_rho", combination_rho)
record("combination_level_gain_p", combination_rho_p)
print("\n  POSITIVE CONTROL: combinations later approved for BRAF V600-mutant tumors")
for row in by_combination.loc[by_combination.approval != ""].itertuples(index=False):
    print(f"    {row.combination} ({row.approval}), {row.evaluations} evaluations with both gains, "
          f"{row.evaluations_depth} with depth")
    print(f"      depth gain       {row.median_depth_gain:.1f} points "
          f"(95% CI {row.depth_low:.1f} to {row.depth_high:.1f}), q = {format_p(row.depth_q)}")
    print(f"      durability gain  {row.median_durability_gain:.4f} per day "
          f"(95% CI {row.durability_low:.4f} to {row.durability_high:.4f}); better in {row.better_durability}, "
          f"worse in {row.worse_durability}, tied in {row.tied_durability}; sign-test q = {format_p(row.durability_q)}")
    print(f"      superior on both {100 * row.share_superior_on_both:.1f}% of evaluations")
    tag = "".join(character if character.isalnum() else "_" for character in row.combination)
    record(f"approved_{tag}_evaluations", int(row.evaluations))
    record(f"approved_{tag}_depth_gain", float(row.median_depth_gain))
    record(f"approved_{tag}_depth_q", float(row.depth_q) if pd.notna(row.depth_q) else np.nan)
    # An infinite median is stored as missing (the register holds finite
    # numbers only) with its sign stored beside it.
    record(f"approved_{tag}_durability_gain",
           float(row.median_durability_gain) if np.isfinite(row.median_durability_gain) else np.nan)
    record(f"approved_{tag}_durability_gain_infinite",
           int(np.sign(row.median_durability_gain)) if np.isinf(row.median_durability_gain) else 0)
    record(f"approved_{tag}_durability_q", float(row.durability_q) if pd.notna(row.durability_q) else np.nan)
    record(f"approved_{tag}_better_durability", int(row.better_durability))
    record(f"approved_{tag}_worse_durability", int(row.worse_durability))
    record(f"approved_{tag}_share_superior_on_both", float(row.share_superior_on_both))
save_record(combination_evaluations, "combination_evaluations")
save_record(all_evaluations, "combination_evaluations_depth")
save_record(first_run_evaluations, "combination_evaluations_first_run")
save_record(by_combination, "combination_summary")
save_record(combination_comparison, "combination_summary_comparison")
```

```python
# ==========================================================================
# 10.3 The approved combinations by BRAF V600 status (added after review)
#     Both approved combinations are approved for BRAF V600-mutant tumors,
#     but the evaluations of 10.1 come from every PDX model in which the
#     combination and its components were tested, most of them BRAF wild
#     type, where a BRAF inhibitor has no mutant target. Here the evaluations
#     of every encorafenib-containing combination are split by the BRAF V600
#     status of the PDX model (Section 8.3: a BRAF finding whose text
#     contains "V600"; models absent from the mutation sheet are "no mutation
#     data"), and the summary of 10.2 is repeated in each group: median gains
#     with 95% bootstrap intervals, a Wilcoxon signed-rank test for depth and
#     a sign test for durability (two-sided; computed with at least
#     V600_MIN_EVALUATIONS evaluations), with Benjamini–Hochberg q values
#     across the combinations within each group. The depth gain of each
#     combination is also compared between V600 and wild-type models
#     (Mann–Whitney, two-sided). Single-agent durability of encorafenib and
#     binimetinib is compared between the groups as well (their depth is in
#     8.3). The groups are small, so all of this is descriptive.
# ==========================================================================

V600_MIN_EVALUATIONS = 5
BRAF_GROUPS = ("BRAF V600", "BRAF wild type", "no mutation data")
v600_rng = np.random.default_rng(SEEDS["v600_combination_bootstrap"])
# BRAF status of every PDX model in the response data.
V600_STATUS = {model: ("BRAF V600" if model in braf_v600_models
                       else "BRAF wild type" if model in model_sets["mutation"] else "no mutation data")
               for model in set(arms.model_id.astype(str))}
finite_or_nan = lambda value: float(value) if np.isfinite(value) else np.nan   # the register holds finite numbers
encorafenib_combinations = sorted(combination for combination in all_evaluations.combination.unique()
                                  if "encorafenib" in [part.strip() for part in combination.split("+")])
v600_rows = []
for combination in encorafenib_combinations:
    depth_block = all_evaluations.loc[all_evaluations.combination == combination]
    durability_block = combination_evaluations.loc[combination_evaluations.combination == combination]
    depth_status = depth_block.model_id.astype(str).map(V600_STATUS)
    durability_status = durability_block.model_id.astype(str).map(V600_STATUS)
    for status in BRAF_GROUPS:
        depth_gains = depth_block.loc[depth_status == status, "depth_gain"]
        group = durability_block.loc[durability_status == status]
        if not len(depth_gains):
            continue
        v600_rows.append({"combination": combination, "braf_status": status,
                          "approval": APPROVED_COMBINATIONS.get(combination, ""),
                          **gain_summary(depth_gains, group.durability_gain, N_BOOTSTRAP, v600_rng,
                                         V600_MIN_EVALUATIONS),
                          "share_superior_on_both": (float((group.category == "superior on both").mean())
                                                     if len(group) else np.nan)})
    # Depth gain, V600 against wild-type models (two-sided Mann–Whitney).
    mutant_gains = depth_block.loc[depth_status == "BRAF V600", "depth_gain"]
    wild_gains = depth_block.loc[depth_status == "BRAF wild type", "depth_gain"]
    if len(mutant_gains) >= 3 and len(wild_gains) >= 3:
        between_p = float(stats.mannwhitneyu(mutant_gains, wild_gains).pvalue)
        for row in v600_rows:
            if row["combination"] == combination:
                row["depth_gain_p_v600_vs_wild_type"] = between_p
combination_v600 = pd.DataFrame(v600_rows)
if len(combination_v600):
    for status in BRAF_GROUPS:
        in_group = combination_v600.braf_status == status
        combination_v600.loc[in_group, "depth_q"] = benjamini_hochberg(combination_v600.loc[in_group, "depth_p"])
        combination_v600.loc[in_group, "durability_q"] = benjamini_hochberg(
            combination_v600.loc[in_group, "durability_p"])
    for row in combination_v600.loc[combination_v600.approval != ""].itertuples(index=False):
        tag = "".join(character if character.isalnum() else "_" for character in row.combination)
        group_tag = {"BRAF V600": "v600", "BRAF wild type": "wild_type", "no mutation data": "no_mutation_data"}[
            row.braf_status]
        stem = f"{group_tag}_{tag}"
        record(f"{stem}_evaluations", int(row.evaluations))
        record(f"{stem}_evaluations_depth", int(row.evaluations_depth))
        record(f"{stem}_median_depth_gain", finite_or_nan(row.median_depth_gain))
        record(f"{stem}_median_depth_gain_low", finite_or_nan(row.depth_low))
        record(f"{stem}_median_depth_gain_high", finite_or_nan(row.depth_high))
        record(f"{stem}_depth_p", finite_or_nan(row.depth_p))
        # An infinite median durability gain is stored as missing, with its sign beside it.
        record(f"{stem}_median_durability_gain", finite_or_nan(row.median_durability_gain))
        record(f"{stem}_median_durability_gain_infinite",
               int(np.sign(row.median_durability_gain)) if np.isinf(row.median_durability_gain) else 0)
        record(f"{stem}_median_durability_gain_low", finite_or_nan(row.durability_low))
        record(f"{stem}_median_durability_gain_high", finite_or_nan(row.durability_high))
        record(f"{stem}_durability_p", finite_or_nan(row.durability_p))
        record(f"{stem}_better_durability", int(row.better_durability))
        record(f"{stem}_worse_durability", int(row.worse_durability))

# --- single agents: durability by BRAF V600 status ------------------------------------
single_v600_rows = []
for agent in ("encorafenib", "binimetinib"):
    agent_arms = treated_all.loc[treated_all.treatment == agent].copy()
    agent_arms["braf_status"] = agent_arms.model_id.astype(str).map(V600_STATUS)
    for status in BRAF_GROUPS:
        group = agent_arms.loc[agent_arms.braf_status == status]
        if not len(group):
            continue
        post_nadir_group = group.loc[group.arm_id.isin(post_nadir_ids)]
        single_v600_rows.append({
            "agent": agent, "braf_status": status, "arms": len(group),
            "median_best_average_response": float(group.best_average_response.median()),
            "post_nadir_arms": len(post_nadir_group),
            "median_post_nadir_growth_rate": (float(post_nadir_group.post_nadir_growth_rate.median())
                                              if len(post_nadir_group) else np.nan),
            "no_regrowth_during_follow_up": int(np.isneginf(group.durability_value).sum())})
    mutant = agent_arms.loc[(agent_arms.braf_status == "BRAF V600") & agent_arms.arm_id.isin(post_nadir_ids),
                            "post_nadir_growth_rate"]
    wild = agent_arms.loc[(agent_arms.braf_status == "BRAF wild type") & agent_arms.arm_id.isin(post_nadir_ids),
                          "post_nadir_growth_rate"]
    if len(mutant) >= 3 and len(wild) >= 3:
        p_value = float(stats.mannwhitneyu(mutant, wild).pvalue)
        record(f"single_agent_{agent}_durability_v600_vs_wild_type_p", p_value)
        for row in single_v600_rows:
            if row["agent"] == agent:
                row["durability_p_v600_vs_wild_type"] = p_value
single_agent_v600 = pd.DataFrame(single_v600_rows)
for row in single_agent_v600.itertuples(index=False):
    group_tag = {"BRAF V600": "v600", "BRAF wild type": "wild_type", "no mutation data": "no_mutation_data"}[
        row.braf_status]
    record(f"single_agent_{row.agent}_{group_tag}_median_post_nadir_growth_rate",
           finite_or_nan(row.median_post_nadir_growth_rate))
    record(f"single_agent_{row.agent}_{group_tag}_post_nadir_arms", int(row.post_nadir_arms))

print("10.3 ENCORAFENIB-CONTAINING COMBINATIONS BY BRAF V600 STATUS (descriptive, added after review)")
print(f"  PDX models: BRAF V600 {sum(status == 'BRAF V600' for status in V600_STATUS.values())}, wild type "
      f"{sum(status == 'BRAF wild type' for status in V600_STATUS.values())}, no mutation data "
      f"{sum(status == 'no mutation data' for status in V600_STATUS.values())}; tests need >= "
      f"{V600_MIN_EVALUATIONS} evaluations; q values within each BRAF group")
if len(combination_v600):
    print(combination_v600.drop(columns=["approval"]).round(4).to_string(index=False))
print("\n  single agents by BRAF V600 status (post-nadir growth rate: arms that regressed and then had "
      ">= 2 readings)")
print(single_agent_v600.round(4).to_string(index=False))
save_record(combination_v600, "combination_v600")
save_record(single_agent_v600, "single_agent_v600")
# Figure data: the BRAF V600 status of every PDX model in the response data
# (Figure 4d, e and Supplementary Figure S9 of the figures notebook).
save_record(pd.DataFrame(sorted(V600_STATUS.items()), columns=["model_id", "braf_status"]), "braf_v600_status")
```

## 11 · Analysis of mechanism class

The single-agent treatments are grouped by compound, with dose variants pooled, and each compound is assigned to one of nine mechanism families from the treatment-target annotation of the source study (listed in `compound_families.csv`). The cell-cycle family holds a CDK4/6 inhibitor (LEE011) and two MDM2 inhibitors (HDM201, CGM097). The family assignments and the contrast were fixed in a plan before testing, and that plan is loaded below and checked against its recorded checksum. The contrast compares cytostatic families (cell cycle, PI3K–mTOR, MAPK) with cytotoxic and apoptosis-inducing families. The inventory of each agent's median depth and durability had been printed before the plan was written.

For every pair of compounds from different families tested in the same PDX model, the difference in each endpoint is taken as cytostatic minus comparator. Differences are averaged within each PDX model and then across models. The plan specified a bootstrap that resamples PDX models (10,000 replicates) and a permutation of family labels across compounds (10,000 permutations). Its hypothesis predicted slower regrowth and *shallower* depth for cytostatic families. Four sensitivity analyses were added after the plan was fixed:

- a bootstrap (20,000 replicates) and a sign-flip randomization test with the compound as the unit (Monte Carlo, 20,000 draws, and, added after review, by exact enumeration of all 2^10 sign patterns);
- an exact randomization test over all 1,260 assignments of three families against two, also run for depth;
- a linear mixed-effects model with crossed random intercepts for PDX model and family pair. Two other random-effect structures are fitted beside it and printed: family pair nested within PDX model (with a PDX-model intercept), and family pair within PDX model without a PDX-model intercept. The last is what a statsmodels formula with a variance component and no `re_formula` fits, because statsmodels then drops the default intercept for the grouping factor, so it is shown to make that difference visible.

The compounds share comparators and PDX models, so they are not fully independent units. Section 11.4 therefore repeats the compound-level estimate with a bootstrap that resamples PDX models, cytostatic compounds and comparator compounds independently (Owen, 2007), refits the mixed model with a PDX-model random intercept only (the crossed fit ended on the boundary of its parameter space), estimates the intercept by least squares with standard errors clustered on PDX model and on PDX model and cytostatic compound, and lists every test in the order of the plan. The planned class-level test is the permutation of family labels. Durability exists only in post-nadir arms, and families differ in how often their arms reach that set (shown below), so each comparison is between the arms of each family that regressed and regrew.

**The encorafenib exception by BRAF V600 status (11.5).** Added after review. Encorafenib was the one cytostatic compound with a negative effect. Adaptive MAPK reactivation after BRAF inhibition is BRAF-mutant biology, while in BRAF wild-type tumors a RAF inhibitor can activate MAPK signaling paradoxically, so the compound-level cells of 11.3 are split by the BRAF V600 status of the PDX model, for encorafenib and for every other cytostatic compound. Descriptive; no test of the class contrast changes.

**Inputs:** `arms`, `post_nadir_ids`, `prespecification/mechanism_family_plan.json`, `V600_STATUS` (Section 10.3).
**Outputs:** `compound_atlas`, `compound_table` (table `compound_families.csv`), `compound_pairs`, `mechanism_results`; figure data `mechanism_class_compound_effects.csv`, `mechanism_class_leave_one_out.csv` and `mechanism_class_contrast_pairs.csv`; tables `mechanism_class_dependence.csv` and `mechanism_class_evidence.csv`; `encorafenib_v600`.

```python
# ==========================================================================
# 11.1 The checksummed plan, the compound atlas and within-model compound pairs
# ==========================================================================

set_section("11")
mechanism_plan, mechanism_plan_sha256 = load_prespecified_plan("mechanism_class")
FAMILY_OF_LABEL = mechanism_plan["family_map"]
COMPOUND_OF_LABEL = mechanism_plan["compound_map"]
CYTOSTATIC = tuple(sorted(mechanism_plan["cytostatic_families"]))
COMPARATOR = tuple(sorted(mechanism_plan["cytotoxic_families"]))
print("11.1 MECHANISM-CLASS PLAN")
print(f"  file      {relative(PRESPEC_DIR / PRESPECIFIED_PLANS['mechanism_class'][0])}")
print(f"  SHA-256   {mechanism_plan_sha256}  (matches the recorded {PRESPECIFIED_PLANS['mechanism_class'][1]})")
print(f"  contrast  {' + '.join(CYTOSTATIC)}  against  {' + '.join(COMPARATOR)}")
print(f"  hypothesis: {mechanism_plan['primary_hypothesis']}")

single_agents = arms.loc[~arms.is_control & ~arms.treatment.str.contains("+", regex=False)].copy()
single_agents["compound"] = single_agents.treatment.map(lambda label: COMPOUND_OF_LABEL.get(label, label))
single_agents["family"] = single_agents.treatment.map(FAMILY_OF_LABEL)
unmapped = sorted(single_agents.loc[single_agents.family.isna(), "treatment"].unique())
compound_atlas = single_agents.loc[single_agents.family.notna()].copy()
assert (compound_atlas.groupby("compound").family.nunique() == 1).all(), "a compound is in two families"
compound_atlas["post_nadir"] = compound_atlas.arm_id.isin(post_nadir_ids)
compound_atlas["depth_score"] = -compound_atlas.best_average_response            # higher = deeper
compound_atlas["durability_score"] = np.where(compound_atlas.post_nadir,
                                              -compound_atlas.post_nadir_growth_rate, np.nan)  # higher = slower

compound_table = compound_atlas.groupby("compound").agg(
    family=("family", "first"), published_target=("treatment_target", "first"),
    labels=("treatment", lambda t: "; ".join(sorted(t.unique()))),
    arms=("arm_id", "size"), pdx_models=("model_id", "nunique"),
    durability_values=("post_nadir", "sum")).reset_index().sort_values(["family", "compound"])
family_table = compound_atlas.groupby("family").agg(
    compounds=("compound", "nunique"), arms=("arm_id", "size"), pdx_models=("model_id", "nunique"),
    post_nadir_arms=("post_nadir", "sum"), post_nadir_share=("post_nadir", "mean")
).sort_values("post_nadir_share", ascending=False)
record("n_single_agent_compounds", compound_atlas.compound.nunique())
record("n_families", compound_atlas.family.nunique())
print(f"\n  single-agent labels without a family (left out): {unmapped if unmapped else 'none'}")
print(f"  compounds {compound_atlas.compound.nunique()} in {compound_atlas.family.nunique()} families; "
      f"arms {len(compound_atlas):,}; post-nadir arms {int(compound_atlas.post_nadir.sum()):,}")
print("\n  families, with the share of arms in the post-nadir set (the arms that carry durability)")
print(family_table.round(3).to_string())

# One value per compound per PDX model (dose variants averaged), then every
# pair of compounds within the model, oriented compound_a − compound_b.
per_compound = (compound_atlas.groupby(["model_id", "compound", "family"], as_index=False)
                .agg(depth_score=("depth_score", "mean"), durability_score=("durability_score", "mean")))
compound_pair_rows = []
for model_id, block in per_compound.groupby("model_id", sort=True):
    records_in_model = block.sort_values("compound").to_dict("records")
    for first, second in combinations(records_in_model, 2):
        compound_pair_rows.append({"model_id": model_id, "compound_a": first["compound"], "compound_b": second["compound"],
                          "family_a": first["family"], "family_b": second["family"],
                          "depth_difference": first["depth_score"] - second["depth_score"],
                          "durability_difference": first["durability_score"] - second["durability_score"]})
compound_pairs = pd.DataFrame(compound_pair_rows)
cross_family = compound_pairs.loc[compound_pairs.family_a != compound_pairs.family_b].copy()
a_first = cross_family.family_a.isin(CYTOSTATIC) & cross_family.family_b.isin(COMPARATOR)
b_first = cross_family.family_b.isin(CYTOSTATIC) & cross_family.family_a.isin(COMPARATOR)
contrast_pairs = cross_family.loc[a_first | b_first].copy()
sign = np.where(a_first.loc[contrast_pairs.index], 1.0, -1.0)
contrast_pairs["depth"] = sign * contrast_pairs.depth_difference
contrast_pairs["durability"] = sign * contrast_pairs.durability_difference
contrast_pairs["cytostatic_compound"] = np.where(sign > 0, contrast_pairs.compound_a, contrast_pairs.compound_b)
contrast_pairs["comparator_compound"] = np.where(sign > 0, contrast_pairs.compound_b, contrast_pairs.compound_a)
contrast_pairs["cytostatic_family"] = contrast_pairs.cytostatic_compound.map(
    compound_table.set_index("compound").family)
contrast_pairs["comparator_family"] = contrast_pairs.comparator_compound.map(
    compound_table.set_index("compound").family)
durability_pairs = contrast_pairs.dropna(subset=["durability"])
cytostatic_compounds = sorted(compound_table.loc[compound_table.family.isin(CYTOSTATIC), "compound"])
record("n_contrast_pairs_durability", len(durability_pairs))
record("n_contrast_models_durability", durability_pairs.model_id.nunique())
record("n_cytostatic_compounds", len(cytostatic_compounds))
record("n_cytostatic_compounds_with_pairs", durability_pairs.cytostatic_compound.nunique())
print(f"\n  within-model compound pairs {len(compound_pairs):,}; between families {len(cross_family):,}")
print(f"  cytostatic against comparator: {len(contrast_pairs):,} pairs in "
      f"{contrast_pairs.model_id.nunique()} PDX models for depth; {len(durability_pairs):,} pairs in "
      f"{durability_pairs.model_id.nunique()} PDX models for durability")
print(f"  cytostatic compounds {len(cytostatic_compounds)}, of which "
      f"{durability_pairs.cytostatic_compound.nunique()} have durability pairs")
save_record(compound_table, "compound_families")
```

```python
# ==========================================================================
# 11.2 Prespecified analyses: bootstrap over PDX models and permutation of
#      family labels across compounds
# ==========================================================================

N_MODEL_BOOTSTRAP, N_LABEL_PERMUTATIONS = 10_000, 10_000
model_means = contrast_pairs.groupby("model_id")[["depth", "durability"]].mean()
depth_result = bootstrap_mean(model_means.depth, N_MODEL_BOOTSTRAP, SEEDS["mechanism_class_model"] + 1)
durability_result = bootstrap_mean(model_means.durability, N_MODEL_BOOTSTRAP,
                                   SEEDS["mechanism_class_model"] + 2)
record("mechanism_durability", durability_result["estimate"])
record("mechanism_durability_low", durability_result["low"])
record("mechanism_durability_high", durability_result["high"])
record("mechanism_durability_p", durability_result["p_two_sided"])
record("mechanism_durability_fold_28", float(np.exp(28 * durability_result["estimate"])))
record("mechanism_depth", depth_result["estimate"])
record("mechanism_depth_low", depth_result["low"])
record("mechanism_depth_high", depth_result["high"])
record("mechanism_depth_p", depth_result["p_two_sided"])
print("11.2 PRESPECIFIED ANALYSES (cytostatic minus comparator; higher is better on both endpoints)")
print(f"  durability  {durability_result['estimate']:.4f} per day (95% CI {durability_result['low']:.4f} to "
      f"{durability_result['high']:.4f}), {p_label(durability_result['p_two_sided'], 1 / N_MODEL_BOOTSTRAP)}, "
      f"{durability_result['n']} PDX models; {np.exp(28 * durability_result['estimate']):.2f}-fold in volume "
      f"28 days after the nadir")
print(f"  depth       {depth_result['estimate']:.1f} points (95% CI {depth_result['low']:.1f} to "
      f"{depth_result['high']:.1f}), {p_label(depth_result['p_two_sided'], 1 / N_MODEL_BOOTSTRAP)}, "
      f"{depth_result['n']} PDX models")
print("  predicted direction: slower regrowth (positive durability) and shallower depth (negative depth)")
print(f"    durability prediction supported: {durability_result['low'] > 0}")
print(f"    depth prediction supported:      {depth_result['high'] < 0}")

# Permutation of family labels across compounds (family sizes kept). Every
# within-model compound pair with durability on both members is eligible, so
# a relabelled pair can enter the contrast whatever its original families.
compound_codes = {compound: i for i, compound in enumerate(sorted(per_compound.compound.unique()))}
family_names = sorted(compound_table.family.unique())
family_code = {family: i for i, family in enumerate(family_names)}
compound_family_codes = np.array([family_code[compound_table.set_index("compound").family[c]]
                                  for c in compound_codes])
durable_pairs = compound_pairs.dropna(subset=["durability_difference"])
pair_model_codes = pd.Categorical(durable_pairs.model_id).codes
pair_a = durable_pairs.compound_a.map(compound_codes).to_numpy()
pair_b = durable_pairs.compound_b.map(compound_codes).to_numpy()
pair_durability = durable_pairs.durability_difference.to_numpy(float)
cytostatic_codes = [family_code[f] for f in CYTOSTATIC]
comparator_codes = [family_code[f] for f in COMPARATOR]
observed_label_statistic, _, _ = family_contrast_statistic(
    pair_model_codes, compound_family_codes[pair_a], compound_family_codes[pair_b], pair_durability,
    cytostatic_codes, comparator_codes)
permutation_rng = np.random.default_rng(SEEDS["mechanism_label_permutation"])
label_null = np.empty(N_LABEL_PERMUTATIONS)
for index in range(N_LABEL_PERMUTATIONS):
    shuffled = permutation_rng.permutation(compound_family_codes)
    label_null[index] = family_contrast_statistic(pair_model_codes, shuffled[pair_a], shuffled[pair_b],
                                                  pair_durability, cytostatic_codes, comparator_codes)[0]
label_null = label_null[np.isfinite(label_null)]
label_p_one = (1 + np.sum(label_null >= observed_label_statistic)) / (1 + len(label_null))
label_p_two = (1 + np.sum(np.abs(label_null) >= abs(observed_label_statistic))) / (1 + len(label_null))
record("label_permutation_p_one_sided", float(label_p_one))
print(f"\n  family labels permuted across {len(compound_codes)} compounds ({len(label_null):,} permutations)")
print(f"    statistic {observed_label_statistic:.4f}; one-sided P = {format_p(label_p_one)}, "
      f"two-sided P = {format_p(label_p_two)}")

# ==========================================================================
# 11.3 Sensitivity analyses added after the plan was fixed
# ==========================================================================

N_COMPOUND_BOOTSTRAP = 20_000
compound_by_model = durability_pairs.groupby(["cytostatic_compound", "comparator_compound", "model_id"],
                                             as_index=False).durability.mean()
compound_effects = compound_by_model.groupby("cytostatic_compound", as_index=False).agg(
    effect=("durability", "mean"), pdx_models=("model_id", "nunique"))
compound_effects["family"] = compound_effects.cytostatic_compound.map(compound_table.set_index("compound").family)
compound_effects = compound_effects.sort_values("effect").reset_index(drop=True)
compound_values = compound_effects.effect.to_numpy(float)
compound_rng = np.random.default_rng(SEEDS["mechanism_class_compound"])
compound_draws = compound_values[compound_rng.integers(0, len(compound_values),
                                                       size=(N_COMPOUND_BOOTSTRAP, len(compound_values)))].mean(axis=1)
compound_low, compound_high = np.percentile(compound_draws, [2.5, 97.5])
sign_rng = np.random.default_rng(SEEDS["mechanism_class_compound"] + 101)
sign_null = (sign_rng.choice([-1.0, 1.0], size=(N_COMPOUND_BOOTSTRAP, len(compound_values)))
             * compound_values).mean(axis=1)
compound_effect = float(compound_values.mean())
loco_rows = []
for position, compound in enumerate(sorted(compound_effects.cytostatic_compound)):
    remaining = compound_effects.loc[compound_effects.cytostatic_compound != compound, "effect"].to_numpy(float)
    loco_rng = np.random.default_rng(SEEDS["mechanism_class_compound"] + 200 + position)
    draws = remaining[loco_rng.integers(0, len(remaining), size=(N_COMPOUND_BOOTSTRAP, len(remaining)))].mean(axis=1)
    loco_rows.append({"omitted": compound, "estimate": remaining.mean(),
                      "low": np.percentile(draws, 2.5), "high": np.percentile(draws, 97.5)})
leave_one_out = pd.DataFrame(loco_rows)
record("compound_effect", compound_effect)
record("compound_effect_low", float(compound_low))
record("compound_effect_high", float(compound_high))
record("compound_effect_p", bootstrap_p_two_sided(compound_draws))
record("n_compounds_positive", int((compound_values > 0).sum()))
record("loco_all_above_zero", int((leave_one_out.low > 0).all()))
if "encorafenib" in set(compound_effects.cytostatic_compound):
    record("encorafenib_effect", float(compound_effects.set_index("cytostatic_compound").effect["encorafenib"]))
print("11.3 SENSITIVITY ANALYSES ADDED AFTER THE PLAN WAS FIXED")
print(f"  (a) compound as the unit ({len(compound_values)} cytostatic compounds)")
print(compound_effects.round(5).to_string(index=False))
print(f"      mean effect {format_ci(compound_effect, compound_low, compound_high, 1, 4, ' per day')}; "
      f"bootstrap P two-sided {format_p(bootstrap_p_two_sided(compound_draws), 1 / N_COMPOUND_BOOTSTRAP)}")
sign_flip_p_one = float(np.mean(sign_null >= compound_effect))
sign_flip_p_two = float(np.mean(np.abs(sign_null) >= abs(compound_effect)))
record("sign_flip_p_one_sided", sign_flip_p_one)
record("sign_flip_p_two_sided", sign_flip_p_two)
print(f"      sign-flip randomization: one-sided {p_label(sign_flip_p_one, 1 / N_COMPOUND_BOOTSTRAP)}, "
      f"two-sided {p_label(sign_flip_p_two, 1 / N_COMPOUND_BOOTSTRAP)}")
# The same test by exact enumeration: with ten compounds there are only
# 2^10 = 1,024 sign patterns, so every one is evaluated and no random draw is
# made (the Monte Carlo values above are kept, as reported). The observed
# pattern (all signs +) is in the enumeration, so the P value is at least
# 1 / 2^n; a small tolerance keeps rounding from dropping it.
if len(compound_values) <= 20:
    all_sign_patterns = np.array(list(product((-1.0, 1.0), repeat=len(compound_values))))
    exact_sign_null = (all_sign_patterns * compound_values).mean(axis=1)
    tolerance = 1e-12 * max(abs(compound_effect), 1e-12)
    sign_flip_exact_one = float(np.mean(exact_sign_null >= compound_effect - tolerance))
    sign_flip_exact_two = float(np.mean(np.abs(exact_sign_null) >= abs(compound_effect) - tolerance))
    record("sign_flip_p_one_sided_exact", sign_flip_exact_one)
    record("sign_flip_p_two_sided_exact", sign_flip_exact_two)
    print(f"      sign-flip randomization, all {len(all_sign_patterns):,} sign patterns: one-sided "
          f"{p_label(sign_flip_exact_one)}, two-sided {p_label(sign_flip_exact_two)}")
print(f"      compounds with a positive effect {int((compound_values > 0).sum())} of {len(compound_values)}")
print("      leaving out one compound at a time")
print(leave_one_out.round(5).to_string(index=False))

# (b) exact randomization over every assignment of three families against two
model_codes_all = pd.Categorical(cross_family.model_id).codes
family_a_codes = cross_family.family_a.map(family_code).to_numpy()
family_b_codes = cross_family.family_b.map(family_code).to_numpy()
exact_rows = []
for group_one in combinations(family_names, len(CYTOSTATIC)):
    remaining_families = [f for f in family_names if f not in group_one]
    for group_two in combinations(remaining_families, len(COMPARATOR)):
        row = {"group_one": " + ".join(group_one), "group_two": " + ".join(group_two),
               "is_planned": set(group_one) == set(CYTOSTATIC) and set(group_two) == set(COMPARATOR),
               "cytostatic_families_kept": len(set(group_one) & set(CYTOSTATIC)),
               "small_family_comparator": bool({"developmental", "hormonal"} & set(group_two))}
        for endpoint in ("durability", "depth"):
            row[endpoint], row[f"{endpoint}_models"], _ = family_contrast_statistic(
                model_codes_all, family_a_codes, family_b_codes,
                cross_family[f"{endpoint}_difference"].to_numpy(float),
                [family_code[f] for f in group_one], [family_code[f] for f in group_two])
        exact_rows.append(row)
exact_splits = pd.DataFrame(exact_rows)
planned = exact_splits.loc[exact_splits.is_planned].iloc[0]
exact_p = {endpoint: {"one": float(np.mean(exact_splits[endpoint] >= planned[endpoint])),
                      "two": float(np.mean(np.abs(exact_splits[endpoint]) >= abs(planned[endpoint])))}
           for endpoint in ("durability", "depth")}
tail = exact_splits.loc[exact_splits.durability >= planned.durability]
record("n_exact_assignments", len(exact_splits))
record("exact_p_durability_one_sided", exact_p["durability"]["one"])
record("exact_p_depth_two_sided", exact_p["depth"]["two"])
print(f"\n  (b) exact randomization over {len(exact_splits):,} assignments of three families against two")
for endpoint in ("durability", "depth"):
    print(f"      {endpoint:<10} planned statistic {planned[endpoint]:.4f}; one-sided P = "
          f"{exact_p[endpoint]['one']:.3f}, two-sided P = {exact_p[endpoint]['two']:.3f}")
print(f"      assignments at or above the planned durability statistic: {len(tail)}; "
      f"{100 * tail.small_family_comparator.mean():.1f}% use developmental or hormonal as a comparator; "
      f"{100 * (tail.cytostatic_families_kept >= 2).mean():.1f}% keep two or three cytostatic families; "
      f"median PDX models {tail.durability_models.median():.0f} (planned {int(planned.durability_models)})")

# (c) mixed-effects model with random intercepts for PDX model and family pair
#     The intercept is the cytostatic-minus-comparator durability difference.
#     Three random-effect structures are fitted:
#       crossed  PDX model and family pair as crossed random intercepts (one
#                group holding every pair, both factors as variance
#                components); this is the model the section note describes
#                and the one recorded;
#       nested   a PDX-model random intercept (re_formula="1") and family
#                pair as a variance component within PDX model;
#       no PDX-model intercept  family pair within PDX model only. statsmodels
#                fits this when a variance component is given without
#                re_formula, because it then leaves out the default
#                intercept for the grouping factor.
mixed_frame = durability_pairs.assign(family_pair=[" | ".join(sorted(pair)) for pair in
                                                   zip(durability_pairs.cytostatic_family,
                                                       durability_pairs.comparator_family)])
MIXED_SPECIFICATIONS = {
    "crossed": dict(groups=np.ones(len(mixed_frame)), re_formula="0",
                    vc_formula={"pdx_model": "0 + C(model_id)", "family_pair": "0 + C(family_pair)"}),
    "nested": dict(groups=mixed_frame.model_id, re_formula="1",
                   vc_formula={"family_pair": "0 + C(family_pair)"}),
    "no PDX-model intercept": dict(groups=mixed_frame.model_id,
                                   vc_formula={"family_pair": "0 + C(family_pair)"}),
}
mixed_rows, mixed_fits = [], {}
for label, specification in MIXED_SPECIFICATIONS.items():
    fitted, converged, messages = fit_mixed_model("durability ~ 1", mixed_frame, **specification)
    mixed_fits[label] = fitted
    components = dict(zip(fitted.model.exog_vc.names, np.asarray(fitted.vcomp, float)))
    if fitted.cov_re.size:                       # the PDX-model intercept of the nested form
        components["pdx_model"] = float(fitted.cov_re.iloc[0, 0])
    mixed_rows.append({"structure": label, "intercept": float(fitted.params["Intercept"]),
                       "se": float(fitted.bse["Intercept"]), "p_value": float(fitted.pvalues["Intercept"]),
                       "pdx_model_variance": components.get("pdx_model", np.nan),
                       "family_pair_variance": components.get("family_pair", np.nan),
                       "residual_variance": float(fitted.scale), "converged": converged,
                       "warnings": "; ".join(messages)})
mixed_models = pd.DataFrame(mixed_rows)
mixed_fit = mixed_fits["crossed"]
record("mixed_model_intercept", float(mixed_fit.params["Intercept"]))
record("mixed_model_se", float(mixed_fit.bse["Intercept"]))
record("mixed_model_p", float(mixed_fit.pvalues["Intercept"]))
record("mixed_model_converged", int(mixed_models.set_index("structure").converged["crossed"]))
nested_row = mixed_models.set_index("structure").loc["nested"]
record("mixed_model_nested_intercept", float(nested_row.intercept))
record("mixed_model_nested_p", float(nested_row.p_value))
print(f"\n  (c) mixed-effects models: {len(mixed_frame)} pairs, {mixed_frame.model_id.nunique()} PDX models, "
      f"{mixed_frame.family_pair.nunique()} family pairs")
print(mixed_models.drop(columns="warnings").assign(
    p_value=lambda d: d.p_value.map(format_p)).to_string(index=False, float_format=lambda v: f"{v:.3g}"))
for row in mixed_models.itertuples(index=False):
    if row.warnings:
        print(f"      {row.structure}: {row.warnings}")
print(f"      recorded (crossed): intercept {mixed_fit.params['Intercept']:.4f} per day "
      f"(SE {mixed_fit.bse['Intercept']:.4f}), {p_label(mixed_fit.pvalues['Intercept'])}")

breakdown = durability_pairs.groupby(["cytostatic_family", "comparator_family"]).agg(
    pairs=("durability", "size"), pdx_models=("model_id", "nunique"),
    mean_difference=("durability", "mean"), median_difference=("durability", "median")).reset_index()
cell_cycle_cytotoxic = breakdown.loc[(breakdown.cytostatic_family == "cell cycle")
                                     & (breakdown.comparator_family == "cytotoxic")]
if len(cell_cycle_cytotoxic):
    record("cell_cycle_vs_cytotoxic_median", float(cell_cycle_cytotoxic.median_difference.iloc[0]))
print("\n  durability difference by family pair")
print(breakdown.round(5).to_string(index=False))
mechanism_results = pd.DataFrame([
    {"analysis": "bootstrap over PDX models, durability (planned)", **durability_result},
    {"analysis": "bootstrap over PDX models, depth (planned)", **depth_result},
    {"analysis": "family labels permuted across compounds (planned)", "estimate": observed_label_statistic,
     "p_one_sided": label_p_one, "p_two_sided": label_p_two, "n": len(label_null)},
    {"analysis": "compound bootstrap (added)", "estimate": compound_effect, "low": compound_low,
     "high": compound_high, "p_two_sided": bootstrap_p_two_sided(compound_draws), "n": len(compound_values)},
    {"analysis": "exact family randomization, durability (added)", "estimate": planned.durability,
     "p_one_sided": exact_p["durability"]["one"], "p_two_sided": exact_p["durability"]["two"],
     "n": len(exact_splits)},
    {"analysis": "exact family randomization, depth (added)", "estimate": planned.depth,
     "p_one_sided": exact_p["depth"]["one"], "p_two_sided": exact_p["depth"]["two"], "n": len(exact_splits)},
    *[{"analysis": f"mixed-effects model, {row.structure} (added)", "estimate": row.intercept,
       "p_two_sided": row.p_value, "n": len(mixed_frame)} for row in mixed_models.itertuples(index=False)]])
save_record(mechanism_results, "mechanism_class_results")
save_record(exact_splits, "mechanism_class_exact_splits")

# Data for the mechanism-class figure: each cytostatic compound's effect,
# the leave-one-compound-out estimates and every cytostatic-comparator pair
# (the exact assignments are saved above).
save_record(compound_effects, "mechanism_class_compound_effects")
save_record(leave_one_out, "mechanism_class_leave_one_out")
save_record(contrast_pairs, "mechanism_class_contrast_pairs")
```

```python
# ==========================================================================
# 11.4 Inference that respects shared compounds, comparators and PDX models
#     The compound bootstrap and sign-flip test of 11.3 treat the cytostatic
#     compounds as independent, but they are compared with the same few
#     comparator compounds in the same PDX models. Here the compound-level
#     effect is resampled over PDX models, cytostatic compounds and
#     comparator compounds, alone and together (the three-way version is a
#     pigeonhole bootstrap; Owen 2007). The crossed mixed model of 11.3
#     ended on the boundary (family-pair variance near zero, Hessian not
#     positive definite), so the model is refitted with the PDX-model random
#     intercept only, and the intercept is also estimated by least squares
#     with standard errors clustered on PDX model, and on PDX model and
#     cytostatic compound. The planned class-level test remains the
#     permutation of family labels (11.2); the table at the end lists every
#     test in the order of the plan.
# ==========================================================================

N_DEPENDENCE_BOOTSTRAP = N_COMPOUND_BOOTSTRAP
dependence_rng = np.random.default_rng(SEEDS["mechanism_dependence_bootstrap"])
cell_durability = compound_by_model.durability.to_numpy(float)
dependence_cytostatic_codes, cytostatic_levels = pd.factorize(compound_by_model.cytostatic_compound)
dependence_comparator_codes, comparator_levels = pd.factorize(compound_by_model.comparator_compound)
dependence_model_codes, dependence_model_levels = pd.factorize(compound_by_model.model_id.astype(str))
point_estimate = weighted_compound_effect(cell_durability, dependence_cytostatic_codes, dependence_comparator_codes,
                                          dependence_model_codes, np.ones(len(cytostatic_levels)),
                                          np.ones(len(comparator_levels)), np.ones(len(dependence_model_levels)))
assert np.isclose(point_estimate, compound_effect), "the weighted compound effect does not reproduce 11.3"
# Resampling weights: how many times each level is drawn (all ones when the
# factor is not resampled). Draw order per replicate: PDX models, cytostatic
# compounds, comparator compounds.
level_weights = lambda levels, resample: (
    np.bincount(dependence_rng.integers(0, len(levels), len(levels)), minlength=len(levels)).astype(float)
    if resample else np.ones(len(levels)))
RESAMPLING_SCHEMES = {                     # (PDX models, cytostatic, comparator)
    # the 11.3 scheme with fresh draws: its interval differs from 11.3's only
    # by Monte Carlo error; the paper reports 11.3's interval for this scheme
    "cytostatic compounds (11.3 scheme, fresh draws)": (False, True, False),
    "PDX models": (True, False, False),
    "cytostatic and comparator compounds": (False, True, True),
    "PDX models and cytostatic compounds": (True, True, False),
    "PDX models, cytostatic and comparator compounds": (True, True, True),
}
dependence_rows = []
for scheme, (resample_models, resample_cytostatic, resample_comparators) in RESAMPLING_SCHEMES.items():
    draws = []
    for _ in range(N_DEPENDENCE_BOOTSTRAP):
        model_weights = level_weights(dependence_model_levels, resample_models)
        cytostatic_weights = level_weights(cytostatic_levels, resample_cytostatic)
        comparator_weights = level_weights(comparator_levels, resample_comparators)
        value = weighted_compound_effect(cell_durability, dependence_cytostatic_codes, dependence_comparator_codes, dependence_model_codes,
                                         cytostatic_weights, comparator_weights, model_weights)
        if np.isfinite(value):
            draws.append(value)
    draws = np.asarray(draws)
    dependence_rows.append({"resampled": scheme, "replicates": len(draws), "estimate": point_estimate,
                            "low": float(np.percentile(draws, 2.5)), "high": float(np.percentile(draws, 97.5)),
                            "p_two_sided": bootstrap_p_two_sided(draws)})
dependence_table = pd.DataFrame(dependence_rows)
three_way = dependence_table.set_index("resampled").loc["PDX models, cytostatic and comparator compounds"]
record("compound_effect_threeway", point_estimate)
record("compound_effect_threeway_low", float(three_way.low))
record("compound_effect_threeway_high", float(three_way.high))
record("compound_effect_threeway_p", float(three_way.p_two_sided))

# Mixed model without the boundary component, and cluster-robust least squares.
model_only_fit, model_only_converged, model_only_messages = fit_mixed_model(
    "durability ~ 1", mixed_frame, groups=mixed_frame.model_id)
contrast_pair_durability = mixed_frame.durability.to_numpy(float)
intercept_only = sm.OLS(contrast_pair_durability, np.ones((len(mixed_frame), 1)))
ols_by_model = intercept_only.fit(cov_type="cluster",
                                  cov_kwds={"groups": pd.factorize(mixed_frame.model_id)[0]})
# Two-way clustering (PDX model and cytostatic compound). With only about ten
# cytostatic compounds this standard error is itself imprecise.
ols_two_way = intercept_only.fit(cov_type="cluster", cov_kwds={"groups": np.column_stack(
    [pd.factorize(mixed_frame.model_id)[0], pd.factorize(mixed_frame.cytostatic_compound)[0]])})
record("mixed_model_pdx_only_intercept", float(model_only_fit.params["Intercept"]))
record("mixed_model_pdx_only_p", float(model_only_fit.pvalues["Intercept"]))
record("mixed_model_pdx_only_converged", int(model_only_converged))
record("ols_cluster_model_p", float(ols_by_model.pvalues[0]))
record("ols_cluster_two_way_p", float(ols_two_way.pvalues[0]))

# Every test of the class contrast, in the order of the plan.
crossed_row = mixed_models.set_index("structure").loc["crossed"]
mechanism_evidence = pd.DataFrame([
    {"test": "family labels permuted across compounds", "status": "planned (11.2), class-level",
     "unit": "compound labels", "estimate": observed_label_statistic, "low": np.nan, "high": np.nan,
     "p_one_sided": label_p_one, "p_two_sided": label_p_two},
    {"test": "bootstrap over PDX models", "status": "planned (11.2)", "unit": "PDX models",
     "estimate": durability_result["estimate"], "low": durability_result["low"], "high": durability_result["high"],
     "p_one_sided": np.nan, "p_two_sided": durability_result["p_two_sided"]},
    {"test": "exact randomization of family assignments", "status": "added (11.3b), class-level",
     "unit": "family assignments", "estimate": planned.durability, "low": np.nan, "high": np.nan,
     "p_one_sided": exact_p["durability"]["one"], "p_two_sided": exact_p["durability"]["two"]},
    {"test": "compound bootstrap", "status": "added (11.3a)", "unit": "cytostatic compounds as independent",
     "estimate": compound_effect, "low": compound_low, "high": compound_high,
     "p_one_sided": np.nan, "p_two_sided": bootstrap_p_two_sided(compound_draws)},
    {"test": "sign-flip randomization", "status": "added (11.3a)", "unit": "cytostatic compounds as independent",
     "estimate": compound_effect, "low": np.nan, "high": np.nan,
     "p_one_sided": sign_flip_p_one, "p_two_sided": sign_flip_p_two},
    {"test": "mixed model, crossed (boundary fit)", "status": "added (11.3c)", "unit": "pairs",
     "estimate": crossed_row.intercept, "low": crossed_row.intercept - 1.96 * crossed_row.se,
     "high": crossed_row.intercept + 1.96 * crossed_row.se, "p_one_sided": np.nan, "p_two_sided": crossed_row.p_value},
    {"test": "three-way bootstrap", "status": "added (11.4)",
     "unit": "PDX models, cytostatic and comparator compounds", "estimate": point_estimate,
     "low": three_way.low, "high": three_way.high, "p_one_sided": np.nan, "p_two_sided": three_way.p_two_sided},
    {"test": "mixed model, PDX-model intercept only", "status": "added (11.4)", "unit": "pairs",
     "estimate": float(model_only_fit.params["Intercept"]),
     "low": float(model_only_fit.params["Intercept"] - 1.96 * model_only_fit.bse["Intercept"]),
     "high": float(model_only_fit.params["Intercept"] + 1.96 * model_only_fit.bse["Intercept"]),
     "p_one_sided": np.nan, "p_two_sided": float(model_only_fit.pvalues["Intercept"])},
    {"test": "least squares, clustered on PDX model", "status": "added (11.4)", "unit": "pairs",
     "estimate": float(ols_by_model.params[0]), "low": float(ols_by_model.conf_int()[0][0]),
     "high": float(ols_by_model.conf_int()[0][1]), "p_one_sided": np.nan, "p_two_sided": float(ols_by_model.pvalues[0])},
    {"test": "least squares, clustered on PDX model and cytostatic compound", "status": "added (11.4)",
     "unit": "pairs", "estimate": float(ols_two_way.params[0]), "low": float(ols_two_way.conf_int()[0][0]),
     "high": float(ols_two_way.conf_int()[0][1]), "p_one_sided": np.nan, "p_two_sided": float(ols_two_way.pvalues[0])},
])

print("11.4 INFERENCE THAT RESPECTS SHARED COMPOUNDS, COMPARATORS AND PDX MODELS")
print(f"  cells (cytostatic compound × comparator compound × PDX model) {len(compound_by_model)}; "
      f"cytostatic compounds {len(cytostatic_levels)}, comparator compounds {len(comparator_levels)}, "
      f"PDX models {len(dependence_model_levels)}")
print(dependence_table.assign(p_two_sided=lambda d: d.p_two_sided.map(
    lambda p: format_p(p, 1 / N_DEPENDENCE_BOOTSTRAP))).round(5).to_string(index=False))
print(f"\n  mixed model with a PDX-model random intercept only: converged {model_only_converged}")
for message in model_only_messages:
    print(f"    reported: {message}")
print("\n  EVERY TEST OF THE CLASS CONTRAST (durability, cytostatic minus comparator, per day)")
print(mechanism_evidence.assign(p_one_sided=lambda d: d.p_one_sided.map(lambda p: format_p(p) if pd.notna(p) else ""),
                                p_two_sided=lambda d: d.p_two_sided.map(lambda p: format_p(p) if pd.notna(p) else ""))
      .round(5).to_string(index=False))
print(f"\n  planned class-level test (family labels permuted): one-sided {p_label(label_p_one)}")
save_record(dependence_table, "mechanism_class_dependence")
save_record(mechanism_evidence, "mechanism_class_evidence")
```

```python
# ==========================================================================
# 11.5 The encorafenib exception by BRAF V600 status (added after review)
#     Encorafenib was the one cytostatic compound with a negative effect in
#     11.3. Adaptive reactivation of MAPK signaling after BRAF inhibition is
#     BRAF-mutant biology; in BRAF wild-type tumors a RAF inhibitor can
#     instead activate MAPK signaling paradoxically, so the exception means
#     different things in the two groups. The compound-level cells of 11.3
#     (cytostatic compound × comparator compound × PDX model) are split by
#     the BRAF V600 status of the PDX model (10.3), for encorafenib and, for
#     comparison, for every cytostatic compound. Each group's effect is the
#     mean of its cells (the 11.3 compound effect restricted to the group),
#     with a 95% interval resampling PDX models. Descriptive: the groups are
#     small, and no test of the class contrast is changed.
# ==========================================================================

encorafenib_rng = np.random.default_rng(SEEDS["encorafenib_v600_bootstrap"])
cells_by_status = compound_by_model.assign(braf_status=compound_by_model.model_id.astype(str).map(V600_STATUS))
v600_compound_rows = []
for compound, block in cells_by_status.groupby("cytostatic_compound", sort=True):
    for status in BRAF_GROUPS:
        group = block.loc[block.braf_status == status].reset_index(drop=True)
        if not len(group):
            continue
        values = group.durability.to_numpy(float)
        low, high = np.nan, np.nan
        if group.model_id.nunique() >= 3:
            # Cells of one PDX model move together in the resample.
            draws = cluster_bootstrap(rows_by_cluster(group.model_id.astype(str)),
                                      lambda rows: values[rows].mean(), N_BOOTSTRAP, encorafenib_rng)
            low, high = float(np.percentile(draws, 2.5)), float(np.percentile(draws, 97.5))
        v600_compound_rows.append({"cytostatic_compound": compound, "braf_status": status,
                                   "cells": len(group), "pdx_models": group.model_id.nunique(),
                                   "effect": float(values.mean()), "low": low, "high": high})
encorafenib_v600 = pd.DataFrame(v600_compound_rows)
# The groups together reproduce the 11.3 effect of encorafenib.
encorafenib_cells = compound_by_model.loc[compound_by_model.cytostatic_compound == "encorafenib"]
if len(encorafenib_cells):
    assert np.isclose(encorafenib_cells.durability.mean(), RESULTS["encorafenib_effect"])
for row in encorafenib_v600.loc[encorafenib_v600.cytostatic_compound == "encorafenib"].itertuples(index=False):
    group_tag = {"BRAF V600": "braf_v600", "BRAF wild type": "braf_wild_type",
                 "no mutation data": "no_mutation_data"}[row.braf_status]
    record(f"encorafenib_effect_{group_tag}", row.effect)
    record(f"encorafenib_effect_{group_tag}_low", row.low)
    record(f"encorafenib_effect_{group_tag}_high", row.high)
    record(f"encorafenib_effect_{group_tag}_pdx_models", int(row.pdx_models))
print("11.5 CYTOSTATIC COMPOUND EFFECTS BY BRAF V600 STATUS (durability, cytostatic minus comparator, per day; "
      "intervals resample PDX models)")
print(encorafenib_v600.round(5).to_string(index=False))
save_record(encorafenib_v600, "encorafenib_v600")
```

## 12 · Drug exposure and observation span

Section 5 compared the two regrowth models under constant drug effect. Here both models are extended so that the kill term follows simulated drug exposure. Exposure q(t) follows a one-compartment model with first-order elimination and instantaneous absorption, with a dose once daily from day 0 to the end of the window. Mouse half-lives were not available for these agents and dosing schedules are not recorded in the source data, so two half-lives are used, 0.25 and 1 day. Results are conditional on that range and do not refer to any agent.

- **Tolerant subpopulation:** sensitive cells grow as dV/dt = [g − k·q(t)]V and tolerant cells as dV/dt = gV.
- **Declining effect:** dV/dt = [g − k·q(t)·e^(−δt)]V. Clearance is carried by q(t), and δ carries any loss of effect beyond clearance.

The cumulative exposure is integrated analytically and scaled so that mean exposure over [0, T] is 1, with T the last measurement day of the schedule being modeled. The scaling is a constant factor and is absorbed into the kill rate k, but the scale of k therefore depends on the schedule. Both models have four parameters. They are fitted by bounded least squares from three starting points (the bounds of Section 5), and a fit that ends on a bound is flagged as pinned.

**Paired replicates.** Each replicate draws one schedule from the source study and one set of shared parameters: baseline volume, growth rate, kill rate and a standard-normal noise vector. Both mechanisms are generated from these, so the mechanism is the only systematic difference between the two simulated arms. The parameter distributions are listed in `paired_exposure_trajectories` (Section 1) and differ from the priors of Section 5. Both models are fitted to each arm. Three quantities are recorded:

- **Structural separation:** the RMSE of the incorrect model fitted to the noise-free arm.
- **Decisive rate:** the share of arms whose AIC difference is at least 2. A smaller difference is recorded as no decision. Section 5.5 had no such category and always chose the lower AIC.
- **Accuracy:** the share of decisive calls that name the generating model. Counts and Wilson 95% intervals are given because some cells have few decisive calls.

**Termination.** With growth rates up to about 0.13 per day, many simulated arms pass 2,000 mm³ before the end of a long window, which the source study would not have observed. Every arm is therefore scored twice: with the full schedule, and ended at its first measurement above 2,000 mm³ (the rule of Section 5). An arm left with fewer than five measurements counts as undecidable. The terminated version matches the source study and is the one recorded; the full-schedule version is printed beside it.

The section has four parts:

- **12.1** repeats the comparison over 28 days on every schedule reaching day 28. This gives the structural separation over the study's typical window.
- **12.2** measures separation at windows from 14 to 84 days on a fixed pool: the schedules that reach day 84, each cut at the window. A fixed pool keeps measurement density from changing with the window. Fold changes from 14 and 28 days to 84 days, and the exponent of a log–log fit of separation on window length, are computed within this pool. Each fold change and exponent carries a 95% interval from resampling the paired replicates of each window (2,000 resamples). The 14-day separation is small and rests on fewer arms, because fewer schedules keep five measurements by day 14, so the fold change from day 14 is the least stable of these quantities.
- **12.3** scores the AIC verdict at windows of 28, 56 and 84 days and four noise levels, including the measured noise of Section 3.6. Decisive rates carry Wilson 95% intervals beside the accuracies.
- **12.4** thins the 84-day schedules to every second or third measurement, so that span and number of measurements can be separated.
- **12.5** quantifies three caveats. The Wilson intervals treat the two arms of a paired replicate as independent, so the decisive rate and accuracy at the measured noise are recomputed with intervals that resample paired replicates. About a third of fits end with a pinned parameter, so the same summaries are given among arms without one. Every result is conditional on the simulated parameter distributions, which give strong kill (about 0.13 per day), so the 28- and 84-day verdicts are repeated with weaker kill (about 0.06 per day).

Each part has its own seed and 50 paired replicates per condition. These simulations bear on the design of future studies. Few of the source study's schedules reach 84 days (Section 3.1), and the arms that do are a selected subset, because termination depends on volume (Section 6.3).

**Inputs:** `empirical_grids`, `MEASUREMENT_NOISE`, and `approximation`, `mechanism_discrimination` and `ILLUSTRATIVE_TOLERANT` from Section 5.
**Outputs:** `exposure_paired`, `exposure_window_sweep`, `exposure_confirmation`, `exposure_thinning` and their summaries; figure data `exposure_residual_example.csv` (a worked example on one typical schedule).

```python
# ==========================================================================
# 12.1 Paired simulations over 28 days on the study's own schedules
#     One schedule is drawn from every schedule reaching day 28 and cut
#     there. Both mechanisms are generated from one parameter draw and scored
#     at four noise levels; zero noise gives the structural separation (the
#     RMSE of the incorrect model). Each arm is scored with the full schedule
#     and with the termination rule applied.
# ==========================================================================

set_section("12")
EXPOSURE_HALF_LIVES = (0.25, 1.0)          # days; two clearance regimes, not agent values
HALF_LIFE_TAG = {0.25: "hl025", 1.0: "hl100"}
DOSING_INTERVAL_DAYS = 1.0                 # once daily from day 0
TIE_THRESHOLD = 2.0                        # |ΔAIC| below this is recorded as no decision
N_EXPOSURE_REPLICATES = 50                 # paired replicates per condition
STUDY_WINDOW_DAYS = 28
PAIRED_NOISE_LEVELS = (0.0, 0.025, 0.05, MEASUREMENT_NOISE)
TERMINATION_RULES = ("applied", "not applied")

paired_rng = np.random.default_rng(SEEDS["exposure_separation"])
study_pool = [grid for grid in empirical_grids if grid.max() >= STUDY_WINDOW_DAYS]
paired_rows = []
for half_life in EXPOSURE_HALF_LIVES:
    dose_days = dosing_days(DOSING_INTERVAL_DAYS, STUDY_WINDOW_DAYS)
    for noise_sd in PAIRED_NOISE_LEVELS:
        for replicate in range(N_EXPOSURE_REPLICATES):
            days = draw_schedule(paired_rng, study_pool, STUDY_WINDOW_DAYS)
            if days is None:
                continue
            clean, noise = paired_exposure_trajectories(paired_rng, days, dose_days, half_life)
            for row in score_exposure_pair(clean, noise, noise_sd, days, dose_days, half_life,
                                           TIE_THRESHOLD):
                paired_rows.append({"half_life_days": half_life, "noise_sd": noise_sd,
                                    "replicate": replicate, **row})
exposure_paired = pd.DataFrame(paired_rows)

# Structural separation: the incorrect model fitted to noise-free arms. The
# correct model's RMSE on the same arms checks the fitter (it should be ~0).
noise_free = exposure_paired.loc[exposure_paired.noise_sd == 0.0]
paired_separation = noise_free.groupby(["termination", "half_life_days", "true_mechanism"]).agg(
    arms=("rmse_wrong", "count"), median_measurements=("n_measurements", "median"),
    median_separation=("rmse_wrong", "median"),
    p10=("rmse_wrong", lambda v: v.quantile(0.10)), p90=("rmse_wrong", lambda v: v.quantile(0.90)),
    median_rmse_correct_model=("rmse_right", "median"), terminated_share=("terminated", "mean"))
separation_28 = noise_free.groupby("termination").rmse_wrong.median()
record("exposure_separation_28", float(separation_28["applied"]))
record("exposure_separation_28_without_termination", float(separation_28["not applied"]))

paired_summary = (exposure_paired.loc[exposure_paired.noise_sd > 0]
                  .groupby(["termination", "half_life_days", "noise_sd"])
                  .apply(summarize_verdicts, include_groups=False).reset_index())
scored = exposure_paired.loc[(exposure_paired.noise_sd > 0) & (exposure_paired.termination == "applied")]

print("12.1 PAIRED SIMULATIONS OVER 28 DAYS (all schedules reaching day 28)")
print(f"  schedules in the pool {len(study_pool):,}; {N_EXPOSURE_REPLICATES} paired replicates per "
      f"half-life and noise level; once-daily dosing; tie threshold |ΔAIC| < {TIE_THRESHOLD}")
print("\n  structural separation (incorrect model fitted to noise-free arms), log units")
print(paired_separation.round(4).to_string())
print(f"\n  median structural separation, both half-lives: {separation_28['applied']:.4f} with the "
      f"termination rule, {separation_28['not applied']:.4f} without; measurement noise "
      f"{MEASUREMENT_NOISE:.4f} (ratio {separation_28['applied'] / MEASUREMENT_NOISE:.3f})")
print("\n  AIC verdicts with noise (accuracy among decisive calls, Wilson 95% interval)")
print(paired_summary.round(3).to_string(index=False))
print("\n  arms with a pinned parameter in either fit (termination rule applied), by generating model")
print(scored.groupby("true_mechanism").any_pinned.agg(["mean", "sum", "size"]).round(3).to_string())
unpinned = scored.loc[~scored.any_pinned & scored.verdict.isin(MECHANISMS)]
print(f"  accuracy among decisive calls without pinned fits: {unpinned.correct.mean():.3f} "
      f"({int(unpinned.correct.sum())} of {len(unpinned)})")
save_record(paired_separation.reset_index(), "exposure_paired_separation")
save_record(paired_summary, "exposure_paired_verdicts")

# ==========================================================================
# 12.2 Structural separation against window length on a fixed pool
#     The pool is every schedule reaching day 84, cut at each window, so the
#     schedules do not change with the window. Fold changes are taken from
#     14 and from 28 days to 84 days, and the growth of separation with the
#     window is summarized by the slope of log median separation on log
#     window.
# ==========================================================================

SWEEP_WINDOWS = (14, 21, 28, 42, 56, 84)

window_rng = np.random.default_rng(SEEDS["exposure_window"])
long_pool = [grid for grid in empirical_grids if grid.max() >= max(SWEEP_WINDOWS)]
sweep_rows = []
for half_life in EXPOSURE_HALF_LIVES:
    for window_days in SWEEP_WINDOWS:
        dose_days = dosing_days(DOSING_INTERVAL_DAYS, window_days)
        for replicate in range(N_EXPOSURE_REPLICATES):
            days = draw_schedule(window_rng, long_pool, window_days)
            if days is None:
                continue
            clean, noise = paired_exposure_trajectories(window_rng, days, dose_days, half_life)
            for row in score_exposure_pair(clean, noise, 0.0, days, dose_days, half_life,
                                           TIE_THRESHOLD):
                sweep_rows.append({"half_life_days": half_life, "window_days": window_days,
                                    "replicate": replicate, **row})
exposure_window_sweep = pd.DataFrame(sweep_rows)
separation_by_window = exposure_window_sweep.groupby(
    ["termination", "half_life_days", "window_days"]).agg(
    arms=("rmse_wrong", "count"), median_measurements=("n_measurements", "median"),
    median_separation=("rmse_wrong", "median"),
    p10=("rmse_wrong", lambda v: v.quantile(0.10)), p90=("rmse_wrong", lambda v: v.quantile(0.90)),
    median_rmse_correct_model=("rmse_right", "median"),
    terminated_share=("terminated", "mean")).reset_index()

growth_rows = []
for (rule, half_life), block in separation_by_window.groupby(["termination", "half_life_days"]):
    by_window = block.set_index("window_days").median_separation
    exponent = np.polyfit(np.log(by_window.index.to_numpy(float)), np.log(by_window.to_numpy()), 1)[0]
    growth_rows.append({"termination": rule, "half_life_days": half_life,
                        "separation_14": by_window[14], "separation_28": by_window[28],
                        "separation_84": by_window[84],
                        "fold_14_to_84": by_window[84] / by_window[14],
                        "fold_28_to_84": by_window[84] / by_window[28],
                        "log_log_exponent": exponent})
separation_growth = pd.DataFrame(growth_rows)
for row in separation_growth.itertuples():
    suffix = "" if row.termination == "applied" else "_without_termination"
    tag = HALF_LIFE_TAG[row.half_life_days]
    record(f"exposure_fold_28_to_84_{tag}{suffix}", float(row.fold_28_to_84))
    record(f"exposure_fold_14_to_84_{tag}{suffix}", float(row.fold_14_to_84))
    record(f"exposure_separation_exponent_{tag}{suffix}", float(row.log_log_exponent))
    record(f"exposure_separation_84_{tag}{suffix}", float(row.separation_84))

# Intervals for the fold changes and the exponent. Within each window the
# paired replicates are resampled with replacement (both arms of a replicate
# move together), the median separation is recomputed per window, and the
# fold changes and the log-log slope are recomputed from those medians.
N_SEPARATION_BOOTSTRAP = N_BOOTSTRAP
separation_rng = np.random.default_rng(SEEDS["separation_growth_bootstrap"])
growth_interval_rows = []
for (rule, half_life), block in exposure_window_sweep.groupby(["termination", "half_life_days"]):
    values_by_window = {}
    for window_days, window_block in block.dropna(subset=["rmse_wrong"]).groupby("window_days"):
        values_by_window[window_days] = [group.rmse_wrong.to_numpy(float)
                                         for _, group in window_block.groupby("replicate", sort=True)]
    windows = np.array(sorted(values_by_window), dtype=float)
    draws = []
    for _ in range(N_SEPARATION_BOOTSTRAP):
        medians = {}
        for window_days in windows:
            groups_in_window = values_by_window[window_days]
            picked = separation_rng.integers(0, len(groups_in_window), size=len(groups_in_window))
            medians[window_days] = float(np.median(np.concatenate([groups_in_window[i] for i in picked])))
        median_values = np.array([medians[w] for w in windows])
        if np.all(median_values > 0):
            draws.append((medians[84] / medians[14], medians[84] / medians[28],
                          np.polyfit(np.log(windows), np.log(median_values), 1)[0]))
    draws = np.array(draws)
    bounds = np.percentile(draws, [2.5, 97.5], axis=0)
    growth_interval_rows.append({"termination": rule, "half_life_days": half_life, "resamples": len(draws),
                                 "fold_14_to_84_low": bounds[0, 0], "fold_14_to_84_high": bounds[1, 0],
                                 "fold_28_to_84_low": bounds[0, 1], "fold_28_to_84_high": bounds[1, 1],
                                 "exponent_low": bounds[0, 2], "exponent_high": bounds[1, 2]})
separation_growth = separation_growth.merge(pd.DataFrame(growth_interval_rows),
                                            on=["termination", "half_life_days"], how="left")
for row in separation_growth.itertuples():
    suffix = "" if row.termination == "applied" else "_without_termination"
    tag = HALF_LIFE_TAG[row.half_life_days]
    for quantity in ("fold_28_to_84", "fold_14_to_84"):
        record(f"exposure_{quantity}_{tag}{suffix}_low", float(getattr(row, f"{quantity}_low")))
        record(f"exposure_{quantity}_{tag}{suffix}_high", float(getattr(row, f"{quantity}_high")))
    record(f"exposure_separation_exponent_{tag}{suffix}_low", float(row.exponent_low))
    record(f"exposure_separation_exponent_{tag}{suffix}_high", float(row.exponent_high))

print("\n12.2 STRUCTURAL SEPARATION AGAINST WINDOW (fixed pool of schedules reaching day 84)")
print(f"  schedules in the pool {len(long_pool):,} of {len(empirical_grids):,} "
      f"({100 * len(long_pool) / len(empirical_grids):.1f}%)")
print(separation_by_window.round(5).to_string(index=False))
print(f"\n  growth of separation with the window; 95% intervals from {N_SEPARATION_BOOTSTRAP:,} resamples "
      "of the paired replicates within each window")
print(separation_growth.round(4).to_string(index=False))
print(f"\n  The 28-day value of 12.1 ({separation_28['applied']:.4f}) comes from every schedule reaching "
      "day 28;\n  the fold changes above compare windows within the fixed pool only.")
save_record(separation_by_window, "exposure_separation_by_window")
save_record(separation_growth, "exposure_separation_growth")
```

```python
# ==========================================================================
# 12.3 AIC verdicts by window and noise level (fixed pool)
#     Windows of 28, 56 and 84 days on the schedules that reach day 84, at
#     four noise levels including the measured noise. The recorded values
#     are at the measured noise with the termination rule applied; ranges
#     are over the two half-lives.
# ==========================================================================

CONFIRM_WINDOWS = (28, 56, 84)
CONFIRM_NOISE_LEVELS = (0.025, 0.05, MEASUREMENT_NOISE, 0.15)

confirm_rng = np.random.default_rng(SEEDS["exposure_confirmation"])
confirm_rows = []
for half_life in EXPOSURE_HALF_LIVES:
    for window_days in CONFIRM_WINDOWS:
        dose_days = dosing_days(DOSING_INTERVAL_DAYS, window_days)
        for noise_sd in CONFIRM_NOISE_LEVELS:
            for replicate in range(N_EXPOSURE_REPLICATES):
                days = draw_schedule(confirm_rng, long_pool, window_days)
                if days is None:
                    continue
                clean, noise = paired_exposure_trajectories(confirm_rng, days, dose_days, half_life)
                for row in score_exposure_pair(clean, noise, noise_sd, days, dose_days, half_life,
                                               TIE_THRESHOLD):
                    confirm_rows.append({"half_life_days": half_life, "window_days": window_days,
                                         "noise_sd": noise_sd, "replicate": replicate, **row})
exposure_confirmation = pd.DataFrame(confirm_rows)
confirmation_summary = (exposure_confirmation
                        .groupby(["termination", "half_life_days", "window_days", "noise_sd"])
                        .apply(summarize_verdicts, include_groups=False).reset_index())
at_measured_noise = confirmation_summary.loc[np.isclose(confirmation_summary.noise_sd, MEASUREMENT_NOISE)]

for rule in TERMINATION_RULES:
    suffix = "" if rule == "applied" else "_without_termination"
    for window_days in (28, 84):
        cell = at_measured_noise.loc[(at_measured_noise.termination == rule)
                                     & (at_measured_noise.window_days == window_days)]
        record(f"exposure_decisive_{window_days}_min{suffix}", float(cell.decisive_rate.min()))
        record(f"exposure_decisive_{window_days}_max{suffix}", float(cell.decisive_rate.max()))
        record(f"exposure_accuracy_{window_days}_min{suffix}", float(cell.accuracy_when_decisive.min()))
        record(f"exposure_accuracy_{window_days}_max{suffix}", float(cell.accuracy_when_decisive.max()))
record("exposure_terminated_share_84", float(at_measured_noise.loc[
    (at_measured_noise.termination == "applied") & (at_measured_noise.window_days == 84),
    "terminated_share"].mean()))

print("12.3 AIC VERDICTS BY WINDOW AND NOISE LEVEL (fixed pool)")
print(f"  {len(exposure_confirmation) // 2:,} scored arms per termination rule")
print(confirmation_summary.round(3).to_string(index=False))
print(f"\n  at the measured noise ({MEASUREMENT_NOISE:.4f} log units)")
columns = ["termination", "half_life_days", "window_days", "n", "median_measurements", "decisive_rate",
           "decisive_low", "decisive_high", "decisive_calls", "correct_calls", "accuracy_when_decisive",
           "accuracy_low", "accuracy_high", "terminated_share"]
print(at_measured_noise[columns].round(3).to_string(index=False))
longest = exposure_confirmation.loc[(exposure_confirmation.window_days == max(CONFIRM_WINDOWS))
                                    & (exposure_confirmation.termination == "applied")
                                    & exposure_confirmation.verdict.isin(MECHANISMS)]
print("\n  accuracy among decisive calls at 84 days by generating model (termination rule applied)")
print(longest.groupby(["noise_sd", "true_mechanism"]).correct.mean().unstack().round(3).to_string())
save_record(confirmation_summary, "exposure_verdicts_by_window")

# ==========================================================================
# 12.4 Span or number of measurements: thinning the 84-day schedules
#     The 84-day schedules are thinned to every second or third measurement,
#     which brings the count near that of a 28-day schedule while keeping the
#     span. Each thinned arm is scored at the measured noise and its
#     structural separation is measured without noise.
# ==========================================================================

THINNING_STEPS = (1, 2, 3)
THINNING_WINDOW_DAYS = 84

thinning_rng = np.random.default_rng(SEEDS["exposure_thinning"])
dose_days = dosing_days(DOSING_INTERVAL_DAYS, THINNING_WINDOW_DAYS)
thinning_rows = []
for half_life in EXPOSURE_HALF_LIVES:
    for keep_every in THINNING_STEPS:
        for replicate in range(N_EXPOSURE_REPLICATES):
            days = draw_schedule(thinning_rng, long_pool, THINNING_WINDOW_DAYS, keep_every)
            if days is None:
                continue
            clean, noise = paired_exposure_trajectories(thinning_rng, days, dose_days, half_life)
            for noise_sd in (0.0, MEASUREMENT_NOISE):
                for row in score_exposure_pair(clean, noise, noise_sd, days, dose_days, half_life,
                                               TIE_THRESHOLD):
                    thinning_rows.append({"half_life_days": half_life, "keep_every": keep_every,
                                          "noise_sd": noise_sd, "replicate": replicate, **row})
exposure_thinning = pd.DataFrame(thinning_rows)
thinning_verdicts = (exposure_thinning.loc[exposure_thinning.noise_sd > 0]
                     .groupby(["termination", "half_life_days", "keep_every"])
                     .apply(summarize_verdicts, include_groups=False).reset_index())
thinning_separation = (exposure_thinning.loc[exposure_thinning.noise_sd == 0]
                       .groupby(["termination", "half_life_days", "keep_every"])
                       .agg(arms=("rmse_wrong", "count"), median_measurements=("n_measurements", "median"),
                            median_separation=("rmse_wrong", "median")).reset_index())
thinned = thinning_verdicts.loc[(thinning_verdicts.termination == "applied")
                                & (thinning_verdicts.keep_every == max(THINNING_STEPS))]
record("exposure_thinned_decisive_min", float(thinned.decisive_rate.min()))
record("exposure_thinned_decisive_max", float(thinned.decisive_rate.max()))
record("exposure_thinned_accuracy_min", float(thinned.accuracy_when_decisive.min()))
record("exposure_thinned_accuracy_max", float(thinned.accuracy_when_decisive.max()))
record("exposure_thinned_median_measurements", float(thinned.median_measurements.median()))

print("\n12.4 THINNING THE 84-DAY SCHEDULES")
print(f"  AIC verdicts at the measured noise ({MEASUREMENT_NOISE:.4f})")
print(thinning_verdicts.round(3).to_string(index=False))
print("\n  structural separation without noise")
print(thinning_separation.round(5).to_string(index=False))
reference_28 = at_measured_noise.loc[(at_measured_noise.termination == "applied")
                                     & (at_measured_noise.window_days == 28)]
print(f"\n  every third measurement over 84 days: median {thinned.median_measurements.median():.0f} "
      f"measurements, decisive rate {thinned.decisive_rate.min():.2f} to {thinned.decisive_rate.max():.2f}, "
      f"accuracy {thinned.accuracy_when_decisive.min():.2f} to {thinned.accuracy_when_decisive.max():.2f}")
print(f"  28-day schedules (12.3): median {reference_28.median_measurements.median():.0f} measurements, "
      f"decisive rate {reference_28.decisive_rate.min():.2f} to {reference_28.decisive_rate.max():.2f}, "
      f"accuracy {reference_28.accuracy_when_decisive.min():.2f} to "
      f"{reference_28.accuracy_when_decisive.max():.2f}")
save_record(thinning_verdicts, "exposure_thinning_verdicts")
save_record(thinning_separation, "exposure_thinning_separation")

# The worked example for the identifiability figure: a noise-free
# tolerant-subpopulation arm under explicit exposure (fixed parameters,
# 0.25-day half-life) on the typical schedule of the fixed pool, cut at 28
# and at 84 days, with the residual left by the best declining-effect fit.
DEMONSTRATION = {"log_baseline": np.log(200.0), "growth_rate": 0.05, "kill_rate": 0.13,
                 "tolerant_fraction": 0.08, "half_life_days": 0.25}
schedule_84 = typical_schedule(long_pool, 84)
example_rows = []
for window_days in (28, 84):
    days = schedule_84[schedule_84 <= window_days]
    doses = dosing_days(DOSING_INTERVAL_DAYS, window_days)
    truth_curve = log_volume_tolerant_exposure(
        days, DEMONSTRATION["log_baseline"], DEMONSTRATION["growth_rate"], DEMONSTRATION["kill_rate"],
        DEMONSTRATION["tolerant_fraction"], doses, DEMONSTRATION["half_life_days"])
    wrong_fit = fit_mechanism_exposure("declining drug effect", days, truth_curve, doses,
                                       DEMONSTRATION["half_life_days"])
    residual = truth_curve - exposure_forward("declining drug effect", doses,
                                              DEMONSTRATION["half_life_days"])(wrong_fit["x"], days)
    example_rows.append(pd.DataFrame({"window_days": window_days, "day": days, "log_volume": truth_curve,
                                      "residual": residual, "rmse_incorrect_model": wrong_fit["rmse"]}))
    record(f"exposure_example_rmse_{window_days}", float(wrong_fit["rmse"]))
exposure_residual_example = pd.concat(example_rows, ignore_index=True)
print("\n  worked example (0.25-day half-life, typical schedule of the fixed pool): RMSE of the "
      "incorrect model " + ", ".join(f"{w} days {RESULTS[f'exposure_example_rmse_{w}']:.4f}" for w in (28, 84)))
save_record(exposure_residual_example, "exposure_residual_example")
```

```python
# ==========================================================================
# 12.5 The caveats of the exposure simulations, quantified
#     (a) The Wilson intervals of 12.3 treat the two arms of a paired
#         replicate as independent, though they share every parameter but
#         the mechanism. The decisive rate and the accuracy are recomputed
#         here with intervals that resample paired replicates.
#     (b) A fit that ends on a fitting bound is pinned; the same summaries
#         are given among arms with no pinned fit.
#     (c) Every result is conditional on the simulated parameter
#         distributions, which give strong kill (|Normal(0.13, 0.04)| per
#         day). The 28- and 84-day verdicts at the measured noise are
#         repeated with weaker kill (|Normal(0.06, 0.02)|).
# ==========================================================================

exposure_cluster_rng = np.random.default_rng(SEEDS["exposure_cluster_bootstrap"])
at_measured = exposure_confirmation.loc[np.isclose(exposure_confirmation.noise_sd, MEASUREMENT_NOISE)]
clustered_rows = []
for (rule, half_life, window_days), block in at_measured.groupby(["termination", "half_life_days", "window_days"]):
    for subset_label, subset in (("all arms", block), ("arms without a pinned fit", block.loc[~block.any_pinned])):
        decisive = subset.verdict.isin(MECHANISMS)
        per_replicate = (subset.assign(decisive=decisive, correct_call=decisive & subset.correct)
                         .groupby("replicate").agg(arms=("verdict", "size"), decisive=("decisive", "sum"),
                                                   correct=("correct_call", "sum")))
        arms_count = per_replicate.arms.to_numpy(float)
        decisive_count = per_replicate.decisive.to_numpy(float)
        correct_count = per_replicate.correct.to_numpy(float)
        rate_draws, accuracy_draws = [], []
        for _ in range(N_BOOTSTRAP):
            picked = exposure_cluster_rng.integers(0, len(per_replicate), len(per_replicate))
            rate_draws.append(decisive_count[picked].sum() / arms_count[picked].sum())
            if decisive_count[picked].sum() > 0:
                accuracy_draws.append(correct_count[picked].sum() / decisive_count[picked].sum())
        clustered_rows.append({
            "termination": rule, "half_life_days": half_life, "window_days": window_days, "arms": subset_label,
            "n": int(arms_count.sum()), "paired_replicates": len(per_replicate),
            "pinned_share": float(block.any_pinned.mean()),
            "decisive_rate": decisive_count.sum() / arms_count.sum(),
            "decisive_low": float(np.percentile(rate_draws, 2.5)), "decisive_high": float(np.percentile(rate_draws, 97.5)),
            "decisive_calls": int(decisive_count.sum()),
            "accuracy_when_decisive": correct_count.sum() / decisive_count.sum() if decisive_count.sum() else np.nan,
            "accuracy_low": float(np.percentile(accuracy_draws, 2.5)) if accuracy_draws else np.nan,
            "accuracy_high": float(np.percentile(accuracy_draws, 97.5)) if accuracy_draws else np.nan})
exposure_clustered = pd.DataFrame(clustered_rows)
for row in exposure_clustered.loc[(exposure_clustered.termination == "applied")
                                  & (exposure_clustered.arms == "all arms")].itertuples(index=False):
    tag = f"{row.window_days}_{HALF_LIFE_TAG[row.half_life_days]}"
    record(f"exposure_pinned_share_{tag}", row.pinned_share)
    record(f"exposure_clustered_decisive_{tag}", row.decisive_rate)
    record(f"exposure_clustered_decisive_{tag}_low", row.decisive_low)
    record(f"exposure_clustered_decisive_{tag}_high", row.decisive_high)

# (c) Weaker kill, at the measured noise, 28 and 84 days, fixed pool.
WEAK_KILL_CENTRE, WEAK_KILL_SPREAD = 0.06, 0.02
weak_kill_rng = np.random.default_rng(SEEDS["exposure_weak_kill"])
weak_rows = []
for half_life in EXPOSURE_HALF_LIVES:
    for window_days in (28, 84):
        dose_days = dosing_days(DOSING_INTERVAL_DAYS, window_days)
        for replicate in range(N_EXPOSURE_REPLICATES):
            days = draw_schedule(weak_kill_rng, long_pool, window_days)
            if days is None:
                continue
            clean, noise = paired_exposure_trajectories(weak_kill_rng, days, dose_days, half_life,
                                                        kill_centre=WEAK_KILL_CENTRE, kill_spread=WEAK_KILL_SPREAD)
            for noise_sd in (0.0, MEASUREMENT_NOISE):
                for row in score_exposure_pair(clean, noise, noise_sd, days, dose_days, half_life, TIE_THRESHOLD):
                    weak_rows.append({"half_life_days": half_life, "window_days": window_days,
                                      "noise_sd": noise_sd, "replicate": replicate, **row})
exposure_weak_kill = pd.DataFrame(weak_rows)
weak_kill_summary = (exposure_weak_kill.loc[exposure_weak_kill.noise_sd > 0]
                     .groupby(["termination", "half_life_days", "window_days"])
                     .apply(summarize_verdicts, include_groups=False).reset_index())
weak_kill_separation = (exposure_weak_kill.loc[exposure_weak_kill.noise_sd == 0]
                        .groupby(["termination", "half_life_days", "window_days"])
                        .rmse_wrong.median().rename("median_separation").reset_index())
weak_kill_summary = weak_kill_summary.merge(weak_kill_separation,
                                            on=["termination", "half_life_days", "window_days"], how="left")
for window_days in (28, 84):
    cell = weak_kill_summary.loc[(weak_kill_summary.termination == "applied")
                                 & (weak_kill_summary.window_days == window_days)]
    record(f"exposure_weak_kill_decisive_{window_days}_min", float(cell.decisive_rate.min()))
    record(f"exposure_weak_kill_decisive_{window_days}_max", float(cell.decisive_rate.max()))
    record(f"exposure_weak_kill_accuracy_{window_days}_min", float(cell.accuracy_when_decisive.min()))
    record(f"exposure_weak_kill_accuracy_{window_days}_max", float(cell.accuracy_when_decisive.max()))
    record(f"exposure_weak_kill_separation_{window_days}", float(cell.median_separation.median()))

print("12.5 CAVEATS OF THE EXPOSURE SIMULATIONS")
print(f"  (a, b) at the measured noise ({MEASUREMENT_NOISE:.4f}); intervals resample paired replicates")
print(exposure_clustered.round(3).to_string(index=False))
print(f"\n  (c) weaker kill, |Normal({WEAK_KILL_CENTRE}, {WEAK_KILL_SPREAD})| per day, at the measured noise")
print(weak_kill_summary.round(3).to_string(index=False))
strong_reference = at_measured_noise.loc[at_measured_noise.window_days.isin([28, 84])]
print("\n  strong kill (12.3) for comparison")
print(strong_reference[["termination", "half_life_days", "window_days", "decisive_rate", "decisive_low",
                        "decisive_high", "accuracy_when_decisive", "accuracy_low", "accuracy_high"]]
      .round(3).to_string(index=False))
save_record(exposure_clustered, "exposure_verdicts_clustered")
save_record(weak_kill_summary, "exposure_weak_kill")
```

## 13 · External cohorts: between-animal variation and measurement error

The source study tested each model–treatment pair in one animal. Some arms have replicate readings on the same day (Section 2.2), but no arm has a second animal, so between-animal variation cannot be estimated from it. Two external cohorts with several animals per arm are used instead. Each is analyzed only when its file is present in `data/` and is skipped with a printed line otherwise.

**13.1 PDXNet** (public export from the PDXNet Portal; caliper volumes for nine PDX models from one contributing center). An animal is identified by model, arm, animal ID and `mouse.id`. Without `mouse.id`, two animals can share an ID within one arm, and the cell checks that the chosen key leaves no animal with two readings on one day. The untreated arm is marked `Control` in `Arms`, and the cell checks that `Type` and `Dose.Level` agree with it.

For each animal with at least four readings over at least five days, a line is fitted to log volume. The slope is the growth rate and its standard error comes from the residuals. In each model–arm group of at least three animals, the between-animal SD of the growth rate is corrected for slope error by subtracting the group's mean squared standard error from the variance of the slopes, with a floor at zero.

The primary comparison uses untreated groups only, because the source-study reference is the between-model SD of untreated growth rates (Section 3.6). The SD over all groups, treated included, is printed beside it. Per-animal growth rates are whole-series slopes, comparable with untreated growth rates and not with post-nadir regrowth. PDXNet series are shorter than the source study's untreated arms.

**13.2 Measurement error without a growth model.** Within each series, each interior reading is predicted by linear interpolation between its two neighbors. With locally linear growth, the residual divided by √(1 + w₀² + w₂²) estimates the measurement SD, where w₀ and w₂ are the interpolation weights. Untreated and treated series are analyzed separately, because the curvature of regression and regrowth inflates the estimate in treated series. Intervals are 95% bootstrap intervals that resample whole series. Each estimate is also given for triples spanning 6 to 10 days, the span of the Oberg cohort's 4-day schedule.

**Oberg 2021** (provided by the authors on request; ultrasound-derived tumor area for five ovarian PDX models, four arms each). Readings below the limit of detection (0.1) are excluded from the noise estimate. The cohort measures area, not volume, so its log-scale error is on a different quantity. If volume scales with area to the power 3/2, the same error on the volume scale would be 1.5 times larger.

**13.3 Like-for-like ratio.** The ratio of 13.1 divides a between-animal SD corrected for slope error by a source-study SD that is not, and that SD already contains between-animal variation, because each source-study PDX model is one animal. Section 13.3 removes the slope-error variance from the source-study variance and then the between-animal variance, and gives the ratio against each, for whole-arm slopes and for slopes refitted over the PDXNet untreated span.

**13.4 Selection judged against a second animal.** Regret is measured against the arm with the best observed day-24 volume, from one animal, which need not be the truly best treatment. Section 13.4 draws the outcome a second animal would give, under measurement noise alone, under the treated-arm interpolation noise, and under measurement noise plus the PDXNet between-animal variation in growth rate, and scores every selection against it, beside the regret of the best observed arm (the floor that outcome noise leaves) and of a random choice.

The Oberg log-area error is also stored on the volume scale (1.5 times the log-area value, if volume scales with area to the power 3/2), labelled as resting on that scaling.

**13.5 Discordance expected from between-animal variation.** Added after review. Section 4.5 gave the share of depth-matched pairs that measurement noise alone would make discordant. The two arms of a pair are also two animals, so here each arm's slope carries, in addition, between-animal variation of SD τ, the PDXNet between-animal SD of the growth rate (13.1), and the expected share is recomputed. This assumes that the between-animal SD of untreated growth also applies to post-nadir regrowth. The interval of the expected share comes from the interval of τ, and a joint bootstrap over PDXNet untreated groups and source-study PDX models gives the interval of the observed share minus the expected one.

**Inputs:** `observations`, `untreated_growth`, `data/pdxnet/PDXNet-PDTC-Export-*.csv`, `data/oberg2021/supp2_data.csv`, `model_pairs` (Section 4.4).
**Outputs:** `pdxnet_animals`, `pdxnet_groups`, `oberg`, `external_noise` (table `external_measurement_error.csv`), `external_ratio` (table `external_variance_ratio.csv`), `selection_against_replicate`, `between_discordance`, `between_discordance_sweep`.

```python
# ==========================================================================
# 13.1 Between-animal variation in growth rate (PDXNet)
# ==========================================================================

set_section("13")
PDXNET_SERIES_KEY = ["PDXNet_Model_ID", "Arms", "Animal_ID", "mouse.id"]
MIN_POINTS_PER_ANIMAL = 4
MIN_SPAN_PER_ANIMAL_DAYS = 5
MIN_ANIMALS_PER_GROUP = 3

pdxe_between_model_sd = record("pdxe_between_model_sd", float(untreated_growth.std()))
pdxnet_files = sorted(PDXNET_DIR.glob("PDXNet-PDTC-Export-*.csv"))
PDXNET_AVAILABLE = bool(pdxnet_files)
print("13.1 BETWEEN-ANIMAL VARIATION IN GROWTH RATE (PDXNet)")
pdxnet_animals, pdxnet_groups, between_animal = pd.DataFrame(), pd.DataFrame(), pd.DataFrame()
if not PDXNET_AVAILABLE:
    print(f"  skipped: no PDXNet-PDTC-Export-*.csv in {relative(PDXNET_DIR)}")
else:
    pdxnet = pd.read_csv(pdxnet_files[-1])
    print(f"  file {relative(pdxnet_files[-1])}; {len(pdxnet):,} rows; "
          f"{pdxnet.PDXNet_Model_ID.nunique()} PDX models; agents "
          f"{', '.join(sorted(pdxnet.AgentName.dropna().astype(str).unique()))}; "
          f"contributors {', '.join(sorted(pdxnet.Contributor.dropna().astype(str).unique()))}")

    # The animal key: the candidate keys, with the number of (series, day)
    # pairs that hold more than one reading. The chosen key must leave none.
    key_rows = []
    for key in (["PDXNet_Model_ID", "Animal_ID"], ["PDXNet_Model_ID", "Arms", "Animal_ID"], PDXNET_SERIES_KEY):
        labels = pdxnet[key].astype(str).agg("|".join, axis=1)
        key_rows.append({"key": " + ".join(key), "series": labels.nunique(),
                         "duplicated_series_days": int(pd.DataFrame({"s": labels, "d": pdxnet.OBS_DAY})
                                                       .duplicated().sum())})
    print("\n  animal key")
    print(pd.DataFrame(key_rows).to_string(index=False))
    pdxnet["series"] = pdxnet[PDXNET_SERIES_KEY].astype(str).agg("|".join, axis=1)
    assert not pdxnet.duplicated(["series", "OBS_DAY"]).any(), "an animal has two readings on one day"
    pdxnet["is_control"] = pdxnet.Arms.astype(str).eq("Control")
    assert pdxnet.is_control.equals(pdxnet.Type.astype(str).eq("Control")), "Arms and Type disagree"
    assert pdxnet.is_control.equals(pdxnet["Dose.Level"].eq(0)), "Arms and Dose.Level disagree"
    nonpositive = int((pdxnet.TUMOR_VOL <= 0).sum())
    print(f"  readings at or below 0 mm³ (left out)  {nonpositive}")
    for column in ("Dose.Level", "Unit", "Route", "Frequency"):
        print(f"  {column:<11} recorded for {100 * pdxnet[column].notna().mean():.0f}% of rows: "
              f"{', '.join(map(str, pdxnet[column].dropna().unique()[:4]))}")

    animal_rows = []
    for series, block in pdxnet.loc[pdxnet.TUMOR_VOL > 0].sort_values("OBS_DAY").groupby("series", sort=True):
        day = block.OBS_DAY.to_numpy(float)
        if len(day) < MIN_POINTS_PER_ANIMAL or np.ptp(day) < MIN_SPAN_PER_ANIMAL_DAYS:
            continue
        slope, _, residual_sd = fit_line(day, np.log(block.TUMOR_VOL.to_numpy(float)))
        animal_rows.append({"series": series, "model_id": block.PDXNet_Model_ID.iloc[0],
                            "arm": block.Arms.iloc[0], "is_control": bool(block.is_control.iloc[0]),
                            "n_points": len(day), "span_days": float(np.ptp(day)), "growth_rate": slope,
                            "residual_sd": residual_sd,
                            "slope_se": residual_sd / np.sqrt(np.sum((day - day.mean()) ** 2))})
    pdxnet_animals = pd.DataFrame(animal_rows)
    pdxnet_groups = pdxnet_animals.groupby(["model_id", "arm"]).agg(
        is_control=("is_control", "first"), animals=("series", "size"),
        mean_growth_rate=("growth_rate", "mean"), sd_raw=("growth_rate", "std"),
        mean_slope_variance=("slope_se", lambda se: float(np.mean(se ** 2)))).reset_index()
    pdxnet_groups = pdxnet_groups.loc[pdxnet_groups.animals >= MIN_ANIMALS_PER_GROUP].copy()
    pdxnet_groups["sd_corrected"] = np.sqrt(np.maximum(
        pdxnet_groups.sd_raw ** 2 - pdxnet_groups.mean_slope_variance, 0.0))

    # Median over groups, with a 95% interval that resamples groups.
    external_rng = np.random.default_rng(SEEDS["external_cohort"])
    between_rows = []
    for label, groups in (("untreated groups", pdxnet_groups.loc[pdxnet_groups.is_control]),
                          ("all groups", pdxnet_groups)):
        low, high = bootstrap_median_interval(groups.sd_corrected.to_numpy(), N_BOOTSTRAP, external_rng)
        median_sd = float(groups.sd_corrected.median())
        between_rows.append({"groups": label, "n_groups": len(groups), "pdx_models": groups.model_id.nunique(),
                             "animals": int(groups.animals.sum()), "median_animals": float(groups.animals.median()),
                             "median_sd_raw": float(groups.sd_raw.median()), "median_sd_corrected": median_sd,
                             "low": low, "high": high, "ratio_to_pdxe_between_model": median_sd / pdxe_between_model_sd,
                             "groups_at_zero": int((groups.sd_corrected == 0).sum())})
    between_animal = pd.DataFrame(between_rows)
    for row in between_animal.itertuples():
        tag = "control" if row.groups == "untreated groups" else "all"
        record(f"pdxnet_n_groups_{tag}", row.n_groups)
        record(f"pdxnet_median_animals_{tag}", row.median_animals)
        record(f"pdxnet_between_sd_{tag}", row.median_sd_corrected)
        record(f"pdxnet_between_sd_{tag}_low", row.low)
        record(f"pdxnet_between_sd_{tag}_high", row.high)
        record(f"pdxnet_between_ratio_{tag}", row.ratio_to_pdxe_between_model)
    record("pdxnet_n_models", pdxnet.PDXNet_Model_ID.nunique())
    record("pdxnet_n_animals", pdxnet.series.nunique())

    print(f"\n  animals {pdxnet.series.nunique()}; with a fitted growth rate {len(pdxnet_animals)} "
          f"(>= {MIN_POINTS_PER_ANIMAL} readings over >= {MIN_SPAN_PER_ANIMAL_DAYS} days); readings per animal, "
          f"median {pdxnet_animals.n_points.median():.0f}; span, median {pdxnet_animals.span_days.median():.0f} days")
    print(f"  model–arm groups {pdxnet_animals.groupby(['model_id', 'arm']).ngroups}, of which "
          f"{len(pdxnet_groups)} have >= {MIN_ANIMALS_PER_GROUP} animals")
    print("\n  between-animal SD of growth rate (per day), corrected for slope error; median over groups")
    print(between_animal.round(5).to_string(index=False))
    print(f"\n  source study: between-model SD of untreated growth rate {pdxe_between_model_sd:.5f} per day "
          f"({len(untreated_growth)} PDX models)")
    print("\n  per group")
    print(pdxnet_groups.round(5).to_string(index=False))
    save_record(pdxnet_animals, "pdxnet_animal_growth_rates")
    save_record(pdxnet_groups, "pdxnet_between_animal_groups")

# ==========================================================================
# 13.2 Measurement error from interpolation residuals
#     Series sets: source-study arms (one per arm, canonical table), PDXNet
#     animals, Oberg mice; untreated and treated separately, all triples and
#     triples spanning 6 to 10 days.
# ==========================================================================

OBERG_LOD = 0.1                                   # detection limit used in raw_lod
OBERG_COLUMNS = ["model", "treatment", "mouse", "initial", "day", "day_c", "raw", "raw_lod"]
MATCHED_SPAN_DAYS = (6.0, 10.0)

OBERG_AVAILABLE = OBERG_PATH.exists()
oberg = pd.DataFrame()
if OBERG_AVAILABLE:
    oberg = pd.read_csv(OBERG_PATH)
    oberg.columns = [column.strip().lower() for column in oberg.columns]
    missing = sorted(set(OBERG_COLUMNS) - set(oberg.columns))
    assert not missing, f"Oberg file lacks columns {missing}"
    oberg["series"] = oberg.model.astype(str) + "|" + oberg.treatment.astype(str) + "|" + oberg.mouse.astype(str)
    oberg["below_lod"] = oberg.raw < OBERG_LOD
    assert oberg.below_lod.equals(oberg.raw != oberg.raw_lod), "raw_lod is not raw floored at the LOD"
    oberg["log_area"] = np.log(oberg.raw_lod)
    oberg["is_control"] = oberg.treatment.eq("Control")
    per_mouse = oberg.groupby("series").agg(cell=("treatment", "first"), model=("model", "first"),
                                            readings=("day", "size"), last_day=("day", "max"))
    per_cell = per_mouse.groupby(["model", "cell"]).size()
    record("oberg_n_mice", len(per_mouse))
    record("oberg_n_models", oberg.model.nunique())
    print("13.2 OBERG 2021 DESIGN")
    print(f"  {relative(OBERG_PATH)}: {len(oberg):,} rows; {oberg.model.nunique()} PDX models x "
          f"{oberg.treatment.nunique()} arms; {len(per_mouse)} mice "
          f"(per model–arm cell: median {per_cell.median():.0f}, range {per_cell.min()} to {per_cell.max()})")
    print(f"  mice not followed to day {oberg.day.max():.0f}: {int((per_mouse.last_day < oberg.day.max()).sum())} "
          f"of {len(per_mouse)}")
    print(f"  readings below the detection limit: {int(oberg.below_lod.sum())} of {len(oberg)} "
          f"({100 * oberg.below_lod.mean():.1f}%), left out of the noise estimate")
else:
    print(f"13.2 Oberg 2021 skipped: {relative(OBERG_PATH)} not found (available from the authors on request)")


noise_sets = {
    ("Source study", "untreated"): series_frames(observations.loc[observations.is_control],
                                                  "arm_id", "day", "log_volume"),
    ("Source study", "treated"): series_frames(observations.loc[~observations.is_control],
                                                "arm_id", "day", "log_volume"),
}
if PDXNET_AVAILABLE:
    positive = pdxnet.loc[pdxnet.TUMOR_VOL > 0].assign(log_volume=lambda f: np.log(f.TUMOR_VOL))
    noise_sets[("PDXNet", "untreated")] = series_frames(positive.loc[positive.is_control],
                                                         "series", "OBS_DAY", "log_volume")
    noise_sets[("PDXNet", "treated")] = series_frames(positive.loc[~positive.is_control],
                                                       "series", "OBS_DAY", "log_volume")
if OBERG_AVAILABLE:
    detected = oberg.loc[~oberg.below_lod]
    noise_sets[("Oberg (log area)", "untreated")] = series_frames(detected.loc[detected.is_control],
                                                                   "series", "day", "log_area")
    noise_sets[("Oberg (log area)", "treated")] = series_frames(detected.loc[~detected.is_control],
                                                                 "series", "day", "log_area")

noise_rng = np.random.default_rng(SEEDS["external_noise_bootstrap"])
noise_rows = []
for (cohort, arm_type), frames in noise_sets.items():
    for span_label, span_range in (("all triples", None), ("span 6-10 days", MATCHED_SPAN_DAYS)):
        noise_estimate = summarize_interpolation_noise(interpolation_noise(frames, span_range), noise_rng)
        noise_rows.append({"cohort": cohort, "arm_type": arm_type, "triples": span_label,
                           "series_available": len(frames), **noise_estimate})
external_noise = pd.DataFrame(noise_rows)
for row in external_noise.loc[external_noise.triples == "all triples"].itertuples():
    tag = f"noise_interp_{row.cohort.split()[0].lower()}_{row.arm_type}"
    record(tag, row.sigma)
    record(f"{tag}_low", row.low)
    record(f"{tag}_high", row.high)
    record(f"{tag}_series", row.n_series)
print("\n13.2 MEASUREMENT ERROR FROM INTERPOLATION RESIDUALS (log scale)")
print(external_noise.round(4).to_string(index=False))
print(f"\n  residual-based measurement noise of the source study (Section 3.6): {MEASUREMENT_NOISE:.4f}")
# The Oberg cohort measures tumor AREA by ultrasound. If volume scales as
# area to the power 3/2, log volume = 1.5 × log area + a constant, so the
# same measurement error on the volume scale is 1.5 times the log-area value.
# It is stored beside the log-area value, labelled, for comparison with the
# volume-scale estimates of the other cohorts; it rests on that scaling.
OBERG_AREA_TO_VOLUME_EXPONENT = 1.5
for row in external_noise.loc[(external_noise.triples == "all triples")
                              & external_noise.cohort.str.startswith("Oberg")].itertuples():
    tag = f"noise_interp_oberg_{row.arm_type}_volume_scale"
    record(tag, OBERG_AREA_TO_VOLUME_EXPONENT * row.sigma)
    record(f"{tag}_low", OBERG_AREA_TO_VOLUME_EXPONENT * row.low)
    record(f"{tag}_high", OBERG_AREA_TO_VOLUME_EXPONENT * row.high)
    print(f"  Oberg {row.arm_type}: {row.sigma:.4f} on the log-area scale, about "
          f"{OBERG_AREA_TO_VOLUME_EXPONENT * row.sigma:.4f} on the log-volume scale if volume ∝ area^1.5")
save_record(external_noise, "external_noise")
```

```python
# ==========================================================================
# 13.3 Between-animal against between-model variation, like for like
#     The ratio of 13.1 divides a PDXNet between-animal SD that is corrected
#     for slope error by a source-study SD that is not, and that SD already
#     contains between-animal variation (each source-study PDX model is one
#     animal). Its variance decomposes as
#         var(observed slopes) = between-model + between-animal + slope error,
#     so the slope-error variance (mean squared slope SE, Section 3.6) is
#     removed first, and the PDXNet between-animal variance (median corrected
#     SD of the untreated groups, squared) second, leaving the between-model
#     component. Ratios against each are given, with 95% intervals that
#     resample source-study untreated arms and PDXNet untreated groups
#     independently. PDXNet series are shorter, and untreated growth slows
#     over time (Section 6.1), so the decomposition is repeated with the
#     source-study slopes refitted over the PDXNet untreated span.
# ==========================================================================

external_ratio = pd.DataFrame()
if not PDXNET_AVAILABLE:
    print("13.3 skipped: no PDXNet export")
else:
    ratio_rng = np.random.default_rng(SEEDS["external_ratio_bootstrap"])
    control_group_sd = pdxnet_groups.loc[pdxnet_groups.is_control, "sd_corrected"].to_numpy(float)
    between_animal_sd = float(np.median(control_group_sd))
    matched_span_days = float(pdxnet_animals.loc[pdxnet_animals.is_control, "span_days"].median())

    # Source-study untreated slopes over the first `matched_span_days` days.
    matched_rows = []
    for arm_id, block in arm_series(observations.loc[observations.is_control]):
        day = block.day.to_numpy(float) - block.day.iloc[0]
        log_volume = block.log_volume.to_numpy(float)
        keep = day <= matched_span_days
        if keep.sum() < MIN_POINTS_PER_ANIMAL or np.ptp(day[keep]) < MIN_SPAN_PER_ANIMAL_DAYS:
            continue
        slope, _, residual_sd = fit_line(day[keep], log_volume[keep])
        matched_rows.append({"growth_rate": slope,
                             "slope_se": residual_sd / np.sqrt(np.sum((day[keep] - day[keep].mean()) ** 2))})
    slope_sets = {"whole untreated arm (Section 3.6)": untreated_fits[["growth_rate", "slope_se"]],
                  f"first {matched_span_days:.0f} days (PDXNet span)": pd.DataFrame(matched_rows)}

    ratio_rows = []
    for (label, frame), tag in zip(slope_sets.items(), ("", "_matched_span")):
        slopes = frame.growth_rate.to_numpy(float)
        slope_error_variance_by_arm = frame.slope_se.to_numpy(float) ** 2
        observed_variance = float(np.var(slopes, ddof=1))
        slope_error_variance = float(np.mean(slope_error_variance_by_arm))
        biological_variance = max(observed_variance - slope_error_variance, 0.0)
        between_model_variance = biological_variance - between_animal_sd ** 2
        draws = []
        for _ in range(N_BOOTSTRAP):
            arms_drawn = ratio_rng.integers(0, len(slopes), len(slopes))
            groups_drawn = ratio_rng.integers(0, len(control_group_sd), len(control_group_sd))
            animal_sd = float(np.median(control_group_sd[groups_drawn]))
            observed = float(np.var(slopes[arms_drawn], ddof=1))
            biological = max(observed - float(np.mean(slope_error_variance_by_arm[arms_drawn])), 0.0)
            between_model = biological - animal_sd ** 2
            draws.append((animal_sd / np.sqrt(observed),
                          animal_sd / np.sqrt(biological) if biological > 0 else np.nan,
                          animal_sd / np.sqrt(between_model) if between_model > 0 else np.nan))
        draws = np.asarray(draws)
        row = {"source_study_slopes": label, "arms": len(slopes),
               "observed_sd": np.sqrt(observed_variance), "slope_error_sd": np.sqrt(slope_error_variance),
               "slope_error_fraction_of_variance": slope_error_variance / observed_variance,
               "biological_sd": np.sqrt(biological_variance),
               "between_animal_sd_pdxnet": between_animal_sd,
               "between_animal_fraction_of_biological": (between_animal_sd ** 2 / biological_variance
                                                         if biological_variance > 0 else np.nan),
               "between_model_sd": np.sqrt(between_model_variance) if between_model_variance > 0 else np.nan,
               "ratio_as_reported": between_animal_sd / np.sqrt(observed_variance),
               "ratio_slope_error_corrected": (between_animal_sd / np.sqrt(biological_variance)
                                               if biological_variance > 0 else np.nan),
               "ratio_between_model": (between_animal_sd / np.sqrt(between_model_variance)
                                       if between_model_variance > 0 else np.nan),
               "share_of_resamples_without_between_model_variance": float(np.mean(np.isnan(draws[:, 2])))}
        for column, index in (("ratio_as_reported", 0), ("ratio_slope_error_corrected", 1), ("ratio_between_model", 2)):
            row[f"{column}_low"] = float(np.nanpercentile(draws[:, index], 2.5)) if np.isfinite(draws[:, index]).any() else np.nan
            row[f"{column}_high"] = float(np.nanpercentile(draws[:, index], 97.5)) if np.isfinite(draws[:, index]).any() else np.nan
        ratio_rows.append(row)
        record(f"pdxe_slope_error_fraction_of_variance{tag}", row["slope_error_fraction_of_variance"])
        for key, column in (("pdxnet_ratio_corrected_control", "ratio_slope_error_corrected"),
                            ("pdxnet_ratio_between_model_control", "ratio_between_model")):
            record(f"{key}{tag}", row[column])
            record(f"{key}{tag}_low", row[f"{column}_low"])
            record(f"{key}{tag}_high", row[f"{column}_high"])
    external_ratio = pd.DataFrame(ratio_rows)
    print("13.3 BETWEEN-ANIMAL AGAINST BETWEEN-MODEL VARIATION, LIKE FOR LIKE (untreated groups only)")
    print(f"  PDXNet between-animal SD (median over untreated groups, corrected for slope error) "
          f"{between_animal_sd:.5f} per day; PDXNet untreated span, median {matched_span_days:.0f} days")
    print(external_ratio.T.to_string(header=False))
    print("  ratio_as_reported is the ratio of 13.1; the main text should use the untreated-group ratio, "
          "corrected on both sides.")
    save_record(external_ratio, "external_variance_ratio")
```

```python
# ==========================================================================
# 13.4 Treatment selection judged against a second animal
#     Regret is measured against the arm with the best OBSERVED day-24
#     volume, from one animal, which is not necessarily the truly best
#     treatment. replicate_outcome_regret (Section 1.3) draws the outcome a
#     second animal would give and scores every selection against it. Three
#     scenarios for the SD of (second animal − first animal):
#       measurement noise only, √2σ with σ from Section 3.6;
#       treated-arm interpolation noise, √2σ (an upper value for noise);
#       measurement noise plus between-animal variation in growth rate from
#       PDXNet (median untreated-group SD, corrected for slope error) over
#       the 24 days to the outcome. Baseline-volume differences between
#       animals are ignored, so this is still a lower bound on replicate
#       variation. Needs the PDXNet export.
#     For each scenario: mean regret of each set's choice, of the best
#     observed arm (the regret outcome noise alone leaves; no selection can
#     beat it on average), and of a random choice; the share of PDX models
#     whose best observed arm stays best; and the relative regret reduction
#     of the trajectory set against the depth set. The 95% ranges are over
#     replicate draws only (they do not resample PDX models).
# ==========================================================================

N_REPLICATE_DRAWS = 2000
replicate_rng = np.random.default_rng(SEEDS["replicate_noise"])
REPLICATE_SCENARIOS = {"measurement": ("measurement noise (σ, Section 3.6)", np.sqrt(2.0) * MEASUREMENT_NOISE),
                       "treated": ("treated-arm interpolation noise (upper value)", np.sqrt(2.0) * TREATED_NOISE)}
if PDXNET_AVAILABLE:
    between_animal_growth_sd = float(pdxnet_groups.loc[pdxnet_groups.is_control, "sd_corrected"].median())
    REPLICATE_SCENARIOS["between_animal"] = (
        "measurement noise and between-animal growth (PDXNet)",
        np.sqrt(2.0) * np.sqrt(MEASUREMENT_NOISE ** 2 + (between_animal_growth_sd * FORECAST_HORIZON_DAY) ** 2))
REPLICATE_COLUMNS = {"gradient boosting: depth": "depth", "gradient boosting: trajectory": "trajectory",
                     "gradient boosting: baseline plus slope": "baseline plus slope",
                     "least squares: baseline plus slope": "least-squares baseline plus slope"}
replicate_rows = []
for tag, selected in comparator_selected.items():
    for scenario_tag, (scenario, replicate_sd) in REPLICATE_SCENARIOS.items():
        per_draw, best_kept = replicate_outcome_regret(selected, list(REPLICATE_COLUMNS), replicate_sd,
                                                       N_REPLICATE_DRAWS, replicate_rng)
        reduction_draws = ((per_draw["gradient boosting: depth"] - per_draw["gradient boosting: trajectory"])
                           / per_draw["gradient boosting: depth"])
        row = {"population": SELECTION_POPULATIONS[tag][0], "scenario": scenario,
               "replicate_sd": replicate_sd, "best_observed_arm_stays_best": best_kept}
        for column, values in per_draw.items():
            row[f"regret, {REPLICATE_COLUMNS.get(column, column)}"] = float(values.mean())
        row["regret_reduction_trajectory_vs_depth"] = float(reduction_draws.mean())
        row["reduction_2_5"] = float(np.percentile(reduction_draws, 2.5))
        row["reduction_97_5"] = float(np.percentile(reduction_draws, 97.5))
        replicate_rows.append(row)
        stem = f"replicate_{tag}_{scenario_tag}"
        record(f"{stem}_regret_reduction", row["regret_reduction_trajectory_vs_depth"])
        # The 2.5th and 97.5th percentiles over simulated second animals,
        # not a confidence interval (PDX models are not resampled), so they
        # are stored under _draws_ names, which Section 15 does not read as
        # a 95% interval.
        record(f"{stem}_regret_reduction_draws_low", row["reduction_2_5"])
        record(f"{stem}_regret_reduction_draws_high", row["reduction_97_5"])
        record(f"{stem}_best_arm_kept", best_kept)
        record(f"{stem}_observed_best_regret", row["regret, observed best"])
selection_against_replicate = pd.DataFrame(replicate_rows)
print("13.4 TREATMENT SELECTION JUDGED AGAINST A SECOND ANIMAL")
print("  reduction_2_5 and reduction_97_5 are percentiles over simulated second animals (PDX models are not "
      "resampled); they are not confidence intervals")
if PDXNET_AVAILABLE:
    print(f"  between-animal growth SD (PDXNet untreated groups) {between_animal_growth_sd:.4f} per day, "
          f"{between_animal_growth_sd * FORECAST_HORIZON_DAY:.3f} log units over {FORECAST_HORIZON_DAY} days")
print(selection_against_replicate.T.to_string(header=False))
save_record(selection_against_replicate, "selection_against_replicate")
```

```python
# ==========================================================================
# 13.5 Discordance expected from between-animal variation (added after review)
#     Section 4.5 asked how many depth-matched pairs would be discordant if
#     the two arms had the same true regrowth rate and only measurement noise
#     separated their slopes. The two arms of a pair are also two different
#     animals. Here each arm's slope carries, in addition, between-animal
#     variation of SD τ, the PDXNet between-animal SD of the growth rate
#     (13.1: median over untreated groups, corrected for slope error), so a
#     pair's slope difference is Normal(0, σ²(1/Sxx_a + 1/Sxx_b) + 2τ²)
#     (slope_difference_exceedance, Section 1.2), and the expected share of
#     discordant pairs is the mean of the pairs' exceedance probabilities.
#     This assumes that the between-animal SD of whole-series untreated
#     growth also applies to post-nadir regrowth after treatment; PDXNet has
#     too few treated regrowth series to check that.
#     Intervals for the expected share: (i) the expected share at the bounds
#     of the 95% interval of τ (13.1), which is the 95% interval of the
#     expected share given the pairs, because the share rises monotonically
#     with τ; (ii) a joint bootstrap that resamples PDXNet untreated groups
#     (for τ) and source-study PDX models (for the pairs), which also gives
#     the interval of the observed share minus the expected one. The
#     threshold sweep of 4.4 is repeated with τ, and the expected share is
#     also given with τ from all PDXNet groups (treated included) and at
#     each noise level of 4.5.
# ==========================================================================

between_discordance, between_discordance_sweep = pd.DataFrame(), pd.DataFrame()
if not PDXNET_AVAILABLE:
    print("13.5 skipped: no PDXNet export")
else:
    tau = float(RESULTS["pdxnet_between_sd_control"])
    tau_low, tau_high = float(RESULTS["pdxnet_between_sd_control_low"]), float(RESULTS["pdxnet_between_sd_control_high"])
    tau_all_groups = float(RESULTS["pdxnet_between_sd_all"])
    arm_sxx = post_nadir_arms.set_index("arm_id").post_nadir_sxx
    pairs_13 = model_pairs.loc[model_pairs.response_gap < RESPONSE_TOLERANCE].reset_index(drop=True)
    sxx_a, sxx_b = pairs_13.arm_a.map(arm_sxx).to_numpy(float), pairs_13.arm_b.map(arm_sxx).to_numpy(float)
    observed_13 = (pairs_13.growth_rate_gap >= GROWTH_RATE_GAP).to_numpy()
    assert np.isclose(observed_13.mean(), RESULTS["discordant_share"]), "the depth-matched pairs of 4.4 changed"
    noise_levels_13 = {"median residual SD (Section 3.6)": ("median", MEASUREMENT_NOISE),
                       "pooled RMS residual SD": ("rms", RESULTS["measurement_noise_rms_linear"]),
                       "treated-arm interpolation (upper value)": ("treated", TREATED_NOISE)}
    tau_levels = {"none (measurement noise only, 4.5)": 0.0,
                  "PDXNet untreated groups (primary)": tau,
                  "PDXNet all groups, treated included": tau_all_groups}

    between_rows_13 = []
    for noise_label, (noise_tag, sigma) in noise_levels_13.items():
        for tau_label, tau_value in tau_levels.items():
            expected = float(slope_difference_exceedance(sxx_a, sxx_b, sigma, GROWTH_RATE_GAP, tau_value).mean())
            row = {"noise": noise_label, "sigma": sigma, "between_animal_sd": tau_label, "tau": tau_value,
                   "depth_matched_pairs": len(pairs_13), "observed_share": float(observed_13.mean()),
                   "expected_share": expected, "observed_minus_expected": float(observed_13.mean()) - expected,
                   "expected_at_tau_low": np.nan, "expected_at_tau_high": np.nan}
            if tau_label.startswith("PDXNet untreated"):
                # The expected share at the bounds of τ's 95% interval.
                row["expected_at_tau_low"] = float(slope_difference_exceedance(
                    sxx_a, sxx_b, sigma, GROWTH_RATE_GAP, tau_low).mean())
                row["expected_at_tau_high"] = float(slope_difference_exceedance(
                    sxx_a, sxx_b, sigma, GROWTH_RATE_GAP, tau_high).mean())
                suffix = "" if noise_tag == "median" else f"_{noise_tag}"
                record(f"discordance_between_animal_expected{suffix}", expected)
                record(f"discordance_between_animal_expected{suffix}_low", row["expected_at_tau_low"])
                record(f"discordance_between_animal_expected{suffix}_high", row["expected_at_tau_high"])
            elif tau_label.startswith("PDXNet all") and noise_tag == "median":
                record("discordance_between_animal_expected_all_groups", expected)
            between_rows_13.append(row)
    between_discordance = pd.DataFrame(between_rows_13)
    # Measurement noise alone must reproduce 4.5.
    assert np.isclose(between_discordance.loc[(between_discordance.tau == 0.0)
                                              & (between_discordance.sigma == MEASUREMENT_NOISE),
                                              "expected_share"].iloc[0],
                      RESULTS["discordance_null_expected_median"])

    # Joint bootstrap: PDXNet untreated groups (τ, as in 13.1) and source-study
    # PDX models (the pairs) are resampled independently in every replicate.
    control_group_sd = pdxnet_groups.loc[pdxnet_groups.is_control, "sd_corrected"].to_numpy(float)
    pair_rows_13 = rows_by_cluster(pairs_13.model_id)
    pair_clusters_13 = np.array(list(pair_rows_13))
    joint_rng = np.random.default_rng(SEEDS["between_animal_discordance"])
    joint_expected, joint_excess = [], []
    for _ in range(N_BOOTSTRAP):
        tau_draw = float(np.median(control_group_sd[joint_rng.integers(0, len(control_group_sd),
                                                                        len(control_group_sd))]))
        drawn = joint_rng.choice(pair_clusters_13, size=len(pair_clusters_13), replace=True)
        rows = np.concatenate([pair_rows_13[cluster] for cluster in drawn])
        expected_draw = float(slope_difference_exceedance(sxx_a[rows], sxx_b[rows], MEASUREMENT_NOISE,
                                                          GROWTH_RATE_GAP, tau_draw).mean())
        joint_expected.append(expected_draw)
        joint_excess.append(float(observed_13[rows].mean()) - expected_draw)
    expected_primary = RESULTS["discordance_between_animal_expected"]
    record("discordance_between_animal_expected_joint_low", float(np.percentile(joint_expected, 2.5)))
    record("discordance_between_animal_expected_joint_high", float(np.percentile(joint_expected, 97.5)))
    record("discordance_between_animal_excess", float(observed_13.mean()) - expected_primary)
    record("discordance_between_animal_excess_low", float(np.percentile(joint_excess, 2.5)))
    record("discordance_between_animal_excess_high", float(np.percentile(joint_excess, 97.5)))

    # The threshold sweep of 4.4 with τ added (σ from Section 3.6).
    sweep_sxx_a = model_pairs.arm_a.map(arm_sxx).to_numpy(float)
    sweep_sxx_b = model_pairs.arm_b.map(arm_sxx).to_numpy(float)
    sweep_rows_13 = []
    for tolerance in (2.5, 5.0, 10.0):
        in_tolerance = (model_pairs.response_gap < tolerance).to_numpy()
        for gap in (0.01, 0.02, 0.03, 0.05):
            sweep_rows_13.append({
                "response tolerance (points)": tolerance, "rate gap (per day)": gap,
                "depth-matched pairs": int(in_tolerance.sum()),
                "observed share": float((model_pairs.growth_rate_gap.to_numpy()[in_tolerance] >= gap).mean()),
                "expected, measurement noise only": float(slope_difference_exceedance(
                    sweep_sxx_a[in_tolerance], sweep_sxx_b[in_tolerance], MEASUREMENT_NOISE, gap).mean()),
                "expected, with between-animal variation": float(slope_difference_exceedance(
                    sweep_sxx_a[in_tolerance], sweep_sxx_b[in_tolerance], MEASUREMENT_NOISE, gap, tau).mean()),
                "expected at the lower bound of tau": float(slope_difference_exceedance(
                    sweep_sxx_a[in_tolerance], sweep_sxx_b[in_tolerance], MEASUREMENT_NOISE, gap, tau_low).mean()),
                "expected at the upper bound of tau": float(slope_difference_exceedance(
                    sweep_sxx_a[in_tolerance], sweep_sxx_b[in_tolerance], MEASUREMENT_NOISE, gap, tau_high).mean())})
    between_discordance_sweep = pd.DataFrame(sweep_rows_13)

    print("13.5 DISCORDANCE EXPECTED FROM BETWEEN-ANIMAL VARIATION (identical true rates, two animals per pair)")
    print(f"  τ (PDXNet untreated groups) {tau:.5f} per day (95% CI {tau_low:.5f} to {tau_high:.5f}); "
          f"all groups {tau_all_groups:.5f}; {len(pairs_13):,} depth-matched pairs; observed discordant share "
          f"{observed_13.mean():.3f}")
    print(between_discordance.round(4).to_string(index=False))
    print(f"\n  expected share with between-animal variation (σ from Section 3.6): "
          f"{format_ci(expected_primary, RESULTS['discordance_between_animal_expected_low'], RESULTS['discordance_between_animal_expected_high'])} "
          f"at the bounds of τ's interval; joint bootstrap "
          f"{100 * RESULTS['discordance_between_animal_expected_joint_low']:.1f} to "
          f"{100 * RESULTS['discordance_between_animal_expected_joint_high']:.1f}%")
    print(f"  observed minus expected "
          f"{format_ci(RESULTS['discordance_between_animal_excess'], RESULTS['discordance_between_animal_excess_low'], RESULTS['discordance_between_animal_excess_high'])}"
          ", joint bootstrap")
    print("\n  threshold sweep of 4.4 with between-animal variation (σ from Section 3.6)")
    print(between_discordance_sweep.round(3).to_string(index=False))
    save_record(between_discordance, "discordance_between_animal")
    save_record(between_discordance_sweep, "discordance_between_animal_sweep")
```

## 14 · Trajectory shape and progression in patients

Tumor burden is the sum of the longest diameters of the target lesions (SLD) fixed at baseline. It comes from the comparator arms of two phase III NSCLC trials released through Project Data Sphere: A6181087 (NCT00457392; erlotinib plus placebo) and CA031 (NCT00540514; carboplatin plus paclitaxel). A third study, EFC10261 (NCT00532155), was screened and excluded before any outcome was computed, because its release contains no tumor-measurement domain; the cell prints its file listing. The data are available under a data use agreement and cannot be redistributed. If they are absent the section prints a skip line.

**Endpoint.** Progression follows RECIST 1.1: a sum at least 20% and at least 5 mm above the reference, or a new lesion.

- **Reference.** The reference is the smallest sum recorded before each assessment, starting at baseline. An assessment with fewer target lesions measured than at baseline cannot set the reference, but it can show progression.
- **Lesion values.** A lesion recorded as 0 mm counts as 0 mm. A lesion recorded as too small to measure counts as 5 mm.
- **Clock.** Time to progression is measured from a landmark at the second scheduled tumor assessment: day 112 in A6181087 (assessments every 8 weeks) and day 84 in CA031 (every 6 weeks). Follow-up is censored at the last assessment.
- **Cohort.** The landmark cohort contains patients assessed on or after the landmark, free of progression (including new lesions) before it, and with at least two complete assessments before it.
- **Predictors.** The predictors come from complete assessments before the landmark: depth (1 − smallest sum / baseline sum), the slope of log SLD over the final interval, and an indicator that SLD rose over that interval.
- **Models.** The comparator Cox model contains depth alone. The trajectory model adds the two shape predictors. Concordance (Harrell's C) is computed out of fold, with five folds stratified on event status (and on study when pooled), and the difference is tested with a paired bootstrap of patients (2,000 resamples, within study when pooled).

Concordance here ranks patients by prognosis. It is not the within-tumor ranking of treatments used for the PDX models, and the treatment-selection endpoint has no counterpart in single-arm data.

**Two plans, both fixed and checksummed before their analyses.** The single-trial plan (A6181087) names the primary analysis and four sensitivity analyses: lowest reading removed, nadir clock, sum only, and baseline-adjusted. Holm correction runs across those five. The pooled plan was written after the single-trial primary analysis had been run and found no difference. CA031's file structure was known when it was written, but its panel had not been built and no outcome had been computed. It names the pooled primary analysis and five secondary analyses: leave-one-study-out, each study separately, sum only, baseline-adjusted and nadir clock. Each plan is loaded from `prespecification/` and checked against its recorded checksum.

**Departures from the plans and from the analysis first run, all of which change results:**

- **RECIST reference.** The analysis first run took the smallest sum from the clock start onward, including assessments after the one being scored. It also ignored new lesions before the landmark when forming the landmark cohort. Here the reference is the running minimum, and patients who had progressed before the landmark are excluded.
- **Lesions recorded as 0 mm** were set to 5 mm in the analysis first run. RECIST 1.1 records them as 0 mm.
- **Incomplete assessments.** In the analysis first run, incomplete assessments could set the nadir or the reference. Here they cannot, and baseline measurements spread over several days before randomization are combined into one baseline assessment.
- **Nadir clock.** Its predictors now come from assessments on or before the nadir. In the analysis first run they came from the landmark, after the clock had started.
- **Sum only.** In both families, follow-up is censored on the day a new lesion appears.
- **Pooled concordance.** The pooled Cox model has a baseline hazard per study. That choice was made at fit time and is not in the plan text. Concordance is therefore computed within study and combined; the analysis first run compared patients across trials. The unstratified pooled comparison was added after the primary analysis and is reported as added.
- **Pooled Holm family.** It has seven members: the primary analysis, leave-one-study-out, each study separately (two tests), sum only, baseline-adjusted and nadir clock. The plan lists the per-study analysis once, so counting each study separately makes the correction stricter. The plan's bootstrap resamples patients within study. Its success rule says "clustered on study", which cannot be done with two studies.
- **New lesions** are taken from an explicit flag in A6181087 and from the first appearance of a non-target lesion at a post-baseline visit in CA031. A sensitivity analysis applies first appearance in both studies (added).
- **CA031 data.** CA031 records the per-assessment sum in centimeters (converted to mm), without a per-lesion value for lesions too small to measure. A scheduled assessment's day is its scheduled week × 7, so the WEEK12 assessment falls on the day-84 landmark named in the plan. An unscheduled assessment takes its recorded scan day. The agreement between recorded and scheduled days is printed.
- **Lymph nodes.** Only the longest diameter is recorded, so the RECIST short-axis rule for lymph nodes cannot be applied. The number of target lesions described as nodes is printed.

CA031's investigator-recorded progression day is not used in either plan's analysis family. It is compared with progression derived from baseline as a check on this implementation of RECIST, and Section 14.5 uses it as the outcome of an added analysis.

**Diagnostics.** Two diagnostics are printed with the results. First, a patient with only a baseline and one later complete assessment before the landmark has a final-interval slope fixed by depth and the interval length, so the share of such patients and the variance inflation factor of each trajectory predictor are given for each study; where the shape predictors are nearly collinear with depth, the comparison cannot separate them. Second, the number of landmark-cohort patients whose outcome changes under the harmonized new-lesion definition is printed, so that a sensitivity analysis that changes no patient is identified as such; when it changes none, its row in the results table is marked as identical to the primary and it is not a separate result.

**Analyses added after both plans (14.5).** Three analyses are added and reported as added. First, derived progression in CA031 is set against the investigator's by the source of the derived event, to show whether the new-lesion rule or the sum rule drives the disagreement. Second, the landmark is moved to the third scheduled assessment (day 168 in A6181087, day 126 in CA031): at the second, almost every CA031 patient has only two complete assessments before the landmark, so the final-interval slope is a function of depth and the comparison cannot separate shape from depth. Third, CA031 is analyzed with the investigator-assessed progression day as the outcome, at both landmarks.

**Shape against a nonlinear function of depth (14.6).** Added after review. With only a baseline and one later complete assessment before the landmark, the final-interval slope is log(1 − depth) / interval, a nonlinear function of depth, so depth plus slope can gain concordance from curvature alone. The share of patients whose slope is exactly that function is printed per study and landmark, and the comparisons are repeated against a comparator that already holds the curvature (depth plus log(1 − depth)), for the primary and the sum-only outcome at the second and third assessments.

**Inputs:** `data/project_data_sphere/A6181087/` (`tmm_p`, `random`, `final`), `data/project_data_sphere/CA031/` (`lesn`, `enrl`, `eos`), `data/project_data_sphere/EFC10261/`, `prespecification/clinical_single_trial_plan.json`, `prespecification/clinical_pooled_plan.json`.
**Outputs:** `clinical_panel`, `clinical_scored`, `clinical_results` (table `clinical_results.csv`), `clinical_coefficients`, `progression_check`, `progression_sources`, `clinical_nonlinear`; figure data `clinical_panel_restricted.csv` (patient-level, derived from restricted data: keep it local and never redistribute it).

```python
# ==========================================================================
# 14.1 Plans, studies and tumor-burden panels
#     Both plans are loaded and checked against their recorded checksums
#     whether or not the trial data are present. The panels hold one row
#     per patient assessment: day since randomization, SLD in mm, and
#     whether every baseline target lesion was measured.
# ==========================================================================

set_section("14")
CLINICAL_STUDIES = {
    "A6181087": {"nct": "NCT00457392", "comparator_arm": "erlotinib plus placebo",
                 "assessment_interval_days": 56,
                 "files": ["tmm_p.sas7bdat", "random.sas7bdat"]},
    "CA031": {"nct": "NCT00540514", "comparator_arm": "carboplatin plus paclitaxel",
              "assessment_interval_days": 42,
              "files": ["lesn.sas7bdat", "enrl.sas7bdat", "eos.sas7bdat"]},
}
TOO_SMALL_TO_MEASURE_MM = 5.0      # RECIST 1.1 default for a lesion too small to measure
BASELINE_WINDOW_DAY = 0            # A6181087: lesions measured on or before day 0 form the baseline
CM_TO_MM = 10.0                    # CA031 records sums in centimeters
NODE_PATTERN = r"LYMPH|NODE|NODAL"

single_plan, single_plan_sha256 = load_prespecified_plan("clinical_single_trial")
pooled_plan, pooled_plan_sha256 = load_prespecified_plan("clinical_pooled")
assert pooled_plan["supersedes"] == single_plan_sha256, "the pooled plan does not name the single-trial plan"
assert single_plan["seed"] == SEEDS["clinical_single_trial"] and pooled_plan["seed"] == SEEDS["clinical_pooled"]
assert pooled_plan["comparator_features"] == single_plan["comparator_features"]
assert pooled_plan["trajectory_features"] == single_plan["trajectory_features"]
LANDMARK_DAYS = {"A6181087": float(pooled_plan["landmark_by_study"]["A6181087"]["landmark_day"]),
                 "CA031": float(pooled_plan["landmark_by_study"]["CA031"]["nominal_day"])}
assert LANDMARK_DAYS["A6181087"] == float(single_plan["landmark_day"])
CLINICAL_COMPARATOR = list(single_plan["comparator_features"])
CLINICAL_TRAJECTORY = list(single_plan["trajectory_features"])

print("14.1 CLINICAL ANALYSIS PLANS")
for label, plan, digest in (("single-trial (A6181087)", single_plan, single_plan_sha256),
                            ("pooled", pooled_plan, pooled_plan_sha256)):
    print(f"  {label:<24} SHA-256 {digest[:16]}  seed {plan['seed']}")
print(f"  comparator {CLINICAL_COMPARATOR}; trajectory {CLINICAL_TRAJECTORY}")
print(f"  landmarks (second scheduled assessment): {LANDMARK_DAYS}")
print(f"  single-trial sensitivity analyses: {'; '.join(single_plan['sensitivities'])}")
print(f"  pooled secondary analyses: {'; '.join(pooled_plan['secondary_analyses'])}")

CLINICAL_AVAILABLE = all((PDS_DIR / study / name).exists()
                         for study, spec in CLINICAL_STUDIES.items() for name in spec["files"])
efc_folder = PDS_DIR / "EFC10261"
if efc_folder.exists():
    efc_tables = sorted(path.stem.upper() for path in efc_folder.glob("*.sas7bdat"))
    print(f"\n  EFC10261 (NCT00532155) tables: {', '.join(efc_tables)}")
    assert not {"TR", "TU"} & set(efc_tables), "EFC10261 has a tumor domain; its exclusion must be revisited"
    print("  no TR (tumor results) or TU (tumor identification) domain, so no lesion measurements: excluded")

clinical_panel, new_lesion_days, investigator_progression = pd.DataFrame(), {}, pd.Series(dtype=float)
if not CLINICAL_AVAILABLE:
    print(f"\n  Section 14 skipped: the Project Data Sphere files are not in {relative(PDS_DIR)} "
          "(available under a data use agreement)")
else:
    # --- A6181087: per-lesion records ------------------------------------
    study_folder = PDS_DIR / "A6181087"
    tmm = read_sas_table(study_folder / "tmm_p.sas7bdat")
    randomization = read_sas_table(study_folder / "random.sas7bdat").set_index("PID_A").RANDDAY
    tmm["day"] = tmm.EFDAY - tmm.PID_A.map(randomization)
    unanchored = int(tmm.day.isna().sum())
    tmm = tmm.loc[tmm.day.notna()]
    targets = tmm.loc[tmm.LESTYPE.eq("Target")].copy()
    wrong_unit = targets.loc[targets.TMMDIA.notna() & targets.TMMDIAU.ne("MM"), "TMMDIAU"]
    assert wrong_unit.empty, f"diameters not in mm: {wrong_unit.value_counts().to_dict()}"
    too_small = targets.TMMNMES.eq("TOO SMALL TO MEASURE")
    targets.loc[too_small, "TMMDIA"] = TOO_SMALL_TO_MEASURE_MM
    zero_mm = int(targets.TMMDIA.eq(0).sum())                    # kept at 0 mm
    targets["baseline_lesion"] = (targets.groupby(["PID_A", "TMMLNM"]).day.transform("min")
                                  <= BASELINE_WINDOW_DAY)
    measured = targets.loc[targets.baseline_lesion & targets.TMMDIA.notna()].copy()
    # Baseline measurements spread over several days before randomization
    # form one baseline assessment (each lesion's latest baseline value).
    in_window = measured.day <= BASELINE_WINDOW_DAY
    multi_day_baseline = int(measured.loc[in_window].groupby("PID_A").day.nunique().gt(1).sum())
    measured["recorded_day"] = measured.day
    measured.loc[in_window, "day"] = 0.0
    duplicated_readings = int(measured.duplicated(["PID_A", "TMMLNM", "day"]).sum())
    measured = (measured.sort_values("recorded_day")
                .drop_duplicates(["PID_A", "TMMLNM", "day"], keep="last"))
    baseline_lesions = measured.loc[measured.day.eq(0)].groupby("PID_A").TMMLNM.nunique()
    panel_a = measured.groupby(["PID_A", "day"], as_index=False).agg(
        sld=("TMMDIA", "sum"), n_lesions=("TMMLNM", "nunique"))
    panel_a["baseline_lesions"] = panel_a.PID_A.map(baseline_lesions)
    panel_a["complete"] = panel_a.n_lesions.eq(panel_a.baseline_lesions)
    panel_a["patient"] = "A6181087:" + panel_a.PID_A.astype(str)
    panel_a["study"], panel_a["day_source"] = "A6181087", "recorded day"
    text_columns = targets.select_dtypes(include=["object", "string"]).columns
    node_lesions_a = targets.loc[targets[text_columns].apply(
        lambda column: column.astype(str).str.contains(NODE_PATTERN, case=False)).any(axis=1)
    ].groupby(["PID_A", "TMMLNM"]).ngroups

    # New lesions: the explicit flag (primary) and the first appearance of a
    # non-target lesion after baseline (the CA031 definition; sensitivity).
    flagged = tmm.loc[tmm.TMMNLES.eq("NEW LESION")].groupby("PID_A").day.min()
    appearance = tmm.loc[tmm.LESTYPE.ne("Target")].groupby(["PID_A", "TMMLNM"]).day.min()
    appeared = appearance.loc[appearance > BASELINE_WINDOW_DAY].groupby(level="PID_A").min()
    new_lesion_days["A6181087"] = flagged.rename(lambda pid: f"A6181087:{pid}")
    new_lesion_days["A6181087 first appearance"] = appeared.rename(lambda pid: f"A6181087:{pid}")

    # --- CA031: per-assessment sums on a ten-slot form ---------------------
    study_folder = PDS_DIR / "CA031"
    lesn = read_sas_table(study_folder / "lesn.sas7bdat")
    randomization_ca = read_sas_table(study_folder / "enrl.sas7bdat").set_index("RPT").RANDDAY
    eos = read_sas_table(study_folder / "eos.sas7bdat").set_index("RPT")
    lesn = lesn.loc[lesn.RPT.isin(randomization_ca.index)].copy()
    lesn["observed_day"] = lesn.SCANDAY - lesn.RPT.map(randomization_ca)
    week = pd.to_numeric(lesn.CPEVENT.str.extract(r"^WEEK(\d+)$", expand=False), errors="coerce")
    lesn["nominal_day"] = np.where(lesn.CPEVENT.eq("BASELINE"), 0.0, week * 7.0)
    targets_ca = lesn.loc[lesn.LESTYP.eq("TARG")]
    assert targets_ca.groupby(["RPT", "CPEVENT"]).SUMDIAM.nunique().le(1).all(), \
        "SUMDIAM differs between rows of one assessment"
    day_agreement = (targets_ca.observed_day - targets_ca.nominal_day).abs().dropna()
    ca031_assessments = targets_ca.groupby(["RPT", "CPEVENT"]).agg(
        sum_cm=("SUMDIAM", "first"), observed_day=("observed_day", "min"),
        nominal_day=("nominal_day", "first")).reset_index()
    ca031_assessments = ca031_assessments.merge(targets_ca.loc[targets_ca.DIAM.notna()].groupby(["RPT", "CPEVENT"]).LESNUM
                          .nunique().rename("n_lesions").reset_index(), how="left").fillna({"n_lesions": 0})
    # Scheduled visits keep their scheduled day (week × 7), so the WEEK12 assessment
    # sits at the day-84 landmark the pooled plan names; unscheduled visits use the
    # recorded scan day.
    ca031_assessments["day"] = ca031_assessments.nominal_day.fillna(ca031_assessments.observed_day)
    ca031_assessments["day_source"] = np.where(ca031_assessments.nominal_day.notna(), "scheduled week",
                                               "recorded day")
    unplaced = int((ca031_assessments.day.isna() | ca031_assessments.sum_cm.isna()).sum())
    ca031_assessments = ca031_assessments.loc[ca031_assessments.day.notna() & ca031_assessments.sum_cm.notna()].copy()
    ca031_assessments["sld"] = ca031_assessments.sum_cm * CM_TO_MM
    ca031_assessments["baseline_lesions"] = ca031_assessments.RPT.map(ca031_assessments.loc[ca031_assessments.CPEVENT.eq("BASELINE")]
                                                .set_index("RPT").n_lesions)
    ca031_assessments["complete"] = ca031_assessments.n_lesions.eq(ca031_assessments.baseline_lesions)
    ca031_assessments["patient"] = "CA031:" + ca031_assessments.RPT.astype(str)
    ca031_assessments["study"] = "CA031"
    node_lesions_ca = targets_ca.loc[targets_ca.LESDESC.str.contains(NODE_PATTERN, case=False, na=False)
                                     ].groupby(["RPT", "LESDESC"]).ngroups

    # New lesions: a non-target lesion first recorded at a post-baseline visit.
    nontarget_ca = lesn.loc[lesn.LESTYP.eq("NTLE") & lesn.LESDESC.fillna("").ne("")].copy()
    nontarget_ca["day"] = nontarget_ca.nominal_day.fillna(nontarget_ca.observed_day)
    by_lesion = nontarget_ca.groupby(["RPT", "LESDESC"]).agg(
        at_baseline=("CPEVENT", lambda visit: bool(visit.eq("BASELINE").any())), first_day=("day", "min"))
    appeared_ca = by_lesion.loc[~by_lesion.at_baseline].first_day.dropna().groupby(level="RPT").min()
    new_lesion_days["CA031"] = appeared_ca.rename(lambda rpt: f"CA031:{rpt}")
    investigator_progression = (eos.PROGDAY - eos.index.map(randomization_ca)).dropna().rename(
        lambda rpt: f"CA031:{rpt}")

    columns = ["study", "patient", "day", "sld", "n_lesions", "baseline_lesions", "complete", "day_source"]
    clinical_panel = pd.concat([panel_a[columns], ca031_assessments[columns]], ignore_index=True)
    panel_rows = []
    for study, panel in clinical_panel.groupby("study"):
        per_patient = panel.groupby("patient")
        spacing = panel.sort_values(["patient", "day"]).groupby("patient").day.diff()
        new_days = new_lesion_days[study]
        panel_rows.append({
            "study": study, "patients": panel.patient.nunique(), "assessments": len(panel),
            "median_assessments": float(per_patient.size().median()),
            "median_spacing_days": float(spacing.median()),
            "complete_share": float(panel.complete.mean()),
            "median_baseline_sld_mm": float(panel.loc[panel.day <= 0].groupby("patient").sld.first().median()),
            "patients_with_new_lesion": int(new_days.index.isin(panel.patient).sum()),
            "recorded_day_share": float(panel.day_source.eq("recorded day").mean())})
    panel_summary = pd.DataFrame(panel_rows)
    print("\n  TUMOR-BURDEN PANELS")
    print(panel_summary.round(3).to_string(index=False))
    print(f"\n  A6181087: lesion rows without a randomization day {unanchored}; target readings too small "
          f"to measure {int(too_small.sum())} (set to {TOO_SMALL_TO_MEASURE_MM:g} mm); at 0 mm {zero_mm} "
          f"(kept at 0); patients with baseline readings on more than one day {multi_day_baseline} "
          f"(combined); repeated lesion readings on one day {duplicated_readings} (latest kept); "
          f"target lesions described as lymph nodes {node_lesions_a}")
    print(f"  A6181087 new lesions: flagged in {len(flagged)} patients; first appearance of a non-target "
          f"lesion after baseline in {len(appeared)}")
    print(f"  CA031: |recorded − scheduled day| median {day_agreement.median():.0f} days (90th percentile "
          f"{day_agreement.quantile(0.9):.0f}) on {len(day_agreement):,} target rows with both; assessments "
          f"without a day or sum {unplaced}; with no recorded lesion diameter "
          f"{int(ca031_assessments.n_lesions.eq(0).sum())}; target lesions described as lymph nodes {node_lesions_ca}")
    print(f"  CA031: investigator progression day recorded for {len(investigator_progression)} of "
          f"{len(eos)} patients")
    save_record(panel_summary, "clinical_panels")
    # Patient-level assessments, for the clinical figure. Derived from Project
    # Data Sphere data, which may not be redistributed: keep this file local.
    save_record(clinical_panel, "clinical_panel_restricted")
```

```python
# ==========================================================================
# 14.2 Landmark cohorts, and derived progression against the investigator's
#      progression day (CA031)
# ==========================================================================

clinical_scored, clinical_cohort, nadir_cohort = pd.DataFrame(), pd.DataFrame(), pd.DataFrame()
progression_check = pd.DataFrame()
clinical_rows, clinical_coefficients = [], []
if not CLINICAL_AVAILABLE:
    print("14.2 skipped: Project Data Sphere files not present")
else:
    clinical_scored = pd.concat([
        score_patients(clinical_panel.loc[clinical_panel.study == "A6181087"], new_lesion_days["A6181087"],
                       LANDMARK_DAYS["A6181087"], "A6181087", new_lesion_days["A6181087 first appearance"]),
        score_patients(clinical_panel.loc[clinical_panel.study == "CA031"], new_lesion_days["CA031"],
                       LANDMARK_DAYS["CA031"], "CA031", new_lesion_days["CA031"]),
    ], ignore_index=True)
    in_cohort = (clinical_scored.reached_landmark & ~clinical_scored.progressed_before_landmark
                 & clinical_scored.has_features)
    clinical_cohort = clinical_scored.loc[in_cohort].reset_index(drop=True)
    assert (clinical_cohort.last_feature_day < clinical_cohort.study.map(LANDMARK_DAYS)).all(), \
        "a predictor used an assessment on or after the landmark"

    # Nadir clock: the same patients, the clock started at the lowest
    # complete sum and the predictors taken on or before it.
    nadir_cohort = clinical_cohort.dropna(subset=["nadir_depth_at_landmark", "nadir_duration"]).copy()
    for name in ("depth_at_landmark", "slope_to_landmark", "already_rising", "log_baseline_sld"):
        nadir_cohort[name] = nadir_cohort[f"nadir_{name}"]
    assert (nadir_cohort.nadir_last_feature_day <= nadir_cohort.nadir_day).all()

    clinical_attrition = clinical_scored.groupby("study").agg(
        patients=("patient", "size"), assessed_at_or_after_landmark=("reached_landmark", "sum"))
    clinical_attrition["progressed_before_landmark"] = clinical_scored.loc[clinical_scored.reached_landmark].groupby(
        "study").progressed_before_landmark.sum()
    clinical_attrition["landmark_cohort"] = clinical_cohort.groupby("study").size()
    clinical_attrition["events"] = clinical_cohort.groupby("study").event.sum()
    clinical_attrition["events_new_lesion"] = clinical_cohort.loc[clinical_cohort.source.eq("new lesion")].groupby(
        "study").size()
    clinical_attrition["events_sum_only"] = clinical_cohort.groupby("study").sld_only_event.sum()
    clinical_attrition["nadir_clock_patients"] = nadir_cohort.groupby("study").size()
    clinical_attrition["median_slope_window_days"] = clinical_cohort.groupby("study").slope_window_days.median()
    clinical_attrition = clinical_attrition.fillna(0)
    print("14.2 LANDMARK COHORTS")
    print(clinical_attrition.to_string())
    print(f"  pooled: {len(clinical_cohort)} patients, {int(clinical_cohort.event.sum())} events")
    print("\n  predictors in the landmark cohort")
    print(clinical_cohort.groupby("study")[CLINICAL_TRAJECTORY + ["log_baseline_sld"]].describe()
          .T.round(4).to_string())
    print("\n  Spearman correlation of each predictor with depth, by study")
    for study, part in clinical_cohort.groupby("study"):
        rhos = {name: spearman(part.depth_at_landmark, part[name])[0]
                for name in ("slope_to_landmark", "already_rising", "log_baseline_sld")}
        print(f"    {study:<9} " + "   ".join(f"{name} {value:+.3f}" for name, value in rhos.items()))

    # Collinearity. A patient with only a baseline and one later complete
    # assessment before the landmark has a final-interval slope that is a
    # function of depth and of the interval length alone (log(1 − depth) /
    # interval when the sum fell), so the two shape predictors add little that
    # depth does not carry, and the Cox coefficients of depth and slope become
    # unstable. The share of such patients and the variance inflation factors
    # are printed for each study.
    collinearity_rows = []
    for study, part in clinical_cohort.groupby("study"):
        factors = variance_inflation(part, CLINICAL_TRAJECTORY)
        share_two = float((part.n_assessments == 2).mean())
        collinearity_rows.append({"study": study, "patients": len(part),
                                  "share_with_two_assessments": share_two,
                                  **{f"vif_{name}": value for name, value in factors.items()}})
        record(f"clinical_{study}_share_two_assessments", share_two)
        record(f"clinical_{study}_max_vif", float(np.nanmax(list(factors.values()))))
    clinical_collinearity = pd.DataFrame(collinearity_rows)
    print("\n  collinearity of the trajectory predictors (share of patients with only two complete "
          "assessments before the landmark; variance inflation factors)")
    print(clinical_collinearity.round(3).to_string(index=False))

    # RECIST progression derived from baseline in CA031, against the day the
    # investigator recorded. A check on the implementation, not a test.
    check_rows = []
    ca_panel = clinical_panel.loc[clinical_panel.study == "CA031"].sort_values("day")
    for patient, block in ca_panel.groupby("patient", sort=True):
        if len(block) < 2:
            continue
        derived_outcome = time_to_progression(block, float(block.day.iloc[0]),
                                      new_lesion_days["CA031"].get(patient, np.nan))
        check_rows.append({"patient": patient, "derived": derived_outcome["event"],
                           "derived_day": float(block.day.iloc[0]) + derived_outcome["duration"] if derived_outcome["event"] else np.nan,
                           "investigator": patient in investigator_progression.index,
                           "investigator_day": investigator_progression.get(patient, np.nan)})
    progression_check = pd.DataFrame(check_rows)
    progressed_both = progression_check.loc[progression_check.derived & progression_check.investigator]
    day_gap = progressed_both.derived_day - progressed_both.investigator_day
    interval = CLINICAL_STUDIES["CA031"]["assessment_interval_days"]
    print("\n  CA031: RECIST progression derived from baseline against the investigator's progression day")
    print(pd.crosstab(progression_check.derived, progression_check.investigator,
                      rownames=["derived"], colnames=["investigator"]).to_string())
    print(f"  progressing by both: {len(progressed_both)}; derived minus investigator day, median {day_gap.median():+.0f}; "
          f"within one assessment interval ({interval} days) {100 * day_gap.abs().le(interval).mean():.1f}%")
    save_record(clinical_scored, "clinical_patients")

# ==========================================================================
# 14.3 The single-trial plan (A6181087)
#     Primary and four sensitivity analyses, Holm correction across the five.
#     Folds stratified on event status; patients resampled without strata.
# ==========================================================================

if CLINICAL_AVAILABLE:
    single = clinical_cohort.loc[clinical_cohort.study == "A6181087"]
    single_nadir = nadir_cohort.loc[nadir_cohort.study == "A6181087"]
    single_seed = single_plan["seed"]
    single_analyses = [
        ("primary", single, "duration", "event", CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY),
        ("lowest reading removed", single.dropna(subset=["nadir_excluded_duration"]),
         "nadir_excluded_duration", "nadir_excluded_event", CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY),
        ("nadir clock", single_nadir, "nadir_duration", "nadir_event", CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY),
        ("sum only", single, "sld_only_duration", "sld_only_event", CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY),
        ("baseline adjusted", single, "duration", "event",
         CLINICAL_COMPARATOR + ["log_baseline_sld"], CLINICAL_TRAJECTORY + ["log_baseline_sld"]),
    ]
    single_results = [compare_cox_models(frame, comparator, trajectory, single_seed, duration, event,
                                         label=name, min_events=5)
                      for name, frame, duration, event, comparator, trajectory in single_analyses]
    single_thresholds, single_rejected = holm([result["p_value"] for result in single_results])
    print("\n14.3 SINGLE-TRIAL PLAN (A6181087, landmark day "
          f"{LANDMARK_DAYS['A6181087']:.0f}; Holm across {len(single_results)})")
    for result, threshold, reject in zip(single_results, single_thresholds, single_rejected):
        report_comparison(result)
        print(f"    Holm threshold {threshold:.4f}; {'rejected' if reject else 'not rejected'}")
        clinical_rows.append({"family": "single-trial plan", "analysis": result["analysis"],
                              "status": "prespecified", **{k: result[k] for k in (
                                  "n", "events", "c_depth", "c_trajectory", "difference", "low", "high",
                                  "p_value", "bootstrap_draws")},
                              "holm_threshold": threshold, "holm_rejected": bool(reject)})
    single_coefficients = single_results[0]["coefficients"]
    print("\n  primary: trajectory-model coefficients by fold (log hazard per SD)")
    print(single_coefficients.round(4).to_string())
    print("  sign the same in every fold: " + ", ".join(
        f"{name} {'yes' if np.sign(single_coefficients[name].dropna()).nunique() == 1 else 'no'}"
        for name in CLINICAL_TRAJECTORY))
    print("\n  primary: concordance by fold (each fold's own patients)")
    print(single_results[0]["checks"].round(3).to_string(index=False))
    record("clinical_single_c_depth", single_results[0]["c_depth"])
    record("clinical_single_c_trajectory", single_results[0]["c_trajectory"])
    record("clinical_single_difference", single_results[0]["difference"])
    record("clinical_single_p", single_results[0]["p_value"])
```

```python
# ==========================================================================
# 14.4 The pooled plan (A6181087 and CA031)
#     Primary: Cox models with a baseline hazard per study, concordance
#     within study, folds stratified on event status and study, bootstrap
#     within study. Holm correction across seven tests. Then the analyses
#     added after the primary and the coefficient diagnostics.
# ==========================================================================

clinical_results = pd.DataFrame()
if not CLINICAL_AVAILABLE:
    print("14.4 skipped: Project Data Sphere files not present")
else:
    pooled_seed = pooled_plan["seed"]
    pooled_results = [
        compare_cox_models(clinical_cohort, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed, stratified=True,
                           label="primary (pooled, baseline hazard per study)", min_events=10),
        leave_one_study_out(clinical_cohort, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed,
                            label="leave-one-study-out", min_events=10),
    ]
    for study in CLINICAL_STUDIES:
        pooled_results.append(compare_cox_models(
            clinical_cohort.loc[clinical_cohort.study == study], CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed,
            label=f"{study} alone", min_events=10))
    pooled_results += [
        compare_cox_models(clinical_cohort, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed, "sld_only_duration",
                           "sld_only_event", stratified=True, label="sum only (censored at a new lesion)",
                           min_events=10),
        compare_cox_models(clinical_cohort, CLINICAL_COMPARATOR + ["log_baseline_sld"], CLINICAL_TRAJECTORY + ["log_baseline_sld"],
                           pooled_seed, stratified=True, label="baseline adjusted", min_events=10),
        compare_cox_models(nadir_cohort, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed, "nadir_duration", "nadir_event",
                           stratified=True, label="nadir clock", min_events=10),
    ]
    pooled_thresholds, pooled_rejected = holm([result["p_value"] for result in pooled_results])
    print("14.4 POOLED PLAN (Holm across "
          f"{len(pooled_results)}; plan lists the per-study analysis once)")
    for result, threshold, reject in zip(pooled_results, pooled_thresholds, pooled_rejected):
        report_comparison(result)
        print(f"    Holm threshold {threshold:.4f}; {'rejected' if reject else 'not rejected'}")
        clinical_rows.append({"family": "pooled plan", "analysis": result["analysis"], "status": "prespecified",
                              **{k: result[k] for k in ("n", "events", "c_depth", "c_trajectory", "difference",
                                                        "low", "high", "p_value", "bootstrap_draws")},
                              "holm_threshold": threshold, "holm_rejected": bool(reject)})
    primary, loso, sum_only = pooled_results[0], pooled_results[1], pooled_results[4]
    print("\n  leave-one-study-out, by held-out study (models fitted on the other study)")
    print(loso["checks"].round(3).to_string(index=False))
    print("\n  primary: concordance by fold and by study (within-study pairs)")
    print(primary["checks"].round(3).to_string(index=False))
    print("  The per-study rows reuse the pooled folds and fits, so they differ from the "
          "per-study analyses above,\n  which have their own folds.")

    # --- analyses added after the primary ----------------------------------
    added = [
        compare_cox_models(clinical_cohort, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed, stratified=False,
                           label="added: pooled, one baseline hazard, concordance across all pairs",
                           min_events=10),
    ]
    harmonized = clinical_scored.loc[clinical_scored.reached_landmark & clinical_scored.has_features
                                     & ~clinical_scored.harmonized_progressed_before_landmark.astype(bool)].copy()
    harmonized["duration"], harmonized["event"] = harmonized.harmonized_duration, harmonized.harmonized_event
    added.append(compare_cox_models(harmonized, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed, stratified=True,
                                    label="added: new lesions by first appearance in both studies",
                                    min_events=10))
    # How many landmark-cohort patients the harmonized new-lesion definition
    # actually changes. If none, the added analysis repeats the primary by
    # construction and its identical result carries no extra information.
    primary_outcomes = clinical_cohort.set_index("patient")[["duration", "event"]]
    harmonized_outcomes = harmonized.set_index("patient")[["duration", "event"]]
    shared_patients = primary_outcomes.index.intersection(harmonized_outcomes.index)
    outcome_changed = ((primary_outcomes.loc[shared_patients, "event"].astype(bool)
                        != harmonized_outcomes.loc[shared_patients, "event"].astype(bool))
                       | ~np.isclose(primary_outcomes.loc[shared_patients, "duration"].to_numpy(float),
                                     harmonized_outcomes.loc[shared_patients, "duration"].to_numpy(float),
                                     equal_nan=True))
    cohort_difference = len(set(primary_outcomes.index) ^ set(harmonized_outcomes.index))
    record("clinical_harmonized_patients_changed", int(outcome_changed.sum()))
    record("clinical_harmonized_cohort_difference", cohort_difference)
    print(f"\n  harmonized new-lesion definition: patients entering or leaving the cohort {cohort_difference}; "
          f"patients whose time or event changes {int(outcome_changed.sum())} of {len(shared_patients)}")
    if cohort_difference == 0 and not outcome_changed.any():
        print("  The definition changes no patient in the landmark cohort, so the added analysis below "
              "reproduces the primary exactly.")

    print("\n  ADDED AFTER THE PRIMARY (outside the Holm family)")
    # When the harmonized new-lesion definition changes no patient, its
    # analysis is the primary repeated, so it is marked as such in the
    # results table instead of being reported as a separate result.
    harmonized_identical = cohort_difference == 0 and not outcome_changed.any()
    for result in added:
        identical = harmonized_identical and "first appearance" in result["analysis"]
        report_comparison(result)
        if identical:
            print("    identical to the primary by construction (no patient changes): not a separate result")
        clinical_rows.append({"family": "pooled plan", "analysis": result["analysis"],
                              "status": "added, identical to the primary" if identical else "added",
                              **{k: result[k] for k in ("n", "events", "c_depth", "c_trajectory", "difference",
                                                        "low", "high", "p_value", "bootstrap_draws")}})

    # --- coefficient diagnostics -------------------------------------------
    # Fold-by-fold coefficients of the pooled models, and Cox models fitted
    # on all patients: pooled with a baseline hazard per study, and each
    # study alone. Coefficients are log hazard per SD of the predictor.
    print("\n  pooled primary: coefficients by fold (log hazard per SD)")
    print(pd.concat([primary["coefficients_depth"].add_suffix(" (depth model)"),
                     primary["coefficients"].add_suffix(" (trajectory model)")], axis=1).round(4).to_string())
    for label, frame, stratified in (("pooled, baseline hazard per study", clinical_cohort, True),
                                     *[(f"{study} alone", clinical_cohort.loc[clinical_cohort.study == study], False)
                                       for study in CLINICAL_STUDIES]):
        for model, predictors in (("depth", CLINICAL_COMPARATOR), ("trajectory", CLINICAL_TRAJECTORY)):
            full_fit = cox_full_fit(frame, predictors, stratified=stratified)
            clinical_coefficients.append(full_fit.assign(fit=label, model=model))
    clinical_coefficients = pd.concat(clinical_coefficients, ignore_index=True)[
        ["fit", "model", "predictor", "coefficient", "se", "low", "high", "p_value"]]
    print("\n  Cox models fitted on all patients")
    print(clinical_coefficients.round(4).to_string(index=False))

    clinical_results = pd.DataFrame(clinical_rows)
    record("clinical_n_trials", clinical_cohort.study.nunique())
    record("clinical_n_patients", primary["n"])
    record("clinical_n_events", primary["events"])
    record("clinical_c_depth", primary["c_depth"])
    record("clinical_c_trajectory", primary["c_trajectory"])
    record("clinical_difference", primary["difference"])
    record("clinical_difference_low", primary["low"])
    record("clinical_difference_high", primary["high"])
    record("clinical_p", primary["p_value"])
    record("clinical_unstratified_c_depth", added[0]["c_depth"])
    record("clinical_unstratified_c_trajectory", added[0]["c_trajectory"])
    record("clinical_sum_only_events", sum_only["events"])
    record("clinical_sum_only_c_depth", sum_only["c_depth"])
    record("clinical_sum_only_c_trajectory", sum_only["c_trajectory"])
    record("clinical_sum_only_difference", sum_only["difference"])
    record("clinical_sum_only_difference_low", sum_only["low"])
    record("clinical_sum_only_difference_high", sum_only["high"])
    record("clinical_sum_only_p", sum_only["p_value"])
    record("clinical_pooled_holm_rejected", int(clinical_results.loc[
        clinical_results.family.eq("pooled plan") & clinical_results.status.eq("prespecified"),
        "holm_rejected"].sum()))
    save_record(clinical_results, "clinical_results")
    save_record(clinical_coefficients, "clinical_coefficients")
```

```python
# ==========================================================================
# 14.5 Analyses added after both plans
#     (a) Derived RECIST progression against the investigator's in CA031,
#         by the source of the derived event: if most progressions found only
#         by the derivation come from the first appearance of a non-target
#         lesion, the new-lesion rule, not the sum rule, drives the
#         disagreement. Agreement is also given for the sum-only derivation.
#     (b) A landmark at the THIRD scheduled assessment (day 168 in A6181087,
#         day 126 in CA031). At the second assessment almost every CA031
#         patient has only a baseline and one later complete assessment
#         before the landmark, so the final-interval slope is a function of
#         depth and the comparison cannot separate shape from depth. A third
#         assessment gives the slope an interval of its own.
#     (c) CA031 with the investigator-assessed progression day as the
#         outcome, at the planned and at the third-assessment landmark.
#     None of these is in either plan; all are reported as added.
# ==========================================================================

progression_sources, progression_source_table = pd.DataFrame(), pd.DataFrame()
if not CLINICAL_AVAILABLE:
    print("14.5 skipped: Project Data Sphere files not present")
else:
    added_seed = pooled_plan["seed"]
    added_rows = []

    # --- (a) derived progression by source, against the investigator ---------
    source_rows = []
    for patient, block in ca_panel.groupby("patient", sort=True):
        if len(block) < 2:
            continue
        start, lesion_day = float(block.day.iloc[0]), new_lesion_days["CA031"].get(patient, np.nan)
        derived = time_to_progression(block, start, lesion_day)
        derived_sum_only = time_to_progression(block, start, lesion_day, new_lesions="censor")
        source_rows.append({"patient": patient,
                            "derived_source": derived["source"] if derived["event"] else "no progression",
                            "derived": bool(derived["event"]), "derived_sum_only": bool(derived_sum_only["event"]),
                            "investigator": patient in investigator_progression.index})
    progression_sources = pd.DataFrame(source_rows)
    progression_source_table = pd.crosstab(progression_sources.derived_source, progression_sources.investigator,
                                           rownames=["derived event source"], colnames=["investigator progression"],
                                           margins=True).reset_index()
    agreement = float((progression_sources.derived == progression_sources.investigator).mean())
    agreement_sum_only = float((progression_sources.derived_sum_only == progression_sources.investigator).mean())
    derived_only = progression_sources.loc[progression_sources.derived & ~progression_sources.investigator]
    record("clinical_ca031_agreement_derived", agreement)
    record("clinical_ca031_agreement_sum_only", agreement_sum_only)
    record("clinical_ca031_derived_only", len(derived_only))
    record("clinical_ca031_derived_only_new_lesion", int((derived_only.derived_source == "new lesion").sum()))
    print("14.5 ANALYSES ADDED AFTER BOTH PLANS")
    print("  (a) CA031: derived progression (from baseline) by event source, against the investigator")
    print(progression_source_table.to_string(index=False))
    print(f"  agreement (progressed or not): derived {100 * agreement:.1f}%, sum-only derivation "
          f"{100 * agreement_sum_only:.1f}%; progressions found only by the derivation {len(derived_only)}, of "
          f"which from a new lesion {int((derived_only.derived_source == 'new lesion').sum())}")

    # --- (b) landmark at the third scheduled assessment ------------------------
    THIRD_LANDMARK_DAYS = {study: 3.0 * spec["assessment_interval_days"] for study, spec in CLINICAL_STUDIES.items()}
    scored_third = pd.concat([score_patients(clinical_panel.loc[clinical_panel.study == study],
                                             new_lesion_days[study], THIRD_LANDMARK_DAYS[study], study)
                              for study in CLINICAL_STUDIES], ignore_index=True)
    cohort_third = scored_third.loc[scored_third.reached_landmark & ~scored_third.progressed_before_landmark
                                    & scored_third.has_features].reset_index(drop=True)
    print(f"\n  (b) landmark at the third scheduled assessment: {THIRD_LANDMARK_DAYS}")
    for study, part in cohort_third.groupby("study"):
        share_three = float((part.n_assessments >= 3).mean())
        slope_rho, _ = spearman(part.depth_at_landmark, part.slope_to_landmark)
        factors = variance_inflation(part, CLINICAL_TRAJECTORY)
        record(f"clinical_third_{study}_share_three_assessments", share_three)
        print(f"    {study:<9} patients {len(part)}, events {int(part.event.sum())}; >= 3 complete assessments "
              f"before the landmark {100 * share_three:.1f}%; ρ(depth, slope) {slope_rho:+.3f}; VIF "
              + ", ".join(f"{name} {value:.2f}" for name, value in factors.items()))
    third_analyses = [("pooled", cohort_third, True)] + [
        (study, cohort_third.loc[cohort_third.study == study], False) for study in CLINICAL_STUDIES]
    for tag, frame, stratified in third_analyses:
        if frame.empty or frame.event.sum() < 20 or (stratified and frame.study.nunique() < 2):
            print(f"    {tag}: too few events for the comparison")
            continue
        # An added analysis that cannot be fitted (too few patients in a fold
        # or stratum) prints its reason instead of stopping the run.
        try:
            result = compare_cox_models(frame, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, added_seed,
                                        stratified=stratified,
                                        label=f"added: landmark at the third assessment, {tag}", min_events=10)
        except (ValueError, np.linalg.LinAlgError) as error:
            print(f"    {tag}: comparison not computed ({error})")
            continue
        report_comparison(result)
        added_rows.append(result)
        record(f"clinical_third_landmark_{tag}_difference", result["difference"])
        record(f"clinical_third_landmark_{tag}_difference_low", result["low"])
        record(f"clinical_third_landmark_{tag}_difference_high", result["high"])
        record(f"clinical_third_landmark_{tag}_p", result["p_value"])

    # --- (c) CA031, investigator-assessed progression as the outcome ------------
    print("\n  (c) CA031 with the investigator-assessed progression day as the outcome")
    for landmark_label, cohort_frame, landmark_day in (
            ("second", clinical_cohort, LANDMARK_DAYS["CA031"]),
            ("third", cohort_third, THIRD_LANDMARK_DAYS["CA031"])):
        ca = cohort_frame.loc[cohort_frame.study == "CA031"].copy()
        investigator_day = ca.patient.map(investigator_progression)
        ca = ca.loc[~(investigator_day < landmark_day)].copy()       # progressed by the investigator before it
        investigator_day = ca.patient.map(investigator_progression)
        ca["investigator_event"] = investigator_day.notna().to_numpy()
        ca["investigator_duration"] = np.where(ca.investigator_event, investigator_day - landmark_day,
                                               ca.last_day - landmark_day)
        if ca.investigator_event.sum() < 20:
            print(f"    landmark at the {landmark_label} assessment: too few events")
            continue
        try:
            result = compare_cox_models(ca, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, added_seed,
                                        "investigator_duration", "investigator_event",
                                        label=f"added: CA031, investigator-assessed progression, "
                                              f"landmark at the {landmark_label} assessment", min_events=10)
        except (ValueError, np.linalg.LinAlgError) as error:
            print(f"    landmark at the {landmark_label} assessment: comparison not computed ({error})")
            continue
        report_comparison(result)
        added_rows.append(result)
        record(f"clinical_ca031_investigator_{landmark_label}_difference", result["difference"])
        record(f"clinical_ca031_investigator_{landmark_label}_difference_low", result["low"])
        record(f"clinical_ca031_investigator_{landmark_label}_difference_high", result["high"])
        record(f"clinical_ca031_investigator_{landmark_label}_p", result["p_value"])

    for result in added_rows:
        clinical_rows.append({"family": "added after both plans", "analysis": result["analysis"], "status": "added",
                              **{k: result[k] for k in ("n", "events", "c_depth", "c_trajectory", "difference",
                                                        "low", "high", "p_value", "bootstrap_draws")}})
    clinical_results = pd.DataFrame(clinical_rows)
    save_record(clinical_results, "clinical_results")
    save_record(progression_source_table, "clinical_progression_sources")
```

```python
# ==========================================================================
# 14.6 Shape against a nonlinear function of depth (added after review)
#     With only a baseline and one later complete assessment before the
#     landmark, the final-interval slope is log(1 − depth) / interval when
#     the sum fell, a nonlinear function of depth. A Cox model with depth and
#     slope can then gain concordance from that curvature alone, without any
#     information about the trajectory's shape. Two things are given here.
#     First, the share of patients whose slope equals log(1 − depth) /
#     interval exactly, per study and landmark. Second, the comparisons are
#     repeated against a comparator that already holds the curvature, depth
#     plus log(1 − depth) (1 − depth floored at 0.01, so that a complete
#     response stays finite). Challengers: the depth-plus-shape set of the
#     plans, and the comparator plus the two shape predictors; and the
#     comparator itself against depth alone. Outcomes: the primary and the
#     sum-only definition, at the planned (second) and the third scheduled
#     assessment, pooled with a baseline hazard per study; at the third
#     assessment the sum-only outcome is also compared with depth alone (the
#     primary outcome there is in 14.5). Same folds and bootstrap seed as the
#     pooled plan. Reported as added, outside every Holm family.
# ==========================================================================

clinical_nonlinear = pd.DataFrame()
if not CLINICAL_AVAILABLE:
    print("14.6 skipped: Project Data Sphere files not present")
else:
    REMAINING_FLOOR = 0.01                              # 1 − depth floored here before the log
    NONLINEAR_COMPARATOR = CLINICAL_COMPARATOR + ["log_remaining_burden"]
    SHAPE_PREDICTORS = [name for name in CLINICAL_TRAJECTORY if name not in CLINICAL_COMPARATOR]
    NONLINEAR_CHALLENGER = NONLINEAR_COMPARATOR + SHAPE_PREDICTORS
    # label -> (tag, comparator predictors, challenger predictors)
    NONLINEAR_COMPARISONS = {
        "depth + shape against depth + log(1 − depth)": ("shape_vs_nonlinear", NONLINEAR_COMPARATOR,
                                                        CLINICAL_TRAJECTORY),
        "depth + log(1 − depth) + shape against depth + log(1 − depth)": ("nonlinear_shape_vs_nonlinear",
                                                                          NONLINEAR_COMPARATOR,
                                                                          NONLINEAR_CHALLENGER),
        "depth + log(1 − depth) against depth": ("nonlinear_vs_depth", CLINICAL_COMPARATOR, NONLINEAR_COMPARATOR),
    }
    nonlinear_rows = []
    print("14.6 SHAPE AGAINST A NONLINEAR FUNCTION OF DEPTH (added after both plans)")
    print("  in each comparison below, 'C, depth' is the comparator's C and 'C, depth plus shape' the "
          "challenger's; the label names both")
    for landmark_label, cohort_frame in (("second", clinical_cohort), ("third", cohort_third)):
        frame = cohort_frame.assign(log_remaining_burden=np.log(np.maximum(1.0 - cohort_frame.depth_at_landmark,
                                                                           REMAINING_FLOOR)))
        # Patients whose final-interval slope is exactly log(1 − depth) / interval.
        implied_slope = (np.log(np.maximum(1.0 - frame.depth_at_landmark, 1e-12)) / frame.slope_window_days)
        slope_from_depth = np.isclose(frame.slope_to_landmark, implied_slope, rtol=1e-9, atol=1e-12)
        for study, part in frame.assign(slope_from_depth=slope_from_depth).groupby("study"):
            share = float(part.slope_from_depth.mean())
            record(f"clinical_{landmark_label}_{study}_share_slope_from_depth", share)
            factors = variance_inflation(part, NONLINEAR_CHALLENGER)
            print(f"  landmark at the {landmark_label} assessment, {study}: slope = log(1 − depth) / interval in "
                  f"{100 * share:.1f}% of {len(part)} patients; VIF " +
                  ", ".join(f"{name} {value:.2f}" for name, value in factors.items()))
        for outcome_label, outcome_tag, duration, event in (("primary", "primary", "duration", "event"),
                                                             ("sum only", "sum_only", "sld_only_duration",
                                                              "sld_only_event")):
            usable = frame.dropna(subset=[duration]).reset_index(drop=True)
            if usable[event].sum() < 20 or usable.study.nunique() < 2:
                print(f"    {outcome_label}, {landmark_label} assessment: too few events or one study only")
                continue
            comparisons = dict(NONLINEAR_COMPARISONS)
            if landmark_label == "third" and outcome_tag == "sum_only":
                comparisons["depth + shape against depth"] = ("shape_vs_depth", CLINICAL_COMPARATOR,
                                                              CLINICAL_TRAJECTORY)
            for comparison_label, (comparison_tag, comparator, challenger) in comparisons.items():
                # A comparison that cannot be fitted (too few patients in a
                # fold or stratum) prints its reason instead of stopping the run.
                try:
                    result = compare_cox_models(usable, comparator, challenger, pooled_plan["seed"], duration, event,
                                                stratified=True, min_events=10,
                                                label=f"added: {comparison_label}, {outcome_label}, landmark at the "
                                                      f"{landmark_label} assessment")
                except (ValueError, np.linalg.LinAlgError) as error:
                    print(f"    {comparison_label}, {outcome_label}, {landmark_label}: not computed ({error})")
                    continue
                report_comparison(result)
                nonlinear_rows.append({"landmark": landmark_label, "outcome": outcome_label,
                                       "comparison": comparison_label, "comparator": " + ".join(comparator),
                                       "challenger": " + ".join(challenger),
                                       **{key: result[key] for key in ("analysis", "n", "events", "c_depth",
                                                                       "c_trajectory", "difference", "low", "high",
                                                                       "p_value", "bootstrap_draws")}})
                stem = f"clinical_nonlinear_{landmark_label}_{outcome_tag}_{comparison_tag}"
                record(f"{stem}_difference", result["difference"])
                record(f"{stem}_difference_low", result["low"])
                record(f"{stem}_difference_high", result["high"])
                record(f"{stem}_p", result["p_value"])
    clinical_nonlinear = pd.DataFrame(nonlinear_rows)
    # Added to the clinical results table as added analyses (c_depth holds the
    # comparator's C and c_trajectory the challenger's; the comparator and
    # challenger columns name them).
    for row in nonlinear_rows:
        clinical_rows.append({"family": "added after both plans", "analysis": row["analysis"], "status": "added",
                              **{key: row[key] for key in ("n", "events", "c_depth", "c_trajectory", "difference",
                                                           "low", "high", "p_value", "bootstrap_draws",
                                                           "comparator", "challenger")}})
    clinical_results = pd.DataFrame(clinical_rows)
    save_record(clinical_results, "clinical_results")
    save_record(clinical_nonlinear, "clinical_nonlinear_depth")
```

## 15 · Results register, tables and run manifest

This section gathers what the run produced. Section 15.1 writes the main result tables to `outputs/tables/` and the results register (every value stored with `record()`, with the section that stored it) to `outputs/results_register.json` and `outputs/results_register.csv`. Optional inputs that were absent are marked as not computed. Section 15.2 checks the stored results for internal consistency: quantities derived from other stored quantities must agree with them, proportions and P values must lie between 0 and 1, and each endpoint decision must follow from its estimate, interval and margin. A failed check stops the run. Checks that are expected but not guaranteed (for example, that a percentile bootstrap interval contains its point estimate) are reported without stopping it. Section 15.3 writes `outputs/manifest.json`, which records the software versions, platform, seeds and key constants, the SHA-256 checksum and size of every input and output file, any file under `outputs/` that this run did not rewrite (figures, which the figures notebook writes later, are left out), the raw and LF-normalized checksums of each plan file, and the run time, and then prints a summary of the main results.

**Inputs:** every object stored above, `RESULTS`.
**Outputs:** `outputs/tables/*.csv`, `outputs/results_register.json`, `outputs/results_register.csv`, `outputs/manifest.json`.

```python
# ==========================================================================
# 15.1 Result tables and the results register
#     Each table is written to outputs/tables/ under a name that says what it
#     holds. Tables of optional inputs that were absent are empty and are
#     listed as not written. The register holds every value stored with
#     record(), the section that stored it, and whether each optional input
#     was present.
# ==========================================================================

set_section("15")
OPTIONAL_INPUTS = {
    "pdxnet_external_cohort": bool(PDXNET_AVAILABLE),
    "oberg_external_cohort": bool(OBERG_AVAILABLE),
    "hallmark_gene_sets": bool(HALLMARK_AVAILABLE),
    "clinical_trial_cohorts": bool(CLINICAL_AVAILABLE),
}
for name, available in OPTIONAL_INPUTS.items():
    record(f"computed_{name}", available)

RESULT_TABLES = {
    "predictor_sets": predictor_set_table,
    "depth_durability_sensitivity": depth_durability_sensitivity,
    "pair_threshold_sweep": pair_sweep,
    "parameter_recovery": parameter_recovery,
    "design_map": design_map,
    "mechanism_discrimination": mechanism_discrimination,
    "mechanism_discrimination_paired": paired_discrimination,
    "simulation_based_calibration": calibration.reset_index(),
    "simulation_based_calibration_converged": calibration_converged.reset_index(),
    "kill_rate_partition": kill_rate_partition,
    "forecasting_predictor_sets": predictor_set_performance,
    "forecasting_contrasts": predictor_contrasts,
    "forecasting_start_days_and_horizons": rolling_origins,
    "forecasting_held_out_tumor_types": tumor_type_holdout,
    "forecasting_censoring_scenarios": censoring_scenarios,
    "forecasting_untreated_and_treated": untreated_forecasting,
    "positive_control_braf": positive_control,
    "ranking_by_separation": separability,
    "baseline_slope_ranking": baseline_slope_selection,
    "gene_sets": gene_set_table,
    "molecular_blocks": molecular_blocks,
    "molecular_blocks_hallmark": hallmark_blocks,
    "hallmark_sets": hallmark_sets_table,
    "combination_summary": by_combination,
    "compound_families": compound_table,
    "mechanism_class_results": mechanism_results,
    "mechanism_class_mixed_models": mixed_models,
    "mechanism_class_by_family_pair": breakdown,
    "mechanism_class_leave_one_compound_out": leave_one_out,
    "exposure_separation_by_window": separation_by_window,
    "exposure_separation_growth": separation_growth,
    "exposure_verdicts_by_window": confirmation_summary,
    "exposure_thinning": thinning_verdicts,
    "external_between_animal": between_animal,
    "external_measurement_error": external_noise,
    "clinical_results": clinical_results,
    "clinical_coefficients": (clinical_coefficients if isinstance(clinical_coefficients, pd.DataFrame)
                              else pd.DataFrame()),
    # Tables added with the review fixes.
    "durability_definition_check": durability_check,
    "measurement_noise_estimates": noise_estimates_table,
    "depth_durability_resampling": rho_resampling,
    "discordance_noise_baseline": noise_baseline,
    "discordance_noise_baseline_by_points": noise_baseline_by_points,
    "discordance_null_sweep": null_sweep,
    "durability_attenuation": attenuation,
    "local_identifiability": local_identifiability,
    "parameter_recovery_conditional": conditional_recovery,
    "forecasting_endpoint_sensitivity": endpoint_sensitivity,
    "forecasting_day10_census": day10_census.reset_index(),
    "ranking_by_separation_all_schemes": separability_all,
    "selection_all_treated_arms": all_arm_selection,
    "molecular_blocks_within_treatment": molecular_blocks_within_treatment,
    "molecular_blocks_hallmark_within_treatment": molecular_blocks_hallmark_within,
    "combination_summary_comparison": combination_comparison,
    "mechanism_class_dependence": dependence_table,
    "mechanism_class_evidence": mechanism_evidence,
    "exposure_verdicts_clustered": exposure_clustered,
    "exposure_weak_kill": weak_kill_summary,
    "external_variance_ratio": external_ratio,
    "clinical_progression_sources": progression_source_table,
    "comparator_contrasts": comparator_contrasts,
    "selection_influence": selection_influence,
    "selection_stability_summary": selection_stability_summary,
    "selection_with_censored_arms": selection_with_censored,
    "fold_partition_summary": fold_partition_summary,
    "selection_against_replicate": selection_against_replicate,
    # Tables added with the second round of review fixes.
    "discordance_responders": discordance_responders,
    "floor_plateau_sensitivity": floor_plateau_sensitivity,
    "interval_reconciliation": interval_reconciliation,
    "learner_contrasts": learner_contrasts,
    "model_free_ranking_levels": model_free_levels,
    "model_free_selection": model_free_selection,
    "selection_change_from_baseline": selection_change_from_baseline,
    "selection_by_horizon": selection_by_horizon,
    "combination_v600": combination_v600,
    "single_agent_v600": single_agent_v600,
    "encorafenib_v600": encorafenib_v600,
    "discordance_between_animal": between_discordance,
    "discordance_between_animal_sweep": between_discordance_sweep,
    "clinical_nonlinear_depth": clinical_nonlinear,
}
written_tables, empty_tables = [], []
for name, table in RESULT_TABLES.items():
    if isinstance(table, pd.DataFrame) and len(table):
        table.to_csv(TABLE_DIR / f"{name}.csv", index=False)
        written_tables.append(name)
    else:
        empty_tables.append(name)
print("15.1 RESULT TABLES")
print(f"  written to {relative(TABLE_DIR)}: {len(written_tables)}")
for name in written_tables:
    print(f"    {name}.csv   {len(RESULT_TABLES[name]):,} rows")
if empty_tables:
    print(f"  not written (input absent or nothing to report): {', '.join(empty_tables)}")

# --- the register --------------------------------------------------------------
register = pd.DataFrame({"key": list(RESULTS),
                         "section": [RESULT_SECTION.get(key, "") for key in RESULTS],
                         "value": [json_ready(RESULTS[key]) for key in RESULTS]})
REGISTER_JSON = OUTPUT_DIR / "results_register.json"
REGISTER_CSV = OUTPUT_DIR / "results_register.csv"
REGISTER_JSON.write_text(json.dumps(
    {"run_started_utc": RUN_STARTED_UTC,
     "optional_inputs": OPTIONAL_INPUTS,
     "values": {row.key: {"section": row.section, "value": row.value}
                for row in register.itertuples(index=False)}},
    indent=1, ensure_ascii=False), encoding="utf-8")
register.to_csv(REGISTER_CSV, index=False)
not_computed = [name for name, available in OPTIONAL_INPUTS.items() if not available]
print(f"\n  results register: {len(register):,} values from {register.section.nunique()} sections")
print(f"    {relative(REGISTER_JSON)} and {relative(REGISTER_CSV)}")
print(f"  not computed (optional input absent): {', '.join(not_computed) if not_computed else 'none'}")
print("\n  values per section")
print(register.groupby("section", sort=False).size().rename("values").to_string())
```

```python
# ==========================================================================
# 15.2 Internal consistency of the stored results
#     Required checks must hold: a quantity derived from other stored
#     quantities must equal what they give, proportions and P values must lie
#     in [0, 1], and each endpoint decision must follow from its estimate,
#     interval and margin. A failed required check stops the run. Expected
#     checks (a percentile interval containing its estimate) are reported
#     without stopping it, because a percentile interval of a ratio can, in
#     principle, exclude its point estimate. A check whose inputs were not
#     computed is marked not applicable.
# ==========================================================================

# (description, keys it needs, required?, condition on the register r)
DERIVED_CHECKS = [
    ("forecast reduction = (depth RMSE − trajectory RMSE) / depth RMSE",
     ["forecast_reduction", "depth_rmse", "trajectory_rmse"], True,
     lambda r: np.isclose(r["forecast_reduction"], (r["depth_rmse"] - r["trajectory_rmse"]) / r["depth_rmse"])),
    ("regret reduction = (depth regret − trajectory regret) / depth regret",
     ["regret_reduction", "depth_regret", "trajectory_regret"], True,
     lambda r: np.isclose(r["regret_reduction"], (r["depth_regret"] - r["trajectory_regret"]) / r["depth_regret"])),
    ("regret fold changes = exp(regret)",
     ["depth_regret_fold", "depth_regret", "trajectory_regret_fold", "trajectory_regret"], True,
     lambda r: np.isclose(r["depth_regret_fold"], np.exp(r["depth_regret"]))
     and np.isclose(r["trajectory_regret_fold"], np.exp(r["trajectory_regret"]))),
    ("share of different choices = count / PDX models",
     ["share_different_choice", "n_different_choice", "regret_n_models"], True,
     lambda r: np.isclose(r["share_different_choice"], r["n_different_choice"] / r["regret_n_models"])),
    ("discordant share = discordant pairs / depth-matched pairs",
     ["discordant_share", "n_discordant_pairs", "n_depth_matched_pairs"], True,
     lambda r: np.isclose(r["discordant_share"], r["n_discordant_pairs"] / r["n_depth_matched_pairs"])),
    ("shared rank variance = ρ²", ["depth_durability_shared_variance", "depth_durability_rho"], True,
     lambda r: np.isclose(r["depth_durability_shared_variance"], r["depth_durability_rho"] ** 2)),
    ("combination shared rank variance = ρ²", ["gain_shared_variance", "gain_rho"], True,
     lambda r: np.isclose(r["gain_shared_variance"], r["gain_rho"] ** 2)),
    ("combination categories sum to one",
     ["share_superior_on_both", "share_superior_in_depth_only", "share_superior_in_durability_only",
      "share_superior_on_neither"], True,
     lambda r: np.isclose(r["share_superior_on_both"] + r["share_superior_in_depth_only"]
                          + r["share_superior_in_durability_only"] + r["share_superior_on_neither"], 1.0)),
    ("28-day fold change of the mechanism-class effect = exp(28 × effect)",
     ["mechanism_durability_fold_28", "mechanism_durability"], True,
     lambda r: np.isclose(r["mechanism_durability_fold_28"], np.exp(28 * r["mechanism_durability"]))),
    ("pair-rate gap fold change = exp(0.02 × 28)", ["fold_change_28_days"], True,
     lambda r: np.isclose(r["fold_change_28_days"], np.exp(GROWTH_RATE_GAP * PAIR_HORIZON_DAYS))),
    ("forecasting arms are post-nadir arms", ["forecast_n_arms", "n_post_nadir_arms"], True,
     lambda r: r["forecast_n_arms"] <= r["n_post_nadir_arms"]),
    ("forecasting endpoint decision follows its estimate, interval and margin",
     ["forecast_endpoint_met", "forecast_reduction", "forecast_reduction_low"], True,
     lambda r: r["forecast_endpoint_met"] == int(r["forecast_reduction"] >= FORECAST_MARGIN
                                                 and r["forecast_reduction_low"] > 0)),
    ("regret endpoint decision follows its estimate, interval and margin",
     ["regret_endpoint_met", "regret_reduction", "regret_reduction_low"], True,
     lambda r: r["regret_endpoint_met"] == int(r["regret_reduction"] >= REGRET_MARGIN
                                               and r["regret_reduction_low"] > 0)),
    ("baseline-plus-slope ranking accuracy plus the difference gives the trajectory accuracy",
     ["ranking_accuracy_baseline_slope", "ranking_trajectory_minus_baseline_slope", "ranking_accuracy_trajectory"],
     True,
     lambda r: np.isclose(r["ranking_accuracy_baseline_slope"] + r["ranking_trajectory_minus_baseline_slope"],
                          r["ranking_accuracy_trajectory"])),
    ("converged calibration replicates do not exceed all replicates",
     ["calibration_converged_replicates", "calibration_replicates"], True,
     lambda r: r["calibration_converged_replicates"] <= r["calibration_replicates"]),
    # Checks added with the review fixes.
    ("forecasting lower-bound decision follows its interval and margin",
     ["forecast_lower_bound_clears_margin", "forecast_reduction_low"], True,
     lambda r: r["forecast_lower_bound_clears_margin"] == int(r["forecast_reduction_low"] >= FORECAST_MARGIN)),
    ("regret lower-bound decision follows its interval and margin",
     ["regret_lower_bound_clears_margin", "regret_reduction_low"], True,
     lambda r: r["regret_lower_bound_clears_margin"] == int(r["regret_reduction_low"] >= REGRET_MARGIN)),
    ("7.4 reproduces the primary forecasting result under the rules of Section 1.1",
     [f"forecast_reduction_{WINDOW_TAG[FORECAST_WINDOW_RESPONSE]}_{FORECAST_OUTCOME_ZEROS}", "forecast_reduction"], True,
     lambda r: np.isclose(r[f"forecast_reduction_{WINDOW_TAG[FORECAST_WINDOW_RESPONSE]}_{FORECAST_OUTCOME_ZEROS}"],
                          r["forecast_reduction"])),
    ("7.4 reproduces the primary regret result under the rules of Section 1.1",
     [f"regret_reduction_{WINDOW_TAG[FORECAST_WINDOW_RESPONSE]}_{FORECAST_OUTCOME_ZEROS}", "regret_reduction"], True,
     lambda r: np.isclose(r[f"regret_reduction_{WINDOW_TAG[FORECAST_WINDOW_RESPONSE]}_{FORECAST_OUTCOME_ZEROS}"],
                          r["regret_reduction"])),
    ("the three-way weighted compound effect reproduces 11.3",
     ["compound_effect_threeway", "compound_effect"], True,
     lambda r: np.isclose(r["compound_effect_threeway"], r["compound_effect"])),
    ("8.7 reproduces the regret reduction of Section 7 in the primary population",
     ["comparator_primary_trajectory_vs_depth_regret_reduction", "regret_reduction"], True,
     lambda r: np.isclose(r["comparator_primary_trajectory_vs_depth_regret_reduction"], r["regret_reduction"])),
    ("8.7 reproduces the forecasting result of Section 7 in the primary population",
     ["comparator_primary_trajectory_vs_depth_rmse_reduction", "forecast_reduction"], True,
     lambda r: np.isclose(r["comparator_primary_trajectory_vs_depth_rmse_reduction"], r["forecast_reduction"])),
    ("no predictor changes when readings after the origin change",
     ["leakage_audit_max_difference"], True, lambda r: r["leakage_audit_max_difference"] == 0),
    ("treated-arm durability recomputed from the canonical table matches pdxe_utils",
     ["n_treated_arms_durability_differs_from_utils"], False,
     lambda r: r["n_treated_arms_durability_differs_from_utils"] == 0),
    # Checks added with the second round of review fixes.
    ("the PDX-model regret tallies add up to the models of 7.3",
     ["regret_models_trajectory_lower", "regret_models_depth_lower", "regret_models_tied", "regret_n_models"], True,
     lambda r: r["regret_models_trajectory_lower"] + r["regret_models_depth_lower"] + r["regret_models_tied"]
     == r["regret_n_models"]),
    ("between-animal variation can only raise the expected discordance (13.5 >= 4.5)",
     ["discordance_between_animal_expected", "discordance_null_expected_median"], True,
     lambda r: r["discordance_between_animal_expected"] >= r["discordance_null_expected_median"] - 1e-12),
    ("one interval per estimate: trajectory against expanded depth in 8.1 and 8.7 (8.7b)",
     ["trajectory_over_expanded_low", "trajectory_over_expanded_high",
      "comparator_primary_trajectory_vs_expanded_depth_rmse_reduction_low",
      "comparator_primary_trajectory_vs_expanded_depth_rmse_reduction_high"], True,
     lambda r: np.isclose(r["trajectory_over_expanded_low"],
                          r["comparator_primary_trajectory_vs_expanded_depth_rmse_reduction_low"])
     and np.isclose(r["trajectory_over_expanded_high"],
                    r["comparator_primary_trajectory_vs_expanded_depth_rmse_reduction_high"])),
    ("one interval per estimate: regret, trajectory against baseline plus slope, in 8.5 and 8.7 (8.7b)",
     ["trajectory_vs_baseline_slope_regret_reduction_low",
      "comparator_primary_trajectory_vs_baseline_slope_regret_reduction_low"], True,
     lambda r: np.isclose(r["trajectory_vs_baseline_slope_regret_reduction_low"],
                          r["comparator_primary_trajectory_vs_baseline_slope_regret_reduction_low"])),
    ("one interval per estimate: regret in the practical pool, 8.6 and 8.7 (8.7b)",
     [f"all_arms_regret_reduction_{WINDOW_TAG[FORECAST_WINDOW_RESPONSE]}_floor_low",
      "comparator_practical_trajectory_vs_depth_regret_reduction_low"], True,
     lambda r: np.isclose(r[f"all_arms_regret_reduction_{WINDOW_TAG[FORECAST_WINDOW_RESPONSE]}_floor_low"],
                          r["comparator_practical_trajectory_vs_depth_regret_reduction_low"])),
]


check_rows = []
for description, keys, required, condition in DERIVED_CHECKS:
    if not all(key in RESULTS and usable_value(RESULTS[key]) for key in keys):
        status = "not applicable"
    else:
        status = "holds" if bool(condition(RESULTS)) else "fails"
    check_rows.append({"check": description, "required": required, "status": status})

# P values and proportions must lie in [0, 1].
P_VALUE_PATTERN = re.compile(r"(^|_)(p|q)(_|$)")
PROPORTION_PATTERN = re.compile(r"share|accuracy|chose_tolerant|^exposure_decisive_|^exposure_accuracy_")
for key, value in RESULTS.items():
    if not isinstance(value, (int, float)) or isinstance(value, bool) or not usable_value(value):
        continue
    if P_VALUE_PATTERN.search(key) or PROPORTION_PATTERN.search(key):
        kind = "P value" if P_VALUE_PATTERN.search(key) else "proportion"
        check_rows.append({"check": f"{kind} in [0, 1]: {key}", "required": True,
                           "status": "holds" if 0.0 <= value <= 1.0 else "fails"})

# No stored number is infinite.
infinite = [key for key, value in RESULTS.items()
            if isinstance(value, float) and np.isinf(value)]
check_rows.append({"check": "no stored value is infinite", "required": True,
                   "status": "holds" if not infinite else f"fails: {infinite}"})

# Each estimate lies inside its own 95% interval (expected, not required).
for key, value in RESULTS.items():
    low, high = RESULTS.get(f"{key}_low"), RESULTS.get(f"{key}_high")
    if all(isinstance(v, (int, float)) and not isinstance(v, bool) and usable_value(v) for v in (value, low, high)):
        check_rows.append({"check": f"estimate inside its interval: {key}", "required": False,
                           "status": "holds" if low <= value <= high else "fails"})

consistency_checks = pd.DataFrame(check_rows)
required_failures = consistency_checks.loc[consistency_checks.required
                                           & consistency_checks.status.str.startswith("fails")]
expected_failures = consistency_checks.loc[~consistency_checks.required
                                           & consistency_checks.status.str.startswith("fails")]
save_record(consistency_checks, "consistency_checks")
print("15.2 INTERNAL CONSISTENCY OF THE STORED RESULTS")
print(consistency_checks.groupby(["required", "status"]).size().rename("checks").to_string())
print("\n  derived quantities and endpoint decisions")
print(consistency_checks.loc[consistency_checks.check.isin([c[0] for c in DERIVED_CHECKS])]
      .to_string(index=False))
if len(expected_failures):
    print("\n  expected checks that did not hold (reported, not stopping the run)")
    print(expected_failures.to_string(index=False))
assert required_failures.empty, f"required consistency checks failed:\n{required_failures.to_string(index=False)}"
print(f"\n  all {int(consistency_checks.required.sum())} required checks hold")
```

```python
# ==========================================================================
# 15.3 Run manifest and summary of the main results
#     The manifest lists the software, platform, seeds and key constants, the
#     size and SHA-256 checksum of every input file read and every output
#     file written (Parquet caches under outputs/interim/ are left out), and
#     the run time. It is written last, so it does not list itself.
# ==========================================================================

MANIFEST_PATH = OUTPUT_DIR / "manifest.json"
input_paths = [WORKBOOK_PATH, PROJECT_DIR / "pdxe_utils.py"]
input_paths += [PRESPEC_DIR / file_name for file_name, _ in PRESPECIFIED_PLANS.values()
                if (PRESPEC_DIR / file_name).exists()]
input_paths += list(pdxnet_files) + ([OBERG_PATH] if OBERG_AVAILABLE else []) + list(hallmark_files)
input_paths += [PDS_DIR / study / name for study, spec in CLINICAL_STUDIES.items() for name in spec["files"]
                if (PDS_DIR / study / name).exists()]
# Figures (outputs/figures/) are written by the figures notebook after this
# run, so they are always older than it: they are left out here and listed
# in that notebook's own manifest.
output_paths = sorted(path for path in OUTPUT_DIR.rglob("*")
                      if path.is_file() and INTERIM_DIR not in path.parents and FIGURE_DIR not in path.parents
                      and path != MANIFEST_PATH)
# Files under outputs/ that this run did not rewrite: records of an analysis
# skipped this time, or of an earlier version of the notebook. The figures
# notebook must not read them; they are listed here and in the manifest.
stale_outputs = [relative(path) for path in output_paths if path.stat().st_mtime < RUN_STARTED_EPOCH]
# Plan files: "inputs" below lists the SHA-256 of their raw bytes, while each
# plan's recorded checksum (Section 1.1) is of its text with line endings
# normalized to LF (plan_sha256), so that a checkout that converts line
# endings still verifies. Both are recorded, so the two can be told apart. A
# .gitattributes line "prespecification/*.json text eol=lf" keeps the raw and
# the normalized checksums equal in the repository.
plan_checksums = {file_name: {"raw_sha256": sha256_of_file(PRESPEC_DIR / file_name),
                              "lf_normalized_sha256": plan_sha256(PRESPEC_DIR / file_name),
                              "recorded_prefix": recorded_prefix}
                  for file_name, recorded_prefix in PRESPECIFIED_PLANS.values()
                  if (PRESPEC_DIR / file_name).exists()}

elapsed_seconds = time.perf_counter() - RUN_STARTED
manifest = {
    "run_started_utc": RUN_STARTED_UTC,
    "run_finished_utc": datetime.now(timezone.utc).isoformat(timespec="seconds"),
    "elapsed_seconds": round(elapsed_seconds, 1),
    "software": SOFTWARE_VERSIONS,
    "platform": PLATFORM,
    "pdxe_utils_sha256": UTILS_SHA256,
    "plan_checksums": plan_checksums,
    "seeds": SEEDS,
    "constants": {
        "volume_floor_mm3": VOLUME_FLOOR, "best_average_response_from_day": MIN_DAY_FOR_BEST_AVERAGE_RESPONSE,
        "forecast_origin_day": FORECAST_ORIGIN_DAY, "forecast_horizon_day": FORECAST_HORIZON_DAY,
        "forecast_outcome_zeros": FORECAST_OUTCOME_ZEROS, "forecast_margin": FORECAST_MARGIN,
        "forecast_window_response": FORECAST_WINDOW_RESPONSE,
        "window_response_tolerance_days": WINDOW_RESPONSE_TOLERANCE_DAYS,
        "durability_includes_nadir": DURABILITY_INCLUDES_NADIR,
        "min_points_after_nadir": MIN_POINTS_AFTER_NADIR,
        "regret_margin": REGRET_MARGIN, "folds": N_FOLDS, "bootstrap_replicates": N_BOOTSTRAP,
        "control_bootstrap_replicates": N_BOOTSTRAP_CONTROLS, "measurement_noise": MEASUREMENT_NOISE,
        "kill_rate_range": KILL_RATE_RANGE, "termination_volume_mm3": TERMINATION_VOLUME,
        "simulation_noise_sd": SIMULATION_NOISE_SD,
    },
    "optional_inputs": OPTIONAL_INPUTS,
    "stale_outputs": stale_outputs,
    "results_register": {"values": len(RESULTS), "json": REGISTER_JSON, "csv": REGISTER_CSV},
    "consistency_checks": {"required": int(consistency_checks.required.sum()),
                           "required_failed": len(required_failures),
                           "expected": int((~consistency_checks.required).sum()),
                           "expected_not_holding": len(expected_failures)},
    "inputs": [file_record(path) for path in input_paths],
    "outputs": [file_record(path) for path in output_paths],
}
MANIFEST_PATH.write_text(json.dumps(json_ready(manifest), indent=1, ensure_ascii=False), encoding="utf-8")

print("15.3 RUN MANIFEST")
print(f"  {relative(MANIFEST_PATH)}")
print(f"  inputs {len(manifest['inputs'])}, outputs {len(manifest['outputs'])}; "
      f"run time {elapsed_seconds / 60:.1f} minutes")
plan_paths = {relative(PRESPEC_DIR / file_name): file_name for file_name in plan_checksums}
for entry in manifest["inputs"]:
    note = ""
    if entry["path"] in plan_paths:
        # The plan's recorded checksum is of its LF-normalized text (Section 1.1).
        plan = plan_checksums[plan_paths[entry["path"]]]
        note = f"   LF-normalized {plan['lf_normalized_sha256'][:16]} (recorded {plan['recorded_prefix']})"
    print(f"    input   {entry['path']:<70} {entry['sha256'][:16]}{note}")
if stale_outputs:
    print(f"\n  WARNING: {len(stale_outputs)} file(s) under outputs/ were not rewritten by this run "
          "(listed in the manifest as stale_outputs); delete them or make sure no figure reads them:")
    for path in stale_outputs[:20]:
        print(f"    {path}")
else:
    print("  every file under outputs/ was written by this run")

# --- summary of the main results ----------------------------------------------
# Each line shows a stored value and, where the register holds its interval,
# the 95% interval beside it. Values of optional sections that were not
# computed are shown as such.
SUMMARY = [
    ("4", "post-nadir arms", "n_post_nadir_arms", "int"),
    ("4", "depth against durability, Spearman ρ (interval: PDX models resampled)", "depth_durability_rho", "f3"),
    ("4", "same, P resampling PDX models / PDX models and treatments",
     ("depth_durability_p_cluster", "depth_durability_p_cluster_twoway"), "pb"),
    ("4", "depth-matched pairs discordant in durability", "discordant_share", "pct"),
    ("4", "discordant share expected from noise alone (σ from 3.6 / treated)",
     ("discordance_null_expected_median", "discordance_null_expected_treated"), "pct"),
    ("4", "discordance in excess of noise (σ from 3.6)", "discordance_excess_median", "pct"),
    ("4", "reliability of durability (σ from 3.6 / treated)",
     ("durability_reliability_median", "durability_reliability_treated"), "f3"),
    ("4", "discordant share, both arms at or below -30% (responders)", "discordant_share_responders_30", "pct"),
    ("4", "same, expected from measurement noise alone (σ from 3.6)", "discordance_null_expected_responders_30",
     "pct"),
    ("4", "post-nadir arms held at the floor for >= 2 readings", "n_floor_plateau_post_nadir", "int"),
    ("4", "depth against durability, floor plateaus left out", "rho_without_floor_plateaus", "f3"),
    ("4", "same, durability of those arms from the end of the plateau", "rho_floor_plateau_from_relapse", "f3"),
    ("5", "mechanism discrimination, mean accuracy (unpaired / paired)",
     ("discrimination_mean_accuracy", "paired_discrimination_mean_accuracy"), "f3"),
    ("5", "simulation-based calibration, converged replicates", "calibration_converged_replicates", "int"),
    ("6", "measurement noise (log units)", "measurement_noise", "f4"),
    ("6", "REML share of log kill-rate variance, treatment / PDX model",
     ("reml_share_treatment", "reml_share_model"), "f3"),
    ("7", "forecasting, relative RMSE reduction", "forecast_reduction", "pct"),
    ("7", "treatment selection, relative regret reduction", "regret_reduction", "pct"),
    ("7", "pairwise ranking accuracy, depth / trajectory",
     ("ranking_accuracy_depth", "ranking_accuracy_trajectory"), "f3"),
    ("7", "PDX models where the two sets choose differently", "share_different_choice", "pct"),
    ("7", "PDX models with lower regret: trajectory / depth / equal",
     ("regret_models_trajectory_lower", "regret_models_depth_lower", "regret_models_tied"), "int"),
    ("7", "PDX models in which each set picks the best arm: depth / trajectory",
     ("best_arm_share_depth", "best_arm_share_trajectory"), "pct"),
    ("7", "net regret reduction carried by the 10 largest model differences",
     "regret_top10_fraction_of_net_reduction", "pct"),
    ("7", "lower bound clears its margin: forecasting / regret (1 = yes)",
     ("forecast_lower_bound_clears_margin", "regret_lower_bound_clears_margin"), "int"),
    ("7", "one-sided P against the margin: forecasting / regret", ("forecast_p_vs_margin", "regret_p_vs_margin"), "pb"),
    ("7", "forecasting, last-reading window rule, floor kept", "forecast_reduction_last_floor", "pct"),
    ("7", "treatment selection, last-reading window rule, floor kept", "regret_reduction_last_floor", "pct"),
    ("8", "trajectory over expanded depth", "trajectory_over_expanded", "pct"),
    ("8", "trajectory over depth plus current volume", "trajectory_over_current_volume", "pct"),
    ("8", "trajectory over depth plus permuted shape", "trajectory_over_permuted_shape", "pct"),
    ("8", "baseline plus slope over depth", "baseline_slope_over_depth", "pct"),
    ("8", "trajectory over baseline plus slope", "trajectory_over_baseline_slope", "pct"),
    ("8", "noise-padded depth over depth", "noise_over_depth", "pct"),
    ("8", "without the post-nadir restriction", "forecast_all_reduction", "pct"),
    ("8", "untreated arms, day 14 to 21", "untreated_forecast_14_21", "pct"),
    ("8", "treated arms, day 14 to 21", "treated_forecast_14_21", "pct"),
    ("8", "censoring scenarios, primary zero rule: growth stopped", "scenario_stopped", "pct"),
    ("8", "censoring scenarios, primary zero rule: growth continued", "scenario_continued", "pct"),
    ("8", "ranking accuracy, day-24 gap > 8 √2σ (depth / trajectory)",
     ("top_stratum_depth_accuracy", "top_stratum_trajectory_accuracy"), "f3"),
    ("8", "same, units of the first run, gap > 8 σ (depth / trajectory)",
     ("top_stratum_depth_accuracy_first_run", "top_stratum_trajectory_accuracy_first_run"), "f3"),
    ("8", "same, day-10 gap > 8 √2σ, known at selection (depth / trajectory)",
     ("top_stratum_depth_accuracy_day10", "top_stratum_trajectory_accuracy_day10"), "f3"),
    ("8", "ranking accuracy, trajectory minus baseline plus slope", "ranking_trajectory_minus_baseline_slope", "f3"),
    ("8", "same, day-24 gap > 8 √2σ", "top_stratum_trajectory_minus_baseline_slope", "f3"),
    ("8", "trajectory over depth plus current volume: regret reduction (primary)",
     "comparator_primary_trajectory_vs_current_volume_regret_reduction", "pct"),
    ("8", "trajectory over baseline plus slope: regret reduction (primary)",
     "comparator_primary_trajectory_vs_baseline_slope_regret_reduction", "pct"),
    ("8", "trajectory over depth: regret reduction (practical pool)",
     "comparator_practical_trajectory_vs_depth_regret_reduction", "pct"),
    ("8", "trajectory over baseline plus slope: regret reduction (practical pool)",
     "comparator_practical_trajectory_vs_baseline_slope_regret_reduction", "pct"),
    ("8", "trajectory over least-squares baseline plus slope: RMSE reduction (practical pool)",
     "comparator_practical_trajectory_vs_ls_baseline_slope_rmse_reduction", "pct"),
    ("8", "stable choice of arm in >= 80% of refits: depth / trajectory (primary)",
     ("stability_primary_depth_share_stable", "stability_primary_trajectory_share_stable"), "pct"),
    ("8", "regret reduction over refits, median (primary)", "refit_primary_regret_reduction", "pct"),
    ("8", "same, 2.5th / 97.5th percentiles over refits (not a confidence interval)",
     ("refit_primary_regret_reduction_refits_low", "refit_primary_regret_reduction_refits_high"), "pct"),
    ("8", "censored arms added, growth stopped: regret reduction", "censored_selection_stopped_regret_reduction", "pct"),
    ("8", "regret reduction over 20 random fold partitions: lowest / median / highest (primary)",
     ("partition_primary_regret_reduction_min", "partition_primary_regret_reduction_median",
      "partition_primary_regret_reduction_max"), "pct"),
    ("8", "random fold partitions whose regret reduction reaches the margin (primary / practical pool)",
     ("partition_primary_regret_reduction_share_at_margin",
      "partition_practical_regret_reduction_share_at_margin"), "pct"),
    ("8", "RMSE reduction over 20 random fold partitions: lowest / median / highest (primary)",
     ("partition_primary_rmse_reduction_min", "partition_primary_rmse_reduction_median",
      "partition_primary_rmse_reduction_max"), "pct"),
    ("8", "selection among all treated arms, regret reduction (floor dropped)",
     "all_arms_regret_reduction_exact_drop", "pct"),
    ("8", "selection among all treated arms, regret reduction (floor kept)",
     "all_arms_regret_reduction_exact_floor", "pct"),
    ("8", "trajectory (boosting) over depth (least squares): RMSE reduction (primary)",
     "learner_primary_trajectory_gb_vs_depth_ls_rmse_reduction", "pct"),
    ("8", "same: regret reduction (primary)", "learner_primary_trajectory_gb_vs_depth_ls_regret_reduction", "pct"),
    ("8", "same: ranking accuracy difference (primary)",
     "learner_primary_trajectory_gb_vs_depth_ls_ranking_difference", "f3"),
    ("8", "same: regret reduction (practical pool)",
     "learner_practical_trajectory_gb_vs_depth_ls_regret_reduction", "pct"),
    ("8", "no model, raw window slope over raw window response: regret reduction (primary)",
     "model_free_primary_slope_vs_response_regret_reduction", "pct"),
    ("8", "no model, raw window slope over raw window response: regret reduction (practical pool)",
     "model_free_practical_slope_vs_response_regret_reduction", "pct"),
    ("8", "change from baseline, trajectory over depth: regret reduction (primary)",
     "change_primary_trajectory_vs_depth_regret_reduction", "pct"),
    ("8", "selection, start day 21 to day 42: regret reduction (post-nadir arms)",
     "horizon_primary_21_42_regret_reduction", "pct"),
    ("8", "selection, start day 21 to day 42: regret reduction (practical pool)",
     "horizon_practical_21_42_regret_reduction", "pct"),
    ("9", "dissociation statistic D, curated signatures", "dissociation", "f3"),
    ("9", "dissociation statistic D, hallmark signatures", "dissociation_hallmark", "f3"),
    ("9", "dissociation statistic D, within treatment", "dissociation_within_treatment", "f3"),
    ("10", "combination evaluations superior on both endpoints", "share_superior_on_both", "pct"),
    ("10", "depth gain against durability gain, Spearman ρ", "gain_rho", "f3"),
    ("10", "same, rule of the analysis first run", "gain_rho_first_run", "f3"),
    ("10", "combinations significantly deeper / shallower (q < 0.05)",
     ("combinations_depth_significant", "combinations_shallower"), "int"),
    ("10", "combinations regrowing significantly slower / faster (q < 0.10, any depth)",
     ("combinations_slower_regrowth_any_depth", "combinations_faster_regrowth_any_depth"), "int"),
    ("10", "encorafenib + binimetinib, BRAF V600 models: evaluations",
     "v600_encorafenib___binimetinib_evaluations", "int"),
    ("10", "same: median depth gain (points)", "v600_encorafenib___binimetinib_median_depth_gain", "f1"),
    ("10", "same: median durability gain (per day)", "v600_encorafenib___binimetinib_median_durability_gain",
     "f4"),
    ("10", "cetuximab + encorafenib, BRAF V600 models: evaluations",
     "v600_cetuximab___encorafenib_evaluations", "int"),
    ("10", "same: median depth gain (points)", "v600_cetuximab___encorafenib_median_depth_gain", "f1"),
    ("11", "cytostatic minus comparator durability (per day)", "mechanism_durability", "f4"),
    ("11", "same, compound as the unit (per day)", "compound_effect", "f4"),
    ("11", "mixed-effects model, crossed: intercept / P", ("mixed_model_intercept", "mixed_model_p"), "f4"),
    ("11", "family labels permuted (the planned class-level test), one-sided P",
     "label_permutation_p_one_sided", "p"),
    ("11", "compound effect, three-way bootstrap (per day)", "compound_effect_threeway", "f4"),
    ("11", "sign-flip test over all sign patterns, one-sided P", "sign_flip_p_one_sided_exact", "p"),
    ("11", "encorafenib effect in BRAF V600 models (per day)", "encorafenib_effect_braf_v600", "f4"),
    ("11", "encorafenib effect in BRAF wild-type models (per day)", "encorafenib_effect_braf_wild_type", "f4"),
    ("12", "structural separation over 28 days (log units)", "exposure_separation_28", "f4"),
    ("12", "separation fold change, day 28 to 84 (half-life 0.25 d)", "exposure_fold_28_to_84_hl025", "f2"),
    ("12", "84-day arms with a pinned fit (half-life 0.25 d)", "exposure_pinned_share_84_hl025", "pct"),
    ("12", "weaker kill, 84 days: decisive rate, range over half-lives",
     ("exposure_weak_kill_decisive_84_min", "exposure_weak_kill_decisive_84_max"), "f3"),
    ("13", "PDXNet between-animal SD / between-model SD (as first run)", "pdxnet_between_ratio_control", "f3"),
    ("13", "same, source-study SD corrected for slope error", "pdxnet_ratio_corrected_control", "f3"),
    ("13", "same, against the between-model component", "pdxnet_ratio_between_model_control", "f3"),
    ("13", "regret reduction judged against a second animal (measurement noise / with between-animal growth)",
     ("replicate_primary_measurement_regret_reduction", "replicate_primary_between_animal_regret_reduction"), "pct"),
    ("13", "best observed arm stays best with a second animal (measurement noise / with between-animal growth)",
     ("replicate_primary_measurement_best_arm_kept", "replicate_primary_between_animal_best_arm_kept"), "pct"),
    ("13", "same, with between-animal growth: 2.5th / 97.5th percentiles of the regret reduction over draws",
     ("replicate_primary_between_animal_regret_reduction_draws_low",
      "replicate_primary_between_animal_regret_reduction_draws_high"), "pct"),
    ("13", "discordant share expected with between-animal variation (σ from 3.6; interval from τ's interval)",
     "discordance_between_animal_expected", "pct"),
    ("13", "observed discordant share minus that expectation (joint bootstrap)",
     "discordance_between_animal_excess", "pct"),
    ("13", "Oberg untreated measurement error, volume scale (1.5 × log area)",
     "noise_interp_oberg_untreated_volume_scale", "f4"),
    ("14", "pooled clinical, C difference (depth plus shape − depth)", "clinical_difference", "f3"),
    ("14", "landmark at the third assessment, pooled C difference", "clinical_third_landmark_pooled_difference", "f3"),
    ("14", "CA031, slope equal to log(1 − depth) / interval at the second assessment",
     "clinical_second_CA031_share_slope_from_depth", "pct"),
    ("14", "measured lesions only: depth + shape against depth + log(1 − depth), C difference",
     "clinical_nonlinear_second_sum_only_shape_vs_nonlinear_difference", "f3"),
    ("14", "measured lesions only, third assessment: depth + shape against depth, C difference",
     "clinical_nonlinear_third_sum_only_shape_vs_depth_difference", "f3"),
]


print("\nSUMMARY OF THE MAIN RESULTS (95% intervals where the register holds them; percentile ranges over "
      "refits or simulated animals are labelled as such)")
current_section = None
for section, label, keys, style in SUMMARY:
    if section != current_section:
        print(f"  Section {section}")
        current_section = section
    keys = keys if isinstance(keys, tuple) else (keys,)
    text = " / ".join(format_number(RESULTS.get(key), style if not key.endswith("_p") else "p")
                      for key in keys)
    if len(keys) == 1 and usable_value(RESULTS.get(f"{keys[0]}_low")):
        text += (f" ({format_number(RESULTS[keys[0] + '_low'], style)} to "
                 f"{format_number(RESULTS[keys[0] + '_high'], style)})")
    print(f"    {label:<74} {text}")
```
