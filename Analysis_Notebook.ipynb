# Depth and durability of response in patient-derived xenografts: analysis notebook

This notebook carries out every analysis of the study, from the source data to the result tables, the result records and the data behind every figure. It draws no figures: they are drawn by the separate figures notebook, which reads only the files this notebook saves, so a change to a figure never requires rerunning the analysis. Each section opens with a note that states what the section does, the inputs it reads and the objects it produces. The code cells then carry out the analysis. Every quantity that the study reports is stored in a results register as it is computed, and Section 15 writes that register, the tables and a run manifest to `outputs/`.

### How to run

1. Create the environment from `environment.yml` and register its kernel under the name `pdxe`.
2. Place the input files under `data/` as listed below, and the plan files under `prespecification/`.
3. Run the notebook from the project folder (the folder that holds `pdxe_utils.py`) with *Restart and run all*.
4. Run the figures notebook, which reads the files written to `outputs/`.

Every random seed is fixed in Section 1.1, and each analysis cell creates its own random generator from its own seed. Section 15.2 checks the internal consistency of the stored results, and Section 15.3 records the software versions, the checksums of every input and output file, and the run time.

## 1 · Configuration and shared functions

Sections 1.2 and 1.3 define every function used by more than one analysis. No later cell imports a package or defines a function.

Every analysis cell creates its own random generator from its seed in `SEEDS`, so the results of one cell do not depend on how many random numbers an earlier cell drew. Seeds added for the robustness analyses are new entries, so they leave every earlier stream unchanged.
```python
# ==========================================================================
# 1.2 Functions: records, checksums, statistics, the canonical
#     observation table, mechanistic models and model fitting
#
# Every function used by more than one analysis is defined here, once.
# ==========================================================================

# --------------------------------------------------------------------------
# Results register. Each analysis stores the quantities it reports under a
# descriptive key as it computes them; Section 15 writes the register to
# outputs/ and checks it for internal consistency. RESULT_SECTION remembers
# the section that stored each key, for the register's section column.
# --------------------------------------------------------------------------
RESULTS = {}
RESULT_SECTION = {}
CURRENT_SECTION = {"label": "1"}


def set_section(label):
    """Name the section whose values the following record() calls store."""
    CURRENT_SECTION["label"] = str(label)


def record(key, value):
    """Store one reported value under `key` and return it unchanged.

    NumPy scalars are converted to Python numbers so that the register can be
    written to JSON. A key stored twice keeps the later value, and the
    section column follows the later call.
    """
    if isinstance(value, np.generic):
        value = value.item()
    RESULTS[key] = value
    RESULT_SECTION[key] = CURRENT_SECTION["label"]
    return value


def relative(path):
    """A path relative to the project folder, so no output shows local folders."""
    try:
        return Path(path).resolve().relative_to(PROJECT_DIR.resolve()).as_posix()
    except ValueError:
        return Path(path).name


def sha256_of_file(path, chunk_bytes=1 << 20):
    """Streaming SHA-256 of a file's bytes."""
    digest = hashlib.sha256()
    with open(path, "rb") as handle:
        for block in iter(lambda: handle.read(chunk_bytes), b""):
            digest.update(block)
    return digest.hexdigest()


def plan_sha256(path):
    """SHA-256 of a plan file with line endings normalized to LF.

    Each plan's checksum was computed on the plan text before it was written,
    and the written file carries the line endings of the system that wrote
    it, so carriage returns are removed before hashing.
    """
    return hashlib.sha256(Path(path).read_bytes().replace(b"\r\n", b"\n")).hexdigest()


def load_prespecified_plan(key):
    """Load a plan from PRESPEC_DIR and assert its recorded checksum."""
    file_name, recorded_prefix = PRESPECIFIED_PLANS[key]
    path = PRESPEC_DIR / file_name
    if not path.exists():
        raise FileNotFoundError(
            f"{relative(path)} is missing. Copy the plan file fixed before "
            "testing into the prespecification folder under this name.")
    digest = plan_sha256(path)
    assert digest.startswith(recorded_prefix), (
        f"{file_name}: SHA-256 {digest[:16]} does not match the recorded "
        f"{recorded_prefix}. The file is not the plan that was fixed.")
    return json.loads(path.read_text(encoding="utf-8")), digest


def save_record(frame, name, index=False):
    """Write one analysis record to FROZEN_DIR as CSV."""
    frame.to_csv(FROZEN_DIR / f"{name}.csv", index=index)


def file_record(path):
    """Relative path, size in bytes and SHA-256 of one file, for the manifest."""
    path = Path(path)
    return {"path": relative(path), "bytes": path.stat().st_size, "sha256": sha256_of_file(path)}


def usable_value(value):
    """True for a stored value that a check can use: not None and, for a
    float, finite."""
    return value is not None and not (isinstance(value, float) and not np.isfinite(value))


def json_ready(value):
    """A value converted for JSON: NumPy scalars to Python numbers, NaN and
    infinities to None, paths to project-relative text, containers
    element by element."""
    if isinstance(value, np.generic):
        value = value.item()
    if isinstance(value, float) and not np.isfinite(value):
        return None
    if isinstance(value, Path):
        return relative(value)
    if isinstance(value, dict):
        return {str(key): json_ready(item) for key, item in value.items()}
    if isinstance(value, (list, tuple, set)):
        return [json_ready(item) for item in value]
    return value


# --------------------------------------------------------------------------
# Number formatting for printed output
# --------------------------------------------------------------------------
_SUPERSCRIPT = str.maketrans("-0123456789", "⁻⁰¹²³⁴⁵⁶⁷⁸⁹")


def format_p(p, resolution=None):
    """A P value with two significant figures, and m × 10⁻ᵏ below 0.001.

    `resolution` is the smallest non-zero value a resampling P value can
    take; below it the result is a bound ("< 0.0005"). A P value that
    underflows to zero is printed as a bound as well. The returned text
    starts with "<" exactly when it is a bound, which p_label relies on.
    """
    if p is None or not np.isfinite(p):
        return "not computed"
    if resolution is not None and p < resolution:
        return f"< {resolution:.2g}"
    if p == 0:
        return "< 1 × 10⁻³⁰⁰"
    if p < 0.001:
        mantissa, exponent = f"{p:.1e}".split("e")
        return f"{mantissa} × 10{str(int(exponent)).translate(_SUPERSCRIPT)}"
    return f"{p:.2g}"


def format_number(value, style):
    """A stored value for the summary: style "int", "pct" (as a percentage
    with one decimal), "p" (a P value) or "fN" (N decimals). A value that
    was not computed is shown as such."""
    if not usable_value(value):
        return "not computed"
    if style == "int":
        return f"{int(value):,}"
    if style == "pct":
        return f"{100 * value:.1f}%"
    if style == "p":
        return format_p(value)
    return f"{value:.{int(style[1:])}f}"


def p_label(p, resolution=None):
    """'P = 0.026' or 'P < 0.0005': the P value with the right relation sign,
    so that a bound is never printed as 'P = < ...'."""
    text = format_p(p, resolution)
    if text == "not computed":
        return "P not computed"
    return f"P {text}" if text.startswith("<") else f"P = {text}"


def format_ci(estimate, low, high, scale=100.0, decimals=1, unit="%"):
    """'19.7% (95% CI 13.1 to 26.0)' style, with a true minus sign."""
    def number(value):
        return f"{scale * value:.{decimals}f}".replace("-", "−")
    return f"{number(estimate)}{unit} (95% CI {number(low)} to {number(high)})"


# --------------------------------------------------------------------------
# Multiple testing and resampling
# --------------------------------------------------------------------------
def benjamini_hochberg(p_values):
    """Benjamini–Hochberg q values. Missing P values pass through and do not
    count toward the number of tests."""
    p = np.asarray(p_values, float)
    finite = np.isfinite(p)
    adjusted = np.full_like(p, np.nan)
    if not finite.any():
        return adjusted
    order = np.argsort(p[finite])
    ranked = p[finite][order]
    m = len(ranked)
    stepped = np.minimum.accumulate((ranked * m / np.arange(1, m + 1))[::-1])[::-1]
    restored = np.empty(m)
    restored[order] = np.clip(stepped, 0, 1)
    adjusted[finite] = restored
    return adjusted


def holm(p_values, alpha=0.05):
    """Holm step-down. Returns the per-test threshold and whether the test is
    rejected; once one test fails, every larger P value also fails."""
    p = np.asarray(p_values, float)
    order = np.argsort(p)
    thresholds = np.empty(len(p))
    rejected = np.zeros(len(p), dtype=bool)
    still_rejecting = True
    for rank, index in enumerate(order):
        thresholds[index] = alpha / (len(p) - rank)
        if still_rejecting and p[index] <= thresholds[index]:
            rejected[index] = True
        else:
            still_rejecting = False
    return thresholds, rejected


def rows_by_cluster(labels):
    """Row positions for each cluster label, clusters in sorted order."""
    labels = np.asarray(labels)
    return {cluster: np.flatnonzero(labels == cluster) for cluster in np.unique(labels)}


def cluster_bootstrap(cluster_rows, statistic, n_bootstrap, rng):
    """Resample whole clusters with replacement and evaluate `statistic` on the
    concatenated rows. Non-finite replicates are dropped."""
    clusters = np.array(list(cluster_rows))
    draws = []
    for _ in range(n_bootstrap):
        drawn = rng.choice(clusters, size=len(clusters), replace=True)
        value = statistic(np.concatenate([cluster_rows[c] for c in drawn]))
        if np.isfinite(value):
            draws.append(value)
    return np.asarray(draws)


def bootstrap_median_interval(values, n_bootstrap, rng):
    """95% percentile interval of the median, resampling values (n >= 4)."""
    values = np.asarray(values, float)
    values = values[np.isfinite(values)]
    if len(values) < 4:
        return np.nan, np.nan
    medians = np.median(values[rng.integers(0, len(values), size=(n_bootstrap, len(values)))], axis=1)
    return float(np.percentile(medians, 2.5)), float(np.percentile(medians, 97.5))


def bootstrap_p_two_sided(draws):
    """Twice the smaller share of bootstrap estimates on either side of zero."""
    draws = np.asarray(draws, float)
    return float(min(1.0, 2.0 * min(np.mean(draws <= 0), np.mean(draws >= 0))))


def rmse(observed, predicted):
    return float(np.sqrt(np.mean((np.asarray(observed) - np.asarray(predicted)) ** 2)))


def relative_rmse_reduction(observed, baseline, challenger):
    """(RMSE of baseline − RMSE of challenger) / RMSE of baseline."""
    base = rmse(observed, baseline)
    return (base - rmse(observed, challenger)) / base if base > 0 else np.nan


def reduction_with_interval(observed, baseline, challenger, cluster_rows, n_bootstrap, rng):
    """Relative RMSE reduction with a 95% percentile interval from resampling
    whole clusters (PDX models), predictions held fixed. Returns the
    estimate, the interval bounds and the bootstrap draws."""
    observed, baseline, challenger = (np.asarray(observed, float), np.asarray(baseline, float),
                                      np.asarray(challenger, float))
    draws = cluster_bootstrap(
        cluster_rows,
        lambda rows: relative_rmse_reduction(observed[rows], baseline[rows], challenger[rows]),
        n_bootstrap, rng)
    low, high = np.percentile(draws, [2.5, 97.5])
    return relative_rmse_reduction(observed, baseline, challenger), float(low), float(high), draws


def wilson_interval(successes, trials, z=1.96):
    """Wilson score interval for a binomial proportion."""
    if trials == 0:
        return np.nan, np.nan
    share = successes / trials
    centre = (share + z * z / (2 * trials)) / (1 + z * z / trials)
    half = z * np.sqrt(share * (1 - share) / trials + z * z / (4 * trials ** 2)) / (1 + z * z / trials)
    return centre - half, centre + half


def spearman(x, y):
    """Spearman correlation and P value, or NaN when either input is constant."""
    x, y = np.asarray(x, float), np.asarray(y, float)
    if len(x) < 3 or np.ptp(x) == 0 or np.ptp(y) == 0:
        return np.nan, np.nan
    result = stats.spearmanr(x, y)
    return float(result.statistic), float(result.pvalue)


def correlation_row(analysis, subset, x, y):
    """One row of a correlation table: Spearman ρ, its P value and n."""
    value, p_value = spearman(x, y)
    return {"analysis": analysis, "subset": subset, "n_arms": len(x),
            "spearman_rho": value, "p_value": p_value}


# --------------------------------------------------------------------------
# Regression fits shared by several sections
# --------------------------------------------------------------------------
MIXED_MODEL_OPTIMIZERS = ["lbfgs", "bfgs", "powell", "nm"]


def fit_mixed_model(formula, data, **model_arguments):
    """REML fit of a statsmodels linear mixed model that tries several
    optimizers in turn and stops at the first that converges.

    statsmodels moves to the next optimizer in the list only when the
    previous one fails, so a model that converges with L-BFGS gives the same
    estimates as a fit with L-BFGS alone. Returns the fit, whether it
    converged, and the distinct warning messages raised while fitting.
    """
    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter("always")
        fitted = smf.mixedlm(formula, data, **model_arguments).fit(
            reml=True, method=MIXED_MODEL_OPTIMIZERS)
    messages = sorted({str(w.message) for w in caught})
    return fitted, bool(getattr(fitted, "converged", False)), messages


def unpenalized_logistic():
    """Logistic regression without a penalty. C = ∞ is the form scikit-learn
    1.8 and later accept for an unpenalized fit (penalty=None is deprecated);
    the estimates are the maximum-likelihood estimates either way."""
    return LogisticRegression(C=np.inf, max_iter=5000)



# --------------------------------------------------------------------------
# The canonical observation table and per-arm summaries
# --------------------------------------------------------------------------
def canonical_observations(raw):
    """Apply the two preprocessing rules once, for every later analysis.

    Volumes below VOLUME_FLOOR (zero or recorded below 1 mm³) are set to the
    floor, and replicate readings on the same day of the same arm are averaged
    on the log scale (geometric mean), giving one row per arm per day.
    """
    frame = raw.copy()
    frame["volume"] = np.maximum(frame.volume.to_numpy(float), VOLUME_FLOOR)
    frame["log_volume"] = np.log(frame.volume)
    collapsed = (frame.groupby(["arm_id", "day"], as_index=False, sort=True)
                 .agg(model_id=("model_id", "first"),
                      tumor_type=("tumor_type", "first"),
                      treatment=("treatment", "first"),
                      is_control=("is_control", "first"),
                      log_volume=("log_volume", "mean"),
                      n_readings=("log_volume", "size")))
    collapsed["volume"] = np.exp(collapsed.log_volume)
    return collapsed


def arm_series(canonical, arm_ids=None):
    """Yield (arm_id, block) for each arm of the canonical table, sorted by day."""
    frame = canonical if arm_ids is None else canonical.loc[canonical.arm_id.isin(arm_ids)]
    for arm_id, block in frame.groupby("arm_id", sort=True):
        yield arm_id, block


def fit_line(day, log_volume):
    """Least-squares line on log volume: slope, intercept and residual SD with
    the two fitted parameters removed from the degrees of freedom."""
    day = np.asarray(day, float)
    log_volume = np.asarray(log_volume, float)
    slope, intercept = np.polyfit(day, log_volume, 1)
    residual = log_volume - (slope * day + intercept)
    residual_sd = (float(np.sqrt(np.sum(residual ** 2) / (len(day) - 2)))
                   if len(day) > 2 else np.nan)
    return float(slope), float(intercept), residual_sd


def summarize_regrowth(day, log_volume):
    """Model-free regrowth summary of one arm's series (sorted by day).

    The nadir is the first occurrence of the smallest volume. The post-nadir
    growth rate is the least-squares slope of log volume from the nadir on.
    """
    day = np.asarray(day, float)
    log_volume = np.asarray(log_volume, float)
    if len(day) < 3:
        return None
    nadir_index = int(np.argmin(log_volume))
    n_after_nadir = len(day) - 1 - nadir_index
    summary = {
        "n_measurements": len(day),
        "nadir_index": nadir_index,
        "n_after_nadir": n_after_nadir,
        "nadir_ratio": float(np.exp(log_volume[nadir_index] - log_volume[0])),
        "regrowth_multiple": float(np.exp(log_volume[-1] - log_volume[nadir_index])),
        "post_nadir_growth_rate": np.nan,
    }
    if n_after_nadir >= MIN_POINTS_AFTER_NADIR:
        summary["post_nadir_growth_rate"] = float(
            np.polyfit(day[nadir_index:] - day[nadir_index], log_volume[nadir_index:], 1)[0])
    return summary


def durability_phenotype(summary):
    """The durability phenotype applied to treated arms: shrank by at least
    REGRESSION_FRACTION, an interior nadir followed by at least two points,
    and regrowth to at least REGROWTH_MULTIPLE times the nadir."""
    return (summary["nadir_ratio"] <= 1 - REGRESSION_FRACTION
            and summary["nadir_index"] > 0
            and summary["n_after_nadir"] >= MIN_POINTS_AFTER_NADIR
            and summary["regrowth_multiple"] >= REGROWTH_MULTIPLE)


def series_frames(frame, id_column, day_column, value_column):
    """{series id: (days, values)} for each series of a long table, by day."""
    return {series: (block[day_column].to_numpy(float), block[value_column].to_numpy(float))
            for series, block in frame.sort_values(day_column).groupby(id_column, sort=True)}


def interpolation_noise(frames, span_range=None):
    """Measurement-error SD from interpolation residuals, no growth model assumed.

    `frames` maps a series id to (days, values) sorted by day. For consecutive
    readings (t0, y0), (t1, y1), (t2, y2) the middle one is predicted by
    linear interpolation between its neighbors. With locally linear growth
    the residual r has variance σ²(1 + w0² + w2²), so each triple contributes
    r²/(1 + w0² + w2²) as an estimate of σ². Equal spacing reduces to the
    second-difference estimator. With `span_range` = (low, high), only
    triples with low ≤ t2 − t0 ≤ high are used. Returns per-series arrays so
    that resampling can be clustered on series.
    """
    per_series = {}
    for series_id, (t, y) in frames.items():
        estimates = []
        for i in range(1, len(y) - 1):
            span = t[i + 1] - t[i - 1]
            if span <= 0:
                continue
            if span_range is not None and not (span_range[0] <= span <= span_range[1]):
                continue
            w0 = (t[i + 1] - t[i]) / span
            w2 = (t[i] - t[i - 1]) / span
            residual = y[i] - (w0 * y[i - 1] + w2 * y[i + 1])
            estimates.append(residual ** 2 / (1.0 + w0 ** 2 + w2 ** 2))
        if estimates:
            per_series[series_id] = np.asarray(estimates)
    return per_series


def summarize_interpolation_noise(per_series, rng, n_bootstrap=N_BOOTSTRAP):
    """Pooled σ with a bootstrap interval that resamples whole series."""
    if not per_series:
        return {"sigma": np.nan, "low": np.nan, "high": np.nan,
                "n_triples": 0, "n_series": 0}
    # The pooled estimate is sqrt(sum of contributions / number of triples),
    # so a resample only needs each series' sum and count.
    sums = np.array([values.sum() for values in per_series.values()])
    counts = np.array([len(values) for values in per_series.values()])
    sigma = float(np.sqrt(sums.sum() / counts.sum()))
    draws = np.empty(n_bootstrap)
    for b in range(n_bootstrap):
        picked = rng.integers(0, len(sums), size=len(sums))
        draws[b] = np.sqrt(sums[picked].sum() / counts[picked].sum())
    low, high = np.percentile(draws, [2.5, 97.5])
    return {"sigma": sigma, "low": float(low), "high": float(high),
            "n_triples": int(counts.sum()), "n_series": len(sums)}


# --------------------------------------------------------------------------
# Mechanistic models, on the log-volume scale
# --------------------------------------------------------------------------
def log_volume_exponential(t, log_baseline, growth_rate):
    """Untreated growth: log V(t) = log V0 + g t."""
    return log_baseline + growth_rate * np.asarray(t, float)


def log_volume_constant_kill(t, log_baseline, growth_rate, kill_rate):
    """Treatment lowers the growth rate by a fixed kill rate: dV/dt = (g − k)V."""
    return log_baseline + (growth_rate - kill_rate) * np.asarray(t, float)


def log_volume_tolerant(t, log_baseline, growth_rate, kill_rate, tolerant_fraction):
    """Tolerant subpopulation: V(t) = V0[(1 − f)e^((g − k)t) + f e^(gt)].

    Evaluated with logaddexp, so neither exponential is formed, and with
    log1p(−f), which keeps precision when f is small.
    """
    t = np.asarray(t, float)
    fraction = np.clip(tolerant_fraction, 1e-12, 1.0 - 1e-12)
    sensitive = np.log1p(-fraction) + (growth_rate - kill_rate) * t
    tolerant = np.log(fraction) + growth_rate * t
    return log_baseline + np.logaddexp(sensitive, tolerant)


def log_volume_declining(t, log_baseline, growth_rate, initial_kill_rate, decay_rate):
    """Declining drug effect: dV/dt = [g − k e^(−δt)]V, integrated in closed form.
    As δ → 0 the model is the constant-kill model, handled explicitly."""
    t = np.asarray(t, float)
    if decay_rate < 1e-10:
        return log_baseline + (growth_rate - initial_kill_rate) * t
    integrated_kill = (initial_kill_rate / decay_rate) * (-np.expm1(-decay_rate * t))
    return log_baseline + growth_rate * t - integrated_kill


def sensitivity_matrix(model, days, parameters, relative_step=1e-5):
    """Central-difference derivatives of predicted log volume (rows: days)
    with respect to each parameter (columns)."""
    parameters = np.asarray(parameters, float)
    columns = []
    for index in range(len(parameters)):
        step = relative_step * max(1.0, abs(parameters[index]))
        forward, backward = parameters.copy(), parameters.copy()
        forward[index] += step
        backward[index] -= step
        columns.append((model(days, *forward) - model(days, *backward)) / (2.0 * step))
    return np.column_stack(columns)


MECHANISMS = ("tolerant subpopulation", "declining drug effect")

# Fitting bounds on the unconstrained scale (log for rates and volume, logit
# for the fraction). A fit that ends on a bound is a degenerate limit and is
# flagged, never used silently.
MECHANISM_BOUNDS = {
    "tolerant subpopulation": [("log_baseline", 0.0, 11.5), ("log_growth", -6.9, 0.0),
                               ("log_kill", -6.9, 0.69), ("logit_fraction", -12.0, 4.0)],
    "declining drug effect": [("log_baseline", 0.0, 11.5), ("log_growth", -6.9, 0.0),
                              ("log_kill", -6.9, 0.69), ("log_decay", -11.5, 0.69)],
}


def constant_exposure_forward(mechanism):
    """Forward model on the unconstrained scale, under constant exposure."""
    def forward(theta, days):
        log_v0, log_growth, log_kill, fourth = theta
        if mechanism == "tolerant subpopulation":
            return log_volume_tolerant(days, log_v0, np.exp(log_growth), np.exp(log_kill),
                                       1.0 / (1.0 + np.exp(-fourth)))
        return log_volume_declining(days, log_v0, np.exp(log_growth), np.exp(log_kill),
                                    float(np.exp(fourth)))
    return forward


def mechanism_starts(observed):
    """Three starting points shared by every mechanistic fit."""
    return [np.array([observed[0], np.log(0.05), np.log(0.08), -2.0]),
            np.array([observed[0], np.log(0.02), np.log(0.15), -1.0]),
            np.array([observed[0], np.log(0.09), np.log(0.04), -3.0])]


def fit_bounded(forward, bounds, days, observed, starts, max_nfev=2000,
                pin_tolerance=1e-6):
    """Bounded multi-start least squares. Returns the best fit as a dict with
    the residual sum of squares, RMSE, Akaike criterion and pinned parameters."""
    names = [b[0] for b in bounds]
    low = np.array([b[1] for b in bounds])
    high = np.array([b[2] for b in bounds])
    margin = 1e-3 * (high - low)
    best = None
    for start in starts:
        start = np.clip(start, low + margin, high - margin)
        try:
            solution = least_squares(lambda theta: forward(theta, days) - observed,
                                     start, bounds=(low, high), max_nfev=max_nfev)
        except (ValueError, np.linalg.LinAlgError):
            continue
        rss = float(np.sum(solution.fun ** 2))
        if np.isfinite(rss) and (best is None or rss < best["rss"]):
            best = {"rss": rss, "x": solution.x}
    if best is None:
        return {"rss": np.inf, "rmse": np.inf, "criterion": np.inf,
                "n_pinned": len(names), "pinned": "all starts failed",
                "x": np.full(len(names), np.nan), "names": names}
    span = np.maximum(high - low, 1e-12)
    at_low = (best["x"] - low) / span < pin_tolerance
    at_high = (high - best["x"]) / span < pin_tolerance
    pinned = [f"{name}@{'low' if lo else 'high'}"
              for name, lo, hi in zip(names, at_low, at_high) if lo or hi]
    n = len(observed)
    return {"rss": best["rss"], "rmse": float(np.sqrt(best["rss"] / n)),
            "criterion": float(n * np.log(max(best["rss"], 1e-12) / n) + 2 * len(names)),
            "n_pinned": len(pinned), "pinned": ";".join(pinned),
            "x": best["x"], "names": names}


def fit_mechanism(mechanism, days, observed):
    """Fit one regrowth mechanism under constant exposure."""
    return fit_bounded(constant_exposure_forward(mechanism), MECHANISM_BOUNDS[mechanism],
                       days, observed, mechanism_starts(observed))


def tolerant_estimates(theta):
    """Growth rate, kill rate and tolerant fraction from a fitted parameter vector."""
    return {"growth_rate": float(np.exp(theta[1])), "kill_rate": float(np.exp(theta[2])),
            "tolerant_fraction": float(1.0 / (1.0 + np.exp(-theta[3])))}


def recovery_summary(log_errors, log_truths):
    """Median absolute log error, bias, error-to-prior ratio (RMSE of the log
    error over the SD of the log true values) and whether the prespecified
    recovery threshold is met."""
    log_errors, log_truths = np.asarray(log_errors), np.asarray(log_truths)
    median_error = float(np.median(np.abs(log_errors)))
    return {"replicates": len(log_errors), "median_abs_log_error": median_error,
            "bias_log": float(np.mean(log_errors)),
            "error_to_prior_ratio": float(np.sqrt(np.mean(log_errors ** 2)) / np.std(log_truths)),
            "recovered": median_error < RECOVERY_THRESHOLD}


def aic_verdict(fits, tie_threshold=None):
    """Which mechanism the Akaike criterion prefers. With `tie_threshold`, a
    difference smaller than it is recorded as no decision."""
    delta = fits["declining drug effect"]["criterion"] - fits["tolerant subpopulation"]["criterion"]
    if not np.isfinite(delta):
        return "undecidable", delta
    if tie_threshold is not None and abs(delta) < tie_threshold:
        return "tie", delta
    return ("tolerant subpopulation" if delta > 0 else "declining drug effect"), delta


def apply_termination(days, log_volume, threshold=TERMINATION_VOLUME):
    """End a simulated arm at its first measurement above the threshold volume,
    keeping that measurement, as the source study removed large tumors."""
    above = np.flatnonzero(log_volume > np.log(threshold))
    if len(above):
        cut = above[0] + 1
        return days[:cut], log_volume[:cut], True
    return days, log_volume, False


def sample_truncated_normal(rng, centre, spread, upper):
    """One draw from Normal(centre, spread) truncated above at `upper`."""
    while True:
        value = rng.normal(centre, spread)
        if value <= upper:
            return value


def draw_prior_parameters(rng, prior):
    """One parameter set of the tolerant-subpopulation model from the priors of
    the Bayesian model in Section 5.6 (prior dictionary built there)."""
    return {
        "log_baseline": rng.normal(np.log(200.0), 0.5),
        "growth_rate": float(np.exp(sample_truncated_normal(
            rng, prior["growth_centre"], prior["growth_spread"],
            np.log(prior["growth_upper"])))),
        "kill_rate": float(abs(rng.normal(0.0, 0.10))),
        "tolerant_fraction": float(1.0 / (1.0 + np.exp(-rng.normal(-2.2, 1.5)))),
    }


def tolerant_subpopulation_model(days, prior, log_volume=None):
    """The tolerant-subpopulation model as a probabilistic program.

    Priors: log baseline volume Normal(log 200, 0.5); log growth rate
    TruncatedNormal(centre, spread, upper) from the untreated arms (Section
    5.6); kill rate HalfNormal(0.10); logit tolerant fraction Normal(−2.2,
    1.5); measurement SD HalfNormal(0.30). The likelihood is on log volume.
    """
    log_baseline = numpyro.sample("log_baseline", dist.Normal(np.log(200.0), 0.5))
    log_growth = numpyro.sample("log_growth_rate", dist.TruncatedNormal(
        loc=prior["growth_centre"], scale=prior["growth_spread"],
        high=np.log(prior["growth_upper"])))
    growth_rate = numpyro.deterministic("growth_rate", jax.numpy.exp(log_growth))
    kill_rate = numpyro.sample("kill_rate", dist.HalfNormal(0.10))
    fraction_logit = numpyro.sample("fraction_logit", dist.Normal(-2.2, 1.5))
    fraction = numpyro.deterministic("tolerant_fraction", jax.nn.sigmoid(fraction_logit))
    noise_scale = numpyro.sample("noise_scale", dist.HalfNormal(0.30))
    sensitive = jax.numpy.log1p(-fraction) + (growth_rate - kill_rate) * days
    tolerant = jax.numpy.log(fraction) + growth_rate * days
    predicted = log_baseline + jax.numpy.logaddexp(sensitive, tolerant)
    numpyro.sample("observed", dist.Normal(predicted, noise_scale), obs=log_volume)


# --------------------------------------------------------------------------
# Drug exposure (Section 12): one-compartment kinetics, analytic integrals
# --------------------------------------------------------------------------
def dosing_days(interval_days, duration_days):
    """Dosing days at a fixed interval (1.0 = once daily) up to duration_days."""
    return np.arange(0.0, duration_days + 1e-9, interval_days)


def cumulative_exposure(t, dose_days, half_life_days, normalize_over=None):
    """∫₀ᵗ q(s) ds for superposed first-order doses with instantaneous
    absorption. With `normalize_over` = T, the result is scaled so that mean
    exposure over [0, T] is 1, which keeps kill rates comparable with the
    constant-exposure models. T is the last day of the grid being modeled, so
    the scaling (a constant factor absorbed by the kill rate) depends on it."""
    t = np.atleast_1d(np.asarray(t, float))
    decay = np.log(2.0) / half_life_days
    elapsed = np.maximum(t[:, None] - dose_days[None, :], 0.0)
    active = t[:, None] >= dose_days[None, :]
    total = np.where(active, (1.0 - np.exp(-decay * elapsed)) / decay, 0.0).sum(axis=1)
    if normalize_over is not None and normalize_over > 0:
        reference = cumulative_exposure(np.array([normalize_over]), dose_days,
                                        half_life_days)[0]
        if reference > 0:
            total = total * (normalize_over / reference)
    return total


def log_volume_tolerant_exposure(t, log_baseline, growth_rate, kill_rate,
                                 tolerant_fraction, dose_days, half_life_days):
    """Tolerant subpopulation with the kill term driven by drug exposure:
    sensitive cells dV/dt = (g − k q(t))V; tolerant cells dV/dt = gV."""
    t = np.asarray(t, float)
    fraction = np.clip(tolerant_fraction, 1e-12, 1.0 - 1e-12)
    integrated = cumulative_exposure(t, dose_days, half_life_days,
                                     normalize_over=float(np.max(t)) or 1.0)
    sensitive = np.log1p(-fraction) + growth_rate * t - kill_rate * integrated
    tolerant = np.log(fraction) + growth_rate * t
    return log_baseline + np.logaddexp(sensitive, tolerant)


def log_volume_declining_exposure(t, log_baseline, growth_rate, kill_rate,
                                  biological_decay, dose_days, half_life_days):
    """Declining effect with clearance separated from biology:
    dV/dt = (g − k q(t) e^(−δt))V. The integral stays analytic, and it is
    scaled by the same exposure-only factor as the tolerant model."""
    t = np.asarray(t, float)
    pk_decay = np.log(2.0) / half_life_days
    elapsed = np.maximum(t[:, None] - dose_days[None, :], 0.0)
    active = t[:, None] >= dose_days[None, :]
    combined = pk_decay + biological_decay
    if combined < 1e-12:
        per_dose = np.where(active, elapsed, 0.0)
    else:
        per_dose = np.where(active, np.exp(-biological_decay * dose_days[None, :])
                            * (1.0 - np.exp(-combined * elapsed)) / combined, 0.0)
    integrated = per_dose.sum(axis=1)
    reference_days = float(np.max(t)) or 1.0
    reference = cumulative_exposure(np.array([reference_days]), dose_days, half_life_days)[0]
    if reference > 0:
        integrated = integrated * (reference_days / reference)
    return log_baseline + growth_rate * t - kill_rate * integrated


def exposure_forward(mechanism, dose_days, half_life_days):
    """Forward model on the unconstrained scale, with explicit exposure."""
    def forward(theta, days):
        log_v0, log_growth, log_kill, fourth = theta
        if mechanism == "tolerant subpopulation":
            return log_volume_tolerant_exposure(
                days, log_v0, np.exp(log_growth), np.exp(log_kill),
                1.0 / (1.0 + np.exp(-fourth)), dose_days, half_life_days)
        return log_volume_declining_exposure(
            days, log_v0, np.exp(log_growth), np.exp(log_kill), float(np.exp(fourth)),
            dose_days, half_life_days)
    return forward


def fit_mechanism_exposure(mechanism, days, observed, dose_days, half_life_days):
    """Fit one regrowth mechanism with explicit exposure."""
    return fit_bounded(exposure_forward(mechanism, dose_days, half_life_days),
                       MECHANISM_BOUNDS[mechanism], days, observed,
                       mechanism_starts(observed))


def draw_schedule(rng, pool, window_days, keep_every=1):
    """One measurement schedule drawn from `pool`, cut at `window_days` and,
    with keep_every = k, thinned to every k-th measurement. Returns None when
    fewer than five measurements remain (no further random draws are made)."""
    grid = pool[rng.integers(len(pool))]
    days = grid[grid <= window_days][::keep_every]
    return days if len(days) >= 5 else None


def typical_schedule(pool, window_days):
    """The schedule, cut at `window_days`, whose number of measurements is
    closest to the median of the pool (the first such schedule in pool order).
    Figures use it so that they show a typical design, not the densest one."""
    cut = [grid[grid <= window_days] for grid in pool]
    counts = np.array([len(days) for days in cut], float)
    return cut[int(np.argmin(np.abs(counts - np.median(counts))))]


def paired_exposure_trajectories(rng, days, dose_days, half_life_days):
    """Both mechanisms from one shared draw of baseline volume, growth rate,
    kill rate and noise realization, so that only the mechanism differs.

    log V0 ~ Normal(log 200, 0.3); growth rate ~ LogNormal(log 0.05, 0.5);
    kill rate ~ |Normal(0.13, 0.04)|; noise ~ Normal(0, 1) per measurement
    (scaled by the noise SD when scored). Mechanism-specific: logit tolerant
    fraction ~ Normal(−2.5, 1.0); biological decay ~ LogNormal(log 0.04, 0.5).
    The draw order is fixed, so a seed reproduces the same arms."""
    log_v0 = np.log(200.0) + rng.normal(0, 0.3)
    growth = float(np.exp(rng.normal(np.log(0.05), 0.5)))
    kill = float(abs(rng.normal(0.13, 0.04)))
    noise = rng.normal(0, 1.0, size=len(days))
    fraction = float(1.0 / (1.0 + np.exp(-rng.normal(-2.5, 1.0))))
    decay = float(np.exp(rng.normal(np.log(0.04), 0.5)))
    clean = {
        "tolerant subpopulation": log_volume_tolerant_exposure(
            days, log_v0, growth, kill, fraction, dose_days, half_life_days),
        "declining drug effect": log_volume_declining_exposure(
            days, log_v0, growth, kill, decay, dose_days, half_life_days),
    }
    return clean, noise


def score_exposure_pair(clean, noise, noise_sd, days, dose_days, half_life_days, tie_threshold):
    """Fit both mechanisms to each simulated arm and record the AIC verdict,
    once with the full schedule and once with the arm ended at its first
    measurement above TERMINATION_VOLUME (`termination` = "not applied" or
    "applied"). An arm left with fewer than five measurements is undecidable.
    When the rule does not shorten the arm, the full-schedule fits are reused.
    """
    rows = []
    for truth in MECHANISMS:
        observed = clean[truth] + noise_sd * noise
        cut_days, cut_observed, stopped = apply_termination(days, observed)
        results_by_length = {}
        for rule, (arm_days, arm_observed) in (("not applied", (days, observed)),
                                               ("applied", (cut_days, cut_observed))):
            if len(arm_days) not in results_by_length:
                if len(arm_days) < 5:
                    result = {"verdict": "undecidable", "criterion_delta": np.nan,
                              "rmse_right": np.nan, "rmse_wrong": np.nan, "any_pinned": False}
                else:
                    fits = {m: fit_mechanism_exposure(m, arm_days, arm_observed, dose_days,
                                                      half_life_days) for m in MECHANISMS}
                    verdict, delta = aic_verdict(fits, tie_threshold)
                    wrong = [m for m in MECHANISMS if m != truth][0]
                    result = {"verdict": verdict, "criterion_delta": delta,
                              "rmse_right": fits[truth]["rmse"], "rmse_wrong": fits[wrong]["rmse"],
                              "any_pinned": (fits[truth]["n_pinned"] + fits[wrong]["n_pinned"]) > 0}
                results_by_length[len(arm_days)] = result
            result = results_by_length[len(arm_days)]
            rows.append({"termination": rule, "true_mechanism": truth,
                         "correct": result["verdict"] == truth,
                         "terminated": stopped and rule == "applied",
                         "n_measurements": len(arm_days), **result})
    return rows


def summarize_verdicts(block):
    """Decisive rate and accuracy among decisive calls, each with counts and a
    Wilson 95% interval, and the share choosing the tolerant-subpopulation
    model. The decisive-rate interval treats the arms of a cell as
    independent; the two arms of one paired replicate share their parameters,
    so it is somewhat narrow."""
    decisive = block.verdict.isin(MECHANISMS)
    n_decisive = int(decisive.sum())
    n_correct = int(block.loc[decisive, "correct"].sum())
    low, high = wilson_interval(n_correct, n_decisive)
    decisive_low, decisive_high = wilson_interval(n_decisive, len(block))
    return pd.Series({
        "n": len(block),
        "median_measurements": float(block.n_measurements.median()),
        "decisive_rate": float(decisive.mean()),
        "decisive_low": decisive_low, "decisive_high": decisive_high,
        "decisive_calls": n_decisive,
        "correct_calls": n_correct,
        "accuracy_when_decisive": n_correct / n_decisive if n_decisive else np.nan,
        "accuracy_low": low, "accuracy_high": high,
        "chose_tolerant": float((block.verdict == "tolerant subpopulation").mean()),
        "terminated_share": float(block.terminated.mean()),
    })
```

```python
# ==========================================================================
# 1.3 Functions: forecasting features and learners, within-model ranking,
#     molecular scores and grouped ridge regression, and the clinical
#     tumor-burden endpoints
# ==========================================================================

# --------------------------------------------------------------------------
# Forecasting (Sections 7 and 8)
# --------------------------------------------------------------------------
def early_window_features(day, log_volume, origin_day):
    """Predictors from one arm's measurements on or before `origin_day`.

    Depth-derived quantities: baseline log volume, best average response in
    the window, nadir depth (log baseline − log minimum), nadir-to-baseline
    ratio, minimum log volume and the current (last in-window) log volume.
    Shape descriptors: overall slope, initial slope (first three points),
    steepest decline, time to nadir, post-nadir slope inside the window
    (0.0 when the in-window minimum is the last point), curvature (quadratic
    coefficient) and change in slope (post-nadir slope − initial slope).
    """
    in_window = day <= origin_day
    window_day, window_log = day[in_window], log_volume[in_window]
    if len(window_day) < 3 or np.ptp(window_day) == 0:
        return None
    nadir_index = int(np.argmin(window_log))
    initial_slope = float(np.polyfit(window_day[:3], window_log[:3], 1)[0])
    post_nadir_slope = (float(np.polyfit(window_day[nadir_index:], window_log[nadir_index:], 1)[0])
                        if len(window_day) - nadir_index >= 2 else 0.0)
    return {
        "baseline_log_volume": float(window_log[0]),
        "response_in_window": utils.best_average_response(
            window_day, np.exp(window_log), min_day=MIN_DAY_FOR_BEST_AVERAGE_RESPONSE),
        "nadir_depth": float(window_log[0] - window_log.min()),
        "nadir_ratio": float(np.exp(window_log.min() - window_log[0])),
        "minimum_log_volume": float(window_log.min()),
        "current_log_volume": float(window_log[-1]),
        "overall_slope": float(np.polyfit(window_day, window_log, 1)[0]),
        "initial_slope": initial_slope,
        "steepest_decline": float(np.min(np.diff(window_log) / np.diff(window_day))),
        "time_to_nadir": float(window_day[nadir_index] - window_day[0]),
        "post_nadir_slope_in_window": post_nadir_slope,
        "curvature": float(np.polyfit(window_day, window_log, 2)[0]),
        "slope_change": post_nadir_slope - initial_slope,
    }


def build_forecast_table(canonical, origin_day, horizon_day, arm_ids=None,
                         outcome_zeros=None):
    """One row per arm: early-window predictors and the log volume at the
    horizon, linearly interpolated between the measurements around it. Arms
    whose observation ended before the horizon are not included.

    outcome_zeros (default FORECAST_OUTCOME_ZEROS): "floor" interpolates the
    outcome from every reading, with readings at or below zero at 1 mm³;
    "drop" interpolates it from readings above the floor only, as in the
    forecasting analysis first run. Predictors always use the floored readings."""
    outcome_zeros = outcome_zeros or FORECAST_OUTCOME_ZEROS
    rows = []
    for arm_id, block in arm_series(canonical, arm_ids):
        day = block.day.to_numpy(float)
        log_volume = block.log_volume.to_numpy(float)
        kept = block.volume.to_numpy(float) > VOLUME_FLOOR if outcome_zeros == "drop" else np.ones(len(day), bool)
        outcome_day, outcome_log_volume = day[kept], log_volume[kept]
        if (len(day) < 3 or len(outcome_day) == 0 or outcome_day.max() < horizon_day
                or outcome_day.min() > horizon_day):
            continue
        features = early_window_features(day, log_volume, origin_day)
        if features is None:
            continue
        rows.append({"arm_id": arm_id, "model_id": block.model_id.iloc[0],
                     "treatment": block.treatment.iloc[0],
                     "tumor_type": block.tumor_type.iloc[0],
                     "outcome_log_volume": float(np.interp(horizon_day, outcome_day, outcome_log_volume)),
                     **features})
    return pd.DataFrame(rows)


def make_forecaster():
    """The one gradient-boosted regressor used for every predictor set."""
    return HistGradientBoostingRegressor(max_iter=300, learning_rate=0.05,
                                         max_leaf_nodes=15, l2_regularization=1.0,
                                         random_state=SEED)


def grouped_folds(frame, n_folds=N_FOLDS):
    """Cross-validation folds that keep every arm of a PDX model in one fold."""
    groups = frame.model_id.astype(str).to_numpy()
    n_splits = min(n_folds, len(np.unique(groups)))
    folds = list(GroupKFold(n_splits=n_splits).split(frame, groups=groups))
    for train, test in folds:
        assert set(groups[train]).isdisjoint(groups[test]), "a PDX model is in two folds"
    return folds


def out_of_fold_predictions(frame, predictors, outcome, folds):
    """Predictions for every row, each from a model that did not see its PDX model."""
    design = frame[predictors].to_numpy(float)
    predicted = np.full(len(outcome), np.nan)
    for train, test in folds:
        learner = make_forecaster()
        learner.fit(design[train], outcome[train])
        predicted[test] = learner.predict(design[test])
    return predicted


def compare_predictor_sets(frame, predictor_sets, folds=None):
    """Out-of-fold predictions and RMSE for several predictor sets on one frame."""
    outcome = frame.outcome_log_volume.to_numpy(float)
    folds = grouped_folds(frame) if folds is None else folds
    predictions = {name: out_of_fold_predictions(frame, predictors, outcome, folds)
                   for name, predictors in predictor_sets.items()}
    errors = {name: rmse(outcome, predicted) for name, predicted in predictions.items()}
    return predictions, errors


def selection_by_model(selection, prediction_columns):
    """For each PDX model with two or more arms: the regret of the arm each
    prediction would select (its observed log volume minus the smallest
    observed log volume among the model's arms) and whether the selections
    differ."""
    rows = []
    for model_id, block in selection.groupby("model_id", sort=True):
        if len(block) < 2:
            continue
        best = float(block.outcome_log_volume.min())
        row = {"model_id": model_id, "n_treatments": len(block)}
        chosen = {}
        for column in prediction_columns:
            choice = block.loc[block[column].idxmin()]
            chosen[column] = choice.arm_id
            row[f"{column}_regret"] = float(choice.outcome_log_volume - best)
        row["different_choice"] = len(set(chosen.values())) > 1
        rows.append(row)
    return pd.DataFrame(rows)


def within_model_pairs(selection, prediction_columns):
    """Every pair of arms within a PDX model with distinct observed outcomes.

    For each prediction column the pair scores 1 when the predicted order
    matches the observed order, 0.5 when the predictions are tied and 0
    otherwise. Pairs with tied observed outcomes have no order and are left
    out.
    """
    rows = []
    for model_id, block in selection.groupby("model_id", sort=True):
        outcome = block.outcome_log_volume.to_numpy(float)
        predictions = {c: block[c].to_numpy(float) for c in prediction_columns}
        for i, j in combinations(range(len(block)), 2):
            observed_gap = outcome[i] - outcome[j]
            if observed_gap == 0:
                continue
            row = {"model_id": model_id, "observed_gap": abs(observed_gap)}
            for column, predicted in predictions.items():
                predicted_gap = predicted[i] - predicted[j]
                row[column] = 0.5 if predicted_gap == 0 else float(
                    np.sign(predicted_gap) == np.sign(observed_gap))
            rows.append(row)
    return pd.DataFrame(rows)


def ranking_difference(pairs, column_a, column_b, n_bootstrap, rng):
    """Pairwise ranking accuracy of `column_a` minus that of `column_b` over
    the rows of a within_model_pairs table, with a 95% percentile interval
    from resampling PDX models (each model's pairs move together)."""
    scores_a, scores_b = pairs[column_a].to_numpy(float), pairs[column_b].to_numpy(float)
    draws = cluster_bootstrap(rows_by_cluster(pairs.model_id),
                              lambda rows: scores_a[rows].mean() - scores_b[rows].mean(),
                              n_bootstrap, rng)
    low, high = np.percentile(draws, [2.5, 97.5])
    return float(scores_a.mean() - scores_b.mean()), float(low), float(high)


# --------------------------------------------------------------------------
# Molecular correlates (Section 9)
# --------------------------------------------------------------------------
RIDGE_ALPHAS = np.logspace(-3, 4, 30)


def _ridge_path(train_x, train_y, test_x, alphas):
    """Predictions on test_x for every penalty in `alphas`, by SVD of the
    standardized training design (the ridge solution path in closed form)."""
    centre, scale = train_x.mean(axis=0), train_x.std(axis=0)
    keep = scale > 0
    if not keep.any():
        return np.repeat(np.full((len(test_x), 1), train_y.mean()), len(alphas), axis=1)
    x = (train_x[:, keep] - centre[keep]) / scale[keep]
    x_test = (test_x[:, keep] - centre[keep]) / scale[keep]
    y_mean = train_y.mean()
    u, s, vt = np.linalg.svd(x, full_matrices=False)
    uty = u.T @ (train_y - y_mean)
    factors = s[None, :] / (s[None, :] ** 2 + alphas[:, None])
    coefficients = (factors * uty[None, :]) @ vt
    return x_test @ coefficients.T + y_mean


def grouped_ridge_r2(design, target, groups, alphas=RIDGE_ALPHAS, n_outer=5, n_inner=5):
    """Out-of-fold R² of ridge regression with PDX-model-grouped folds.

    The penalty is chosen inside each training fold by an inner grouped
    cross-validation, and missing feature values are imputed with
    training-fold medians, so nothing from a test fold reaches its own
    prediction.
    """
    outer_splits = min(n_outer, len(np.unique(groups)))
    if outer_splits < 2:
        return np.nan
    predicted = np.full(len(target), np.nan)
    for train, test in GroupKFold(n_splits=outer_splits).split(design, groups=groups):
        medians = np.nanmedian(design[train], axis=0)
        medians = np.where(np.isfinite(medians), medians, 0.0)
        train_x = np.where(np.isfinite(design[train]), design[train], medians)
        test_x = np.where(np.isfinite(design[test]), design[test], medians)
        train_y, train_groups = target[train], groups[train]
        inner_splits = min(n_inner, len(np.unique(train_groups)))
        inner_error = np.zeros(len(alphas))
        for inner_train, inner_test in GroupKFold(n_splits=inner_splits).split(
                train_x, groups=train_groups):
            path = _ridge_path(train_x[inner_train], train_y[inner_train],
                               train_x[inner_test], alphas)
            inner_error += ((path - train_y[inner_test][:, None]) ** 2).sum(axis=0)
        best_alpha = alphas[[int(np.argmin(inner_error))]]
        predicted[test] = _ridge_path(train_x, train_y, test_x, best_alpha)[:, 0]
    return float(r2_score(target, predicted))


def block_r2(frame, features, outcome_column):
    """Out-of-fold R² of one feature block for one outcome."""
    return grouped_ridge_r2(frame[features].to_numpy(float),
                            frame[outcome_column].to_numpy(float),
                            frame.model_id.astype(str).to_numpy())


def block_features(block):
    """All feature columns of one block specification."""
    return block["model"] + block["arm"] + block["fixed"]


def dissociation_statistic(frame, feature_blocks, outcomes):
    """D = [R²(signatures, durability) − R²(context, durability)]
         − [R²(signatures, depth) − R²(context, depth)]."""
    r2 = {(block, label): block_r2(frame, block_features(feature_blocks[block]), column)
          for block in ("signatures", "context") for label, column in outcomes.items()}
    return ((r2[("signatures", "durability")] - r2[("context", "durability")])
            - (r2[("signatures", "depth")] - r2[("context", "depth")]))


def normalized_set_name(name):
    """A gene-set name reduced to upper-case letters and digits, without the
    HALLMARK prefix, so that the MSigDB GMT name
    (HALLMARK_EPITHELIAL_MESENCHYMAL_TRANSITION) and the Enrichr name
    (Epithelial Mesenchymal Transition) of one set compare equal."""
    text = "".join(character for character in str(name).upper() if character.isalnum())
    return text[len("HALLMARK"):] if text.startswith("HALLMARK") else text


def read_gene_set_library(path):
    """{normalized name: (name as written, [genes])} from a GMT file or an
    Enrichr text library.

    Both formats hold one set per line, tab-separated: the name, a second
    field (a description or URL in GMT, usually empty in Enrichr) and then
    the genes. Some Enrichr libraries append ",weight" to each gene; the
    weight is removed. Gene symbols are upper-cased and duplicates dropped.
    """
    library = {}
    for line in Path(path).read_text(encoding="utf-8", errors="replace").splitlines():
        fields = [field.strip() for field in line.rstrip("\n").split("\t")]
        if len(fields) < 3 or not fields[0]:
            continue
        genes = [field.split(",")[0].strip().upper() for field in fields[2:] if field.strip()]
        library[normalized_set_name(fields[0])] = (fields[0], list(dict.fromkeys(genes)))
    return library


def gene_set_score(expression_ranks, genes):
    """Mean within-model percentile rank of the genes present, minus 0.5 (the
    scoring rule of Section 9.1), and the number of genes present."""
    present = [gene for gene in genes if gene in expression_ranks.index]
    if not present:
        return pd.Series(np.nan, index=expression_ranks.columns), 0
    return expression_ranks.loc[present].mean(axis=0) - 0.5, len(present)


def permute_within_tumor_type(frame, model_features, arm_features, rng):
    """Break the feature–outcome link while keeping tumor-type structure.

    PDX-model-level features are reassigned from another PDX model of the
    same tumor type (every arm of a model receives the donor's values);
    arm-level features are shuffled across arms of the same tumor type.
    Tumor-type indicators cannot change under either shuffle.
    """
    permuted = frame.copy()
    for _, positions in frame.groupby("tumor_type", sort=True).indices.items():
        block = frame.iloc[positions]
        if model_features:
            models = block.model_id.unique()
            donor = dict(zip(models, rng.permutation(models)))
            values = block.groupby("model_id", sort=False)[model_features].first()
            permuted.iloc[positions, [frame.columns.get_loc(c) for c in model_features]] = (
                values.reindex(block.model_id.map(donor)).to_numpy())
        if arm_features:
            permuted.iloc[positions, [frame.columns.get_loc(c) for c in arm_features]] = (
                block[arm_features].to_numpy()[rng.permutation(len(block))])
    return permuted


# --------------------------------------------------------------------------
# Mechanism class (Section 11)
# --------------------------------------------------------------------------
def bootstrap_mean(values, n_bootstrap, seed):
    """Mean with a 95% percentile interval and two-sided bootstrap P value."""
    rng = np.random.default_rng(seed)
    values = np.asarray(values, float)
    values = values[np.isfinite(values)]
    draws = np.array([rng.choice(values, size=len(values), replace=True).mean()
                      for _ in range(n_bootstrap)])
    low, high = np.percentile(draws, [2.5, 97.5])
    return {"estimate": float(values.mean()), "low": float(low), "high": float(high),
            "p_two_sided": bootstrap_p_two_sided(draws), "n": len(values)}


def family_contrast_statistic(model_codes, code_a, code_b, differences, group_one, group_two):
    """Mean over PDX models of the mean oriented difference (group one minus
    group two) across within-model pairs with one member in each group.
    `differences` are member a minus member b. Returns the statistic, the
    number of PDX models and the number of pairs."""
    a_first = np.isin(code_a, group_one) & np.isin(code_b, group_two)
    b_first = np.isin(code_b, group_one) & np.isin(code_a, group_two)
    usable = (a_first | b_first) & np.isfinite(differences)
    if not usable.any():
        return np.nan, 0, 0
    oriented = np.where(a_first, differences, -differences)[usable]
    models = model_codes[usable]
    sums = np.bincount(models, weights=oriented)
    counts = np.bincount(models)
    present = counts > 0
    return float((sums[present] / counts[present]).mean()), int(present.sum()), int(usable.sum())


# --------------------------------------------------------------------------
# Clinical tumor-burden endpoints (Section 14)
# --------------------------------------------------------------------------
PROGRESSION_RATIO = 1.20          # RECIST 1.1: >= 20% above the reference sum
PROGRESSION_ABSOLUTE_MM = 5.0     # and >= 5 mm absolute increase


def read_sas_table(path):
    """A Project Data Sphere SAS table, with the trailing spaces of
    fixed-width character storage stripped."""
    frame = pd.read_sas(path, encoding="latin-1")
    for column in frame.select_dtypes(include=["object", "string"]).columns:
        frame[column] = frame[column].str.strip()
    return frame


def crosses_progression(sld, reference):
    """RECIST 1.1 sum criterion: at least 20% and at least 5 mm above the reference."""
    return bool(np.isfinite(reference) and sld >= max(PROGRESSION_RATIO * reference,
                                                       reference + PROGRESSION_ABSOLUTE_MM))


def progressed_before(block, day_limit, new_lesion_day):
    """True when RECIST progression occurred strictly before `day_limit`: a new
    lesion, or a sum that crossed the threshold relative to the smallest
    complete-assessment sum recorded before it. `block` holds one patient's
    assessments (day, sld, complete) sorted by day."""
    if pd.notna(new_lesion_day) and new_lesion_day < day_limit:
        return True
    reference = np.inf
    for day, sld, complete in block.loc[block.day < day_limit, ["day", "sld", "complete"]].itertuples(index=False):
        if crosses_progression(sld, reference):
            return True
        if complete:
            reference = min(reference, sld)
    return False


def time_to_progression(block, origin_day, new_lesion_day, new_lesions="event",
                        drop_lowest=False):
    """Time from `origin_day` to RECIST 1.1 progression, or to censoring.

    Each assessment is compared with the smallest complete-assessment sum
    recorded before it, from baseline on, as RECIST 1.1 specifies; no later
    assessment enters the reference. An incomplete assessment (fewer target
    lesions measured than at baseline) cannot set the reference but can show
    progression. Only assessments on or after the origin can be events.

    new_lesions  "event": a new lesion on or after the origin is progression;
                 "censor": follow-up ends, censored, on the day a new lesion
                 appears (the sum-only definition).
    drop_lowest  remove the patient's lowest complete sum before scoring, so
                 that no single low reading sets the reference.
    Returns {"duration", "event", "source"}, or None without an assessment
    on or after the origin.
    """
    block = block.sort_values("day")
    if drop_lowest and block.complete.any():
        block = block.drop(index=block.loc[block.complete, "sld"].idxmin())
    forward = block.loc[block.day >= origin_day]
    if forward.empty:
        return None
    reference, sld_event_day = np.inf, np.nan
    for day, sld, complete in block[["day", "sld", "complete"]].itertuples(index=False):
        if day >= origin_day and crosses_progression(sld, reference):
            sld_event_day = float(day)
            break
        if complete:
            reference = min(reference, sld)
    last_seen = float(forward.day.max())
    lesion_day = (float(new_lesion_day) if pd.notna(new_lesion_day) and new_lesion_day >= origin_day
                  else np.nan)
    if new_lesions == "censor":
        if np.isfinite(sld_event_day) and not (np.isfinite(lesion_day) and lesion_day < sld_event_day):
            return {"duration": sld_event_day - origin_day, "event": True, "source": "SLD"}
        end = min(lesion_day, last_seen) if np.isfinite(lesion_day) else last_seen
        return {"duration": end - origin_day, "event": False, "source": None}
    candidates = {source: day for source, day in (("new lesion", lesion_day), ("SLD", sld_event_day))
                  if np.isfinite(day)}
    if candidates:
        source = min(candidates, key=candidates.get)
        return {"duration": candidates[source] - origin_day, "event": True, "source": source}
    return {"duration": last_seen - origin_day, "event": False, "source": None}


def burden_features(block, cutoff_day, inclusive=False):
    """Depth, final-interval slope of log burden, and an indicator that burden
    rose over the final interval, from complete assessments before the cutoff
    (on or before it when `inclusive`). None with fewer than two."""
    window = block.loc[(block.day <= cutoff_day) if inclusive else (block.day < cutoff_day)]
    window = window.loc[window.complete].sort_values("day")
    if len(window) < 2 or window.sld.iloc[0] <= 0:
        return None
    baseline, last, previous = (float(window.sld.iloc[0]), float(window.sld.iloc[-1]),
                                float(window.sld.iloc[-2]))
    span = max(float(window.day.iloc[-1] - window.day.iloc[-2]), 1.0)
    return {
        "depth_at_landmark": 1.0 - float(window.sld.min()) / baseline,
        "slope_to_landmark": (np.log(max(last, 1.0)) - np.log(max(previous, 1.0))) / span,
        "already_rising": float(last > previous),
        "log_baseline_sld": float(np.log(max(baseline, 1.0))),
        "n_assessments": len(window),
        "last_feature_day": float(window.day.iloc[-1]),
        "slope_window_days": span,
    }


def _outcome_columns(outcome, prefix):
    if outcome is None:
        return {f"{prefix}duration": np.nan, f"{prefix}event": False, f"{prefix}source": None}
    return {f"{prefix}{key}": value for key, value in outcome.items()}


def score_patients(panel, new_lesion_days, landmark_day, study, alternative_new_lesion_days=None):
    """Landmark and nadir-clock outcomes and features for every patient of one study.

    `panel` has one row per assessment (patient, day, sld, complete);
    `new_lesion_days` gives each patient's first new-lesion day (absent =
    none). Outcomes from the landmark: the primary (new lesions are events),
    sum only (censored at a new lesion) and with the lowest complete sum
    removed. The nadir clock starts at the lowest complete sum, with features
    from assessments on or before it. With `alternative_new_lesion_days`, the
    primary outcome and the pre-landmark progression check are repeated with
    that definition of a new lesion (prefix `harmonized_`).
    """
    rows = []
    for patient, block in panel.sort_values("day").groupby("patient", sort=True):
        new_lesion_day = new_lesion_days.get(patient, np.nan)
        features = burden_features(block, landmark_day)
        row = {"study": study, "patient": patient, "assessments": len(block),
               "last_day": float(block.day.max()),
               "reached_landmark": bool(block.day.max() >= landmark_day),
               "progressed_before_landmark": progressed_before(block, landmark_day, new_lesion_day),
               "has_features": features is not None, **(features or {})}
        row.update(_outcome_columns(time_to_progression(block, landmark_day, new_lesion_day), ""))
        row.update(_outcome_columns(time_to_progression(block, landmark_day, new_lesion_day,
                                                        new_lesions="censor"), "sld_only_"))
        row.update(_outcome_columns(time_to_progression(block, landmark_day, new_lesion_day,
                                                        drop_lowest=True), "nadir_excluded_"))
        complete = block.loc[block.complete]
        if len(complete):
            nadir_day = float(complete.loc[complete.sld.idxmin(), "day"])
            row["nadir_day"] = nadir_day
            row.update(_outcome_columns(time_to_progression(block, nadir_day, new_lesion_day), "nadir_"))
            nadir_features = burden_features(block, nadir_day, inclusive=True)
            row.update({f"nadir_{key}": value for key, value in (nadir_features or {}).items()})
        if alternative_new_lesion_days is not None:
            alternative = alternative_new_lesion_days.get(patient, np.nan)
            row["harmonized_progressed_before_landmark"] = progressed_before(block, landmark_day, alternative)
            row.update(_outcome_columns(time_to_progression(block, landmark_day, alternative), "harmonized_"))
        rows.append(row)
    return pd.DataFrame(rows)


def harrell_c(durations, events, risk, strata=None):
    """Harrell's concordance index. A pair is comparable when the member with
    the shorter time had an event; tied risks count one half. With `strata`,
    only pairs within the same stratum are compared."""
    durations = np.asarray(durations, float)
    events = np.asarray(events, bool)
    risk = np.asarray(risk, float)
    same = (np.ones((len(durations), len(durations)), bool) if strata is None
            else np.asarray(strata)[:, None] == np.asarray(strata)[None, :])
    comparable = (durations[None, :] > durations[:, None]) & events[:, None] & same
    n_comparable = comparable.sum()
    if n_comparable == 0:
        return np.nan
    concordant = ((risk[:, None] > risk[None, :]) & comparable).sum()
    tied = ((risk[:, None] == risk[None, :]) & comparable).sum()
    return float((concordant + 0.5 * tied) / n_comparable)


def variance_inflation(frame, predictors):
    """Variance inflation factor of each predictor: 1 / (1 − R²) from the
    least-squares regression of that predictor on the others (with an
    intercept). Values above about 10 mean the predictor is close to a
    linear function of the others, so its Cox coefficient is unstable. A
    predictor that is constant in `frame` has no defined factor (NaN)."""
    design = frame[predictors].to_numpy(float)
    factors = {}
    for index, name in enumerate(predictors):
        target = design[:, index]
        total = np.sum((target - target.mean()) ** 2)
        if total <= 0:
            factors[name] = np.nan
            continue
        others = np.column_stack([np.ones(len(design)), np.delete(design, index, axis=1)])
        fitted = others @ np.linalg.lstsq(others, target, rcond=None)[0]
        r2 = 1.0 - np.sum((target - fitted) ** 2) / total
        factors[name] = float(1.0 / max(1.0 - r2, 1e-12))
    return factors


def _standardize(train, test):
    """Training-set centring and scaling; constant columns are left out."""
    centre, scale = train.mean(axis=0), train.std(axis=0)
    keep = scale > 1e-12
    return (train[:, keep] - centre[keep]) / scale[keep], (test[:, keep] - centre[keep]) / scale[keep], keep


def cox_out_of_fold_risk(design, durations, events, fold_labels, seed, cox_strata=None):
    """Cross-validated Cox linear predictors (five folds stratified on
    `fold_labels`). Predictors are standardized with training-fold
    statistics; a predictor constant within a training fold is left out of
    that fold's fit. Returns risk scores, each patient's fold and the
    per-fold coefficients (NaN where a predictor was left out)."""
    risk = np.full(len(durations), np.nan)
    fold_of = np.full(len(durations), -1)
    coefficients = []
    splitter = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=seed)
    for fold, (train, test) in enumerate(splitter.split(design, fold_labels)):
        train_design, test_design, keep = _standardize(design[train], design[test])
        fitted = sm.PHReg(durations[train], train_design, status=events[train].astype(int), ties="efron",
                          strata=None if cox_strata is None else cox_strata[train]).fit()
        params = np.asarray(fitted.params, float)
        risk[test] = test_design @ params
        fold_of[test] = fold
        full = np.full(design.shape[1], np.nan)
        full[keep] = params
        coefficients.append(full)
    return risk, fold_of, np.vstack(coefficients)


def paired_bootstrap_c(durations, events, risk_a, risk_b, resample_strata, c_strata,
                       seed, min_events):
    """Paired bootstrap of C(risk_b) − C(risk_a), resampling patients within
    each stratum of `resample_strata`; resamples with fewer than `min_events`
    events are skipped. Returns the 95% interval, two-sided P and the number
    of resamples used."""
    rng = np.random.default_rng(seed + 7)
    index_by_stratum = [np.flatnonzero(resample_strata == s) for s in np.unique(resample_strata)]
    differences = []
    for _ in range(N_BOOTSTRAP):
        drawn = np.concatenate([rng.choice(idx, size=len(idx), replace=True)
                                for idx in index_by_stratum])
        if events[drawn].sum() < min_events:
            continue
        strata = None if c_strata is None else c_strata[drawn]
        c_a = harrell_c(durations[drawn], events[drawn], risk_a[drawn], strata)
        c_b = harrell_c(durations[drawn], events[drawn], risk_b[drawn], strata)
        if np.isfinite(c_a) and np.isfinite(c_b):
            differences.append(c_b - c_a)
    differences = np.asarray(differences)
    low, high = np.percentile(differences, [2.5, 97.5])
    return float(low), float(high), bootstrap_p_two_sided(differences), len(differences)


def _survival_arrays(frame, duration, event):
    """Durations (a time of 0 or less becomes 0.5 day, so that a patient who
    progressed at the clock start stays in the cohort), events and study codes."""
    durations = frame[duration].to_numpy(float)
    durations = np.where(durations <= 0, 0.5, durations)
    return durations, frame[event].to_numpy(bool), pd.Categorical(frame.study.astype(str)).codes


def compare_cox_models(frame, comparator, trajectory, seed, duration="duration",
                       event="event", stratified=False, label="", min_events=5):
    """Depth-only against depth-plus-shape Cox models on the same patients and
    folds, scored by out-of-fold concordance.

    Folds are stratified on event status and, with more than one study, on
    study; bootstrap resamples are drawn within study. With `stratified`,
    the Cox model has a baseline hazard per study and the concordance index
    compares patients only within the same study. Also returns the
    concordance by fold and by study as checks.
    """
    durations, events, study_codes = _survival_arrays(frame, duration, event)
    several_studies = len(np.unique(study_codes)) > 1
    fold_labels = (np.char.add(events.astype(int).astype(str), study_codes.astype(str))
                   if several_studies else events.astype(int))
    cox_strata = study_codes if stratified else None
    c_strata = study_codes if stratified else None
    risk_a, fold_of, coef_a = cox_out_of_fold_risk(frame[comparator].to_numpy(float), durations,
                                                   events, fold_labels, seed, cox_strata)
    risk_b, _, coef_b = cox_out_of_fold_risk(frame[trajectory].to_numpy(float), durations,
                                             events, fold_labels, seed, cox_strata)
    c_a = harrell_c(durations, events, risk_a, c_strata)
    c_b = harrell_c(durations, events, risk_b, c_strata)
    low, high, p_value, n_draws = paired_bootstrap_c(durations, events, risk_a, risk_b,
                                                     study_codes, c_strata, seed, min_events)
    checks = []
    for kind, labels in (("fold", fold_of), ("study", frame.study.astype(str).to_numpy())):
        for value in np.unique(labels):
            rows = labels == value
            strata = None if c_strata is None else c_strata[rows]
            checks.append({"by": kind, "value": value, "patients": int(rows.sum()),
                           "events": int(events[rows].sum()),
                           "c_depth": harrell_c(durations[rows], events[rows], risk_a[rows], strata),
                           "c_trajectory": harrell_c(durations[rows], events[rows], risk_b[rows], strata)})
    return {"analysis": label, "n": len(frame), "events": int(events.sum()),
            "c_depth": c_a, "c_trajectory": c_b, "difference": c_b - c_a,
            "low": low, "high": high, "p_value": p_value, "bootstrap_draws": n_draws,
            "coefficients_depth": pd.DataFrame(coef_a, columns=comparator),
            "coefficients": pd.DataFrame(coef_b, columns=trajectory), "checks": pd.DataFrame(checks)}


def leave_one_study_out(frame, comparator, trajectory, seed, duration="duration", event="event",
                        label="", min_events=5):
    """Each study's patients scored by Cox models fitted on the other studies.
    Concordance is computed within study and the paired bootstrap resamples
    within study, as for the stratified primary."""
    durations, events, study_codes = _survival_arrays(frame, duration, event)
    risk = {"depth": np.full(len(frame), np.nan), "trajectory": np.full(len(frame), np.nan)}
    by_study = []
    for held_out in np.unique(study_codes):
        train, test = study_codes != held_out, study_codes == held_out
        for name, predictors in (("depth", comparator), ("trajectory", trajectory)):
            design = frame[predictors].to_numpy(float)
            train_design, test_design, _ = _standardize(design[train], design[test])
            fitted = sm.PHReg(durations[train], train_design, status=events[train].astype(int),
                              ties="efron").fit()
            risk[name][test] = test_design @ np.asarray(fitted.params, float)
        by_study.append({"held_out": frame.study.astype(str).to_numpy()[test][0],
                         "patients": int(test.sum()), "events": int(events[test].sum()),
                         "c_depth": harrell_c(durations[test], events[test], risk["depth"][test]),
                         "c_trajectory": harrell_c(durations[test], events[test], risk["trajectory"][test])})
    c_a = harrell_c(durations, events, risk["depth"], study_codes)
    c_b = harrell_c(durations, events, risk["trajectory"], study_codes)
    low, high, p_value, n_draws = paired_bootstrap_c(durations, events, risk["depth"], risk["trajectory"],
                                                     study_codes, study_codes, seed, min_events)
    return {"analysis": label, "n": len(frame), "events": int(events.sum()),
            "c_depth": c_a, "c_trajectory": c_b, "difference": c_b - c_a,
            "low": low, "high": high, "p_value": p_value, "bootstrap_draws": n_draws,
            "checks": pd.DataFrame(by_study)}


def cox_full_fit(frame, predictors, duration="duration", event="event", stratified=False):
    """Cox model on all patients, predictors standardized: coefficient (log
    hazard per SD), standard error, 95% Wald interval and P value."""
    durations, events, study_codes = _survival_arrays(frame, duration, event)
    design = frame[predictors].to_numpy(float)
    standardized, _, keep = _standardize(design, design)
    fitted = sm.PHReg(durations, standardized, status=events.astype(int), ties="efron",
                      strata=study_codes if stratified else None).fit()
    params, errors = np.asarray(fitted.params, float), np.asarray(fitted.bse, float)
    return pd.DataFrame({"predictor": np.asarray(predictors)[keep], "coefficient": params,
                         "se": errors, "low": params - 1.96 * errors, "high": params + 1.96 * errors,
                         "p_value": np.asarray(fitted.pvalues, float)})


def report_comparison(result):
    """Print one depth-only against depth-plus-shape comparison: patients,
    events, both concordance indices and their difference with its 95%
    interval and paired-bootstrap P value."""
    print(f"  {result['analysis']}")
    print(f"    patients {result['n']:,}   events {result['events']:,}")
    print(f"    C, depth {result['c_depth']:.3f}   C, depth plus shape "
          f"{result['c_trajectory']:.3f}   difference "
          f"{format_ci(result['difference'], result['low'], result['high'], 1, 3, '')}"
          f"   {p_label(result['p_value'], 1 / result['bootstrap_draws'])}")
```

## 2 · Data and preprocessing

Serial tumor volumes are read from the supplementary workbook of the Novartis PDX encyclopedia (Gao et al., 2015). Each PDX model received each treatment in a single animal, apart from a small number of arms with replicate readings on the same day, which are counted below. The workbook's SHA-256 checksum is compared with that of the published file before anything else is read. Treatment labels are normalized before arm identifiers are built, because one label in the workbook carries a stray quotation mark and would otherwise fail to join the curve-metric sheet.

Two preprocessing rules are then applied once, to one canonical table that every later section uses. Volumes below 1 mm³ (recorded as zero or below 1 mm³) are set to 1 mm³, and replicate readings of an arm on the same day are averaged on the log scale. An arm is one PDX model–treatment pair. The per-arm endpoint table is built by `pdxe_utils.arm_features` from the readings as recorded, and that function applies the same two rules internally. The source study's own response metrics (best average response, treatment target) are then merged in by arm.

**Inputs:** `data/gao2015/41591_2015_BFnm3954_MOESM10_ESM.xlsx`.
**Outputs:** `raw_observations` (readings as recorded), `observations` (canonical: one row per arm per day), `arms` (one row per arm), `model_sets` (PDX models with each molecular data type).

```python
# ==========================================================================
# 2.1 Source workbook: checksum and conversion of the sheets to Parquet
#     The workbook must be the published file. Sheets are converted to
#     Parquet once and reused while the workbook checksum is unchanged.
# ==========================================================================

set_section("2")
assert WORKBOOK_PATH.exists(), (f"{relative(WORKBOOK_PATH)} not found; place the published workbook "
                                "there (see the project layout at the top of the notebook)")
workbook_sha256 = sha256_of_file(WORKBOOK_PATH)
print("SOURCE WORKBOOK")
print(f"  file      {relative(WORKBOOK_PATH)}")
print(f"  size      {WORKBOOK_PATH.stat().st_size / 2 ** 20:.1f} MiB")
print(f"  SHA-256   {workbook_sha256}")
assert workbook_sha256 == EXPECTED_WORKBOOK_SHA256, "the workbook is not the published file"
print("  matches the published file")

WORKBOOK_SHEETS = {"raw": "PCT raw data", "metrics": "PCT curve metrics",
                   "expression": "RNAseq_fpkm", "mutation": "pdxe_mut_and_cn2"}
COPY_NUMBER_SHEET = "copy number"     # only its header row (PDX model ids) is read
sheet_paths = {key: INTERIM_DIR / f"{key}.parquet" for key in WORKBOOK_SHEETS}
stamp_path = INTERIM_DIR / "workbook_sha256.txt"
cache_valid = (stamp_path.exists() and stamp_path.read_text().strip() == workbook_sha256
               and all(path.exists() for path in sheet_paths.values()))

if not cache_valid:
    for key, sheet in WORKBOOK_SHEETS.items():
        frame = pd.read_excel(WORKBOOK_PATH, sheet_name=sheet, engine="openpyxl")
        # Text columns hold mixed cell types in places; they are stored as text.
        for column in frame.columns[frame.dtypes == object]:
            frame[column] = frame[column].map(lambda v: None if pd.isna(v) else str(v))
        frame.to_parquet(sheet_paths[key], index=False)
    stamp_path.write_text(workbook_sha256)
sheets = {key: pd.read_parquet(path) for key, path in sheet_paths.items()}
print(f"\n  sheets {'read from the Parquet copies' if cache_valid else 'converted to Parquet'}")
for key, sheet in WORKBOOK_SHEETS.items():
    print(f"    {sheet:<20} {sheets[key].shape[0]:>9,} rows x {sheets[key].shape[1]:,} columns")

workbook = openpyxl.load_workbook(WORKBOOK_PATH, read_only=True, data_only=True)
copy_number_header = next(workbook[COPY_NUMBER_SHEET].iter_rows(max_row=1, values_only=True))
workbook.close()

# ==========================================================================
# 2.2 The canonical observation table
#     Columns are mapped by name, counts are asserted, and the floor and
#     same-day averaging are applied once (canonical_observations).
# ==========================================================================

RAW_COLUMNS = {"Model": "model_id", "Tumor Type": "tumor_type", "Treatment": "treatment",
               "Days Post T0": "day", "Volume (mm3)": "volume"}
utils.require_columns(sheets["raw"], list(RAW_COLUMNS), "PCT raw data")
raw_observations = sheets["raw"].rename(columns=RAW_COLUMNS)[list(RAW_COLUMNS.values())].copy()
for column in ("day", "volume"):
    raw_observations[column] = pd.to_numeric(raw_observations[column], errors="coerce")
non_numeric = int(raw_observations[["day", "volume"]].isna().any(axis=1).sum())
assert non_numeric == 0, f"{non_numeric} readings have a non-numeric day or volume"

# Treatment labels are normalized before the arm identifier is built: one
# label in the sheet carries a stray quotation mark.
raw_observations["treatment"] = raw_observations.treatment.map(utils.normalise_treatment)
raw_observations["arm_id"] = (raw_observations.model_id.astype(str) + " || "
                              + raw_observations.treatment.astype(str))
raw_observations["is_control"] = raw_observations.treatment.eq(CONTROL_LABEL)

# Tumor type is blank on a few rows; each PDX model has one tumor type, so a
# blank is filled from the same model's other rows.
missing_type = raw_observations.tumor_type.isna()
type_by_model = (raw_observations.loc[~missing_type].groupby("model_id").tumor_type
                 .agg(lambda values: values.mode().iloc[0]))
assert (raw_observations.loc[~missing_type].groupby("model_id").tumor_type.nunique() == 1).all()
filled_arms = sorted(raw_observations.loc[missing_type, "arm_id"].unique())
raw_observations["tumor_type"] = raw_observations.tumor_type.fillna(
    raw_observations.model_id.map(type_by_model))
assert raw_observations.tumor_type.notna().all(), "a PDX model has no recorded tumor type"
raw_observations = raw_observations.sort_values(["arm_id", "day"], kind="stable").reset_index(drop=True)

n_nonpositive = int((raw_observations.volume <= 0).sum())
n_below_floor = int(((raw_observations.volume > 0) & (raw_observations.volume < VOLUME_FLOOR)).sum())
floored_arms = raw_observations.loc[raw_observations.volume < VOLUME_FLOOR, "arm_id"].nunique()

observations = canonical_observations(raw_observations)
replicate_days = observations.loc[observations.n_readings > 1]
replicate_arms = replicate_days.groupby("arm_id").agg(
    days_with_replicates=("day", "size"), max_readings=("n_readings", "max"))

treatments = sorted(raw_observations.treatment.unique())
regimens = [t for t in treatments if t != CONTROL_LABEL]
combination_regimens = [t for t in regimens if "+" in t]

print("\nCANONICAL OBSERVATION TABLE")
print(f"  readings as recorded                 {record('n_readings', len(raw_observations)):,}")
print(f"  PDX models                           {record('n_models', raw_observations.model_id.nunique()):,}")
print(f"  tumor types                          {record('n_tumor_types', raw_observations.tumor_type.nunique())}"
      f"   {sorted(raw_observations.tumor_type.unique())}")
print(f"  arms                                 {record('n_arms', raw_observations.arm_id.nunique()):,}")
print(f"  untreated arms                       {record('n_untreated_arms', raw_observations.loc[raw_observations.is_control, 'arm_id'].nunique()):,}")
print(f"  treatment regimens                   {record('n_regimens', len(regimens))}"
      f"  ({record('n_single_agent_regimens', len(regimens) - len(combination_regimens))} single-agent, "
      f"{record('n_combination_regimens', len(combination_regimens))} combination)")
print(f"  rows with a blank tumor type         {int(missing_type.sum())}, in arm(s) {filled_arms}; filled from the PDX model")
print(f"  readings at or below 0 mm³           {n_nonpositive:,}")
print(f"  readings between 0 and 1 mm³         {n_below_floor:,}")
print(f"  arms with a reading set to the floor {floored_arms:,}")
print(f"  rows after same-day averaging        {len(observations):,}")
print(f"  arms with replicate same-day readings {record('n_arms_with_replicate_readings', len(replicate_arms)):,} "
      f"({int(replicate_arms.days_with_replicates.sum())} arm-days, up to "
      f"{int(replicate_arms.max_readings.max()) if len(replicate_arms) else 0} readings in a day)")
if len(replicate_arms):
    print(replicate_arms.sort_values("days_with_replicates", ascending=False).head(10).to_string())

assert len(raw_observations) == 67_445
assert raw_observations.arm_id.nunique() == 4_706
assert raw_observations.model_id.nunique() == 277
assert raw_observations.loc[raw_observations.is_control, "arm_id"].nunique() == 223
assert observations.groupby(["arm_id", "day"]).size().max() == 1
# The canonical table, for the figures notebook (example trajectories).
save_record(observations, "observations_canonical")

# ==========================================================================
# 2.3 The per-arm endpoint table and the source study's response metrics
#     pdxe_utils.arm_features receives each arm's readings as recorded (it
#     applies the floor and same-day averaging itself) and returns nothing
#     for arms that cannot support its features; those arms are listed.
# ==========================================================================

utils_input = raw_observations.rename(columns={"tumor_type": "indication"})[
    ["model_id", "indication", "treatment", "day", "volume", "arm_id", "is_control"]]
feature_rows, arms_without_features = [], []
for arm_id, block in utils_input.groupby("arm_id", sort=False):
    row = utils.arm_features(block)
    if row is None:
        arms_without_features.append(arm_id)
    else:
        feature_rows.append(row)
# pdxe_utils uses compact column names; the notebook uses spelled-out ones,
# renamed once here so every later section reads one convention.
UTILITY_COLUMN_NAMES = {
    "indication": "tumor_type",
    "BAR_self": "response_reimplemented",           # best average response, recomputed
    "g_late": "post_nadir_growth_rate",              # durability: slope of log volume after the nadir
    "post_n": "n_after_nadir",
    "n_obs": "n_measurements",
    "duration": "duration_days",
    "baseline": "baseline_volume",
    "logV_baseline": "baseline_log_volume",
    "nadir_depth_log": "nadir_depth",
    "durable_estimable": "durability_estimable",
}
arms = pd.DataFrame(feature_rows).rename(columns=UTILITY_COLUMN_NAMES)
arms["treatment"] = arms.treatment.map(utils.normalise_treatment)
arms["is_control"] = arms.treatment.eq(CONTROL_LABEL)
utils.require_columns(arms, ["arm_id", "model_id", "tumor_type", "treatment",
                             "post_nadir_growth_rate", "interior_nadir", "n_after_nadir",
                             "nadir_day", "baseline_volume", "response_reimplemented",
                             "n_measurements", "duration_days"], "arm feature table")

print("\nARM FEATURE TABLE")
print(f"  arms with features                   {len(arms):,}")
print(f"  arms without features                {len(arms_without_features)}")
if arms_without_features:
    print(observations.loc[observations.arm_id.isin(arms_without_features)]
          .groupby("arm_id").agg(readings=("day", "size"), first_day=("day", "min"),
                                 last_day=("day", "max")).to_string())

METRIC_COLUMNS = {"BestAvgResponse": "best_average_response", "BestResponse": "best_response",
                  "Day_BestAvgResponse": "day_best_average_response",
                  "TimeToDouble": "time_to_double_published",
                  "ResponseCategory": "response_category", "Day_Last": "day_last",
                  "Treatment target": "treatment_target", "Treatment type": "treatment_type"}
metrics = sheets["metrics"].rename(columns={"Model": "model_id", "Treatment": "treatment"})
utils.require_columns(metrics, list(METRIC_COLUMNS), "PCT curve metrics")
metrics["treatment"] = metrics.treatment.map(utils.normalise_treatment)
metrics["arm_id"] = metrics.model_id.astype(str) + " || " + metrics.treatment.astype(str)
metrics = metrics.rename(columns=METRIC_COLUMNS)[["arm_id", *METRIC_COLUMNS.values()]]
duplicated_rows = metrics.duplicated("arm_id", keep=False)
disagreeing = metrics.loc[duplicated_rows].groupby("arm_id").nunique(dropna=False).gt(1).any(axis=1)
print(f"\n  curve-metric rows                    {len(metrics):,} for {metrics.arm_id.nunique():,} arms")
print(f"  arms listed more than once           {int(duplicated_rows.groupby(metrics.arm_id).any().sum()):,}"
      f" ({int(disagreeing.sum())} with differing values; the first row is used)")
arms = arms.merge(metrics.drop_duplicates("arm_id"), on="arm_id", how="left")
print(f"  arms matched to a published metric   {int(arms.best_average_response.notna().sum()):,} of {len(arms):,}")

both = arms.dropna(subset=["response_reimplemented", "best_average_response"])
agreement, _ = spearman(both.response_reimplemented, both.best_average_response)
print(f"  best average response, reimplemented against published: "
      f"Spearman {agreement:.3f} (n = {len(both):,}); the published value is used throughout")

# PDX models carrying each molecular data type (expression and copy number
# hold one column per model; the mutation sheet holds one row per finding).
known_models = set(raw_observations.model_id.astype(str))
model_sets = {
    "expression": known_models & set(map(str, sheets["expression"].columns[1:])),
    "copy number": known_models & {str(value) for value in copy_number_header[1:] if value is not None},
    "mutation": known_models & set(sheets["mutation"].iloc[:, 0].astype(str)),
}
```

## 3 · Design characteristics and attrition

This section describes what the design of the source study can support before any model is fitted. It covers the measurement schedules (which the simulations in Sections 5 and 12 reuse), arms with replicate same-day readings, and how many PDX models have an untreated arm. It also covers how the PDX model × treatment matrix is connected, how often treated arms show the durability phenotype, three signatures of termination at large volumes, and which PDX models have molecular data. It ends by fitting a line to log volume in every untreated arm with at least five measurements. The slope is that PDX model's untreated growth rate and the median residual SD is the measurement noise. Both are used from Section 4 onward and reported in Section 6.

**Inputs:** `observations`, `arms`, `model_sets`.
**Outputs:** `arm_design` (per-arm schedule summary), `empirical_grids` (measurement days of arms with ≥5 measurements), `regrowth` (model-free regrowth summary of treated arms), `untreated_fits`, `untreated_growth` (per PDX model), `MEASUREMENT_NOISE`; figure data `arm_design.csv` and `duration_quartiles.csv`.

```python
# ==========================================================================
# 3.1 Measurement schedules
#     The measurement days of every arm with at least five measurements are
#     kept as `empirical_grids`; the simulations draw schedules from them.
# ==========================================================================

set_section("3")
by_arm = observations.groupby("arm_id", sort=True)
arm_design = pd.DataFrame({
    "model_id": by_arm.model_id.first(), "tumor_type": by_arm.tumor_type.first(),
    "treatment": by_arm.treatment.first(), "is_control": by_arm.is_control.first(),
    "n_measurements": by_arm.size(), "first_day": by_arm.day.min(),
    "last_day": by_arm.day.max(), "last_volume": by_arm.volume.last(),
    "max_volume": by_arm.volume.max(),
    "median_gap_days": by_arm.day.agg(lambda d: float(np.median(np.diff(d))) if len(d) > 1 else np.nan),
})
arm_design["duration_days"] = arm_design.last_day - arm_design.first_day

print("3.1 MEASUREMENT SCHEDULES (canonical table, one measurement per arm per day)")
print(arm_design[["n_measurements", "duration_days", "median_gap_days"]]
      .describe(percentiles=[.05, .25, .5, .75, .95]).round(1).to_string())
print("\n  arms with at least k measurements")
for k in (3, 4, 5, 6, 8, 10):
    count = int((arm_design.n_measurements >= k).sum())
    print(f"    k = {k:>2}   {count:>5,}  ({100 * count / len(arm_design):.1f}%)")
print("\n  duration (days) by tumor type")
print(arm_design.groupby("tumor_type").duration_days
      .describe()[["count", "25%", "50%", "75%", "max"]].round(0).to_string())

empirical_grids = [block.day.to_numpy(float) for _, block in arm_series(observations)
                   if len(block) >= MIN_MEASUREMENTS_PER_ARM]
share_reaching_84 = float(np.mean([grid.max() >= 84 for grid in empirical_grids]))
print(f"\n  measurement schedules kept for simulation      {len(empirical_grids):,}")
print(f"  of those extending to day 84                   "
      f"{100 * record('share_schedules_to_day_84', share_reaching_84):.1f}%")

# ==========================================================================
# 3.2 Untreated-arm coverage and connectivity of the design
#     A treated arm can be paired with an untreated growth rate only when its
#     PDX model has an untreated arm. Connectivity is the number of connected
#     components of the bipartite PDX model–treatment graph.
# ==========================================================================

models_with_untreated = set(arm_design.loc[arm_design.is_control, "model_id"])
treated_design = arm_design.loc[~arm_design.is_control]
treated_without_untreated = int((~treated_design.model_id.isin(models_with_untreated)).sum())
print("\n3.2 UNTREATED-ARM COVERAGE AND CONNECTIVITY")
print(f"  PDX models with an untreated arm                {len(models_with_untreated)} of "
      f"{arm_design.model_id.nunique()}")
print(f"  treated arms in PDX models without one          {treated_without_untreated:,} of "
      f"{len(treated_design):,} ({100 * treated_without_untreated / len(treated_design):.1f}%)")

treatments_per_model = treated_design.groupby("model_id").treatment.nunique()
print(f"  treatments per PDX model, median [range]        {treatments_per_model.median():.0f} "
      f"[{treatments_per_model.min()}, {treatments_per_model.max()}]")
print(f"  PDX models with at least two treatments         {int((treatments_per_model >= 2).sum())}")
# Categorical codes are stored in the smallest integer type that holds them
# (int8 below 128 categories); they are widened before the node indices of
# the two node types are added, so the sum cannot overflow.
model_codes = pd.Categorical(treated_design.model_id).codes.astype(np.int64)
treatment_codes = pd.Categorical(treated_design.treatment).codes.astype(np.int64)
n_model_nodes = model_codes.max() + 1
graph = coo_matrix((np.ones(len(treated_design)), (model_codes, n_model_nodes + treatment_codes)),
                   shape=(n_model_nodes + treatment_codes.max() + 1,) * 2)
n_components, _ = connected_components(graph, directed=False)
print(f"  connected components of the design              {n_components}")
save_record(pd.crosstab(treated_design.tumor_type, treated_design.treatment),
            "tumor_type_by_treatment", index=True)

# ==========================================================================
# 3.3 Regrowth base rate (treated arms, canonical table)
#     The durability phenotype: shrank by at least 1%, an interior nadir with
#     at least two later measurements, and regrowth to at least 1.2 times the
#     nadir. The thresholds are varied in the grid below.
# ==========================================================================

regrowth_rows = []
for arm_id, block in arm_series(observations.loc[~observations.is_control]):
    summary = summarize_regrowth(block.day.to_numpy(float), block.log_volume.to_numpy(float))
    if summary is not None:
        regrowth_rows.append({"arm_id": arm_id, "tumor_type": block.tumor_type.iloc[0], **summary})
regrowth = pd.DataFrame(regrowth_rows).set_index("arm_id")
shrank = regrowth.nadir_ratio <= 1 - REGRESSION_FRACTION
interior = (regrowth.nadir_index > 0) & (regrowth.n_after_nadir >= MIN_POINTS_AFTER_NADIR)
regrew = regrowth.regrowth_multiple >= REGROWTH_MULTIPLE
regrowth["durability_phenotype"] = shrank & interior & regrew

print("\n3.3 REGROWTH BASE RATE (treated arms with at least three measurements)")
print(f"  arms                                            {len(regrowth):,}")
for label, mask in [("shrank by at least 1%", shrank),
                    ("interior nadir, >= 2 later measurements", interior),
                    ("regrew to >= 1.2 times the nadir", regrew),
                    ("all three (durability phenotype)", regrowth.durability_phenotype)]:
    print(f"  {label:<47} {int(mask.sum()):>5,}  ({100 * mask.mean():.1f}%)")
record("treated_phenotype_share", float(regrowth.durability_phenotype.mean()))
record("treated_interior_nadir_share", float(interior.mean()))

grid_rows = []
for regression in (0.0, 0.05, 0.10, 0.20, 0.30):
    for multiple in (1.10, 1.20, 1.50, 2.00):
        share = ((regrowth.nadir_ratio <= 1 - regression) & interior
                 & (regrowth.regrowth_multiple >= multiple)).mean()
        grid_rows.append({"minimum regression": regression, "minimum regrowth multiple": multiple,
                          "percent of arms": round(100 * share, 1)})
print("\n  percent of arms with the phenotype, by threshold")
print(pd.DataFrame(grid_rows).pivot(index="minimum regression",
                                    columns="minimum regrowth multiple",
                                    values="percent of arms").to_string())
print("\n  arms with the phenotype, by tumor type")
print(regrowth.groupby("tumor_type").durability_phenotype
      .agg(arms="size", with_phenotype="sum", share="mean").round(3).to_string())

# ==========================================================================
# 3.4 Termination at large volumes
#     Three signatures: arms ending at their largest volume, shorter arms in
#     faster-growing tumors, and larger final volumes in shorter arms.
# ==========================================================================

overall_rate = {}
for arm_id, block in arm_series(observations):
    if len(block) >= 3:
        overall_rate[arm_id] = fit_line(block.day.to_numpy(float) - block.day.iloc[0],
                                        block.log_volume.to_numpy(float))[0]
arm_design["overall_growth_rate"] = pd.Series(overall_rate)
complete_design = arm_design.dropna(subset=["overall_growth_rate"]).copy()
duration_growth_r = float(np.corrcoef(complete_design.duration_days,
                                      complete_design.overall_growth_rate)[0, 1])
complete_design["duration_quartile"] = pd.qcut(complete_design.duration_days, 4,
                                               labels=["Q1", "Q2", "Q3", "Q4"])
quartile_summary = complete_design.groupby("duration_quartile", observed=True).agg(
    days=("duration_days", lambda d: f"{d.min():.0f}-{d.max():.0f}"),
    mean_last_volume=("last_volume", "mean"), mean_growth_rate=("overall_growth_rate", "mean"),
    mean_measurements=("n_measurements", "mean"))

print("\n3.4 TERMINATION")
print(f"  arms whose last volume is their largest         "
      f"{100 * (arm_design.last_volume >= arm_design.max_volume).mean():.1f}%")
print("  final volume (mm³), arms per band")
for low, high in [(800, 1200), (1200, 1600), (1600, 2000), (2000, 2400), (2400, 3200)]:
    print(f"    {low:>5}-{high:<5} {int(arm_design.last_volume.between(low, high, inclusive='left').sum()):>6,}")
print(f"  Pearson r, duration against whole-arm growth rate  {duration_growth_r:.3f} "
      f"(n = {len(complete_design):,})")
print("  by duration quartile")
print(quartile_summary.round(4).to_string())
print("  untreated against treated arms")
print(arm_design.groupby("is_control")[["duration_days", "n_measurements", "last_volume"]]
      .mean().rename(index={False: "treated", True: "untreated"}).round(1).to_string())

# Data for the design and termination figure (drawn by the figures
# notebook): the per-arm schedule summary, with each arm's whole-arm growth
# rate, and the duration-quartile summary.
save_record(arm_design.reset_index(), "arm_design")
save_record(quartile_summary.reset_index(), "duration_quartiles")

# ==========================================================================
# 3.5 Molecular data coverage
# ==========================================================================

treated_models = set(treated_design.model_id.astype(str))
print("\n3.5 MOLECULAR DATA (PDX models in the response data)")
combinations_of_types = {
    "expression": model_sets["expression"], "copy number": model_sets["copy number"],
    "mutation": model_sets["mutation"],
    "all three": model_sets["expression"] & model_sets["copy number"] & model_sets["mutation"],
}
for label, model_ids in combinations_of_types.items():
    arms_covered = int(treated_design.model_id.astype(str).isin(model_ids).sum())
    print(f"  {label:<12} {len(model_ids):>4} PDX models   {arms_covered:>5,} treated arms")
expression_types = (arm_design.loc[arm_design.model_id.astype(str).isin(model_sets["expression"])]
                    .groupby("tumor_type").model_id.nunique())
print("  PDX models with expression data, by tumor type")
print(expression_types.reindex(sorted(arm_design.tumor_type.unique()), fill_value=0).to_string())

# ==========================================================================
# 3.6 Untreated growth and measurement noise
#     A least-squares line on log volume in each untreated arm with at least
#     five measurements. Slope: the PDX model's untreated growth rate.
#     Residual SD (two parameters removed): that arm's measurement noise.
#     A quadratic fit is recorded for the growth-law comparison in Section 6.
# ==========================================================================

untreated_rows = []
for arm_id, block in arm_series(observations.loc[observations.is_control]):
    if len(block) < MIN_MEASUREMENTS_PER_ARM:
        continue
    day = block.day.to_numpy(float) - block.day.iloc[0]
    log_volume = block.log_volume.to_numpy(float)
    slope, _, residual_sd = fit_line(day, log_volume)
    quadratic = np.polyfit(day, log_volume, 2)
    quadratic_residual = log_volume - np.polyval(quadratic, day)
    untreated_rows.append({
        "arm_id": arm_id, "model_id": block.model_id.iloc[0],
        "tumor_type": block.tumor_type.iloc[0], "n_measurements": len(block),
        "growth_rate": slope, "residual_sd": residual_sd,
        "quadratic_residual_sd": float(np.sqrt(np.sum(quadratic_residual ** 2) / (len(day) - 3))),
        "curvature": float(quadratic[0]),
    })
untreated_fits = pd.DataFrame(untreated_rows)
assert untreated_fits.model_id.is_unique, "a PDX model has more than one untreated arm"
untreated_growth = untreated_fits.set_index("model_id").growth_rate
MEASUREMENT_NOISE = record("measurement_noise", float(untreated_fits.residual_sd.median()))
print("\n3.6 UNTREATED GROWTH AND MEASUREMENT NOISE")
print(f"  untreated arms with >= {MIN_MEASUREMENTS_PER_ARM} measurements          {len(untreated_fits)}")
print(f"  measurement noise (median residual SD)          {MEASUREMENT_NOISE:.4f} log units "
      f"({np.exp(MEASUREMENT_NOISE):.3f}-fold)")
print(f"  untreated growth rate, median [5th, 95th]       {untreated_growth.median():.4f} "
      f"[{untreated_growth.quantile(.05):.4f}, {untreated_growth.quantile(.95):.4f}] per day")
```

## 4 · Endpoints

Depth is the best average response published by the source study: from day 10 onward, the lowest value of the running average of percentage change in volume from baseline. Durability is the post-nadir growth rate, the least-squares slope of log volume from the nadir (the smallest observed volume) onward. When the nadir is the first measurement this slope is the growth rate of the whole trajectory, so durability analyses use **post-nadir arms**: treated arms whose nadir comes after the first measurement and is followed by at least two measurements. Depth and durability are compared by Spearman correlation. The sensitivity analyses vary the minimum arm duration and the observation window, split by tumor type and by treatment, and adjust for the untreated growth rate of each PDX model. Treatments are then compared in pairs within each PDX model. A pair is depth-matched when best average response differs by less than 5 percentage points, and durability-discordant when post-nadir growth rates differ by at least 0.02 per day. Both thresholds are varied. Post-nadir arms whose post-nadir slope is not positive are left out of the pair analysis, because their nadir is not followed by regrowth.

All quantities here use the complete trajectory, so this section is descriptive. Forecasting is in Section 7.

**Inputs:** `arms`, `observations`, `untreated_growth`.
**Outputs:** `treated_with_endpoints`, `post_nadir_arms`, `depth_durability_sensitivity` (table `depth_durability_sensitivity.csv`), `model_pairs`; figure data `arm_endpoints.csv` and `treatment_pairs.csv`.

```python
# ==========================================================================
# 4.1 Analysis sets and attrition
# ==========================================================================

set_section("4")
treated_arms = arms.loc[~arms.is_control].copy()
treated_with_endpoints = treated_arms.loc[treated_arms.best_average_response.notna()
                                          & treated_arms.post_nadir_growth_rate.notna()].copy()
post_nadir_arms = treated_with_endpoints.loc[
    treated_with_endpoints.interior_nadir.astype(bool)
    & (treated_with_endpoints.n_after_nadir >= MIN_POINTS_AFTER_NADIR)].copy()
post_nadir_ids = set(post_nadir_arms.arm_id)

attrition = pd.DataFrame([
    ("arms", observations.arm_id.nunique()),
    ("arms with an endpoint table entry", len(arms)),
    ("treated arms", len(treated_arms)),
    ("treated arms with both endpoints", len(treated_with_endpoints)),
    ("post-nadir arms", record("n_post_nadir_arms", len(post_nadir_arms))),
], columns=["stage", "arms"])
print("4.1 ATTRITION TO THE DURABILITY ANALYSIS SET")
print(attrition.to_string(index=False))
print(f"  PDX models with post-nadir arms                 {post_nadir_arms.model_id.nunique()}")

# ==========================================================================
# 4.2 Depth against durability
# ==========================================================================

rho, rho_p = spearman(post_nadir_arms.best_average_response, post_nadir_arms.post_nadir_growth_rate)
pearson_r, pearson_p = stats.pearsonr(post_nadir_arms.best_average_response,
                                      post_nadir_arms.post_nadir_growth_rate)
rho_all, rho_all_p = spearman(treated_with_endpoints.best_average_response,
                              treated_with_endpoints.post_nadir_growth_rate)
record("depth_durability_rho", rho)
record("depth_durability_p", rho_p)
record("depth_durability_shared_variance", rho ** 2)
print("\n4.2 DEPTH AGAINST DURABILITY")
print(f"  post-nadir arms (n = {len(post_nadir_arms):,})")
print(f"    Spearman ρ   {rho:.3f}   {p_label(rho_p)}   shared rank variance {100 * rho ** 2:.0f}%")
print(f"    Pearson r    {pearson_r:.3f}   {p_label(pearson_p)}")
print(f"  all treated arms with both endpoints (n = {len(treated_with_endpoints):,}), including arms "
      f"whose nadir is the first measurement")
print(f"    Spearman ρ   {rho_all:.3f}   {p_label(rho_all_p)}")

# ==========================================================================
# 4.3 Sensitivity analyses
# ==========================================================================

DEPTH, DURABILITY = "best_average_response", "post_nadir_growth_rate"
sensitivity_rows = []
for minimum_days in (0, 21, 28, 35, 42, 56):
    subset = post_nadir_arms.loc[post_nadir_arms.duration_days >= minimum_days]
    sensitivity_rows.append(correlation_row("minimum arm duration", f">= {minimum_days} days",
                                            subset[DEPTH], subset[DURABILITY]))
record("rho_duration_56", sensitivity_rows[-1]["spearman_rho"])
record("n_duration_56", sensitivity_rows[-1]["n_arms"])
record("p_duration_56", sensitivity_rows[-1]["p_value"])

# The same two endpoints recomputed with every arm truncated at a common day.
for window_days in (14, 21, 28, 35):
    window_rows = []
    for arm_id, block in arm_series(observations.loc[~observations.is_control]):
        block = block.loc[block.day <= window_days]
        summary = summarize_regrowth(block.day.to_numpy(float), block.log_volume.to_numpy(float))
        if (summary is None or summary["nadir_index"] == 0
                or summary["n_after_nadir"] < MIN_POINTS_AFTER_NADIR):
            continue
        response = utils.best_average_response(block.day.to_numpy(float), block.volume.to_numpy(float),
                                               min_day=MIN_DAY_FOR_BEST_AVERAGE_RESPONSE)
        if np.isfinite(response):
            window_rows.append((response, summary["post_nadir_growth_rate"]))
    window_values = np.array(window_rows, dtype=float).reshape(-1, 2)
    sensitivity_rows.append(correlation_row("common observation window", f"first {window_days} days",
                                            window_values[:, 0], window_values[:, 1]))

for tumor_type, subset in post_nadir_arms.groupby("tumor_type"):
    sensitivity_rows.append(correlation_row("tumor type", tumor_type, subset[DEPTH], subset[DURABILITY]))
for treatment, subset in post_nadir_arms.groupby("treatment"):
    if len(subset) >= 25:
        sensitivity_rows.append(correlation_row("treatment (>= 25 arms)", treatment,
                                                subset[DEPTH], subset[DURABILITY]))

# Adjustment for the PDX model's untreated growth rate: rank-partial
# correlation, with the unadjusted correlation on the same arms beside it.
adjusted = post_nadir_arms.assign(untreated_growth=post_nadir_arms.model_id.map(untreated_growth)
                                  ).dropna(subset=["untreated_growth"])
sensitivity_rows.append(correlation_row(
    "untreated growth rate", "arms in PDX models with an untreated fit, unadjusted",
    adjusted[DEPTH], adjusted[DURABILITY]))
unadjusted_rho = sensitivity_rows[-1]["spearman_rho"]
ranks = adjusted[["best_average_response", "post_nadir_growth_rate", "untreated_growth"]].rank()
design = np.column_stack([np.ones(len(ranks)), ranks.untreated_growth])
residuals = [ranks[c] - design @ np.linalg.lstsq(design, ranks[c], rcond=None)[0]
             for c in ("best_average_response", "post_nadir_growth_rate")]
partial_rho = float(np.corrcoef(*residuals)[0, 1])
partial_df = len(adjusted) - 3
partial_p = float(2 * stats.t.sf(abs(partial_rho) * np.sqrt(partial_df / (1 - partial_rho ** 2)), partial_df))
sensitivity_rows.append({"analysis": "untreated growth rate",
                         "subset": "same arms, partial correlation adjusted for untreated growth",
                         "n_arms": len(adjusted), "spearman_rho": partial_rho, "p_value": partial_p})
record("partial_rho_untreated_growth", partial_rho)
record("unadjusted_rho_untreated_growth_subset", unadjusted_rho)

depth_durability_sensitivity = pd.DataFrame(sensitivity_rows)
treatment_rows = depth_durability_sensitivity.analysis.str.startswith("treatment")
depth_durability_sensitivity.loc[treatment_rows, "q_value"] = benjamini_hochberg(
    depth_durability_sensitivity.loc[treatment_rows, "p_value"])
by_type = depth_durability_sensitivity.loc[depth_durability_sensitivity.analysis == "tumor type"]
record("rho_tumor_type_min", by_type.spearman_rho.min())
record("rho_tumor_type_max", by_type.spearman_rho.max())

print("\n4.3 SENSITIVITY ANALYSES")
print(depth_durability_sensitivity.assign(
    spearman_rho=lambda d: d.spearman_rho.round(3),
    p_value=lambda d: d.p_value.map(format_p),
    q_value=lambda d: d.q_value.map(lambda q: format_p(q) if pd.notna(q) else "")
).to_string(index=False))
print(f"\n  adjustment for untreated growth: {len(adjusted):,} arms in "
      f"{adjusted.model_id.nunique()} PDX models; ρ {unadjusted_rho:.3f} unadjusted, "
      f"{partial_rho:.3f} adjusted ({p_label(partial_p)}, {partial_df} df)")

# ==========================================================================
# 4.4 Treatment pairs within PDX models
# ==========================================================================

RESPONSE_TOLERANCE = 5.0     # percentage points of best average response
GROWTH_RATE_GAP = 0.02       # per day
PAIR_HORIZON_DAYS = 28       # days over which a rate gap is expressed as a fold change

positive_slope = post_nadir_arms.post_nadir_growth_rate > 0
pair_arms = post_nadir_arms.loc[positive_slope]
pair_rows = []
for model_id, block in pair_arms.groupby("model_id", sort=True):
    response = block.best_average_response.to_numpy(float)
    growth = block.post_nadir_growth_rate.to_numpy(float)
    for i, j in combinations(range(len(block)), 2):
        pair_rows.append({"model_id": model_id,
                          "response_gap": abs(response[i] - response[j]),
                          "growth_rate_gap": abs(growth[i] - growth[j])})
model_pairs = pd.DataFrame(pair_rows)
depth_matched = model_pairs.response_gap < RESPONSE_TOLERANCE
discordant = depth_matched & (model_pairs.growth_rate_gap >= GROWTH_RATE_GAP)
discordant_share = discordant.sum() / depth_matched.sum()
pair_rows_by_model = rows_by_cluster(model_pairs.model_id)
matched_values, discordant_values = depth_matched.to_numpy(), discordant.to_numpy()
pair_draws = cluster_bootstrap(
    pair_rows_by_model,
    lambda rows: discordant_values[rows].sum() / max(matched_values[rows].sum(), 1),
    N_BOOTSTRAP, np.random.default_rng(SEEDS["pair_bootstrap"]))
pair_low, pair_high = np.percentile(pair_draws, [2.5, 97.5])

record("n_depth_matched_pairs", int(depth_matched.sum()))
record("n_discordant_pairs", int(discordant.sum()))
record("discordant_share", float(discordant_share))
record("fold_change_28_days", float(np.exp(GROWTH_RATE_GAP * PAIR_HORIZON_DAYS)))
print("\n4.4 TREATMENT PAIRS WITHIN PDX MODELS")
print(f"  post-nadir arms with a non-positive post-nadir slope, left out   "
      f"{int((~positive_slope).sum()):,} ({100 * (~positive_slope).mean():.1f}%)")
print(f"  arms in the pair analysis  {len(pair_arms):,} in {pair_arms.model_id.nunique()} PDX models")
print(f"  pairs                      {len(model_pairs):,}")
print(f"  depth-matched pairs        {int(depth_matched.sum()):,}")
print(f"  of which discordant        {int(discordant.sum()):,}  "
      f"{format_ci(discordant_share, pair_low, pair_high)}, interval resampling PDX models")
print(f"  a rate gap of {GROWTH_RATE_GAP} per day is a {np.exp(GROWTH_RATE_GAP * PAIR_HORIZON_DAYS):.2f}-fold "
      f"volume difference over {PAIR_HORIZON_DAYS} days after the nadir")

sweep_rows = []
for tolerance in (2.5, 5.0, 10.0):
    for gap in (0.01, 0.02, 0.03, 0.05):
        matched = model_pairs.response_gap < tolerance
        hits = matched & (model_pairs.growth_rate_gap >= gap)
        sweep_rows.append({"response tolerance (points)": tolerance, "rate gap (per day)": gap,
                           "fold change over 28 days": round(float(np.exp(gap * PAIR_HORIZON_DAYS)), 2),
                           "depth-matched pairs": int(matched.sum()),
                           "discordant pairs": int(hits.sum()),
                           "share of depth-matched": round(hits.sum() / max(matched.sum(), 1), 3)})
pair_sweep = pd.DataFrame(sweep_rows)
print("\n  threshold sweep")
print(pair_sweep.to_string(index=False))
save_record(depth_durability_sensitivity, "depth_durability_sensitivity")
save_record(pair_sweep, "pair_threshold_sweep")

# Data for the depth and durability figure (drawn by the figures notebook):
# every arm's endpoints with its analysis-set membership, and every
# within-model treatment pair of 4.4.
save_record(arms.assign(treated_with_endpoints=arms.arm_id.isin(treated_with_endpoints.arm_id),
                        post_nadir=arms.arm_id.isin(post_nadir_ids)), "arm_endpoints")
save_record(model_pairs, "treatment_pairs")
```

## 5 · Mechanistic models and identifiability

Tumor volume is described by four models: exponential growth, constant kill (treatment lowers the growth rate g by a fixed kill rate k), and two models in which shrinkage is followed by regrowth. In the tolerant-subpopulation model a fraction f of the tumor does not respond: V(t) = V0[(1 − f)e^((g − k)t) + f·e^(gt)]. In the declining-effect model the whole tumor responds and the kill rate decays at rate δ: dV/dt = [g − k·e^(−δt)]V. The declining-effect model is not nested within the tolerant-subpopulation model, so the two are compared as competing explanations.

Section 5.1 measures how closely the declining-effect model can reproduce a tolerant-subpopulation trajectory on the study's own measurement schedules. Section 5.2 checks local identifiability from sensitivity matrices. Section 5.3 tests whether the tolerant-subpopulation parameters can be recovered. Parameter sets are drawn from the priors of Section 5.6 and simulated on the study's measurement schedules, with measurement error of SD 0.15 log units (deliberately above the measured noise). Each simulated arm ends at its first measurement above 2,000 mm³, and the model is refitted by bounded least squares from three starting points. A parameter counts as recovered when the median absolute log error is below 0.25. Section 5.4 repeats this on regular designs that vary the measurement interval, duration and number of animals per arm. Additional animals are modeled as repeated measurements of one shared trajectory, so the design map is an upper bound on what replication can achieve. Section 5.5 simulates from each regrowth model, fits both, and keeps the model with the lower Akaike information criterion (AIC). In 5.5 the two generating models draw their kill rates from different distributions (centred on 0.12 and 0.14 per day), so part of what the criterion separates can be kill-rate magnitude rather than mechanism. Section 5.5b therefore repeats the comparison with paired replicates: baseline volume, growth rate, kill rate, schedule and noise vector are drawn once and shared by both mechanisms, so the mechanism is the only systematic difference, as in Section 12. Section 5.6 checks a Bayesian version of the tolerant-subpopulation model by prior predictive simulation and simulation-based calibration (Talts et al., 2018); that model is not fitted to observed data. Each calibration replicate runs one chain, so convergence is judged by split R̂ within the chain; the uniformity tests are repeated on the replicates whose split R̂ is at most 1.05 for every calibrated parameter and that had no divergent transitions, to show whether poorly mixing replicates drive the result.

The design map uses 60 replicates per design, so a design whose median error sits near the recovery threshold can change its call between seeds. The median absolute log errors are printed beside the recovery calls for that reason.

**Inputs:** `empirical_grids`, `untreated_growth`, `MEASUREMENT_NOISE`.
**Outputs:** `GROWTH_PRIOR`, `approximation`, `local_identifiability`, `parameter_recovery`, `design_map`, `mechanism_discrimination`, `calibration`, `paired_discrimination`; figure data `approximation_example.csv` and `prior_predictive.csv`.

```python
# ==========================================================================
# 5.0 Priors shared by the simulations and the Bayesian model
#     The growth-rate prior is centred on the untreated arms: log-normal with
#     the median and 1.15 times the SD of the positive untreated log growth
#     rates, truncated at 0.20 per day (a doubling time of 3.5 days).
# ==========================================================================

set_section("5")
positive_untreated = untreated_growth.loc[untreated_growth > 0]
GROWTH_PRIOR = {
    "growth_centre": float(np.median(np.log(positive_untreated))),
    "growth_spread": float(1.15 * np.std(np.log(positive_untreated))),
    "growth_upper": 0.20,
}
upper_z = (np.log(GROWTH_PRIOR["growth_upper"]) - GROWTH_PRIOR["growth_centre"]) / GROWTH_PRIOR["growth_spread"]
prior_rate_quantiles = np.exp(GROWTH_PRIOR["growth_centre"] + GROWTH_PRIOR["growth_spread"]
                              * stats.truncnorm.ppf([0.025, 0.5, 0.975], -np.inf, upper_z))
print("5.0 PRIORS")
print(f"  untreated arms with a positive growth rate      {len(positive_untreated)} "
      f"(fastest {positive_untreated.max():.3f} per day)")
print(f"  log growth rate ~ TruncatedNormal({GROWTH_PRIOR['growth_centre']:.3f}, "
      f"{GROWTH_PRIOR['growth_spread']:.3f}), upper limit log {GROWTH_PRIOR['growth_upper']}")
print(f"  implied doubling time, median [95% range]       {np.log(2) / prior_rate_quantiles[1]:.1f} "
      f"[{np.log(2) / prior_rate_quantiles[2]:.1f}, {np.log(2) / prior_rate_quantiles[0]:.1f}] days")
print("  kill rate ~ HalfNormal(0.10); logit tolerant fraction ~ Normal(−2.2, 1.5) "
      f"(median fraction {1 / (1 + np.exp(2.2)):.3f}); log baseline ~ Normal(log 200, 0.5)")

# ==========================================================================
# 5.1 How closely the declining-effect model reproduces a tolerant-
#     subpopulation trajectory on the study's measurement schedules
#     The declining-effect model is fitted to noiseless tolerant-subpopulation
#     curves; the largest absolute residual is the closest the two models
#     come on that schedule. Two settings: one fixed parameter set on 200
#     schedules, and 200 parameter sets drawn from the priors.
# ==========================================================================

ILLUSTRATIVE_TOLERANT = {"log_baseline": np.log(200.0), "growth_rate": 0.05,
                         "kill_rate": 0.12, "tolerant_fraction": 0.05}
demonstration_rng = np.random.default_rng(SEEDS["mechanism_demonstration"])
declining_forward = constant_exposure_forward("declining drug effect")
schedules_56 = [grid[grid <= 56] for grid in empirical_grids if grid.max() >= 56]
approximation_rows = []
for setting in ("fixed parameters", "prior draws"):
    picks = demonstration_rng.choice(len(schedules_56), size=min(200, len(schedules_56)), replace=False)
    for index in picks:
        days = schedules_56[index]
        truth = (ILLUSTRATIVE_TOLERANT if setting == "fixed parameters"
                 else draw_prior_parameters(demonstration_rng, GROWTH_PRIOR))
        curve = log_volume_tolerant(days, truth["log_baseline"], truth["growth_rate"],
                                    truth["kill_rate"], truth["tolerant_fraction"])
        fit = fit_mechanism("declining drug effect", days, curve)
        residual = declining_forward(fit["x"], days) - curve
        approximation_rows.append({"setting": setting, "n_measurements": len(days),
                                   "max_abs_residual": float(np.max(np.abs(residual))),
                                   "rmse": float(np.sqrt(np.mean(residual ** 2)))})
approximation = pd.DataFrame(approximation_rows)
approximation_summary = approximation.groupby("setting").agg(
    schedules=("max_abs_residual", "size"),
    median_measurements=("n_measurements", "median"),
    median_max_abs_residual=("max_abs_residual", "median"),
    p90_max_abs_residual=("max_abs_residual", lambda v: v.quantile(0.9)),
    median_rmse=("rmse", "median"))
fixed_separation = record("approximation_max_residual",
                          float(approximation_summary.loc["fixed parameters", "median_max_abs_residual"]))
record("approximation_to_noise_ratio", fixed_separation / MEASUREMENT_NOISE)
print("\n5.1 DECLINING-EFFECT FIT TO TOLERANT-SUBPOPULATION CURVES (schedules up to day 56)")
print(approximation_summary.round(4).to_string())
print(f"  fixed parameters: median largest residual {fixed_separation:.4f} log units, "
      f"{fixed_separation / MEASUREMENT_NOISE:.2f} times the measurement noise")

# ==========================================================================
# 5.2 Local identifiability
#     Sensitivity matrices on a dense grid (a property of the models, not of
#     the schedules). Numerical rank uses a relative tolerance of 1e-8 times
#     the largest singular value.
# ==========================================================================

REPRESENTATIVE_PARAMETERS = {
    "constant kill": (log_volume_constant_kill, [np.log(100.0), 0.05, 0.10]),
    "tolerant subpopulation": (log_volume_tolerant, [np.log(100.0), 0.05, 0.10, 0.20]),
    "declining drug effect": (log_volume_declining, [np.log(100.0), 0.05, 0.15, 0.05]),
}
identifiability_rows = []
for window_days in (7, 14, 21, 28, 42, 56):
    days = np.linspace(0.0, float(window_days), max(5, window_days + 1))
    for model_name, (model, parameters) in REPRESENTATIVE_PARAMETERS.items():
        singular = np.linalg.svd(sensitivity_matrix(model, days, parameters), compute_uv=False)
        identifiability_rows.append({
            "model": model_name, "window_days": window_days, "parameters": len(parameters),
            "numerical_rank": int(np.sum(singular > 1e-8 * singular.max())),
            "condition_number": float(singular.max() / singular.min()) if singular.min() > 0 else np.inf})
local_identifiability = pd.DataFrame(identifiability_rows)
print("\n5.2 LOCAL IDENTIFIABILITY")
print(local_identifiability.pivot(index="window_days", columns="model",
                                  values="numerical_rank").to_string())
print(local_identifiability.pivot(index="window_days", columns="model", values="condition_number")
      .map(lambda v: f"{v:.3g}").to_string())
print("  In the constant-kill model growth and kill rates enter only as g − k, so its rank is 2 of 3.")

# ==========================================================================
# 5.3 Recovery of the tolerant-subpopulation parameters on the study's schedules
#     Error-to-prior ratio: RMSE of the log error over the SD of the log true
#     values; below 1 the estimates carry information beyond the prior.
# ==========================================================================

RECOVERY_WINDOWS = (14, 21, 28, 42, 56)
N_RECOVERY_REPLICATES = 200
RECOVERED_PARAMETERS = ("growth_rate", "kill_rate", "tolerant_fraction")

recovery_rng = np.random.default_rng(SEEDS["parameter_recovery"])
recovery_rows = []
for window_days in RECOVERY_WINDOWS:
    usable = [grid[grid <= window_days] for grid in empirical_grids if grid.max() >= window_days]
    log_errors = {name: [] for name in RECOVERED_PARAMETERS}
    log_truths = {name: [] for name in RECOVERED_PARAMETERS}
    terminated = 0
    for _ in range(N_RECOVERY_REPLICATES):
        days = usable[recovery_rng.integers(len(usable))]
        if len(days) < 4:
            continue
        truth = draw_prior_parameters(recovery_rng, GROWTH_PRIOR)
        noisy = (log_volume_tolerant(days, truth["log_baseline"], truth["growth_rate"],
                                     truth["kill_rate"], truth["tolerant_fraction"])
                 + recovery_rng.normal(0, SIMULATION_NOISE_SD, size=len(days)))
        fit_days, noisy, stopped = apply_termination(days, noisy)
        terminated += stopped
        if len(fit_days) < 4:
            continue
        fit = fit_mechanism("tolerant subpopulation", fit_days, noisy)
        if not np.isfinite(fit["rss"]):
            continue
        estimate = tolerant_estimates(fit["x"])
        for name in RECOVERED_PARAMETERS:
            log_errors[name].append(np.log(max(estimate[name], 1e-12)) - np.log(max(truth[name], 1e-12)))
            log_truths[name].append(np.log(max(truth[name], 1e-12)))
    for name in RECOVERED_PARAMETERS:
        recovery_rows.append({"parameter": name, "window_days": window_days,
                              "terminated_share": terminated / N_RECOVERY_REPLICATES,
                              **recovery_summary(log_errors[name], log_truths[name])})
parameter_recovery = pd.DataFrame(recovery_rows)
record("n_recovered_parameter_windows", int(parameter_recovery.recovered.sum()))
print("\n5.3 PARAMETER RECOVERY (tolerant-subpopulation model, study schedules)")
print(f"  simulation noise SD {SIMULATION_NOISE_SD} log units; arms end above "
      f"{TERMINATION_VOLUME:,.0f} mm³; recovered when the median |log error| < {RECOVERY_THRESHOLD}")
print(parameter_recovery.round(3).to_string(index=False))
print(f"  parameter-window combinations recovered         {int(parameter_recovery.recovered.sum())} "
      f"of {len(parameter_recovery)}")

# ==========================================================================
# 5.4 Design map: regular schedules with varied interval, duration and
#     number of animals per arm
#     Extra animals enter as repeated measurements of one shared trajectory
#     (noise SD divided by the square root of the number of animals), with no
#     between-animal variation; the map is therefore an upper bound.
# ==========================================================================

DESIGN_INTERVALS = (2, 3, 4, 7)
DESIGN_DURATIONS = (14, 21, 28, 42, 56)
DESIGN_ANIMALS = (1, 2, 3, 5)
N_DESIGN_REPLICATES = 60

design_rng = np.random.default_rng(SEEDS["design_map"])
design_rows = []
for interval_days in DESIGN_INTERVALS:
    for duration_days in DESIGN_DURATIONS:
        days = np.arange(0.0, duration_days + 1, interval_days)
        if len(days) < 4:
            continue
        for animals in DESIGN_ANIMALS:
            noise_sd = SIMULATION_NOISE_SD / np.sqrt(animals)
            log_errors = {name: [] for name in RECOVERED_PARAMETERS}
            log_truths = {name: [] for name in RECOVERED_PARAMETERS}
            for _ in range(N_DESIGN_REPLICATES):
                truth = draw_prior_parameters(design_rng, GROWTH_PRIOR)
                noisy = (log_volume_tolerant(days, truth["log_baseline"], truth["growth_rate"],
                                             truth["kill_rate"], truth["tolerant_fraction"])
                         + design_rng.normal(0, noise_sd, size=len(days)))
                fit_days, noisy, _ = apply_termination(days, noisy)
                if len(fit_days) < 4:
                    continue
                fit = fit_mechanism("tolerant subpopulation", fit_days, noisy)
                if not np.isfinite(fit["rss"]):
                    continue
                estimate = tolerant_estimates(fit["x"])
                for name in RECOVERED_PARAMETERS:
                    log_errors[name].append(np.log(max(estimate[name], 1e-12))
                                            - np.log(max(truth[name], 1e-12)))
                    log_truths[name].append(np.log(max(truth[name], 1e-12)))
            for name in RECOVERED_PARAMETERS:
                if len(log_errors[name]) >= 10:
                    design_rows.append({"parameter": name, "interval_days": interval_days,
                                        "duration_days": duration_days, "animals_per_arm": animals,
                                        **recovery_summary(log_errors[name], log_truths[name])})
design_map = pd.DataFrame(design_rows)
n_designs = design_map.groupby(["interval_days", "duration_days", "animals_per_arm"]).ngroups
fraction_designs = design_map.loc[design_map.parameter == "tolerant_fraction"]
record("n_alternative_designs", n_designs)
record("n_designs_fraction_recovered", int(fraction_designs.recovered.sum()))
print("\n5.4 DESIGN MAP")
print(f"  designs evaluated                               {n_designs}")
for name in RECOVERED_PARAMETERS:
    subset = design_map.loc[design_map.parameter == name]
    print(f"  {name:<18} recovered in {int(subset.recovered.sum())} of {len(subset)} designs")
for name in RECOVERED_PARAMETERS:
    print(f"\n  {name}: recovered (rows: interval, animals per arm; columns: duration in days)")
    print(design_map.loc[design_map.parameter == name]
          .pivot_table(index=["interval_days", "animals_per_arm"], columns="duration_days",
                       values="recovered", aggfunc="first").to_string())
    # The continuous error behind each call: a call can flip between seeds
    # when the median error sits near the threshold, the error itself less so.
    print(f"  {name}: median absolute log error (threshold {RECOVERY_THRESHOLD})")
    print(design_map.loc[design_map.parameter == name]
          .pivot_table(index=["interval_days", "animals_per_arm"], columns="duration_days",
                       values="median_abs_log_error", aggfunc="first").round(2).to_string())
save_record(parameter_recovery, "parameter_recovery")
save_record(design_map, "design_map")
save_record(approximation, "declining_effect_approximation")

# The worked example of 5.1, for the identifiability figure: the fixed
# tolerant-subpopulation curve on the typical 56-day schedule and its best
# declining-effect fit, on that schedule and on a dense grid of days.
schedule_56 = typical_schedule(schedules_56, 56)
example_fit = fit_mechanism("declining drug effect", schedule_56,
                            log_volume_tolerant(schedule_56, **ILLUSTRATIVE_TOLERANT))
dense_days = np.linspace(0.0, float(schedule_56.max()), 300)
approximation_example = pd.concat([
    pd.DataFrame({"grid": grid, "day": days,
                  "tolerant": log_volume_tolerant(days, **ILLUSTRATIVE_TOLERANT),
                  "declining_fit": declining_forward(example_fit["x"], days)})
    for grid, days in (("schedule", schedule_56), ("dense", dense_days))], ignore_index=True)
save_record(approximation_example, "approximation_example")
```

```python
# ==========================================================================
# 5.5 Discrimination between the two regrowth models
#     Trajectories are simulated from each model on the study's measurement
#     schedules, with SD 0.15 log units of noise, and end at their first
#     measurement above 2,000 mm³. Both models are fitted to every
#     trajectory by bounded multi-start least squares, and the model with the
#     lower AIC is selected. Both models have four parameters, so the AIC
#     comparison is a comparison of residual fit.
# ==========================================================================

DISCRIMINATION_WINDOWS = (14, 21, 28, 42, 56)
N_DISCRIMINATION_REPLICATES = 200

discrimination_rng = np.random.default_rng(SEEDS["mechanism_discrimination"])
discrimination_rows = []
for window_days in DISCRIMINATION_WINDOWS:
    usable = [grid[grid <= window_days] for grid in empirical_grids if grid.max() >= window_days]
    for truth in MECHANISMS:
        for _ in range(N_DISCRIMINATION_REPLICATES):
            days = usable[discrimination_rng.integers(len(usable))]
            if len(days) < 5:
                continue
            log_v0 = np.log(200.0) + discrimination_rng.normal(0, 0.3)
            growth = float(np.exp(discrimination_rng.normal(np.log(0.05), 0.5)))
            if truth == "tolerant subpopulation":
                curve = log_volume_tolerant(
                    days, log_v0, growth, float(abs(discrimination_rng.normal(0.12, 0.04))),
                    float(1.0 / (1.0 + np.exp(-discrimination_rng.normal(-2.5, 1.0)))))
            else:
                curve = log_volume_declining(
                    days, log_v0, growth, float(abs(discrimination_rng.normal(0.14, 0.04))),
                    float(np.exp(discrimination_rng.normal(np.log(0.06), 0.5))))
            noisy = curve + discrimination_rng.normal(0, SIMULATION_NOISE_SD, size=len(days))
            fit_days, noisy, stopped = apply_termination(days, noisy)
            if len(fit_days) < 5:
                discrimination_rows.append({"true_mechanism": truth, "window_days": window_days,
                                            "verdict": "too few measurements", "terminated": stopped})
                continue
            fits = {mechanism: fit_mechanism(mechanism, fit_days, noisy) for mechanism in MECHANISMS}
            verdict, delta = aic_verdict(fits)
            discrimination_rows.append({"true_mechanism": truth, "window_days": window_days,
                                        "verdict": verdict, "terminated": stopped,
                                        "criterion_gap": abs(delta)})
discrimination_fits = pd.DataFrame(discrimination_rows)
scored_fits = discrimination_fits.loc[discrimination_fits.verdict.isin(MECHANISMS)]
mechanism_discrimination = scored_fits.groupby(["true_mechanism", "window_days"]).agg(
    replicates=("verdict", "size"),
    chose_tolerant=("verdict", lambda v: float((v == "tolerant subpopulation").mean())),
    median_criterion_gap=("criterion_gap", "median")).reset_index()
# Accuracy is the share choosing the model that generated the data.
mechanism_discrimination["accuracy"] = np.where(
    mechanism_discrimination.true_mechanism == "tolerant subpopulation",
    mechanism_discrimination.chose_tolerant, 1 - mechanism_discrimination.chose_tolerant)
excluded = discrimination_fits.groupby("window_days").verdict.apply(
    lambda v: int((v == "too few measurements").sum()))

record("chose_tolerant_min", float(mechanism_discrimination.chose_tolerant.min()))
record("chose_tolerant_max", float(mechanism_discrimination.chose_tolerant.max()))
record("discrimination_mean_accuracy", float(mechanism_discrimination.accuracy.mean()))
print("5.5 DISCRIMINATION BETWEEN THE TWO REGROWTH MODELS")
print(mechanism_discrimination.round(3).to_string(index=False))
print(f"\n  share selecting the tolerant-subpopulation model, range over conditions   "
      f"{mechanism_discrimination.chose_tolerant.min():.3f} to {mechanism_discrimination.chose_tolerant.max():.3f}")
print(f"  mean accuracy over conditions                    {mechanism_discrimination.accuracy.mean():.3f}")
print(f"  simulated arms ending with fewer than five measurements, by window: "
      f"{excluded.to_dict()}")
save_record(mechanism_discrimination, "mechanism_discrimination")

# ==========================================================================
# 5.5b Paired discrimination: the mechanism is the only systematic difference
#     In 5.5 the two generating models draw kill rates from different
#     distributions (centred on 0.12 and 0.14 per day). Here each replicate
#     draws one schedule, baseline volume, growth rate, kill rate (centred on
#     0.13, between the two) and standard-normal noise vector, and generates
#     both mechanisms from them. Only the tolerant fraction (tolerant model)
#     and the decay rate (declining model) are mechanism-specific, with the
#     distributions of 5.5. Each arm ends at its first measurement above
#     2,000 mm³, as in 5.5. Verdicts are recorded twice: always choosing the
#     lower AIC, as in 5.5, and with |ΔAIC| < 2 recorded as no decision, as
#     in Section 12.
# ==========================================================================

N_PAIRED_DISCRIMINATION_REPLICATES = 200        # pairs per window
PAIRED_DISCRIMINATION_TIE = 2.0                 # |ΔAIC| below this is no decision

paired_discrimination_rng = np.random.default_rng(SEEDS["paired_discrimination"])
paired_discrimination_rows = []
for window_days in DISCRIMINATION_WINDOWS:
    usable = [grid[grid <= window_days] for grid in empirical_grids if grid.max() >= window_days]
    for replicate in range(N_PAIRED_DISCRIMINATION_REPLICATES):
        # Fixed draw order, so the seed reproduces the same pairs.
        days = usable[paired_discrimination_rng.integers(len(usable))]
        log_v0 = np.log(200.0) + paired_discrimination_rng.normal(0, 0.3)
        growth = float(np.exp(paired_discrimination_rng.normal(np.log(0.05), 0.5)))
        kill = float(abs(paired_discrimination_rng.normal(0.13, 0.04)))
        noise = paired_discrimination_rng.normal(0, 1.0, size=len(days))
        fraction = float(1.0 / (1.0 + np.exp(-paired_discrimination_rng.normal(-2.5, 1.0))))
        decay = float(np.exp(paired_discrimination_rng.normal(np.log(0.06), 0.5)))
        if len(days) < 5:
            continue
        curves = {"tolerant subpopulation": log_volume_tolerant(days, log_v0, growth, kill, fraction),
                  "declining drug effect": log_volume_declining(days, log_v0, growth, kill, decay)}
        for truth, curve in curves.items():
            noisy = curve + SIMULATION_NOISE_SD * noise
            fit_days, fit_values, stopped = apply_termination(days, noisy)
            row = {"window_days": window_days, "replicate": replicate, "true_mechanism": truth,
                   "terminated": stopped, "n_measurements": len(fit_days)}
            if len(fit_days) < 5:
                paired_discrimination_rows.append({**row, "verdict": "too few measurements",
                                                   "verdict_with_tie": "too few measurements",
                                                   "criterion_gap": np.nan})
                continue
            fits = {mechanism: fit_mechanism(mechanism, fit_days, fit_values) for mechanism in MECHANISMS}
            verdict, delta = aic_verdict(fits)
            verdict_with_tie, _ = aic_verdict(fits, PAIRED_DISCRIMINATION_TIE)
            paired_discrimination_rows.append({**row, "verdict": verdict, "verdict_with_tie": verdict_with_tie,
                                               "criterion_gap": abs(delta)})
paired_discrimination_fits = pd.DataFrame(paired_discrimination_rows)
paired_scored = paired_discrimination_fits.loc[paired_discrimination_fits.verdict.isin(MECHANISMS)]
paired_discrimination = paired_scored.groupby(["true_mechanism", "window_days"]).agg(
    replicates=("verdict", "size"),
    chose_tolerant=("verdict", lambda v: float((v == "tolerant subpopulation").mean())),
    decisive_rate=("verdict_with_tie", lambda v: float(v.isin(MECHANISMS).mean())),
    median_criterion_gap=("criterion_gap", "median")).reset_index()
paired_discrimination["accuracy"] = np.where(
    paired_discrimination.true_mechanism == "tolerant subpopulation",
    paired_discrimination.chose_tolerant, 1 - paired_discrimination.chose_tolerant)
decisive_calls = paired_scored.loc[paired_scored.verdict_with_tie.isin(MECHANISMS)]
paired_decisive_accuracy = (decisive_calls.assign(correct=decisive_calls.verdict_with_tie
                                                   == decisive_calls.true_mechanism)
                            .groupby("window_days").correct.agg(["mean", "sum", "size"]))
# Balanced accuracy per window: the mean of the two per-mechanism accuracies,
# so that a fitter preferring one model scores 0.5 however strong the bias.
balanced_by_window = paired_discrimination.groupby("window_days").accuracy.mean()

record("paired_chose_tolerant_min", float(paired_discrimination.chose_tolerant.min()))
record("paired_chose_tolerant_max", float(paired_discrimination.chose_tolerant.max()))
record("paired_discrimination_mean_accuracy", float(paired_discrimination.accuracy.mean()))
record("paired_balanced_accuracy_min", float(balanced_by_window.min()))
record("paired_balanced_accuracy_max", float(balanced_by_window.max()))
print("\n5.5b PAIRED DISCRIMINATION (shared baseline, growth, kill rate, schedule and noise)")
print(paired_discrimination.round(3).to_string(index=False))
print("\n  balanced accuracy by window (mean of the two per-mechanism accuracies)")
print(balanced_by_window.round(3).to_string())
print(f"\n  with |ΔAIC| < {PAIRED_DISCRIMINATION_TIE:g} recorded as no decision: accuracy among decisive calls by window")
print(paired_decisive_accuracy.rename(columns={"mean": "accuracy", "sum": "correct", "size": "decisive_calls"})
      .round(3).to_string())
print(f"\n  unpaired (5.5) against paired: mean accuracy over conditions "
      f"{RESULTS['discrimination_mean_accuracy']:.3f} against {RESULTS['paired_discrimination_mean_accuracy']:.3f}; "
      f"share selecting the tolerant model {RESULTS['chose_tolerant_min']:.3f}-{RESULTS['chose_tolerant_max']:.3f} "
      f"against {RESULTS['paired_chose_tolerant_min']:.3f}-{RESULTS['paired_chose_tolerant_max']:.3f}")
save_record(paired_discrimination, "mechanism_discrimination_paired")
```

```python
# ==========================================================================
# 5.6 Bayesian tolerant-subpopulation model: prior predictive simulation and
#     simulation-based calibration
#     Calibration: a parameter set is drawn from the priors, data are
#     simulated on a 28-day schedule, the posterior is sampled with NUTS,
#     and the rank of the true value among posterior draws is recorded.
#     Posterior draws are thinned to roughly their effective sample size
#     before ranking, so that ranks are computed among nearly independent
#     draws. Under correct calibration the normalized ranks are uniform.
# ==========================================================================

prior_days = np.linspace(0.0, 42.0, 15)
prior_draws = Predictive(tolerant_subpopulation_model, num_samples=400)(
    jax.random.PRNGKey(SEEDS["prior_predictive"]), days=prior_days, prior=GROWTH_PRIOR)
prior_volumes = np.exp(np.asarray(prior_draws["observed"]))
print("5.6 PRIOR PREDICTIVE SIMULATION (400 trajectories, days 0 to 42)")
for label, column in (("day 0", 0), ("day 42", -1)):
    values = prior_volumes[:, column]
    print(f"  volume at {label:<7} median {np.median(values):,.0f} mm³ "
          f"[5th {np.percentile(values, 5):,.0f}, 95th {np.percentile(values, 95):,.0f}]")
print(f"  trajectories above 20,000 mm³ at day 42          "
      f"{100 * np.mean(prior_volumes[:, -1] > 20_000):.1f}% (the study removed animals near 2,000 mm³)")

CALIBRATION_DAYS = np.array([0.0, 3.0, 7.0, 10.0, 14.0, 17.0, 21.0, 24.0, 28.0])
N_CALIBRATION_REPLICATES = 100
CALIBRATION_WARMUP, CALIBRATION_DRAWS = 300, 500
CALIBRATED_PARAMETERS = ("growth_rate", "kill_rate", "tolerant_fraction", "noise_scale")

uniform_rng = np.random.default_rng(SEEDS["calibration"])
calibration_rows = []
for replicate in range(N_CALIBRATION_REPLICATES):
    truth_key, sampler_key = jax.random.split(jax.random.PRNGKey(SEEDS["calibration"] + replicate))
    truth = Predictive(tolerant_subpopulation_model, num_samples=1)(
        truth_key, days=CALIBRATION_DAYS, prior=GROWTH_PRIOR)
    sampler = MCMC(NUTS(tolerant_subpopulation_model, target_accept_prob=0.90),
                   num_warmup=CALIBRATION_WARMUP, num_samples=CALIBRATION_DRAWS,
                   num_chains=1, progress_bar=False)
    sampler.run(sampler_key, days=CALIBRATION_DAYS, prior=GROWTH_PRIOR,
                log_volume=np.asarray(truth["observed"])[0], extra_fields=("diverging",))
    posterior = sampler.get_samples()
    divergences = int(np.asarray(sampler.get_extra_fields()["diverging"]).sum())
    for name in CALIBRATED_PARAMETERS:
        draws = np.asarray(posterior[name], dtype=float)
        ess = float(effective_sample_size(draws[None, :]))
        thin = max(1, int(np.ceil(len(draws) / max(ess, 1.0))))
        kept = draws[::thin]
        rank = int(np.sum(kept < float(np.asarray(truth[name])[0])))
        calibration_rows.append({
            "replicate": replicate, "parameter": name, "divergences": divergences,
            "ess": ess, "split_rhat": float(split_gelman_rubin(draws[None, :])),
            "kept_draws": len(kept), "rank": rank,
            "normalized_rank": (rank + uniform_rng.uniform()) / (len(kept) + 1)})
calibration_ranks = pd.DataFrame(calibration_rows)
calibration = calibration_ranks.groupby("parameter").agg(
    replicates=("rank", "size"), median_ess=("ess", "median"),
    median_kept_draws=("kept_draws", "median"), max_split_rhat=("split_rhat", "max"),
    ks_statistic=("normalized_rank", lambda r: stats.kstest(r, "uniform").statistic),
    ks_p_value=("normalized_rank", lambda r: stats.kstest(r, "uniform").pvalue))
replicate_divergences = calibration_ranks.drop_duplicates("replicate").divergences
print("\n5.6 SIMULATION-BASED CALIBRATION")
print(f"  replicates {N_CALIBRATION_REPLICATES}; NUTS, {CALIBRATION_WARMUP} warm-up and "
      f"{CALIBRATION_DRAWS} draws, one chain; schedule {CALIBRATION_DAYS.astype(int).tolist()}")
print(f"  divergent transitions: {int(replicate_divergences.sum())} in total, in "
      f"{int((replicate_divergences > 0).sum())} replicates")
print(calibration.round(3).to_string())
save_record(calibration_ranks, "calibration_ranks")

# Convergence. With one chain per replicate, split R̂ compares the two halves
# of that chain. A replicate is counted as converged when every calibrated
# parameter has split R̂ <= 1.05 and the chain had no divergent transition.
# The uniformity tests are repeated on the converged replicates only; if the
# poorly mixing replicates were driving the result, the two sets would differ.
MAX_CALIBRATION_RHAT = 1.05
by_replicate = calibration_ranks.groupby("replicate").agg(max_rhat=("split_rhat", "max"),
                                                          divergences=("divergences", "first"))
converged_replicates = by_replicate.index[(by_replicate.max_rhat <= MAX_CALIBRATION_RHAT)
                                          & (by_replicate.divergences == 0)]
converged_ranks = calibration_ranks.loc[calibration_ranks.replicate.isin(converged_replicates)]
calibration_converged = converged_ranks.groupby("parameter").agg(
    replicates=("rank", "size"),
    ks_statistic=("normalized_rank", lambda r: stats.kstest(r, "uniform").statistic),
    ks_p_value=("normalized_rank", lambda r: stats.kstest(r, "uniform").pvalue))
record("calibration_replicates", N_CALIBRATION_REPLICATES)
record("calibration_converged_replicates", len(converged_replicates))
record("calibration_replicates_with_divergences", int((replicate_divergences > 0).sum()))
record("calibration_min_ks_p", float(calibration.ks_p_value.min()))
record("calibration_converged_min_ks_p", float(calibration_converged.ks_p_value.min())
       if len(calibration_converged) else np.nan)
print(f"\n  replicates with split R̂ > {MAX_CALIBRATION_RHAT} for some parameter: "
      f"{int((by_replicate.max_rhat > MAX_CALIBRATION_RHAT).sum())}; with a divergent transition: "
      f"{int((by_replicate.divergences > 0).sum())}; converged on both counts: {len(converged_replicates)} "
      f"of {N_CALIBRATION_REPLICATES}")
print("  uniformity of normalized ranks, converged replicates only")
print(calibration_converged.round(3).to_string())

# Data for the Bayesian model-checks figure: the prior predictive
# trajectories, one row per draw and day (the SBC ranks are saved above).
save_record(pd.DataFrame({"draw": np.repeat(np.arange(prior_volumes.shape[0]), len(prior_days)),
                          "day": np.tile(prior_days, prior_volumes.shape[0]),
                          "volume": prior_volumes.ravel()}), "prior_predictive")
```

## 6 · Measurement noise, kinetic estimates and termination hazard

Measurement noise is the median residual SD across untreated arms after a straight line is fitted to log volume (Section 3.6). The intrinsic growth rate of each PDX model is the slope of that line. Exponential growth is the working growth law; the curvature check below reports how often a quadratic fits the untreated arms better.

Kill rates are estimated for each treated arm by least squares under the tolerant-subpopulation model, with the growth rate held at the PDX model's untreated value. The fit is bounded (log kill rate between −18.4 and 6.9), so a fit that finds no signal stops at a bound instead of overflowing. Fits with kill rates below 0.0001 or above 1 per day are failed fits. The variance of the log kill rate over the remaining fits is divided into between-treatment, between-model and remainder parts, by two methods. The first is descriptive marginal shares (the weighted variance of treatment means and of model means, each over the total), which overlap and do not sum to one. The second is a crossed random-effects model fitted by restricted maximum likelihood (REML). Each PDX model–treatment pair is one animal, so the model × treatment interaction cannot be separated from residual error, and both sit in the remainder. The REML fit tries L-BFGS first and moves on to BFGS, Powell and Nelder–Mead only if the previous optimizer fails to converge; whether the final fit converged is printed and stored with the shares, because variance components from a fit that has not converged are not estimates.

Arms were ended when tumors became large. The per-visit probability that a measurement is the arm's last is modeled by logistic regression on standardized log volume and elapsed day, without a penalty, with a bootstrap that resamples arms. The fitted probabilities give inverse-probability-of-censoring weights for the mean post-nadir growth rate. Each post-nadir arm is weighted by the inverse of its estimated probability of remaining under observation through its second post-nadir measurement, the point at which it entered the analysis set (probabilities below 0.05 are set to 0.05).

**Inputs:** `observations`, `untreated_fits`, `untreated_growth`, `post_nadir_arms`.
**Outputs:** `kinetic_estimates`, `kill_rate_partition`, `visits`, `hazard_model`; figure data `termination_hazard_curve.csv`, `termination_hazard_points.csv`, `termination_by_volume_decile.csv`, `last_visits.csv` and `censoring_weights.csv`.

```python
# ==========================================================================
# 6.1 Measurement noise, untreated growth and the growth law
#     The quadratic is counted as fitting better when its residual SD (three
#     parameters removed) is more than 10% below the linear fit's (two removed).
# ==========================================================================

set_section("6")
doubling_days = np.log(2) / untreated_growth.loc[untreated_growth > 0]
curved = untreated_fits.quadratic_residual_sd < 0.9 * untreated_fits.residual_sd
print("6.1 MEASUREMENT NOISE AND UNTREATED GROWTH")
print(f"  untreated arms fitted                           {len(untreated_fits)} "
      f"(>= {MIN_MEASUREMENTS_PER_ARM} measurements)")
print(f"  measurement noise, median residual SD           {MEASUREMENT_NOISE:.4f} log units")
print(f"  residual SD, interquartile range                {untreated_fits.residual_sd.quantile(.25):.4f}"
      f" to {untreated_fits.residual_sd.quantile(.75):.4f}")
print(f"  untreated growth rate, median [5th, 95th]       {untreated_growth.median():.4f} "
      f"[{untreated_growth.quantile(.05):.4f}, {untreated_growth.quantile(.95):.4f}] per day")
print(f"  doubling time, median [5th, 95th]               {doubling_days.median():.1f} "
      f"[{doubling_days.quantile(.05):.1f}, {doubling_days.quantile(.95):.1f}] days")
print(f"  arms where a quadratic fits better              {int(curved.sum())} of {len(untreated_fits)} "
      f"({100 * curved.mean():.1f}%), decelerating in "
      f"{int((curved & (untreated_fits.curvature < 0)).sum())}")
print("  untreated growth rate by tumor type")
print(untreated_fits.groupby("tumor_type").growth_rate.describe()[["count", "25%", "50%", "75%"]]
      .round(4).to_string())

# ==========================================================================
# 6.2 Kill rates and the variance partition
# ==========================================================================

KILL_FIT_BOUNDS = (np.array([-5.0, np.log(1e-8), -30.0]), np.array([15.0, np.log(1e3), 30.0]))
kill_rows = []
for arm_id, block in arm_series(observations.loc[~observations.is_control]):
    model_id = block.model_id.iloc[0]
    if model_id not in untreated_growth.index or len(block) < MIN_MEASUREMENTS_PER_ARM:
        continue
    day = block.day.to_numpy(float) - block.day.iloc[0]
    log_volume = block.log_volume.to_numpy(float)
    growth = float(untreated_growth[model_id])
    solution = least_squares(
        lambda theta: log_volume_tolerant(day, theta[0], growth, np.exp(theta[1]),
                                          1.0 / (1.0 + np.exp(-theta[2]))) - log_volume,
        np.clip([log_volume[0], np.log(0.05), -2.0], *KILL_FIT_BOUNDS),
        bounds=KILL_FIT_BOUNDS, max_nfev=2000)
    kill_rows.append({"arm_id": arm_id, "model_id": model_id, "treatment": block.treatment.iloc[0],
                      "tumor_type": block.tumor_type.iloc[0], "kill_rate": float(np.exp(solution.x[1])),
                      "tolerant_fraction": float(1.0 / (1.0 + np.exp(-solution.x[2]))),
                      "residual_sd": float(np.sqrt(np.sum(solution.fun ** 2) / max(len(day) - 3, 1)))})
kinetic_estimates = pd.DataFrame(kill_rows)
kinetic_estimates["log_kill_rate"] = np.log(kinetic_estimates.kill_rate)
kinetic_estimates["failed_fit"] = ~kinetic_estimates.kill_rate.between(*KILL_RATE_RANGE)
retained = kinetic_estimates.loc[~kinetic_estimates.failed_fit].copy()
below = float((kinetic_estimates.kill_rate < KILL_RATE_RANGE[0]).mean())
above = float((kinetic_estimates.kill_rate > KILL_RATE_RANGE[1]).mean())
record("failed_kill_fit_share", below + above)

print("\n6.2 KILL RATES (tolerant-subpopulation model, growth held at the untreated rate)")
print(f"  treated arms fitted                             {len(kinetic_estimates):,} in "
      f"{kinetic_estimates.model_id.nunique()} PDX models, {kinetic_estimates.treatment.nunique()} treatments")
print(f"  kill rate below {KILL_RATE_RANGE[0]:g} per day (failed)          {100 * below:.1f}%")
print(f"  kill rate above {KILL_RATE_RANGE[1]:g} per day (failed)              {100 * above:.1f}%")
print(f"  retained fits                                   {len(retained):,}; median kill rate "
      f"{retained.kill_rate.median():.4f} per day; log kill rate mean {retained.log_kill_rate.mean():.3f}, "
      f"SD {retained.log_kill_rate.std():.3f}")
kill_quantiles = kinetic_estimates.kill_rate.quantile([0, .01, .05, .25, .5, .75, .95, .99, 1])
print("  kill rate quantiles, all fits: " + ", ".join(f"p{100 * q:g} {v:.3g}" for q, v in kill_quantiles.items()))

total_variance = float(retained.log_kill_rate.var(ddof=1))
grand_mean = float(retained.log_kill_rate.mean())
marginal_shares = {}
for factor, column in (("treatment", "treatment"), ("PDX model", "model_id")):
    means = retained.groupby(column).log_kill_rate.agg(["mean", "size"])
    marginal_shares[factor] = float(np.average((means["mean"] - grand_mean) ** 2,
                                               weights=means["size"]) / total_variance)

# Crossed random intercepts for treatment and PDX model: one group holding
# every arm, with both factors as variance components (the statsmodels form
# for crossed effects). fit_mixed_model moves to another optimizer only if
# L-BFGS fails to converge.
reml_fit, reml_converged, reml_warnings = fit_mixed_model(
    "log_kill_rate ~ 1", retained, groups=np.ones(len(retained)), re_formula="0",
    vc_formula={"treatment": "0 + C(treatment)", "model": "0 + C(model_id)"})
reml_components = dict(zip(reml_fit.model.exog_vc.names, np.asarray(reml_fit.vcomp, float)))
reml_total = sum(reml_components.values()) + float(reml_fit.scale)
kill_rate_partition = pd.DataFrame({
    "component": ["treatment", "PDX model", "remainder (interaction and residual)"],
    "marginal_share": [marginal_shares["treatment"], marginal_shares["PDX model"], np.nan],
    "reml_variance": [reml_components["treatment"], reml_components["model"], float(reml_fit.scale)]})
kill_rate_partition["reml_share"] = kill_rate_partition.reml_variance / reml_total
print(f"\n  variance partition of the log kill rate, retained fits (total variance {total_variance:.3f})")
print(kill_rate_partition.round(3).to_string(index=False))
print(f"  REML fit converged: {reml_converged}")
for message in reml_warnings:
    print(f"  REML fit reported: {message}")
record("reml_converged", int(reml_converged))
record("reml_share_treatment", float(kill_rate_partition.reml_share.iloc[0]))
record("reml_share_model", float(kill_rate_partition.reml_share.iloc[1]))
record("marginal_share_treatment", marginal_shares["treatment"])
record("marginal_share_model", marginal_shares["PDX model"])

# The kill-rate figure is drawn by the figures notebook from
# kinetic_estimates.csv and kill_rate_partition.csv (saved at the end of
# this cell).

# ==========================================================================
# 6.3 Termination hazard and inverse-probability-of-censoring weights
# ==========================================================================

visit_rows = []
for arm_id, block in arm_series(observations):
    if len(block) < 3:
        continue
    for index, (day, log_volume) in enumerate(zip(block.day, block.log_volume)):
        visit_rows.append((arm_id, index, day, log_volume, int(index == len(block) - 1),
                           bool(block.is_control.iloc[0])))
visits = pd.DataFrame(visit_rows, columns=["arm_id", "visit", "day", "log_volume",
                                           "arm_ends_here", "is_control"])
hazard_scaler = StandardScaler().fit(visits[["log_volume", "day"]])
hazard_design = hazard_scaler.transform(visits[["log_volume", "day"]])
hazard_outcome = visits.arm_ends_here.to_numpy()
hazard_model = unpenalized_logistic().fit(hazard_design, hazard_outcome)
volume_coefficient, time_coefficient = hazard_model.coef_[0]
hazard_rng = np.random.default_rng(SEEDS["hazard_bootstrap"])
coefficient_draws = cluster_bootstrap(
    rows_by_cluster(visits.arm_id),
    lambda rows: unpenalized_logistic().fit(hazard_design[rows], hazard_outcome[rows]).coef_[0][0],
    N_BOOTSTRAP, hazard_rng)
coefficient_low, coefficient_high = np.percentile(coefficient_draws, [2.5, 97.5])

median_day = float(visits.day.median())
volume_points = np.percentile(np.exp(visits.log_volume), [5, 25, 50, 75, 95])
hazard_points = hazard_model.predict_proba(hazard_scaler.transform(pd.DataFrame(
    {"log_volume": np.log(volume_points), "day": median_day})))[:, 1]
record("hazard_low_volume", float(volume_points[0]))
record("hazard_low", float(hazard_points[0]))
record("hazard_high_volume", float(volume_points[-1]))
record("hazard_high", float(hazard_points[-1]))
print("\n6.3 TERMINATION HAZARD")
print(f"  visits {len(visits):,} in {visits.arm_id.nunique():,} arms; last visits "
      f"{int(hazard_outcome.sum()):,}")
print(f"  standardized log-volume coefficient            {volume_coefficient:.3f} "
      f"(95% CI {coefficient_low:.3f} to {coefficient_high:.3f}, {len(coefficient_draws):,} replicates)")
print(f"  standardized elapsed-day coefficient           {time_coefficient:.3f}")
print(f"  probability a visit is the arm's last, at the median elapsed day ({median_day:.0f})")
for volume, hazard in zip(volume_points, hazard_points):
    print(f"    {volume:>7,.0f} mm³   {100 * hazard:.1f}%")

continue_probability = 1.0 - hazard_model.predict_proba(hazard_design)[:, 1]
visits["continue_probability"] = continue_probability
continuation = visits.groupby("arm_id").continue_probability.apply(np.asarray)
visit_log_volume = visits.groupby("arm_id").log_volume.apply(np.asarray)
weights = {}
for arm_id in post_nadir_arms.arm_id:
    # The arm entered the analysis set at its second measurement after the
    # nadir, so it had to remain under observation at every earlier visit.
    entry_index = int(np.argmin(visit_log_volume[arm_id])) + MIN_POINTS_AFTER_NADIR
    weights[arm_id] = 1.0 / max(np.prod(continuation[arm_id][:entry_index]), 0.05)
weighted_set = post_nadir_arms.assign(ipc_weight=post_nadir_arms.arm_id.map(weights))
unweighted_mean = float(weighted_set.post_nadir_growth_rate.mean())
weighted_mean = float(np.average(weighted_set.post_nadir_growth_rate, weights=weighted_set.ipc_weight))
record("ipcw_relative_change", (weighted_mean - unweighted_mean) / unweighted_mean)
print(f"\n  mean post-nadir growth rate, unweighted          {unweighted_mean:.5f} per day")
print(f"  weighted for censoring                           {weighted_mean:.5f} per day "
      f"({100 * (weighted_mean - unweighted_mean) / unweighted_mean:+.1f}%)")
print(f"  weights: median {weighted_set.ipc_weight.median():.2f}, 99th percentile "
      f"{weighted_set.ipc_weight.quantile(.99):.2f}, at the cap {int((weighted_set.ipc_weight >= 20).sum())}")

# Data for the termination-hazard figure: the fitted probability of a last
# visit across volumes at the median elapsed day, the same at the five
# volume percentiles printed above, the observed share by volume decile,
# each arm's last measurement, and the censoring weights.
dense_volume = np.exp(np.linspace(*np.log(np.exp(visits.log_volume).quantile([0.02, 0.98])), 120))
dense_hazard = hazard_model.predict_proba(hazard_scaler.transform(pd.DataFrame(
    {"log_volume": np.log(dense_volume), "day": median_day})))[:, 1]
deciles = visits.assign(decile=pd.qcut(visits.log_volume, 10, labels=False))
empirical = deciles.groupby("decile").agg(volume=("log_volume", lambda v: np.exp(v.median())),
                                          end_rate=("arm_ends_here", "mean"),
                                          visits=("arm_ends_here", "size"))
save_record(pd.DataFrame({"volume": dense_volume, "probability_last_visit": dense_hazard,
                          "day": median_day}), "termination_hazard_curve")
save_record(pd.DataFrame({"percentile": [5, 25, 50, 75, 95], "volume": volume_points,
                          "probability_last_visit": hazard_points, "day": median_day}),
            "termination_hazard_points")
save_record(empirical.reset_index(), "termination_by_volume_decile")
save_record(visits.loc[visits.arm_ends_here == 1, ["arm_id", "day", "is_control"]], "last_visits")
save_record(weighted_set[["arm_id", "post_nadir_growth_rate", "ipc_weight"]], "censoring_weights")
save_record(kinetic_estimates, "kinetic_estimates")
save_record(kill_rate_partition, "kill_rate_partition")
```

## 7 · Forecasting and selection of treatment

Tumor volume at day 24 is forecast from measurements taken on or before day 10, the first day on which best average response is defined. The outcome is log volume at day 24, linearly interpolated between the positive readings around it. Readings at or below zero are left out of the outcome, as in the analysis on which the endpoints were first evaluated; the predictors use readings floored at 1 mm³ (`FORECAST_OUTCOME_ZEROS` in Section 1.1 switches the outcome to the floor for the sensitivity analysis). Arms observed for less than 24 days are not included here (Section 8.1 reinserts them). Two predictor sets are compared here; they and the control sets of Section 8 are listed in `predictor_sets.csv`. The depth set holds baseline log volume and best average response within the window. The trajectory set adds eight shape descriptors; one of them, nadir depth, is itself depth-derived and is also in the expanded depth set. Volume at the end of the window is in neither set. Both sets are fitted with the same gradient-boosted regressor (300 iterations, learning rate 0.05, L2 penalty 1.0, 15 leaves per tree) on the same five folds, each PDX model kept within one fold. The primary population is post-nadir arms. That criterion uses each arm's complete trajectory, but no measurement after day 10 enters any predictor.

**Prespecified endpoints.** The first is the relative reduction in out-of-fold root-mean-square error (RMSE), with a margin of 10%. The second is the relative reduction in mean regret, with a margin of 15%. For each PDX model with at least two eligible arms, each predictor set selects the arm with the smallest predicted day-24 volume. Regret is that arm's observed log volume minus the smallest observed log volume among the model's eligible arms. An endpoint is met when its margin is reached and its 95% bootstrap interval, resampling PDX models 2,000 times with out-of-fold predictions held fixed, excludes zero. Holm correction is applied across the two endpoints. Pairwise ranking accuracy counts tied predictions as one half and leaves out pairs whose observed outcomes are tied.

**Inputs:** `observations`, `post_nadir_ids`.
**Outputs:** `PREDICTOR_SETS`, `forecast_primary`, `selection`, `regret`, `forecast_pairs`, table `predictor_sets.csv`; figure data `forecast_predictions.csv`, `forecast_bootstrap_draws.csv` and `ranking_pairs.csv`.

```python
# ==========================================================================
# 7.1 Predictor sets and the forecasting table
#     The depth and trajectory sets are the two compared by the prespecified
#     endpoints; the others are controls, fitted in Section 8.
# ==========================================================================

set_section("7")
DEPTH_SET = ["baseline_log_volume", "response_in_window"]
SHAPE_DESCRIPTORS = ["overall_slope", "initial_slope", "steepest_decline", "nadir_depth",
                     "time_to_nadir", "post_nadir_slope_in_window", "curvature", "slope_change"]
TRAJECTORY_SET = DEPTH_SET + SHAPE_DESCRIPTORS
PREDICTOR_SETS = {
    "depth": DEPTH_SET,
    "trajectory": TRAJECTORY_SET,
    "baseline only": ["baseline_log_volume"],
    "expanded depth": DEPTH_SET + ["nadir_depth", "nadir_ratio", "minimum_log_volume"],
    "noise-padded depth": DEPTH_SET + [f"noise_{i}" for i in range(len(SHAPE_DESCRIPTORS))],
    "permuted shape": ["baseline_log_volume"] + [f"permuted_{c}" for c in SHAPE_DESCRIPTORS],
    "depth plus current volume": DEPTH_SET + ["current_log_volume"],
    # Robustness sets (Section 8): the permuted descriptors added to the full
    # depth set, so that the comparison with the trajectory set differs only
    # in whether the shape descriptors carry their real values; and the
    # simplest trajectory summary, baseline volume and the net slope of log
    # volume over the window (in treated arms the slope includes the drug's
    # effect, so this is not a growth-only set).
    "depth plus permuted shape": DEPTH_SET + [f"permuted_{c}" for c in SHAPE_DESCRIPTORS],
    "baseline plus slope": ["baseline_log_volume", "overall_slope"],
}
PREDICTOR_SET_ROLES = {
    "depth": "Comparator for both primary endpoints",
    "trajectory": "Tested set",
    "baseline only": "Floor",
    "expanded depth": "Tests whether shape adds to all depth information",
    "noise-padded depth": "Controls for the number of predictors",
    "permuted shape": "Keeps each descriptor's distribution, removes its association with outcome",
    "depth plus current volume": "Tests whether the advantage is the tumor's size at the end of the window",
    "depth plus permuted shape": "Same predictors as the trajectory set, shape descriptors permuted across arms",
    "baseline plus slope": "Simplest trajectory summary; shows how much of the advantage one early slope carries",
}
predictor_set_table = pd.DataFrame({"set": list(PREDICTOR_SETS),
                        "predictors": ["; ".join(v) for v in PREDICTOR_SETS.values()],
                        "role": [PREDICTOR_SET_ROLES[k] for k in PREDICTOR_SETS]})
predictor_set_table.to_csv(TABLE_DIR / "predictor_sets.csv", index=False)

forecast_table = build_forecast_table(observations.loc[~observations.is_control],
                                      FORECAST_ORIGIN_DAY, FORECAST_HORIZON_DAY)
forecast_primary = (forecast_table.loc[forecast_table.arm_id.isin(post_nadir_ids)]
                    .dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"]).reset_index(drop=True))
outcome = forecast_primary.outcome_log_volume.to_numpy(float)
groups = forecast_primary.model_id.astype(str).to_numpy()
print("7.1 FORECASTING TABLE")
print(f"  treated arms with a day-{FORECAST_ORIGIN_DAY} window and an observed day-{FORECAST_HORIZON_DAY} "
      f"volume   {len(forecast_table):,}")
print(f"  post-nadir arms with every predictor            "
      f"{record('forecast_n_arms', len(forecast_primary)):,} in "
      f"{record('forecast_n_models', len(np.unique(groups)))} PDX models")

# ==========================================================================
# 7.2 First endpoint: out-of-fold forecasting error
# ==========================================================================

forecast_folds = grouped_folds(forecast_primary)
depth_prediction = out_of_fold_predictions(forecast_primary, DEPTH_SET, outcome, forecast_folds)
trajectory_prediction = out_of_fold_predictions(forecast_primary, TRAJECTORY_SET, outcome, forecast_folds)
forecast_reduction = relative_rmse_reduction(outcome, depth_prediction, trajectory_prediction)
forecast_rows_by_model = rows_by_cluster(groups)
forecast_draws = cluster_bootstrap(
    forecast_rows_by_model,
    lambda rows: relative_rmse_reduction(outcome[rows], depth_prediction[rows], trajectory_prediction[rows]),
    N_BOOTSTRAP, np.random.default_rng(SEEDS["forecast_bootstrap"]))
forecast_low, forecast_high = np.percentile(forecast_draws, [2.5, 97.5])
forecast_p = bootstrap_p_two_sided(forecast_draws)

record("depth_rmse", rmse(outcome, depth_prediction))
record("trajectory_rmse", rmse(outcome, trajectory_prediction))
record("depth_r2", r2_score(outcome, depth_prediction))
record("trajectory_r2", r2_score(outcome, trajectory_prediction))
record("forecast_reduction", forecast_reduction)
record("forecast_reduction_low", forecast_low)
record("forecast_reduction_high", forecast_high)
print(f"\n7.2 FORECASTING, day {FORECAST_ORIGIN_DAY} to day {FORECAST_HORIZON_DAY} "
      f"({len(forecast_primary)} arms, {len(np.unique(groups))} PDX models)")
print(f"  depth set        RMSE {RESULTS['depth_rmse']:.3f}   R² {RESULTS['depth_r2']:.3f}")
print(f"  trajectory set   RMSE {RESULTS['trajectory_rmse']:.3f}   R² {RESULTS['trajectory_r2']:.3f}")
print(f"  relative reduction in RMSE   {format_ci(forecast_reduction, forecast_low, forecast_high)}")
print(f"  margin {FORECAST_MARGIN:.0%} met: {forecast_reduction >= FORECAST_MARGIN}   "
      f"interval excludes zero: {forecast_low > 0}")

# ==========================================================================
# 7.3 Second endpoint: treatment-selection regret, and ranking accuracy
# ==========================================================================

selection = forecast_primary.assign(depth_prediction=depth_prediction,
                                    trajectory_prediction=trajectory_prediction)
regret = selection_by_model(selection, ["depth_prediction", "trajectory_prediction"])
depth_regret = float(regret.depth_prediction_regret.mean())
trajectory_regret = float(regret.trajectory_prediction_regret.mean())
regret_reduction = (depth_regret - trajectory_regret) / depth_regret
regret_rng = np.random.default_rng(SEEDS["regret_bootstrap"])
regret_draws = []
for _ in range(N_BOOTSTRAP):
    drawn = regret.sample(len(regret), replace=True, random_state=int(regret_rng.integers(1 << 31)))
    depth_mean = drawn.depth_prediction_regret.mean()
    if depth_mean > 0:
        regret_draws.append((depth_mean - drawn.trajectory_prediction_regret.mean()) / depth_mean)
regret_low, regret_high = np.percentile(regret_draws, [2.5, 97.5])
regret_p = bootstrap_p_two_sided(regret_draws)
holm_thresholds, holm_rejected = holm([forecast_p, regret_p])

forecast_pairs = within_model_pairs(selection, ["depth_prediction", "trajectory_prediction"])
record("regret_n_models", len(regret))
record("depth_regret", depth_regret)
record("trajectory_regret", trajectory_regret)
record("depth_regret_fold", float(np.exp(depth_regret)))
record("trajectory_regret_fold", float(np.exp(trajectory_regret)))
record("regret_reduction", regret_reduction)
record("regret_reduction_low", regret_low)
record("regret_reduction_high", regret_high)
record("forecast_p", forecast_p)
record("regret_p", regret_p)
record("n_different_choice", int(regret.different_choice.sum()))
record("share_different_choice", float(regret.different_choice.mean()))
record("ranking_accuracy_depth", float(forecast_pairs.depth_prediction.mean()))
record("ranking_accuracy_trajectory", float(forecast_pairs.trajectory_prediction.mean()))
record("n_ranking_pairs", len(forecast_pairs))
# Each endpoint is met when its margin is reached and its interval excludes
# zero; the Holm decisions are stored beside the endpoint decisions.
record("forecast_endpoint_met", int(forecast_reduction >= FORECAST_MARGIN and forecast_low > 0))
record("regret_endpoint_met", int(regret_reduction >= REGRET_MARGIN and regret_low > 0))
record("forecast_holm_significant", int(holm_rejected[0]))
record("regret_holm_significant", int(holm_rejected[1]))

print(f"\n7.3 TREATMENT SELECTION ({len(regret)} PDX models with at least two eligible arms)")
print(f"  mean regret, depth set        {depth_regret:.3f} log units ({np.exp(depth_regret):.2f}-fold)")
print(f"  mean regret, trajectory set   {trajectory_regret:.3f} log units ({np.exp(trajectory_regret):.2f}-fold)")
print(f"  relative reduction in regret  {format_ci(regret_reduction, regret_low, regret_high)}")
print(f"  margin {REGRET_MARGIN:.0%} met: {regret_reduction >= REGRET_MARGIN}   "
      f"interval excludes zero: {regret_low > 0}")
print(f"  PDX models where the two sets select different arms   "
      f"{int(regret.different_choice.sum())} ({100 * regret.different_choice.mean():.1f}%)")
print(f"  pairwise ranking accuracy ({len(forecast_pairs):,} pairs)   depth "
      f"{forecast_pairs.depth_prediction.mean():.3f}   trajectory {forecast_pairs.trajectory_prediction.mean():.3f}")
print("\n  Holm correction across the two endpoints")
for name, p_value, threshold, rejected in zip(("forecasting", "regret"), (forecast_p, regret_p),
                                              holm_thresholds, holm_rejected):
    print(f"    {name:<12} {p_label(p_value, 1 / N_BOOTSTRAP):<12} threshold {threshold:.3f}   "
          f"significant: {rejected}")
save_record(pd.DataFrame([{
    "n_arms": len(forecast_primary), "n_models": len(np.unique(groups)),
    "depth_rmse": RESULTS["depth_rmse"], "trajectory_rmse": RESULTS["trajectory_rmse"],
    "rmse_reduction": forecast_reduction, "rmse_low": forecast_low, "rmse_high": forecast_high,
    "rmse_p": forecast_p, "regret_models": len(regret), "depth_regret": depth_regret,
    "trajectory_regret": trajectory_regret, "regret_reduction": regret_reduction,
    "regret_low": regret_low, "regret_high": regret_high, "regret_p": regret_p}]),
    "forecasting_endpoints")
save_record(regret, "treatment_selection_regret")

# Data for the forecasting and selection figure: each arm's observed day-24
# log volume with both out-of-fold predictions, the bootstrap distribution
# of the relative RMSE reduction, and every within-model pair of 7.3 (the
# per-model regret table is saved above).
save_record(selection[["arm_id", "model_id", "treatment", "tumor_type", "outcome_log_volume",
                       "depth_prediction", "trajectory_prediction"]], "forecast_predictions")
save_record(pd.DataFrame({"relative_reduction": forecast_draws}), "forecast_bootstrap_draws")
save_record(forecast_pairs, "ranking_pairs")
```

## 8 · Controls and robustness analyses

**Control predictor sets (8.1).** Seven control sets are fitted with the same learner and folds as Section 7 (all sets are listed in `predictor_sets.csv`). Baseline only is the floor. Expanded depth tests whether shape adds to every depth-derived quantity in the window. Noise-padded depth adds eight Gaussian noise predictors to the depth set, so its predictor count matches the trajectory set's. Permuted shape is baseline volume plus the eight shape descriptors permuted across arms: it keeps each descriptor's distribution but removes its association with the outcome. Because it also lacks best average response, its comparison with the trajectory set mixes the loss of shape with the loss of depth. Depth plus permuted shape removes that mixture: it holds exactly the predictors of the trajectory set, with the shape descriptors permuted, so its comparison with the trajectory set isolates the information carried by the real shape values. Depth plus current volume tests whether the advantage reduces to the tumor's size at the end of the window. Baseline plus slope holds the two simplest trajectory quantities, baseline volume and the slope of log volume over the window; it shows how much of the advantage one early slope already carries. In a treated arm that slope is the net rate of change under treatment, so it includes the drug's effect and is not a growth-only control. Each comparison is a relative reduction in RMSE with a 95% interval from 1,500 bootstrap resamples of PDX models.

**Robustness (8.1).** Forecasts are repeated at three start days (10, 14 and 21) and three prediction horizons (10, 14 and 21 days later), and with each tumor type held out of training in turn. Each of these comparisons carries a 95% interval from 2,000 resamples of PDX models with predictions held fixed. Arms that ended before day 24 are reinserted under two scenarios: growth stopped at the last observed volume, or growth continued at the arm's own whole-arm slope (non-negative). These are scenarios, not bounds.

**Without the post-nadir restriction (8.2).** The primary population is restricted to post-nadir arms, a criterion that uses the complete trajectory. The forecast is repeated on every treated arm with a complete day-10 window and an observed day-24 volume.

**Negative and positive controls (8.3).** Untreated arms cannot truly regress, so they serve as negative controls in three ways. First, the durability-phenotype rule applied to treated arms (Section 3.3) is applied unchanged to untreated arms. Second, the tolerant-subpopulation model is fitted to untreated arms, whose true kill rate is zero; treated arms are fitted the same way (growth rate free) for comparison. Third, the forecast is repeated in untreated arms and in all treated arms over the same windows. Forecasting starts at day 10 or later, because best average response is not defined before day 10. The positive control compares depth between BRAF V600-mutant and wild-type PDX models under MAPK-pathway inhibitors. The approved combinations are the second positive control and appear in Section 10.

**Ranking accuracy by separation (8.4).** Within-model treatment pairs from Section 7 are grouped by the observed gap between their day-24 log volumes, in units of the measurement noise (Section 3.6). Ranking accuracy is reported within each group, with the trajectory-minus-depth difference and a 95% interval from resampling PDX models.

**Baseline plus slope and treatment selection (8.5).** The baseline-plus-slope set is taken through the selection endpoints of Section 7 (regret and pairwise ranking accuracy, overall and by separation) and compared with the trajectory set, with intervals from resampling PDX models. The comparison shows what the remaining shape descriptors add beyond a single early slope. It does not separate drug effect from intrinsic growth; that separation comes from the design of the selection endpoint itself: treatments are ranked within a PDX model, and a PDX model's intrinsic growth rate is shared by all of its arms, so it cannot order them.

**Inputs:** `forecast_primary`, `observations`, `post_nadir_ids`, `PREDICTOR_SETS`, `forecast_pairs`, `regrowth`, `MEASUREMENT_NOISE`, `sheets["mutation"]`.
**Outputs:** `predictor_set_performance`, `predictor_contrasts`, `rolling_origins`, `tumor_type_holdout`, `censoring_scenarios`, `untreated_controls`, `untreated_forecasting`, `positive_control`, `separability`, `baseline_slope_selection`; figure data `forecast_predictions_all_sets.csv`, `untreated_regrowth_summary.csv` and `treated_regrowth_summary.csv`.

```python
# ==========================================================================
# 8.1 Control predictor sets, start days and horizons, held-out tumor types,
#     and reinsertion of arms that ended before day 24
# ==========================================================================

set_section("8")
control_rng = np.random.default_rng(SEEDS["predictor_controls"])
validation = forecast_primary.copy()
for index in range(len(SHAPE_DESCRIPTORS)):
    validation[f"noise_{index}"] = control_rng.normal(size=len(validation))
for column in SHAPE_DESCRIPTORS:
    validation[f"permuted_{column}"] = validation[column].to_numpy()[control_rng.permutation(len(validation))]

set_predictions, set_errors = compare_predictor_sets(validation, PREDICTOR_SETS, forecast_folds)
predictor_set_performance = pd.DataFrame([
    {"predictor_set": name, "predictors": len(PREDICTOR_SETS[name]), "rmse": set_errors[name],
     "r2": r2_score(outcome, set_predictions[name])} for name in PREDICTOR_SETS]).sort_values("rmse")
print("8.1 PREDICTOR SETS, same learner and folds as Section 7")
print(predictor_set_performance.round(3).to_string(index=False))

CONTRASTS = [("depth", "trajectory"), ("expanded depth", "trajectory"),
             ("noise-padded depth", "trajectory"), ("permuted shape", "trajectory"),
             ("depth", "expanded depth"), ("depth", "noise-padded depth"),
             ("depth plus current volume", "trajectory"), ("depth", "depth plus current volume"),
             # Robustness contrasts, appended so that the resamples drawn for
             # the contrasts above are unchanged.
             ("depth plus permuted shape", "trajectory"), ("baseline plus slope", "trajectory"),
             ("depth", "baseline plus slope")]
contrast_rows = []
for baseline, challenger in CONTRASTS:
    draws = cluster_bootstrap(
        forecast_rows_by_model,
        lambda rows: relative_rmse_reduction(outcome[rows], set_predictions[baseline][rows],
                                             set_predictions[challenger][rows]),
        N_BOOTSTRAP_CONTROLS, control_rng)
    low, high = np.percentile(draws, [2.5, 97.5])
    contrast_rows.append({"challenger": challenger, "baseline": baseline,
                          "relative_reduction": relative_rmse_reduction(
                              outcome, set_predictions[baseline], set_predictions[challenger]),
                          "low": low, "high": high})
predictor_contrasts = pd.DataFrame(contrast_rows)
print(f"\n  relative reduction in RMSE, 95% interval from {N_BOOTSTRAP_CONTROLS:,} resamples of PDX models")
for row in predictor_contrasts.itertuples(index=False):
    print(f"    {row.challenger:<26} against {row.baseline:<26} "
          f"{format_ci(row.relative_reduction, row.low, row.high)}")
contrast_lookup = predictor_contrasts.set_index(["baseline", "challenger"])
for key, (baseline, challenger) in {"expanded_over_depth": ("depth", "expanded depth"),
                                    "trajectory_over_expanded": ("expanded depth", "trajectory"),
                                    "noise_over_depth": ("depth", "noise-padded depth"),
                                    "trajectory_over_current_volume": ("depth plus current volume", "trajectory"),
                                    "trajectory_over_permuted_shape": ("depth plus permuted shape", "trajectory"),
                                    "trajectory_over_baseline_slope": ("baseline plus slope", "trajectory"),
                                    "baseline_slope_over_depth": ("depth", "baseline plus slope"),
                                    }.items():
    row = contrast_lookup.loc[(baseline, challenger)]
    record(key, row.relative_reduction)
    record(f"{key}_low", row.low)
    record(f"{key}_high", row.high)
record("permuted_shape_rmse", set_errors["permuted shape"])
record("baseline_only_rmse", set_errors["baseline only"])
record("baseline_slope_rmse", set_errors["baseline plus slope"])
record("depth_plus_permuted_shape_rmse", set_errors["depth plus permuted shape"])

# --- start days and horizons -----------------------------------------------
# Each window carries a 95% interval from resampling PDX models with its
# out-of-fold predictions held fixed (one generator for all windows).
rolling_rng = np.random.default_rng(SEEDS["rolling_bootstrap"])
rolling_rows = []
for origin_day in (10, 14, 21):
    for lead_days in (10, 14, 21):
        table = build_forecast_table(observations.loc[~observations.is_control], origin_day,
                                     origin_day + lead_days, arm_ids=post_nadir_ids)
        table = table.dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"]).reset_index(drop=True)
        if len(table) < 60 or table.model_id.nunique() < 2:
            continue
        predictions, errors = compare_predictor_sets(table, {"depth": DEPTH_SET, "trajectory": TRAJECTORY_SET})
        _, low, high, _ = reduction_with_interval(
            table.outcome_log_volume.to_numpy(float), predictions["depth"], predictions["trajectory"],
            rows_by_cluster(table.model_id.astype(str)), N_BOOTSTRAP, rolling_rng)
        rolling_rows.append({"start_day": origin_day, "horizon_day": origin_day + lead_days,
                             "arms": len(table), "pdx_models": table.model_id.nunique(),
                             "depth_rmse": errors["depth"], "trajectory_rmse": errors["trajectory"],
                             "relative_reduction": (errors["depth"] - errors["trajectory"]) / errors["depth"],
                             "low": low, "high": high})
rolling_origins = pd.DataFrame(rolling_rows)
record("n_rolling_windows", len(rolling_origins))
record("rolling_reduction_min", rolling_origins.relative_reduction.min())
record("rolling_reduction_max", rolling_origins.relative_reduction.max())
record("rolling_windows_interval_above_zero", int((rolling_origins.low > 0).sum()))
record("rolling_windows_above_margin", int((rolling_origins.relative_reduction >= FORECAST_MARGIN).sum()))
print("\n  start days and horizons (post-nadir arms)")
print(rolling_origins.round(3).to_string(index=False))

# --- each tumor type held out of training ----------------------------------
# Intervals resample the held-out PDX models with predictions held fixed.
holdout_rng = np.random.default_rng(SEEDS["holdout_bootstrap"])
holdout_rows = []
for held_out in sorted(forecast_primary.tumor_type.unique()):
    train = forecast_primary.loc[forecast_primary.tumor_type != held_out]
    test = forecast_primary.loc[forecast_primary.tumor_type == held_out]
    if len(test) < 30 or len(train) < 100:
        continue
    row = {"held_out_tumor_type": held_out, "train_arms": len(train), "test_arms": len(test),
           "test_pdx_models": test.model_id.nunique()}
    held_out_predictions = {}
    for name, predictors in (("depth", DEPTH_SET), ("trajectory", TRAJECTORY_SET)):
        learner = make_forecaster().fit(train[predictors].to_numpy(float),
                                        train.outcome_log_volume.to_numpy(float))
        predicted = learner.predict(test[predictors].to_numpy(float))
        held_out_predictions[name] = predicted
        row[f"{name}_rmse"] = rmse(test.outcome_log_volume, predicted)
        row[f"{name}_r2"] = r2_score(test.outcome_log_volume, predicted)
    row["relative_reduction"] = (row["depth_rmse"] - row["trajectory_rmse"]) / row["depth_rmse"]
    _, row["low"], row["high"], _ = reduction_with_interval(
        test.outcome_log_volume.to_numpy(float), held_out_predictions["depth"],
        held_out_predictions["trajectory"], rows_by_cluster(test.model_id.astype(str)),
        N_BOOTSTRAP, holdout_rng)
    holdout_rows.append(row)
    record(f"holdout_{held_out}", row["relative_reduction"])
tumor_type_holdout = pd.DataFrame(holdout_rows)
lowest_holdout = tumor_type_holdout.loc[tumor_type_holdout.relative_reduction.idxmin()]
others = tumor_type_holdout.drop(index=lowest_holdout.name)
record("holdout_types_above_margin", int((tumor_type_holdout.relative_reduction >= FORECAST_MARGIN).sum()))
record("holdout_others_min", others.relative_reduction.min())
record("holdout_others_max", others.relative_reduction.max())
record("holdout_lowest_type", str(lowest_holdout.held_out_tumor_type))
record("holdout_types_interval_above_zero", int((tumor_type_holdout.low > 0).sum()))
print("\n  each tumor type held out of training")
print(tumor_type_holdout.round(3).to_string(index=False))

# --- arms that ended before day 24, reinserted ------------------------------
scenario_rows = []
for arm_id, block in arm_series(observations, post_nadir_ids):
    day, log_volume = block.day.to_numpy(float), block.log_volume.to_numpy(float)
    features = early_window_features(day, log_volume, FORECAST_ORIGIN_DAY) if len(day) >= 3 else None
    if features is None:
        continue
    if day.max() >= FORECAST_HORIZON_DAY:
        observed = float(np.interp(FORECAST_HORIZON_DAY, day, log_volume))
        stopped = continued = observed
    else:
        slope = float(np.polyfit(day, log_volume, 1)[0])
        stopped = float(log_volume[-1])
        continued = stopped + (FORECAST_HORIZON_DAY - day.max()) * max(0.0, slope)
    scenario_rows.append({"arm_id": arm_id, "model_id": block.model_id.iloc[0],
                          "ended_early": day.max() < FORECAST_HORIZON_DAY,
                          "growth_stopped": stopped, "growth_continued": continued, **features})
scenario_frame = pd.DataFrame(scenario_rows).dropna(subset=TRAJECTORY_SET).reset_index(drop=True)
scenario_results = []
for scenario in ("growth_stopped", "growth_continued"):
    frame = scenario_frame.assign(outcome_log_volume=scenario_frame[scenario])
    _, errors = compare_predictor_sets(frame, {"depth": DEPTH_SET, "trajectory": TRAJECTORY_SET})
    scenario_results.append({"scenario": scenario.replace("_", " "), "arms": len(frame),
                             "depth_rmse": errors["depth"], "trajectory_rmse": errors["trajectory"],
                             "relative_reduction": (errors["depth"] - errors["trajectory"]) / errors["depth"]})
censoring_scenarios = pd.DataFrame(scenario_results)
record("n_ended_early", int(scenario_frame.ended_early.sum()))
record("share_ended_early", float(scenario_frame.ended_early.mean()))
record("scenario_stopped", censoring_scenarios.relative_reduction.iloc[0])
record("scenario_continued", censoring_scenarios.relative_reduction.iloc[1])
print(f"\n  arms that ended before day {FORECAST_HORIZON_DAY}, reinserted: "
      f"{int(scenario_frame.ended_early.sum())} of {len(scenario_frame)} "
      f"({100 * scenario_frame.ended_early.mean():.1f}%)")
print(censoring_scenarios.round(3).to_string(index=False))
for name, frame in (("predictor_set_performance", predictor_set_performance),
                    ("predictor_contrasts", predictor_contrasts), ("rolling_origins", rolling_origins),
                    ("tumor_type_holdout", tumor_type_holdout), ("censoring_scenarios", censoring_scenarios)):
    save_record(frame, name)

# Data for the controls figure: the out-of-fold predictions of every
# predictor set on the Section 7 arms (the performance, contrast, window and
# held-out tables are saved above).
save_record(validation[["arm_id", "model_id", "outcome_log_volume"]].assign(
    **{f"prediction_{name.replace(' ', '_')}": values for name, values in set_predictions.items()}),
    "forecast_predictions_all_sets")
```

```python
# ==========================================================================
# 8.2 Forecasting without the post-nadir restriction, day 10 to day 24
#     Same learner, predictor sets and grouped folds as Section 7; every
#     treated arm with a complete day-10 window and an observed day-24 volume.
# ==========================================================================

forecast_all = (forecast_table.dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"])
                .reset_index(drop=True))
outcome_all = forecast_all.outcome_log_volume.to_numpy(float)
groups_all = forecast_all.model_id.astype(str).to_numpy()
folds_all = grouped_folds(forecast_all)
depth_all = out_of_fold_predictions(forecast_all, DEPTH_SET, outcome_all, folds_all)
trajectory_all = out_of_fold_predictions(forecast_all, TRAJECTORY_SET, outcome_all, folds_all)
reduction_all = relative_rmse_reduction(outcome_all, depth_all, trajectory_all)
# Resample PDX models as for the primary endpoint. The seed is the forecasting
# seed plus one, which leaves the Section 7 stream unchanged.
draws_all = cluster_bootstrap(
    rows_by_cluster(groups_all),
    lambda rows: relative_rmse_reduction(outcome_all[rows], depth_all[rows], trajectory_all[rows]),
    N_BOOTSTRAP, np.random.default_rng(SEEDS["forecast_bootstrap"] + 1))
low_all, high_all = np.percentile(draws_all, [2.5, 97.5])
record("forecast_all_n_arms", len(forecast_all))
record("forecast_all_reduction", reduction_all)
record("forecast_all_reduction_low", low_all)
record("forecast_all_reduction_high", high_all)
record("forecast_all_n_models", len(np.unique(groups_all)))
print(f"8.2 FORECASTING WITHOUT THE POST-NADIR RESTRICTION, "
      f"day {FORECAST_ORIGIN_DAY} to day {FORECAST_HORIZON_DAY}")
print(f"  arms {len(forecast_all):,} in {len(np.unique(groups_all))} PDX models")
print(f"  relative reduction in RMSE   {format_ci(reduction_all, low_all, high_all)}")
```

```python
# ==========================================================================
# 8.3 Negative controls (untreated arms) and the mutation-matched positive
#     control
# ==========================================================================

# --- (a) the durability-phenotype rule applied to untreated arms -----------
untreated_regrowth_rows = []
for arm_id, block in arm_series(observations.loc[observations.is_control]):
    summary = summarize_regrowth(block.day.to_numpy(float), block.log_volume.to_numpy(float))
    if summary is not None:
        untreated_regrowth_rows.append({"arm_id": arm_id, **summary,
                               "durability_phenotype": durability_phenotype(summary)})
untreated_controls = pd.DataFrame(untreated_regrowth_rows)
untreated_interior = ((untreated_controls.nadir_index > 0)
                      & (untreated_controls.n_after_nadir >= MIN_POINTS_AFTER_NADIR))
record("untreated_phenotype_share", float(untreated_controls.durability_phenotype.mean()))
print("8.3 NEGATIVE CONTROLS")
print("  (a) durability-phenotype rule          untreated            treated")
print(f"      arms assessed                      {len(untreated_controls):>9,}         {len(regrowth):>9,}")
print(f"      interior nadir with >= 2 later      {100 * untreated_interior.mean():>8.1f}%"
      f"         {100 * RESULTS['treated_interior_nadir_share']:>8.1f}%")
print(f"      all three conditions (phenotype)    {100 * untreated_controls.durability_phenotype.mean():>8.1f}%"
      f"         {100 * RESULTS['treated_phenotype_share']:>8.1f}%")
spurious = untreated_controls.loc[untreated_controls.durability_phenotype, "post_nadir_growth_rate"]
print(f"      post-nadir growth rate, median      {spurious.median():>9.4f}         "
      f"{regrowth.loc[regrowth.durability_phenotype, 'post_nadir_growth_rate'].median():>9.4f}  per day")

# --- (b) apparent kill rates, growth rate free in both groups ---------------
tolerant_forward = constant_exposure_forward("tolerant subpopulation")
free_fit_rows = []
for arm_id, block in arm_series(observations):
    if len(block) < MIN_MEASUREMENTS_PER_ARM:
        continue
    day = block.day.to_numpy(float) - block.day.iloc[0]
    log_volume = block.log_volume.to_numpy(float)
    fit = fit_bounded(tolerant_forward, MECHANISM_BOUNDS["tolerant subpopulation"], day,
                      log_volume, mechanism_starts(log_volume))
    free_fit_rows.append({"arm_id": arm_id, "is_control": bool(block.is_control.iloc[0]),
                          "kill_rate": tolerant_estimates(fit["x"])["kill_rate"],
                          "pinned": fit["pinned"]})
free_fits = pd.DataFrame(free_fit_rows)
treated_free_median = float(free_fits.loc[~free_fits.is_control, "kill_rate"].median())
untreated_free = free_fits.loc[free_fits.is_control, "kill_rate"]
record("untreated_kill_above_treated_median", float((untreated_free > treated_free_median).mean()))
print(f"\n  (b) tolerant-subpopulation fits with the growth rate free "
      f"({int(free_fits.is_control.sum())} untreated, {int((~free_fits.is_control).sum()):,} treated arms)")
print(f"      median apparent kill rate, untreated   {untreated_free.median():.4f} per day")
print(f"      median kill rate, treated              {treated_free_median:.4f} per day")
print(f"      untreated arms above the treated median   "
      f"{100 * (untreated_free > treated_free_median).mean():.1f}%")
print(f"      fits ending on the lower kill-rate bound (0.001 per day): untreated "
      f"{100 * free_fits.loc[free_fits.is_control, 'pinned'].str.contains('log_kill@low').mean():.1f}%, "
      f"treated {100 * free_fits.loc[~free_fits.is_control, 'pinned'].str.contains('log_kill@low').mean():.1f}%")

# --- (c) forecasting in untreated arms and in all treated arms --------------
CONTROL_WINDOWS = [(10, 17), (10, 20), (10, 24), (14, 21), (14, 28)]
MIN_UNTREATED_ARMS, MIN_UNTREATED_MODELS = 80, 20
forecast_rows = []
window_census = []
for origin_day, horizon_day in CONTROL_WINDOWS:
    populations = {}
    for label, is_control in (("untreated", True), ("treated", False)):
        table = build_forecast_table(observations.loc[observations.is_control == is_control],
                                     origin_day, horizon_day)
        populations[label] = (table.dropna(subset=TRAJECTORY_SET + ["outcome_log_volume"])
                              .reset_index(drop=True) if len(table) else table)
    window_census.append({
        "start_day": origin_day, "horizon_day": horizon_day,
        **{f"{label}_arms": len(frame) for label, frame in populations.items()},
        **{f"{label}_interior_nadir_share": float((frame.time_to_nadir > 0).mean()) if len(frame) else np.nan
           for label, frame in populations.items()}})
    untreated_frame = populations["untreated"]
    if len(untreated_frame) < MIN_UNTREATED_ARMS or untreated_frame.model_id.nunique() < MIN_UNTREATED_MODELS:
        continue
    for label, frame in populations.items():
        frame_outcome = frame.outcome_log_volume.to_numpy(float)
        predictions, errors = compare_predictor_sets(frame, {"depth": DEPTH_SET, "trajectory": TRAJECTORY_SET})
        draws = cluster_bootstrap(
            rows_by_cluster(frame.model_id.astype(str)),
            lambda rows: relative_rmse_reduction(frame_outcome[rows], predictions["depth"][rows],
                                                 predictions["trajectory"][rows]),
            N_BOOTSTRAP, np.random.default_rng(SEEDS["untreated_forecast_bootstrap"]))
        low, high = np.percentile(draws, [2.5, 97.5])
        forecast_rows.append({"population": label, "start_day": origin_day, "horizon_day": horizon_day,
                              "arms": len(frame), "pdx_models": frame.model_id.nunique(),
                              "relative_reduction": (errors["depth"] - errors["trajectory"]) / errors["depth"],
                              "low": low, "high": high})
untreated_forecasting = pd.DataFrame(forecast_rows)
print("\n  (c) forecasting in untreated and treated arms over the same windows")
print(pd.DataFrame(window_census).round(3).to_string(index=False))
print(f"  windows with >= {MIN_UNTREATED_ARMS} untreated arms in >= {MIN_UNTREATED_MODELS} PDX models are fitted")
for row in untreated_forecasting.itertuples(index=False):
    print(f"    {row.population:<9} day {row.start_day} to {row.horizon_day}   {row.arms:>5,} arms   "
          f"{format_ci(row.relative_reduction, row.low, row.high)}")
for row in untreated_forecasting.loc[(untreated_forecasting.start_day == 14)
                                     & (untreated_forecasting.horizon_day == 21)].itertuples(index=False):
    record(f"{row.population}_forecast_14_21", row.relative_reduction)
    record(f"{row.population}_forecast_14_21_low", row.low)
    record(f"{row.population}_forecast_14_21_high", row.high)

# --- positive control: depth by BRAF V600 status under MAPK-pathway agents ---
# The mutation sheet holds one row per finding; the first column is the PDX
# model. The gene column is the one in which "BRAF" occurs most often. A
# finding is BRAF V600 when any of its text fields contains "V600".
mutation = sheets["mutation"]
mutation_model_column = mutation.columns[0]
braf_counts = {column: int((mutation[column].astype(str).str.upper() == "BRAF").sum())
               for column in mutation.columns[1:]}
gene_column = max(braf_counts, key=braf_counts.get)
braf_rows = mutation.loc[mutation[gene_column].astype(str).str.upper() == "BRAF"]
other_text = braf_rows.drop(columns=[mutation_model_column, gene_column]).astype(str)
v600_rows = braf_rows.loc[other_text.apply(lambda c: c.str.contains("V600", case=False)).any(axis=1)]
braf_v600_models = set(v600_rows[mutation_model_column].astype(str))
MAPK_AGENTS = ["encorafenib", "binimetinib", "binimetinib-3.5mpk", "trametinib", "encorafenib + binimetinib"]
positive_rows = []
for agent in MAPK_AGENTS:
    agent_arms = arms.loc[(arms.treatment == agent) & arms.model_id.astype(str).isin(model_sets["mutation"])
                          & arms.best_average_response.notna()]
    mutant = agent_arms.model_id.astype(str).isin(braf_v600_models)
    if mutant.sum() < 3 or (~mutant).sum() < 3:
        continue
    test = stats.mannwhitneyu(agent_arms.loc[mutant, "best_average_response"],
                              agent_arms.loc[~mutant, "best_average_response"], alternative="less")
    positive_rows.append({"agent": agent, "braf_v600_models": int(mutant.sum()),
                          "wild_type_models": int((~mutant).sum()),
                          "median_response_v600": agent_arms.loc[mutant, "best_average_response"].median(),
                          "median_response_wild_type": agent_arms.loc[~mutant, "best_average_response"].median(),
                          "one_sided_p": test.pvalue})
positive_control = pd.DataFrame(positive_rows)
record("braf_v600_models", len(braf_v600_models))
for row in positive_control.itertuples(index=False):
    tag = "".join(character if character.isalnum() else "_" for character in row.agent)
    record(f"positive_control_{tag}_median_v600", float(row.median_response_v600))
    record(f"positive_control_{tag}_median_wild_type", float(row.median_response_wild_type))
    record(f"positive_control_{tag}_p", float(row.one_sided_p))
print("\n  POSITIVE CONTROL: best average response by BRAF V600 status")
print(f"    gene column '{gene_column}'; BRAF V600 findings in {len(braf_v600_models)} PDX models "
      f"of {len(model_sets['mutation'])} with mutation data")
if len(positive_control):
    print(positive_control.assign(one_sided_p=positive_control.one_sided_p.map(format_p))
          .round(2).to_string(index=False))
save_record(untreated_forecasting, "untreated_forecasting")
# The phenotype rule's inputs for untreated and treated arms, for the
# negative-control panels of the figures notebook.
save_record(untreated_controls, "untreated_regrowth_summary")
save_record(regrowth.reset_index(), "treated_regrowth_summary")
save_record(free_fits, "growth_free_kill_rates")
save_record(positive_control, "positive_control_braf")
```

```python
# ==========================================================================
# 8.4 Ranking accuracy by the observed separation of the two treatments
#     The observed gap between two arms' day-24 log volumes is expressed in
#     units of the measurement noise (Section 3.6). A pair's observed gap is
#     itself measured with error (SD about √2 times the noise), so the lowest
#     strata hold pairs whose order the data cannot establish reliably.
# ==========================================================================

SEPARATION_EDGES = [0, 1, 2, 4, 8, np.inf]
SEPARATION_LABELS = ["< 1", "1-2", "2-4", "4-8", "> 8"]
separation_pairs = forecast_pairs.assign(separation=pd.cut(
    forecast_pairs.observed_gap / MEASUREMENT_NOISE, SEPARATION_EDGES, labels=SEPARATION_LABELS))
separability_rows = []
for label in SEPARATION_LABELS:
    stratum = separation_pairs.loc[separation_pairs.separation == label].reset_index(drop=True)
    if stratum.empty:
        continue
    depth_scores = stratum.depth_prediction.to_numpy()
    trajectory_scores = stratum.trajectory_prediction.to_numpy()
    draws = cluster_bootstrap(rows_by_cluster(stratum.model_id),
                              lambda rows: trajectory_scores[rows].mean() - depth_scores[rows].mean(),
                              N_BOOTSTRAP, np.random.default_rng(SEEDS["separability_bootstrap"]))
    low, high = np.percentile(draws, [2.5, 97.5])
    separability_rows.append({
        "noise_sd_units": label, "pairs": len(stratum), "share_of_pairs": len(stratum) / len(separation_pairs),
        "median_gap": stratum.observed_gap.median(), "depth_accuracy": depth_scores.mean(),
        "trajectory_accuracy": trajectory_scores.mean(),
        "difference": trajectory_scores.mean() - depth_scores.mean(), "low": low, "high": high})
separability = pd.DataFrame(separability_rows)
top = separability.iloc[-1]
record("share_pairs_below_2_sd", float(separability.share_of_pairs.iloc[:2].sum()))
record("top_stratum_share", float(top.share_of_pairs))
record("top_stratum_depth_accuracy", float(top.depth_accuracy))
record("top_stratum_trajectory_accuracy", float(top.trajectory_accuracy))
record("top_stratum_difference", float(top.difference))
record("top_stratum_difference_low", float(top.low))
record("top_stratum_difference_high", float(top.high))
print("8.4 RANKING ACCURACY BY OBSERVED SEPARATION")
print(f"  {len(separation_pairs):,} pairs in {separation_pairs.model_id.nunique()} PDX models; "
      f"measurement noise {MEASUREMENT_NOISE:.4f} log units")
print(separability.round(3).to_string(index=False))
save_record(separability, "ranking_by_separation")
```

```python
# ==========================================================================
# 8.5 Baseline volume and the window slope: forecasting, treatment selection
#     and ranking
#     The baseline-plus-slope set (baseline log volume and the least-squares
#     slope of log volume over the window) was fitted in 8.1 with the learner
#     and folds of Section 7. In a treated arm the slope is the net rate of
#     change under treatment, so it carries the drug's effect as well as the
#     tumor's growth: this set is the simplest trajectory summary, not a
#     growth-only control. Its out-of-fold predictions select an arm in each
#     PDX model and rank every within-model pair, exactly as the depth and
#     trajectory sets do in Section 7, so the comparison with the trajectory
#     set shows what the remaining shape descriptors add beyond one early
#     slope. Intervals resample PDX models with predictions held fixed.
# ==========================================================================

slope_rng = np.random.default_rng(SEEDS["baseline_slope_bootstrap"])
slope_prediction = set_predictions["baseline plus slope"]
# validation (8.1) is a copy of forecast_primary, so its rows are in the order
# of the Section 7 selection table.
assert len(slope_prediction) == len(selection)
assert (validation.arm_id.to_numpy() == selection.arm_id.to_numpy()).all()
slope_selection = selection.assign(slope_prediction=slope_prediction)

# --- regret of the arm each set selects --------------------------------------
slope_regret = selection_by_model(slope_selection, ["slope_prediction", "trajectory_prediction"])
slope_mean_regret = float(slope_regret.slope_prediction_regret.mean())
trajectory_mean_regret = float(slope_regret.trajectory_prediction_regret.mean())
slope_regret_reduction = (slope_mean_regret - trajectory_mean_regret) / slope_mean_regret
slope_regret_draws = []
for _ in range(N_BOOTSTRAP):
    drawn = slope_regret.iloc[slope_rng.integers(0, len(slope_regret), size=len(slope_regret))]
    slope_mean = drawn.slope_prediction_regret.mean()
    if slope_mean > 0:
        slope_regret_draws.append((slope_mean - drawn.trajectory_prediction_regret.mean()) / slope_mean)
slope_regret_low, slope_regret_high = np.percentile(slope_regret_draws, [2.5, 97.5])

# --- pairwise ranking accuracy, overall and by separation ---------------------
ranking_columns = ["depth_prediction", "trajectory_prediction", "slope_prediction"]
slope_pairs = within_model_pairs(slope_selection, ranking_columns)
assert len(slope_pairs) == len(forecast_pairs), "the pairs differ from those of Section 7"
slope_pairs["separation"] = pd.cut(slope_pairs.observed_gap / MEASUREMENT_NOISE,
                                   SEPARATION_EDGES, labels=SEPARATION_LABELS)

slope_rows = []
for label, stratum in [("all pairs", slope_pairs)] + [
        (label, slope_pairs.loc[slope_pairs.separation == label]) for label in SEPARATION_LABELS]:
    if stratum.empty:
        continue
    stratum = stratum.reset_index(drop=True)
    difference, low, high = ranking_difference(stratum, "trajectory_prediction", "slope_prediction",
                                               N_BOOTSTRAP, slope_rng)
    slope_rows.append({"pairs": label if label == "all pairs" else f"{label} noise SDs",
                       "n_pairs": len(stratum), "depth_accuracy": stratum.depth_prediction.mean(),
                       "baseline_slope_accuracy": stratum.slope_prediction.mean(),
                       "trajectory_accuracy": stratum.trajectory_prediction.mean(),
                       "trajectory_minus_baseline_slope": difference, "low": low, "high": high})
baseline_slope_selection = pd.DataFrame(slope_rows)
overall_row = baseline_slope_selection.iloc[0]
top_row = baseline_slope_selection.iloc[-1]

record("baseline_slope_regret", slope_mean_regret)
record("trajectory_vs_baseline_slope_regret_reduction", slope_regret_reduction)
record("trajectory_vs_baseline_slope_regret_reduction_low", float(slope_regret_low))
record("trajectory_vs_baseline_slope_regret_reduction_high", float(slope_regret_high))
record("share_different_choice_trajectory_baseline_slope", float(slope_regret.different_choice.mean()))
record("ranking_accuracy_baseline_slope", float(overall_row.baseline_slope_accuracy))
record("ranking_trajectory_minus_baseline_slope", float(overall_row.trajectory_minus_baseline_slope))
record("ranking_trajectory_minus_baseline_slope_low", float(overall_row.low))
record("ranking_trajectory_minus_baseline_slope_high", float(overall_row.high))
record("top_stratum_baseline_slope_accuracy", float(top_row.baseline_slope_accuracy))
record("top_stratum_trajectory_minus_baseline_slope", float(top_row.trajectory_minus_baseline_slope))
record("top_stratum_trajectory_minus_baseline_slope_low", float(top_row.low))
record("top_stratum_trajectory_minus_baseline_slope_high", float(top_row.high))

print("8.5 BASELINE VOLUME AND WINDOW SLOPE (the simplest trajectory summary)")
print(f"  forecasting RMSE: baseline plus slope {RESULTS['baseline_slope_rmse']:.3f}, "
      f"depth {RESULTS['depth_rmse']:.3f}, trajectory {RESULTS['trajectory_rmse']:.3f}")
print(f"  baseline plus slope against depth, relative reduction in RMSE      "
      f"{format_ci(RESULTS['baseline_slope_over_depth'], RESULTS['baseline_slope_over_depth_low'], RESULTS['baseline_slope_over_depth_high'])}")
print(f"  trajectory against baseline plus slope, relative reduction in RMSE "
      f"{format_ci(RESULTS['trajectory_over_baseline_slope'], RESULTS['trajectory_over_baseline_slope_low'], RESULTS['trajectory_over_baseline_slope_high'])}")
print(f"\n  treatment selection ({len(slope_regret)} PDX models)")
print(f"    mean regret, baseline plus slope   {slope_mean_regret:.3f} log units ({np.exp(slope_mean_regret):.2f}-fold)")
print(f"    mean regret, trajectory set        {trajectory_mean_regret:.3f} log units "
      f"({np.exp(trajectory_mean_regret):.2f}-fold)")
print(f"    relative reduction in regret, trajectory against baseline plus slope   "
      f"{format_ci(slope_regret_reduction, slope_regret_low, slope_regret_high)}")
print(f"    PDX models where the two select different arms   {int(slope_regret.different_choice.sum())} "
      f"({100 * slope_regret.different_choice.mean():.1f}%)")
print("\n  pairwise ranking accuracy; trajectory minus baseline plus slope, with a 95% interval "
      "(PDX models resampled)")
print(baseline_slope_selection.round(3).to_string(index=False))
save_record(baseline_slope_selection, "baseline_slope_ranking")
save_record(slope_regret, "baseline_slope_regret")
```

## 9 · Molecular correlates of depth and durability

Expression signatures are scored from the RNA-seq data of the source study. Genes are ranked within each PDX model, and a signature score is the mean percentile rank of its genes minus 0.5. Three tumor-cell programs are scored: epithelial–mesenchymal transition (EMT), stemness, and cell-cycle arrest. The arrest score is signed: the mean rank of four arrest genes minus the mean rank of six proliferation genes. A stromal score serves as a proxy for stromal content. The genes of every set are listed in `gene_sets.csv`. The signature set holds the three program scores. The context set holds the stromal score, baseline log volume, the PDX model's untreated growth rate, and tumor-type indicators.

The analysis set is post-nadir arms with at least five measurements, in PDX models that have both an untreated growth rate and expression data. Gastric cancer models have no expression data, so gastric arms are not included. Depth and durability are each regressed on each set by ridge regression with five folds grouped by PDX model. The penalty is chosen within each training fold by an inner grouped cross-validation. Performance is the out-of-fold R².

The null distribution comes from 500 shuffles that break the link between features and outcome but keep tumor-type structure. PDX-model-level features are reassigned between PDX models of the same tumor type, and baseline volume is shuffled across arms of the same tumor type. Tumor-type indicators cannot change under either shuffle, so a context set that beats its null carries information beyond tumor type. Permutation P values are (1 + number of null values ≥ observed) / (1 + 500).

The dissociation statistic is D = [R²(signatures, durability) − R²(context, durability)] − [R²(signatures, depth) − R²(context, depth)]. D is positive when the signatures explain durability relatively better than they explain depth. Its interval comes from 2,000 bootstrap resamples of PDX models.

**Robustness to the choice of gene sets (9.3).** The three program scores of 9.1 come from curated marker genes. Section 9.3 rebuilds them from the MSigDB hallmark collection, scored by the same rank rule, and repeats the block R², shuffle null and dissociation statistic with the hallmark signature block in place of the curated one; the context block is unchanged. EMT is the hallmark epithelial–mesenchymal transition set. Stemness, which has no hallmark set, is represented by the mean score of the three developmental self-renewal pathways (Wnt/β-catenin, Notch and Hedgehog signalling). Cell-cycle arrest is the hallmark p53 pathway score minus the mean of the E2F-target and G2/M-checkpoint scores, the same arrest-minus-proliferation construction as in 9.1. The section runs when a hallmark library is present in `data/msigdb/` (a GMT file, or the Enrichr `MSigDB_Hallmark_2020` text library) and prints a skip line otherwise.

**Inputs:** `sheets["expression"]`, `post_nadir_arms`, `arm_design`, `untreated_growth`, `data/msigdb/` (optional).
**Outputs:** `signature_scores`, `gene_set_table` (table `gene_sets.csv`), `molecular_frame`, `molecular_blocks`, `dissociation`, `hallmark_blocks`.

```python
# ==========================================================================
# 9.1 Signature scores
# ==========================================================================

set_section("9")
expression = sheets["expression"].rename(columns={sheets["expression"].columns[0]: "gene"})
expression["gene"] = expression.gene.astype(str).str.strip().str.upper()
duplicated_genes = int(expression.gene.duplicated().sum())
expression = expression.groupby("gene").mean(numeric_only=True)       # duplicate symbols averaged
expression = expression.loc[expression.notna().mean(axis=1) > 0.5]
expression_ranks = expression.rank(axis=0, pct=True)

GENE_SETS = {
    "EMT": ["VIM", "CDH2", "FN1", "SNAI1", "SNAI2", "TWIST1", "ZEB1", "ZEB2", "MMP2", "MMP9",
            "SPARC", "TGFB1", "COL1A1", "COL1A2", "SERPINE1"],
    "stemness": ["SOX2", "POU5F1", "NANOG", "KLF4", "MYC", "ALDH1A1", "PROM1", "CD44", "LGR5",
                 "BMI1", "NOTCH1"],
    "cell-cycle arrest": ["CDKN1A", "CDKN1B", "CDKN2A", "RB1"],
    "proliferation": ["MKI67", "PCNA", "CCNB1", "TOP2A", "AURKA", "PLK1"],
    "stromal": ["PECAM1", "VWF", "PTPRC", "CD68", "COL3A1", "ACTA2", "PDGFRB", "THY1"],
}
# Source for each gene set, written to gene_sets.csv. An empty entry is
# printed below as a set without a recorded source.
GENE_SET_SOURCES = {name: "" for name in GENE_SETS}

set_scores, gene_set_rows = {}, []
for name, genes in GENE_SETS.items():
    present = [gene for gene in genes if gene in expression_ranks.index]
    set_scores[name] = expression_ranks.loc[present].mean(axis=0) - 0.5
    gene_set_rows.append({"gene_set": name, "genes": ", ".join(genes), "genes_in_data": len(present),
                          "genes_in_set": len(genes), "source": GENE_SET_SOURCES[name]})
gene_set_table = pd.DataFrame(gene_set_rows)
signature_scores = pd.DataFrame({
    "signature_emt": set_scores["EMT"],
    "signature_stemness": set_scores["stemness"],
    "signature_cell_cycle_arrest": set_scores["cell-cycle arrest"] - set_scores["proliferation"],
    "context_stromal": set_scores["stromal"],
})
signature_scores.index = signature_scores.index.astype(str)

print("9.1 SIGNATURE SCORES")
print(f"  expression matrix {expression.shape[0]:,} genes x {expression.shape[1]:,} PDX models "
      f"({duplicated_genes} duplicate gene symbols averaged)")
print(gene_set_table[["gene_set", "genes_in_data", "genes_in_set"]].to_string(index=False))
missing_sources = [name for name, source in GENE_SET_SOURCES.items() if not source]
if missing_sources:
    print(f"  gene sets without a recorded source (GENE_SET_SOURCES): {missing_sources}")
print("  Spearman correlation of each program score with the stromal score")
for column in ("signature_emt", "signature_stemness", "signature_cell_cycle_arrest"):
    value, _ = spearman(signature_scores[column], signature_scores.context_stromal)
    print(f"    {column:<30} ρ = {value:.3f}")

# ==========================================================================
# 9.2 Analysis set, block R² against the shuffle null, and the dissociation
# ==========================================================================

molecular_frame = post_nadir_arms.loc[
    post_nadir_arms.arm_id.map(arm_design.n_measurements).ge(MIN_MEASUREMENTS_PER_ARM)
    & post_nadir_arms.model_id.isin(untreated_growth.index)
    & post_nadir_arms.model_id.astype(str).isin(signature_scores.index),
    ["arm_id", "model_id", "tumor_type", "best_average_response", "post_nadir_growth_rate",
     "baseline_volume"]].reset_index(drop=True)
molecular_frame = molecular_frame.join(signature_scores, on=molecular_frame.model_id.astype(str))
molecular_frame["context_untreated_growth"] = molecular_frame.model_id.map(untreated_growth)
molecular_frame["context_baseline_log_volume"] = np.log(molecular_frame.baseline_volume.clip(lower=VOLUME_FLOOR))
tumor_type_columns = []
for tumor_type in sorted(molecular_frame.tumor_type.unique()):
    molecular_frame[f"context_type_{tumor_type}"] = (molecular_frame.tumor_type == tumor_type).astype(float)
    tumor_type_columns.append(f"context_type_{tumor_type}")
FEATURE_BLOCKS = {
    "signatures": {"model": ["signature_emt", "signature_stemness", "signature_cell_cycle_arrest"],
                   "arm": [], "fixed": []},
    "context": {"model": ["context_stromal", "context_untreated_growth"],
                "arm": ["context_baseline_log_volume"], "fixed": tumor_type_columns},
}
MOLECULAR_OUTCOMES = {"depth": "best_average_response", "durability": "post_nadir_growth_rate"}
N_PERMUTATIONS = 500
print("\n9.2 ANALYSIS SET")
print(f"  arms {record('molecular_n_arms', len(molecular_frame)):,} in "
      f"{record('molecular_n_models', molecular_frame.model_id.nunique())} PDX models; tumor types "
      f"{sorted(molecular_frame.tumor_type.unique())}")

null_rng = np.random.default_rng(SEEDS["molecular_null"])
block_rows = []
for outcome_label, outcome_column in MOLECULAR_OUTCOMES.items():
    for block_name, block in FEATURE_BLOCKS.items():
        features = block_features(block)
        observed_r2 = block_r2(molecular_frame, features, outcome_column)
        null_r2 = np.array([block_r2(permute_within_tumor_type(molecular_frame, block["model"],
                                                               block["arm"], null_rng),
                                     features, outcome_column)
                            for _ in range(N_PERMUTATIONS)])
        block_rows.append({"outcome": outcome_label, "block": block_name, "features": len(features),
                           "r2": observed_r2, "null_mean": null_r2.mean(),
                           "null_95th": np.percentile(null_r2, 95),
                           "permutation_p": (1 + np.sum(null_r2 >= observed_r2)) / (1 + N_PERMUTATIONS)})
        record(f"r2_{block_name}_{outcome_label}", observed_r2)
        record(f"null95_{block_name}_{outcome_label}", float(np.percentile(null_r2, 95)))
        record(f"permutation_p_{block_name}_{outcome_label}", float(block_rows[-1]["permutation_p"]))
molecular_blocks = pd.DataFrame(block_rows)
print(f"\n  out-of-fold R² against {N_PERMUTATIONS} within-tumor-type shuffles")
print(molecular_blocks.round(4).to_string(index=False))

dissociation = dissociation_statistic(molecular_frame, FEATURE_BLOCKS, MOLECULAR_OUTCOMES)
dissociation_draws = cluster_bootstrap(
    rows_by_cluster(molecular_frame.model_id.astype(str)),
    lambda rows: dissociation_statistic(molecular_frame.iloc[rows].reset_index(drop=True),
                                        FEATURE_BLOCKS, MOLECULAR_OUTCOMES),
    N_BOOTSTRAP, np.random.default_rng(SEEDS["molecular_bootstrap"]))
dissociation_low, dissociation_high = np.percentile(dissociation_draws, [2.5, 97.5])
record("dissociation", dissociation)
record("dissociation_low", dissociation_low)
record("dissociation_high", dissociation_high)
print(f"\n  dissociation statistic D   {format_ci(dissociation, dissociation_low, dissociation_high, 1, 3, '')}"
      f", {len(dissociation_draws):,} resamples of PDX models")
save_record(molecular_blocks, "molecular_blocks")
save_record(molecular_frame, "molecular_frame")
```

```python
# ==========================================================================
# 9.3 Robustness to the choice of gene sets: MSigDB hallmark programs
#     The curated signature block of 9.1 is rebuilt from hallmark sets and
#     scored by the same rank rule. The analysis set, the context block, the
#     ridge regression, the shuffle null and the dissociation statistic are
#     those of 9.2; only the values in the three signature columns change.
#     The context block's R² is therefore the one printed in 9.2.
# ==========================================================================

# Hallmark sets (normalized names, see normalized_set_name) behind each
# program. A program score is the mean score of its "plus" sets minus the mean
# score of its "minus" sets.
HALLMARK_PROGRAMS = {
    "signature_emt": {"plus": ["EPITHELIALMESENCHYMALTRANSITION"], "minus": []},
    "signature_stemness": {"plus": ["WNTBETACATENINSIGNALING", "NOTCHSIGNALING", "HEDGEHOGSIGNALING"],
                           "minus": []},
    "signature_cell_cycle_arrest": {"plus": ["P53PATHWAY"], "minus": ["E2FTARGETS", "G2MCHECKPOINT"]},
}
assert set(HALLMARK_PROGRAMS) == set(FEATURE_BLOCKS["signatures"]["model"]), \
    "the hallmark programs must replace the curated signature columns one for one"

hallmark_files = (sorted([*MSIGDB_DIR.glob("*.gmt"), *MSIGDB_DIR.glob("*.txt")])
                  if MSIGDB_DIR.exists() else [])
HALLMARK_AVAILABLE = False
hallmark_blocks, hallmark_sets_table = pd.DataFrame(), pd.DataFrame()
print("9.3 ROBUSTNESS TO THE CHOICE OF GENE SETS (MSigDB hallmark collection)")
if not hallmark_files:
    print(f"  skipped: no .gmt or .txt hallmark library in {relative(MSIGDB_DIR)}")
else:
    library_path = hallmark_files[0]
    library = read_gene_set_library(library_path)
    needed = sorted({name for program in HALLMARK_PROGRAMS.values()
                     for name in program["plus"] + program["minus"]})
    missing_sets = [name for name in needed if name not in library]
    print(f"  library {relative(library_path)}: {len(library)} sets "
          f"(SHA-256 {sha256_of_file(library_path)[:16]})")
    if missing_sets:
        print(f"  skipped: the library lacks {missing_sets}")
    else:
        HALLMARK_AVAILABLE = True
        set_rows, set_score = [], {}
        for name in needed:
            written, genes = library[name]
            set_score[name], present = gene_set_score(expression_ranks, genes)
            set_rows.append({"hallmark_set": written, "genes_in_set": len(genes), "genes_in_data": present,
                             "coverage": present / len(genes) if genes else np.nan})
        hallmark_sets_table = pd.DataFrame(set_rows)
        print(hallmark_sets_table.round(3).to_string(index=False))

        hallmark_scores = pd.DataFrame({
            program: (pd.concat([set_score[name] for name in spec["plus"]], axis=1).mean(axis=1)
                      - (pd.concat([set_score[name] for name in spec["minus"]], axis=1).mean(axis=1)
                         if spec["minus"] else 0.0))
            for program, spec in HALLMARK_PROGRAMS.items()})
        hallmark_scores.index = hallmark_scores.index.astype(str)

        # How closely each hallmark program follows its curated counterpart and
        # the stromal proxy, across PDX models with expression data.
        print("\n  Spearman correlation across PDX models: hallmark program with the curated program, "
              "and with the stromal score")
        for program in HALLMARK_PROGRAMS:
            shared = hallmark_scores.index.intersection(signature_scores.index)
            with_curated, _ = spearman(hallmark_scores.loc[shared, program], signature_scores.loc[shared, program])
            with_stroma, _ = spearman(hallmark_scores.loc[shared, program],
                                      signature_scores.loc[shared, "context_stromal"])
            print(f"    {program:<30} curated ρ = {with_curated:+.3f}   stromal ρ = {with_stroma:+.3f}")

        # The 9.2 analysis frame with the hallmark values in the signature columns.
        hallmark_frame = molecular_frame.copy()
        for program in HALLMARK_PROGRAMS:
            hallmark_frame[program] = hallmark_frame.model_id.astype(str).map(hallmark_scores[program])
        print(f"\n  arms {len(hallmark_frame):,} in {hallmark_frame.model_id.nunique()} PDX models "
              f"(the 9.2 analysis set); arms without a hallmark score "
              f"{int(hallmark_frame[list(HALLMARK_PROGRAMS)].isna().any(axis=1).sum())}")

        hallmark_null_rng = np.random.default_rng(SEEDS["hallmark_null"])
        signature_features = block_features(FEATURE_BLOCKS["signatures"])
        hallmark_rows = []
        for outcome_label, outcome_column in MOLECULAR_OUTCOMES.items():
            observed_r2 = block_r2(hallmark_frame, signature_features, outcome_column)
            null_r2 = np.array([block_r2(permute_within_tumor_type(
                hallmark_frame, FEATURE_BLOCKS["signatures"]["model"], [], hallmark_null_rng),
                signature_features, outcome_column) for _ in range(N_PERMUTATIONS)])
            curated = molecular_blocks.set_index(["outcome", "block"]).loc[(outcome_label, "signatures")]
            hallmark_rows.append({"outcome": outcome_label, "r2_hallmark": observed_r2,
                                  "null_mean": null_r2.mean(), "null_95th": np.percentile(null_r2, 95),
                                  "permutation_p": (1 + np.sum(null_r2 >= observed_r2)) / (1 + N_PERMUTATIONS),
                                  "r2_curated": curated.r2, "permutation_p_curated": curated.permutation_p})
            record(f"r2_hallmark_signatures_{outcome_label}", observed_r2)
            record(f"null95_hallmark_signatures_{outcome_label}", float(np.percentile(null_r2, 95)))
            record(f"permutation_p_hallmark_signatures_{outcome_label}", float(hallmark_rows[-1]["permutation_p"]))
        hallmark_blocks = pd.DataFrame(hallmark_rows)
        print(f"\n  signature block, out-of-fold R² against {N_PERMUTATIONS} within-tumor-type shuffles "
              "(hallmark beside curated)")
        print(hallmark_blocks.round(4).to_string(index=False))

        hallmark_dissociation = dissociation_statistic(hallmark_frame, FEATURE_BLOCKS, MOLECULAR_OUTCOMES)
        hallmark_draws = cluster_bootstrap(
            rows_by_cluster(hallmark_frame.model_id.astype(str)),
            lambda rows: dissociation_statistic(hallmark_frame.iloc[rows].reset_index(drop=True),
                                                FEATURE_BLOCKS, MOLECULAR_OUTCOMES),
            N_BOOTSTRAP, np.random.default_rng(SEEDS["hallmark_bootstrap"]))
        hallmark_low, hallmark_high = np.percentile(hallmark_draws, [2.5, 97.5])
        record("dissociation_hallmark", hallmark_dissociation)
        record("dissociation_hallmark_low", float(hallmark_low))
        record("dissociation_hallmark_high", float(hallmark_high))
        print(f"\n  dissociation statistic D, hallmark signatures   "
              f"{format_ci(hallmark_dissociation, hallmark_low, hallmark_high, 1, 3, '')}, "
              f"{len(hallmark_draws):,} resamples of PDX models")
        print(f"  dissociation statistic D, curated signatures (9.2)   "
              f"{format_ci(RESULTS['dissociation'], RESULTS['dissociation_low'], RESULTS['dissociation_high'], 1, 3, '')}")
        save_record(hallmark_blocks, "molecular_blocks_hallmark")
        save_record(hallmark_sets_table, "hallmark_sets")
```

## 10 · Analysis of combinations

A combination is evaluated in a PDX model only if every one of its components was also tested alone in that model. The combination is then compared with its own single agents on each endpoint. Depth gain is the lowest best average response among the single agents minus the combination's best average response. Durability gain is the lowest post-nadir growth rate among the single agents minus the combination's post-nadir growth rate. A positive gain means the combination is better than its best component on that endpoint. Depth is available for every treated arm with both endpoints. Durability is available only for post-nadir arms, so the best single component for durability is the best among components that have a post-nadir growth rate; the number of evaluations where a component lacked one is printed. Evaluations are classified by depth gain > 5 percentage points and durability gain > 0.005 per day.

For combinations with at least eight evaluations, the median of each gain is compared with zero by a Wilcoxon signed-rank test, with the Benjamini–Hochberg procedure applied across those combinations. Medians carry 95% bootstrap intervals. The association between the two gains is the Spearman correlation across evaluations and across combinations. Two combinations in the panel were later approved for patients carrying BRAF V600 mutations, and they serve as a positive control.

**Inputs:** `treated_with_endpoints`, `post_nadir_arms`.
**Outputs:** `combination_evaluations`, `by_combination`.

```python
# ==========================================================================
# 10.1 Combination–PDX model evaluations against the best single component
# ==========================================================================

set_section("10")
DEPTH_GAIN_THRESHOLD = 5.0          # percentage points of best average response
DURABILITY_GAIN_THRESHOLD = 0.005   # per day
MIN_EVALUATIONS_TO_TEST = 8
APPROVED_COMBINATIONS = {
    "encorafenib + binimetinib": "FDA 2018, BRAF V600E/K melanoma",
    "cetuximab + encorafenib": "FDA 2020, BRAF V600E colorectal cancer",
}

depth_lookup = treated_with_endpoints.groupby(["model_id", "treatment"]).best_average_response.mean()
durability_lookup = post_nadir_arms.groupby(["model_id", "treatment"]).post_nadir_growth_rate.mean()
combination_labels = sorted(t for t in treated_with_endpoints.treatment.unique() if "+" in t)
evaluation_rows = []
for combination in combination_labels:
    components = [part.strip() for part in combination.split("+")]
    for model_id in depth_lookup.xs(combination, level="treatment").index:
        if not all((model_id, part) in depth_lookup.index for part in components):
            continue
        component_growth = [durability_lookup.get((model_id, part), np.nan) for part in components]
        evaluation_rows.append({
            "model_id": model_id, "combination": combination, "components": len(components),
            "depth_gain": min(depth_lookup[(model_id, part)] for part in components)
                          - depth_lookup[(model_id, combination)],
            "durability_gain": (np.nanmin(component_growth) - durability_lookup[(model_id, combination)]
                                if (model_id, combination) in durability_lookup.index
                                and np.isfinite(component_growth).any() else np.nan),
            "component_without_durability": bool(np.isnan(component_growth).any())})
all_evaluations = pd.DataFrame(evaluation_rows)
combination_evaluations = all_evaluations.dropna(subset=["durability_gain"]).reset_index(drop=True)
deeper = combination_evaluations.depth_gain > DEPTH_GAIN_THRESHOLD
longer = combination_evaluations.durability_gain > DURABILITY_GAIN_THRESHOLD
combination_evaluations["category"] = np.select(
    [deeper & longer, deeper & ~longer, ~deeper & longer],
    ["superior on both", "superior in depth only", "superior in durability only"], "superior on neither")

print("10.1 COMBINATION EVALUATIONS")
print(f"  combination labels                                {len(combination_labels)}")
print(f"  evaluations with depth gain                       {len(all_evaluations):,}")
print(f"  evaluations with both gains                       "
      f"{record('combination_evaluations', len(combination_evaluations)):,} across "
      f"{record('combinations_evaluated', combination_evaluations.combination.nunique())} combinations")
print(f"  of which a component had no post-nadir growth rate  "
      f"{int(combination_evaluations.component_without_durability.sum())}")
category_shares = combination_evaluations.category.value_counts(normalize=True)
for category in ("superior on both", "superior in depth only", "superior in durability only",
                 "superior on neither"):
    share = float(category_shares.get(category, 0.0))
    record(f"share_{category.replace(' ', '_')}", share)
    print(f"  {category:<30} {int(round(share * len(combination_evaluations))):>4}  ({100 * share:.1f}%)")

gain_rho, gain_p = spearman(combination_evaluations.depth_gain, combination_evaluations.durability_gain)
record("gain_rho", gain_rho)
record("gain_p", gain_p)
record("gain_shared_variance", gain_rho ** 2)
print(f"\n  depth gain against durability gain, evaluations   ρ = {gain_rho:.3f}, {p_label(gain_p)}, "
      f"shared rank variance {100 * gain_rho ** 2:.0f}%")

# ==========================================================================
# 10.2 Per-combination tests and the approved-combination positive control
# ==========================================================================

combination_rng = np.random.default_rng(SEEDS["combination_bootstrap"])
combination_rows = []
for combination, group in combination_evaluations.groupby("combination"):
    depth_values = group.depth_gain.to_numpy(float)
    durability_values = group.durability_gain.to_numpy(float)
    testable = len(group) >= MIN_EVALUATIONS_TO_TEST
    depth_low, depth_high = bootstrap_median_interval(depth_values, N_BOOTSTRAP, combination_rng)
    durability_low, durability_high = bootstrap_median_interval(durability_values, N_BOOTSTRAP, combination_rng)
    combination_rows.append({
        "combination": combination, "evaluations": len(group),
        "median_depth_gain": np.median(depth_values), "depth_low": depth_low, "depth_high": depth_high,
        "depth_p": stats.wilcoxon(depth_values).pvalue if testable and np.ptp(depth_values) > 0 else np.nan,
        "median_durability_gain": np.median(durability_values),
        "durability_low": durability_low, "durability_high": durability_high,
        "durability_p": (stats.wilcoxon(durability_values).pvalue
                         if testable and np.ptp(durability_values) > 0 else np.nan),
        "share_superior_on_both": float((group.category == "superior on both").mean()),
        "approval": APPROVED_COMBINATIONS.get(combination, "")})
by_combination = pd.DataFrame(combination_rows).sort_values("evaluations", ascending=False)
by_combination["depth_q"] = benjamini_hochberg(by_combination.depth_p)
by_combination["durability_q"] = benjamini_hochberg(by_combination.durability_p)
tested = by_combination.loc[by_combination.evaluations >= MIN_EVALUATIONS_TO_TEST]
depth_significant = tested.loc[tested.depth_q < 0.05]
also_durable = depth_significant.loc[depth_significant.durability_q < 0.10]
combination_rho, combination_rho_p = spearman(tested.median_depth_gain, tested.median_durability_gain)
record("combinations_tested", len(tested))
record("combinations_depth_significant", len(depth_significant))
record("combinations_also_durable", len(also_durable))
if len(also_durable):
    record("also_durable_gain_min", also_durable.median_durability_gain.min())
    record("also_durable_gain_max", also_durable.median_durability_gain.max())
    record("also_durable_q_min", also_durable.durability_q.min())
    record("also_durable_q_max", also_durable.durability_q.max())
not_durable = depth_significant.loc[depth_significant.durability_q >= 0.10]
if len(not_durable):
    record("not_durable_q_min", not_durable.durability_q.min())

print(f"\n10.2 COMBINATIONS WITH >= {MIN_EVALUATIONS_TO_TEST} EVALUATIONS "
      f"({len(tested)}; Wilcoxon tests, Benjamini–Hochberg across these)")
print(tested.drop(columns=["depth_p", "durability_p"]).round(4).to_string(index=False))
print(f"\n  deeper response, q < 0.05                         {len(depth_significant)}: "
      f"{', '.join(depth_significant.combination)}")
print(f"  of these, slower regrowth with q < 0.10           {len(also_durable)}: "
      f"{', '.join(also_durable.combination)}")
print(f"  depth gain against durability gain, combinations  ρ = {combination_rho:.3f}, "
      f"{p_label(combination_rho_p)} (n = {len(tested)})")
record("combination_level_gain_rho", combination_rho)
record("combination_level_gain_p", combination_rho_p)
print("\n  POSITIVE CONTROL: combinations later approved for BRAF V600-mutant tumors")
for row in by_combination.loc[by_combination.approval != ""].itertuples(index=False):
    print(f"    {row.combination} ({row.approval}), {row.evaluations} evaluations")
    print(f"      depth gain       {row.median_depth_gain:.1f} points "
          f"(95% CI {row.depth_low:.1f} to {row.depth_high:.1f}), q = {format_p(row.depth_q)}")
    print(f"      durability gain  {row.median_durability_gain:.4f} per day "
          f"(95% CI {row.durability_low:.4f} to {row.durability_high:.4f}), q = {format_p(row.durability_q)}")
    print(f"      superior on both {100 * row.share_superior_on_both:.1f}% of evaluations")
    tag = "".join(character if character.isalnum() else "_" for character in row.combination)
    record(f"approved_{tag}_evaluations", int(row.evaluations))
    record(f"approved_{tag}_depth_gain", float(row.median_depth_gain))
    record(f"approved_{tag}_depth_q", float(row.depth_q) if pd.notna(row.depth_q) else np.nan)
    record(f"approved_{tag}_durability_gain", float(row.median_durability_gain))
    record(f"approved_{tag}_durability_q", float(row.durability_q) if pd.notna(row.durability_q) else np.nan)
    record(f"approved_{tag}_share_superior_on_both", float(row.share_superior_on_both))
save_record(combination_evaluations, "combination_evaluations")
save_record(by_combination, "combination_summary")
```

## 11 · Analysis of mechanism class

The single-agent treatments are grouped by compound, with dose variants pooled, and each compound is assigned to one of nine mechanism families from the treatment-target annotation of the source study (listed in `compound_families.csv`). The cell-cycle family holds a CDK4/6 inhibitor (LEE011) and two MDM2 inhibitors (HDM201, CGM097). The family assignments and the contrast were fixed in a plan before testing, and that plan is loaded below and checked against its recorded checksum. The contrast compares cytostatic families (cell cycle, PI3K–mTOR, MAPK) with cytotoxic and apoptosis-inducing families. The inventory of each agent's median depth and durability had been printed before the plan was written.

For every pair of compounds from different families tested in the same PDX model, the difference in each endpoint is taken as cytostatic minus comparator. Differences are averaged within each PDX model and then across models. The plan specified a bootstrap that resamples PDX models (10,000 replicates) and a permutation of family labels across compounds (10,000 permutations). Its hypothesis predicted slower regrowth and *shallower* depth for cytostatic families. Four sensitivity analyses were added after the plan was fixed:

- a bootstrap (20,000 replicates) and a sign-flip randomization test with the compound as the unit;
- an exact randomization test over all 1,260 assignments of three families against two, also run for depth;
- a linear mixed-effects model with crossed random intercepts for PDX model and family pair. Two other random-effect structures are fitted beside it and printed: family pair nested within PDX model (with a PDX-model intercept), and family pair within PDX model without a PDX-model intercept. The last is what a statsmodels formula with a variance component and no `re_formula` fits, because statsmodels then drops the default intercept for the grouping factor, so it is shown to make that difference visible.

The compounds share comparators and PDX models, so they are not fully independent units. Durability exists only in post-nadir arms, and families differ in how often their arms reach that set (shown below), so each comparison is between the arms of each family that regressed and regrew.

**Inputs:** `arms`, `post_nadir_ids`, `prespecification/mechanism_family_plan.json`.
**Outputs:** `compound_atlas`, `compound_table` (table `compound_families.csv`), `compound_pairs`, `mechanism_results`; figure data `mechanism_class_compound_effects.csv`, `mechanism_class_leave_one_out.csv` and `mechanism_class_contrast_pairs.csv`.

```python
# ==========================================================================
# 11.1 The checksummed plan, the compound atlas and within-model compound pairs
# ==========================================================================

set_section("11")
mechanism_plan, mechanism_plan_sha256 = load_prespecified_plan("mechanism_class")
FAMILY_OF_LABEL = mechanism_plan["family_map"]
COMPOUND_OF_LABEL = mechanism_plan["compound_map"]
CYTOSTATIC = tuple(sorted(mechanism_plan["cytostatic_families"]))
COMPARATOR = tuple(sorted(mechanism_plan["cytotoxic_families"]))
print("11.1 MECHANISM-CLASS PLAN")
print(f"  file      {relative(PRESPEC_DIR / PRESPECIFIED_PLANS['mechanism_class'][0])}")
print(f"  SHA-256   {mechanism_plan_sha256}  (matches the recorded {PRESPECIFIED_PLANS['mechanism_class'][1]})")
print(f"  contrast  {' + '.join(CYTOSTATIC)}  against  {' + '.join(COMPARATOR)}")
print(f"  hypothesis: {mechanism_plan['primary_hypothesis']}")

single_agents = arms.loc[~arms.is_control & ~arms.treatment.str.contains("+", regex=False)].copy()
single_agents["compound"] = single_agents.treatment.map(lambda label: COMPOUND_OF_LABEL.get(label, label))
single_agents["family"] = single_agents.treatment.map(FAMILY_OF_LABEL)
unmapped = sorted(single_agents.loc[single_agents.family.isna(), "treatment"].unique())
compound_atlas = single_agents.loc[single_agents.family.notna()].copy()
assert (compound_atlas.groupby("compound").family.nunique() == 1).all(), "a compound is in two families"
compound_atlas["post_nadir"] = compound_atlas.arm_id.isin(post_nadir_ids)
compound_atlas["depth_score"] = -compound_atlas.best_average_response            # higher = deeper
compound_atlas["durability_score"] = np.where(compound_atlas.post_nadir,
                                              -compound_atlas.post_nadir_growth_rate, np.nan)  # higher = slower

compound_table = compound_atlas.groupby("compound").agg(
    family=("family", "first"), published_target=("treatment_target", "first"),
    labels=("treatment", lambda t: "; ".join(sorted(t.unique()))),
    arms=("arm_id", "size"), pdx_models=("model_id", "nunique"),
    durability_values=("post_nadir", "sum")).reset_index().sort_values(["family", "compound"])
family_table = compound_atlas.groupby("family").agg(
    compounds=("compound", "nunique"), arms=("arm_id", "size"), pdx_models=("model_id", "nunique"),
    post_nadir_arms=("post_nadir", "sum"), post_nadir_share=("post_nadir", "mean")
).sort_values("post_nadir_share", ascending=False)
record("n_single_agent_compounds", compound_atlas.compound.nunique())
record("n_families", compound_atlas.family.nunique())
print(f"\n  single-agent labels without a family (left out): {unmapped if unmapped else 'none'}")
print(f"  compounds {compound_atlas.compound.nunique()} in {compound_atlas.family.nunique()} families; "
      f"arms {len(compound_atlas):,}; post-nadir arms {int(compound_atlas.post_nadir.sum()):,}")
print("\n  families, with the share of arms in the post-nadir set (the arms that carry durability)")
print(family_table.round(3).to_string())

# One value per compound per PDX model (dose variants averaged), then every
# pair of compounds within the model, oriented compound_a − compound_b.
per_compound = (compound_atlas.groupby(["model_id", "compound", "family"], as_index=False)
                .agg(depth_score=("depth_score", "mean"), durability_score=("durability_score", "mean")))
compound_pair_rows = []
for model_id, block in per_compound.groupby("model_id", sort=True):
    records_in_model = block.sort_values("compound").to_dict("records")
    for first, second in combinations(records_in_model, 2):
        compound_pair_rows.append({"model_id": model_id, "compound_a": first["compound"], "compound_b": second["compound"],
                          "family_a": first["family"], "family_b": second["family"],
                          "depth_difference": first["depth_score"] - second["depth_score"],
                          "durability_difference": first["durability_score"] - second["durability_score"]})
compound_pairs = pd.DataFrame(compound_pair_rows)
cross_family = compound_pairs.loc[compound_pairs.family_a != compound_pairs.family_b].copy()
a_first = cross_family.family_a.isin(CYTOSTATIC) & cross_family.family_b.isin(COMPARATOR)
b_first = cross_family.family_b.isin(CYTOSTATIC) & cross_family.family_a.isin(COMPARATOR)
contrast_pairs = cross_family.loc[a_first | b_first].copy()
sign = np.where(a_first.loc[contrast_pairs.index], 1.0, -1.0)
contrast_pairs["depth"] = sign * contrast_pairs.depth_difference
contrast_pairs["durability"] = sign * contrast_pairs.durability_difference
contrast_pairs["cytostatic_compound"] = np.where(sign > 0, contrast_pairs.compound_a, contrast_pairs.compound_b)
contrast_pairs["comparator_compound"] = np.where(sign > 0, contrast_pairs.compound_b, contrast_pairs.compound_a)
contrast_pairs["cytostatic_family"] = contrast_pairs.cytostatic_compound.map(
    compound_table.set_index("compound").family)
contrast_pairs["comparator_family"] = contrast_pairs.comparator_compound.map(
    compound_table.set_index("compound").family)
durability_pairs = contrast_pairs.dropna(subset=["durability"])
cytostatic_compounds = sorted(compound_table.loc[compound_table.family.isin(CYTOSTATIC), "compound"])
record("n_contrast_pairs_durability", len(durability_pairs))
record("n_contrast_models_durability", durability_pairs.model_id.nunique())
record("n_cytostatic_compounds", len(cytostatic_compounds))
record("n_cytostatic_compounds_with_pairs", durability_pairs.cytostatic_compound.nunique())
print(f"\n  within-model compound pairs {len(compound_pairs):,}; between families {len(cross_family):,}")
print(f"  cytostatic against comparator: {len(contrast_pairs):,} pairs in "
      f"{contrast_pairs.model_id.nunique()} PDX models for depth; {len(durability_pairs):,} pairs in "
      f"{durability_pairs.model_id.nunique()} PDX models for durability")
print(f"  cytostatic compounds {len(cytostatic_compounds)}, of which "
      f"{durability_pairs.cytostatic_compound.nunique()} have durability pairs")
save_record(compound_table, "compound_families")
```

```python
# ==========================================================================
# 11.2 Prespecified analyses: bootstrap over PDX models and permutation of
#      family labels across compounds
# ==========================================================================

N_MODEL_BOOTSTRAP, N_LABEL_PERMUTATIONS = 10_000, 10_000
model_means = contrast_pairs.groupby("model_id")[["depth", "durability"]].mean()
depth_result = bootstrap_mean(model_means.depth, N_MODEL_BOOTSTRAP, SEEDS["mechanism_class_model"] + 1)
durability_result = bootstrap_mean(model_means.durability, N_MODEL_BOOTSTRAP,
                                   SEEDS["mechanism_class_model"] + 2)
record("mechanism_durability", durability_result["estimate"])
record("mechanism_durability_low", durability_result["low"])
record("mechanism_durability_high", durability_result["high"])
record("mechanism_durability_p", durability_result["p_two_sided"])
record("mechanism_durability_fold_28", float(np.exp(28 * durability_result["estimate"])))
record("mechanism_depth", depth_result["estimate"])
record("mechanism_depth_low", depth_result["low"])
record("mechanism_depth_high", depth_result["high"])
record("mechanism_depth_p", depth_result["p_two_sided"])
print("11.2 PRESPECIFIED ANALYSES (cytostatic minus comparator; higher is better on both endpoints)")
print(f"  durability  {durability_result['estimate']:.4f} per day (95% CI {durability_result['low']:.4f} to "
      f"{durability_result['high']:.4f}), {p_label(durability_result['p_two_sided'], 1 / N_MODEL_BOOTSTRAP)}, "
      f"{durability_result['n']} PDX models; {np.exp(28 * durability_result['estimate']):.2f}-fold in volume "
      f"28 days after the nadir")
print(f"  depth       {depth_result['estimate']:.1f} points (95% CI {depth_result['low']:.1f} to "
      f"{depth_result['high']:.1f}), {p_label(depth_result['p_two_sided'], 1 / N_MODEL_BOOTSTRAP)}, "
      f"{depth_result['n']} PDX models")
print("  predicted direction: slower regrowth (positive durability) and shallower depth (negative depth)")
print(f"    durability prediction supported: {durability_result['low'] > 0}")
print(f"    depth prediction supported:      {depth_result['high'] < 0}")

# Permutation of family labels across compounds (family sizes kept). Every
# within-model compound pair with durability on both members is eligible, so
# a relabelled pair can enter the contrast whatever its original families.
compound_codes = {compound: i for i, compound in enumerate(sorted(per_compound.compound.unique()))}
family_names = sorted(compound_table.family.unique())
family_code = {family: i for i, family in enumerate(family_names)}
compound_family_codes = np.array([family_code[compound_table.set_index("compound").family[c]]
                                  for c in compound_codes])
durable_pairs = compound_pairs.dropna(subset=["durability_difference"])
pair_model_codes = pd.Categorical(durable_pairs.model_id).codes
pair_a = durable_pairs.compound_a.map(compound_codes).to_numpy()
pair_b = durable_pairs.compound_b.map(compound_codes).to_numpy()
pair_durability = durable_pairs.durability_difference.to_numpy(float)
cytostatic_codes = [family_code[f] for f in CYTOSTATIC]
comparator_codes = [family_code[f] for f in COMPARATOR]
observed_label_statistic, _, _ = family_contrast_statistic(
    pair_model_codes, compound_family_codes[pair_a], compound_family_codes[pair_b], pair_durability,
    cytostatic_codes, comparator_codes)
permutation_rng = np.random.default_rng(SEEDS["mechanism_label_permutation"])
label_null = np.empty(N_LABEL_PERMUTATIONS)
for index in range(N_LABEL_PERMUTATIONS):
    shuffled = permutation_rng.permutation(compound_family_codes)
    label_null[index] = family_contrast_statistic(pair_model_codes, shuffled[pair_a], shuffled[pair_b],
                                                  pair_durability, cytostatic_codes, comparator_codes)[0]
label_null = label_null[np.isfinite(label_null)]
label_p_one = (1 + np.sum(label_null >= observed_label_statistic)) / (1 + len(label_null))
label_p_two = (1 + np.sum(np.abs(label_null) >= abs(observed_label_statistic))) / (1 + len(label_null))
record("label_permutation_p_one_sided", float(label_p_one))
print(f"\n  family labels permuted across {len(compound_codes)} compounds ({len(label_null):,} permutations)")
print(f"    statistic {observed_label_statistic:.4f}; one-sided P = {format_p(label_p_one)}, "
      f"two-sided P = {format_p(label_p_two)}")

# ==========================================================================
# 11.3 Sensitivity analyses added after the plan was fixed
# ==========================================================================

N_COMPOUND_BOOTSTRAP = 20_000
compound_by_model = durability_pairs.groupby(["cytostatic_compound", "comparator_compound", "model_id"],
                                             as_index=False).durability.mean()
compound_effects = compound_by_model.groupby("cytostatic_compound", as_index=False).agg(
    effect=("durability", "mean"), pdx_models=("model_id", "nunique"))
compound_effects["family"] = compound_effects.cytostatic_compound.map(compound_table.set_index("compound").family)
compound_effects = compound_effects.sort_values("effect").reset_index(drop=True)
compound_values = compound_effects.effect.to_numpy(float)
compound_rng = np.random.default_rng(SEEDS["mechanism_class_compound"])
compound_draws = compound_values[compound_rng.integers(0, len(compound_values),
                                                       size=(N_COMPOUND_BOOTSTRAP, len(compound_values)))].mean(axis=1)
compound_low, compound_high = np.percentile(compound_draws, [2.5, 97.5])
sign_rng = np.random.default_rng(SEEDS["mechanism_class_compound"] + 101)
sign_null = (sign_rng.choice([-1.0, 1.0], size=(N_COMPOUND_BOOTSTRAP, len(compound_values)))
             * compound_values).mean(axis=1)
compound_effect = float(compound_values.mean())
loco_rows = []
for position, compound in enumerate(sorted(compound_effects.cytostatic_compound)):
    remaining = compound_effects.loc[compound_effects.cytostatic_compound != compound, "effect"].to_numpy(float)
    loco_rng = np.random.default_rng(SEEDS["mechanism_class_compound"] + 200 + position)
    draws = remaining[loco_rng.integers(0, len(remaining), size=(N_COMPOUND_BOOTSTRAP, len(remaining)))].mean(axis=1)
    loco_rows.append({"omitted": compound, "estimate": remaining.mean(),
                      "low": np.percentile(draws, 2.5), "high": np.percentile(draws, 97.5)})
leave_one_out = pd.DataFrame(loco_rows)
record("compound_effect", compound_effect)
record("compound_effect_low", float(compound_low))
record("compound_effect_high", float(compound_high))
record("compound_effect_p", bootstrap_p_two_sided(compound_draws))
record("n_compounds_positive", int((compound_values > 0).sum()))
record("loco_all_above_zero", int((leave_one_out.low > 0).all()))
if "encorafenib" in set(compound_effects.cytostatic_compound):
    record("encorafenib_effect", float(compound_effects.set_index("cytostatic_compound").effect["encorafenib"]))
print("11.3 SENSITIVITY ANALYSES ADDED AFTER THE PLAN WAS FIXED")
print(f"  (a) compound as the unit ({len(compound_values)} cytostatic compounds)")
print(compound_effects.round(5).to_string(index=False))
print(f"      mean effect {format_ci(compound_effect, compound_low, compound_high, 1, 4, ' per day')}; "
      f"bootstrap P two-sided {format_p(bootstrap_p_two_sided(compound_draws), 1 / N_COMPOUND_BOOTSTRAP)}")
sign_flip_p_one = float(np.mean(sign_null >= compound_effect))
sign_flip_p_two = float(np.mean(np.abs(sign_null) >= abs(compound_effect)))
record("sign_flip_p_one_sided", sign_flip_p_one)
record("sign_flip_p_two_sided", sign_flip_p_two)
print(f"      sign-flip randomization: one-sided {p_label(sign_flip_p_one, 1 / N_COMPOUND_BOOTSTRAP)}, "
      f"two-sided {p_label(sign_flip_p_two, 1 / N_COMPOUND_BOOTSTRAP)}")
print(f"      compounds with a positive effect {int((compound_values > 0).sum())} of {len(compound_values)}")
print("      leaving out one compound at a time")
print(leave_one_out.round(5).to_string(index=False))

# (b) exact randomization over every assignment of three families against two
model_codes_all = pd.Categorical(cross_family.model_id).codes
family_a_codes = cross_family.family_a.map(family_code).to_numpy()
family_b_codes = cross_family.family_b.map(family_code).to_numpy()
exact_rows = []
for group_one in combinations(family_names, len(CYTOSTATIC)):
    remaining_families = [f for f in family_names if f not in group_one]
    for group_two in combinations(remaining_families, len(COMPARATOR)):
        row = {"group_one": " + ".join(group_one), "group_two": " + ".join(group_two),
               "is_planned": set(group_one) == set(CYTOSTATIC) and set(group_two) == set(COMPARATOR),
               "cytostatic_families_kept": len(set(group_one) & set(CYTOSTATIC)),
               "small_family_comparator": bool({"developmental", "hormonal"} & set(group_two))}
        for endpoint in ("durability", "depth"):
            row[endpoint], row[f"{endpoint}_models"], _ = family_contrast_statistic(
                model_codes_all, family_a_codes, family_b_codes,
                cross_family[f"{endpoint}_difference"].to_numpy(float),
                [family_code[f] for f in group_one], [family_code[f] for f in group_two])
        exact_rows.append(row)
exact_splits = pd.DataFrame(exact_rows)
planned = exact_splits.loc[exact_splits.is_planned].iloc[0]
exact_p = {endpoint: {"one": float(np.mean(exact_splits[endpoint] >= planned[endpoint])),
                      "two": float(np.mean(np.abs(exact_splits[endpoint]) >= abs(planned[endpoint])))}
           for endpoint in ("durability", "depth")}
tail = exact_splits.loc[exact_splits.durability >= planned.durability]
record("n_exact_assignments", len(exact_splits))
record("exact_p_durability_one_sided", exact_p["durability"]["one"])
record("exact_p_depth_two_sided", exact_p["depth"]["two"])
print(f"\n  (b) exact randomization over {len(exact_splits):,} assignments of three families against two")
for endpoint in ("durability", "depth"):
    print(f"      {endpoint:<10} planned statistic {planned[endpoint]:.4f}; one-sided P = "
          f"{exact_p[endpoint]['one']:.3f}, two-sided P = {exact_p[endpoint]['two']:.3f}")
print(f"      assignments at or above the planned durability statistic: {len(tail)}; "
      f"{100 * tail.small_family_comparator.mean():.1f}% use developmental or hormonal as a comparator; "
      f"{100 * (tail.cytostatic_families_kept >= 2).mean():.1f}% keep two or three cytostatic families; "
      f"median PDX models {tail.durability_models.median():.0f} (planned {int(planned.durability_models)})")

# (c) mixed-effects model with random intercepts for PDX model and family pair
#     The intercept is the cytostatic-minus-comparator durability difference.
#     Three random-effect structures are fitted:
#       crossed  PDX model and family pair as crossed random intercepts (one
#                group holding every pair, both factors as variance
#                components); this is the model the section note describes
#                and the one recorded;
#       nested   a PDX-model random intercept (re_formula="1") and family
#                pair as a variance component within PDX model;
#       no PDX-model intercept  family pair within PDX model only. statsmodels
#                fits this when a variance component is given without
#                re_formula, because it then leaves out the default
#                intercept for the grouping factor.
mixed_frame = durability_pairs.assign(family_pair=[" | ".join(sorted(pair)) for pair in
                                                   zip(durability_pairs.cytostatic_family,
                                                       durability_pairs.comparator_family)])
MIXED_SPECIFICATIONS = {
    "crossed": dict(groups=np.ones(len(mixed_frame)), re_formula="0",
                    vc_formula={"pdx_model": "0 + C(model_id)", "family_pair": "0 + C(family_pair)"}),
    "nested": dict(groups=mixed_frame.model_id, re_formula="1",
                   vc_formula={"family_pair": "0 + C(family_pair)"}),
    "no PDX-model intercept": dict(groups=mixed_frame.model_id,
                                   vc_formula={"family_pair": "0 + C(family_pair)"}),
}
mixed_rows, mixed_fits = [], {}
for label, specification in MIXED_SPECIFICATIONS.items():
    fitted, converged, messages = fit_mixed_model("durability ~ 1", mixed_frame, **specification)
    mixed_fits[label] = fitted
    components = dict(zip(fitted.model.exog_vc.names, np.asarray(fitted.vcomp, float)))
    if fitted.cov_re.size:                       # the PDX-model intercept of the nested form
        components["pdx_model"] = float(fitted.cov_re.iloc[0, 0])
    mixed_rows.append({"structure": label, "intercept": float(fitted.params["Intercept"]),
                       "se": float(fitted.bse["Intercept"]), "p_value": float(fitted.pvalues["Intercept"]),
                       "pdx_model_variance": components.get("pdx_model", np.nan),
                       "family_pair_variance": components.get("family_pair", np.nan),
                       "residual_variance": float(fitted.scale), "converged": converged,
                       "warnings": "; ".join(messages)})
mixed_models = pd.DataFrame(mixed_rows)
mixed_fit = mixed_fits["crossed"]
record("mixed_model_intercept", float(mixed_fit.params["Intercept"]))
record("mixed_model_se", float(mixed_fit.bse["Intercept"]))
record("mixed_model_p", float(mixed_fit.pvalues["Intercept"]))
record("mixed_model_converged", int(mixed_models.set_index("structure").converged["crossed"]))
nested_row = mixed_models.set_index("structure").loc["nested"]
record("mixed_model_nested_intercept", float(nested_row.intercept))
record("mixed_model_nested_p", float(nested_row.p_value))
print(f"\n  (c) mixed-effects models: {len(mixed_frame)} pairs, {mixed_frame.model_id.nunique()} PDX models, "
      f"{mixed_frame.family_pair.nunique()} family pairs")
print(mixed_models.drop(columns="warnings").assign(
    p_value=lambda d: d.p_value.map(format_p)).to_string(index=False, float_format=lambda v: f"{v:.3g}"))
for row in mixed_models.itertuples(index=False):
    if row.warnings:
        print(f"      {row.structure}: {row.warnings}")
print(f"      recorded (crossed): intercept {mixed_fit.params['Intercept']:.4f} per day "
      f"(SE {mixed_fit.bse['Intercept']:.4f}), {p_label(mixed_fit.pvalues['Intercept'])}")

breakdown = durability_pairs.groupby(["cytostatic_family", "comparator_family"]).agg(
    pairs=("durability", "size"), pdx_models=("model_id", "nunique"),
    mean_difference=("durability", "mean"), median_difference=("durability", "median")).reset_index()
cell_cycle_cytotoxic = breakdown.loc[(breakdown.cytostatic_family == "cell cycle")
                                     & (breakdown.comparator_family == "cytotoxic")]
if len(cell_cycle_cytotoxic):
    record("cell_cycle_vs_cytotoxic_median", float(cell_cycle_cytotoxic.median_difference.iloc[0]))
print("\n  durability difference by family pair")
print(breakdown.round(5).to_string(index=False))
mechanism_results = pd.DataFrame([
    {"analysis": "bootstrap over PDX models, durability (planned)", **durability_result},
    {"analysis": "bootstrap over PDX models, depth (planned)", **depth_result},
    {"analysis": "family labels permuted across compounds (planned)", "estimate": observed_label_statistic,
     "p_one_sided": label_p_one, "p_two_sided": label_p_two, "n": len(label_null)},
    {"analysis": "compound bootstrap (added)", "estimate": compound_effect, "low": compound_low,
     "high": compound_high, "p_two_sided": bootstrap_p_two_sided(compound_draws), "n": len(compound_values)},
    {"analysis": "exact family randomization, durability (added)", "estimate": planned.durability,
     "p_one_sided": exact_p["durability"]["one"], "p_two_sided": exact_p["durability"]["two"],
     "n": len(exact_splits)},
    {"analysis": "exact family randomization, depth (added)", "estimate": planned.depth,
     "p_one_sided": exact_p["depth"]["one"], "p_two_sided": exact_p["depth"]["two"], "n": len(exact_splits)},
    *[{"analysis": f"mixed-effects model, {row.structure} (added)", "estimate": row.intercept,
       "p_two_sided": row.p_value, "n": len(mixed_frame)} for row in mixed_models.itertuples(index=False)]])
save_record(mechanism_results, "mechanism_class_results")
save_record(exact_splits, "mechanism_class_exact_splits")

# Data for the mechanism-class figure: each cytostatic compound's effect,
# the leave-one-compound-out estimates and every cytostatic-comparator pair
# (the exact assignments are saved above).
save_record(compound_effects, "mechanism_class_compound_effects")
save_record(leave_one_out, "mechanism_class_leave_one_out")
save_record(contrast_pairs, "mechanism_class_contrast_pairs")
```

## 12 · Drug exposure and observation span

Section 5 compared the two regrowth models under constant drug effect. Here both models are extended so that the kill term follows simulated drug exposure. Exposure q(t) follows a one-compartment model with first-order elimination and instantaneous absorption, with a dose once daily from day 0 to the end of the window. Mouse half-lives were not available for these agents and dosing schedules are not recorded in the source data, so two half-lives are used, 0.25 and 1 day. Results are conditional on that range and do not refer to any agent.

- **Tolerant subpopulation:** sensitive cells grow as dV/dt = [g − k·q(t)]V and tolerant cells as dV/dt = gV.
- **Declining effect:** dV/dt = [g − k·q(t)·e^(−δt)]V. Clearance is carried by q(t), and δ carries any loss of effect beyond clearance.

The cumulative exposure is integrated analytically and scaled so that mean exposure over [0, T] is 1, with T the last measurement day of the schedule being modeled. The scaling is a constant factor and is absorbed into the kill rate k, but the scale of k therefore depends on the schedule. Both models have four parameters. They are fitted by bounded least squares from three starting points (the bounds of Section 5), and a fit that ends on a bound is flagged as pinned.

**Paired replicates.** Each replicate draws one schedule from the source study and one set of shared parameters: baseline volume, growth rate, kill rate and a standard-normal noise vector. Both mechanisms are generated from these, so the mechanism is the only systematic difference between the two simulated arms. The parameter distributions are listed in `paired_exposure_trajectories` (Section 1) and differ from the priors of Section 5. Both models are fitted to each arm. Three quantities are recorded:

- **Structural separation:** the RMSE of the incorrect model fitted to the noise-free arm.
- **Decisive rate:** the share of arms whose AIC difference is at least 2. A smaller difference is recorded as no decision. Section 5.5 had no such category and always chose the lower AIC.
- **Accuracy:** the share of decisive calls that name the generating model. Counts and Wilson 95% intervals are given because some cells have few decisive calls.

**Termination.** With growth rates up to about 0.13 per day, many simulated arms pass 2,000 mm³ before the end of a long window, which the source study would not have observed. Every arm is therefore scored twice: with the full schedule, and ended at its first measurement above 2,000 mm³ (the rule of Section 5). An arm left with fewer than five measurements counts as undecidable. The terminated version matches the source study and is the one recorded; the full-schedule version is printed beside it.

The section has four parts:

- **12.1** repeats the comparison over 28 days on every schedule reaching day 28. This gives the structural separation over the study's typical window.
- **12.2** measures separation at windows from 14 to 84 days on a fixed pool: the schedules that reach day 84, each cut at the window. A fixed pool keeps measurement density from changing with the window. Fold changes from 14 and 28 days to 84 days, and the exponent of a log–log fit of separation on window length, are computed within this pool. Each fold change and exponent carries a 95% interval from resampling the paired replicates of each window (2,000 resamples). The 14-day separation is small and rests on fewer arms, because fewer schedules keep five measurements by day 14, so the fold change from day 14 is the least stable of these quantities.
- **12.3** scores the AIC verdict at windows of 28, 56 and 84 days and four noise levels, including the measured noise of Section 3.6. Decisive rates carry Wilson 95% intervals beside the accuracies.
- **12.4** thins the 84-day schedules to every second or third measurement, so that span and number of measurements can be separated.

Each part has its own seed and 50 paired replicates per condition. These simulations bear on the design of future studies. Few of the source study's schedules reach 84 days (Section 3.1), and the arms that do are a selected subset, because termination depends on volume (Section 6.3).

**Inputs:** `empirical_grids`, `MEASUREMENT_NOISE`, and `approximation`, `mechanism_discrimination` and `ILLUSTRATIVE_TOLERANT` from Section 5.
**Outputs:** `exposure_paired`, `exposure_window_sweep`, `exposure_confirmation`, `exposure_thinning` and their summaries; figure data `exposure_residual_example.csv` (a worked example on one typical schedule).

```python
# ==========================================================================
# 12.1 Paired simulations over 28 days on the study's own schedules
#     One schedule is drawn from every schedule reaching day 28 and cut
#     there. Both mechanisms are generated from one parameter draw and scored
#     at four noise levels; zero noise gives the structural separation (the
#     RMSE of the incorrect model). Each arm is scored with the full schedule
#     and with the termination rule applied.
# ==========================================================================

set_section("12")
EXPOSURE_HALF_LIVES = (0.25, 1.0)          # days; two clearance regimes, not agent values
HALF_LIFE_TAG = {0.25: "hl025", 1.0: "hl100"}
DOSING_INTERVAL_DAYS = 1.0                 # once daily from day 0
TIE_THRESHOLD = 2.0                        # |ΔAIC| below this is recorded as no decision
N_EXPOSURE_REPLICATES = 50                 # paired replicates per condition
STUDY_WINDOW_DAYS = 28
PAIRED_NOISE_LEVELS = (0.0, 0.025, 0.05, MEASUREMENT_NOISE)
TERMINATION_RULES = ("applied", "not applied")

paired_rng = np.random.default_rng(SEEDS["exposure_separation"])
study_pool = [grid for grid in empirical_grids if grid.max() >= STUDY_WINDOW_DAYS]
paired_rows = []
for half_life in EXPOSURE_HALF_LIVES:
    dose_days = dosing_days(DOSING_INTERVAL_DAYS, STUDY_WINDOW_DAYS)
    for noise_sd in PAIRED_NOISE_LEVELS:
        for replicate in range(N_EXPOSURE_REPLICATES):
            days = draw_schedule(paired_rng, study_pool, STUDY_WINDOW_DAYS)
            if days is None:
                continue
            clean, noise = paired_exposure_trajectories(paired_rng, days, dose_days, half_life)
            for row in score_exposure_pair(clean, noise, noise_sd, days, dose_days, half_life,
                                           TIE_THRESHOLD):
                paired_rows.append({"half_life_days": half_life, "noise_sd": noise_sd,
                                    "replicate": replicate, **row})
exposure_paired = pd.DataFrame(paired_rows)

# Structural separation: the incorrect model fitted to noise-free arms. The
# correct model's RMSE on the same arms checks the fitter (it should be ~0).
noise_free = exposure_paired.loc[exposure_paired.noise_sd == 0.0]
paired_separation = noise_free.groupby(["termination", "half_life_days", "true_mechanism"]).agg(
    arms=("rmse_wrong", "count"), median_measurements=("n_measurements", "median"),
    median_separation=("rmse_wrong", "median"),
    p10=("rmse_wrong", lambda v: v.quantile(0.10)), p90=("rmse_wrong", lambda v: v.quantile(0.90)),
    median_rmse_correct_model=("rmse_right", "median"), terminated_share=("terminated", "mean"))
separation_28 = noise_free.groupby("termination").rmse_wrong.median()
record("exposure_separation_28", float(separation_28["applied"]))
record("exposure_separation_28_without_termination", float(separation_28["not applied"]))

paired_summary = (exposure_paired.loc[exposure_paired.noise_sd > 0]
                  .groupby(["termination", "half_life_days", "noise_sd"])
                  .apply(summarize_verdicts, include_groups=False).reset_index())
scored = exposure_paired.loc[(exposure_paired.noise_sd > 0) & (exposure_paired.termination == "applied")]

print("12.1 PAIRED SIMULATIONS OVER 28 DAYS (all schedules reaching day 28)")
print(f"  schedules in the pool {len(study_pool):,}; {N_EXPOSURE_REPLICATES} paired replicates per "
      f"half-life and noise level; once-daily dosing; tie threshold |ΔAIC| < {TIE_THRESHOLD}")
print("\n  structural separation (incorrect model fitted to noise-free arms), log units")
print(paired_separation.round(4).to_string())
print(f"\n  median structural separation, both half-lives: {separation_28['applied']:.4f} with the "
      f"termination rule, {separation_28['not applied']:.4f} without; measurement noise "
      f"{MEASUREMENT_NOISE:.4f} (ratio {separation_28['applied'] / MEASUREMENT_NOISE:.3f})")
print("\n  AIC verdicts with noise (accuracy among decisive calls, Wilson 95% interval)")
print(paired_summary.round(3).to_string(index=False))
print("\n  arms with a pinned parameter in either fit (termination rule applied), by generating model")
print(scored.groupby("true_mechanism").any_pinned.agg(["mean", "sum", "size"]).round(3).to_string())
unpinned = scored.loc[~scored.any_pinned & scored.verdict.isin(MECHANISMS)]
print(f"  accuracy among decisive calls without pinned fits: {unpinned.correct.mean():.3f} "
      f"({int(unpinned.correct.sum())} of {len(unpinned)})")
save_record(paired_separation.reset_index(), "exposure_paired_separation")
save_record(paired_summary, "exposure_paired_verdicts")

# ==========================================================================
# 12.2 Structural separation against window length on a fixed pool
#     The pool is every schedule reaching day 84, cut at each window, so the
#     schedules do not change with the window. Fold changes are taken from
#     14 and from 28 days to 84 days, and the growth of separation with the
#     window is summarized by the slope of log median separation on log
#     window.
# ==========================================================================

SWEEP_WINDOWS = (14, 21, 28, 42, 56, 84)

window_rng = np.random.default_rng(SEEDS["exposure_window"])
long_pool = [grid for grid in empirical_grids if grid.max() >= max(SWEEP_WINDOWS)]
sweep_rows = []
for half_life in EXPOSURE_HALF_LIVES:
    for window_days in SWEEP_WINDOWS:
        dose_days = dosing_days(DOSING_INTERVAL_DAYS, window_days)
        for replicate in range(N_EXPOSURE_REPLICATES):
            days = draw_schedule(window_rng, long_pool, window_days)
            if days is None:
                continue
            clean, noise = paired_exposure_trajectories(window_rng, days, dose_days, half_life)
            for row in score_exposure_pair(clean, noise, 0.0, days, dose_days, half_life,
                                           TIE_THRESHOLD):
                sweep_rows.append({"half_life_days": half_life, "window_days": window_days,
                                    "replicate": replicate, **row})
exposure_window_sweep = pd.DataFrame(sweep_rows)
separation_by_window = exposure_window_sweep.groupby(
    ["termination", "half_life_days", "window_days"]).agg(
    arms=("rmse_wrong", "count"), median_measurements=("n_measurements", "median"),
    median_separation=("rmse_wrong", "median"),
    p10=("rmse_wrong", lambda v: v.quantile(0.10)), p90=("rmse_wrong", lambda v: v.quantile(0.90)),
    median_rmse_correct_model=("rmse_right", "median"),
    terminated_share=("terminated", "mean")).reset_index()

growth_rows = []
for (rule, half_life), block in separation_by_window.groupby(["termination", "half_life_days"]):
    by_window = block.set_index("window_days").median_separation
    exponent = np.polyfit(np.log(by_window.index.to_numpy(float)), np.log(by_window.to_numpy()), 1)[0]
    growth_rows.append({"termination": rule, "half_life_days": half_life,
                        "separation_14": by_window[14], "separation_28": by_window[28],
                        "separation_84": by_window[84],
                        "fold_14_to_84": by_window[84] / by_window[14],
                        "fold_28_to_84": by_window[84] / by_window[28],
                        "log_log_exponent": exponent})
separation_growth = pd.DataFrame(growth_rows)
for row in separation_growth.itertuples():
    suffix = "" if row.termination == "applied" else "_without_termination"
    tag = HALF_LIFE_TAG[row.half_life_days]
    record(f"exposure_fold_28_to_84_{tag}{suffix}", float(row.fold_28_to_84))
    record(f"exposure_fold_14_to_84_{tag}{suffix}", float(row.fold_14_to_84))
    record(f"exposure_separation_exponent_{tag}{suffix}", float(row.log_log_exponent))
    record(f"exposure_separation_84_{tag}{suffix}", float(row.separation_84))

# Intervals for the fold changes and the exponent. Within each window the
# paired replicates are resampled with replacement (both arms of a replicate
# move together), the median separation is recomputed per window, and the
# fold changes and the log-log slope are recomputed from those medians.
N_SEPARATION_BOOTSTRAP = N_BOOTSTRAP
separation_rng = np.random.default_rng(SEEDS["separation_growth_bootstrap"])
growth_interval_rows = []
for (rule, half_life), block in exposure_window_sweep.groupby(["termination", "half_life_days"]):
    values_by_window = {}
    for window_days, window_block in block.dropna(subset=["rmse_wrong"]).groupby("window_days"):
        values_by_window[window_days] = [group.rmse_wrong.to_numpy(float)
                                         for _, group in window_block.groupby("replicate", sort=True)]
    windows = np.array(sorted(values_by_window), dtype=float)
    draws = []
    for _ in range(N_SEPARATION_BOOTSTRAP):
        medians = {}
        for window_days in windows:
            groups_in_window = values_by_window[window_days]
            picked = separation_rng.integers(0, len(groups_in_window), size=len(groups_in_window))
            medians[window_days] = float(np.median(np.concatenate([groups_in_window[i] for i in picked])))
        median_values = np.array([medians[w] for w in windows])
        if np.all(median_values > 0):
            draws.append((medians[84] / medians[14], medians[84] / medians[28],
                          np.polyfit(np.log(windows), np.log(median_values), 1)[0]))
    draws = np.array(draws)
    bounds = np.percentile(draws, [2.5, 97.5], axis=0)
    growth_interval_rows.append({"termination": rule, "half_life_days": half_life, "resamples": len(draws),
                                 "fold_14_to_84_low": bounds[0, 0], "fold_14_to_84_high": bounds[1, 0],
                                 "fold_28_to_84_low": bounds[0, 1], "fold_28_to_84_high": bounds[1, 1],
                                 "exponent_low": bounds[0, 2], "exponent_high": bounds[1, 2]})
separation_growth = separation_growth.merge(pd.DataFrame(growth_interval_rows),
                                            on=["termination", "half_life_days"], how="left")
for row in separation_growth.itertuples():
    suffix = "" if row.termination == "applied" else "_without_termination"
    tag = HALF_LIFE_TAG[row.half_life_days]
    for quantity in ("fold_28_to_84", "fold_14_to_84"):
        record(f"exposure_{quantity}_{tag}{suffix}_low", float(getattr(row, f"{quantity}_low")))
        record(f"exposure_{quantity}_{tag}{suffix}_high", float(getattr(row, f"{quantity}_high")))
    record(f"exposure_separation_exponent_{tag}{suffix}_low", float(row.exponent_low))
    record(f"exposure_separation_exponent_{tag}{suffix}_high", float(row.exponent_high))

print("\n12.2 STRUCTURAL SEPARATION AGAINST WINDOW (fixed pool of schedules reaching day 84)")
print(f"  schedules in the pool {len(long_pool):,} of {len(empirical_grids):,} "
      f"({100 * len(long_pool) / len(empirical_grids):.1f}%)")
print(separation_by_window.round(5).to_string(index=False))
print(f"\n  growth of separation with the window; 95% intervals from {N_SEPARATION_BOOTSTRAP:,} resamples "
      "of the paired replicates within each window")
print(separation_growth.round(4).to_string(index=False))
print(f"\n  The 28-day value of 12.1 ({separation_28['applied']:.4f}) comes from every schedule reaching "
      "day 28;\n  the fold changes above compare windows within the fixed pool only.")
save_record(separation_by_window, "exposure_separation_by_window")
save_record(separation_growth, "exposure_separation_growth")
```

```python
# ==========================================================================
# 12.3 AIC verdicts by window and noise level (fixed pool)
#     Windows of 28, 56 and 84 days on the schedules that reach day 84, at
#     four noise levels including the measured noise. The recorded values
#     are at the measured noise with the termination rule applied; ranges
#     are over the two half-lives.
# ==========================================================================

CONFIRM_WINDOWS = (28, 56, 84)
CONFIRM_NOISE_LEVELS = (0.025, 0.05, MEASUREMENT_NOISE, 0.15)

confirm_rng = np.random.default_rng(SEEDS["exposure_confirmation"])
confirm_rows = []
for half_life in EXPOSURE_HALF_LIVES:
    for window_days in CONFIRM_WINDOWS:
        dose_days = dosing_days(DOSING_INTERVAL_DAYS, window_days)
        for noise_sd in CONFIRM_NOISE_LEVELS:
            for replicate in range(N_EXPOSURE_REPLICATES):
                days = draw_schedule(confirm_rng, long_pool, window_days)
                if days is None:
                    continue
                clean, noise = paired_exposure_trajectories(confirm_rng, days, dose_days, half_life)
                for row in score_exposure_pair(clean, noise, noise_sd, days, dose_days, half_life,
                                               TIE_THRESHOLD):
                    confirm_rows.append({"half_life_days": half_life, "window_days": window_days,
                                         "noise_sd": noise_sd, "replicate": replicate, **row})
exposure_confirmation = pd.DataFrame(confirm_rows)
confirmation_summary = (exposure_confirmation
                        .groupby(["termination", "half_life_days", "window_days", "noise_sd"])
                        .apply(summarize_verdicts, include_groups=False).reset_index())
at_measured_noise = confirmation_summary.loc[np.isclose(confirmation_summary.noise_sd, MEASUREMENT_NOISE)]

for rule in TERMINATION_RULES:
    suffix = "" if rule == "applied" else "_without_termination"
    for window_days in (28, 84):
        cell = at_measured_noise.loc[(at_measured_noise.termination == rule)
                                     & (at_measured_noise.window_days == window_days)]
        record(f"exposure_decisive_{window_days}_min{suffix}", float(cell.decisive_rate.min()))
        record(f"exposure_decisive_{window_days}_max{suffix}", float(cell.decisive_rate.max()))
        record(f"exposure_accuracy_{window_days}_min{suffix}", float(cell.accuracy_when_decisive.min()))
        record(f"exposure_accuracy_{window_days}_max{suffix}", float(cell.accuracy_when_decisive.max()))
record("exposure_terminated_share_84", float(at_measured_noise.loc[
    (at_measured_noise.termination == "applied") & (at_measured_noise.window_days == 84),
    "terminated_share"].mean()))

print("12.3 AIC VERDICTS BY WINDOW AND NOISE LEVEL (fixed pool)")
print(f"  {len(exposure_confirmation) // 2:,} scored arms per termination rule")
print(confirmation_summary.round(3).to_string(index=False))
print(f"\n  at the measured noise ({MEASUREMENT_NOISE:.4f} log units)")
columns = ["termination", "half_life_days", "window_days", "n", "median_measurements", "decisive_rate",
           "decisive_low", "decisive_high", "decisive_calls", "correct_calls", "accuracy_when_decisive",
           "accuracy_low", "accuracy_high", "terminated_share"]
print(at_measured_noise[columns].round(3).to_string(index=False))
longest = exposure_confirmation.loc[(exposure_confirmation.window_days == max(CONFIRM_WINDOWS))
                                    & (exposure_confirmation.termination == "applied")
                                    & exposure_confirmation.verdict.isin(MECHANISMS)]
print("\n  accuracy among decisive calls at 84 days by generating model (termination rule applied)")
print(longest.groupby(["noise_sd", "true_mechanism"]).correct.mean().unstack().round(3).to_string())
save_record(confirmation_summary, "exposure_verdicts_by_window")

# ==========================================================================
# 12.4 Span or number of measurements: thinning the 84-day schedules
#     The 84-day schedules are thinned to every second or third measurement,
#     which brings the count near that of a 28-day schedule while keeping the
#     span. Each thinned arm is scored at the measured noise and its
#     structural separation is measured without noise.
# ==========================================================================

THINNING_STEPS = (1, 2, 3)
THINNING_WINDOW_DAYS = 84

thinning_rng = np.random.default_rng(SEEDS["exposure_thinning"])
dose_days = dosing_days(DOSING_INTERVAL_DAYS, THINNING_WINDOW_DAYS)
thinning_rows = []
for half_life in EXPOSURE_HALF_LIVES:
    for keep_every in THINNING_STEPS:
        for replicate in range(N_EXPOSURE_REPLICATES):
            days = draw_schedule(thinning_rng, long_pool, THINNING_WINDOW_DAYS, keep_every)
            if days is None:
                continue
            clean, noise = paired_exposure_trajectories(thinning_rng, days, dose_days, half_life)
            for noise_sd in (0.0, MEASUREMENT_NOISE):
                for row in score_exposure_pair(clean, noise, noise_sd, days, dose_days, half_life,
                                               TIE_THRESHOLD):
                    thinning_rows.append({"half_life_days": half_life, "keep_every": keep_every,
                                          "noise_sd": noise_sd, "replicate": replicate, **row})
exposure_thinning = pd.DataFrame(thinning_rows)
thinning_verdicts = (exposure_thinning.loc[exposure_thinning.noise_sd > 0]
                     .groupby(["termination", "half_life_days", "keep_every"])
                     .apply(summarize_verdicts, include_groups=False).reset_index())
thinning_separation = (exposure_thinning.loc[exposure_thinning.noise_sd == 0]
                       .groupby(["termination", "half_life_days", "keep_every"])
                       .agg(arms=("rmse_wrong", "count"), median_measurements=("n_measurements", "median"),
                            median_separation=("rmse_wrong", "median")).reset_index())
thinned = thinning_verdicts.loc[(thinning_verdicts.termination == "applied")
                                & (thinning_verdicts.keep_every == max(THINNING_STEPS))]
record("exposure_thinned_decisive_min", float(thinned.decisive_rate.min()))
record("exposure_thinned_decisive_max", float(thinned.decisive_rate.max()))
record("exposure_thinned_accuracy_min", float(thinned.accuracy_when_decisive.min()))
record("exposure_thinned_accuracy_max", float(thinned.accuracy_when_decisive.max()))
record("exposure_thinned_median_measurements", float(thinned.median_measurements.median()))

print("\n12.4 THINNING THE 84-DAY SCHEDULES")
print(f"  AIC verdicts at the measured noise ({MEASUREMENT_NOISE:.4f})")
print(thinning_verdicts.round(3).to_string(index=False))
print("\n  structural separation without noise")
print(thinning_separation.round(5).to_string(index=False))
reference_28 = at_measured_noise.loc[(at_measured_noise.termination == "applied")
                                     & (at_measured_noise.window_days == 28)]
print(f"\n  every third measurement over 84 days: median {thinned.median_measurements.median():.0f} "
      f"measurements, decisive rate {thinned.decisive_rate.min():.2f} to {thinned.decisive_rate.max():.2f}, "
      f"accuracy {thinned.accuracy_when_decisive.min():.2f} to {thinned.accuracy_when_decisive.max():.2f}")
print(f"  28-day schedules (12.3): median {reference_28.median_measurements.median():.0f} measurements, "
      f"decisive rate {reference_28.decisive_rate.min():.2f} to {reference_28.decisive_rate.max():.2f}, "
      f"accuracy {reference_28.accuracy_when_decisive.min():.2f} to "
      f"{reference_28.accuracy_when_decisive.max():.2f}")
save_record(thinning_verdicts, "exposure_thinning_verdicts")
save_record(thinning_separation, "exposure_thinning_separation")

# The worked example for the identifiability figure: a noise-free
# tolerant-subpopulation arm under explicit exposure (fixed parameters,
# 0.25-day half-life) on the typical schedule of the fixed pool, cut at 28
# and at 84 days, with the residual left by the best declining-effect fit.
DEMONSTRATION = {"log_baseline": np.log(200.0), "growth_rate": 0.05, "kill_rate": 0.13,
                 "tolerant_fraction": 0.08, "half_life_days": 0.25}
schedule_84 = typical_schedule(long_pool, 84)
example_rows = []
for window_days in (28, 84):
    days = schedule_84[schedule_84 <= window_days]
    doses = dosing_days(DOSING_INTERVAL_DAYS, window_days)
    truth_curve = log_volume_tolerant_exposure(
        days, DEMONSTRATION["log_baseline"], DEMONSTRATION["growth_rate"], DEMONSTRATION["kill_rate"],
        DEMONSTRATION["tolerant_fraction"], doses, DEMONSTRATION["half_life_days"])
    wrong_fit = fit_mechanism_exposure("declining drug effect", days, truth_curve, doses,
                                       DEMONSTRATION["half_life_days"])
    residual = truth_curve - exposure_forward("declining drug effect", doses,
                                              DEMONSTRATION["half_life_days"])(wrong_fit["x"], days)
    example_rows.append(pd.DataFrame({"window_days": window_days, "day": days, "log_volume": truth_curve,
                                      "residual": residual, "rmse_incorrect_model": wrong_fit["rmse"]}))
    record(f"exposure_example_rmse_{window_days}", float(wrong_fit["rmse"]))
exposure_residual_example = pd.concat(example_rows, ignore_index=True)
print("\n  worked example (0.25-day half-life, typical schedule of the fixed pool): RMSE of the "
      "incorrect model " + ", ".join(f"{w} days {RESULTS[f'exposure_example_rmse_{w}']:.4f}" for w in (28, 84)))
save_record(exposure_residual_example, "exposure_residual_example")
```

## 13 · External cohorts: between-animal variation and measurement error

The source study tested each model–treatment pair in one animal. Some arms have replicate readings on the same day (Section 2.2), but no arm has a second animal, so between-animal variation cannot be estimated from it. Two external cohorts with several animals per arm are used instead. Each is analyzed only when its file is present in `data/` and is skipped with a printed line otherwise.

**13.1 PDXNet** (public export from the PDXNet Portal; caliper volumes for nine PDX models from one contributing center). An animal is identified by model, arm, animal ID and `mouse.id`. Without `mouse.id`, two animals can share an ID within one arm, and the cell checks that the chosen key leaves no animal with two readings on one day. The untreated arm is marked `Control` in `Arms`, and the cell checks that `Type` and `Dose.Level` agree with it.

For each animal with at least four readings over at least five days, a line is fitted to log volume. The slope is the growth rate and its standard error comes from the residuals. In each model–arm group of at least three animals, the between-animal SD of the growth rate is corrected for slope error by subtracting the group's mean squared standard error from the variance of the slopes, with a floor at zero.

The primary comparison uses untreated groups only, because the source-study reference is the between-model SD of untreated growth rates (Section 3.6). The SD over all groups, treated included, is printed beside it. Per-animal growth rates are whole-series slopes, comparable with untreated growth rates and not with post-nadir regrowth. PDXNet series are shorter than the source study's untreated arms.

**13.2 Measurement error without a growth model.** Within each series, each interior reading is predicted by linear interpolation between its two neighbors. With locally linear growth, the residual divided by √(1 + w₀² + w₂²) estimates the measurement SD, where w₀ and w₂ are the interpolation weights. Untreated and treated series are analyzed separately, because the curvature of regression and regrowth inflates the estimate in treated series. Intervals are 95% bootstrap intervals that resample whole series. Each estimate is also given for triples spanning 6 to 10 days, the span of the Oberg cohort's 4-day schedule.

**Oberg 2021** (provided by the authors on request; ultrasound-derived tumor area for five ovarian PDX models, four arms each). Readings below the limit of detection (0.1) are excluded from the noise estimate. The cohort measures area, not volume, so its log-scale error is on a different quantity. If volume scales with area to the power 3/2, the same error on the volume scale would be 1.5 times larger.

**Inputs:** `observations`, `untreated_growth`, `data/pdxnet/PDXNet-PDTC-Export-*.csv`, `data/oberg2021/supp2_data.csv`.
**Outputs:** `pdxnet_animals`, `pdxnet_groups`, `oberg`, `external_noise` (table `external_measurement_error.csv`).

```python
# ==========================================================================
# 13.1 Between-animal variation in growth rate (PDXNet)
# ==========================================================================

set_section("13")
PDXNET_SERIES_KEY = ["PDXNet_Model_ID", "Arms", "Animal_ID", "mouse.id"]
MIN_POINTS_PER_ANIMAL = 4
MIN_SPAN_PER_ANIMAL_DAYS = 5
MIN_ANIMALS_PER_GROUP = 3

pdxe_between_model_sd = record("pdxe_between_model_sd", float(untreated_growth.std()))
pdxnet_files = sorted(PDXNET_DIR.glob("PDXNet-PDTC-Export-*.csv"))
PDXNET_AVAILABLE = bool(pdxnet_files)
print("13.1 BETWEEN-ANIMAL VARIATION IN GROWTH RATE (PDXNet)")
pdxnet_animals, pdxnet_groups, between_animal = pd.DataFrame(), pd.DataFrame(), pd.DataFrame()
if not PDXNET_AVAILABLE:
    print(f"  skipped: no PDXNet-PDTC-Export-*.csv in {relative(PDXNET_DIR)}")
else:
    pdxnet = pd.read_csv(pdxnet_files[-1])
    print(f"  file {relative(pdxnet_files[-1])}; {len(pdxnet):,} rows; "
          f"{pdxnet.PDXNet_Model_ID.nunique()} PDX models; agents "
          f"{', '.join(sorted(pdxnet.AgentName.dropna().astype(str).unique()))}; "
          f"contributors {', '.join(sorted(pdxnet.Contributor.dropna().astype(str).unique()))}")

    # The animal key: the candidate keys, with the number of (series, day)
    # pairs that hold more than one reading. The chosen key must leave none.
    key_rows = []
    for key in (["PDXNet_Model_ID", "Animal_ID"], ["PDXNet_Model_ID", "Arms", "Animal_ID"], PDXNET_SERIES_KEY):
        labels = pdxnet[key].astype(str).agg("|".join, axis=1)
        key_rows.append({"key": " + ".join(key), "series": labels.nunique(),
                         "duplicated_series_days": int(pd.DataFrame({"s": labels, "d": pdxnet.OBS_DAY})
                                                       .duplicated().sum())})
    print("\n  animal key")
    print(pd.DataFrame(key_rows).to_string(index=False))
    pdxnet["series"] = pdxnet[PDXNET_SERIES_KEY].astype(str).agg("|".join, axis=1)
    assert not pdxnet.duplicated(["series", "OBS_DAY"]).any(), "an animal has two readings on one day"
    pdxnet["is_control"] = pdxnet.Arms.astype(str).eq("Control")
    assert pdxnet.is_control.equals(pdxnet.Type.astype(str).eq("Control")), "Arms and Type disagree"
    assert pdxnet.is_control.equals(pdxnet["Dose.Level"].eq(0)), "Arms and Dose.Level disagree"
    nonpositive = int((pdxnet.TUMOR_VOL <= 0).sum())
    print(f"  readings at or below 0 mm³ (left out)  {nonpositive}")
    for column in ("Dose.Level", "Unit", "Route", "Frequency"):
        print(f"  {column:<11} recorded for {100 * pdxnet[column].notna().mean():.0f}% of rows: "
              f"{', '.join(map(str, pdxnet[column].dropna().unique()[:4]))}")

    animal_rows = []
    for series, block in pdxnet.loc[pdxnet.TUMOR_VOL > 0].sort_values("OBS_DAY").groupby("series", sort=True):
        day = block.OBS_DAY.to_numpy(float)
        if len(day) < MIN_POINTS_PER_ANIMAL or np.ptp(day) < MIN_SPAN_PER_ANIMAL_DAYS:
            continue
        slope, _, residual_sd = fit_line(day, np.log(block.TUMOR_VOL.to_numpy(float)))
        animal_rows.append({"series": series, "model_id": block.PDXNet_Model_ID.iloc[0],
                            "arm": block.Arms.iloc[0], "is_control": bool(block.is_control.iloc[0]),
                            "n_points": len(day), "span_days": float(np.ptp(day)), "growth_rate": slope,
                            "residual_sd": residual_sd,
                            "slope_se": residual_sd / np.sqrt(np.sum((day - day.mean()) ** 2))})
    pdxnet_animals = pd.DataFrame(animal_rows)
    pdxnet_groups = pdxnet_animals.groupby(["model_id", "arm"]).agg(
        is_control=("is_control", "first"), animals=("series", "size"),
        mean_growth_rate=("growth_rate", "mean"), sd_raw=("growth_rate", "std"),
        mean_slope_variance=("slope_se", lambda se: float(np.mean(se ** 2)))).reset_index()
    pdxnet_groups = pdxnet_groups.loc[pdxnet_groups.animals >= MIN_ANIMALS_PER_GROUP].copy()
    pdxnet_groups["sd_corrected"] = np.sqrt(np.maximum(
        pdxnet_groups.sd_raw ** 2 - pdxnet_groups.mean_slope_variance, 0.0))

    # Median over groups, with a 95% interval that resamples groups.
    external_rng = np.random.default_rng(SEEDS["external_cohort"])
    between_rows = []
    for label, groups in (("untreated groups", pdxnet_groups.loc[pdxnet_groups.is_control]),
                          ("all groups", pdxnet_groups)):
        low, high = bootstrap_median_interval(groups.sd_corrected.to_numpy(), N_BOOTSTRAP, external_rng)
        median_sd = float(groups.sd_corrected.median())
        between_rows.append({"groups": label, "n_groups": len(groups), "pdx_models": groups.model_id.nunique(),
                             "animals": int(groups.animals.sum()), "median_animals": float(groups.animals.median()),
                             "median_sd_raw": float(groups.sd_raw.median()), "median_sd_corrected": median_sd,
                             "low": low, "high": high, "ratio_to_pdxe_between_model": median_sd / pdxe_between_model_sd,
                             "groups_at_zero": int((groups.sd_corrected == 0).sum())})
    between_animal = pd.DataFrame(between_rows)
    for row in between_animal.itertuples():
        tag = "control" if row.groups == "untreated groups" else "all"
        record(f"pdxnet_n_groups_{tag}", row.n_groups)
        record(f"pdxnet_median_animals_{tag}", row.median_animals)
        record(f"pdxnet_between_sd_{tag}", row.median_sd_corrected)
        record(f"pdxnet_between_sd_{tag}_low", row.low)
        record(f"pdxnet_between_sd_{tag}_high", row.high)
        record(f"pdxnet_between_ratio_{tag}", row.ratio_to_pdxe_between_model)
    record("pdxnet_n_models", pdxnet.PDXNet_Model_ID.nunique())
    record("pdxnet_n_animals", pdxnet.series.nunique())

    print(f"\n  animals {pdxnet.series.nunique()}; with a fitted growth rate {len(pdxnet_animals)} "
          f"(>= {MIN_POINTS_PER_ANIMAL} readings over >= {MIN_SPAN_PER_ANIMAL_DAYS} days); readings per animal, "
          f"median {pdxnet_animals.n_points.median():.0f}; span, median {pdxnet_animals.span_days.median():.0f} days")
    print(f"  model–arm groups {pdxnet_animals.groupby(['model_id', 'arm']).ngroups}, of which "
          f"{len(pdxnet_groups)} have >= {MIN_ANIMALS_PER_GROUP} animals")
    print("\n  between-animal SD of growth rate (per day), corrected for slope error; median over groups")
    print(between_animal.round(5).to_string(index=False))
    print(f"\n  source study: between-model SD of untreated growth rate {pdxe_between_model_sd:.5f} per day "
          f"({len(untreated_growth)} PDX models)")
    print("\n  per group")
    print(pdxnet_groups.round(5).to_string(index=False))
    save_record(pdxnet_animals, "pdxnet_animal_growth_rates")
    save_record(pdxnet_groups, "pdxnet_between_animal_groups")

# ==========================================================================
# 13.2 Measurement error from interpolation residuals
#     Series sets: source-study arms (one per arm, canonical table), PDXNet
#     animals, Oberg mice; untreated and treated separately, all triples and
#     triples spanning 6 to 10 days.
# ==========================================================================

OBERG_LOD = 0.1                                   # detection limit used in raw_lod
OBERG_COLUMNS = ["model", "treatment", "mouse", "initial", "day", "day_c", "raw", "raw_lod"]
MATCHED_SPAN_DAYS = (6.0, 10.0)

OBERG_AVAILABLE = OBERG_PATH.exists()
oberg = pd.DataFrame()
if OBERG_AVAILABLE:
    oberg = pd.read_csv(OBERG_PATH)
    oberg.columns = [column.strip().lower() for column in oberg.columns]
    missing = sorted(set(OBERG_COLUMNS) - set(oberg.columns))
    assert not missing, f"Oberg file lacks columns {missing}"
    oberg["series"] = oberg.model.astype(str) + "|" + oberg.treatment.astype(str) + "|" + oberg.mouse.astype(str)
    oberg["below_lod"] = oberg.raw < OBERG_LOD
    assert oberg.below_lod.equals(oberg.raw != oberg.raw_lod), "raw_lod is not raw floored at the LOD"
    oberg["log_area"] = np.log(oberg.raw_lod)
    oberg["is_control"] = oberg.treatment.eq("Control")
    per_mouse = oberg.groupby("series").agg(cell=("treatment", "first"), model=("model", "first"),
                                            readings=("day", "size"), last_day=("day", "max"))
    per_cell = per_mouse.groupby(["model", "cell"]).size()
    record("oberg_n_mice", len(per_mouse))
    record("oberg_n_models", oberg.model.nunique())
    print("13.2 OBERG 2021 DESIGN")
    print(f"  {relative(OBERG_PATH)}: {len(oberg):,} rows; {oberg.model.nunique()} PDX models x "
          f"{oberg.treatment.nunique()} arms; {len(per_mouse)} mice "
          f"(per model–arm cell: median {per_cell.median():.0f}, range {per_cell.min()} to {per_cell.max()})")
    print(f"  mice not followed to day {oberg.day.max():.0f}: {int((per_mouse.last_day < oberg.day.max()).sum())} "
          f"of {len(per_mouse)}")
    print(f"  readings below the detection limit: {int(oberg.below_lod.sum())} of {len(oberg)} "
          f"({100 * oberg.below_lod.mean():.1f}%), left out of the noise estimate")
else:
    print(f"13.2 Oberg 2021 skipped: {relative(OBERG_PATH)} not found (available from the authors on request)")


noise_sets = {
    ("Source study", "untreated"): series_frames(observations.loc[observations.is_control],
                                                  "arm_id", "day", "log_volume"),
    ("Source study", "treated"): series_frames(observations.loc[~observations.is_control],
                                                "arm_id", "day", "log_volume"),
}
if PDXNET_AVAILABLE:
    positive = pdxnet.loc[pdxnet.TUMOR_VOL > 0].assign(log_volume=lambda f: np.log(f.TUMOR_VOL))
    noise_sets[("PDXNet", "untreated")] = series_frames(positive.loc[positive.is_control],
                                                         "series", "OBS_DAY", "log_volume")
    noise_sets[("PDXNet", "treated")] = series_frames(positive.loc[~positive.is_control],
                                                       "series", "OBS_DAY", "log_volume")
if OBERG_AVAILABLE:
    detected = oberg.loc[~oberg.below_lod]
    noise_sets[("Oberg (log area)", "untreated")] = series_frames(detected.loc[detected.is_control],
                                                                   "series", "day", "log_area")
    noise_sets[("Oberg (log area)", "treated")] = series_frames(detected.loc[~detected.is_control],
                                                                 "series", "day", "log_area")

noise_rng = np.random.default_rng(SEEDS["external_noise_bootstrap"])
noise_rows = []
for (cohort, arm_type), frames in noise_sets.items():
    for span_label, span_range in (("all triples", None), ("span 6-10 days", MATCHED_SPAN_DAYS)):
        noise_estimate = summarize_interpolation_noise(interpolation_noise(frames, span_range), noise_rng)
        noise_rows.append({"cohort": cohort, "arm_type": arm_type, "triples": span_label,
                           "series_available": len(frames), **noise_estimate})
external_noise = pd.DataFrame(noise_rows)
for row in external_noise.loc[external_noise.triples == "all triples"].itertuples():
    tag = f"noise_interp_{row.cohort.split()[0].lower()}_{row.arm_type}"
    record(tag, row.sigma)
    record(f"{tag}_low", row.low)
    record(f"{tag}_high", row.high)
    record(f"{tag}_series", row.n_series)
print("\n13.2 MEASUREMENT ERROR FROM INTERPOLATION RESIDUALS (log scale)")
print(external_noise.round(4).to_string(index=False))
print(f"\n  residual-based measurement noise of the source study (Section 3.6): {MEASUREMENT_NOISE:.4f}")
save_record(external_noise, "external_noise")
```

## 14 · Trajectory shape and progression in patients

Tumor burden is the sum of the longest diameters of the target lesions (SLD) fixed at baseline. It comes from the comparator arms of two phase III NSCLC trials released through Project Data Sphere: A6181087 (NCT00457392; erlotinib plus placebo) and CA031 (NCT00540514; carboplatin plus paclitaxel). A third study, EFC10261 (NCT00532155), was screened and excluded before any outcome was computed, because its release contains no tumor-measurement domain; the cell prints its file listing. The data are available under a data use agreement and cannot be redistributed. If they are absent the section prints a skip line.

**Endpoint.** Progression follows RECIST 1.1: a sum at least 20% and at least 5 mm above the reference, or a new lesion.

- **Reference.** The reference is the smallest sum recorded before each assessment, starting at baseline. An assessment with fewer target lesions measured than at baseline cannot set the reference, but it can show progression.
- **Lesion values.** A lesion recorded as 0 mm counts as 0 mm. A lesion recorded as too small to measure counts as 5 mm.
- **Clock.** Time to progression is measured from a landmark at the second scheduled tumor assessment: day 112 in A6181087 (assessments every 8 weeks) and day 84 in CA031 (every 6 weeks). Follow-up is censored at the last assessment.
- **Cohort.** The landmark cohort contains patients assessed on or after the landmark, free of progression (including new lesions) before it, and with at least two complete assessments before it.
- **Predictors.** The predictors come from complete assessments before the landmark: depth (1 − smallest sum / baseline sum), the slope of log SLD over the final interval, and an indicator that SLD rose over that interval.
- **Models.** The comparator Cox model contains depth alone. The trajectory model adds the two shape predictors. Concordance (Harrell's C) is computed out of fold, with five folds stratified on event status (and on study when pooled), and the difference is tested with a paired bootstrap of patients (2,000 resamples, within study when pooled).

Concordance here ranks patients by prognosis. It is not the within-tumor ranking of treatments used for the PDX models, and the treatment-selection endpoint has no counterpart in single-arm data.

**Two plans, both fixed and checksummed before their analyses.** The single-trial plan (A6181087) names the primary analysis and four sensitivity analyses: lowest reading removed, nadir clock, sum only, and baseline-adjusted. Holm correction runs across those five. The pooled plan was written after the single-trial primary analysis had been run and found no difference. CA031's file structure was known when it was written, but its panel had not been built and no outcome had been computed. It names the pooled primary analysis and five secondary analyses: leave-one-study-out, each study separately, sum only, baseline-adjusted and nadir clock. Each plan is loaded from `prespecification/` and checked against its recorded checksum.

**Departures from the plans and from the analysis first run, all of which change results:**

- **RECIST reference.** The analysis first run took the smallest sum from the clock start onward, including assessments after the one being scored. It also ignored new lesions before the landmark when forming the landmark cohort. Here the reference is the running minimum, and patients who had progressed before the landmark are excluded.
- **Lesions recorded as 0 mm** were set to 5 mm in the analysis first run. RECIST 1.1 records them as 0 mm.
- **Incomplete assessments.** In the analysis first run, incomplete assessments could set the nadir or the reference. Here they cannot, and baseline measurements spread over several days before randomization are combined into one baseline assessment.
- **Nadir clock.** Its predictors now come from assessments on or before the nadir. In the analysis first run they came from the landmark, after the clock had started.
- **Sum only.** In both families, follow-up is censored on the day a new lesion appears.
- **Pooled concordance.** The pooled Cox model has a baseline hazard per study. That choice was made at fit time and is not in the plan text. Concordance is therefore computed within study and combined; the analysis first run compared patients across trials. The unstratified pooled comparison was added after the primary analysis and is reported as added.
- **Pooled Holm family.** It has seven members: the primary analysis, leave-one-study-out, each study separately (two tests), sum only, baseline-adjusted and nadir clock. The plan lists the per-study analysis once, so counting each study separately makes the correction stricter. The plan's bootstrap resamples patients within study. Its success rule says "clustered on study", which cannot be done with two studies.
- **New lesions** are taken from an explicit flag in A6181087 and from the first appearance of a non-target lesion at a post-baseline visit in CA031. A sensitivity analysis applies first appearance in both studies (added).
- **CA031 data.** CA031 records the per-assessment sum in centimeters (converted to mm), without a per-lesion value for lesions too small to measure. A scheduled assessment's day is its scheduled week × 7, so the WEEK12 assessment falls on the day-84 landmark named in the plan. An unscheduled assessment takes its recorded scan day. The agreement between recorded and scheduled days is printed.
- **Lymph nodes.** Only the longest diameter is recorded, so the RECIST short-axis rule for lymph nodes cannot be applied. The number of target lesions described as nodes is printed.

CA031's investigator-recorded progression day is not used in any analysis family. It is compared with progression derived from baseline as a check on this implementation of RECIST.

**Diagnostics.** Two diagnostics are printed with the results. First, a patient with only a baseline and one later complete assessment before the landmark has a final-interval slope fixed by depth and the interval length, so the share of such patients and the variance inflation factor of each trajectory predictor are given for each study; where the shape predictors are nearly collinear with depth, the comparison cannot separate them. Second, the number of landmark-cohort patients whose outcome changes under the harmonized new-lesion definition is printed, so that a sensitivity analysis that changes no patient is identified as such.

**Inputs:** `data/project_data_sphere/A6181087/` (`tmm_p`, `random`, `final`), `data/project_data_sphere/CA031/` (`lesn`, `enrl`, `eos`), `data/project_data_sphere/EFC10261/`, `prespecification/clinical_single_trial_plan.json`, `prespecification/clinical_pooled_plan.json`.
**Outputs:** `clinical_panel`, `clinical_scored`, `clinical_results` (table `clinical_results.csv`), `clinical_coefficients`, `progression_check`; figure data `clinical_panel_restricted.csv` (patient-level, derived from restricted data: keep it local and never redistribute it).

```python
# ==========================================================================
# 14.1 Plans, studies and tumor-burden panels
#     Both plans are loaded and checked against their recorded checksums
#     whether or not the trial data are present. The panels hold one row
#     per patient assessment: day since randomization, SLD in mm, and
#     whether every baseline target lesion was measured.
# ==========================================================================

set_section("14")
CLINICAL_STUDIES = {
    "A6181087": {"nct": "NCT00457392", "comparator_arm": "erlotinib plus placebo",
                 "assessment_interval_days": 56,
                 "files": ["tmm_p.sas7bdat", "random.sas7bdat"]},
    "CA031": {"nct": "NCT00540514", "comparator_arm": "carboplatin plus paclitaxel",
              "assessment_interval_days": 42,
              "files": ["lesn.sas7bdat", "enrl.sas7bdat", "eos.sas7bdat"]},
}
TOO_SMALL_TO_MEASURE_MM = 5.0      # RECIST 1.1 default for a lesion too small to measure
BASELINE_WINDOW_DAY = 0            # A6181087: lesions measured on or before day 0 form the baseline
CM_TO_MM = 10.0                    # CA031 records sums in centimeters
NODE_PATTERN = r"LYMPH|NODE|NODAL"

single_plan, single_plan_sha256 = load_prespecified_plan("clinical_single_trial")
pooled_plan, pooled_plan_sha256 = load_prespecified_plan("clinical_pooled")
assert pooled_plan["supersedes"] == single_plan_sha256, "the pooled plan does not name the single-trial plan"
assert single_plan["seed"] == SEEDS["clinical_single_trial"] and pooled_plan["seed"] == SEEDS["clinical_pooled"]
assert pooled_plan["comparator_features"] == single_plan["comparator_features"]
assert pooled_plan["trajectory_features"] == single_plan["trajectory_features"]
LANDMARK_DAYS = {"A6181087": float(pooled_plan["landmark_by_study"]["A6181087"]["landmark_day"]),
                 "CA031": float(pooled_plan["landmark_by_study"]["CA031"]["nominal_day"])}
assert LANDMARK_DAYS["A6181087"] == float(single_plan["landmark_day"])
CLINICAL_COMPARATOR = list(single_plan["comparator_features"])
CLINICAL_TRAJECTORY = list(single_plan["trajectory_features"])

print("14.1 CLINICAL ANALYSIS PLANS")
for label, plan, digest in (("single-trial (A6181087)", single_plan, single_plan_sha256),
                            ("pooled", pooled_plan, pooled_plan_sha256)):
    print(f"  {label:<24} SHA-256 {digest[:16]}  seed {plan['seed']}")
print(f"  comparator {CLINICAL_COMPARATOR}; trajectory {CLINICAL_TRAJECTORY}")
print(f"  landmarks (second scheduled assessment): {LANDMARK_DAYS}")
print(f"  single-trial sensitivity analyses: {'; '.join(single_plan['sensitivities'])}")
print(f"  pooled secondary analyses: {'; '.join(pooled_plan['secondary_analyses'])}")

CLINICAL_AVAILABLE = all((PDS_DIR / study / name).exists()
                         for study, spec in CLINICAL_STUDIES.items() for name in spec["files"])
efc_folder = PDS_DIR / "EFC10261"
if efc_folder.exists():
    efc_tables = sorted(path.stem.upper() for path in efc_folder.glob("*.sas7bdat"))
    print(f"\n  EFC10261 (NCT00532155) tables: {', '.join(efc_tables)}")
    assert not {"TR", "TU"} & set(efc_tables), "EFC10261 has a tumor domain; its exclusion must be revisited"
    print("  no TR (tumor results) or TU (tumor identification) domain, so no lesion measurements: excluded")

clinical_panel, new_lesion_days, investigator_progression = pd.DataFrame(), {}, pd.Series(dtype=float)
if not CLINICAL_AVAILABLE:
    print(f"\n  Section 14 skipped: the Project Data Sphere files are not in {relative(PDS_DIR)} "
          "(available under a data use agreement)")
else:
    # --- A6181087: per-lesion records ------------------------------------
    study_folder = PDS_DIR / "A6181087"
    tmm = read_sas_table(study_folder / "tmm_p.sas7bdat")
    randomization = read_sas_table(study_folder / "random.sas7bdat").set_index("PID_A").RANDDAY
    tmm["day"] = tmm.EFDAY - tmm.PID_A.map(randomization)
    unanchored = int(tmm.day.isna().sum())
    tmm = tmm.loc[tmm.day.notna()]
    targets = tmm.loc[tmm.LESTYPE.eq("Target")].copy()
    wrong_unit = targets.loc[targets.TMMDIA.notna() & targets.TMMDIAU.ne("MM"), "TMMDIAU"]
    assert wrong_unit.empty, f"diameters not in mm: {wrong_unit.value_counts().to_dict()}"
    too_small = targets.TMMNMES.eq("TOO SMALL TO MEASURE")
    targets.loc[too_small, "TMMDIA"] = TOO_SMALL_TO_MEASURE_MM
    zero_mm = int(targets.TMMDIA.eq(0).sum())                    # kept at 0 mm
    targets["baseline_lesion"] = (targets.groupby(["PID_A", "TMMLNM"]).day.transform("min")
                                  <= BASELINE_WINDOW_DAY)
    measured = targets.loc[targets.baseline_lesion & targets.TMMDIA.notna()].copy()
    # Baseline measurements spread over several days before randomization
    # form one baseline assessment (each lesion's latest baseline value).
    in_window = measured.day <= BASELINE_WINDOW_DAY
    multi_day_baseline = int(measured.loc[in_window].groupby("PID_A").day.nunique().gt(1).sum())
    measured["recorded_day"] = measured.day
    measured.loc[in_window, "day"] = 0.0
    duplicated_readings = int(measured.duplicated(["PID_A", "TMMLNM", "day"]).sum())
    measured = (measured.sort_values("recorded_day")
                .drop_duplicates(["PID_A", "TMMLNM", "day"], keep="last"))
    baseline_lesions = measured.loc[measured.day.eq(0)].groupby("PID_A").TMMLNM.nunique()
    panel_a = measured.groupby(["PID_A", "day"], as_index=False).agg(
        sld=("TMMDIA", "sum"), n_lesions=("TMMLNM", "nunique"))
    panel_a["baseline_lesions"] = panel_a.PID_A.map(baseline_lesions)
    panel_a["complete"] = panel_a.n_lesions.eq(panel_a.baseline_lesions)
    panel_a["patient"] = "A6181087:" + panel_a.PID_A.astype(str)
    panel_a["study"], panel_a["day_source"] = "A6181087", "recorded day"
    text_columns = targets.select_dtypes(include=["object", "string"]).columns
    node_lesions_a = targets.loc[targets[text_columns].apply(
        lambda column: column.astype(str).str.contains(NODE_PATTERN, case=False)).any(axis=1)
    ].groupby(["PID_A", "TMMLNM"]).ngroups

    # New lesions: the explicit flag (primary) and the first appearance of a
    # non-target lesion after baseline (the CA031 definition; sensitivity).
    flagged = tmm.loc[tmm.TMMNLES.eq("NEW LESION")].groupby("PID_A").day.min()
    appearance = tmm.loc[tmm.LESTYPE.ne("Target")].groupby(["PID_A", "TMMLNM"]).day.min()
    appeared = appearance.loc[appearance > BASELINE_WINDOW_DAY].groupby(level="PID_A").min()
    new_lesion_days["A6181087"] = flagged.rename(lambda pid: f"A6181087:{pid}")
    new_lesion_days["A6181087 first appearance"] = appeared.rename(lambda pid: f"A6181087:{pid}")

    # --- CA031: per-assessment sums on a ten-slot form ---------------------
    study_folder = PDS_DIR / "CA031"
    lesn = read_sas_table(study_folder / "lesn.sas7bdat")
    randomization_ca = read_sas_table(study_folder / "enrl.sas7bdat").set_index("RPT").RANDDAY
    eos = read_sas_table(study_folder / "eos.sas7bdat").set_index("RPT")
    lesn = lesn.loc[lesn.RPT.isin(randomization_ca.index)].copy()
    lesn["observed_day"] = lesn.SCANDAY - lesn.RPT.map(randomization_ca)
    week = pd.to_numeric(lesn.CPEVENT.str.extract(r"^WEEK(\d+)$", expand=False), errors="coerce")
    lesn["nominal_day"] = np.where(lesn.CPEVENT.eq("BASELINE"), 0.0, week * 7.0)
    targets_ca = lesn.loc[lesn.LESTYP.eq("TARG")]
    assert targets_ca.groupby(["RPT", "CPEVENT"]).SUMDIAM.nunique().le(1).all(), \
        "SUMDIAM differs between rows of one assessment"
    day_agreement = (targets_ca.observed_day - targets_ca.nominal_day).abs().dropna()
    ca031_assessments = targets_ca.groupby(["RPT", "CPEVENT"]).agg(
        sum_cm=("SUMDIAM", "first"), observed_day=("observed_day", "min"),
        nominal_day=("nominal_day", "first")).reset_index()
    ca031_assessments = ca031_assessments.merge(targets_ca.loc[targets_ca.DIAM.notna()].groupby(["RPT", "CPEVENT"]).LESNUM
                          .nunique().rename("n_lesions").reset_index(), how="left").fillna({"n_lesions": 0})
    # Scheduled visits keep their scheduled day (week × 7), so the WEEK12 assessment
    # sits at the day-84 landmark the pooled plan names; unscheduled visits use the
    # recorded scan day.
    ca031_assessments["day"] = ca031_assessments.nominal_day.fillna(ca031_assessments.observed_day)
    ca031_assessments["day_source"] = np.where(ca031_assessments.nominal_day.notna(), "scheduled week",
                                               "recorded day")
    unplaced = int((ca031_assessments.day.isna() | ca031_assessments.sum_cm.isna()).sum())
    ca031_assessments = ca031_assessments.loc[ca031_assessments.day.notna() & ca031_assessments.sum_cm.notna()].copy()
    ca031_assessments["sld"] = ca031_assessments.sum_cm * CM_TO_MM
    ca031_assessments["baseline_lesions"] = ca031_assessments.RPT.map(ca031_assessments.loc[ca031_assessments.CPEVENT.eq("BASELINE")]
                                                .set_index("RPT").n_lesions)
    ca031_assessments["complete"] = ca031_assessments.n_lesions.eq(ca031_assessments.baseline_lesions)
    ca031_assessments["patient"] = "CA031:" + ca031_assessments.RPT.astype(str)
    ca031_assessments["study"] = "CA031"
    node_lesions_ca = targets_ca.loc[targets_ca.LESDESC.str.contains(NODE_PATTERN, case=False, na=False)
                                     ].groupby(["RPT", "LESDESC"]).ngroups

    # New lesions: a non-target lesion first recorded at a post-baseline visit.
    nontarget_ca = lesn.loc[lesn.LESTYP.eq("NTLE") & lesn.LESDESC.fillna("").ne("")].copy()
    nontarget_ca["day"] = nontarget_ca.nominal_day.fillna(nontarget_ca.observed_day)
    by_lesion = nontarget_ca.groupby(["RPT", "LESDESC"]).agg(
        at_baseline=("CPEVENT", lambda visit: bool(visit.eq("BASELINE").any())), first_day=("day", "min"))
    appeared_ca = by_lesion.loc[~by_lesion.at_baseline].first_day.dropna().groupby(level="RPT").min()
    new_lesion_days["CA031"] = appeared_ca.rename(lambda rpt: f"CA031:{rpt}")
    investigator_progression = (eos.PROGDAY - eos.index.map(randomization_ca)).dropna().rename(
        lambda rpt: f"CA031:{rpt}")

    columns = ["study", "patient", "day", "sld", "n_lesions", "baseline_lesions", "complete", "day_source"]
    clinical_panel = pd.concat([panel_a[columns], ca031_assessments[columns]], ignore_index=True)
    panel_rows = []
    for study, panel in clinical_panel.groupby("study"):
        per_patient = panel.groupby("patient")
        spacing = panel.sort_values(["patient", "day"]).groupby("patient").day.diff()
        new_days = new_lesion_days[study]
        panel_rows.append({
            "study": study, "patients": panel.patient.nunique(), "assessments": len(panel),
            "median_assessments": float(per_patient.size().median()),
            "median_spacing_days": float(spacing.median()),
            "complete_share": float(panel.complete.mean()),
            "median_baseline_sld_mm": float(panel.loc[panel.day <= 0].groupby("patient").sld.first().median()),
            "patients_with_new_lesion": int(new_days.index.isin(panel.patient).sum()),
            "recorded_day_share": float(panel.day_source.eq("recorded day").mean())})
    panel_summary = pd.DataFrame(panel_rows)
    print("\n  TUMOR-BURDEN PANELS")
    print(panel_summary.round(3).to_string(index=False))
    print(f"\n  A6181087: lesion rows without a randomization day {unanchored}; target readings too small "
          f"to measure {int(too_small.sum())} (set to {TOO_SMALL_TO_MEASURE_MM:g} mm); at 0 mm {zero_mm} "
          f"(kept at 0); patients with baseline readings on more than one day {multi_day_baseline} "
          f"(combined); repeated lesion readings on one day {duplicated_readings} (latest kept); "
          f"target lesions described as lymph nodes {node_lesions_a}")
    print(f"  A6181087 new lesions: flagged in {len(flagged)} patients; first appearance of a non-target "
          f"lesion after baseline in {len(appeared)}")
    print(f"  CA031: |recorded − scheduled day| median {day_agreement.median():.0f} days (90th percentile "
          f"{day_agreement.quantile(0.9):.0f}) on {len(day_agreement):,} target rows with both; assessments "
          f"without a day or sum {unplaced}; with no recorded lesion diameter "
          f"{int(ca031_assessments.n_lesions.eq(0).sum())}; target lesions described as lymph nodes {node_lesions_ca}")
    print(f"  CA031: investigator progression day recorded for {len(investigator_progression)} of "
          f"{len(eos)} patients")
    save_record(panel_summary, "clinical_panels")
    # Patient-level assessments, for the clinical figure. Derived from Project
    # Data Sphere data, which may not be redistributed: keep this file local.
    save_record(clinical_panel, "clinical_panel_restricted")
```

```python
# ==========================================================================
# 14.2 Landmark cohorts, and derived progression against the investigator's
#      progression day (CA031)
# ==========================================================================

clinical_scored, clinical_cohort, nadir_cohort = pd.DataFrame(), pd.DataFrame(), pd.DataFrame()
progression_check = pd.DataFrame()
clinical_rows, clinical_coefficients = [], []
if not CLINICAL_AVAILABLE:
    print("14.2 skipped: Project Data Sphere files not present")
else:
    clinical_scored = pd.concat([
        score_patients(clinical_panel.loc[clinical_panel.study == "A6181087"], new_lesion_days["A6181087"],
                       LANDMARK_DAYS["A6181087"], "A6181087", new_lesion_days["A6181087 first appearance"]),
        score_patients(clinical_panel.loc[clinical_panel.study == "CA031"], new_lesion_days["CA031"],
                       LANDMARK_DAYS["CA031"], "CA031", new_lesion_days["CA031"]),
    ], ignore_index=True)
    in_cohort = (clinical_scored.reached_landmark & ~clinical_scored.progressed_before_landmark
                 & clinical_scored.has_features)
    clinical_cohort = clinical_scored.loc[in_cohort].reset_index(drop=True)
    assert (clinical_cohort.last_feature_day < clinical_cohort.study.map(LANDMARK_DAYS)).all(), \
        "a predictor used an assessment on or after the landmark"

    # Nadir clock: the same patients, the clock started at the lowest
    # complete sum and the predictors taken on or before it.
    nadir_cohort = clinical_cohort.dropna(subset=["nadir_depth_at_landmark", "nadir_duration"]).copy()
    for name in ("depth_at_landmark", "slope_to_landmark", "already_rising", "log_baseline_sld"):
        nadir_cohort[name] = nadir_cohort[f"nadir_{name}"]
    assert (nadir_cohort.nadir_last_feature_day <= nadir_cohort.nadir_day).all()

    clinical_attrition = clinical_scored.groupby("study").agg(
        patients=("patient", "size"), assessed_at_or_after_landmark=("reached_landmark", "sum"))
    clinical_attrition["progressed_before_landmark"] = clinical_scored.loc[clinical_scored.reached_landmark].groupby(
        "study").progressed_before_landmark.sum()
    clinical_attrition["landmark_cohort"] = clinical_cohort.groupby("study").size()
    clinical_attrition["events"] = clinical_cohort.groupby("study").event.sum()
    clinical_attrition["events_new_lesion"] = clinical_cohort.loc[clinical_cohort.source.eq("new lesion")].groupby(
        "study").size()
    clinical_attrition["events_sum_only"] = clinical_cohort.groupby("study").sld_only_event.sum()
    clinical_attrition["nadir_clock_patients"] = nadir_cohort.groupby("study").size()
    clinical_attrition["median_slope_window_days"] = clinical_cohort.groupby("study").slope_window_days.median()
    clinical_attrition = clinical_attrition.fillna(0)
    print("14.2 LANDMARK COHORTS")
    print(clinical_attrition.to_string())
    print(f"  pooled: {len(clinical_cohort)} patients, {int(clinical_cohort.event.sum())} events")
    print("\n  predictors in the landmark cohort")
    print(clinical_cohort.groupby("study")[CLINICAL_TRAJECTORY + ["log_baseline_sld"]].describe()
          .T.round(4).to_string())
    print("\n  Spearman correlation of each predictor with depth, by study")
    for study, part in clinical_cohort.groupby("study"):
        rhos = {name: spearman(part.depth_at_landmark, part[name])[0]
                for name in ("slope_to_landmark", "already_rising", "log_baseline_sld")}
        print(f"    {study:<9} " + "   ".join(f"{name} {value:+.3f}" for name, value in rhos.items()))

    # Collinearity. A patient with only a baseline and one later complete
    # assessment before the landmark has a final-interval slope that is a
    # function of depth and of the interval length alone (log(1 − depth) /
    # interval when the sum fell), so the two shape predictors add little that
    # depth does not carry, and the Cox coefficients of depth and slope become
    # unstable. The share of such patients and the variance inflation factors
    # are printed for each study.
    collinearity_rows = []
    for study, part in clinical_cohort.groupby("study"):
        factors = variance_inflation(part, CLINICAL_TRAJECTORY)
        share_two = float((part.n_assessments == 2).mean())
        collinearity_rows.append({"study": study, "patients": len(part),
                                  "share_with_two_assessments": share_two,
                                  **{f"vif_{name}": value for name, value in factors.items()}})
        record(f"clinical_{study}_share_two_assessments", share_two)
        record(f"clinical_{study}_max_vif", float(np.nanmax(list(factors.values()))))
    clinical_collinearity = pd.DataFrame(collinearity_rows)
    print("\n  collinearity of the trajectory predictors (share of patients with only two complete "
          "assessments before the landmark; variance inflation factors)")
    print(clinical_collinearity.round(3).to_string(index=False))

    # RECIST progression derived from baseline in CA031, against the day the
    # investigator recorded. A check on the implementation, not a test.
    check_rows = []
    ca_panel = clinical_panel.loc[clinical_panel.study == "CA031"].sort_values("day")
    for patient, block in ca_panel.groupby("patient", sort=True):
        if len(block) < 2:
            continue
        derived_outcome = time_to_progression(block, float(block.day.iloc[0]),
                                      new_lesion_days["CA031"].get(patient, np.nan))
        check_rows.append({"patient": patient, "derived": derived_outcome["event"],
                           "derived_day": float(block.day.iloc[0]) + derived_outcome["duration"] if derived_outcome["event"] else np.nan,
                           "investigator": patient in investigator_progression.index,
                           "investigator_day": investigator_progression.get(patient, np.nan)})
    progression_check = pd.DataFrame(check_rows)
    progressed_both = progression_check.loc[progression_check.derived & progression_check.investigator]
    day_gap = progressed_both.derived_day - progressed_both.investigator_day
    interval = CLINICAL_STUDIES["CA031"]["assessment_interval_days"]
    print("\n  CA031: RECIST progression derived from baseline against the investigator's progression day")
    print(pd.crosstab(progression_check.derived, progression_check.investigator,
                      rownames=["derived"], colnames=["investigator"]).to_string())
    print(f"  progressing by both: {len(progressed_both)}; derived minus investigator day, median {day_gap.median():+.0f}; "
          f"within one assessment interval ({interval} days) {100 * day_gap.abs().le(interval).mean():.1f}%")
    save_record(clinical_scored, "clinical_patients")

# ==========================================================================
# 14.3 The single-trial plan (A6181087)
#     Primary and four sensitivity analyses, Holm correction across the five.
#     Folds stratified on event status; patients resampled without strata.
# ==========================================================================

if CLINICAL_AVAILABLE:
    single = clinical_cohort.loc[clinical_cohort.study == "A6181087"]
    single_nadir = nadir_cohort.loc[nadir_cohort.study == "A6181087"]
    single_seed = single_plan["seed"]
    single_analyses = [
        ("primary", single, "duration", "event", CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY),
        ("lowest reading removed", single.dropna(subset=["nadir_excluded_duration"]),
         "nadir_excluded_duration", "nadir_excluded_event", CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY),
        ("nadir clock", single_nadir, "nadir_duration", "nadir_event", CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY),
        ("sum only", single, "sld_only_duration", "sld_only_event", CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY),
        ("baseline adjusted", single, "duration", "event",
         CLINICAL_COMPARATOR + ["log_baseline_sld"], CLINICAL_TRAJECTORY + ["log_baseline_sld"]),
    ]
    single_results = [compare_cox_models(frame, comparator, trajectory, single_seed, duration, event,
                                         label=name, min_events=5)
                      for name, frame, duration, event, comparator, trajectory in single_analyses]
    single_thresholds, single_rejected = holm([result["p_value"] for result in single_results])
    print("\n14.3 SINGLE-TRIAL PLAN (A6181087, landmark day "
          f"{LANDMARK_DAYS['A6181087']:.0f}; Holm across {len(single_results)})")
    for result, threshold, reject in zip(single_results, single_thresholds, single_rejected):
        report_comparison(result)
        print(f"    Holm threshold {threshold:.4f}; {'rejected' if reject else 'not rejected'}")
        clinical_rows.append({"family": "single-trial plan", "analysis": result["analysis"],
                              "status": "prespecified", **{k: result[k] for k in (
                                  "n", "events", "c_depth", "c_trajectory", "difference", "low", "high",
                                  "p_value", "bootstrap_draws")},
                              "holm_threshold": threshold, "holm_rejected": bool(reject)})
    single_coefficients = single_results[0]["coefficients"]
    print("\n  primary: trajectory-model coefficients by fold (log hazard per SD)")
    print(single_coefficients.round(4).to_string())
    print("  sign the same in every fold: " + ", ".join(
        f"{name} {'yes' if np.sign(single_coefficients[name].dropna()).nunique() == 1 else 'no'}"
        for name in CLINICAL_TRAJECTORY))
    print("\n  primary: concordance by fold (each fold's own patients)")
    print(single_results[0]["checks"].round(3).to_string(index=False))
    record("clinical_single_c_depth", single_results[0]["c_depth"])
    record("clinical_single_c_trajectory", single_results[0]["c_trajectory"])
    record("clinical_single_difference", single_results[0]["difference"])
    record("clinical_single_p", single_results[0]["p_value"])
```

```python
# ==========================================================================
# 14.4 The pooled plan (A6181087 and CA031)
#     Primary: Cox models with a baseline hazard per study, concordance
#     within study, folds stratified on event status and study, bootstrap
#     within study. Holm correction across seven tests. Then the analyses
#     added after the primary and the coefficient diagnostics.
# ==========================================================================

clinical_results = pd.DataFrame()
if not CLINICAL_AVAILABLE:
    print("14.4 skipped: Project Data Sphere files not present")
else:
    pooled_seed = pooled_plan["seed"]
    pooled_results = [
        compare_cox_models(clinical_cohort, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed, stratified=True,
                           label="primary (pooled, baseline hazard per study)", min_events=10),
        leave_one_study_out(clinical_cohort, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed,
                            label="leave-one-study-out", min_events=10),
    ]
    for study in CLINICAL_STUDIES:
        pooled_results.append(compare_cox_models(
            clinical_cohort.loc[clinical_cohort.study == study], CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed,
            label=f"{study} alone", min_events=10))
    pooled_results += [
        compare_cox_models(clinical_cohort, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed, "sld_only_duration",
                           "sld_only_event", stratified=True, label="sum only (censored at a new lesion)",
                           min_events=10),
        compare_cox_models(clinical_cohort, CLINICAL_COMPARATOR + ["log_baseline_sld"], CLINICAL_TRAJECTORY + ["log_baseline_sld"],
                           pooled_seed, stratified=True, label="baseline adjusted", min_events=10),
        compare_cox_models(nadir_cohort, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed, "nadir_duration", "nadir_event",
                           stratified=True, label="nadir clock", min_events=10),
    ]
    pooled_thresholds, pooled_rejected = holm([result["p_value"] for result in pooled_results])
    print("14.4 POOLED PLAN (Holm across "
          f"{len(pooled_results)}; plan lists the per-study analysis once)")
    for result, threshold, reject in zip(pooled_results, pooled_thresholds, pooled_rejected):
        report_comparison(result)
        print(f"    Holm threshold {threshold:.4f}; {'rejected' if reject else 'not rejected'}")
        clinical_rows.append({"family": "pooled plan", "analysis": result["analysis"], "status": "prespecified",
                              **{k: result[k] for k in ("n", "events", "c_depth", "c_trajectory", "difference",
                                                        "low", "high", "p_value", "bootstrap_draws")},
                              "holm_threshold": threshold, "holm_rejected": bool(reject)})
    primary, loso, sum_only = pooled_results[0], pooled_results[1], pooled_results[4]
    print("\n  leave-one-study-out, by held-out study (models fitted on the other study)")
    print(loso["checks"].round(3).to_string(index=False))
    print("\n  primary: concordance by fold and by study (within-study pairs)")
    print(primary["checks"].round(3).to_string(index=False))
    print("  The per-study rows reuse the pooled folds and fits, so they differ from the "
          "per-study analyses above,\n  which have their own folds.")

    # --- analyses added after the primary ----------------------------------
    added = [
        compare_cox_models(clinical_cohort, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed, stratified=False,
                           label="added: pooled, one baseline hazard, concordance across all pairs",
                           min_events=10),
    ]
    harmonized = clinical_scored.loc[clinical_scored.reached_landmark & clinical_scored.has_features
                                     & ~clinical_scored.harmonized_progressed_before_landmark.astype(bool)].copy()
    harmonized["duration"], harmonized["event"] = harmonized.harmonized_duration, harmonized.harmonized_event
    added.append(compare_cox_models(harmonized, CLINICAL_COMPARATOR, CLINICAL_TRAJECTORY, pooled_seed, stratified=True,
                                    label="added: new lesions by first appearance in both studies",
                                    min_events=10))
    # How many landmark-cohort patients the harmonized new-lesion definition
    # actually changes. If none, the added analysis repeats the primary by
    # construction and its identical result carries no extra information.
    primary_outcomes = clinical_cohort.set_index("patient")[["duration", "event"]]
    harmonized_outcomes = harmonized.set_index("patient")[["duration", "event"]]
    shared_patients = primary_outcomes.index.intersection(harmonized_outcomes.index)
    outcome_changed = ((primary_outcomes.loc[shared_patients, "event"].astype(bool)
                        != harmonized_outcomes.loc[shared_patients, "event"].astype(bool))
                       | ~np.isclose(primary_outcomes.loc[shared_patients, "duration"].to_numpy(float),
                                     harmonized_outcomes.loc[shared_patients, "duration"].to_numpy(float),
                                     equal_nan=True))
    cohort_difference = len(set(primary_outcomes.index) ^ set(harmonized_outcomes.index))
    record("clinical_harmonized_patients_changed", int(outcome_changed.sum()))
    record("clinical_harmonized_cohort_difference", cohort_difference)
    print(f"\n  harmonized new-lesion definition: patients entering or leaving the cohort {cohort_difference}; "
          f"patients whose time or event changes {int(outcome_changed.sum())} of {len(shared_patients)}")
    if cohort_difference == 0 and not outcome_changed.any():
        print("  The definition changes no patient in the landmark cohort, so the added analysis below "
              "reproduces the primary exactly.")

    print("\n  ADDED AFTER THE PRIMARY (outside the Holm family)")
    for result in added:
        report_comparison(result)
        clinical_rows.append({"family": "pooled plan", "analysis": result["analysis"], "status": "added",
                              **{k: result[k] for k in ("n", "events", "c_depth", "c_trajectory", "difference",
                                                        "low", "high", "p_value", "bootstrap_draws")}})

    # --- coefficient diagnostics -------------------------------------------
    # Fold-by-fold coefficients of the pooled models, and Cox models fitted
    # on all patients: pooled with a baseline hazard per study, and each
    # study alone. Coefficients are log hazard per SD of the predictor.
    print("\n  pooled primary: coefficients by fold (log hazard per SD)")
    print(pd.concat([primary["coefficients_depth"].add_suffix(" (depth model)"),
                     primary["coefficients"].add_suffix(" (trajectory model)")], axis=1).round(4).to_string())
    for label, frame, stratified in (("pooled, baseline hazard per study", clinical_cohort, True),
                                     *[(f"{study} alone", clinical_cohort.loc[clinical_cohort.study == study], False)
                                       for study in CLINICAL_STUDIES]):
        for model, predictors in (("depth", CLINICAL_COMPARATOR), ("trajectory", CLINICAL_TRAJECTORY)):
            full_fit = cox_full_fit(frame, predictors, stratified=stratified)
            clinical_coefficients.append(full_fit.assign(fit=label, model=model))
    clinical_coefficients = pd.concat(clinical_coefficients, ignore_index=True)[
        ["fit", "model", "predictor", "coefficient", "se", "low", "high", "p_value"]]
    print("\n  Cox models fitted on all patients")
    print(clinical_coefficients.round(4).to_string(index=False))

    clinical_results = pd.DataFrame(clinical_rows)
    record("clinical_n_trials", clinical_cohort.study.nunique())
    record("clinical_n_patients", primary["n"])
    record("clinical_n_events", primary["events"])
    record("clinical_c_depth", primary["c_depth"])
    record("clinical_c_trajectory", primary["c_trajectory"])
    record("clinical_difference", primary["difference"])
    record("clinical_difference_low", primary["low"])
    record("clinical_difference_high", primary["high"])
    record("clinical_p", primary["p_value"])
    record("clinical_unstratified_c_depth", added[0]["c_depth"])
    record("clinical_unstratified_c_trajectory", added[0]["c_trajectory"])
    record("clinical_sum_only_events", sum_only["events"])
    record("clinical_sum_only_c_depth", sum_only["c_depth"])
    record("clinical_sum_only_c_trajectory", sum_only["c_trajectory"])
    record("clinical_sum_only_difference", sum_only["difference"])
    record("clinical_sum_only_difference_low", sum_only["low"])
    record("clinical_sum_only_difference_high", sum_only["high"])
    record("clinical_sum_only_p", sum_only["p_value"])
    record("clinical_pooled_holm_rejected", int(clinical_results.loc[
        clinical_results.family.eq("pooled plan") & clinical_results.status.eq("prespecified"),
        "holm_rejected"].sum()))
    save_record(clinical_results, "clinical_results")
    save_record(clinical_coefficients, "clinical_coefficients")
```

## 15 · Results register, tables and run manifest

This section gathers what the run produced. Section 15.1 writes the main result tables to `outputs/tables/` and the results register (every value stored with `record()`, with the section that stored it) to `outputs/results_register.json` and `outputs/results_register.csv`. Optional inputs that were absent are marked as not computed. Section 15.2 checks the stored results for internal consistency: quantities derived from other stored quantities must agree with them, proportions and P values must lie between 0 and 1, and each endpoint decision must follow from its estimate, interval and margin. A failed check stops the run. Checks that are expected but not guaranteed (for example, that a percentile bootstrap interval contains its point estimate) are reported without stopping it. Section 15.3 writes `outputs/manifest.json`, which records the software versions, platform, seeds and key constants, the SHA-256 checksum and size of every input and output file, and the run time, and then prints a summary of the main results.

**Inputs:** every object stored above, `RESULTS`.
**Outputs:** `outputs/tables/*.csv`, `outputs/results_register.json`, `outputs/results_register.csv`, `outputs/manifest.json`.

```python
# ==========================================================================
# 15.1 Result tables and the results register
#     Each table is written to outputs/tables/ under a name that says what it
#     holds. Tables of optional inputs that were absent are empty and are
#     listed as not written. The register holds every value stored with
#     record(), the section that stored it, and whether each optional input
#     was present.
# ==========================================================================

set_section("15")
OPTIONAL_INPUTS = {
    "pdxnet_external_cohort": bool(PDXNET_AVAILABLE),
    "oberg_external_cohort": bool(OBERG_AVAILABLE),
    "hallmark_gene_sets": bool(HALLMARK_AVAILABLE),
    "clinical_trial_cohorts": bool(CLINICAL_AVAILABLE),
}
for name, available in OPTIONAL_INPUTS.items():
    record(f"computed_{name}", available)

RESULT_TABLES = {
    "predictor_sets": predictor_set_table,
    "depth_durability_sensitivity": depth_durability_sensitivity,
    "pair_threshold_sweep": pair_sweep,
    "parameter_recovery": parameter_recovery,
    "design_map": design_map,
    "mechanism_discrimination": mechanism_discrimination,
    "mechanism_discrimination_paired": paired_discrimination,
    "simulation_based_calibration": calibration.reset_index(),
    "simulation_based_calibration_converged": calibration_converged.reset_index(),
    "kill_rate_partition": kill_rate_partition,
    "forecasting_predictor_sets": predictor_set_performance,
    "forecasting_contrasts": predictor_contrasts,
    "forecasting_start_days_and_horizons": rolling_origins,
    "forecasting_held_out_tumor_types": tumor_type_holdout,
    "forecasting_censoring_scenarios": censoring_scenarios,
    "forecasting_untreated_and_treated": untreated_forecasting,
    "positive_control_braf": positive_control,
    "ranking_by_separation": separability,
    "baseline_slope_ranking": baseline_slope_selection,
    "gene_sets": gene_set_table,
    "molecular_blocks": molecular_blocks,
    "molecular_blocks_hallmark": hallmark_blocks,
    "hallmark_sets": hallmark_sets_table,
    "combination_summary": by_combination,
    "compound_families": compound_table,
    "mechanism_class_results": mechanism_results,
    "mechanism_class_mixed_models": mixed_models,
    "mechanism_class_by_family_pair": breakdown,
    "mechanism_class_leave_one_compound_out": leave_one_out,
    "exposure_separation_by_window": separation_by_window,
    "exposure_separation_growth": separation_growth,
    "exposure_verdicts_by_window": confirmation_summary,
    "exposure_thinning": thinning_verdicts,
    "external_between_animal": between_animal,
    "external_measurement_error": external_noise,
    "clinical_results": clinical_results,
    "clinical_coefficients": (clinical_coefficients if isinstance(clinical_coefficients, pd.DataFrame)
                              else pd.DataFrame()),
}
written_tables, empty_tables = [], []
for name, table in RESULT_TABLES.items():
    if isinstance(table, pd.DataFrame) and len(table):
        table.to_csv(TABLE_DIR / f"{name}.csv", index=False)
        written_tables.append(name)
    else:
        empty_tables.append(name)
print("15.1 RESULT TABLES")
print(f"  written to {relative(TABLE_DIR)}: {len(written_tables)}")
for name in written_tables:
    print(f"    {name}.csv   {len(RESULT_TABLES[name]):,} rows")
if empty_tables:
    print(f"  not written (input absent or nothing to report): {', '.join(empty_tables)}")

# --- the register --------------------------------------------------------------
register = pd.DataFrame({"key": list(RESULTS),
                         "section": [RESULT_SECTION.get(key, "") for key in RESULTS],
                         "value": [json_ready(RESULTS[key]) for key in RESULTS]})
REGISTER_JSON = OUTPUT_DIR / "results_register.json"
REGISTER_CSV = OUTPUT_DIR / "results_register.csv"
REGISTER_JSON.write_text(json.dumps(
    {"run_started_utc": RUN_STARTED_UTC,
     "optional_inputs": OPTIONAL_INPUTS,
     "values": {row.key: {"section": row.section, "value": row.value}
                for row in register.itertuples(index=False)}},
    indent=1, ensure_ascii=False), encoding="utf-8")
register.to_csv(REGISTER_CSV, index=False)
not_computed = [name for name, available in OPTIONAL_INPUTS.items() if not available]
print(f"\n  results register: {len(register):,} values from {register.section.nunique()} sections")
print(f"    {relative(REGISTER_JSON)} and {relative(REGISTER_CSV)}")
print(f"  not computed (optional input absent): {', '.join(not_computed) if not_computed else 'none'}")
print("\n  values per section")
print(register.groupby("section", sort=False).size().rename("values").to_string())
```

```python
# ==========================================================================
# 15.2 Internal consistency of the stored results
#     Required checks must hold: a quantity derived from other stored
#     quantities must equal what they give, proportions and P values must lie
#     in [0, 1], and each endpoint decision must follow from its estimate,
#     interval and margin. A failed required check stops the run. Expected
#     checks (a percentile interval containing its estimate) are reported
#     without stopping it, because a percentile interval of a ratio can, in
#     principle, exclude its point estimate. A check whose inputs were not
#     computed is marked not applicable.
# ==========================================================================

# (description, keys it needs, required?, condition on the register r)
DERIVED_CHECKS = [
    ("forecast reduction = (depth RMSE − trajectory RMSE) / depth RMSE",
     ["forecast_reduction", "depth_rmse", "trajectory_rmse"], True,
     lambda r: np.isclose(r["forecast_reduction"], (r["depth_rmse"] - r["trajectory_rmse"]) / r["depth_rmse"])),
    ("regret reduction = (depth regret − trajectory regret) / depth regret",
     ["regret_reduction", "depth_regret", "trajectory_regret"], True,
     lambda r: np.isclose(r["regret_reduction"], (r["depth_regret"] - r["trajectory_regret"]) / r["depth_regret"])),
    ("regret fold changes = exp(regret)",
     ["depth_regret_fold", "depth_regret", "trajectory_regret_fold", "trajectory_regret"], True,
     lambda r: np.isclose(r["depth_regret_fold"], np.exp(r["depth_regret"]))
     and np.isclose(r["trajectory_regret_fold"], np.exp(r["trajectory_regret"]))),
    ("share of different choices = count / PDX models",
     ["share_different_choice", "n_different_choice", "regret_n_models"], True,
     lambda r: np.isclose(r["share_different_choice"], r["n_different_choice"] / r["regret_n_models"])),
    ("discordant share = discordant pairs / depth-matched pairs",
     ["discordant_share", "n_discordant_pairs", "n_depth_matched_pairs"], True,
     lambda r: np.isclose(r["discordant_share"], r["n_discordant_pairs"] / r["n_depth_matched_pairs"])),
    ("shared rank variance = ρ²", ["depth_durability_shared_variance", "depth_durability_rho"], True,
     lambda r: np.isclose(r["depth_durability_shared_variance"], r["depth_durability_rho"] ** 2)),
    ("combination shared rank variance = ρ²", ["gain_shared_variance", "gain_rho"], True,
     lambda r: np.isclose(r["gain_shared_variance"], r["gain_rho"] ** 2)),
    ("combination categories sum to one",
     ["share_superior_on_both", "share_superior_in_depth_only", "share_superior_in_durability_only",
      "share_superior_on_neither"], True,
     lambda r: np.isclose(r["share_superior_on_both"] + r["share_superior_in_depth_only"]
                          + r["share_superior_in_durability_only"] + r["share_superior_on_neither"], 1.0)),
    ("28-day fold change of the mechanism-class effect = exp(28 × effect)",
     ["mechanism_durability_fold_28", "mechanism_durability"], True,
     lambda r: np.isclose(r["mechanism_durability_fold_28"], np.exp(28 * r["mechanism_durability"]))),
    ("pair-rate gap fold change = exp(0.02 × 28)", ["fold_change_28_days"], True,
     lambda r: np.isclose(r["fold_change_28_days"], np.exp(GROWTH_RATE_GAP * PAIR_HORIZON_DAYS))),
    ("forecasting arms are post-nadir arms", ["forecast_n_arms", "n_post_nadir_arms"], True,
     lambda r: r["forecast_n_arms"] <= r["n_post_nadir_arms"]),
    ("forecasting endpoint decision follows its estimate, interval and margin",
     ["forecast_endpoint_met", "forecast_reduction", "forecast_reduction_low"], True,
     lambda r: r["forecast_endpoint_met"] == int(r["forecast_reduction"] >= FORECAST_MARGIN
                                                 and r["forecast_reduction_low"] > 0)),
    ("regret endpoint decision follows its estimate, interval and margin",
     ["regret_endpoint_met", "regret_reduction", "regret_reduction_low"], True,
     lambda r: r["regret_endpoint_met"] == int(r["regret_reduction"] >= REGRET_MARGIN
                                               and r["regret_reduction_low"] > 0)),
    ("baseline-plus-slope ranking accuracy plus the difference gives the trajectory accuracy",
     ["ranking_accuracy_baseline_slope", "ranking_trajectory_minus_baseline_slope", "ranking_accuracy_trajectory"],
     True,
     lambda r: np.isclose(r["ranking_accuracy_baseline_slope"] + r["ranking_trajectory_minus_baseline_slope"],
                          r["ranking_accuracy_trajectory"])),
    ("converged calibration replicates do not exceed all replicates",
     ["calibration_converged_replicates", "calibration_replicates"], True,
     lambda r: r["calibration_converged_replicates"] <= r["calibration_replicates"]),
]


check_rows = []
for description, keys, required, condition in DERIVED_CHECKS:
    if not all(key in RESULTS and usable_value(RESULTS[key]) for key in keys):
        status = "not applicable"
    else:
        status = "holds" if bool(condition(RESULTS)) else "fails"
    check_rows.append({"check": description, "required": required, "status": status})

# P values and proportions must lie in [0, 1].
P_VALUE_PATTERN = re.compile(r"(^|_)(p|q)(_|$)")
PROPORTION_PATTERN = re.compile(r"share|accuracy|chose_tolerant|^exposure_decisive_|^exposure_accuracy_")
for key, value in RESULTS.items():
    if not isinstance(value, (int, float)) or isinstance(value, bool) or not usable_value(value):
        continue
    if P_VALUE_PATTERN.search(key) or PROPORTION_PATTERN.search(key):
        kind = "P value" if P_VALUE_PATTERN.search(key) else "proportion"
        check_rows.append({"check": f"{kind} in [0, 1]: {key}", "required": True,
                           "status": "holds" if 0.0 <= value <= 1.0 else "fails"})

# No stored number is infinite.
infinite = [key for key, value in RESULTS.items()
            if isinstance(value, float) and np.isinf(value)]
check_rows.append({"check": "no stored value is infinite", "required": True,
                   "status": "holds" if not infinite else f"fails: {infinite}"})

# Each estimate lies inside its own 95% interval (expected, not required).
for key, value in RESULTS.items():
    low, high = RESULTS.get(f"{key}_low"), RESULTS.get(f"{key}_high")
    if all(isinstance(v, (int, float)) and not isinstance(v, bool) and usable_value(v) for v in (value, low, high)):
        check_rows.append({"check": f"estimate inside its interval: {key}", "required": False,
                           "status": "holds" if low <= value <= high else "fails"})

consistency_checks = pd.DataFrame(check_rows)
required_failures = consistency_checks.loc[consistency_checks.required
                                           & consistency_checks.status.str.startswith("fails")]
expected_failures = consistency_checks.loc[~consistency_checks.required
                                           & consistency_checks.status.str.startswith("fails")]
save_record(consistency_checks, "consistency_checks")
print("15.2 INTERNAL CONSISTENCY OF THE STORED RESULTS")
print(consistency_checks.groupby(["required", "status"]).size().rename("checks").to_string())
print("\n  derived quantities and endpoint decisions")
print(consistency_checks.loc[consistency_checks.check.isin([c[0] for c in DERIVED_CHECKS])]
      .to_string(index=False))
if len(expected_failures):
    print("\n  expected checks that did not hold (reported, not stopping the run)")
    print(expected_failures.to_string(index=False))
assert required_failures.empty, f"required consistency checks failed:\n{required_failures.to_string(index=False)}"
print(f"\n  all {int(consistency_checks.required.sum())} required checks hold")
```

```python
# ==========================================================================
# 15.3 Run manifest and summary of the main results
#     The manifest lists the software, platform, seeds and key constants, the
#     size and SHA-256 checksum of every input file read and every output
#     file written (Parquet caches under outputs/interim/ are left out), and
#     the run time. It is written last, so it does not list itself.
# ==========================================================================

MANIFEST_PATH = OUTPUT_DIR / "manifest.json"
input_paths = [WORKBOOK_PATH, PROJECT_DIR / "pdxe_utils.py"]
input_paths += [PRESPEC_DIR / file_name for file_name, _ in PRESPECIFIED_PLANS.values()
                if (PRESPEC_DIR / file_name).exists()]
input_paths += list(pdxnet_files) + ([OBERG_PATH] if OBERG_AVAILABLE else []) + list(hallmark_files)
input_paths += [PDS_DIR / study / name for study, spec in CLINICAL_STUDIES.items() for name in spec["files"]
                if (PDS_DIR / study / name).exists()]
output_paths = sorted(path for path in OUTPUT_DIR.rglob("*")
                      if path.is_file() and INTERIM_DIR not in path.parents and path != MANIFEST_PATH)

elapsed_seconds = time.perf_counter() - RUN_STARTED
manifest = {
    "run_started_utc": RUN_STARTED_UTC,
    "run_finished_utc": datetime.now(timezone.utc).isoformat(timespec="seconds"),
    "elapsed_seconds": round(elapsed_seconds, 1),
    "software": SOFTWARE_VERSIONS,
    "platform": PLATFORM,
    "pdxe_utils_sha256": UTILS_SHA256,
    "seeds": SEEDS,
    "constants": {
        "volume_floor_mm3": VOLUME_FLOOR, "best_average_response_from_day": MIN_DAY_FOR_BEST_AVERAGE_RESPONSE,
        "forecast_origin_day": FORECAST_ORIGIN_DAY, "forecast_horizon_day": FORECAST_HORIZON_DAY,
        "forecast_outcome_zeros": FORECAST_OUTCOME_ZEROS, "forecast_margin": FORECAST_MARGIN,
        "regret_margin": REGRET_MARGIN, "folds": N_FOLDS, "bootstrap_replicates": N_BOOTSTRAP,
        "control_bootstrap_replicates": N_BOOTSTRAP_CONTROLS, "measurement_noise": MEASUREMENT_NOISE,
        "kill_rate_range": KILL_RATE_RANGE, "termination_volume_mm3": TERMINATION_VOLUME,
        "simulation_noise_sd": SIMULATION_NOISE_SD,
    },
    "optional_inputs": OPTIONAL_INPUTS,
    "results_register": {"values": len(RESULTS), "json": REGISTER_JSON, "csv": REGISTER_CSV},
    "consistency_checks": {"required": int(consistency_checks.required.sum()),
                           "required_failed": len(required_failures),
                           "expected": int((~consistency_checks.required).sum()),
                           "expected_not_holding": len(expected_failures)},
    "inputs": [file_record(path) for path in input_paths],
    "outputs": [file_record(path) for path in output_paths],
}
MANIFEST_PATH.write_text(json.dumps(json_ready(manifest), indent=1, ensure_ascii=False), encoding="utf-8")

print("15.3 RUN MANIFEST")
print(f"  {relative(MANIFEST_PATH)}")
print(f"  inputs {len(manifest['inputs'])}, outputs {len(manifest['outputs'])}; "
      f"run time {elapsed_seconds / 60:.1f} minutes")
for entry in manifest["inputs"]:
    print(f"    input   {entry['path']:<70} {entry['sha256'][:16]}")

# --- summary of the main results ----------------------------------------------
# Each line shows a stored value and, where the register holds its interval,
# the 95% interval beside it. Values of optional sections that were not
# computed are shown as such.
SUMMARY = [
    ("4", "post-nadir arms", "n_post_nadir_arms", "int"),
    ("4", "depth against durability, Spearman ρ", "depth_durability_rho", "f3"),
    ("4", "depth-matched pairs discordant in durability", "discordant_share", "pct"),
    ("5", "mechanism discrimination, mean accuracy (unpaired / paired)",
     ("discrimination_mean_accuracy", "paired_discrimination_mean_accuracy"), "f3"),
    ("5", "simulation-based calibration, converged replicates", "calibration_converged_replicates", "int"),
    ("6", "measurement noise (log units)", "measurement_noise", "f4"),
    ("6", "REML share of log kill-rate variance, treatment / PDX model",
     ("reml_share_treatment", "reml_share_model"), "f3"),
    ("7", "forecasting, relative RMSE reduction", "forecast_reduction", "pct"),
    ("7", "treatment selection, relative regret reduction", "regret_reduction", "pct"),
    ("7", "pairwise ranking accuracy, depth / trajectory",
     ("ranking_accuracy_depth", "ranking_accuracy_trajectory"), "f3"),
    ("7", "PDX models where the two sets choose differently", "share_different_choice", "pct"),
    ("8", "trajectory over expanded depth", "trajectory_over_expanded", "pct"),
    ("8", "trajectory over depth plus current volume", "trajectory_over_current_volume", "pct"),
    ("8", "trajectory over depth plus permuted shape", "trajectory_over_permuted_shape", "pct"),
    ("8", "baseline plus slope over depth", "baseline_slope_over_depth", "pct"),
    ("8", "trajectory over baseline plus slope", "trajectory_over_baseline_slope", "pct"),
    ("8", "noise-padded depth over depth", "noise_over_depth", "pct"),
    ("8", "without the post-nadir restriction", "forecast_all_reduction", "pct"),
    ("8", "untreated arms, day 14 to 21", "untreated_forecast_14_21", "pct"),
    ("8", "treated arms, day 14 to 21", "treated_forecast_14_21", "pct"),
    ("8", "ranking accuracy, pairs separated by > 8 noise SDs (depth / trajectory)",
     ("top_stratum_depth_accuracy", "top_stratum_trajectory_accuracy"), "f3"),
    ("8", "ranking accuracy, trajectory minus baseline plus slope", "ranking_trajectory_minus_baseline_slope", "f3"),
    ("8", "same, pairs separated by > 8 noise SDs", "top_stratum_trajectory_minus_baseline_slope", "f3"),
    ("9", "dissociation statistic D, curated signatures", "dissociation", "f3"),
    ("9", "dissociation statistic D, hallmark signatures", "dissociation_hallmark", "f3"),
    ("10", "combination evaluations superior on both endpoints", "share_superior_on_both", "pct"),
    ("10", "depth gain against durability gain, Spearman ρ", "gain_rho", "f3"),
    ("11", "cytostatic minus comparator durability (per day)", "mechanism_durability", "f4"),
    ("11", "same, compound as the unit (per day)", "compound_effect", "f4"),
    ("11", "mixed-effects model, crossed: intercept / P", ("mixed_model_intercept", "mixed_model_p"), "f4"),
    ("12", "structural separation over 28 days (log units)", "exposure_separation_28", "f4"),
    ("12", "separation fold change, day 28 to 84 (half-life 0.25 d)", "exposure_fold_28_to_84_hl025", "f2"),
    ("13", "PDXNet between-animal SD / between-model SD", "pdxnet_between_ratio_control", "f3"),
    ("14", "pooled clinical, C difference (depth plus shape − depth)", "clinical_difference", "f3"),
]


print("\nSUMMARY OF THE MAIN RESULTS (95% intervals where the register holds them)")
current_section = None
for section, label, keys, style in SUMMARY:
    if section != current_section:
        print(f"  Section {section}")
        current_section = section
    keys = keys if isinstance(keys, tuple) else (keys,)
    text = " / ".join(format_number(RESULTS.get(key), style if not key.endswith("_p") else "p")
                      for key in keys)
    if len(keys) == 1 and usable_value(RESULTS.get(f"{keys[0]}_low")):
        text += (f" ({format_number(RESULTS[keys[0] + '_low'], style)} to "
                 f"{format_number(RESULTS[keys[0] + '_high'], style)})")
    print(f"    {label:<74} {text}")
```
